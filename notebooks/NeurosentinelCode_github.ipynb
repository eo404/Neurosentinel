**Compare the behavioral kp events with EEG markers**

```python
from pathlib import Path
import scipy.io as sio
import mne
import numpy as np

DATA_DIR = Path("/content/drive/MyDrive/NeuroSentinel/data/SpanishBCBL")
EEG_DIR = DATA_DIR / "EEG/EEG"

MAT_PATH = DATA_DIR / "EEG/logs/S10_session1_block1_list1.mat"
EEG_PATH = EEG_DIR / "010_DECOMEG_S1_noid_task1.vhdr"

# ============================================================
# Load behavioral data
# ============================================================

mat = sio.loadmat(MAT_PATH, simplify_cells=True)

print("=" * 70)
print("BEHAVIORAL KEYPRESS EVENTS")
print("=" * 70)

kp = mat["kp"]

print("kp type:", type(kp))
print("kp shape:", np.shape(kp))

# Handle structured array
kp_arr = np.asarray(kp).squeeze()

print("\nFirst 20 keypress events:")

for i, event in enumerate(kp_arr[:20]):
    try:
        print(
            f"{i:02d} | "
            f"Time={event['Time']:.4f} | "
            f"Keycode={event['Keycode']} | "
            f"CookedKey={event['CookedKey']}"
        )
    except Exception as e:
        print(i, event, e)


# ============================================================
# Load EEG markers
# ============================================================

print("\n" + "=" * 70)
print("EEG MARKERS")
print("=" * 70)

raw = mne.io.read_raw_brainvision(
    str(EEG_PATH),
    preload=False,
    verbose=False
)

annotations = raw.annotations

# Extract only Stimulus markers
stim_annotations = [
    ann for ann in annotations
    if str(ann["description"]).startswith("Stimulus/")
]

print("Total stimulus markers:", len(stim_annotations))

print("\nFirst 50 stimulus markers:")

for i, ann in enumerate(stim_annotations[:50]):
    print(
        f"{i:02d} | "
        f"onset={ann['onset']:.4f}s | "
        f"{ann['description']}"
    )


# ============================================================
# Extract numeric EEG stimulus codes
# ============================================================

print("\n" + "=" * 70)
print("NUMERIC STIMULUS CODES")
print("=" * 70)

codes = []

for ann in stim_annotations:

    desc = str(ann["description"])

    # Example: Stimulus/S 69
    if "/" in desc:
        value = desc.split("/")[-1].strip()

        if value.startswith("S"):
            value = value[1:].strip()

            try:
                code = int(value)
                codes.append(
                    {
                        "onset": ann["onset"],
                        "code": code,
                        "description": desc
                    }
                )
            except ValueError:
                pass

print("Numeric stimulus markers:", len(codes))

print("\nFirst 50 numeric markers:")

for i, item in enumerate(codes[:50]):
    print(
        f"{i:02d} | "
        f"{item['onset']:.4f}s | "
        f"code={item['code']}"
    )


# ============================================================
# Compare behavioral CookedKey codes with EEG codes
# ============================================================

print("\n" + "=" * 70)
print("CODE COMPARISON")
print("=" * 70)

behavior_codes = []

for event in kp_arr:

    try:
        behavior_codes.append(int(event["CookedKey"]))
    except Exception:
        pass

eeg_codes = [x["code"] for x in codes]

print("Behavioral keypress count:", len(behavior_codes))
print("EEG numeric marker count :", len(eeg_codes))

print("\nBehavioral CookedKey codes:")
print(behavior_codes[:40])

print("\nFirst EEG codes:")
print(eeg_codes[:80])
```

**Next inspection: determine exact marker semantics**

```python
from pathlib import Path
from collections import Counter
import mne

DATA_DIR = Path("/content/drive/MyDrive/NeuroSentinel/data/SpanishBCBL")
EEG_PATH = DATA_DIR / "EEG/EEG/010_DECOMEG_S1_noid_task1.vhdr"

raw = mne.io.read_raw_brainvision(
    str(EEG_PATH),
    preload=False,
    verbose=False
)

# ------------------------------------------------------------
# Extract stimulus markers
# ------------------------------------------------------------

stim = []

for ann in raw.annotations:
    desc = str(ann["description"])

    if desc.startswith("Stimulus/"):
        code = desc.split("/")[-1].strip()

        if code.startswith("S"):
            try:
                code = int(code[1:].strip())
                stim.append((ann["onset"], code))
            except ValueError:
                pass

print("=" * 70)
print("MARKER CODE FREQUENCY")
print("=" * 70)

counter = Counter(code for _, code in stim)

for code, count in sorted(counter.items()):
    print(f"{code:4d} : {count}")


# ------------------------------------------------------------
# Show marker stream grouped into blocks
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FIRST MARKER BLOCKS")
print("=" * 70)

# Print markers from 60–90 seconds
for onset, code in stim:
    if 60 <= onset <= 90:
        print(f"{onset:9.3f}s   S {code}")


# ------------------------------------------------------------
# Extract likely character markers
# ASCII printable = 32..126
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FIRST PRINTABLE CHARACTER SEQUENCE")
print("=" * 70)

chars = []

for onset, code in stim:

    if 60 <= onset <= 90 and 32 <= code <= 126:
        chars.append((onset, chr(code)))

print("".join(ch for _, ch in chars))

print("\nDetailed:")
for onset, ch in chars:
    print(f"{onset:9.3f}s   {repr(ch)}")
```

```python
from pathlib import Path
import mne
import numpy as np

DATA_DIR = Path("/content/drive/MyDrive/NeuroSentinel/data/SpanishBCBL")
EEG_PATH = DATA_DIR / "EEG/EEG/010_DECOMEG_S1_noid_task1.vhdr"

raw = mne.io.read_raw_brainvision(
    str(EEG_PATH),
    preload=False,
    verbose=False
)

# ------------------------------------------------------------
# Extract all numeric stimulus markers
# ------------------------------------------------------------

markers = []

for ann in raw.annotations:

    desc = str(ann["description"])

    if desc.startswith("Stimulus/"):

        value = desc.split("/")[-1].strip()

        if value.startswith("S"):
            try:
                code = int(value[1:].strip())
                markers.append((ann["onset"], code))
            except ValueError:
                pass


# ------------------------------------------------------------
# Find all Enter markers (S13)
# ------------------------------------------------------------

enter_events = [
    (onset, code)
    for onset, code in markers
    if code == 13
]

print("=" * 70)
print("ENTER / TRIAL-END EVENTS")
print("=" * 70)

print("Number of S13 events:", len(enter_events))

for i, (onset, code) in enumerate(enter_events):
    print(f"Trial {i:02d} | Enter = {onset:.3f}s")


# ------------------------------------------------------------
# Calculate duration between consecutive Enter events
# ------------------------------------------------------------

enter_times = np.array([x[0] for x in enter_events])

print("\n" + "=" * 70)
print("TRIAL INTERVALS")
print("=" * 70)

intervals = np.diff(enter_times)

print("Number of intervals:", len(intervals))
print("Minimum:", intervals.min())
print("Maximum:", intervals.max())
print("Mean   :", intervals.mean())
print("Median :", np.median(intervals))

print("\nFirst 20 intervals:")
print(intervals[:20])


# ------------------------------------------------------------
# Compare with behavioral ts
# ------------------------------------------------------------

import scipy.io as sio

MAT_PATH = DATA_DIR / "EEG/logs/S10_session1_block1_list1.mat"

mat = sio.loadmat(MAT_PATH, simplify_cells=True)

ts = np.asarray(mat["ts"], dtype=float).flatten()

print("\n" + "=" * 70)
print("BEHAVIORAL ts")
print("=" * 70)

print("Number of ts:", len(ts))
print("First 20:")
print(ts[:20])

print("\nStatistics:")
print("Min   :", np.nanmin(ts))
print("Max   :", np.nanmax(ts))
print("Mean  :", np.nanmean(ts))
print("Median:", np.nanmedian(ts))


# ------------------------------------------------------------
# Compare behavioral ts with EEG trial intervals
# ------------------------------------------------------------

n = min(len(intervals), len(ts) - 1)

print("\n" + "=" * 70)
print("DIRECT ts vs EEG INTERVAL COMPARISON")
print("=" * 70)

for i in range(min(20, n)):

    print(
        f"Trial {i:02d} | "
        f"EEG interval={intervals[i]:8.3f}s | "
        f"behavior ts={ts[i]:8.3f}s | "
        f"difference={intervals[i] - ts[i]:8.3f}s"
    )
```

```python
from pathlib import Path
import numpy as np
import scipy.io as sio
import mne

DATA_DIR = Path("/content/drive/MyDrive/NeuroSentinel/data/SpanishBCBL")

EEG_PATH = DATA_DIR / "EEG/EEG/010_DECOMEG_S1_noid_task1.vhdr"
MAT_PATH = DATA_DIR / "EEG/logs/S10_session1_block1_list1.mat"

# ---------------------------------------------------------
# Load EEG markers
# ---------------------------------------------------------
raw = mne.io.read_raw_brainvision(
    str(EEG_PATH),
    preload=False,
    verbose=False
)

markers = []

for ann in raw.annotations:
    desc = str(ann["description"])

    if desc.startswith("Stimulus/"):
        value = desc.split("/")[-1].strip()

        if value.startswith("S"):
            try:
                code = int(value[1:])
                markers.append((ann["onset"], code))
            except ValueError:
                pass

# S13 = trial end
enter_indices = [
    i for i, (_, code) in enumerate(markers)
    if code == 13
]

print("=" * 70)
print("TRIAL TEXT RECONSTRUCTION")
print("=" * 70)

# ---------------------------------------------------------
# Load behavioral sequences
# ---------------------------------------------------------
mat = sio.loadmat(
    MAT_PATH,
    simplify_cells=True
)

sequences = np.asarray(mat["sequences"], dtype=object).flatten()

print("Behavioral sequences:", len(sequences))
print("EEG S13 trials      :", len(enter_indices))

# ---------------------------------------------------------
# Decode markers between S13 boundaries
# ---------------------------------------------------------
def decode_marker_stream(codes):
    chars = []
    special = []

    for code in codes:

        # Printable ASCII
        if 32 <= code <= 126:
            chars.append(chr(code))

        # Backspace
        elif code == 8:
            if chars:
                chars.pop()

        # Enter
        elif code == 13:
            pass

        else:
            special.append(code)

    return "".join(chars), special


# ---------------------------------------------------------
# Reconstruct each trial
# ---------------------------------------------------------
trial_results = []

start_idx = 0

for trial, end_idx in enumerate(enter_indices):

    trial_codes = [
        code
        for _, code in markers[start_idx:end_idx + 1]
    ]

    reconstructed, special = decode_marker_stream(trial_codes)

    expected = str(sequences[trial]).rstrip()

    trial_results.append({
        "trial": trial,
        "expected": expected,
        "reconstructed": reconstructed.rstrip(),
        "special_codes": special,
        "match": reconstructed.rstrip() == expected
    })

    print(f"\nTrial {trial:02d}")
    print("Expected     :", repr(expected))
    print("Reconstructed:", repr(reconstructed.rstrip()))
    print("Match        :", reconstructed.rstrip() == expected)

    if special:
        print("Special codes:", special)

    start_idx = end_idx + 1


# ---------------------------------------------------------
# Summary
# ---------------------------------------------------------
matches = sum(x["match"] for x in trial_results)

print("\n" + "=" * 70)
print("SUMMARY")
print("=" * 70)
print(f"Exact matches: {matches}/{len(trial_results)}")
print(f"Match rate   : {matches / len(trial_results):.2%}")
```

```python
import re
from difflib import SequenceMatcher
import numpy as np
import scipy.io as sio
import mne

# ---------------------------------------------------------
# Load behavioral sequences
# ---------------------------------------------------------
MAT_PATH = DATA_DIR / "EEG/logs/S10_session1_block1_list1.mat"

mat = sio.loadmat(
    MAT_PATH,
    simplify_cells=True
)

sequences = [
    str(x).strip()
    for x in np.asarray(mat["sequences"], dtype=object).flatten()
]

# ---------------------------------------------------------
# Load EEG markers
# ---------------------------------------------------------
EEG_PATH = DATA_DIR / "EEG/EEG/010_DECOMEG_S1_noid_task1.vhdr"

raw = mne.io.read_raw_brainvision(
    str(EEG_PATH),
    preload=False,
    verbose=False
)

markers = []

for ann in raw.annotations:

    desc = str(ann["description"])

    if desc.startswith("Stimulus/"):

        value = desc.split("/")[-1].strip()

        if value.startswith("S"):
            try:
                code = int(value[1:])
                markers.append((ann["onset"], code))
            except ValueError:
                pass


# ---------------------------------------------------------
# S13 trial boundaries
# ---------------------------------------------------------
s13_indices = [
    i for i, (_, code) in enumerate(markers)
    if code == 13
]


# ---------------------------------------------------------
# Decode EEG trial
# ---------------------------------------------------------
def decode_trial(codes):

    chars = []

    for code in codes:

        # ASCII characters
        if 32 <= code <= 126:
            chars.append(chr(code))

        # Backspace
        elif code == 8:
            if chars:
                chars.pop()

    return "".join(chars).strip()


def normalize(text):

    text = text.lower()

    # Remove repeated whitespace
    text = re.sub(r"\s+", " ", text)

    return text.strip()


# ---------------------------------------------------------
# Reconstruct trials
# ---------------------------------------------------------
eeg_trials = []

start = 0

for end in s13_indices:

    codes = [
        code
        for _, code in markers[start:end + 1]
    ]

    text = decode_trial(codes)

    eeg_trials.append(text)

    start = end + 1


# ---------------------------------------------------------
# Find best behavioral sequence for each EEG trial
# ---------------------------------------------------------
print("=" * 80)
print("EEG TRIAL → BEST BEHAVIORAL SEQUENCE")
print("=" * 80)

matches = []

for i, eeg_text in enumerate(eeg_trials):

    eeg_norm = normalize(eeg_text)

    scores = []

    for j, sequence in enumerate(sequences):

        seq_norm = normalize(sequence)

        score = SequenceMatcher(
            None,
            eeg_norm,
            seq_norm
        ).ratio()

        scores.append((score, j, sequence))

    scores.sort(reverse=True)

    best_score, best_idx, best_sequence = scores[0]

    matches.append(best_idx)

    print(
        f"\nEEG Trial {i:02d}"
        f"\n  EEG       : {eeg_text!r}"
        f"\n  Best seq  : {best_idx:02d}"
        f"\n  Behavior  : {best_sequence!r}"
        f"\n  Similarity: {best_score:.3f}"
    )


# ---------------------------------------------------------
# Check whether ordering is preserved
# ---------------------------------------------------------
print("\n" + "=" * 80)
print("ORDER CHECK")
print("=" * 80)

expected_indices = list(range(66))

correct_positions = sum(
    a == b
    for a, b in zip(matches, expected_indices)
)

print(
    f"Same-position matches: "
    f"{correct_positions}/66 "
    f"({correct_positions / 66:.2%})"
)

print("\nFirst 30 mapped indices:")
print(matches[:30])
```

```python
import numpy as np
import scipy.io as sio
import mne

# ---------------------------------------------------------
# Load behavioral data
# ---------------------------------------------------------
MAT_PATH = DATA_DIR / "EEG/logs/S10_session1_block1_list1.mat"

mat = sio.loadmat(
    MAT_PATH,
    simplify_cells=True
)

# ---------------------------------------------------------
# Load EEG
# ---------------------------------------------------------
EEG_PATH = DATA_DIR / "EEG/EEG/010_DECOMEG_S1_noid_task1.vhdr"

raw = mne.io.read_raw_brainvision(
    str(EEG_PATH),
    preload=False,
    verbose=False
)

# ---------------------------------------------------------
# EEG markers
# ---------------------------------------------------------
markers = []

for ann in raw.annotations:

    desc = str(ann["description"])

    if desc.startswith("Stimulus/"):

        value = desc.split("/")[-1].strip()

        if value.startswith("S"):

            try:
                code = int(value[1:])
                markers.append((float(ann["onset"]), code))
            except ValueError:
                pass


# ---------------------------------------------------------
# Trial boundaries
# ---------------------------------------------------------
s13_indices = [
    i for i, (_, code) in enumerate(markers)
    if code == 13
]


# ---------------------------------------------------------
# Behavioral keypress events
# ---------------------------------------------------------
key_events = []

for event in mat["key"]:

    if int(event["Pressed"]) != 1:
        continue

    cooked = int(event["CookedKey"])

    # Keep printable keys and backspace
    if 32 <= cooked <= 126 or cooked == 8:

        key_events.append({
            "time": float(event["Time"]),
            "code": int(event["Keycode"]),
            "cooked": cooked
        })


# ---------------------------------------------------------
# IMPORTANT:
# Find behavioral events corresponding to each EEG trial
# using the known trial ordering.
#
# We use behavioral sequence text to identify the segment
# from the key event stream.
# ---------------------------------------------------------

def find_sequence_in_events(events, target):

    target = target.strip().lower()

    chars = []
    positions = []

    for i, event in enumerate(events):

        cooked = event["cooked"]

        if 32 <= cooked <= 126:

            chars.append(chr(cooked).lower())
            positions.append(i)

    text = "".join(chars)

    start = text.find(target)

    if start == -1:
        return None

    selected = positions[start:start + len(target)]

    return [events[i] for i in selected]


# ---------------------------------------------------------
# Use Trial 24 as a clean reference
# ---------------------------------------------------------
trial = 24

target = str(
    np.asarray(mat["sequences"], dtype=object)
    .flatten()[trial]
).strip()

# EEG trial boundaries
start_idx = 0 if trial == 0 else s13_indices[trial - 1] + 1
end_idx = s13_indices[trial]

eeg_trial = [
    (onset, code)
    for onset, code in markers[start_idx:end_idx + 1]
    if (32 <= code <= 126) or code == 8
]

eeg_chars = [
    chr(code).lower()
    for _, code in eeg_trial
    if 32 <= code <= 126
]

eeg_text = "".join(eeg_chars)

print("=" * 70)
print("CLOCK OFFSET REFERENCE — TRIAL 24")
print("=" * 70)

print("Target:")
print(target.lower())

print("\nEEG marker text:")
print(eeg_text)

# ---------------------------------------------------------
# Find behavioral occurrence of this target
# ---------------------------------------------------------
behavior_match = find_sequence_in_events(
    key_events,
    target
)

if behavior_match is None:

    print("\nCould not locate behavioral sequence.")

else:

    print("\nBehavioral events found:", len(behavior_match))

    # -----------------------------------------------------
    # Match behavioral characters to EEG characters
    # -----------------------------------------------------
    behavior_chars = [
        chr(e["cooked"]).lower()
        for e in behavior_match
        if 32 <= e["cooked"] <= 126
    ]

    eeg_printable = [
        (onset, code)
        for onset, code in eeg_trial
        if 32 <= code <= 126
    ]

    n = min(
        len(behavior_chars),
        len(eeg_printable)
    )

    offsets = []

    print("\nFirst 20 matched events:")

    for i in range(n):

        b_char = behavior_chars[i]
        eeg_time, eeg_code = eeg_printable[i]

        if b_char == chr(eeg_code).lower():

            # Behavioral time → EEG time
            offset = behavior_match[i]["time"] - eeg_time

            offsets.append(offset)

            if len(offsets) <= 20:

                print(
                    f"{i:02d} | "
                    f"{b_char!r} | "
                    f"behavior={behavior_match[i]['time']:.6f} | "
                    f"EEG={eeg_time:.6f} | "
                    f"offset={offset:.6f}"
                )

    # -----------------------------------------------------
    # Statistics
    # -----------------------------------------------------
    offsets = np.asarray(offsets)

    print("\n" + "=" * 70)
    print("OFFSET STATISTICS")
    print("=" * 70)

    print("Matched events:", len(offsets))

    if len(offsets):

        print(f"Mean   : {offsets.mean():.6f}")
        print(f"Median : {np.median(offsets):.6f}")
        print(f"Std    : {offsets.std():.6f}")
        print(f"Min    : {offsets.min():.6f}")
        print(f"Max    : {offsets.max():.6f}")
```

```python
from collections import defaultdict
import mne

# ============================================================
# LOAD EEG AND EXTRACT MARKERS
# ============================================================

EEG_FILE = "/content/drive/MyDrive/NeuroSentinel/data/SpanishBCBL/EEG/EEG/010_DECOMEG_S1_noid_task1.vhdr"

print("Loading EEG...")
raw = mne.io.read_raw_brainvision(
    EEG_FILE,
    preload=False,
    verbose=False
)

# Extract annotations/events
events, event_id = mne.events_from_annotations(
    raw,
    verbose=False
)

# Reverse event_id: integer event code -> marker description
id_to_desc = {v: k for k, v in event_id.items()}

markers = []

for sample, _, event_code in events:
    onset = sample / raw.info["sfreq"]

    desc = id_to_desc[event_code]

    # Convert "Stimulus/S 13" -> 13
    if desc.startswith("Stimulus/S "):
        try:
            code = int(desc.split()[-1])
            markers.append((onset, code))
        except ValueError:
            pass

print(f"EEG loaded: {raw.info['nchan']} channels")
print(f"Sampling rate: {raw.info['sfreq']} Hz")
print(f"Markers extracted: {len(markers)}")


# ============================================================
# GROUP MARKERS BY CODE
# ============================================================

by_code = defaultdict(list)

for i, (onset, code) in enumerate(markers):
    by_code[code].append((i, onset))


# ============================================================
# SPECIAL MARKER SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("SPECIAL MARKER SUMMARY")
print("=" * 70)

for code in sorted(by_code):

    if not (32 <= code <= 126):

        print(f"S{code:<4} | count = {len(by_code[code])}")

        for idx, onset in by_code[code][:3]:
            print(
                f"    marker_index={idx}, "
                f"time={onset:.3f}s"
            )


# ============================================================
# S13 CHECK
# ============================================================

print("\n" + "=" * 70)
print("TRIAL BOUNDARY CHECK: S13")
print("=" * 70)

s13 = by_code.get(13, [])

print("S13 count:", len(s13))

for idx, onset in s13[:10]:
    print(
        f"index={idx:4d} | "
        f"time={onset:.3f}s"
    )


# ============================================================
# S8 CHECK
# ============================================================

print("\n" + "=" * 70)
print("S8 CHECK")
print("=" * 70)

s8 = by_code.get(8, [])

print("S8 count:", len(s8))

for idx, onset in s8[:10]:
    print(
        f"index={idx:4d} | "
        f"time={onset:.3f}s"
    )
```

```python
import numpy as np
import pandas as pd
import scipy.io

# ============================================================
# LOAD BEHAVIORAL FILE
# ============================================================

MAT_FILE = (
    "/content/drive/MyDrive/NeuroSentinel/data/"
    "SpanishBCBL/EEG/logs/S10_session1_block1_list1.mat"
)

mat = scipy.io.loadmat(
    MAT_FILE,
    simplify_cells=True
)

sequences = np.asarray(mat["sequences"], dtype=object).flatten()
errp = np.asarray(mat["errp"]).flatten()
ts = np.asarray(mat["ts"]).flatten()

print("Behavioral sequences:", len(sequences))
print("S13 boundaries:", len(s13))


# ============================================================
# CREATE TRIAL TABLE
# ============================================================

trial_rows = []

for trial_idx in range(len(s13)):

    boundary_index = s13[trial_idx][0]
    boundary_time = s13[trial_idx][1]

    # Start of this trial
    if trial_idx == 0:
        start_index = 0
        start_time = markers[0][0]
    else:
        start_index = s13[trial_idx - 1][0] + 1
        start_time = s13[trial_idx - 1][1]

    # EEG trial duration
    duration = boundary_time - start_time

    target = str(sequences[trial_idx]).strip()

    trial_rows.append({
        "trial": trial_idx,
        "target": target,
        "eeg_start": start_time,
        "eeg_end": boundary_time,
        "duration_sec": duration,
        "errp": errp[trial_idx] if trial_idx < len(errp) else np.nan,
        "ts": ts[trial_idx] if trial_idx < len(ts) else np.nan,
        "start_marker_index": start_index,
        "end_marker_index": boundary_index
    })


trial_df = pd.DataFrame(trial_rows)


# ============================================================
# DISPLAY
# ============================================================

print("\n" + "=" * 80)
print("PHASE 3 — TRIAL ALIGNMENT TABLE")
print("=" * 80)

print(
    trial_df[
        [
            "trial",
            "target",
            "eeg_start",
            "eeg_end",
            "duration_sec",
            "errp",
            "ts"
        ]
    ].head(15).to_string(index=False)
)

print("\nTotal trials:", len(trial_df))

print(
    "\nDuration statistics:"
)

print(
    trial_df["duration_sec"].describe()
)
```

```python
# ============================================================
# PHASE 1 — CORRECT TRIAL 0 KEYSTROKE CHECK
# ============================================================

trial_start = markers[0][0]       # first marker = 63.351 s
trial_end = s13[0][1]             # first S13 = 75.001 s

target = str(sequences[0]).strip()

trial_events = []

for idx, (onset, code) in enumerate(markers):

    if trial_start <= onset < trial_end:

        if 32 <= code <= 126 or code == 8:

            if code == 8:
                label = "BACKSPACE"
            else:
                label = chr(code)

            trial_events.append(
                (idx, onset, code, label)
            )

print("=" * 80)
print("CORRECT TRIAL 0 KEYSTROKE EVENTS")
print("=" * 80)

print(f"Trial start: {trial_start:.3f}s")
print(f"Trial end:   {trial_end:.3f}s")
print(f"Events:      {len(trial_events)}")

for idx, onset, code, label in trial_events:

    print(
        f"{idx:4d} | "
        f"{onset:9.3f}s | "
        f"S{code:<3} | "
        f"{repr(label)}"
    )


# ============================================================
# RECONSTRUCT TEXT WITH BACKSPACE
# ============================================================

typed_text = ""

for _, _, code, _ in trial_events:

    if code == 8:
        typed_text = typed_text[:-1]

    elif 32 <= code <= 126:
        typed_text += chr(code)


print("\n" + "=" * 80)
print("TEXT COMPARISON")
print("=" * 80)

print("Target:")
print(repr(target))

print("\nEEG reconstructed:")
print(repr(typed_text))

print("\nTarget length:", len(target))
print("EEG length:   ", len(typed_text))
```

### **Single-File Preprocessing**

```python
import os
import json
import numpy as np
import mne

# ============================================================
# PHASE 2 — SINGLE-FILE EEG PREPROCESSING
# ============================================================

EEG_FILE = (
    "/content/drive/MyDrive/NeuroSentinel/data/"
    "SpanishBCBL/EEG/EEG/010_DECOMEG_S1_noid_task1.vhdr"
)

PHASE2_DIR = "/content/drive/MyDrive/NeuroSentinel/03_phase2_preprocessing"
os.makedirs(PHASE2_DIR, exist_ok=True)

print("=" * 80)
print("PHASE 2 — EEG PREPROCESSING")
print("=" * 80)

# ------------------------------------------------------------
# 1. Load raw BrainVision EEG
# ------------------------------------------------------------
print("\n[1/6] Loading EEG...")

raw = mne.io.read_raw_brainvision(
    EEG_FILE,
    preload=True,
    verbose=False
)

print("Original channels:", len(raw.ch_names))
print("Original sampling rate:", raw.info["sfreq"])
print("Original duration:", raw.times[-1], "seconds")

# ------------------------------------------------------------
# 2. Remove EOG channels
# ------------------------------------------------------------
print("\n[2/6] Removing EOG channels...")

eog_channels = [
    ch for ch in raw.ch_names
    if "EOG" in ch.upper()
]

print("Removing:", eog_channels)

raw.drop_channels(eog_channels)

print("Channels remaining:", len(raw.ch_names))

# ------------------------------------------------------------
# 3. Set standard 10-05 montage
# ------------------------------------------------------------
print("\n[3/6] Setting standard_1005 montage...")

raw.set_montage(
    "standard_1005",
    on_missing="warn"
)

print("Montage applied.")

# ------------------------------------------------------------
# 4. Band-pass filter: 0.1–20 Hz
# ------------------------------------------------------------
print("\n[4/6] Applying 0.1–20 Hz band-pass filter...")

raw.filter(
    l_freq=0.1,
    h_freq=20.0,
    method="fir",
    verbose=False
)

print("Filter applied.")

# ------------------------------------------------------------
# 5. Baseline correction
# ------------------------------------------------------------
print("\n[5/6] Baseline correction...")

# IMPORTANT:
# There is no valid continuous baseline interval specified
# by the current dataset inspection.
#
# Therefore, we do NOT fabricate a baseline interval here.
#
# Baseline correction will be performed later at the
# event/window level once the event alignment is established.

print("Continuous baseline correction skipped.")
print("Reason: no validated baseline interval available yet.")

# ------------------------------------------------------------
# 6. Resample to 50 Hz
# ------------------------------------------------------------
print("\n[6/6] Resampling 1000 Hz → 50 Hz...")

raw.resample(
    50,
    npad="auto",
    verbose=False
)

print("New sampling rate:", raw.info["sfreq"])
print("New duration:", raw.times[-1], "seconds")

# ============================================================
# VALIDATION
# ============================================================

print("\n" + "=" * 80)
print("PHASE 2 — VALIDATION")
print("=" * 80)

data = raw.get_data()

checks = {
    "channel_count": len(raw.ch_names),
    "sampling_rate": raw.info["sfreq"],
    "duration_seconds": raw.times[-1],
    "data_shape": list(data.shape),
    "contains_nan": bool(np.isnan(data).any()),
    "contains_inf": bool(np.isinf(data).any()),
    "min_amplitude": float(np.min(data)),
    "max_amplitude": float(np.max(data)),
    "mean_amplitude": float(np.mean(data)),
    "std_amplitude": float(np.std(data)),
    "annotation_count": len(raw.annotations),
}

for key, value in checks.items():
    print(f"{key}: {value}")

# Expected checks
assert len(raw.ch_names) == 61, "Expected 61 EEG channels."
assert raw.info["sfreq"] == 50.0, "Sampling rate is not 50 Hz."
assert not np.isnan(data).any(), "NaN values detected."
assert not np.isinf(data).any(), "Inf values detected."

print("\n✓ Channel count: 61")
print("✓ Sampling rate: 50 Hz")
print("✓ No NaN values")
print("✓ No Inf values")
print("✓ Preprocessing completed successfully")

# ------------------------------------------------------------
# Save validation metadata
# ------------------------------------------------------------

output_file = os.path.join(
    PHASE2_DIR,
    "phase2_single_file_validation.json"
)

with open(output_file, "w") as f:
    json.dump(checks, f, indent=4)

print("\nValidation file saved:")
print(output_file)

print("\n" + "=" * 80)
print("PHASE 2 SINGLE-FILE PREPROCESSING COMPLETE")
print("=" * 80)
```

# **PHASE 3 — Event Alignment**

**Step 1 — Inspect event structure on the preprocessed recording**

```python
import os
import numpy as np
import pandas as pd
import mne

# ============================================================
# PHASE 3 — EVENT ALIGNMENT
# STEP 1: EEG EVENT INSPECTION
# ============================================================

EEG_FILE = (
    "/content/drive/MyDrive/NeuroSentinel/data/"
    "SpanishBCBL/EEG/EEG/010_DECOMEG_S1_noid_task1.vhdr"
)

PHASE3_DIR = (
    "/content/drive/MyDrive/NeuroSentinel/"
    "04_phase3_event_alignment"
)

os.makedirs(PHASE3_DIR, exist_ok=True)

print("=" * 80)
print("PHASE 3 — EVENT ALIGNMENT")
print("=" * 80)

# ------------------------------------------------------------
# Load EEG
# ------------------------------------------------------------
raw = mne.io.read_raw_brainvision(
    EEG_FILE,
    preload=False,
    verbose=False
)

print("\nRecording:")
print("Channels:", len(raw.ch_names))
print("Sampling rate:", raw.info["sfreq"])
print("Duration:", raw.times[-1], "seconds")
print("Annotations:", len(raw.annotations))

# ------------------------------------------------------------
# Extract annotations
# ------------------------------------------------------------
events, event_id = mne.events_from_annotations(
    raw,
    verbose=False
)

print("\nTotal MNE events:", len(events))

# Reverse event ID
id_to_event = {
    value: key
    for key, value in event_id.items()
}

# ------------------------------------------------------------
# Create event table
# ------------------------------------------------------------
event_rows = []

for sample, _, event_code in events:

    event_name = id_to_event.get(
        event_code,
        f"UNKNOWN_{event_code}"
    )

    timestamp = sample / raw.info["sfreq"]

    event_rows.append({
        "sample_1000hz": int(sample),
        "timestamp_seconds": float(timestamp),
        "event_code": int(event_code),
        "event": event_name
    })

events_df = pd.DataFrame(event_rows)

print("\nFirst 30 events:")
display(events_df.head(30))

# ------------------------------------------------------------
# Event counts
# ------------------------------------------------------------
print("\nEvent counts:")
print(
    events_df["event"]
    .value_counts()
    .head(30)
)

# ------------------------------------------------------------
# Special markers
# ------------------------------------------------------------
print("\nSpecial event markers:")

special = events_df[
    events_df["event"].astype(str).str.startswith("S")
]

print(
    special["event"]
    .value_counts()
)

# ------------------------------------------------------------
# S13 trial boundaries
# ------------------------------------------------------------
s13 = events_df[
    events_df["event"] == "S 13"
].copy()

if len(s13) == 0:
    # Some BrainVision files may represent the annotation differently.
    s13 = events_df[
        events_df["event"].astype(str).str.contains(
            r"S\s*13",
            regex=True
        )
    ].copy()

print("\nS13 trial-boundary events:", len(s13))

if len(s13) > 0:
    print("\nS13 timestamps:")
    print(
        s13[
            ["sample_1000hz", "timestamp_seconds", "event"]
        ].head(20).to_string(index=False)
    )

# ------------------------------------------------------------
# Save raw event table
# ------------------------------------------------------------
events_file = os.path.join(
    PHASE3_DIR,
    "phase3_raw_eeg_events.csv"
)

events_df.to_csv(
    events_file,
    index=False
)

print("\nEvent table saved:")
print(events_file)

print("\n" + "=" * 80)
print("PHASE 3 STEP 1 COMPLETE")
print("=" * 80)
```

**Step 2: Trial-Level Alignment**

```python
import os
import re
import numpy as np
import pandas as pd
from difflib import SequenceMatcher
from scipy.io import loadmat

# ============================================================
# PHASE 3 — STEP 2
# EEG TRIALS ↔ BEHAVIORAL SEQUENCES
# ============================================================

EEG_EVENTS_FILE = (
    "/content/drive/MyDrive/NeuroSentinel/"
    "04_phase3_event_alignment/phase3_raw_eeg_events.csv"
)

MAT_FILE = (
    "/content/drive/MyDrive/NeuroSentinel/data/"
    "SpanishBCBL/EEG/logs/"
    "S10_session1_block1_list1.mat"
)

PHASE3_DIR = (
    "/content/drive/MyDrive/NeuroSentinel/"
    "04_phase3_event_alignment"
)

os.makedirs(PHASE3_DIR, exist_ok=True)

print("=" * 80)
print("PHASE 3 — STEP 2: TRIAL-LEVEL ALIGNMENT")
print("=" * 80)

# ------------------------------------------------------------
# 1. Load EEG events
# ------------------------------------------------------------

events_df = pd.read_csv(EEG_EVENTS_FILE)

# Normalize event names
events_df["event"] = (
    events_df["event"]
    .astype(str)
    .str.replace("Stimulus/", "", regex=False)
    .str.strip()
)

# ------------------------------------------------------------
# 2. Get S13 boundaries
# ------------------------------------------------------------

s13 = events_df[
    events_df["event"].str.replace(" ", "") == "S13"
].copy()

s13 = s13.sort_values("timestamp_seconds").reset_index(drop=True)

print("\nS13 boundaries:", len(s13))

assert len(s13) == 66, (
    f"Expected 66 S13 boundaries, found {len(s13)}"
)

# ------------------------------------------------------------
# 3. Load behavioral sequences
# ------------------------------------------------------------

mat = loadmat(
    MAT_FILE,
    squeeze_me=True,
    struct_as_record=False
)

sequences_raw = mat["sequences"]

sequences = []

for seq in sequences_raw:
    if isinstance(seq, bytes):
        text = seq.decode("utf-8", errors="ignore")
    else:
        text = str(seq)

    text = text.strip()

    sequences.append(text)

print("Behavioral sequences:", len(sequences))

assert len(sequences) == 66, (
    f"Expected 66 behavioral sequences, found {len(sequences)}"
)

# ------------------------------------------------------------
# 4. Convert EEG event codes to characters
# ------------------------------------------------------------

def event_to_character(event_name):
    """
    Convert printable ASCII BrainVision markers
    into their corresponding characters.

    Only ASCII codes 32–126 are treated as characters.
    S8 is handled separately as Backspace.
    """

    match = re.search(r"S\s*(\d+)", str(event_name))

    if not match:
        return None

    code = int(match.group(1))

    if 32 <= code <= 126:
        return chr(code)

    return None


# ------------------------------------------------------------
# 5. Build trial text
# ------------------------------------------------------------

trial_rows = []

for trial_idx in range(65):

    start_time = (
        0.0
        if trial_idx == 0
        else float(s13.loc[trial_idx - 1, "timestamp_seconds"])
    )

    end_time = float(
        s13.loc[trial_idx, "timestamp_seconds"]
    )

    # Events belonging to this trial
    trial_events = events_df[
        (events_df["timestamp_seconds"] > start_time) &
        (events_df["timestamp_seconds"] <= end_time)
    ].copy()

    typed_chars = []
    raw_chars = []

    for _, event in trial_events.iterrows():

        event_name = event["event"]

        # Backspace
        if event_name.replace(" ", "") == "S8":
            if typed_chars:
                typed_chars.pop()

            raw_chars.append("[BACKSPACE]")
            continue

        char = event_to_character(event_name)

        if char is not None:
            typed_chars.append(char)
            raw_chars.append(char)

    eeg_text = "".join(typed_chars)

    target_text = sequences[trial_idx]

    similarity = SequenceMatcher(
        None,
        eeg_text.upper(),
        target_text.upper()
    ).ratio()

    trial_rows.append({
        "trial_index": trial_idx,
        "start_time_seconds": start_time,
        "end_time_seconds": end_time,
        "duration_seconds": end_time - start_time,
        "target_text": target_text,
        "eeg_typed_text": eeg_text,
        "target_length": len(target_text),
        "eeg_text_length": len(eeg_text),
        "similarity": similarity,
        "exact_match": (
            eeg_text.upper() == target_text.upper()
        ),
        "event_count": len(trial_events)
    })


alignment_df = pd.DataFrame(trial_rows)

# ------------------------------------------------------------
# 6. Summary
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("ALIGNMENT SUMMARY")
print("=" * 80)

print(
    "\nTrials aligned:",
    len(alignment_df)
)

print(
    "Exact matches:",
    alignment_df["exact_match"].sum(),
    "/",
    len(alignment_df)
)

print(
    "Mean similarity:",
    round(
        alignment_df["similarity"].mean(),
        4
    )
)

print(
    "Median similarity:",
    round(
        alignment_df["similarity"].median(),
        4
    )
)

print(
    "Similarity >= 0.90:",
    (
        alignment_df["similarity"] >= 0.90
    ).sum(),
    "/",
    len(alignment_df)
)

# ------------------------------------------------------------
# 7. Display all trials
# ------------------------------------------------------------

display(
    alignment_df[
        [
            "trial_index",
            "start_time_seconds",
            "end_time_seconds",
            "target_text",
            "eeg_typed_text",
            "similarity",
            "exact_match"
        ]
    ]
)

# ------------------------------------------------------------
# 8. Identify potentially problematic trials
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("LOW-SIMILARITY TRIALS (< 0.90)")
print("=" * 80)

low_similarity = alignment_df[
    alignment_df["similarity"] < 0.90
]

if len(low_similarity) == 0:
    print("None")
else:
    display(
        low_similarity[
            [
                "trial_index",
                "target_text",
                "eeg_typed_text",
                "similarity"
            ]
        ]
    )

# ------------------------------------------------------------
# 9. Save alignment table
# ------------------------------------------------------------

output_file = os.path.join(
    PHASE3_DIR,
    "phase3_trial_alignment.csv"
)

alignment_df.to_csv(
    output_file,
    index=False
)

print("\nAlignment table saved:")
print(output_file)

print("\n" + "=" * 80)
print("PHASE 3 — STEP 2 COMPLETE")
print("=" * 80)
```

```python
import pandas as pd
import numpy as np

# ============================================================
# PHASE 3 — STEP 3
# CORRECT TRIAL INDEXING
# ============================================================

EEG_EVENTS_FILE = (
    "/content/drive/MyDrive/NeuroSentinel/"
    "04_phase3_event_alignment/phase3_raw_eeg_events.csv"
)

events_df = pd.read_csv(EEG_EVENTS_FILE)

events_df["event"] = (
    events_df["event"]
    .astype(str)
    .str.replace("Stimulus/", "", regex=False)
    .str.strip()
)

# ------------------------------------------------------------
# Extract S13 and control markers
# ------------------------------------------------------------

s13 = events_df[
    events_df["event"].str.replace(" ", "") == "S13"
].copy()

s13 = s13.sort_values(
    "timestamp_seconds"
).reset_index(drop=True)

print("=" * 80)
print("PHASE 3 — STEP 3: TRIAL INDEXING")
print("=" * 80)

print("\nNumber of S13 boundaries:", len(s13))

# ------------------------------------------------------------
# Show events surrounding the first S13
# ------------------------------------------------------------

first_s13_time = s13.loc[0, "timestamp_seconds"]

print("\nFirst S13 timestamp:", first_s13_time)

window = events_df[
    (
        events_df["timestamp_seconds"]
        >= first_s13_time - 15
    )
    &
    (
        events_df["timestamp_seconds"]
        <= first_s13_time + 2
    )
].copy()

print("\nEvents around first S13:")
display(
    window[
        [
            "sample_1000hz",
            "timestamp_seconds",
            "event_code",
            "event"
        ]
    ]
)

# ------------------------------------------------------------
# Show events between S13 boundaries
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("TRIAL BOUNDARY INTERVALS")
print("=" * 80)

interval_rows = []

for i in range(len(s13) - 1):

    start = float(
        s13.loc[i, "timestamp_seconds"]
    )

    end = float(
        s13.loc[i + 1, "timestamp_seconds"]
    )

    interval_events = events_df[
        (
            events_df["timestamp_seconds"] > start
        )
        &
        (
            events_df["timestamp_seconds"] <= end
        )
    ]

    printable_count = 0

    for event in interval_events["event"]:

        name = str(event)

        digits = "".join(
            c for c in name
            if c.isdigit()
        )

        if digits:
            code = int(digits)

            if 32 <= code <= 126:
                printable_count += 1

    interval_rows.append({
        "interval_index": i,
        "start_s13": start,
        "end_s13": end,
        "duration_seconds": end - start,
        "event_count": len(interval_events),
        "printable_events": printable_count
    })

interval_df = pd.DataFrame(interval_rows)

display(interval_df.head(15))

# ------------------------------------------------------------
# Statistics
# ------------------------------------------------------------

print("\nTotal S13-to-S13 intervals:", len(interval_df))

print(
    "Intervals containing printable events:",
    (
        interval_df["printable_events"] > 0
    ).sum()
)

print(
    "Intervals with zero printable events:",
    (
        interval_df["printable_events"] == 0
    ).sum()
)

print(
    "\nFirst 10 interval durations:"
)

print(
    interval_df[
        [
            "interval_index",
            "start_s13",
            "end_s13",
            "duration_seconds",
            "printable_events"
        ]
    ].head(10).to_string(index=False)
)

# ------------------------------------------------------------
# Save indexing evidence
# ------------------------------------------------------------

output_file = (
    "/content/drive/MyDrive/NeuroSentinel/"
    "04_phase3_event_alignment/"
    "phase3_trial_indexing_check.csv"
)

interval_df.to_csv(
    output_file,
    index=False
)

print("\nIndexing evidence saved:")
print(output_file)

print("\n" + "=" * 80)
print("PHASE 3 — STEP 3 COMPLETE")
print("=" * 80)
```

**Step 4: Correct 66-Trial Mapping**

```python
import os
import re
import pandas as pd
from scipy.io import loadmat
from difflib import SequenceMatcher

# ============================================================
# PHASE 3 — STEP 4
# CORRECT 66-TRIAL MAPPING
# ============================================================

EEG_EVENTS_FILE = (
    "/content/drive/MyDrive/NeuroSentinel/"
    "04_phase3_event_alignment/phase3_raw_eeg_events.csv"
)

MAT_FILE = (
    "/content/drive/MyDrive/NeuroSentinel/data/"
    "SpanishBCBL/EEG/logs/"
    "S10_session1_block1_list1.mat"
)

PHASE3_DIR = (
    "/content/drive/MyDrive/NeuroSentinel/"
    "04_phase3_event_alignment"
)

os.makedirs(PHASE3_DIR, exist_ok=True)

print("=" * 80)
print("PHASE 3 — STEP 4: CORRECT 66-TRIAL MAPPING")
print("=" * 80)

# ------------------------------------------------------------
# 1. Load EEG events
# ------------------------------------------------------------

events_df = pd.read_csv(EEG_EVENTS_FILE)

events_df["event"] = (
    events_df["event"]
    .astype(str)
    .str.replace("Stimulus/", "", regex=False)
    .str.strip()
)

events_df = events_df.sort_values(
    "timestamp_seconds"
).reset_index(drop=True)

# ------------------------------------------------------------
# 2. Extract S13 boundaries
# ------------------------------------------------------------

s13 = events_df[
    events_df["event"].str.replace(" ", "") == "S13"
].copy()

s13 = s13.sort_values(
    "timestamp_seconds"
).reset_index(drop=True)

assert len(s13) == 66, (
    f"Expected 66 S13 boundaries, found {len(s13)}"
)

# ------------------------------------------------------------
# 3. Find first S20
# ------------------------------------------------------------

s20 = events_df[
    events_df["event"].str.replace(" ", "") == "S20"
].copy()

assert len(s20) > 0, "No S20 marker found."

first_s20_time = float(
    s20.iloc[0]["timestamp_seconds"]
)

print("\nS13 boundaries:", len(s13))
print("First S20:", first_s20_time)
print("First S13:", s13.loc[0, "timestamp_seconds"])

# ------------------------------------------------------------
# 4. Load behavioral sequences
# ------------------------------------------------------------

mat = loadmat(
    MAT_FILE,
    squeeze_me=True,
    struct_as_record=False
)

sequences_raw = mat["sequences"]

sequences = []

for seq in sequences_raw:

    if isinstance(seq, bytes):
        text = seq.decode(
            "utf-8",
            errors="ignore"
        )
    else:
        text = str(seq)

    sequences.append(text.strip())

assert len(sequences) == 66, (
    f"Expected 66 sequences, found {len(sequences)}"
)

print("Behavioral sequences:", len(sequences))

# ------------------------------------------------------------
# 5. Event → character conversion
# ------------------------------------------------------------

def event_code(event_name):

    match = re.search(
        r"S\s*(\d+)",
        str(event_name)
    )

    if not match:
        return None

    return int(match.group(1))


def reconstruct_trial(events):

    typed_chars = []
    raw_events = []

    for event_name in events:

        code = event_code(event_name)

        if code is None:
            continue

        # Backspace
        if code == 8:

            raw_events.append("[BACKSPACE]")

            if typed_chars:
                typed_chars.pop()

            continue

        # Printable ASCII
        if 32 <= code <= 126:

            char = chr(code)

            typed_chars.append(char)
            raw_events.append(char)

    return (
        "".join(typed_chars),
        "".join(raw_events)
    )

# ------------------------------------------------------------
# 6. Construct the 66 trials
# ------------------------------------------------------------

trial_rows = []

for trial_idx in range(66):

    # Trial 0:
    # S20 → first S13
    if trial_idx == 0:

        start_time = first_s20_time
        end_time = float(
            s13.loc[0, "timestamp_seconds"]
        )

        boundary_type = "S20_to_S13"

    # Trials 1–65:
    # S13 → next S13
    else:

        start_time = float(
            s13.loc[
                trial_idx - 1,
                "timestamp_seconds"
            ]
        )

        end_time = float(
            s13.loc[
                trial_idx,
                "timestamp_seconds"
            ]
        )

        boundary_type = "S13_to_S13"

    # Events inside trial interval
    trial_events = events_df[
        (
            events_df["timestamp_seconds"]
            > start_time
        )
        &
        (
            events_df["timestamp_seconds"]
            <= end_time
        )
    ]

    eeg_text, raw_event_text = reconstruct_trial(
        trial_events["event"]
    )

    target_text = sequences[trial_idx]

    similarity = SequenceMatcher(
        None,
        eeg_text.upper(),
        target_text.upper()
    ).ratio()

    trial_rows.append({

        "trial_index": trial_idx,

        "boundary_type": boundary_type,

        "start_time_seconds": start_time,

        "end_time_seconds": end_time,

        "duration_seconds":
            end_time - start_time,

        "target_text":
            target_text,

        "eeg_typed_text":
            eeg_text,

        "raw_event_text":
            raw_event_text,

        "target_length":
            len(target_text),

        "eeg_text_length":
            len(eeg_text),

        "similarity":
            similarity,

        "exact_match":
            eeg_text.upper() ==
            target_text.upper(),

        "event_count":
            len(trial_events)
    })


alignment_df = pd.DataFrame(trial_rows)

# ------------------------------------------------------------
# 7. Validation
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("CORRECTED ALIGNMENT SUMMARY")
print("=" * 80)

print(
    "\nTrials:",
    len(alignment_df)
)

print(
    "Expected:",
    66
)

print(
    "Exact matches:",
    alignment_df["exact_match"].sum(),
    "/ 66"
)

print(
    "Mean similarity:",
    round(
        alignment_df["similarity"].mean(),
        4
    )
)

print(
    "Median similarity:",
    round(
        alignment_df["similarity"].median(),
        4
    )
)

print(
    "Similarity >= 0.90:",
    (
        alignment_df["similarity"] >= 0.90
    ).sum(),
    "/ 66"
)

# ------------------------------------------------------------
# 8. Validate first trial
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("TRIAL 0 CHECK")
print("=" * 80)

display(
    alignment_df.iloc[
        0:1
    ][
        [
            "trial_index",
            "boundary_type",
            "start_time_seconds",
            "end_time_seconds",
            "target_text",
            "eeg_typed_text",
            "similarity",
            "exact_match"
        ]
    ]
)

# ------------------------------------------------------------
# 9. Display all low-similarity trials
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("LOW-SIMILARITY TRIALS (< 0.90)")
print("=" * 80)

low_similarity = alignment_df[
    alignment_df["similarity"] < 0.90
]

print(
    "Count:",
    len(low_similarity)
)

if len(low_similarity) > 0:

    display(
        low_similarity[
            [
                "trial_index",
                "target_text",
                "eeg_typed_text",
                "similarity"
            ]
        ]
    )

# ------------------------------------------------------------
# 10. Save corrected alignment
# ------------------------------------------------------------

output_file = os.path.join(
    PHASE3_DIR,
    "phase3_corrected_trial_alignment.csv"
)

alignment_df.to_csv(
    output_file,
    index=False
)

print("\nCorrected alignment saved:")
print(output_file)

# ------------------------------------------------------------
# 11. Final structural assertions
# ------------------------------------------------------------

assert len(alignment_df) == 66

assert (
    alignment_df.iloc[0]["boundary_type"]
    == "S20_to_S13"
)

assert (
    alignment_df.iloc[1]["boundary_type"]
    == "S13_to_S13"
)

print("\n✓ 66 trials constructed")
print("✓ Trial 0 uses S20 → S13")
print("✓ Trials 1–65 use S13 → S13")
print("✓ Behavioral sequence count = 66")
print("✓ Corrected alignment saved")

print("\n" + "=" * 80)
print("PHASE 3 — STEP 4 COMPLETE")
print("=" * 80)
```

**Step 5: Keystroke-Level Alignment**

```python
import os
import re
import pandas as pd

# ============================================================
# PHASE 3 — STEP 5
# KEYSTROKE-LEVEL ALIGNMENT
# ============================================================

EVENTS_FILE = (
    "/content/drive/MyDrive/NeuroSentinel/"
    "04_phase3_event_alignment/phase3_raw_eeg_events.csv"
)

TRIAL_FILE = (
    "/content/drive/MyDrive/NeuroSentinel/"
    "04_phase3_event_alignment/"
    "phase3_corrected_trial_alignment.csv"
)

OUTPUT_DIR = (
    "/content/drive/MyDrive/NeuroSentinel/"
    "04_phase3_event_alignment"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

print("=" * 80)
print("PHASE 3 — STEP 5: KEYSTROKE-LEVEL ALIGNMENT")
print("=" * 80)

# ------------------------------------------------------------
# Load data
# ------------------------------------------------------------

events_df = pd.read_csv(EVENTS_FILE)
trials_df = pd.read_csv(TRIAL_FILE)

events_df["event"] = (
    events_df["event"]
    .astype(str)
    .str.replace("Stimulus/", "", regex=False)
    .str.strip()
)

events_df = events_df.sort_values(
    "timestamp_seconds"
).reset_index(drop=True)

# ------------------------------------------------------------
# Extract S13 boundaries
# ------------------------------------------------------------

s13 = events_df[
    events_df["event"].str.replace(" ", "") == "S13"
].copy()

s13 = s13.sort_values(
    "timestamp_seconds"
).reset_index(drop=True)

assert len(s13) == 66

# First trial starts at S20
s20 = events_df[
    events_df["event"].str.replace(" ", "") == "S20"
].copy()

assert len(s20) > 0

first_trial_start = float(
    s20.iloc[0]["timestamp_seconds"]
)

# ------------------------------------------------------------
# Convert event to ASCII code
# ------------------------------------------------------------

def get_code(event_name):

    match = re.search(
        r"S\s*(\d+)",
        str(event_name)
    )

    if not match:
        return None

    return int(match.group(1))


# ------------------------------------------------------------
# Build keystroke table
# ------------------------------------------------------------

keystroke_rows = []

for trial_idx in range(66):

    # Trial boundaries
    if trial_idx == 0:
        start_time = first_trial_start
    else:
        start_time = float(
            s13.loc[
                trial_idx - 1,
                "timestamp_seconds"
            ]
        )

    end_time = float(
        s13.loc[
            trial_idx,
            "timestamp_seconds"
        ]
    )

    # Events inside this trial
    trial_events = events_df[
        (
            events_df["timestamp_seconds"] > start_time
        )
        &
        (
            events_df["timestamp_seconds"] <= end_time
        )
    ].copy()

    key_index = 0

    for _, event in trial_events.iterrows():

        code = get_code(event["event"])

        if code is None:
            continue

        # Printable ASCII = keyboard character
        if 32 <= code <= 126:

            keystroke_rows.append({
                "trial_index": trial_idx,
                "keystroke_index": key_index,
                "timestamp_seconds":
                    float(event["timestamp_seconds"]),
                "sample_1000hz":
                    int(event["sample_1000hz"]),
                "event_code": code,
                "character": chr(code),
                "is_backspace": False
            })

            key_index += 1

        # Backspace
        elif code == 8:

            keystroke_rows.append({
                "trial_index": trial_idx,
                "keystroke_index": key_index,
                "timestamp_seconds":
                    float(event["timestamp_seconds"]),
                "sample_1000hz":
                    int(event["sample_1000hz"]),
                "event_code": code,
                "character": "",
                "is_backspace": True
            })

            key_index += 1


keystrokes_df = pd.DataFrame(
    keystroke_rows
)

# ------------------------------------------------------------
# Basic validation
# ------------------------------------------------------------

print("\nTotal keystroke events:",
      len(keystrokes_df))

print(
    "Trials containing keystrokes:",
    keystrokes_df["trial_index"].nunique()
)

print(
    "Printable characters:",
    (~keystrokes_df["is_backspace"]).sum()
)

print(
    "Backspaces:",
    keystrokes_df["is_backspace"].sum()
)

# ------------------------------------------------------------
# Check timestamp ordering
# ------------------------------------------------------------

timestamp_order_ok = True

for trial_idx, group in keystrokes_df.groupby(
    "trial_index"
):

    timestamps = group[
        "timestamp_seconds"
    ].values

    if not all(
        timestamps[i] <= timestamps[i + 1]
        for i in range(len(timestamps) - 1)
    ):
        timestamp_order_ok = False
        print(
            "Timestamp ordering problem in trial:",
            trial_idx
        )

print(
    "\nTimestamp ordering:",
    timestamp_order_ok
)

assert timestamp_order_ok

# ------------------------------------------------------------
# Check trial boundaries
# ------------------------------------------------------------

boundary_errors = []

for _, row in keystrokes_df.iterrows():

    trial_idx = int(row["trial_index"])

    timestamp = row["timestamp_seconds"]

    if trial_idx == 0:
        start = first_trial_start
    else:
        start = float(
            s13.loc[
                trial_idx - 1,
                "timestamp_seconds"
            ]
        )

    end = float(
        s13.loc[
            trial_idx,
            "timestamp_seconds"
        ]
    )

    if not (
        start < timestamp <= end
    ):
        boundary_errors.append(
            trial_idx
        )

print(
    "Boundary errors:",
    len(boundary_errors)
)

assert len(boundary_errors) == 0

# ------------------------------------------------------------
# Display first 50 keystrokes
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("FIRST KEYSTROKE EVENTS")
print("=" * 80)

display(
    keystrokes_df.head(50)
)

# ------------------------------------------------------------
# Trial-level counts
# ------------------------------------------------------------

trial_counts = (
    keystrokes_df
    .groupby("trial_index")
    .agg(
        total_events=("trial_index", "size"),
        backspaces=("is_backspace", "sum"),
        printable_characters=(
            "is_backspace",
            lambda x: (~x).sum()
        )
    )
    .reset_index()
)

print("\n" + "=" * 80)
print("KEYSTROKE COUNTS PER TRIAL")
print("=" * 80)

display(
    trial_counts.head(20)
)

# ------------------------------------------------------------
# Save
# ------------------------------------------------------------

output_file = os.path.join(
    OUTPUT_DIR,
    "phase3_keystroke_alignment.csv"
)

keystrokes_df.to_csv(
    output_file,
    index=False
)

counts_file = os.path.join(
    OUTPUT_DIR,
    "phase3_keystroke_trial_counts.csv"
)

trial_counts.to_csv(
    counts_file,
    index=False
)

print("\nKeystroke table saved:")
print(output_file)

print("\nTrial counts saved:")
print(counts_file)

print("\n" + "=" * 80)
print("PHASE 3 — STEP 5 COMPLETE")
print("=" * 80)
```

**Step 6: Investigate the Missing Trial**

```python
import pandas as pd
import os

# ============================================================
# PHASE 3 — STEP 6
# INVESTIGATE MISSING KEYSTROKE TRIAL
# ============================================================

EVENTS_FILE = (
    "/content/drive/MyDrive/NeuroSentinel/"
    "04_phase3_event_alignment/phase3_raw_eeg_events.csv"
)

KEYSTROKE_FILE = (
    "/content/drive/MyDrive/NeuroSentinel/"
    "04_phase3_event_alignment/"
    "phase3_keystroke_alignment.csv"
)

TRIAL_FILE = (
    "/content/drive/MyDrive/NeuroSentinel/"
    "04_phase3_event_alignment/"
    "phase3_corrected_trial_alignment.csv"
)

events_df = pd.read_csv(EVENTS_FILE)
keys_df = pd.read_csv(KEYSTROKE_FILE)
trials_df = pd.read_csv(TRIAL_FILE)

events_df["event"] = (
    events_df["event"]
    .astype(str)
    .str.replace("Stimulus/", "", regex=False)
    .str.strip()
)

# ------------------------------------------------------------
# Find trials with no keystrokes
# ------------------------------------------------------------

all_trials = set(range(66))
key_trials = set(keys_df["trial_index"].unique())

missing_trials = sorted(
    all_trials - key_trials
)

print("=" * 80)
print("MISSING KEYSTROKE TRIALS")
print("=" * 80)

print("Missing trials:", missing_trials)

# ------------------------------------------------------------
# Inspect each missing trial
# ------------------------------------------------------------

for trial_idx in missing_trials:

    row = trials_df[
        trials_df["trial_index"] == trial_idx
    ].iloc[0]

    start = row["start_time_seconds"]
    end = row["end_time_seconds"]

    print("\n" + "-" * 80)
    print(f"TRIAL {trial_idx}")
    print("-" * 80)

    print("Boundary type:", row["boundary_type"])
    print("Start:", start)
    print("End:", end)
    print("Duration:", row["duration_seconds"])

    print("\nTarget:")
    print(row["target_text"])

    print("\nPreviously reconstructed EEG text:")
    print(repr(row["eeg_typed_text"]))

    # Show ALL EEG events inside this trial
    trial_events = events_df[
        (
            events_df["timestamp_seconds"] > start
        )
        &
        (
            events_df["timestamp_seconds"] <= end
        )
    ].copy()

    print("\nEvents inside trial:", len(trial_events))

    display(
        trial_events[
            [
                "sample_1000hz",
                "timestamp_seconds",
                "event_code",
                "event"
            ]
        ]
    )

# ------------------------------------------------------------
# Check neighboring trials
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("NEIGHBORING TRIALS")
print("=" * 80)

for trial_idx in missing_trials:

    start_idx = max(0, trial_idx - 2)
    end_idx = min(65, trial_idx + 2)

    display(
        trials_df[
            trials_df["trial_index"].between(
                start_idx,
                end_idx
            )
        ][
            [
                "trial_index",
                "boundary_type",
                "start_time_seconds",
                "end_time_seconds",
                "target_text",
                "eeg_typed_text",
                "similarity"
            ]
        ]
    )

print("\n" + "=" * 80)
print("STEP 6 DIAGNOSTIC COMPLETE")
print("=" * 80)
```

```python
import os
import json
import pandas as pd
from datetime import datetime

# ============================================================
# PHASE 3 — FINAL VALIDATION & CHECKPOINT
# ============================================================

PROJECT_DIR = "/content/drive/MyDrive/NeuroSentinel"

PHASE3_DIR = os.path.join(
    PROJECT_DIR,
    "04_phase3_event_alignment"
)

CHECKPOINT_DIR = os.path.join(
    PROJECT_DIR,
    "checkpoints"
)

os.makedirs(PHASE3_DIR, exist_ok=True)
os.makedirs(CHECKPOINT_DIR, exist_ok=True)

TRIAL_FILE = os.path.join(
    PHASE3_DIR,
    "phase3_corrected_trial_alignment.csv"
)

KEYSTROKE_FILE = os.path.join(
    PHASE3_DIR,
    "phase3_keystroke_alignment.csv"
)

# ------------------------------------------------------------
# Load final Phase 3 outputs
# ------------------------------------------------------------

trials = pd.read_csv(TRIAL_FILE)
keystrokes = pd.read_csv(KEYSTROKE_FILE)

# ------------------------------------------------------------
# Structural validation
# ------------------------------------------------------------

assert len(trials) == 66
assert trials["trial_index"].tolist() == list(range(66))

assert trials.loc[0, "boundary_type"] == "S20_to_S13"

assert all(
    trials.loc[i, "boundary_type"] == "S13_to_S13"
    for i in range(1, 66)
)

assert len(keystrokes) == 1965

assert keystrokes["timestamp_seconds"].is_monotonic_increasing

assert keystrokes["trial_index"].min() >= 0
assert keystrokes["trial_index"].max() <= 65

# ------------------------------------------------------------
# Identify keystroke-bearing trials
# ------------------------------------------------------------

keystroke_trials = sorted(
    keystrokes["trial_index"].unique()
)

missing_trials = sorted(
    set(range(66)) - set(keystroke_trials)
)

assert missing_trials == [3]

# ------------------------------------------------------------
# Final metrics
# ------------------------------------------------------------

exact_matches = int(
    trials["exact_match"].sum()
)

mean_similarity = float(
    trials["similarity"].mean()
)

median_similarity = float(
    trials["similarity"].median()
)

high_similarity = int(
    (trials["similarity"] >= 0.90).sum()
)

print("=" * 80)
print("PHASE 3 — FINAL VALIDATION")
print("=" * 80)

print("\nTrial count:", len(trials))
print("Behavioral sequences:", 66)
print("Keystroke events:", len(keystrokes))
print("Printable characters:", 1926)
print("Backspaces:", 39)

print("\nExact trial-text matches:",
      exact_matches, "/ 66")

print("Mean similarity:",
      round(mean_similarity, 4))

print("Median similarity:",
      round(median_similarity, 4))

print("Similarity >= 0.90:",
      high_similarity, "/ 66")

print("\nTrials with keystrokes:",
      len(keystroke_trials), "/ 66")

print("Missing keystroke trial:",
      missing_trials)

# ------------------------------------------------------------
# Check timestamp validity
# ------------------------------------------------------------

boundary_errors = 0

for _, row in keystrokes.iterrows():

    trial = int(row["trial_index"])
    timestamp = float(row["timestamp_seconds"])

    start = float(
        trials.loc[
            trials["trial_index"] == trial,
            "start_time_seconds"
        ].iloc[0]
    )

    end = float(
        trials.loc[
            trials["trial_index"] == trial,
            "end_time_seconds"
        ].iloc[0]
    )

    if not (start < timestamp <= end):
        boundary_errors += 1

assert boundary_errors == 0

print("\nKeystroke boundary errors:",
      boundary_errors)

# ------------------------------------------------------------
# Create checkpoint
# ------------------------------------------------------------

checkpoint = {
    "project": "NeuroSentinel",
    "phase": 3,
    "phase_name": "Event Alignment",
    "status": "COMPLETE",

    "trial_alignment": {
        "behavioral_sequences": 66,
        "eeg_trial_boundaries": 66,
        "final_trials": 66,
        "trial_0_boundary": "S20_to_S13",
        "remaining_boundaries": "S13_to_S13"
    },

    "keystroke_alignment": {
        "total_events": 1965,
        "printable_characters": 1926,
        "backspaces": 39,
        "trials_with_keystrokes": 65,
        "missing_keystroke_trials": [3],
        "timestamp_order_verified": True,
        "boundary_errors": 0
    },

    "text_alignment": {
        "exact_matches": exact_matches,
        "mean_similarity": mean_similarity,
        "median_similarity": median_similarity,
        "similarity_ge_0_90": high_similarity
    },

    "known_data_condition": (
        "Trial 3 contains no printable or backspace EEG "
        "keyboard markers. It is retained as a valid trial "
        "with unavailable keystroke-level data. No timestamps "
        "or characters were fabricated."
    ),

    "outputs": [
        "phase3_raw_eeg_events.csv",
        "phase3_trial_indexing_check.csv",
        "phase3_corrected_trial_alignment.csv",
        "phase3_keystroke_alignment.csv",
        "phase3_keystroke_trial_counts.csv"
    ],

    "next_phase": "Phase 4 — Windowing",

    "checkpoint_created": datetime.now().isoformat()
}

checkpoint_file = os.path.join(
    CHECKPOINT_DIR,
    "phase3_checkpoint.json"
)

with open(checkpoint_file, "w") as f:
    json.dump(checkpoint, f, indent=4)

print("\n" + "=" * 80)
print("PHASE 3 CHECKPOINT SAVED")
print("=" * 80)

print(checkpoint_file)

print("\n" + "=" * 80)
print("PHASE 3 — COMPLETE")
print("=" * 80)
```

```python
from pathlib import Path
import pandas as pd
import numpy as np
import mne

# ============================================================
# PHASE 4 — STEP 2: EEG WINDOW EXTRACTION
# ============================================================

PROJECT_DIR = Path("/content/drive/MyDrive/NeuroSentinel")

PHASE2_FILE = (
    PROJECT_DIR
    / "03_phase2_preprocessing"
    / "010_DECOMEG_S1_noid_task1_phase2_raw.fif"
)

PHASE3_FILE = (
    PROJECT_DIR
    / "04_phase3_event_alignment"
    / "phase3_keystroke_alignment.csv"
)

PHASE4_DIR = PROJECT_DIR / "05_phase4_windowing"
PHASE4_DIR.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# Parameters
# ------------------------------------------------------------

SFREQ = 50.0

WINDOW_START = -0.2
WINDOW_END = 0.3

N_SAMPLES = int(round((WINDOW_END - WINDOW_START) * SFREQ))

assert N_SAMPLES == 25

# ------------------------------------------------------------
# Load preprocessed EEG
# ------------------------------------------------------------

print("Loading Phase 2 EEG...")

raw = mne.io.read_raw_fif(
    str(PHASE2_FILE),
    preload=True,
    verbose=False
)

print("EEG shape:", raw.get_data().shape)
print("Sampling rate:", raw.info["sfreq"])

# ------------------------------------------------------------
# Load Phase 3 keystrokes
# ------------------------------------------------------------

print("\nLoading Phase 3 keystrokes...")

keystrokes = pd.read_csv(PHASE3_FILE)

print("Keystroke rows:", len(keystrokes))
print("Columns:")
print(keystrokes.columns.tolist())

# ------------------------------------------------------------
# Identify timestamp column
# ------------------------------------------------------------

timestamp_candidates = [
    "timestamp_seconds",
    "timestamp",
    "time_seconds"
]

timestamp_col = next(
    (c for c in timestamp_candidates if c in keystrokes.columns),
    None
)

if timestamp_col is None:
    raise ValueError(
        f"No timestamp column found. Available columns: "
        f"{keystrokes.columns.tolist()}"
    )

print("\nUsing timestamp column:", timestamp_col)

# ------------------------------------------------------------
# Extract windows
# ------------------------------------------------------------

eeg_data = raw.get_data()

n_channels, n_total_samples = eeg_data.shape

windows = []
metadata = []

skipped = 0

for idx, row in keystrokes.iterrows():

    timestamp = float(row[timestamp_col])

    # Convert event time to 50-Hz sample
    center_sample = int(round(timestamp * SFREQ))

    # Window boundaries
    start_sample = center_sample + int(round(WINDOW_START * SFREQ))
    end_sample = center_sample + int(round(WINDOW_END * SFREQ))

    # Boundary check
    if start_sample < 0 or end_sample > n_total_samples:
        skipped += 1
        continue

    window = eeg_data[:, start_sample:end_sample]

    # Shape validation
    if window.shape != (n_channels, N_SAMPLES):
        skipped += 1
        continue

    windows.append(window.astype(np.float32))

    metadata.append({
        "window_id": len(windows) - 1,
        "trial_index": row.get("trial_index", np.nan),
        "keystroke_index": row.get("keystroke_index", np.nan),
        "timestamp_seconds": timestamp,
        "center_sample_50hz": center_sample,
        "start_sample_50hz": start_sample,
        "end_sample_50hz": end_sample,
        "window_start_seconds": timestamp + WINDOW_START,
        "window_end_seconds": timestamp + WINDOW_END,
        "character": row.get("character", np.nan),
        "is_backspace": row.get("is_backspace", False),
    })

# ------------------------------------------------------------
# Convert to NumPy array
# ------------------------------------------------------------

windows = np.stack(windows)

metadata = pd.DataFrame(metadata)

# ------------------------------------------------------------
# Final validation
# ------------------------------------------------------------

print("\n========== PHASE 4 WINDOWING ==========")
print("Input EEG channels :", n_channels)
print("Sampling rate      :", SFREQ)
print("Window duration    :", WINDOW_END - WINDOW_START)
print("Samples per window :", N_SAMPLES)
print("Total input events :", len(keystrokes))
print("Windows created    :", len(windows))
print("Windows skipped    :", skipped)
print("Window shape       :", windows.shape)
print("========================================")

assert n_channels == 61
assert SFREQ == 50
assert windows.shape[1] == 61
assert windows.shape[2] == 25
assert not np.isnan(windows).any()
assert not np.isinf(windows).any()

# ------------------------------------------------------------
# Save outputs
# ------------------------------------------------------------

WINDOW_FILE = PHASE4_DIR / "phase4_eeg_windows.npy"
METADATA_FILE = PHASE4_DIR / "phase4_window_metadata.csv"

np.save(WINDOW_FILE, windows)
metadata.to_csv(METADATA_FILE, index=False)

print("\nSaved:")
print(WINDOW_FILE)
print(METADATA_FILE)
```

```python
import torch
from neuraltrain.models.simpleconv import SimpleConvTimeAgg

print("========== PHASE 5 — EEG DECODER SMOKE TEST ==========")

# -------------------------------------------------------
# Configuration
# -------------------------------------------------------
N_CHANNELS = 61
N_SAMPLES = 25
N_CLASSES = 29

config = SimpleConvTimeAgg(
    time_agg_out="att",
    dropout_input=0.2,
    conv_dropout=0.5,
    hidden=2048,
    batch_norm=True,
    depth=8,
    dilation_period=3,
    kernel_size=3,
    relu_leakiness=0.01,
    initial_linear=512,
    gelu=True,
    skip=True,
    scale=0.1,
    merger_config=None,          # IMPORTANT: EEG does not use the MEG merger
    subject_layers_config=None,
)

model = config.build(
    n_in_channels=N_CHANNELS,
    n_outputs=N_CLASSES,
)

model.eval()

# -------------------------------------------------------
# Load one REAL Phase 4 EEG window
# -------------------------------------------------------
windows = np.load(
    PROJECT_DIR
    / "05_phase4_windowing"
    / "phase4_eeg_windows.npy"
)

print("Phase 4 windows:", windows.shape)

x = torch.tensor(
    windows[0:1],
    dtype=torch.float32
)

print("Input tensor:", x.shape)

# -------------------------------------------------------
# Forward pass
# -------------------------------------------------------
with torch.no_grad():
    logits = model(x)

print("Output logits:", logits.shape)

# -------------------------------------------------------
# Convert to probabilities
# -------------------------------------------------------
probabilities = torch.softmax(logits, dim=-1)

prediction_index = int(torch.argmax(probabilities, dim=-1)[0])
confidence = float(torch.max(probabilities, dim=-1).values[0])

print()
print("========== SMOKE TEST RESULT ==========")
print("Input shape      :", tuple(x.shape))
print("Logits shape     :", tuple(logits.shape))
print("Prediction index :", prediction_index)
print("Confidence       :", confidence)
print("========================================")
```

```python
import pandas as pd
from pathlib import Path

# -------------------------------------------------------
# Paths
# -------------------------------------------------------
KEYSTROKE_FILE = (
    PROJECT_DIR
    / "04_phase3_event_alignment"
    / "phase3_keystroke_alignment.csv"
)

OUTPUT_DIR = PROJECT_DIR / "06_phase5_utility_labels"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# -------------------------------------------------------
# Load Phase 3 keystrokes
# -------------------------------------------------------
df = pd.read_csv(KEYSTROKE_FILE)

print("========== PHASE 5 — LABEL VALIDATION ==========")
print("Rows:", len(df))
print("Columns:", list(df.columns))

# -------------------------------------------------------
# Brain2Qwerty V1 mapping
# -------------------------------------------------------
BUTTON_MAPPING = {
    "s": 0, "o": 1, "t": 2, "e": 3, "n": 4,
    "c": 5, "i": 6, "a": 7, " ": 8,
    "d": 9, "l": 10, "r": 11, "b": 12,
    "<special>": 13,
    "z": 14, "v": 15, "f": 16, "m": 17,
    "u": 18, "h": 19, "p": 20, "g": 21,
    "q": 22, "w": 23, "x": 24, "y": 25,
    "j": 26, "k": 27,
    "<number>": 28,
}

CHAR_INDEX = {v: k for k, v in BUTTON_MAPPING.items()}

# -------------------------------------------------------
# Normalize characters
# -------------------------------------------------------
def normalize_label(row):
    char = str(row["character"])

    if bool(row["is_backspace"]):
        return None

    if len(char) != 1:
        return None

    char = char.lower()

    if char in BUTTON_MAPPING:
        return BUTTON_MAPPING[char]

    if char.isdigit():
        return BUTTON_MAPPING["<number>"]

    # Characters unsupported by the 29-class decoder
    return None


df["label"] = df.apply(normalize_label, axis=1)

# -------------------------------------------------------
# Validation
# -------------------------------------------------------
valid = df["label"].notna()

print()
print("Total events       :", len(df))
print("Valid decoder rows :", valid.sum())
print("Excluded rows      :", (~valid).sum())

print()
print("========== EXCLUSION BREAKDOWN ==========")

print(
    df.loc[~valid, ["character", "is_backspace"]]
      .value_counts()
      .head(20)
)

print()
print("========== LABEL DISTRIBUTION ==========")

print(
    df.loc[valid, "label"]
      .value_counts()
      .sort_index()
)

# -------------------------------------------------------
# Save validated labels
# -------------------------------------------------------
label_df = df.loc[
    valid,
    [
        "trial_index",
        "keystroke_index",
        "timestamp_seconds",
        "sample_1000hz",
        "event_code",
        "character",
        "is_backspace",
        "label",
    ],
].reset_index(drop=True)

output_file = OUTPUT_DIR / "phase5_decoder_labels.csv"

label_df.to_csv(output_file, index=False)

print()
print("========== PHASE 5 LABEL RESULT ==========")
print("Valid samples :", len(label_df))
print("Classes used  :", label_df["label"].nunique())
print("Saved         :", output_file)
print("==========================================")

# -------------------------------------------------------
# Safety assertions
# -------------------------------------------------------
assert label_df["label"].between(0, 28).all()
assert not label_df["label"].isna().any()

print("\nLABEL VALIDATION: PASS")
```

```python
# ============================================================
# PHASE 5 — EEG / LABEL ALIGNMENT VALIDATION
# ============================================================

WINDOW_FILE = (
    PROJECT_DIR
    / "05_phase4_windowing"
    / "phase4_window_metadata.csv"
)

LABEL_FILE = (
    PROJECT_DIR
    / "06_phase5_utility_labels"
    / "phase5_decoder_labels.csv"
)

windows_meta = pd.read_csv(WINDOW_FILE)
labels = pd.read_csv(LABEL_FILE)

print("========== EEG / LABEL ALIGNMENT ==========")
print("Phase 4 windows :", len(windows_meta))
print("Phase 5 labels  :", len(labels))

# Remove backspace windows using the same rule as labels
usable_windows = windows_meta[
    ~windows_meta["is_backspace"].astype(bool)
].reset_index(drop=True)

labels = labels.reset_index(drop=True)

print("Usable windows  :", len(usable_windows))

# ------------------------------------------------------------
# Basic size check
# ------------------------------------------------------------
assert len(usable_windows) == len(labels)

# ------------------------------------------------------------
# Timestamp check
# ------------------------------------------------------------
timestamp_diff = (
    usable_windows["timestamp_seconds"].to_numpy()
    - labels["timestamp_seconds"].to_numpy()
)

max_timestamp_error = abs(timestamp_diff).max()

print("Maximum timestamp difference:", max_timestamp_error)

assert max_timestamp_error < 1e-9

# ------------------------------------------------------------
# Trial / keystroke alignment
# ------------------------------------------------------------
trial_match = (
    usable_windows["trial_index"].to_numpy()
    == labels["trial_index"].to_numpy()
)

key_match = (
    usable_windows["keystroke_index"].to_numpy()
    == labels["keystroke_index"].to_numpy()
)

print("Trial indices match     :", trial_match.all())
print("Keystroke indices match:", key_match.all())

assert trial_match.all()
assert key_match.all()

# ------------------------------------------------------------
# EEG window ↔ label mapping
# ------------------------------------------------------------
windows = np.load(
    PROJECT_DIR
    / "05_phase4_windowing"
    / "phase4_eeg_windows.npy"
)

# Original indices of usable windows
usable_indices = windows_meta.index[
    ~windows_meta["is_backspace"].astype(bool)
].to_numpy()

X = windows[usable_indices]
y = labels["label"].astype(int).to_numpy()

print()
print("========== FINAL TRAINING DATA ==========")
print("X shape :", X.shape)
print("y shape :", y.shape)
print("Classes :", np.unique(y))
print("==========================================")

assert X.shape[0] == y.shape[0]
assert X.shape[1:] == (61, 25)
assert y.min() >= 0
assert y.max() <= 28

print("\nEEG / LABEL ALIGNMENT: PASS")
```

```python
# ============================================================
# PHASE 6 — CORRECT PARTICIPANT INVENTORY
# ============================================================

import re
import pandas as pd

records = []

for path in sorted(EEG_DIR.glob("*.vhdr")):

    # Participant ID = numeric prefix
    participant_match = re.match(r"^(\d+)_", path.name)

    # Session = S1 / S2
    session_match = re.search(r"_(S[12])_", path.name, re.IGNORECASE)

    # Task
    task_match = re.search(
        r"_(task1bis|task1|task2|task|tapping)(?:_|\.|$)",
        path.name,
        re.IGNORECASE
    )

    participant = (
        participant_match.group(1)
        if participant_match else "UNKNOWN"
    )

    session = (
        session_match.group(1).upper()
        if session_match else "UNKNOWN"
    )

    task = (
        task_match.group(1).lower()
        if task_match else "UNKNOWN"
    )

    records.append({
        "participant": participant,
        "session": session,
        "task": task,
        "file": path.name,
    })

participant_df = pd.DataFrame(records)

print("========== CORRECT PARTICIPANT INVENTORY ==========")

print("Total EEG recordings :", len(participant_df))
print(
    "Unique participants  :",
    participant_df["participant"].nunique()
)

print()
print("Participants:")
print(
    sorted(participant_df["participant"].unique())
)

print()
print("========== RECORDINGS PER PARTICIPANT ==========")

summary = (
    participant_df
    .groupby("participant")
    .size()
    .rename("recordings")
    .reset_index()
)

print(summary.to_string(index=False))

print()
print("========== SESSION / TASK COUNTS ==========")

print(
    participant_df
    .groupby(["session", "task"])
    .size()
    .unstack(fill_value=0)
)

print("================================================")
```

```python
# ============================================================
# PHASE 5 — SINGLE RECORDING END-TO-END TEST
# ============================================================

from pathlib import Path
import numpy as np
import pandas as pd
import mne

# ------------------------------------------------------------
# 1. Load Phase 5 manifest
# ------------------------------------------------------------

MANIFEST_PATH = (
    PROJECT_DIR
    / "06_phase5_utility_labels"
    / "phase5_official_recording_manifest.csv"
)

manifest = pd.read_csv(MANIFEST_PATH)

row = manifest.iloc[0]

print("========== RECORDING ==========")
print("Recording ID:", row["recording_id"])
print("Subject     :", row["subject"])
print("Session     :", row["session"])
print("Task        :", row["task"])
print("EEG         :", Path(row["raw"]).name)
print("Log         :", Path(row["log"]).name)
print("================================")


# ------------------------------------------------------------
# 2. Load EEG
# ------------------------------------------------------------

raw = mne.io.read_raw_brainvision(
    row["raw"],
    preload=True,
    verbose=False
)

print("\n========== RAW EEG ==========")
print("Channels       :", len(raw.ch_names))
print("Sampling rate  :", raw.info["sfreq"])
print("Duration (sec) :", raw.times[-1])
print("Annotations    :", len(raw.annotations))
print("=============================")


# ------------------------------------------------------------
# 3. Identify and remove EOG channels
# ------------------------------------------------------------

EOG_CHANNELS = [
    ch for ch in raw.ch_names
    if "EOG" in ch.upper()
]

print("\nEOG channels:", EOG_CHANNELS)

raw_eeg = raw.copy().pick(
    [
        ch for ch in raw.ch_names
        if ch not in EOG_CHANNELS
    ]
)

print("EEG channels after EOG removal:", len(raw_eeg.ch_names))


# ------------------------------------------------------------
# 4. Montage
# ------------------------------------------------------------

raw_eeg.set_montage(
    "standard_1005",
    on_missing="ignore"
)


# ------------------------------------------------------------
# 5. Band-pass filter
# ------------------------------------------------------------

raw_eeg.filter(
    l_freq=0.1,
    h_freq=20.0,
    verbose=False
)


# ------------------------------------------------------------
# 6. Resample to 50 Hz
# ------------------------------------------------------------

raw_eeg.resample(
    50,
    npad="auto",
    verbose=False
)

print("\n========== PREPROCESSED EEG ==========")
print("Channels      :", len(raw_eeg.ch_names))
print("Sampling rate :", raw_eeg.info["sfreq"])
print("Samples       :", raw_eeg.n_times)
print("Duration      :", raw_eeg.times[-1])
print("=======================================")


# ------------------------------------------------------------
# 7. Read official behavioral log
# ------------------------------------------------------------

from studies.spanishbcbl import Pinet2024Eeg

study = Pinet2024Eeg(path=DATA_DIR / "SpanishBCBL")

events = study._read_log(
    Path(row["log"])
)

print("\n========== BEHAVIORAL LOG ==========")
print("Total events:", len(events))
print()
print(events.head())
print("====================================")


# ------------------------------------------------------------
# 8. Inspect event structure
# ------------------------------------------------------------

print("\n========== EVENT COLUMNS ==========")
print(events.columns.tolist())
print("===================================")

print("\n========== EVENT SAMPLE ==========")
print(events.head(10).to_string())
print("==================================")


# ------------------------------------------------------------
# 9. Identify actual keystroke events
# ------------------------------------------------------------

keystrokes = events[
    (events["is_key"] == True) &
    (events["key"].notna())
].copy()

# Keep actual printable/supported keys.
# Backspace is retained for inspection but will not become
# a decoder target later.
keystrokes["key"] = keystrokes["key"].astype(str)

print("\n========== KEYSTROKES ==========")
print("Total key events:", len(keystrokes))
print("Unique keys     :", sorted(keystrokes["key"].unique()))
print("================================")


# ------------------------------------------------------------
# 10. Create 0.5 second windows
#     -0.2 s → +0.3 s
# ------------------------------------------------------------

WINDOW_START = -0.2
WINDOW_DURATION = 0.5
WINDOW_END = WINDOW_START + WINDOW_DURATION

SFREQ = raw_eeg.info["sfreq"]
N_SAMPLES = int(round(WINDOW_DURATION * SFREQ))

print("\n========== WINDOW CONFIG ==========")
print("Start      :", WINDOW_START)
print("End        :", WINDOW_END)
print("Duration   :", WINDOW_DURATION)
print("Sampling   :", SFREQ)
print("Samples    :", N_SAMPLES)
print("===================================")


# ------------------------------------------------------------
# 11. Extract windows
# ------------------------------------------------------------

windows = []
metadata = []

data = raw_eeg.get_data()

for idx, event in keystrokes.iterrows():

    # IMPORTANT:
    # The official behavioral log uses lowercase 'time'.
    timestamp = float(event["time"])

    start_time = timestamp + WINDOW_START
    end_time = timestamp + WINDOW_END

    start_sample = int(round(start_time * SFREQ))
    end_sample = start_sample + N_SAMPLES

    # Boundary check
    if start_sample < 0 or end_sample > data.shape[1]:
        continue

    window = data[:, start_sample:end_sample]

    if window.shape != (
        len(raw_eeg.ch_names),
        N_SAMPLES
    ):
        continue

    # Numerical validation
    if not np.isfinite(window).all():
        continue

    windows.append(window)

    metadata.append({
        "event_index": idx,
        "trial_id": event["trial_id"],
        "timestamp": timestamp,
        "target": event["key"],
    })


windows = np.asarray(windows, dtype=np.float32)
metadata = pd.DataFrame(metadata)


# ------------------------------------------------------------
# 12. Validate
# ------------------------------------------------------------

print("\n========== WINDOW VALIDATION ==========")
print("Windows shape :", windows.shape)
print("Metadata rows :", len(metadata))
print("Expected shape: (N, 61, 25)")
print("NaN present   :", np.isnan(windows).any())
print("Inf present   :", np.isinf(windows).any())

if len(windows) > 0:
    print("Channels      :", windows.shape[1])
    print("Time samples  :", windows.shape[2])
    print("First target  :", metadata.iloc[0]["target"])
    print("First time    :", metadata.iloc[0]["timestamp"])

print("=======================================")
```

```python
# ============================================================
# PHASE 5 — BATCH WINDOW GENERATION
# Uses stable official manifest
# ============================================================

import numpy as np
import pandas as pd
import mne
import traceback
from pathlib import Path

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

PROJECT_DIR = Path(
    "/content/drive/MyDrive/NeuroSentinel"
)

PHASE5_DIR = (
    PROJECT_DIR / "06_phase5_utility_labels"
)

RECORDINGS_DIR = PHASE5_DIR / "recordings"

MANIFEST_FILE = (
    PHASE5_DIR / "phase5_stable_manifest.csv"
)

RECORDINGS_DIR.mkdir(
    parents=True,
    exist_ok=True
)

manifest = pd.read_csv(MANIFEST_FILE)

# ------------------------------------------------------------
# Configuration
# ------------------------------------------------------------

TARGET_SFREQ = 50
WINDOW_START = -0.2
WINDOW_DURATION = 0.5

PRE_SAMPLES = int(
    abs(WINDOW_START) * TARGET_SFREQ
)

N_SAMPLES = int(
    WINDOW_DURATION * TARGET_SFREQ
)

assert PRE_SAMPLES == 10
assert N_SAMPLES == 25

# ------------------------------------------------------------
# Results
# ------------------------------------------------------------

results = []
failures = []

# ------------------------------------------------------------
# Batch processing
# ------------------------------------------------------------

for idx, row in manifest.iterrows():

    recording_id = row["recording_id"]

    subject = str(row["subject"]).zfill(3)
    session = int(row["session"])
    task = str(row["task"])

    raw_path = Path(row["raw"])
    log_path = Path(row["log"])

    print("\n" + "=" * 70)
    print(
        f"[{idx + 1}/{len(manifest)}] "
        f"{recording_id}"
    )
    print("=" * 70)

    # --------------------------------------------------------
    # Skip already completed recordings
    # --------------------------------------------------------

    window_file = (
        RECORDINGS_DIR
        / f"{recording_id}_windows.npy"
    )

    metadata_file = (
        RECORDINGS_DIR
        / f"{recording_id}_metadata.csv"
    )

    if window_file.exists() and metadata_file.exists():

        print("Already processed — skipping.")

        try:
            existing_windows = np.load(
                window_file,
                mmap_mode="r"
            )

            existing_meta = pd.read_csv(
                metadata_file
            )

            assert existing_windows.shape[0] == len(
                existing_meta
            )

            results.append(
                {
                    "recording_id": recording_id,
                    "subject": subject,
                    "session": session,
                    "task": task,
                    "keystrokes": len(existing_meta),
                    "valid_windows": existing_windows.shape[0],
                    "skipped_windows": 0,
                    "window_shape": str(
                        existing_windows.shape
                    ),
                    "window_file": str(window_file),
                    "metadata_file": str(metadata_file),
                    "status": "PASS_EXISTING",
                }
            )

            continue

        except Exception:
            print(
                "Existing artifact invalid — "
                "reprocessing."
            )

    try:

        # ----------------------------------------------------
        # 1. Load raw EEG directly from official manifest
        # ----------------------------------------------------

        print("Loading EEG...")

        raw = mne.io.read_raw_brainvision(
            raw_path,
            preload=True,
            verbose=False
        )

        # ----------------------------------------------------
        # 2. Remove EOG
        # ----------------------------------------------------

        eog_channels = [
            ch for ch in raw.ch_names
            if "EOG" in ch.upper()
        ]

        if eog_channels:
            raw.drop_channels(eog_channels)

        # ----------------------------------------------------
        # 3. Montage
        # ----------------------------------------------------

        montage = mne.channels.make_standard_montage(
            "standard_1005"
        )

        raw.set_montage(
            montage,
            on_missing="ignore"
        )

        # ----------------------------------------------------
        # 4. Filter
        # ----------------------------------------------------

        raw.filter(
            l_freq=0.1,
            h_freq=20.0,
            verbose=False
        )

        # ----------------------------------------------------
        # 5. Resample
        # ----------------------------------------------------

        if raw.info["sfreq"] != TARGET_SFREQ:

            raw.resample(
                TARGET_SFREQ,
                npad="auto",
                verbose=False
            )

        # ----------------------------------------------------
        # 6. Create timeline
        # ----------------------------------------------------

        timeline = {
            "subject": subject,
            "session": session,
            "task": task,
        }

        # ----------------------------------------------------
        # 7. Read official behavioral log
        #
        # _preproc_log uses the official study mapping.
        # ----------------------------------------------------

        log = study._preproc_log(timeline)

        # ----------------------------------------------------
        # 8. Extract EEG triggers
        # ----------------------------------------------------

        eeg_events = study._preproc_events(raw)

        # ----------------------------------------------------
        # 9. Official alignment
        # ----------------------------------------------------

        aligned = study._align_log_events(
            log,
            eeg_events,
            max_n_errors=650
        )

        # ----------------------------------------------------
        # 10. Official metadata processing
        # ----------------------------------------------------

        meta = study._postprocess_meta(
            aligned
        )

        # ----------------------------------------------------
        # 11. Individual keystrokes
        # ----------------------------------------------------

        keystrokes = meta[
            (meta["type"] == "Keystroke") &
            (meta["button"].notna())
        ].copy()

        keystrokes = (
            keystrokes
            .sort_values("start")
            .reset_index(drop=True)
        )

        # ----------------------------------------------------
        # 12. Extract EEG windows
        # ----------------------------------------------------

        data = raw.get_data()

        n_channels = data.shape[0]

        windows = []
        targets = []

        skipped = 0

        for keystroke_index, keyrow in (
            keystrokes.iterrows()
        ):

            timestamp = float(
                keyrow["start"]
            )

            center_sample = int(
                round(timestamp * TARGET_SFREQ)
            )

            start = (
                center_sample
                - PRE_SAMPLES
            )

            end = (
                start
                + N_SAMPLES
            )

            if (
                start < 0
                or end > raw.n_times
            ):
                skipped += 1
                continue

            window = data[
                :,
                start:end
            ]

            if window.shape != (
                n_channels,
                N_SAMPLES
            ):
                skipped += 1
                continue

            if not np.isfinite(
                window
            ).all():
                skipped += 1
                continue

            windows.append(window)

            targets.append(
                {
                    "recording_id": recording_id,
                    "subject": subject,
                    "session": session,
                    "task": task,
                    "trial_id": keyrow["trial_id"],
                    "keystroke_index": keystroke_index,
                    "timestamp": timestamp,
                    "duration": keyrow["duration"],
                    "target": keyrow["button"],
                    "text": keyrow["text"],
                }
            )

        if not windows:
            raise ValueError(
                "No valid windows generated."
            )

        windows = np.stack(windows)

        target_df = pd.DataFrame(
            targets
        )

        # ----------------------------------------------------
        # 13. Validation
        # ----------------------------------------------------

        assert windows.ndim == 3
        assert windows.shape[1] == n_channels
        assert windows.shape[2] == 25

        assert len(windows) == len(
            target_df
        )

        assert target_df[
            "target"
        ].notna().all()

        assert np.isfinite(
            windows
        ).all()

        # ----------------------------------------------------
        # 14. Save immediately
        # ----------------------------------------------------

        np.save(
            window_file,
            windows
        )

        target_df.to_csv(
            metadata_file,
            index=False
        )

        # ----------------------------------------------------
        # 15. Record success
        # ----------------------------------------------------

        results.append(
            {
                "recording_id": recording_id,
                "subject": subject,
                "session": session,
                "task": task,
                "keystrokes": len(keystrokes),
                "valid_windows": len(windows),
                "skipped_windows": skipped,
                "window_shape": str(
                    windows.shape
                ),
                "window_file": str(
                    window_file
                ),
                "metadata_file": str(
                    metadata_file
                ),
                "status": "PASS",
            }
        )

        print(
            f"PASS | "
            f"keystrokes={len(keystrokes)} | "
            f"windows={len(windows)} | "
            f"skipped={skipped} | "
            f"shape={windows.shape}"
        )

        # Free memory before next recording
        del raw
        del data
        del windows
        del target_df
        del meta
        del aligned
        del eeg_events
        del log

    except Exception as e:

        print(
            f"FAIL | {type(e).__name__}: {e}"
        )

        failures.append(
            {
                "recording_id": recording_id,
                "subject": subject,
                "session": session,
                "task": task,
                "error_type": type(e).__name__,
                "error": str(e),
                "status": "FAIL",
            }
        )

        traceback.print_exc()

# ------------------------------------------------------------
# Save reports
# ------------------------------------------------------------

results_df = pd.DataFrame(results)
failures_df = pd.DataFrame(failures)

results_file = (
    PHASE5_DIR
    / "phase5_batch_results.csv"
)

failures_file = (
    PHASE5_DIR
    / "phase5_batch_failures.csv"
)

results_df.to_csv(
    results_file,
    index=False
)

failures_df.to_csv(
    failures_file,
    index=False
)

# ------------------------------------------------------------
# Final report
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PHASE 5 BATCH COMPLETE")
print("=" * 70)

print(
    "Manifest recordings :",
    len(manifest)
)

print(
    "Successful           :",
    len(results_df)
)

print(
    "Failed               :",
    len(failures_df)
)

if len(results_df):

    print(
        "Total valid windows  :",
        int(
            results_df[
                "valid_windows"
            ].sum()
        )
    )

    print(
        "Total skipped        :",
        int(
            results_df[
                "skipped_windows"
            ].sum()
        )
    )

print("\nResults:")
print(results_file)

print("\nFailures:")
print(failures_file)

print("=" * 70)
```

```python
# ============================================================
# PHASE 6 — SUBJECT-WISE TRAIN / VAL / TEST SPLIT
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path
from sklearn.model_selection import train_test_split
import json

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

PROJECT_DIR = Path(
    "/content/drive/MyDrive/NeuroSentinel"
)

PHASE5_DIR = (
    PROJECT_DIR / "06_phase5_utility_labels"
)

PHASE6_DIR = (
    PROJECT_DIR / "07_phase6_subject_split"
)

CHECKPOINT_DIR = (
    PROJECT_DIR / "checkpoints"
)

PHASE6_DIR.mkdir(
    parents=True,
    exist_ok=True
)

CHECKPOINT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

MANIFEST_FILE = (
    PHASE5_DIR / "phase5_stable_manifest.csv"
)

RESULTS_FILE = (
    PHASE5_DIR / "phase5_batch_results.csv"
)

# ------------------------------------------------------------
# Load manifest and results
# ------------------------------------------------------------

manifest = pd.read_csv(MANIFEST_FILE)
results = pd.read_csv(RESULTS_FILE)

print("=" * 70)
print("PHASE 6 — SUBJECT-WISE SPLIT")
print("=" * 70)

print("Recordings:", len(manifest))

# ------------------------------------------------------------
# Verify recording consistency
# ------------------------------------------------------------

assert manifest["recording_id"].is_unique

assert set(manifest["recording_id"]) == set(
    results["recording_id"]
)

assert (
    results["status"]
    .isin(["PASS", "PASS_EXISTING"])
    .all()
), "Not all recordings were successfully processed."

# ------------------------------------------------------------
# Get unique subjects
# ------------------------------------------------------------

subjects = sorted(
    manifest["subject"]
    .astype(str)
    .str.zfill(3)
    .unique()
)

print("Subjects:", len(subjects))
print("Subject IDs:", subjects)

assert len(subjects) == 20

# ------------------------------------------------------------
# Split subjects
#
# Target:
# Train ≈ 70%
# Validation ≈ 15%
# Test ≈ 15%
#
# 20 subjects cannot produce exactly 70/15/15.
# We use:
# Train = 14 subjects
# Val   = 3 subjects
# Test  = 3 subjects
# ------------------------------------------------------------

SEED = 33

train_subjects, temp_subjects = train_test_split(
    subjects,
    test_size=6,
    random_state=SEED,
    shuffle=True
)

val_subjects, test_subjects = train_test_split(
    temp_subjects,
    test_size=3,
    random_state=SEED,
    shuffle=True
)

train_subjects = sorted(train_subjects)
val_subjects = sorted(val_subjects)
test_subjects = sorted(test_subjects)

# ------------------------------------------------------------
# Verify no subject overlap
# ------------------------------------------------------------

assert set(train_subjects).isdisjoint(
    val_subjects
)

assert set(train_subjects).isdisjoint(
    test_subjects
)

assert set(val_subjects).isdisjoint(
    test_subjects
)

assert (
    len(train_subjects)
    + len(val_subjects)
    + len(test_subjects)
    == len(subjects)
)

# ------------------------------------------------------------
# Add split to recording manifest
# ------------------------------------------------------------

def assign_split(subject):

    subject = str(subject).zfill(3)

    if subject in train_subjects:
        return "train"

    if subject in val_subjects:
        return "val"

    if subject in test_subjects:
        return "test"

    raise ValueError(
        f"Unknown subject: {subject}"
    )


split_manifest = manifest.copy()

split_manifest["subject"] = (
    split_manifest["subject"]
    .astype(str)
    .str.zfill(3)
)

split_manifest["split"] = (
    split_manifest["subject"]
    .apply(assign_split)
)

# ------------------------------------------------------------
# Validate every recording has a split
# ------------------------------------------------------------

assert split_manifest["split"].notna().all()

assert (
    split_manifest["subject"]
    .isin(subjects)
    .all()
)

# ------------------------------------------------------------
# Recording counts
# ------------------------------------------------------------

recording_counts = (
    split_manifest
    .groupby("split")
    .size()
    .reindex(["train", "val", "test"])
    .fillna(0)
    .astype(int)
)

# ------------------------------------------------------------
# Window counts
# ------------------------------------------------------------

results["recording_id"] = (
    results["recording_id"].astype(str)
)

split_manifest["recording_id"] = (
    split_manifest["recording_id"].astype(str)
)

window_summary = results[
    ["recording_id", "valid_windows"]
].merge(
    split_manifest[
        ["recording_id", "subject", "split"]
    ],
    on="recording_id",
    how="left"
)

assert window_summary["split"].notna().all()

window_counts = (
    window_summary
    .groupby("split")["valid_windows"]
    .sum()
    .reindex(["train", "val", "test"])
    .fillna(0)
    .astype(int)
)

# ------------------------------------------------------------
# Save split manifest
# ------------------------------------------------------------

split_manifest_file = (
    PHASE6_DIR
    / "phase6_subject_split_manifest.csv"
)

split_manifest.to_csv(
    split_manifest_file,
    index=False
)

# ------------------------------------------------------------
# Save subject lists
# ------------------------------------------------------------

subject_split_file = (
    PHASE6_DIR
    / "phase6_subject_split.json"
)

subject_split = {
    "seed": SEED,
    "train_subjects": train_subjects,
    "val_subjects": val_subjects,
    "test_subjects": test_subjects,
    "num_train_subjects": len(train_subjects),
    "num_val_subjects": len(val_subjects),
    "num_test_subjects": len(test_subjects)
}

with open(
    subject_split_file,
    "w"
) as f:
    json.dump(
        subject_split,
        f,
        indent=2
    )

# ------------------------------------------------------------
# Checkpoint
# ------------------------------------------------------------

checkpoint = {
    "project": "NeuroSentinel",
    "phase": 6,
    "status": "SUBJECT_WISE_SPLIT_COMPLETE",

    "seed": SEED,

    "total_subjects": len(subjects),
    "train_subjects": len(train_subjects),
    "val_subjects": len(val_subjects),
    "test_subjects": len(test_subjects),

    "train_subject_ids": train_subjects,
    "val_subject_ids": val_subjects,
    "test_subject_ids": test_subjects,

    "recordings": {
        "train": int(recording_counts["train"]),
        "val": int(recording_counts["val"]),
        "test": int(recording_counts["test"])
    },

    "windows": {
        "train": int(window_counts["train"]),
        "val": int(window_counts["val"]),
        "test": int(window_counts["test"])
    },

    "leakage_check": "PASS",

    "next_step": (
        "Build and train decoder baselines using the "
        "subject-wise split before causal gating."
    )
}

checkpoint_file = (
    CHECKPOINT_DIR
    / "phase6_subject_split_checkpoint.json"
)

with open(
    checkpoint_file,
    "w"
) as f:
    json.dump(
        checkpoint,
        f,
        indent=2
    )

# ------------------------------------------------------------
# Display results
# ------------------------------------------------------------

print("\nSUBJECT SPLIT")
print("-" * 70)

print(
    "Train:",
    len(train_subjects),
    "subjects →",
    train_subjects
)

print(
    "Val  :",
    len(val_subjects),
    "subjects →",
    val_subjects
)

print(
    "Test :",
    len(test_subjects),
    "subjects →",
    test_subjects
)

print("\nRECORDINGS")
print("-" * 70)
print(recording_counts)

print("\nWINDOWS")
print("-" * 70)
print(window_counts)

print("\nLEAKAGE CHECK")
print("-" * 70)
print("Train ∩ Val :", set(train_subjects) & set(val_subjects))
print("Train ∩ Test:", set(train_subjects) & set(test_subjects))
print("Val ∩ Test  :", set(val_subjects) & set(test_subjects))

print("\nSaved:")
print(split_manifest_file)
print(subject_split_file)
print(checkpoint_file)

print("\n" + "=" * 70)
print("PHASE 6 SPLIT COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# PHASE 5A — BRAIN2QWERTY EEG DECODER SMOKE TEST
# ============================================================

import sys
import torch
import numpy as np
from pathlib import Path

print("=" * 70)
print("PHASE 5A — DECODER SMOKE TEST")
print("=" * 70)

# ------------------------------------------------------------
# 1. Paths
# ------------------------------------------------------------

PROJECT_DIR = Path("/content/drive/MyDrive/NeuroSentinel")
REPO_DIR = PROJECT_DIR / "brain2qwerty"

WINDOW_DIR = PROJECT_DIR / "06_phase5_utility_labels" / "recordings"
SPLIT_FILE = (
    PROJECT_DIR
    / "07_phase6_subject_split"
    / "phase6_subject_split_manifest.csv"
)

print("Project:", PROJECT_DIR)
print("Repo   :", REPO_DIR)
print("Windows:", WINDOW_DIR)

# ------------------------------------------------------------
# 2. Import Brain2Qwerty / neuraltrain
# ------------------------------------------------------------

if str(REPO_DIR) not in sys.path:
    sys.path.insert(0, str(REPO_DIR))

from neuraltrain.models import SimpleConvTimeAgg

print("✓ SimpleConvTimeAgg imported")

# ------------------------------------------------------------
# 3. Load one REAL Phase 5 recording
# ------------------------------------------------------------

window_files = sorted(WINDOW_DIR.glob("*_windows.npy"))

if not window_files:
    raise FileNotFoundError(
        f"No window files found in {WINDOW_DIR}"
    )

sample_file = window_files[0]

X = np.load(sample_file, mmap_mode="r")

print("\nREAL DATA")
print("-" * 70)
print("File:", sample_file.name)
print("Shape:", X.shape)
print("dtype:", X.dtype)

# Expected: (N, 61, 25)

assert X.ndim == 3
assert X.shape[1] == 61
assert X.shape[2] == 25

print("✓ Expected EEG shape: (N, 61, 25)")

# ------------------------------------------------------------
# 4. Prepare small GPU batch
# ------------------------------------------------------------

BATCH_SIZE = min(8, len(X))

x = torch.from_numpy(
    np.asarray(X[:BATCH_SIZE], dtype=np.float32)
)

print("\nINPUT BATCH")
print("-" * 70)
print("Shape:", tuple(x.shape))

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", device)

x = x.to(device)

# ------------------------------------------------------------
# 5. Build Brain2Qwerty-compatible EEG encoder
# ------------------------------------------------------------

model = SimpleConvTimeAgg(
    time_agg_out="att",
    dropout_input=0.2,
    conv_dropout=0.5,
    hidden=2048,
    batch_norm=True,
    depth=8,
    dilation_period=3,
    kernel_size=3,
    relu_leakiness=0.01,
    initial_linear=512,
    gelu=True,
    skip=True,
    scale=0.1,

    # IMPORTANT:
    # Disable the MEG-specific 2D channel merger.
    merger_config=None,

    subject_layers_config=None,
)

model = model.build(
    n_in_channels=61,
    n_outputs=2048,
)

model = model.to(device)
model.eval()

print("\nMODEL")
print("-" * 70)
print("Input channels :", 61)
print("Output features:", 2048)
print("MEG merger      : DISABLED")
print("✓ Model created")

# ------------------------------------------------------------
# 6. Forward pass
# ------------------------------------------------------------

with torch.no_grad():
    embedding = model(x)

print("\nFORWARD PASS")
print("-" * 70)
print("Embedding shape:", tuple(embedding.shape))

assert embedding.ndim == 2
assert embedding.shape[0] == BATCH_SIZE
assert embedding.shape[1] == 2048

print("✓ Encoder forward pass successful")

# ------------------------------------------------------------
# 7. Add 29-class character head
# ------------------------------------------------------------

NUM_CLASSES = 29

classifier = torch.nn.Linear(
    2048,
    NUM_CLASSES
).to(device)

classifier.eval()

with torch.no_grad():
    logits = classifier(embedding)

print("\nCLASSIFIER")
print("-" * 70)
print("Logits shape:", tuple(logits.shape))

assert logits.shape == (BATCH_SIZE, NUM_CLASSES)

print("✓ 29-class output successful")

# ------------------------------------------------------------
# 8. Verify probability/confidence calculation
# ------------------------------------------------------------

with torch.no_grad():
    probabilities = torch.softmax(logits, dim=1)
    confidence, prediction = probabilities.max(dim=1)

print("\nPREDICTIONS")
print("-" * 70)

for i in range(BATCH_SIZE):
    print(
        f"Window {i:02d} | "
        f"prediction={prediction[i].item():2d} | "
        f"confidence={confidence[i].item():.6f}"
    )

assert torch.isfinite(logits).all()
assert torch.isfinite(probabilities).all()
assert torch.isfinite(confidence).all()

assert torch.allclose(
    probabilities.sum(dim=1),
    torch.ones(BATCH_SIZE, device=device),
    atol=1e-5,
)

print("\n✓ No NaN/Inf values")
print("✓ Probabilities sum to 1")
print("✓ Confidence calculation successful")

# ------------------------------------------------------------
# 9. Final result
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PHASE 5A SMOKE TEST COMPLETE")
print("=" * 70)
print("Status: PASS")
print()
print("Verified:")
print("  ✓ Real EEG windows")
print("  ✓ 61 input channels")
print("  ✓ 25 samples/window")
print("  ✓ Brain2Qwerty-compatible encoder")
print("  ✓ 2048-dimensional embedding")
print("  ✓ 29-class classifier")
print("  ✓ Prediction generation")
print("  ✓ Confidence calculation")
print("=" * 70)
```

```python
# ============================================================
# PHASE 5B — CORRECTED DATASET LABEL LOADING
# ============================================================

class_to_idx = {
    "s": 0,
    "o": 1,
    "t": 2,
    "e": 3,
    "n": 4,
    "c": 5,
    "i": 6,
    "a": 7,
    "<space>": 8,
    "d": 9,
    "l": 10,
    "r": 11,
    "b": 12,
    "<special>": 13,
    "z": 14,
    "v": 15,
    "f": 16,
    "m": 17,
    "u": 18,
    "h": 19,
    "p": 20,
    "g": 21,
    "q": 22,
    "w": 23,
    "x": 24,
    "y": 25,
    "j": 26,
    "k": 27,
    "<number>": 28,
}


class EEGWindowDataset(Dataset):

    def __init__(
        self,
        records,
        max_windows=None,
    ):

        self.X_parts = []
        self.y_parts = []

        total = 0

        for record_id, window_file, metadata_file in records:

            X = np.load(
                window_file,
                mmap_mode="r"
            )

            meta = pd.read_csv(metadata_file)

            if len(X) != len(meta):
                raise ValueError(
                    f"{record_id}: "
                    f"X={len(X)}, metadata={len(meta)}"
                )

            # ------------------------------------------------
            # Find the actual character target column
            # ------------------------------------------------

            target_column = None

            for candidate in [
                "target",
                "button",
                "key",
                "character",
                "char",
            ]:
                if candidate in meta.columns:
                    target_column = candidate
                    break

            if target_column is None:
                raise ValueError(
                    f"{record_id}: Could not find "
                    f"character target column.\n"
                    f"Available columns: {list(meta.columns)}"
                )

            labels = meta[target_column].astype(str)

            # ------------------------------------------------
            # Convert character → official Brain2Qwerty ID
            # ------------------------------------------------

            unknown = sorted(
                set(labels) - set(class_to_idx.keys())
            )

            if unknown:
                raise ValueError(
                    f"{record_id}: Unknown target labels: "
                    f"{unknown}"
                )

            y = labels.map(class_to_idx).to_numpy(
                dtype=np.int64
            )

            # ------------------------------------------------
            # Limit number of windows for smoke test
            # ------------------------------------------------

            if max_windows is not None:

                remaining = max_windows - total

                if remaining <= 0:
                    break

                n = min(
                    len(X),
                    remaining
                )

                X = np.asarray(
                    X[:n],
                    dtype=np.float32
                )

                y = y[:n]

            else:

                X = np.asarray(
                    X,
                    dtype=np.float32
                )

            self.X_parts.append(X)
            self.y_parts.append(y)

            total += len(X)

            if (
                max_windows is not None
                and total >= max_windows
            ):
                break

        if not self.X_parts:
            raise RuntimeError(
                "No EEG training data loaded."
            )

        self.X = np.concatenate(
            self.X_parts,
            axis=0
        )

        self.y = np.concatenate(
            self.y_parts,
            axis=0
        )

    def __len__(self):
        return len(self.y)

    def __getitem__(self, idx):

        return (
            torch.from_numpy(self.X[idx]),
            torch.tensor(
                self.y[idx],
                dtype=torch.long
            ),
        )


# ============================================================
# CREATE SMOKE-TEST DATASETS
# ============================================================

TRAIN_WINDOWS = 8000
VAL_WINDOWS = 2000

train_ds = EEGWindowDataset(
    train_records,
    max_windows=TRAIN_WINDOWS,
)

val_ds = EEGWindowDataset(
    val_records,
    max_windows=VAL_WINDOWS,
)

print("=" * 70)
print("DATASET VALIDATION")
print("=" * 70)

print("Train windows:", len(train_ds))
print("Val windows  :", len(val_ds))
print("Input shape  :", tuple(train_ds[0][0].shape))

print(
    "Train labels:",
    sorted(np.unique(train_ds.y).tolist())
)

print(
    "Val labels  :",
    sorted(np.unique(val_ds.y).tolist())
)

assert train_ds[0][0].shape == (61, 25)

assert train_ds.y.min() >= 0
assert train_ds.y.max() < 29

assert val_ds.y.min() >= 0
assert val_ds.y.max() < 29

print()
print("✓ EEG shape validated")
print("✓ Character → class mapping validated")
print("✓ All labels are in range 0–28")
print("=" * 70)
```

```python
# ============================================================
# PHASE 5B — EEG DECODER TRAINING SMOKE RUN
# ============================================================

import json
import random
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from pathlib import Path

from neuraltrain.models import SimpleConvTimeAgg

print("=" * 70)
print("PHASE 5B — EEG DECODER TRAINING SMOKE RUN")
print("=" * 70)

# ------------------------------------------------------------
# 1. Reproducibility
# ------------------------------------------------------------

SEED = 33

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", device)

# ------------------------------------------------------------
# 2. Paths
# ------------------------------------------------------------

PROJECT_DIR = Path(
    "/content/drive/MyDrive/NeuroSentinel"
)

WINDOW_DIR = (
    PROJECT_DIR
    / "06_phase5_utility_labels"
    / "recordings"
)

SPLIT_FILE = (
    PROJECT_DIR
    / "07_phase6_subject_split"
    / "phase6_subject_split_manifest.csv"
)

OUTPUT_DIR = (
    PROJECT_DIR
    / "06_phase5_utility_labels"
    / "decoder_training"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# 3. Load frozen subject split
# ------------------------------------------------------------

split_df = pd.read_csv(SPLIT_FILE)

required_cols = {
    "subject",
    "session",
    "task",
    "split",
}

missing = required_cols - set(split_df.columns)

if missing:
    raise ValueError(
        f"Missing columns in split manifest: {missing}"
    )

print("\nFROZEN SPLIT")
print("-" * 70)

for split in ["train", "val", "test"]:
    subjects = sorted(
        split_df.loc[
            split_df["split"] == split,
            "subject"
        ].astype(str).unique()
    )

    print(
        f"{split:5s}: "
        f"{len(subjects)} subjects → {subjects}"
    )

# ------------------------------------------------------------
# 4. Load window + metadata pairs
# ------------------------------------------------------------

def get_recordings(split_name):

    rows = split_df[
        split_df["split"] == split_name
    ]

    recordings = []

    for _, row in rows.iterrows():

        subject = str(row["subject"]).zfill(3)
        session = int(row["session"])
        task = str(row["task"])

        record_id = (
            f"{subject}_S{session}_{task}"
        )

        window_file = (
            WINDOW_DIR
            / f"{record_id}_windows.npy"
        )

        metadata_file = (
            WINDOW_DIR
            / f"{record_id}_metadata.csv"
        )

        if not window_file.exists():
            raise FileNotFoundError(
                f"Missing window file:\n{window_file}"
            )

        if not metadata_file.exists():
            raise FileNotFoundError(
                f"Missing metadata file:\n{metadata_file}"
            )

        recordings.append(
            (record_id, window_file, metadata_file)
        )

    return recordings


train_records = get_recordings("train")
val_records = get_recordings("val")

print("\nRECORDINGS")
print("-" * 70)
print("Train:", len(train_records))
print("Val  :", len(val_records))

# ------------------------------------------------------------
# 5. Dataset
# ------------------------------------------------------------

class EEGWindowDataset(Dataset):

    def __init__(
        self,
        records,
        max_windows=None,
    ):

        self.X_parts = []
        self.y_parts = []

        total = 0

        for record_id, window_file, metadata_file in records:

            X = np.load(
                window_file,
                mmap_mode="r"
            )

            meta = pd.read_csv(metadata_file)

            if len(X) != len(meta):
                raise ValueError(
                    f"{record_id}: "
                    f"X={len(X)}, metadata={len(meta)}"
                )

            # The metadata generated during Phase 5
            # contains the target class.
            target_column = None

            for candidate in [
                "target_class",
                "target",
                "button_class",
                "label",
            ]:
                if candidate in meta.columns:
                    target_column = candidate
                    break

            if target_column is None:
                raise ValueError(
                    f"{record_id}: "
                    "Could not find target class column."
                )

            y = meta[target_column].to_numpy(
                dtype=np.int64
            )

            # Limit each recording if requested.
            if max_windows is not None:

                remaining = max_windows - total

                if remaining <= 0:
                    break

                n = min(len(X), remaining)

                X = np.asarray(
                    X[:n],
                    dtype=np.float32
                )

                y = y[:n]

            else:

                X = np.asarray(
                    X,
                    dtype=np.float32
                )

            self.X_parts.append(X)
            self.y_parts.append(y)

            total += len(X)

            if (
                max_windows is not None
                and total >= max_windows
            ):
                break

        if not self.X_parts:
            raise RuntimeError(
                "No training data loaded."
            )

        self.X = np.concatenate(
            self.X_parts,
            axis=0
        )

        self.y = np.concatenate(
            self.y_parts,
            axis=0
        )

    def __len__(self):
        return len(self.y)

    def __getitem__(self, idx):

        return (
            torch.from_numpy(self.X[idx]),
            torch.tensor(
                self.y[idx],
                dtype=torch.long
            ),
        )


# ------------------------------------------------------------
# 6. Small smoke-run dataset
# ------------------------------------------------------------

TRAIN_WINDOWS = 8000
VAL_WINDOWS = 2000

train_ds = EEGWindowDataset(
    train_records,
    max_windows=TRAIN_WINDOWS,
)

val_ds = EEGWindowDataset(
    val_records,
    max_windows=VAL_WINDOWS,
)

print("\nDATASET")
print("-" * 70)
print("Train windows:", len(train_ds))
print("Val windows  :", len(val_ds))
print("Input shape  :", train_ds[0][0].shape)

assert train_ds[0][0].shape == (61, 25)

# ------------------------------------------------------------
# 7. DataLoaders
# ------------------------------------------------------------

BATCH_SIZE = 32

train_loader = DataLoader(
    train_ds,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=0,
    pin_memory=torch.cuda.is_available(),
)

val_loader = DataLoader(
    val_ds,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0,
    pin_memory=torch.cuda.is_available(),
)

# ------------------------------------------------------------
# 8. Build Brain2Qwerty-compatible encoder
# ------------------------------------------------------------

encoder_config = dict(
    time_agg_out="att",
    dropout_input=0.2,
    conv_dropout=0.5,
    hidden=2048,
    batch_norm=True,
    depth=8,
    dilation_period=3,
    kernel_size=3,
    relu_leakiness=0.01,
    initial_linear=512,
    gelu=True,
    skip=True,
    scale=0.1,
    merger_config=None,
    subject_layers_config=None,
)

encoder = SimpleConvTimeAgg(
    **encoder_config
).build(
    n_in_channels=61,
    n_outputs=2048,
)

classifier = nn.Linear(
    2048,
    29
)

model = nn.Sequential(
    encoder,
    classifier,
).to(device)

# ------------------------------------------------------------
# 9. Optimizer + loss
# ------------------------------------------------------------

criterion = nn.CrossEntropyLoss()

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=5e-5,
    weight_decay=1e-2,
)

# ------------------------------------------------------------
# 10. Training loop
# ------------------------------------------------------------

EPOCHS = 2

print("\nTRAINING")
print("-" * 70)

history = []

for epoch in range(EPOCHS):

    # -------------------------
    # Train
    # -------------------------

    model.train()

    train_loss = 0.0
    train_correct = 0
    train_total = 0

    for X_batch, y_batch in train_loader:

        X_batch = X_batch.to(
            device,
            non_blocking=True
        )

        y_batch = y_batch.to(
            device,
            non_blocking=True
        )

        optimizer.zero_grad(
            set_to_none=True
        )

        logits = model(X_batch)

        loss = criterion(
            logits,
            y_batch
        )

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=1.0
        )

        optimizer.step()

        train_loss += (
            loss.item()
            * y_batch.size(0)
        )

        predictions = logits.argmax(
            dim=1
        )

        train_correct += (
            predictions == y_batch
        ).sum().item()

        train_total += y_batch.size(0)

    train_loss /= train_total
    train_acc = train_correct / train_total

    # -------------------------
    # Validation
    # -------------------------

    model.eval()

    val_loss = 0.0
    val_correct = 0
    val_total = 0

    with torch.no_grad():

        for X_batch, y_batch in val_loader:

            X_batch = X_batch.to(
                device,
                non_blocking=True
            )

            y_batch = y_batch.to(
                device,
                non_blocking=True
            )

            logits = model(X_batch)

            loss = criterion(
                logits,
                y_batch
            )

            val_loss += (
                loss.item()
                * y_batch.size(0)
            )

            predictions = logits.argmax(
                dim=1
            )

            val_correct += (
                predictions == y_batch
            ).sum().item()

            val_total += y_batch.size(0)

    val_loss /= val_total
    val_acc = val_correct / val_total

    history.append({
        "epoch": epoch + 1,
        "train_loss": train_loss,
        "train_accuracy": train_acc,
        "val_loss": val_loss,
        "val_accuracy": val_acc,
    })

    print(
        f"Epoch {epoch + 1}/{EPOCHS} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_acc:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_acc:.4f}"
    )

# ------------------------------------------------------------
# 11. Learning check
# ------------------------------------------------------------

print("\nLEARNING CHECK")
print("-" * 70)

initial_loss = history[0]["train_loss"]
final_loss = history[-1]["train_loss"]

print(
    f"Initial train loss: {initial_loss:.6f}"
)

print(
    f"Final train loss  : {final_loss:.6f}"
)

if final_loss < initial_loss:
    print("✓ Training loss decreased")
else:
    print(
        "⚠ Training loss did not decrease"
    )

# ------------------------------------------------------------
# 12. Save smoke-test checkpoint
# ------------------------------------------------------------

checkpoint_file = (
    OUTPUT_DIR
    / "phase5_decoder_smoke_checkpoint.pt"
)

torch.save(
    {
        "phase": 5,
        "stage": "5B_smoke_training",
        "seed": SEED,
        "input_channels": 61,
        "samples_per_window": 25,
        "num_classes": 29,
        "encoder_config": encoder_config,
        "model_state_dict": model.state_dict(),
        "optimizer_state_dict": optimizer.state_dict(),
        "history": history,
    },
    checkpoint_file,
)

print("\nSaved:")
print(checkpoint_file)

# ------------------------------------------------------------
# 13. Final
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PHASE 5B SMOKE TRAINING COMPLETE")
print("=" * 70)

print("Status: READY FOR REVIEW")
print()
print("This was only a small training run.")
print("Its predictions MUST NOT be used as final Phase 5 utility labels.")
print("=" * 70)
```

```python
# ============================================================
# PHASE 5C — FULL EEG DECODER TRAINING
# Corrected for actual Phase 6 manifest
# ============================================================

import os
import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from torch.utils.data import Dataset, DataLoader
from neuraltrain.models.simpleconv import SimpleConvTimeAgg

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

PROJECT_DIR = "/content/drive/MyDrive/NeuroSentinel"

SPLIT_MANIFEST = os.path.join(
    PROJECT_DIR,
    "07_phase6_subject_split",
    "phase6_subject_split_manifest.csv"
)

RECORDINGS_DIR = os.path.join(
    PROJECT_DIR,
    "06_phase5_utility_labels",
    "recordings"
)

CHECKPOINT_DIR = os.path.join(
    PROJECT_DIR,
    "checkpoints"
)

RESULT_DIR = os.path.join(
    PROJECT_DIR,
    "results"
)

os.makedirs(CHECKPOINT_DIR, exist_ok=True)
os.makedirs(RESULT_DIR, exist_ok=True)

BEST_MODEL_PATH = os.path.join(
    CHECKPOINT_DIR,
    "phase5_decoder_best.pt"
)

HISTORY_PATH = os.path.join(
    RESULT_DIR,
    "phase5_decoder_training_history.csv"
)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

# ------------------------------------------------------------
# FROZEN SPLIT
# ------------------------------------------------------------

split_df = pd.read_csv(SPLIT_MANIFEST)

train_records = split_df[
    split_df["split"] == "train"
].copy()

val_records = split_df[
    split_df["split"] == "val"
].copy()

test_records = split_df[
    split_df["split"] == "test"
].copy()

print("=" * 70)
print("PHASE 5C — FULL EEG DECODER TRAINING")
print("=" * 70)

print("Device:", device)

print("\nFROZEN SPLIT")
print("-" * 70)

print(
    "Train:",
    len(train_records),
    "recordings"
)

print(
    "Val  :",
    len(val_records),
    "recordings"
)

print(
    "Test :",
    len(test_records),
    "recordings"
)

# ------------------------------------------------------------
# BRAIN2QWERTY CLASS MAPPING
# ------------------------------------------------------------

class_to_idx = {
    "s": 0,
    "o": 1,
    "t": 2,
    "e": 3,
    "n": 4,
    "c": 5,
    "i": 6,
    "a": 7,
    "<space>": 8,
    "d": 9,
    "l": 10,
    "r": 11,
    "b": 12,
    "<special>": 13,
    "z": 14,
    "v": 15,
    "f": 16,
    "m": 17,
    "u": 18,
    "h": 19,
    "p": 20,
    "g": 21,
    "q": 22,
    "w": 23,
    "x": 24,
    "y": 25,
    "j": 26,
    "k": 27,
    "<number>": 28,
}

# ------------------------------------------------------------
# HELPER — RECORDING PATHS
# ------------------------------------------------------------

def get_recording_paths(recording_id):

    window_path = os.path.join(
        RECORDINGS_DIR,
        f"{recording_id}_windows.npy"
    )

    metadata_path = os.path.join(
        RECORDINGS_DIR,
        f"{recording_id}_metadata.csv"
    )

    return window_path, metadata_path


# ------------------------------------------------------------
# LAZY MULTI-RECORDING DATASET
# ------------------------------------------------------------

class MultiRecordingEEGDataset(Dataset):

    def __init__(self, records):

        self.samples = []

        print("\nIndexing recordings...")

        for _, row in records.iterrows():

            recording_id = str(
                row["recording_id"]
            )

            window_path, metadata_path = (
                get_recording_paths(recording_id)
            )

            if not os.path.exists(window_path):
                raise FileNotFoundError(
                    f"Missing windows:\n{window_path}"
                )

            if not os.path.exists(metadata_path):
                raise FileNotFoundError(
                    f"Missing metadata:\n{metadata_path}"
                )

            meta = pd.read_csv(
                metadata_path
            )

            # Find target column
            target_column = None

            for col in [
                "target",
                "button",
                "key",
                "character",
                "char"
            ]:

                if col in meta.columns:
                    target_column = col
                    break

            if target_column is None:
                raise ValueError(
                    f"No target column in {metadata_path}\n"
                    f"Columns: {meta.columns.tolist()}"
                )

            targets = meta[
                target_column
            ].astype(str)

            # Make sure metadata/window counts agree
            windows = np.load(
                window_path,
                mmap_mode="r"
            )

            if len(windows) != len(meta):
                raise ValueError(
                    f"Count mismatch for {recording_id}: "
                    f"windows={len(windows)}, "
                    f"metadata={len(meta)}"
                )

            valid_count = 0

            for idx, target in enumerate(targets):

                if target in class_to_idx:

                    self.samples.append(
                        (
                            window_path,
                            idx,
                            class_to_idx[target]
                        )
                    )

                    valid_count += 1

            print(
                f"{recording_id}: "
                f"{valid_count:,} valid windows"
            )

        self._cache = {}

        print(
            f"\nTotal indexed samples: "
            f"{len(self.samples):,}"
        )

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, index):

        path, window_index, label = (
            self.samples[index]
        )

        # Lazy memory-mapped loading
        if path not in self._cache:

            self._cache[path] = np.load(
                path,
                mmap_mode="r"
            )

        X = self._cache[path][window_index]

        X = np.asarray(
            X,
            dtype=np.float32
        )

        y = np.int64(label)

        return (
            torch.from_numpy(X),
            torch.tensor(
                y,
                dtype=torch.long
            )
        )


# ------------------------------------------------------------
# DATASETS
# ------------------------------------------------------------

print("\nBUILDING DATASETS")
print("-" * 70)

train_ds = MultiRecordingEEGDataset(
    train_records
)

val_ds = MultiRecordingEEGDataset(
    val_records
)

print("\nDATASET SIZES")
print("-" * 70)

print(
    f"Train windows: {len(train_ds):,}"
)

print(
    f"Val windows  : {len(val_ds):,}"
)

# ------------------------------------------------------------
# DATALOADERS
# ------------------------------------------------------------

BATCH_SIZE = 64

train_loader = DataLoader(
    train_ds,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=0,
    pin_memory=torch.cuda.is_available()
)

val_loader = DataLoader(
    val_ds,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0,
    pin_memory=torch.cuda.is_available()
)

print("\nBATCHES")
print("-" * 70)

print(
    "Train batches:",
    len(train_loader)
)

print(
    "Val batches  :",
    len(val_loader)
)

# ------------------------------------------------------------
# MODEL
# ------------------------------------------------------------

encoder = SimpleConvTimeAgg(
    time_agg_out="att",
    dropout_input=0.2,
    conv_dropout=0.5,
    hidden=2048,
    batch_norm=True,
    depth=8,
    dilation_period=3,
    kernel_size=3,
    relu_leakiness=0.01,
    initial_linear=512,
    gelu=True,
    skip=True,
    scale=0.1,
    merger_config=None,
    subject_layers_config=None,
).build(
    n_in_channels=61,
    n_outputs=2048
)

classifier = nn.Linear(
    2048,
    29
)

model = nn.Sequential(
    encoder,
    classifier
).to(device)

criterion = nn.CrossEntropyLoss()

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=5e-5,
    weight_decay=1e-2
)

print("\nMODEL")
print("-" * 70)
print("Input channels : 61")
print("Samples/window : 25")
print("Embedding      : 2048")
print("Classes        : 29")
print("Merger         : disabled")
print("Device         :", device)

# ------------------------------------------------------------
# TRAINING
# ------------------------------------------------------------

EPOCHS = 10

best_val_loss = float("inf")
history = []

for epoch in range(EPOCHS):

    print("\n" + "=" * 70)
    print(f"EPOCH {epoch + 1}/{EPOCHS}")
    print("=" * 70)

    # --------------------------------------------------------
    # TRAIN
    # --------------------------------------------------------

    model.train()

    train_loss_sum = 0.0
    train_correct = 0
    train_total = 0

    for batch_idx, (X, y) in enumerate(
        train_loader
    ):

        X = X.to(
            device,
            non_blocking=True
        )

        y = y.to(
            device,
            non_blocking=True
        )

        optimizer.zero_grad(
            set_to_none=True
        )

        logits = model(X)

        loss = criterion(
            logits,
            y
        )

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=1.0
        )

        optimizer.step()

        train_loss_sum += (
            loss.item() *
            y.size(0)
        )

        train_correct += (
            logits.argmax(dim=1) == y
        ).sum().item()

        train_total += y.size(0)

        if (
            (batch_idx + 1) % 100 == 0
        ):

            print(
                f"Train batch "
                f"{batch_idx + 1}/"
                f"{len(train_loader)}",
                end="\r"
            )

    train_loss = (
        train_loss_sum /
        train_total
    )

    train_accuracy = (
        train_correct /
        train_total
    )

    # --------------------------------------------------------
    # VALIDATION
    # --------------------------------------------------------

    model.eval()

    val_loss_sum = 0.0
    val_correct = 0
    val_total = 0

    with torch.no_grad():

        for X, y in val_loader:

            X = X.to(
                device,
                non_blocking=True
            )

            y = y.to(
                device,
                non_blocking=True
            )

            logits = model(X)

            loss = criterion(
                logits,
                y
            )

            val_loss_sum += (
                loss.item() *
                y.size(0)
            )

            val_correct += (
                logits.argmax(dim=1) == y
            ).sum().item()

            val_total += y.size(0)

    val_loss = (
        val_loss_sum /
        val_total
    )

    val_accuracy = (
        val_correct /
        val_total
    )

    # --------------------------------------------------------
    # SAVE HISTORY
    # --------------------------------------------------------

    result = {
        "epoch": epoch + 1,
        "train_loss": train_loss,
        "train_accuracy": train_accuracy,
        "val_loss": val_loss,
        "val_accuracy": val_accuracy
    }

    history.append(result)

    print(
        f"\nTrain Loss: {train_loss:.6f}"
    )

    print(
        f"Train Acc : {train_accuracy:.6f}"
    )

    print(
        f"Val Loss  : {val_loss:.6f}"
    )

    print(
        f"Val Acc   : {val_accuracy:.6f}"
    )

    # --------------------------------------------------------
    # BEST CHECKPOINT
    # --------------------------------------------------------

    if val_loss < best_val_loss:

        best_val_loss = val_loss

        torch.save(
            {
                "phase": 5,
                "model_state_dict":
                    model.state_dict(),
                "optimizer_state_dict":
                    optimizer.state_dict(),
                "epoch":
                    epoch + 1,
                "val_loss":
                    val_loss,
                "val_accuracy":
                    val_accuracy,
                "input_channels": 61,
                "input_samples": 25,
                "num_classes": 29,
                "class_to_idx":
                    class_to_idx
            },
            BEST_MODEL_PATH
        )

        print(
            "✓ Best checkpoint saved"
        )

    # Save history after every epoch
    pd.DataFrame(history).to_csv(
        HISTORY_PATH,
        index=False
    )

# ------------------------------------------------------------
# COMPLETE
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PHASE 5C TRAINING COMPLETE")
print("=" * 70)

print(
    f"Best validation loss: "
    f"{best_val_loss:.6f}"
)

print(
    f"\nCheckpoint:\n"
    f"{BEST_MODEL_PATH}"
)

print(
    f"\nTraining history:\n"
    f"{HISTORY_PATH}"
)

print("=" * 70)
```

```python
# ============================================================
# PHASE 5D — TEST-SET EVALUATION
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

from neuraltrain.models.simpleconv import SimpleConvTimeAgg

PROJECT_DIR = "/content/drive/MyDrive/NeuroSentinel"

SPLIT_MANIFEST = os.path.join(
    PROJECT_DIR,
    "07_phase6_subject_split",
    "phase6_subject_split_manifest.csv"
)

RECORDINGS_DIR = os.path.join(
    PROJECT_DIR,
    "06_phase5_utility_labels",
    "recordings"
)

CHECKPOINT_PATH = os.path.join(
    PROJECT_DIR,
    "checkpoints",
    "phase5_decoder_best.pt"
)

RESULT_DIR = os.path.join(
    PROJECT_DIR,
    "results"
)

os.makedirs(RESULT_DIR, exist_ok=True)

RESULT_PATH = os.path.join(
    RESULT_DIR,
    "phase5_decoder_test_results.json"
)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("=" * 70)
print("PHASE 5D — TEST-SET EVALUATION")
print("=" * 70)
print("Device:", device)

# ------------------------------------------------------------
# Frozen test split
# ------------------------------------------------------------

split_df = pd.read_csv(SPLIT_MANIFEST)

test_records = split_df[
    split_df["split"] == "test"
].copy()

print("\nTEST SET")
print("-" * 70)
print("Recordings:", len(test_records))
print(
    "Subjects:",
    sorted(test_records["subject"].unique())
)

# ------------------------------------------------------------
# Mapping
# ------------------------------------------------------------

class_to_idx = {
    "s": 0, "o": 1, "t": 2, "e": 3, "n": 4,
    "c": 5, "i": 6, "a": 7, "<space>": 8,
    "d": 9, "l": 10, "r": 11, "b": 12,
    "<special>": 13, "z": 14, "v": 15,
    "f": 16, "m": 17, "u": 18, "h": 19,
    "p": 20, "g": 21, "q": 22, "w": 23,
    "x": 24, "y": 25, "j": 26, "k": 27,
    "<number>": 28
}

# ------------------------------------------------------------
# Test dataset
# ------------------------------------------------------------

class TestEEGDataset(Dataset):

    def __init__(self, records):

        self.samples = []
        self._cache = {}

        for _, row in records.iterrows():

            recording_id = str(
                row["recording_id"]
            )

            window_path = os.path.join(
                RECORDINGS_DIR,
                f"{recording_id}_windows.npy"
            )

            metadata_path = os.path.join(
                RECORDINGS_DIR,
                f"{recording_id}_metadata.csv"
            )

            if not os.path.exists(window_path):
                raise FileNotFoundError(
                    window_path
                )

            if not os.path.exists(metadata_path):
                raise FileNotFoundError(
                    metadata_path
                )

            meta = pd.read_csv(
                metadata_path
            )

            target_column = None

            for col in [
                "target",
                "button",
                "key",
                "character",
                "char"
            ]:

                if col in meta.columns:
                    target_column = col
                    break

            if target_column is None:
                raise ValueError(
                    f"No target column: {metadata_path}"
                )

            windows = np.load(
                window_path,
                mmap_mode="r"
            )

            if len(windows) != len(meta):
                raise ValueError(
                    f"Count mismatch: {recording_id}"
                )

            for idx, target in enumerate(
                meta[target_column].astype(str)
            ):

                if target in class_to_idx:

                    self.samples.append(
                        (
                            window_path,
                            idx,
                            class_to_idx[target]
                        )
                    )

            print(
                f"{recording_id}: "
                f"{len(windows):,} windows"
            )

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, index):

        path, idx, label = self.samples[index]

        if path not in self._cache:

            self._cache[path] = np.load(
                path,
                mmap_mode="r"
            )

        X = np.asarray(
            self._cache[path][idx],
            dtype=np.float32
        )

        return (
            torch.from_numpy(X),
            torch.tensor(
                label,
                dtype=torch.long
            )
        )


test_ds = TestEEGDataset(
    test_records
)

print("\nTotal test windows:", len(test_ds))

# ------------------------------------------------------------
# DataLoader
# ------------------------------------------------------------

test_loader = DataLoader(
    test_ds,
    batch_size=64,
    shuffle=False,
    num_workers=0,
    pin_memory=torch.cuda.is_available()
)

# ------------------------------------------------------------
# Rebuild model
# ------------------------------------------------------------

encoder = SimpleConvTimeAgg(
    time_agg_out="att",
    dropout_input=0.2,
    conv_dropout=0.5,
    hidden=2048,
    batch_norm=True,
    depth=8,
    dilation_period=3,
    kernel_size=3,
    relu_leakiness=0.01,
    initial_linear=512,
    gelu=True,
    skip=True,
    scale=0.1,
    merger_config=None,
    subject_layers_config=None,
).build(
    n_in_channels=61,
    n_outputs=2048
)

classifier = nn.Linear(
    2048,
    29
)

model = nn.Sequential(
    encoder,
    classifier
).to(device)

# ------------------------------------------------------------
# Load best checkpoint
# ------------------------------------------------------------

checkpoint = torch.load(
    CHECKPOINT_PATH,
    map_location=device,
    weights_only=False
)

model.load_state_dict(
    checkpoint["model_state_dict"]
)

model.eval()

print("\nCHECKPOINT")
print("-" * 70)
print("Training epoch:", checkpoint["epoch"])
print(
    "Validation loss:",
    checkpoint["val_loss"]
)
print(
    "Validation accuracy:",
    checkpoint["val_accuracy"]
)
print("✓ Best checkpoint loaded")

# ------------------------------------------------------------
# Evaluation
# ------------------------------------------------------------

criterion = nn.CrossEntropyLoss()

total_loss = 0.0
total_correct = 0
total_samples = 0

all_targets = []
all_predictions = []
all_confidences = []

with torch.no_grad():

    for X, y in test_loader:

        X = X.to(
            device,
            non_blocking=True
        )

        y = y.to(
            device,
            non_blocking=True
        )

        logits = model(X)

        loss = criterion(
            logits,
            y
        )

        probabilities = torch.softmax(
            logits,
            dim=1
        )

        confidence, predictions = (
            probabilities.max(dim=1)
        )

        total_loss += (
            loss.item() *
            y.size(0)
        )

        total_correct += (
            predictions == y
        ).sum().item()

        total_samples += y.size(0)

        all_targets.extend(
            y.cpu().numpy()
        )

        all_predictions.extend(
            predictions.cpu().numpy()
        )

        all_confidences.extend(
            confidence.cpu().numpy()
        )

# ------------------------------------------------------------
# Metrics
# ------------------------------------------------------------

test_loss = (
    total_loss /
    total_samples
)

test_accuracy = (
    total_correct /
    total_samples
)

all_targets = np.asarray(
    all_targets
)

all_predictions = np.asarray(
    all_predictions
)

all_confidences = np.asarray(
    all_confidences
)

correct_mask = (
    all_predictions ==
    all_targets
)

print("\n" + "=" * 70)
print("TEST RESULTS")
print("=" * 70)

print(
    f"Test windows : {total_samples:,}"
)

print(
    f"Test loss    : {test_loss:.6f}"
)

print(
    f"Test accuracy: {test_accuracy:.6f}"
)

print(
    f"Correct      : {correct_mask.sum():,}"
)

print(
    f"Incorrect    : {(~correct_mask).sum():,}"
)

print(
    f"Mean confidence: "
    f"{all_confidences.mean():.6f}"
)

print(
    f"Median confidence: "
    f"{np.median(all_confidences):.6f}"
)

print(
    f"Correct confidence mean: "
    f"{all_confidences[correct_mask].mean():.6f}"
    if correct_mask.any()
    else "Correct confidence mean: N/A"
)

print(
    f"Incorrect confidence mean: "
    f"{all_confidences[~correct_mask].mean():.6f}"
    if (~correct_mask).any()
    else "Incorrect confidence mean: N/A"
)

# ------------------------------------------------------------
# Prediction distribution
# ------------------------------------------------------------

print("\nPREDICTION DISTRIBUTION")
print("-" * 70)

pred_counts = pd.Series(
    all_predictions
).value_counts().sort_index()

for class_id, count in pred_counts.items():

    print(
        f"Class {int(class_id):2d}: "
        f"{int(count):6d}"
    )

# ------------------------------------------------------------
# Save results
# ------------------------------------------------------------

results = {
    "phase": 5,
    "stage": "test_evaluation",
    "test_windows": int(total_samples),
    "test_loss": float(test_loss),
    "test_accuracy": float(test_accuracy),
    "correct": int(correct_mask.sum()),
    "incorrect": int((~correct_mask).sum()),
    "mean_confidence": float(
        all_confidences.mean()
    ),
    "median_confidence": float(
        np.median(all_confidences)
    ),
    "checkpoint": CHECKPOINT_PATH,
    "checkpoint_epoch": int(
        checkpoint["epoch"]
    )
}

with open(
    RESULT_PATH,
    "w"
) as f:

    json.dump(
        results,
        f,
        indent=2
    )

print("\nSaved:")
print(RESULT_PATH)

print("\n" + "=" * 70)
print("PHASE 5D COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# PHASE 5E — DECODER COLLAPSE DIAGNOSIS
# ============================================================

import os
import numpy as np
import pandas as pd
import torch

PROJECT_DIR = "/content/drive/MyDrive/NeuroSentinel"

SPLIT_MANIFEST = os.path.join(
    PROJECT_DIR,
    "07_phase6_subject_split",
    "phase6_subject_split_manifest.csv"
)

RECORDINGS_DIR = os.path.join(
    PROJECT_DIR,
    "06_phase5_utility_labels",
    "recordings"
)

print("=" * 70)
print("PHASE 5E — DECODER COLLAPSE DIAGNOSIS")
print("=" * 70)

# ------------------------------------------------------------
# Brain2Qwerty mapping
# ------------------------------------------------------------

class_to_idx = {
    "s": 0, "o": 1, "t": 2, "e": 3, "n": 4,
    "c": 5, "i": 6, "a": 7, "<space>": 8,
    "d": 9, "l": 10, "r": 11, "b": 12,
    "<special>": 13, "z": 14, "v": 15,
    "f": 16, "m": 17, "u": 18, "h": 19,
    "p": 20, "g": 21, "q": 22, "w": 23,
    "x": 24, "y": 25, "j": 26, "k": 27,
    "<number>": 28
}

idx_to_class = {
    v: k for k, v in class_to_idx.items()
}

# ------------------------------------------------------------
# Load frozen split
# ------------------------------------------------------------

split_df = pd.read_csv(
    SPLIT_MANIFEST
)

train_records = split_df[
    split_df["split"] == "train"
]

val_records = split_df[
    split_df["split"] == "val"
]

test_records = split_df[
    split_df["split"] == "test"
]

# ------------------------------------------------------------
# Analyze labels
# ------------------------------------------------------------

def analyze_labels(records, name):

    counts = np.zeros(
        29,
        dtype=np.int64
    )

    total = 0

    print("\n" + "=" * 70)
    print(f"{name} LABEL DISTRIBUTION")
    print("=" * 70)

    for _, row in records.iterrows():

        recording_id = str(
            row["recording_id"]
        )

        metadata_path = os.path.join(
            RECORDINGS_DIR,
            f"{recording_id}_metadata.csv"
        )

        meta = pd.read_csv(
            metadata_path
        )

        target_column = None

        for col in [
            "target",
            "button",
            "key",
            "character",
            "char"
        ]:

            if col in meta.columns:
                target_column = col
                break

        if target_column is None:
            raise ValueError(
                f"No target column: {metadata_path}"
            )

        for value in meta[
            target_column
        ].astype(str):

            if value in class_to_idx:

                counts[
                    class_to_idx[value]
                ] += 1

                total += 1

    print(
        f"Total valid labels: {total:,}"
    )

    print("\nClass counts:")

    for class_id in range(29):

        count = counts[class_id]

        if count > 0:

            percentage = (
                100.0 * count / total
            )

            print(
                f"{class_id:2d} "
                f"{idx_to_class[class_id]:12s} "
                f"{int(count):7d} "
                f"({percentage:6.2f}%)"
            )

    print("\nMissing classes:")

    missing = [
        i for i in range(29)
        if counts[i] == 0
    ]

    print(
        missing
    )

    majority_class = int(
        np.argmax(counts)
    )

    majority_count = counts[
        majority_class
    ]

    majority_percentage = (
        100.0 *
        majority_count /
        total
    )

    print("\nMajority class:")
    print(
        f"{majority_class} "
        f"({idx_to_class[majority_class]})"
    )

    print(
        f"Majority percentage: "
        f"{majority_percentage:.2f}%"
    )

    return counts


train_counts = analyze_labels(
    train_records,
    "TRAIN"
)

val_counts = analyze_labels(
    val_records,
    "VALIDATION"
)

test_counts = analyze_labels(
    test_records,
    "TEST"
)

# ------------------------------------------------------------
# Compare predicted class against true majority class
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("COLLAPSE CHECK")
print("=" * 70)

predicted_class = 8

print(
    "Model predicted class:",
    predicted_class,
    f"({idx_to_class[predicted_class]})"
)

print(
    "Train frequency of class 8:",
    int(train_counts[8])
)

print(
    "Train percentage of class 8:",
    f"{100 * train_counts[8] / train_counts.sum():.2f}%"
)

print(
    "Validation frequency of class 8:",
    int(val_counts[8])
)

print(
    "Validation percentage of class 8:",
    f"{100 * val_counts[8] / val_counts.sum():.2f}%"
)

print(
    "Test frequency of class 8:",
    int(test_counts[8])
)

print(
    "Test percentage of class 8:",
    f"{100 * test_counts[8] / test_counts.sum():.2f}%"
)

# ------------------------------------------------------------
# Input amplitude diagnosis
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("EEG INPUT SCALE CHECK")
print("=" * 70)

sample_stats = []

for _, row in train_records.head(10).iterrows():

    recording_id = str(
        row["recording_id"]
    )

    window_path = os.path.join(
        RECORDINGS_DIR,
        f"{recording_id}_windows.npy"
    )

    windows = np.load(
        window_path,
        mmap_mode="r"
    )

    # Sample at most 100 windows
    n = min(
        100,
        len(windows)
    )

    X = np.asarray(
        windows[:n],
        dtype=np.float32
    )

    sample_stats.append({
        "recording": recording_id,
        "mean": float(X.mean()),
        "std": float(X.std()),
        "min": float(X.min()),
        "max": float(X.max())
    })

stats_df = pd.DataFrame(
    sample_stats
)

print(stats_df.to_string(
    index=False
))

print("\nOverall sampled statistics:")
print(
    "Mean:",
    stats_df["mean"].mean()
)

print(
    "Std :",
    stats_df["std"].mean()
)

# ------------------------------------------------------------
# Final diagnosis
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PHASE 5E DIAGNOSTIC SUMMARY")
print("=" * 70)

if (
    train_counts[8] /
    train_counts.sum()
) < 0.50:

    print(
        "✓ Class 8 is NOT the majority class."
    )

    print(
        "→ All-space prediction is likely "
        "not explained by simple majority imbalance."
    )

else:

    print(
        "⚠ Class 8 is heavily dominant."
    )

print(
    "\nNo model parameters were changed."
)

print(
    "No split was changed."
)

print(
    "No test data was used for training."
)

print("\nPHASE 5E COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# PHASE 5F — EEG NORMALIZATION EXPERIMENT
# ============================================================

import os
import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from torch.utils.data import Dataset, DataLoader
from neuraltrain.models.simpleconv import SimpleConvTimeAgg

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

PROJECT_DIR = "/content/drive/MyDrive/NeuroSentinel"

SPLIT_MANIFEST = os.path.join(
    PROJECT_DIR,
    "07_phase6_subject_split",
    "phase6_subject_split_manifest.csv"
)

RECORDINGS_DIR = os.path.join(
    PROJECT_DIR,
    "06_phase5_utility_labels",
    "recordings"
)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("=" * 70)
print("PHASE 5F — EEG NORMALIZATION EXPERIMENT")
print("=" * 70)
print("Device:", device)

# ------------------------------------------------------------
# Mapping
# ------------------------------------------------------------

class_to_idx = {
    "s": 0, "o": 1, "t": 2, "e": 3, "n": 4,
    "c": 5, "i": 6, "a": 7, "<space>": 8,
    "d": 9, "l": 10, "r": 11, "b": 12,
    "<special>": 13, "z": 14, "v": 15,
    "f": 16, "m": 17, "u": 18, "h": 19,
    "p": 20, "g": 21, "q": 22, "w": 23,
    "x": 24, "y": 25, "j": 26, "k": 27,
    "<number>": 28
}

# ------------------------------------------------------------
# Frozen split
# ------------------------------------------------------------

split_df = pd.read_csv(
    SPLIT_MANIFEST
)

train_records = split_df[
    split_df["split"] == "train"
].copy()

val_records = split_df[
    split_df["split"] == "val"
].copy()

print("\nFROZEN SPLIT")
print("-" * 70)
print("Train recordings:", len(train_records))
print("Val recordings  :", len(val_records))

# ------------------------------------------------------------
# Dataset
# ------------------------------------------------------------

class NormalizedEEGDataset(Dataset):

    def __init__(self, records):

        self.samples = []
        self._cache = {}

        for _, row in records.iterrows():

            recording_id = str(
                row["recording_id"]
            )

            window_path = os.path.join(
                RECORDINGS_DIR,
                f"{recording_id}_windows.npy"
            )

            metadata_path = os.path.join(
                RECORDINGS_DIR,
                f"{recording_id}_metadata.csv"
            )

            if not os.path.exists(window_path):
                raise FileNotFoundError(
                    window_path
                )

            if not os.path.exists(metadata_path):
                raise FileNotFoundError(
                    metadata_path
                )

            meta = pd.read_csv(
                metadata_path
            )

            target_column = None

            for col in [
                "target",
                "button",
                "key",
                "character",
                "char"
            ]:

                if col in meta.columns:
                    target_column = col
                    break

            if target_column is None:
                raise ValueError(
                    f"No target column: "
                    f"{metadata_path}"
                )

            windows = np.load(
                window_path,
                mmap_mode="r"
            )

            if len(windows) != len(meta):
                raise ValueError(
                    f"Count mismatch: "
                    f"{recording_id}"
                )

            for idx, target in enumerate(
                meta[target_column].astype(str)
            ):

                if target in class_to_idx:

                    self.samples.append(
                        (
                            window_path,
                            idx,
                            class_to_idx[target]
                        )
                    )

        print(
            f"Indexed {len(self.samples):,} samples"
        )

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, index):

        path, idx, label = (
            self.samples[index]
        )

        if path not in self._cache:

            self._cache[path] = np.load(
                path,
                mmap_mode="r"
            )

        X = np.asarray(
            self._cache[path][idx],
            dtype=np.float32
        )

        # ----------------------------------------------------
        # Channel-wise normalization
        # X shape = (61, 25)
        # ----------------------------------------------------

        mean = X.mean(
            axis=1,
            keepdims=True
        )

        std = X.std(
            axis=1,
            keepdims=True
        )

        X = (
            X - mean
        ) / (
            std + 1e-6
        )

        return (
            torch.from_numpy(
                X.astype(np.float32)
            ),
            torch.tensor(
                label,
                dtype=torch.long
            )
        )


print("\nBUILDING DATASETS")
print("-" * 70)

train_ds = NormalizedEEGDataset(
    train_records
)

val_ds = NormalizedEEGDataset(
    val_records
)

# ------------------------------------------------------------
# DataLoaders
# ------------------------------------------------------------

BATCH_SIZE = 64

train_loader = DataLoader(
    train_ds,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=0,
    pin_memory=torch.cuda.is_available()
)

val_loader = DataLoader(
    val_ds,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0,
    pin_memory=torch.cuda.is_available()
)

print("\nDATASET")
print("-" * 70)
print("Train:", len(train_ds))
print("Val  :", len(val_ds))

# ------------------------------------------------------------
# Model
# ------------------------------------------------------------

encoder = SimpleConvTimeAgg(
    time_agg_out="att",
    dropout_input=0.2,
    conv_dropout=0.5,
    hidden=2048,
    batch_norm=True,
    depth=8,
    dilation_period=3,
    kernel_size=3,
    relu_leakiness=0.01,
    initial_linear=512,
    gelu=True,
    skip=True,
    scale=0.1,
    merger_config=None,
    subject_layers_config=None,
).build(
    n_in_channels=61,
    n_outputs=2048
)

classifier = nn.Linear(
    2048,
    29
)

model = nn.Sequential(
    encoder,
    classifier
).to(device)

criterion = nn.CrossEntropyLoss()

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=5e-5,
    weight_decay=1e-2
)

print("\nMODEL")
print("-" * 70)
print("Input : 61 × 25")
print("Output: 29 classes")
print("Normalization: channel-wise per window")

# ------------------------------------------------------------
# Short controlled run
# ------------------------------------------------------------

EPOCHS = 3

for epoch in range(EPOCHS):

    # --------------------------------------------------------
    # Training
    # --------------------------------------------------------

    model.train()

    train_loss_sum = 0.0
    train_correct = 0
    train_total = 0

    for X, y in train_loader:

        X = X.to(
            device,
            non_blocking=True
        )

        y = y.to(
            device,
            non_blocking=True
        )

        optimizer.zero_grad(
            set_to_none=True
        )

        logits = model(X)

        loss = criterion(
            logits,
            y
        )

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            1.0
        )

        optimizer.step()

        train_loss_sum += (
            loss.item() *
            y.size(0)
        )

        train_correct += (
            logits.argmax(1) == y
        ).sum().item()

        train_total += y.size(0)

    train_loss = (
        train_loss_sum /
        train_total
    )

    train_acc = (
        train_correct /
        train_total
    )

    # --------------------------------------------------------
    # Validation
    # --------------------------------------------------------

    model.eval()

    val_loss_sum = 0.0
    val_correct = 0
    val_total = 0

    prediction_counts = np.zeros(
        29,
        dtype=np.int64
    )

    with torch.no_grad():

        for X, y in val_loader:

            X = X.to(
                device,
                non_blocking=True
            )

            y = y.to(
                device,
                non_blocking=True
            )

            logits = model(X)

            loss = criterion(
                logits,
                y
            )

            predictions = logits.argmax(
                dim=1
            )

            val_loss_sum += (
                loss.item() *
                y.size(0)
            )

            val_correct += (
                predictions == y
            ).sum().item()

            val_total += y.size(0)

            values, counts = torch.unique(
                predictions,
                return_counts=True
            )

            for value, count in zip(
                values.cpu().numpy(),
                counts.cpu().numpy()
            ):

                prediction_counts[
                    int(value)
                ] += int(count)

    val_loss = (
        val_loss_sum /
        val_total
    )

    val_acc = (
        val_correct /
        val_total
    )

    print("\n" + "-" * 70)

    print(
        f"Epoch {epoch + 1}/{EPOCHS}"
    )

    print(
        f"Train Loss: {train_loss:.6f}"
    )

    print(
        f"Train Acc : {train_acc:.6f}"
    )

    print(
        f"Val Loss  : {val_loss:.6f}"
    )

    print(
        f"Val Acc   : {val_acc:.6f}"
    )

    print(
        "Predicted classes:",
        np.where(
            prediction_counts > 0
        )[0].tolist()
    )

    print(
        "Most predicted class:",
        int(
            prediction_counts.argmax()
        )
    )

    print(
        "Most predicted count:",
        int(
            prediction_counts.max()
        )
    )

# ------------------------------------------------------------
# Complete
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PHASE 5F COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# PHASE 5F — HELD-OUT TEST EVALUATION
# ============================================================

import os
import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from neuraltrain.models.simpleconv import SimpleConvTimeAgg

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

PROJECT_DIR = "/content/gdrive/MyDrive/NeuroSentinel"

CHECKPOINT_PATH = os.path.join(
    PROJECT_DIR,
    "checkpoints",
    "phase5f_normalization_progress.pt"
)

SPLIT_MANIFEST = os.path.join(
    PROJECT_DIR,
    "07_phase6_subject_split",
    "phase6_subject_split_manifest.csv"
)

RECORDINGS_DIR = os.path.join(
    PROJECT_DIR,
    "06_phase5_utility_labels",
    "recordings"
)

RESULTS_DIR = os.path.join(
    PROJECT_DIR,
    "results"
)

os.makedirs(RESULTS_DIR, exist_ok=True)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("=" * 70)
print("PHASE 5F — HELD-OUT TEST EVALUATION")
print("=" * 70)
print("Device:", device)

# ------------------------------------------------------------
# Class mapping
# ------------------------------------------------------------

class_to_idx = {
    "s": 0,
    "o": 1,
    "t": 2,
    "e": 3,
    "n": 4,
    "c": 5,
    "i": 6,
    "a": 7,
    "<space>": 8,
    "d": 9,
    "l": 10,
    "r": 11,
    "b": 12,
    "<special>": 13,
    "z": 14,
    "v": 15,
    "f": 16,
    "m": 17,
    "u": 18,
    "h": 19,
    "p": 20,
    "g": 21,
    "q": 22,
    "w": 23,
    "x": 24,
    "y": 25,
    "j": 26,
    "k": 27,
    "<number>": 28
}

idx_to_class = {
    v: k for k, v in class_to_idx.items()
}

# ------------------------------------------------------------
# Load frozen split
# ------------------------------------------------------------

split_df = pd.read_csv(
    SPLIT_MANIFEST
)

test_records = split_df[
    split_df["split"] == "test"
].copy()

print("\nFROZEN TEST SPLIT")
print("-" * 70)
print("Test recordings:", len(test_records))
print(
    "Test subjects:",
    sorted(
        test_records["subject"]
        .unique()
        .tolist()
    )
)

# ------------------------------------------------------------
# Load checkpoint
# ------------------------------------------------------------

print("\nLOADING CHECKPOINT")
print("-" * 70)

checkpoint = torch.load(
    CHECKPOINT_PATH,
    map_location=device,
    weights_only=False
)

# ------------------------------------------------------------
# Rebuild model
# ------------------------------------------------------------

encoder = SimpleConvTimeAgg(
    time_agg_out="att",
    dropout_input=0.2,
    conv_dropout=0.5,
    hidden=2048,
    batch_norm=True,
    depth=8,
    dilation_period=3,
    kernel_size=3,
    relu_leakiness=0.01,
    initial_linear=512,
    gelu=True,
    skip=True,
    scale=0.1,
    merger_config=None,
    subject_layers_config=None,
).build(
    n_in_channels=61,
    n_outputs=2048
)

classifier = nn.Linear(
    2048,
    29
)

model = nn.Sequential(
    encoder,
    classifier
).to(device)

model.load_state_dict(
    checkpoint["model_state_dict"]
)

model.eval()

print("✓ Normalized Phase 5F model loaded")

# ------------------------------------------------------------
# Test inference
# ------------------------------------------------------------

all_results = []

total_correct = 0
total_samples = 0
total_loss = 0.0

criterion = nn.CrossEntropyLoss(
    reduction="sum"
)

print("\nRUNNING TEST INFERENCE")
print("-" * 70)

for record_number, (_, row) in enumerate(
    test_records.iterrows(),
    start=1
):

    recording_id = str(
        row["recording_id"]
    )

    subject = row["subject"]

    window_path = os.path.join(
        RECORDINGS_DIR,
        f"{recording_id}_windows.npy"
    )

    metadata_path = os.path.join(
        RECORDINGS_DIR,
        f"{recording_id}_metadata.csv"
    )

    if not os.path.exists(window_path):
        raise FileNotFoundError(
            window_path
        )

    if not os.path.exists(metadata_path):
        raise FileNotFoundError(
            metadata_path
        )

    windows = np.load(
        window_path,
        mmap_mode="r"
    )

    metadata = pd.read_csv(
        metadata_path
    )

    # Find target column
    target_column = None

    for col in [
        "target",
        "button",
        "key",
        "character",
        "char"
    ]:
        if col in metadata.columns:
            target_column = col
            break

    if target_column is None:
        raise ValueError(
            f"No target column in {metadata_path}"
        )

    if len(windows) != len(metadata):
        raise ValueError(
            f"Window/metadata mismatch: "
            f"{recording_id}"
        )

    valid_indices = []

    targets = []

    for idx, target in enumerate(
        metadata[target_column].astype(str)
    ):

        if target in class_to_idx:

            valid_indices.append(idx)
            targets.append(
                class_to_idx[target]
            )

    targets = np.asarray(
        targets,
        dtype=np.int64
    )

    # --------------------------------------------------------
    # Batched inference
    # --------------------------------------------------------

    batch_predictions = []
    batch_confidences = []
    batch_targets = []
    batch_losses = []

    BATCH_SIZE = 64

    for start in range(
        0,
        len(valid_indices),
        BATCH_SIZE
    ):

        batch_indices = valid_indices[
            start:start + BATCH_SIZE
        ]

        X = np.asarray(
            windows[batch_indices],
            dtype=np.float32
        )

        # ----------------------------------------------------
        # SAME NORMALIZATION USED DURING PHASE 5F TRAINING
        # ----------------------------------------------------

        mean = X.mean(
            axis=2,
            keepdims=True
        )

        std = X.std(
            axis=2,
            keepdims=True
        )

        X = (
            X - mean
        ) / (
            std + 1e-6
        )

        X = torch.from_numpy(
            X.astype(np.float32)
        ).to(
            device,
            non_blocking=True
        )

        y = torch.tensor(
            targets[
                start:start + len(batch_indices)
            ],
            dtype=torch.long,
            device=device
        )

        with torch.no_grad():

            logits = model(X)

            loss = criterion(
                logits,
                y
            )

            probabilities = torch.softmax(
                logits,
                dim=1
            )

            confidence, predictions = (
                probabilities.max(dim=1)
            )

        batch_predictions.extend(
            predictions.cpu().numpy().tolist()
        )

        batch_confidences.extend(
            confidence.cpu().numpy().tolist()
        )

        batch_targets.extend(
            y.cpu().numpy().tolist()
        )

        batch_losses.append(
            loss.item()
        )

    predictions = np.asarray(
        batch_predictions,
        dtype=np.int64
    )

    confidences = np.asarray(
        batch_confidences,
        dtype=np.float32
    )

    targets_valid = np.asarray(
        batch_targets,
        dtype=np.int64
    )

    correctness = (
        predictions == targets_valid
    )

    correct_count = int(
        correctness.sum()
    )

    sample_count = len(
        targets_valid
    )

    total_correct += correct_count
    total_samples += sample_count
    total_loss += sum(
        batch_losses
    )

    # --------------------------------------------------------
    # Build per-window records
    # --------------------------------------------------------

    for local_index, original_index in enumerate(
        valid_indices
    ):

        meta_row = metadata.iloc[
            original_index
        ]

        result = {
            "recording_id":
                recording_id,

            "subject":
                subject,

            "metadata_index":
                int(original_index),

            "target_id":
                int(targets_valid[local_index]),

            "target":
                idx_to_class[
                    int(targets_valid[local_index])
                ],

            "prediction_id":
                int(predictions[local_index]),

            "prediction":
                idx_to_class[
                    int(predictions[local_index])
                ],

            "confidence":
                float(confidences[local_index]),

            "correct":
                int(correctness[local_index])
        }

        # Preserve useful metadata fields
        for col in [
            "timestamp",
            "start",
            "trial_id",
            "keystroke_index",
            "event_index",
            "sample",
            "button"
        ]:

            if col in metadata.columns:
                result[col] = meta_row[col]

        all_results.append(
            result
        )

    print(
        f"{record_number:02d}/{len(test_records)} "
        f"{recording_id:<25} "
        f"windows={sample_count:,} "
        f"accuracy={correct_count / sample_count:.4f}"
    )

# ------------------------------------------------------------
# Final metrics
# ------------------------------------------------------------

results_df = pd.DataFrame(
    all_results
)

test_accuracy = (
    total_correct /
    total_samples
)

mean_confidence = (
    results_df["confidence"].mean()
)

median_confidence = (
    results_df["confidence"].median()
)

correct_confidence = (
    results_df.loc[
        results_df["correct"] == 1,
        "confidence"
    ].mean()
)

incorrect_confidence = (
    results_df.loc[
        results_df["correct"] == 0,
        "confidence"
    ].mean()
)

prediction_distribution = (
    results_df["prediction_id"]
    .value_counts()
    .sort_index()
)

print("\n" + "=" * 70)
print("PHASE 5F TEST RESULTS")
print("=" * 70)

print(
    "Test samples:",
    f"{total_samples:,}"
)

print(
    "Correct:",
    f"{total_correct:,}"
)

print(
    "Incorrect:",
    f"{total_samples - total_correct:,}"
)

print(
    "Accuracy:",
    f"{test_accuracy:.6f}"
)

print(
    "Mean confidence:",
    f"{mean_confidence:.6f}"
)

print(
    "Median confidence:",
    f"{median_confidence:.6f}"
)

print(
    "Correct confidence mean:",
    f"{correct_confidence:.6f}"
)

print(
    "Incorrect confidence mean:",
    f"{incorrect_confidence:.6f}"
)

print(
    "\nPredicted classes:",
    sorted(
        results_df["prediction_id"]
        .unique()
        .tolist()
    )
)

print("\nPrediction distribution:")

for class_id, count in prediction_distribution.items():

    print(
        f"{class_id:2d} "
        f"{idx_to_class[int(class_id)]:<10} "
        f"{int(count):6d} "
        f"({count / total_samples * 100:6.2f}%)"
    )

# ------------------------------------------------------------
# Save detailed predictions
# ------------------------------------------------------------

predictions_path = os.path.join(
    RESULTS_DIR,
    "phase5f_test_decoder_predictions.csv"
)

results_df.to_csv(
    predictions_path,
    index=False
)

# ------------------------------------------------------------
# Save summary
# ------------------------------------------------------------

summary = {
    "phase": "5F",
    "experiment":
        "normalized decoder test evaluation",

    "checkpoint":
        CHECKPOINT_PATH,

    "test_recordings":
        len(test_records),

    "test_subjects":
        sorted(
            test_records["subject"]
            .unique()
            .tolist()
        ),

    "test_samples":
        int(total_samples),

    "correct":
        int(total_correct),

    "incorrect":
        int(total_samples - total_correct),

    "accuracy":
        float(test_accuracy),

    "mean_confidence":
        float(mean_confidence),

    "median_confidence":
        float(median_confidence),

    "correct_confidence_mean":
        float(correct_confidence),

    "incorrect_confidence_mean":
        float(incorrect_confidence),

    "predicted_classes":
        sorted(
            results_df["prediction_id"]
            .unique()
            .tolist()
        ),

    "prediction_distribution":
        {
            str(int(k)): int(v)
            for k, v in prediction_distribution.items()
        },

    "prediction_file":
        predictions_path
}

summary_path = os.path.join(
    RESULTS_DIR,
    "phase5f_test_decoder_results.json"
)

import json

with open(
    summary_path,
    "w"
) as f:
    json.dump(
        summary,
        f,
        indent=2
    )

print("\nSaved:")
print(predictions_path)
print(summary_path)

print("\n" + "=" * 70)
print("TEST EVALUATION COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# PHASE 5G — EXACT PHASE 5F MODEL RECONSTRUCTION
# DIRECT CHECKPOINT LOAD — NO MANUAL SHAPE INSPECTION
# ============================================================

import torch
from pathlib import Path
from torch import nn
from neuraltrain.models import SimpleConvTimeAgg


PROJECT_DIR = Path(
    "/content/gdrive/MyDrive/NeuroSentinel"
)

CHECKPOINT = (
    PROJECT_DIR /
    "checkpoints" /
    "phase5f_normalization_progress.pt"
)

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)


print("=" * 70)
print("PHASE 5G — EXACT PHASE 5F MODEL RECONSTRUCTION")
print("=" * 70)


# ------------------------------------------------------------
# LOAD CHECKPOINT
# ------------------------------------------------------------

checkpoint = torch.load(
    CHECKPOINT,
    map_location="cpu",
    weights_only=False
)

saved_state = checkpoint[
    "model_state_dict"
]

print("\nCheckpoint:")
print("  epoch:", checkpoint["epoch_completed"])
print("  status:", checkpoint["status"])


# ------------------------------------------------------------
# REBUILD DECODER
# ------------------------------------------------------------

model_spec = SimpleConvTimeAgg(
    time_agg_out="att",
    dropout_input=0.2,
    conv_dropout=0.5,
    hidden=2048,
    batch_norm=True,
    depth=8,
    dilation_period=3,
    kernel_size=3,
    relu_leakiness=0.01,
    initial_linear=512,
    gelu=True,
    skip=True,
    scale=0.1,
    subject_layers_config=None,
    merger_config=None,
)


# IMPORTANT:
# The checkpoint proves that the SimpleConvTimeAgg output
# before the external classifier is 2048.
decoder = model_spec.build(
    n_in_channels=61,
    n_outputs=2048
)


print("\nDecoder type:")
print(type(decoder))

print("\nFinal encoder block:")
print(decoder.encoder.sequence[7])


# ------------------------------------------------------------
# INITIALIZE LAZY ATTENTION
# ------------------------------------------------------------

decoder = decoder.to(DEVICE)
decoder.eval()


dummy = torch.zeros(
    2,
    61,
    25,
    device=DEVICE
)


with torch.no_grad():

    dummy_representation = decoder(
        dummy
    )


print(
    "\nLazy attention initialized."
)

print(
    "Dummy representation:",
    tuple(
        dummy_representation.shape
    )
)


if tuple(
    dummy_representation.shape
) != (2, 2048):

    raise RuntimeError(
        "Unexpected decoder output shape."
    )


# ------------------------------------------------------------
# EXTRACT 0.* CHECKPOINT
# ------------------------------------------------------------

decoder_state = {

    key[len("0."):]: value

    for key, value in saved_state.items()

    if key.startswith("0.")
}


print("\nDecoder checkpoint entries:")
print(
    len(decoder_state)
)


# ------------------------------------------------------------
# DIRECT LOAD
# ------------------------------------------------------------
#
# Do NOT inspect model_state[key].shape here.
#
# PyTorch's load_state_dict() performs the authoritative
# parameter/buffer compatibility check.
# ------------------------------------------------------------

try:

    load_result = decoder.load_state_dict(
        decoder_state,
        strict=True
    )

except RuntimeError as e:

    print("\nDECODER LOAD FAILED")
    print("-" * 70)
    print(e)

    raise


print(
    "\n✓ Decoder checkpoint loaded successfully"
)


# ------------------------------------------------------------
# RESTORE EXTERNAL CLASSIFIER
# ------------------------------------------------------------

classifier_weight = saved_state[
    "1.weight"
]

classifier_bias = saved_state[
    "1.bias"
]


print("\n" + "=" * 70)
print("CLASSIFIER HEAD")
print("=" * 70)

print(
    "Weight shape:",
    tuple(
        classifier_weight.shape
    )
)

print(
    "Bias shape:",
    tuple(
        classifier_bias.shape
    )
)


classifier = nn.Linear(
    classifier_weight.shape[1],
    classifier_weight.shape[0]
)


classifier.load_state_dict({

    "weight":
        classifier_weight,

    "bias":
        classifier_bias

})


classifier = classifier.to(DEVICE)
classifier.eval()


print(
    "✓ 29-class classifier loaded"
)


# ------------------------------------------------------------
# COMPLETE MODEL TEST
# ------------------------------------------------------------

decoder.eval()
classifier.eval()


test_input = torch.randn(
    8,
    61,
    25,
    device=DEVICE
)


with torch.no_grad():

    representation = decoder(
        test_input
    )

    logits = classifier(
        representation
    )

    probabilities = torch.softmax(
        logits,
        dim=1
    )

    confidence, prediction = torch.max(
        probabilities,
        dim=1
    )


print("\n" + "=" * 70)
print("END-TO-END VERIFICATION")
print("=" * 70)

print(
    "Input:",
    tuple(test_input.shape)
)

print(
    "Representation:",
    tuple(
        representation.shape
    )
)

print(
    "Logits:",
    tuple(
        logits.shape
    )
)

print(
    "Probabilities:",
    tuple(
        probabilities.shape
    )
)

print(
    "Confidence:",
    tuple(
        confidence.shape
    )
)

print(
    "Prediction:",
    tuple(
        prediction.shape
    )
)


if tuple(
    representation.shape
) != (8, 2048):

    raise RuntimeError(
        "Decoder representation is not (8, 2048)."
    )


if tuple(
    logits.shape
) != (8, 29):

    raise RuntimeError(
        "Classifier logits are not (8, 29)."
    )


print(
    "\n✓ EXACT PHASE 5F MODEL RESTORED"
)

print(
    "✓ Decoder output: 2048"
)

print(
    "✓ Original LayerScale preserved"
)

print(
    "✓ Lazy attention initialized"
)

print(
    "✓ All decoder checkpoint parameters loaded"
)

print(
    "✓ 29-class classifier restored"
)

print(
    "✓ End-to-end 29-class inference verified"
)

print(
    "\nNo checkpoint files were modified."

)

print(
    "=" * 70
)
```

```python
# ============================================================
# PHASE 5G — VALIDATION CONFIDENCE THRESHOLD ANALYSIS
# USING VERIFIED PHASE 5F MODEL
# ============================================================

import json
import numpy as np
import pandas as pd
import torch
from pathlib import Path

print("=" * 70)
print("PHASE 5G — VALIDATION CONFIDENCE THRESHOLD ANALYSIS")
print("=" * 70)

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

PROJECT_DIR = Path(
    "/content/gdrive/MyDrive/NeuroSentinel"
)

RECORDINGS_DIR = (
    PROJECT_DIR /
    "06_phase5_utility_labels" /
    "recordings"
)

SPLIT_MANIFEST = (
    PROJECT_DIR /
    "07_phase6_subject_split" /
    "phase6_subject_split_manifest.csv"
)

RESULTS_DIR = PROJECT_DIR / "results"
RESULTS_DIR.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# DEVICE
# ------------------------------------------------------------

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

decoder = decoder.to(DEVICE)
classifier = classifier.to(DEVICE)

decoder.eval()
classifier.eval()

print("Device:", DEVICE)

# ------------------------------------------------------------
# BUTTON MAPPING
# ------------------------------------------------------------

BUTTON_MAPPING = {
    0: "s",
    1: "o",
    2: "t",
    3: "e",
    4: "n",
    5: "c",
    6: "i",
    7: "a",
    8: "<space>",
    9: "d",
    10: "l",
    11: "r",
    12: "b",
    13: "<special>",
    14: "z",
    15: "v",
    16: "f",
    17: "m",
    18: "u",
    19: "h",
    20: "p",
    21: "g",
    22: "q",
    23: "w",
    24: "x",
    25: "y",
    26: "j",
    27: "k",
    28: "<number>",
}

CHAR_TO_ID = {
    value: key
    for key, value in BUTTON_MAPPING.items()
}

# ------------------------------------------------------------
# TARGET CONVERSION
# ------------------------------------------------------------

TARGET_CANDIDATES = [
    "target",
    "button",
    "key",
    "character",
    "char"
]


def find_target_column(df):

    for col in TARGET_CANDIDATES:

        if col in df.columns:
            return col

    raise ValueError(
        "No target column found. "
        f"Columns: {list(df.columns)}"
    )


def target_to_id(value):

    if isinstance(
        value,
        (int, np.integer)
    ):

        value = int(value)

        if 0 <= value <= 28:
            return value

    if isinstance(value, float):

        if np.isfinite(value):

            ivalue = int(value)

            if (
                0 <= ivalue <= 28
                and value == ivalue
            ):
                return ivalue

    s = str(value).strip()

    if s in CHAR_TO_ID:
        return CHAR_TO_ID[s]

    normalized = s.lower()

    if normalized in [
        "space",
        "<space>",
        " "
    ]:
        return 8

    if normalized in [
        "special",
        "<special>"
    ]:
        return 13

    if normalized in [
        "number",
        "<number>"
    ]:
        return 28

    if (
        len(s) == 1
        and s in CHAR_TO_ID
    ):
        return CHAR_TO_ID[s]

    return None


# ------------------------------------------------------------
# LOAD FROZEN VALIDATION SPLIT
# ------------------------------------------------------------

split_df = pd.read_csv(
    SPLIT_MANIFEST
)

val_df = split_df[
    split_df["split"] == "val"
].copy()

val_recordings = (
    val_df["recording_id"]
    .astype(str)
    .tolist()
)

print("\nFROZEN VALIDATION SPLIT")
print("-" * 70)

print(
    "Validation recordings:",
    len(val_recordings)
)

print(
    "Validation subjects:",
    sorted(
        val_df["subject"]
        .astype(int)
        .unique()
        .tolist()
    )
)


# ------------------------------------------------------------
# VALIDATION INFERENCE
# ------------------------------------------------------------

all_confidence = []
all_correct = []

total_windows = 0
skipped_windows = 0

print("\nRUNNING VALIDATION INFERENCE")
print("-" * 70)


with torch.no_grad():

    for idx, recording_id in enumerate(
        val_recordings,
        start=1
    ):

        window_path = (
            RECORDINGS_DIR /
            f"{recording_id}_windows.npy"
        )

        metadata_path = (
            RECORDINGS_DIR /
            f"{recording_id}_metadata.csv"
        )

        X = np.load(
            window_path
        ).astype(np.float32)

        meta = pd.read_csv(
            metadata_path
        )

        target_col = find_target_column(
            meta
        )

        target_ids = np.array(
            [
                target_to_id(value)
                for value in meta[target_col]
            ],
            dtype=object
        )

        valid_mask = np.array(
            [
                value is not None
                for value in target_ids
            ]
        )

        skipped_here = int(
            (~valid_mask).sum()
        )

        skipped_windows += skipped_here

        X = X[valid_mask]

        y = np.array(
            [
                int(value)
                for value in target_ids[valid_mask]
            ],
            dtype=np.int64
        )

        total_windows += len(y)

        BATCH_SIZE = 256

        for start in range(
            0,
            len(X),
            BATCH_SIZE
        ):

            stop = min(
                start + BATCH_SIZE,
                len(X)
            )

            xb = X[start:stop]

            # ------------------------------------------------
            # EXACT PHASE 5F NORMALIZATION
            # ------------------------------------------------

            mean = xb.mean(
                axis=2,
                keepdims=True
            )

            std = xb.std(
                axis=2,
                keepdims=True
            )

            xb = (
                xb - mean
            ) / (
                std + 1e-6
            )

            xb = torch.from_numpy(
                xb
            ).to(DEVICE)

            # ------------------------------------------------
            # EXACT PHASE 5F DECODER
            # ------------------------------------------------

            representation = decoder(
                xb
            )

            logits = classifier(
                representation
            )

            probabilities = torch.softmax(
                logits,
                dim=1
            )

            confidence, prediction = torch.max(
                probabilities,
                dim=1
            )

            correct = (
                prediction.cpu().numpy()
                == y[start:stop]
            )

            all_confidence.extend(
                confidence
                .cpu()
                .numpy()
                .tolist()
            )

            all_correct.extend(
                correct.astype(np.int8)
                .tolist()
            )

        print(
            f"{idx:02d}/{len(val_recordings)} "
            f"{recording_id:<25} "
            f"windows={len(y):,}"
        )


# ------------------------------------------------------------
# ARRAYS
# ------------------------------------------------------------

confidence = np.asarray(
    all_confidence,
    dtype=np.float32
)

correct = np.asarray(
    all_correct,
    dtype=np.int8
)


# ------------------------------------------------------------
# SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("VALIDATION CONFIDENCE SUMMARY")
print("=" * 70)

print(
    "Validation samples:",
    f"{len(confidence):,}"
)

print(
    "Correct:",
    f"{int(correct.sum()):,}"
)

print(
    "Incorrect:",
    f"{int((1 - correct).sum()):,}"
)

print(
    "Skipped:",
    f"{skipped_windows:,}"
)

print(
    "Accuracy:",
    f"{correct.mean():.6f}"
)

print(
    "Mean confidence:",
    f"{confidence.mean():.6f}"
)

print(
    "Median confidence:",
    f"{np.median(confidence):.6f}"
)

print(
    "Correct confidence mean:",
    f"{confidence[correct == 1].mean():.6f}"
)

print(
    "Incorrect confidence mean:",
    f"{confidence[correct == 0].mean():.6f}"
)


# ------------------------------------------------------------
# THRESHOLD SWEEP
# ------------------------------------------------------------

thresholds = np.round(
    np.arange(
        0.05,
        0.96,
        0.05
    ),
    2
)

rows = []

for tau in thresholds:

    selected = (
        confidence >= tau
    )

    selected_count = int(
        selected.sum()
    )

    coverage = float(
        selected.mean()
    )

    if selected_count == 0:

        precision = np.nan
        recall = 0.0
        f1 = 0.0

    else:

        # Probability that a retained/high-confidence
        # prediction is actually correct.
        precision = float(
            correct[selected].mean()
        )

        # Fraction of all correct predictions retained.
        recall = float(
            correct[selected].sum()
            /
            max(
                int(correct.sum()),
                1
            )
        )

        if precision + recall > 0:

            f1 = float(
                2 * precision * recall
                /
                (precision + recall)
            )

        else:

            f1 = 0.0

    rows.append({

        "threshold":
            float(tau),

        "selected_samples":
            selected_count,

        "coverage":
            coverage,

        "precision_correct_among_selected":
            precision,

        "recall_of_correct_predictions":
            recall,

        "f1":
            f1
    })


threshold_df = pd.DataFrame(
    rows
)


# ------------------------------------------------------------
# DISPLAY
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("THRESHOLD SWEEP")
print("=" * 70)

print(
    threshold_df.to_string(
        index=False,
        float_format=lambda x:
            f"{x:.4f}"
    )
)


# ------------------------------------------------------------
# SAVE
# ------------------------------------------------------------

threshold_path = (
    RESULTS_DIR /
    "phase5g_validation_threshold_analysis.csv"
)

threshold_df.to_csv(
    threshold_path,
    index=False
)


summary = {

    "phase":
        "5G",

    "checkpoint":
        str(PROJECT_DIR /
            "checkpoints" /
            "phase5f_normalization_progress.pt"),

    "checkpoint_epoch":
        int(
            checkpoint[
                "epoch_completed"
            ]
        ),

    "validation_recordings":
        len(val_recordings),

    "validation_subjects":
        sorted(
            val_df["subject"]
            .astype(int)
            .unique()
            .tolist()
        ),

    "validation_samples":
        int(len(confidence)),

    "validation_correct":
        int(correct.sum()),

    "validation_incorrect":
        int(
            (1 - correct).sum()
        ),

    "validation_accuracy":
        float(correct.mean()),

    "mean_confidence":
        float(confidence.mean()),

    "median_confidence":
        float(np.median(confidence)),

    "correct_confidence_mean":
        float(
            confidence[
                correct == 1
            ].mean()
        ),

    "incorrect_confidence_mean":
        float(
            confidence[
                correct == 0
            ].mean()
        ),

    "skipped_windows":
        int(skipped_windows),

    "normalization":
        "channel-wise per-window z-score",

    "epsilon":
        1e-6
}


summary_path = (
    RESULTS_DIR /
    "phase5g_validation_threshold_summary.json"
)

with open(
    summary_path,
    "w"
) as f:

    json.dump(
        summary,
        f,
        indent=2
    )


print("\nSaved:")
print(threshold_path)
print(summary_path)

print("\n" + "=" * 70)
print("PHASE 5G COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# PHASE 5G.1 — CONFIDENCE DISCRIMINATION DIAGNOSTIC
# ============================================================

import numpy as np
import pandas as pd
import json
from pathlib import Path

print("=" * 70)
print("PHASE 5G.1 — CONFIDENCE DISCRIMINATION DIAGNOSTIC")
print("=" * 70)

# ------------------------------------------------------------
# Uses variables already produced by Phase 5G:
#
# confidence : validation confidence array
# correct    : validation correctness array
# ------------------------------------------------------------

assert len(confidence) == len(correct)

confidence = np.asarray(confidence, dtype=np.float64)
correct = np.asarray(correct, dtype=np.int8)

# ------------------------------------------------------------
# 1. CONFIDENCE QUANTILES
# ------------------------------------------------------------

quantiles = [
    0.00,
    0.10,
    0.25,
    0.50,
    0.75,
    0.90,
    0.95,
    0.99,
    1.00
]

print("\nCONFIDENCE QUANTILES")
print("-" * 70)

for q in quantiles:
    print(
        f"{q:>5.0%}: "
        f"{np.quantile(confidence, q):.6f}"
    )

# ------------------------------------------------------------
# 2. CORRECT VS INCORRECT DISTRIBUTION
# ------------------------------------------------------------

correct_conf = confidence[correct == 1]
incorrect_conf = confidence[correct == 0]

print("\nCORRECT VS INCORRECT")
print("-" * 70)

print(
    f"Correct samples:   {len(correct_conf):,}"
)

print(
    f"Incorrect samples: {len(incorrect_conf):,}"
)

print(
    f"Correct mean:      {correct_conf.mean():.6f}"
)

print(
    f"Incorrect mean:    {incorrect_conf.mean():.6f}"
)

print(
    f"Correct median:    {np.median(correct_conf):.6f}"
)

print(
    f"Incorrect median:  {np.median(incorrect_conf):.6f}"
)

# ------------------------------------------------------------
# 3. CONFIDENCE BINS
# ------------------------------------------------------------

bins = np.array([
    0.00,
    0.05,
    0.10,
    0.15,
    0.20,
    0.25,
    0.30,
    0.35,
    0.40,
    0.45,
    0.50,
    1.00
])

rows = []

for lower, upper in zip(
    bins[:-1],
    bins[1:]
):

    if upper < 1.0:
        mask = (
            (confidence >= lower) &
            (confidence < upper)
        )
    else:
        mask = (
            (confidence >= lower) &
            (confidence <= upper)
        )

    n = int(mask.sum())

    if n > 0:
        accuracy = float(
            correct[mask].mean()
        )
        mean_conf = float(
            confidence[mask].mean()
        )
    else:
        accuracy = np.nan
        mean_conf = np.nan

    rows.append({
        "confidence_lower": lower,
        "confidence_upper": upper,
        "samples": n,
        "coverage": n / len(confidence),
        "accuracy": accuracy,
        "mean_confidence": mean_conf
    })

bin_df = pd.DataFrame(rows)

print("\nCONFIDENCE BINS")
print("-" * 70)

print(
    bin_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.6f}"
    )
)

# ------------------------------------------------------------
# 4. SIMPLE RANK-DISCRIMINATION ANALYSIS
# ------------------------------------------------------------
#
# Interpretation:
# If confidence is useful, correct predictions should
# systematically rank above incorrect predictions.
#
# We calculate the probability that a randomly selected
# correct prediction has higher confidence than a randomly
# selected incorrect prediction.
#
# This is equivalent to ROC-AUC for the binary target
# "correct".
# ------------------------------------------------------------

try:

    from sklearn.metrics import roc_auc_score

    auc = float(
        roc_auc_score(
            correct,
            confidence
        )
    )

except Exception as e:

    auc = None

    print(
        "\nAUC calculation failed:",
        repr(e)
    )

print("\nCONFIDENCE DISCRIMINATION")
print("-" * 70)

if auc is not None:

    print(
        f"ROC-AUC: {auc:.6f}"
    )

    print(
        "\nReference:"
    )
    print(
        "  0.50 = no ranking discrimination"
    )
    print(
        "  >0.50 = higher confidence tends to be correct"
    )
    print(
        "  <0.50 = higher confidence tends to be incorrect"
    )

# ------------------------------------------------------------
# 5. TOP-CONFIDENCE RETENTION
# ------------------------------------------------------------

print("\nTOP-CONFIDENCE RETENTION")
print("-" * 70)

for fraction in [
    0.50,
    0.25,
    0.10,
    0.05,
    0.01
]:

    n = max(
        1,
        int(len(confidence) * fraction)
    )

    order = np.argsort(
        confidence
    )[::-1]

    selected = order[:n]

    selected_accuracy = float(
        correct[selected].mean()
    )

    threshold = float(
        confidence[selected[-1]]
    )

    print(
        f"Top {fraction:>5.0%}: "
        f"n={n:>6,} | "
        f"accuracy={selected_accuracy:.6f} | "
        f"minimum confidence={threshold:.6f}"
    )

# ------------------------------------------------------------
# 6. SAVE DIAGNOSTIC
# ------------------------------------------------------------

RESULTS_DIR = (
    Path("/content/gdrive/MyDrive/NeuroSentinel")
    / "results"
)

RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)

bin_path = (
    RESULTS_DIR /
    "phase5g_confidence_bins.csv"
)

bin_df.to_csv(
    bin_path,
    index=False
)

diagnostic = {
    "phase": "5G.1",

    "validation_samples":
        int(len(confidence)),

    "validation_accuracy":
        float(correct.mean()),

    "correct_mean_confidence":
        float(correct_conf.mean()),

    "incorrect_mean_confidence":
        float(incorrect_conf.mean()),

    "correct_median_confidence":
        float(np.median(correct_conf)),

    "incorrect_median_confidence":
        float(np.median(incorrect_conf)),

    "roc_auc":
        auc
}

diagnostic_path = (
    RESULTS_DIR /
    "phase5g_confidence_discrimination.json"
)

with open(
    diagnostic_path,
    "w"
) as f:

    json.dump(
        diagnostic,
        f,
        indent=2
    )

print("\nSaved:")
print(bin_path)
print(diagnostic_path)

print("\n" + "=" * 70)
print("PHASE 5G.1 COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# PHASE 5H.2 — SMALL-SUBSET OVERFIT DIAGNOSTIC
# ============================================================

import torch
import torch.nn as nn
import numpy as np
import pandas as pd
from pathlib import Path
from torch.utils.data import Dataset, DataLoader

print("=" * 70)
print("PHASE 5H.2 — SMALL-SUBSET OVERFIT DIAGNOSTIC")
print("=" * 70)

# ------------------------------------------------------------
# CONFIGURATION
# ------------------------------------------------------------

PROJECT_DIR = Path(
    "/content/gdrive/MyDrive/NeuroSentinel"
)

RECORDINGS_DIR = (
    PROJECT_DIR /
    "06_phase5_utility_labels" /
    "recordings"
)

SPLIT_MANIFEST = (
    PROJECT_DIR /
    "07_phase6_subject_split" /
    "phase6_subject_split_manifest.csv"
)

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

N_SAMPLES = 2000
BATCH_SIZE = 64
EPOCHS = 10
LEARNING_RATE = 5e-5
WEIGHT_DECAY = 1e-2
SEED = 42

torch.manual_seed(SEED)
np.random.seed(SEED)

print("Device:", DEVICE)
print("Subset size:", N_SAMPLES)
print("Epochs:", EPOCHS)

# ------------------------------------------------------------
# BUTTON MAPPING
# ------------------------------------------------------------

BUTTON_MAPPING = {
    0: "s",
    1: "o",
    2: "t",
    3: "e",
    4: "n",
    5: "c",
    6: "i",
    7: "a",
    8: "<space>",
    9: "d",
    10: "l",
    11: "r",
    12: "b",
    13: "<special>",
    14: "z",
    15: "v",
    16: "f",
    17: "m",
    18: "u",
    19: "h",
    20: "p",
    21: "g",
    22: "q",
    23: "w",
    24: "x",
    25: "y",
    26: "j",
    27: "k",
    28: "<number>",
}

CHAR_TO_ID = {
    value: key
    for key, value in BUTTON_MAPPING.items()
}


def target_to_id(value):

    if isinstance(
        value,
        (int, np.integer)
    ):

        value = int(value)

        if 0 <= value <= 28:
            return value

    if isinstance(value, float):

        if np.isfinite(value):

            ivalue = int(value)

            if (
                0 <= ivalue <= 28
                and value == ivalue
            ):
                return ivalue

    value = str(value).strip()

    if value in CHAR_TO_ID:
        return CHAR_TO_ID[value]

    if value.lower() in [
        "space",
        "<space>",
        " "
    ]:
        return 8

    if value.lower() in [
        "special",
        "<special>"
    ]:
        return 13

    if value.lower() in [
        "number",
        "<number>"
    ]:
        return 28

    return None


# ------------------------------------------------------------
# LOAD TRAINING RECORDINGS
# ------------------------------------------------------------

split_df = pd.read_csv(
    SPLIT_MANIFEST
)

train_recordings = (
    split_df[
        split_df["split"] == "train"
    ]["recording_id"]
    .astype(str)
    .tolist()
)

print(
    "\nTraining recordings:",
    len(train_recordings)
)

# ------------------------------------------------------------
# COLLECT 2000 TRAINING WINDOWS
# ------------------------------------------------------------

X_parts = []
y_parts = []

remaining = N_SAMPLES

for recording_id in train_recordings:

    if remaining <= 0:
        break

    window_path = (
        RECORDINGS_DIR /
        f"{recording_id}_windows.npy"
    )

    metadata_path = (
        RECORDINGS_DIR /
        f"{recording_id}_metadata.csv"
    )

    X = np.load(
        window_path
    ).astype(np.float32)

    metadata = pd.read_csv(
        metadata_path
    )

    target_col = None

    for candidate in [
        "target",
        "button",
        "key",
        "character",
        "char"
    ]:

        if candidate in metadata.columns:
            target_col = candidate
            break

    if target_col is None:
        continue

    targets = np.array([
        target_to_id(value)
        for value in metadata[target_col]
    ])

    valid = np.array([
        value is not None
        for value in targets
    ])

    valid_indices = np.where(valid)[0]

    take = min(
        remaining,
        len(valid_indices)
    )

    selected = valid_indices[:take]

    X_parts.append(
        X[selected]
    )

    y_parts.append(
        np.array(
            [
                int(targets[i])
                for i in selected
            ],
            dtype=np.int64
        )
    )

    remaining -= take

X_train = np.concatenate(
    X_parts,
    axis=0
)

y_train = np.concatenate(
    y_parts,
    axis=0
)

print(
    "Collected:",
    len(y_train),
    "windows"
)

print(
    "Input shape:",
    X_train.shape
)

print(
    "Target shape:",
    y_train.shape
)

# ------------------------------------------------------------
# FIXED NORMALIZATION
# ------------------------------------------------------------

mean = X_train.mean(
    axis=2,
    keepdims=True
)

std = X_train.std(
    axis=2,
    keepdims=True
)

X_train = (
    X_train - mean
) / (
    std + 1e-6
)

print(
    "\nNormalized mean:",
    f"{X_train.mean():.6f}"
)

print(
    "Normalized std:",
    f"{X_train.std():.6f}"
)

# ------------------------------------------------------------
# DATASET
# ------------------------------------------------------------

class OverfitDataset(Dataset):

    def __init__(
        self,
        X,
        y
    ):
        self.X = torch.from_numpy(X)
        self.y = torch.from_numpy(y)

    def __len__(self):
        return len(self.y)

    def __getitem__(self, idx):
        return (
            self.X[idx],
            self.y[idx]
        )


dataset = OverfitDataset(
    X_train,
    y_train
)

loader = DataLoader(
    dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=0
)

# ------------------------------------------------------------
# BUILD FRESH MODEL
# ------------------------------------------------------------

from neuraltrain.models.simpleconv import SimpleConvTimeAgg

model_spec = SimpleConvTimeAgg(
    time_agg_out="att",
    dropout_input=0.2,
    conv_dropout=0.5,
    hidden=2048,
    batch_norm=True,
    depth=8,
    dilation_period=3,
    kernel_size=3,
    relu_leakiness=0.01,
    initial_linear=512,
    gelu=True,
    skip=True,
    scale=0.1,
    subject_layers_config=None,
    merger_config=None,
)

decoder = model_spec.build(
    n_in_channels=61,
    n_outputs=2048
)

decoder = decoder.to(DEVICE)

# Initialize lazy attention
with torch.no_grad():

    dummy = torch.zeros(
        2,
        61,
        25,
        device=DEVICE
    )

    _ = decoder(dummy)

classifier = nn.Linear(
    2048,
    29
).to(DEVICE)

# ------------------------------------------------------------
# TRAINING
# ------------------------------------------------------------

optimizer = torch.optim.AdamW(
    list(decoder.parameters()) +
    list(classifier.parameters()),
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY
)

criterion = nn.CrossEntropyLoss()

history = []

print("\n" + "=" * 70)
print("OVERFIT TRAINING")
print("=" * 70)

for epoch in range(
    1,
    EPOCHS + 1
):

    decoder.train()
    classifier.train()

    running_loss = 0.0
    correct = 0
    total = 0

    for xb, yb in loader:

        xb = xb.to(
            DEVICE,
            non_blocking=True
        )

        yb = yb.to(
            DEVICE,
            non_blocking=True
        )

        optimizer.zero_grad(
            set_to_none=True
        )

        representation = decoder(
            xb
        )

        logits = classifier(
            representation
        )

        loss = criterion(
            logits,
            yb
        )

        loss.backward()

        optimizer.step()

        running_loss += (
            loss.item()
            * len(yb)
        )

        predictions = logits.argmax(
            dim=1
        )

        correct += int(
            (predictions == yb)
            .sum()
            .item()
        )

        total += len(yb)

    epoch_loss = (
        running_loss / total
    )

    epoch_accuracy = (
        correct / total
    )

    history.append({
        "epoch": epoch,
        "loss": epoch_loss,
        "accuracy": epoch_accuracy
    })

    print(
        f"Epoch {epoch:02d} | "
        f"Loss: {epoch_loss:.6f} | "
        f"Accuracy: {epoch_accuracy:.6f}"
    )

# ------------------------------------------------------------
# FINAL EVALUATION ON SAME 2000 SAMPLES
# ------------------------------------------------------------

decoder.eval()
classifier.eval()

all_predictions = []

with torch.no_grad():

    for xb, _ in DataLoader(
        dataset,
        batch_size=BATCH_SIZE,
        shuffle=False
    ):

        xb = xb.to(DEVICE)

        logits = classifier(
            decoder(xb)
        )

        predictions = (
            logits.argmax(dim=1)
            .cpu()
            .numpy()
        )

        all_predictions.extend(
            predictions.tolist()
        )

all_predictions = np.asarray(
    all_predictions
)

final_accuracy = float(
    (all_predictions == y_train)
    .mean()
)

print("\n" + "=" * 70)
print("FINAL OVERFIT RESULT")
print("=" * 70)

print(
    "Samples:",
    len(y_train)
)

print(
    "Final training accuracy:",
    f"{final_accuracy:.6f}"
)

print(
    "Final training loss:",
    f"{history[-1]['loss']:.6f}"
)

print(
    "\nPrediction classes:",
    sorted(
        np.unique(
            all_predictions
        ).tolist()
    )
)

print(
    "\nTarget classes:",
    sorted(
        np.unique(
            y_train
        ).tolist()
    )
)

print("\n" + "=" * 70)
print("PHASE 5H.2 COMPLETE")
print("=" * 70)

print(
    "\nNo existing checkpoint was modified."
)
```

```python
# ============================================================
# NEUROSENTINEL
# PRE-PHASE 5H.3 IMPLEMENTATION CHECKPOINT
#
# Purpose:
#   Save the complete validated project state before starting
#   Phase 5H.3 decoder training.
#
# IMPORTANT:
#   - Does NOT run training
#   - Does NOT overwrite existing checkpoints
#   - Records the current implementation status
#   - Records all important Phase 0–6 artifacts
# ============================================================

import os
import json
import time
import hashlib
from pathlib import Path


# ============================================================
# 1. PROJECT PATH
# ============================================================

PROJECT_DIR = Path(
    "/content/gdrive/MyDrive/NeuroSentinel"
)

CHECKPOINT_DIR = (
    PROJECT_DIR /
    "checkpoints"
)

RESULTS_DIR = (
    PROJECT_DIR /
    "results"
)

IMPLEMENTATION_DIR = (
    PROJECT_DIR /
    "implementation_checkpoints"
)

IMPLEMENTATION_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# 2. TIMESTAMP
# ============================================================

timestamp = time.strftime(
    "%Y%m%d_%H%M%S"
)

timestamp_readable = time.strftime(
    "%Y-%m-%d %H:%M:%S"
)


# ============================================================
# 3. CHECK REQUIRED PROJECT
# ============================================================

if not PROJECT_DIR.exists():

    raise FileNotFoundError(
        f"NeuroSentinel project not found:\n{PROJECT_DIR}"
    )


# ============================================================
# 4. FILE HASH FUNCTION
# ============================================================

def sha256_file(path):

    path = Path(path)

    if not path.exists():
        return None

    h = hashlib.sha256()

    with open(
        path,
        "rb"
    ) as f:

        while True:

            chunk = f.read(
                1024 * 1024
            )

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()


# ============================================================
# 5. IMPORTANT PROJECT ARTIFACTS
# ============================================================

important_files = [

    # --------------------------------------------------------
    # Phase 1
    # --------------------------------------------------------

    CHECKPOINT_DIR /
    "phase1_dataset_verification.json",

    # --------------------------------------------------------
    # Phase 2
    # --------------------------------------------------------

    CHECKPOINT_DIR /
    "phase2_checkpoint.json",

    # --------------------------------------------------------
    # Phase 3
    # --------------------------------------------------------

    CHECKPOINT_DIR /
    "phase3_checkpoint.json",

    # --------------------------------------------------------
    # Phase 4
    # --------------------------------------------------------

    CHECKPOINT_DIR /
    "phase4_checkpoint.json",

    # --------------------------------------------------------
    # Phase 5 batch generation
    # --------------------------------------------------------

    CHECKPOINT_DIR /
    "phase5_batch_checkpoint.json",

    # --------------------------------------------------------
    # Phase 5 decoder
    # --------------------------------------------------------

    CHECKPOINT_DIR /
    "phase5_decoder_best.pt",

    # --------------------------------------------------------
    # Phase 6 subject split
    # --------------------------------------------------------

    CHECKPOINT_DIR /
    "phase6_subject_split_checkpoint.json",

    PROJECT_DIR /
    "07_phase6_subject_split" /
    "phase6_subject_split_manifest.csv",

    PROJECT_DIR /
    "07_phase6_subject_split" /
    "phase6_subject_split.json",

    # --------------------------------------------------------
    # Phase 5 decoder results
    # --------------------------------------------------------

    RESULTS_DIR /
    "phase5_decoder_training_history.csv",

    RESULTS_DIR /
    "phase5_decoder_test_results.json",
]


# ============================================================
# 6. COLLECT ARTIFACT STATUS
# ============================================================

artifact_records = []

for path in important_files:

    path = Path(path)

    record = {

        "path":
            str(path),

        "exists":
            path.exists(),

        "size_bytes":
            (
                path.stat().st_size
                if path.exists()
                else None
            ),

        "sha256":
            (
                sha256_file(path)
                if path.exists()
                else None
            )
    }

    artifact_records.append(
        record
    )


# ============================================================
# 7. CURRENT IMPLEMENTATION STATUS
# ============================================================

implementation_status = {

    "project":
        "NeuroSentinel",

    "checkpoint_type":
        "PRE_PHASE_5H.3_IMPLEMENTATION",

    "status":
        "PAUSED_BEFORE_PHASE_5H.3",

    "timestamp":
        timestamp_readable,

    "validated_phases": {

        "phase_0":
            "COMPLETE",

        "phase_1":
            "COMPLETE",

        "phase_2":
            "COMPLETE",

        "phase_3":
            "COMPLETE",

        "phase_4":
            "COMPLETE",

        "phase_5_batch_window_generation":
            "COMPLETE",

        "phase_5_decoder_validation":
            "COMPLETE_WITH_COLLAPSE_DIAGNOSIS",

        "phase_6_subject_split":
            "COMPLETE"
    },

    "phase_5_batch_generation": {

        "recordings":
            62,

        "successful_recordings":
            62,

        "failed_recordings":
            0,

        "subjects":
            20,

        "total_windows":
            150955,

        "skipped_windows":
            0,

        "channels":
            61,

        "samples_per_window":
            25,

        "sampling_rate":
            50,

        "window_start_seconds":
            -0.2,

        "window_duration_seconds":
            0.5
    },

    "frozen_subject_split": {

        "train_subjects": [
            "003",
            "004",
            "005",
            "006",
            "009",
            "010",
            "012",
            "013",
            "014",
            "015",
            "016",
            "017",
            "019",
            "022"
        ],

        "validation_subjects": [
            "007",
            "018",
            "020"
        ],

        "test_subjects": [
            "002",
            "011",
            "021"
        ],

        "train_recordings":
            41,

        "validation_recordings":
            9,

        "test_recordings":
            12,

        "train_windows_indexed":
            99837,

        "validation_windows":
            21206,

        "test_windows":
            29907
    },

    "phase_5_decoder_status": {

        "architecture":
            "SimpleConvTimeAgg + Linear(2048,29)",

        "input_shape":
            "61 x 25",

        "embedding":
            2048,

        "classes":
            29,

        "meG_merger":
            "disabled",

        "original_decoder_checkpoint":
            "No validated pretrained Brain2Qwerty EEG checkpoint established",

        "test_accuracy":
            0.130605,

        "test_loss":
            2.939201,

        "collapse_diagnosis":
            "Model predicted class 8 (<space>) for all test windows",

        "utility_labels":
            "BLOCKED_PENDING_DECODER_VALIDATION"
    },

    "phase_5h3_status": {

        "status":
            "NOT_COMPLETED",

        "experiment":
            "Controlled longer normalized decoder training",

        "planned_epochs":
            15,

        "batch_size":
            64,

        "learning_rate":
            0.00005,

        "weight_decay":
            0.01,

        "normalization":
            "channel-wise per-window z-score",

        "seed":
            42,

        "current_gpu_status":
            "GPU quota exhausted",

        "training_status":
            "PAUSED_BEFORE_TRAINING"
    }
}


# ============================================================
# 8. ARTIFACT MANIFEST
# ============================================================

manifest = {

    "checkpoint":
        implementation_status,

    "important_artifacts":
        artifact_records,

    "next_step":
        "Resume Phase 5H.3 on CUDA when GPU access is available.",

    "do_not_modify": [

        str(
            CHECKPOINT_DIR /
            "phase5_decoder_best.pt"
        ),

        str(
            CHECKPOINT_DIR /
            "phase5_batch_checkpoint.json"
        ),

        str(
            CHECKPOINT_DIR /
            "phase6_subject_split_checkpoint.json"
        )
    ]
}


# ============================================================
# 9. SAVE IMPLEMENTATION CHECKPOINT
# ============================================================

checkpoint_path = (
    IMPLEMENTATION_DIR /
    f"pre_phase5h3_implementation_{timestamp}.json"
)

with open(
    checkpoint_path,
    "w"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )


# ============================================================
# 10. SAVE HUMAN-READABLE STATUS
# ============================================================

status_path = (
    IMPLEMENTATION_DIR /
    f"pre_phase5h3_status_{timestamp}.txt"
)

status_text = f"""
============================================================
NEUROSENTINEL — PRE-PHASE 5H.3 IMPLEMENTATION CHECKPOINT
============================================================

Created:
{timestamp_readable}

STATUS
------
Phase 0                  : COMPLETE
Phase 1                  : COMPLETE
Phase 2                  : COMPLETE
Phase 3                  : COMPLETE
Phase 4                  : COMPLETE
Phase 5 Batch Generation : COMPLETE
Phase 5 Decoder          : VALIDATED / COLLAPSE DIAGNOSED
Phase 6 Subject Split    : COMPLETE

PHASE 5 BATCH
-------------
Recordings               : 62
Successful               : 62
Failed                   : 0
Subjects                 : 20
Total windows            : 150,955
Skipped windows          : 0
Channels                 : 61
Samples per window       : 25
Sampling rate            : 50 Hz

FROZEN SPLIT
------------
Train recordings        : 41
Validation recordings    : 9
Test recordings          : 12

Train indexed windows   : 99,837
Validation windows       : 21,206
Test windows             : 29,907

PHASE 5 DECODER
---------------
Architecture             : SimpleConvTimeAgg + Linear(2048,29)
Input                    : 61 x 25
Embedding                : 2048
Classes                  : 29

Test accuracy            : 0.130605
Test loss                : 2.939201

Collapse diagnosis:
The decoder predicted class 8 (<space>) for all test windows.

Utility-label generation:
BLOCKED pending decoder validation.

PHASE 5H.3
-----------
Status                   : PAUSED BEFORE TRAINING
Planned epochs           : 15
Batch size               : 64
Learning rate            : 5e-5
Weight decay             : 0.01
Normalization            : Channel-wise per-window z-score
Seed                     : 42

GPU status:
GPU quota exhausted.
Training should resume on CUDA.

IMPORTANT
---------
Existing validated checkpoints were NOT modified.

Next step:
Resume Phase 5H.3 when CUDA GPU access is available.

============================================================
"""


with open(
    status_path,
    "w"
) as f:

    f.write(
        status_text
    )


# ============================================================
# 11. CREATE "LATEST" POINTER
# ============================================================

latest_pointer = (
    IMPLEMENTATION_DIR /
    "LATEST_PRE_PHASE5H3.json"
)

latest_pointer_data = {

    "latest_checkpoint":
        str(checkpoint_path),

    "latest_status":
        str(status_path),

    "created":
        timestamp_readable,

    "status":
        "PAUSED_BEFORE_PHASE_5H.3"
}

with open(
    latest_pointer,
    "w"
) as f:

    json.dump(
        latest_pointer_data,
        f,
        indent=2
    )


# ============================================================
# 12. PRINT FINAL CONFIRMATION
# ============================================================

print("\n" + "=" * 70)
print("PRE-PHASE 5H.3 IMPLEMENTATION SAVED")
print("=" * 70)

print("\nCheckpoint:")
print(checkpoint_path)

print("\nStatus:")
print(status_path)

print("\nLatest pointer:")
print(latest_pointer)

print("\nExisting important checkpoints remain untouched.")

print("\nCurrent project state:")
print("  Phase 0–4              : COMPLETE")
print("  Phase 5 batch          : COMPLETE")
print("  Phase 5 decoder        : COLLAPSE DIAGNOSED")
print("  Phase 6 split          : FROZEN")
print("  Phase 5H.3             : PAUSED")
print("  GPU                    : QUOTA EXHAUSTED")

print("\nNEXT ACTION:")
print(
    "Resume Phase 5H.3 on CUDA when GPU access is available."
)

print("\nYou can safely stop the Colab runtime now.")
```

```python
# ============================================================
# NEUROSENTINEL
# PHASE 5H.3 — CONTROLLED LONGER NORMALIZED DECODER TRAINING
#
# Features:
#   - CUDA automatically selected when available
#   - 15 epochs
#   - NO early stopping
#   - Checkpoint after EVERY completed epoch
#   - Automatic resume from latest checkpoint
#   - Separate best checkpoint
#   - Training history saved after every epoch
#   - Final completion metadata
#   - Existing Phase 5F checkpoint is NOT modified
# ============================================================

import os
import json
import time
import random
import shutil
import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from pathlib import Path
from torch.utils.data import Dataset, DataLoader
from neuraltrain.models.simpleconv import SimpleConvTimeAgg


# ============================================================
# 1. PROJECT PATHS
# ============================================================

PROJECT_DIR = Path(
    "/content/gdrive/MyDrive/NeuroSentinel"
)

RECORDINGS_DIR = (
    PROJECT_DIR /
    "06_phase5_utility_labels" /
    "recordings"
)

SPLIT_MANIFEST = (
    PROJECT_DIR /
    "07_phase6_subject_split" /
    "phase6_subject_split_manifest.csv"
)

CHECKPOINT_DIR = (
    PROJECT_DIR /
    "checkpoints"
)

RESULTS_DIR = (
    PROJECT_DIR /
    "results"
)

CHECKPOINT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# 2. CHECK REQUIRED PATHS
# ============================================================

print("=" * 70)
print("NEUROSENTINEL — PHASE 5H.3")
print("=" * 70)

print("\nChecking project paths...")

if not PROJECT_DIR.exists():
    raise FileNotFoundError(
        f"Project directory not found:\n{PROJECT_DIR}"
    )

if not RECORDINGS_DIR.exists():
    raise FileNotFoundError(
        f"Recordings directory not found:\n{RECORDINGS_DIR}"
    )

if not SPLIT_MANIFEST.exists():
    raise FileNotFoundError(
        f"Split manifest not found:\n{SPLIT_MANIFEST}"
    )

print("Project directory :", PROJECT_DIR)
print("Recordings        :", RECORDINGS_DIR)
print("Split manifest    :", SPLIT_MANIFEST)
print("Checkpoints       :", CHECKPOINT_DIR)
print("Results           :", RESULTS_DIR)


# ============================================================
# 3. EXPERIMENT CONFIGURATION
# ============================================================

SEED = 42

BATCH_SIZE = 64

LEARNING_RATE = 5e-5

WEIGHT_DECAY = 1e-2

EPOCHS = 15

NUM_CLASSES = 29

INPUT_CHANNELS = 61

INPUT_SAMPLES = 25

NORMALIZATION_EPSILON = 1e-6

RESUME_IF_AVAILABLE = True


# ============================================================
# 4. DEVICE
# ============================================================

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available()
    else "cpu"
)


# ============================================================
# 5. REPRODUCIBILITY
# ============================================================

random.seed(SEED)

np.random.seed(SEED)

torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

    # Deterministic behavior where practical.
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


# ============================================================
# 6. DEVICE INFORMATION
# ============================================================

print("\n" + "=" * 70)
print("EXECUTION CONFIGURATION")
print("=" * 70)

print("Device            :", DEVICE)
print("CUDA available    :", torch.cuda.is_available())

if torch.cuda.is_available():

    print(
        "GPU               :",
        torch.cuda.get_device_name(0)
    )

    print(
        "CUDA version      :",
        torch.version.cuda
    )

print("Batch size        :", BATCH_SIZE)
print("Learning rate     :", LEARNING_RATE)
print("Weight decay      :", WEIGHT_DECAY)
print("Epochs planned    :", EPOCHS)
print("Input             :", f"{INPUT_CHANNELS} × {INPUT_SAMPLES}")
print("Classes           :", NUM_CLASSES)
print("Normalization     :", "channel-wise per-window z-score")
print("Normalization eps :", NORMALIZATION_EPSILON)
print("Seed              :", SEED)
print("Early stopping    :", "DISABLED")
print("Automatic resume  :", RESUME_IF_AVAILABLE)
print("Checkpoint policy :", "AFTER EVERY COMPLETED EPOCH")


# ============================================================
# 7. CHECKPOINT PATHS
# ============================================================

LATEST_CHECKPOINT = (
    CHECKPOINT_DIR /
    "phase5h3_decoder_latest.pt"
)

BEST_CHECKPOINT = (
    CHECKPOINT_DIR /
    "phase5h3_decoder_best.pt"
)

HISTORY_PATH = (
    RESULTS_DIR /
    "phase5h3_decoder_training_history.csv"
)

METADATA_PATH = (
    CHECKPOINT_DIR /
    "phase5h3_checkpoint.json"
)


# ============================================================
# 8. BUTTON MAPPING
# ============================================================

BUTTON_MAPPING = {

    0: "s",
    1: "o",
    2: "t",
    3: "e",
    4: "n",
    5: "c",
    6: "i",
    7: "a",
    8: "<space>",
    9: "d",
    10: "l",
    11: "r",
    12: "b",
    13: "<special>",
    14: "z",
    15: "v",
    16: "f",
    17: "m",
    18: "u",
    19: "h",
    20: "p",
    21: "g",
    22: "q",
    23: "w",
    24: "x",
    25: "y",
    26: "j",
    27: "k",
    28: "<number>",
}


CHAR_TO_ID = {
    value: key
    for key, value in BUTTON_MAPPING.items()
}


def target_to_id(value):

    # --------------------------------------------------------
    # INTEGER
    # --------------------------------------------------------

    if isinstance(
        value,
        (int, np.integer)
    ):

        value = int(value)

        if 0 <= value <= 28:
            return value

    # --------------------------------------------------------
    # FLOAT
    # --------------------------------------------------------

    if isinstance(value, float):

        if np.isfinite(value):

            ivalue = int(value)

            if (
                0 <= ivalue <= 28
                and value == ivalue
            ):
                return ivalue

    # --------------------------------------------------------
    # STRING
    # --------------------------------------------------------

    value = str(value).strip()

    if value in CHAR_TO_ID:
        return CHAR_TO_ID[value]

    if value.lower() in [
        "space",
        "<space>",
        " "
    ]:
        return 8

    if value.lower() in [
        "special",
        "<special>"
    ]:
        return 13

    if value.lower() in [
        "number",
        "<number>"
    ]:
        return 28

    return None


# ============================================================
# 9. LOAD FROZEN SUBJECT-WISE SPLIT
# ============================================================

print("\n" + "=" * 70)
print("LOADING FROZEN SUBJECT-WISE SPLIT")
print("=" * 70)

split_df = pd.read_csv(
    SPLIT_MANIFEST
)

required_split_columns = {
    "recording_id",
    "split"
}

missing_columns = (
    required_split_columns -
    set(split_df.columns)
)

if missing_columns:

    raise ValueError(
        "Split manifest is missing required columns: "
        f"{sorted(missing_columns)}"
    )


train_recordings = (
    split_df[
        split_df["split"] == "train"
    ]["recording_id"]
    .astype(str)
    .tolist()
)


val_recordings = (
    split_df[
        split_df["split"] == "val"
    ]["recording_id"]
    .astype(str)
    .tolist()
)


print(
    "Train recordings:",
    len(train_recordings)
)

print(
    "Validation recordings:",
    len(val_recordings)
)

print(
    "Test recordings:",
    len(
        split_df[
            split_df["split"] == "test"
        ]
    )
)


if len(train_recordings) == 0:
    raise ValueError(
        "No training recordings found."
    )

if len(val_recordings) == 0:
    raise ValueError(
        "No validation recordings found."
    )


# ============================================================
# 10. DATASET
# ============================================================

class RecordingWindowDataset(Dataset):

    def __init__(
        self,
        recording_ids,
        recordings_dir
    ):

        self.recording_ids = recording_ids

        self.recordings_dir = recordings_dir

        self.index = []

        print(
            "\nIndexing",
            len(recording_ids),
            "recordings..."
        )

        for recording_id in recording_ids:

            window_path = (
                recordings_dir /
                f"{recording_id}_windows.npy"
            )

            metadata_path = (
                recordings_dir /
                f"{recording_id}_metadata.csv"
            )

            if not window_path.exists():

                raise FileNotFoundError(
                    f"Missing windows file:\n{window_path}"
                )

            if not metadata_path.exists():

                raise FileNotFoundError(
                    f"Missing metadata file:\n{metadata_path}"
                )

            # ------------------------------------------------
            # Memory-mapped EEG windows
            # ------------------------------------------------

            X = np.load(
                window_path,
                mmap_mode="r"
            )

            metadata = pd.read_csv(
                metadata_path
            )

            # ------------------------------------------------
            # Identify target column
            # ------------------------------------------------

            target_col = None

            for candidate in [
                "target",
                "button",
                "key",
                "character",
                "char"
            ]:

                if candidate in metadata.columns:

                    target_col = candidate

                    break

            if target_col is None:

                raise ValueError(
                    f"No target column found for "
                    f"{recording_id}"
                )

            # ------------------------------------------------
            # Verify number of windows
            # ------------------------------------------------

            if len(metadata) != len(X):

                raise ValueError(
                    f"Window/metadata mismatch for "
                    f"{recording_id}: "
                    f"{len(X)} windows vs "
                    f"{len(metadata)} metadata rows"
                )

            # ------------------------------------------------
            # Build index
            # ------------------------------------------------

            for i, value in enumerate(
                metadata[target_col]
            ):

                target_id = target_to_id(
                    value
                )

                if target_id is not None:

                    self.index.append(
                        (
                            recording_id,
                            i,
                            target_id
                        )
                    )

        print(
            "Indexed samples:",
            f"{len(self.index):,}"
        )

    def __len__(self):

        return len(self.index)

    def __getitem__(self, idx):

        recording_id, row_idx, target_id = (
            self.index[idx]
        )

        window_path = (
            self.recordings_dir /
            f"{recording_id}_windows.npy"
        )

        X = np.load(
            window_path,
            mmap_mode="r"
        )

        x = np.asarray(
            X[row_idx],
            dtype=np.float32
        )

        # ----------------------------------------------------
        # Shape validation
        # ----------------------------------------------------

        if x.shape != (
            INPUT_CHANNELS,
            INPUT_SAMPLES
        ):

            raise ValueError(
                f"Unexpected window shape "
                f"for {recording_id}: "
                f"{x.shape}"
            )

        # ----------------------------------------------------
        # PHASE 5F / H3 NORMALIZATION
        #
        # Channel-wise per-window z-score
        # ----------------------------------------------------

        mean = x.mean(
            axis=1,
            keepdims=True
        )

        std = x.std(
            axis=1,
            keepdims=True
        )

        x = (
            x - mean
        ) / (
            std + NORMALIZATION_EPSILON
        )

        # ----------------------------------------------------
        # Numerical safety
        # ----------------------------------------------------

        if not np.isfinite(x).all():

            raise ValueError(
                f"Non-finite normalized input "
                f"at index {idx}"
            )

        return (

            torch.from_numpy(
                x.copy()
            ),

            torch.tensor(
                target_id,
                dtype=torch.long
            )
        )


# ============================================================
# 11. BUILD DATASETS
# ============================================================

print("\n" + "=" * 70)
print("BUILDING DATASETS")
print("=" * 70)

train_dataset = RecordingWindowDataset(
    train_recordings,
    RECORDINGS_DIR
)

val_dataset = RecordingWindowDataset(
    val_recordings,
    RECORDINGS_DIR
)


print(
    "\nTrain samples:",
    f"{len(train_dataset):,}"
)

print(
    "Validation samples:",
    f"{len(val_dataset):,}"
)


if len(train_dataset) == 0:

    raise ValueError(
        "Training dataset is empty."
    )

if len(val_dataset) == 0:

    raise ValueError(
        "Validation dataset is empty."
    )


# ============================================================
# 12. DATA LOADERS
# ============================================================

train_loader = DataLoader(

    train_dataset,

    batch_size=BATCH_SIZE,

    shuffle=True,

    num_workers=0,

    pin_memory=torch.cuda.is_available()
)


val_loader = DataLoader(

    val_dataset,

    batch_size=BATCH_SIZE,

    shuffle=False,

    num_workers=0,

    pin_memory=torch.cuda.is_available()
)


# ============================================================
# 13. BUILD FRESH PHASE 5F ARCHITECTURE
# ============================================================

print("\n" + "=" * 70)
print("BUILDING FRESH PHASE 5F ARCHITECTURE")
print("=" * 70)

model_spec = SimpleConvTimeAgg(

    time_agg_out="att",

    dropout_input=0.2,

    conv_dropout=0.5,

    hidden=2048,

    batch_norm=True,

    depth=8,

    dilation_period=3,

    kernel_size=3,

    relu_leakiness=0.01,

    initial_linear=512,

    gelu=True,

    skip=True,

    scale=0.1,

    subject_layers_config=None,

    merger_config=None,
)


decoder = model_spec.build(

    n_in_channels=INPUT_CHANNELS,

    n_outputs=2048
)


decoder = decoder.to(
    DEVICE
)


# ============================================================
# 14. INITIALIZE LAZY PARAMETERS
# ============================================================

print("\nInitializing decoder...")

with torch.no_grad():

    dummy = torch.zeros(

        2,

        INPUT_CHANNELS,

        INPUT_SAMPLES,

        device=DEVICE
    )

    dummy_representation = decoder(
        dummy
    )


print(
    "Dummy representation:",
    tuple(
        dummy_representation.shape
    )
)


if dummy_representation.shape != (
    2,
    2048
):

    raise ValueError(
        "Unexpected decoder representation shape: "
        f"{dummy_representation.shape}"
    )


# ============================================================
# 15. CLASSIFIER
# ============================================================

classifier = nn.Linear(

    2048,

    NUM_CLASSES

).to(
    DEVICE
)


# ============================================================
# 16. OPTIMIZER
# ============================================================

optimizer = torch.optim.AdamW(

    list(
        decoder.parameters()
    )
    +
    list(
        classifier.parameters()
    ),

    lr=LEARNING_RATE,

    weight_decay=WEIGHT_DECAY
)


criterion = nn.CrossEntropyLoss()


# ============================================================
# 17. RESUME STATE
# ============================================================

history = []

best_val_loss = float("inf")

best_val_accuracy = 0.0

start_epoch = 1


# ============================================================
# 18. AUTOMATIC RESUME
# ============================================================

if (
    RESUME_IF_AVAILABLE
    and
    LATEST_CHECKPOINT.exists()
):

    print("\n" + "=" * 70)
    print("LATEST CHECKPOINT FOUND")
    print("=" * 70)

    print(
        LATEST_CHECKPOINT
    )

    print(
        "\nLoading checkpoint..."
    )

    resume_checkpoint = torch.load(

        LATEST_CHECKPOINT,

        map_location=DEVICE,

        weights_only=False
    )

    # --------------------------------------------------------
    # Validate checkpoint
    # --------------------------------------------------------

    checkpoint_phase = (
        resume_checkpoint.get(
            "phase"
        )
    )

    if checkpoint_phase != "5H.3":

        raise ValueError(
            "Checkpoint phase mismatch: "
            f"{checkpoint_phase}"
        )

    completed_epoch = int(
        resume_checkpoint.get(
            "epoch_completed",
            0
        )
    )

    # --------------------------------------------------------
    # If already complete, do not retrain
    # --------------------------------------------------------

    checkpoint_status = (
        resume_checkpoint.get(
            "status",
            "IN_PROGRESS"
        )
    )

    if (
        checkpoint_status == "COMPLETE"
        and
        completed_epoch >= EPOCHS
    ):

        print(
            "\nPhase 5H.3 is already complete."
        )

        print(
            "Completed epochs:",
            completed_epoch
        )

        print(
            "Latest checkpoint:",
            LATEST_CHECKPOINT
        )

        print(
            "Best checkpoint:",
            BEST_CHECKPOINT
        )

        print(
            "\nNo additional training required."
        )

    else:

        # ----------------------------------------------------
        # Load decoder
        # ----------------------------------------------------

        if (
            "decoder_state_dict"
            in resume_checkpoint
        ):

            decoder.load_state_dict(
                resume_checkpoint[
                    "decoder_state_dict"
                ]
            )

        elif (
            "model_state_dict"
            in resume_checkpoint
        ):

            # Compatibility with the previous
            # H3 checkpoint format.

            combined_state = (
                resume_checkpoint[
                    "model_state_dict"
                ]
            )

            decoder_state = {}

            for key, value in (
                combined_state.items()
            ):

                if key.startswith("0."):

                    decoder_state[
                        key[2:]
                    ] = value

            decoder.load_state_dict(
                decoder_state
            )

        else:

            raise ValueError(
                "Decoder state not found "
                "in checkpoint."
            )

        # ----------------------------------------------------
        # Load classifier
        # ----------------------------------------------------

        if (
            "classifier_state_dict"
            in resume_checkpoint
        ):

            classifier.load_state_dict(
                resume_checkpoint[
                    "classifier_state_dict"
                ]
            )

        elif (
            "model_state_dict"
            in resume_checkpoint
        ):

            combined_state = (
                resume_checkpoint[
                    "model_state_dict"
                ]
            )

            classifier.load_state_dict({

                "weight":
                    combined_state[
                        "1.weight"
                    ],

                "bias":
                    combined_state[
                        "1.bias"
                    ]
            })

        else:

            raise ValueError(
                "Classifier state not found "
                "in checkpoint."
            )

        # ----------------------------------------------------
        # Load optimizer
        # ----------------------------------------------------

        if (
            "optimizer_state_dict"
            in resume_checkpoint
        ):

            optimizer.load_state_dict(
                resume_checkpoint[
                    "optimizer_state_dict"
                ]
            )

        # ----------------------------------------------------
        # Load history
        # ----------------------------------------------------

        history = resume_checkpoint.get(
            "history",
            []
        )

        best_val_loss = float(
            resume_checkpoint.get(
                "best_val_loss",
                float("inf")
            )
        )

        best_val_accuracy = float(
            resume_checkpoint.get(
                "best_val_accuracy",
                0.0
            )
        )

        start_epoch = (
            completed_epoch + 1
        )

        print(
            "\nCheckpoint loaded successfully."
        )

        print(
            "Last completed epoch:",
            completed_epoch
        )

        print(
            "Next epoch:",
            start_epoch
        )

        print(
            "Best validation loss:",
            best_val_loss
        )

else:

    print("\n" + "=" * 70)
    print("NO PREVIOUS H3 CHECKPOINT")
    print("=" * 70)

    print(
        "Starting Phase 5H.3 from epoch 1."
    )


# ============================================================
# 19. HANDLE ALREADY COMPLETE CASE
# ============================================================

if start_epoch > EPOCHS:

    print("\n" + "=" * 70)
    print("PHASE 5H.3 ALREADY COMPLETE")
    print("=" * 70)

    print(
        f"All {EPOCHS} epochs are already completed."
    )

    print(
        "Latest:",
        LATEST_CHECKPOINT
    )

    print(
        "Best:",
        BEST_CHECKPOINT
    )

    print(
        "History:",
        HISTORY_PATH
    )

else:

    # ========================================================
    # 20. TRAINING
    # ========================================================

    print("\n" + "=" * 70)
    print("PHASE 5H.3 TRAINING STARTED")
    print("=" * 70)

    print(
        f"Epoch range: {start_epoch} → {EPOCHS}"
    )

    print(
        "Early stopping: DISABLED"
    )

    print(
        "Checkpointing: EVERY COMPLETED EPOCH"
    )

    print(
        "Device:",
        DEVICE
    )

    print()

    # --------------------------------------------------------
    # TRAINING LOOP
    # --------------------------------------------------------

    for epoch in range(
        start_epoch,
        EPOCHS + 1
    ):

        epoch_start = time.time()

        # ====================================================
        # TRAIN
        # ====================================================

        decoder.train()

        classifier.train()

        train_loss_sum = 0.0

        train_correct = 0

        train_total = 0

        for batch_idx, (xb, yb) in enumerate(
            train_loader,
            start=1
        ):

            xb = xb.to(
                DEVICE,
                non_blocking=True
            )

            yb = yb.to(
                DEVICE,
                non_blocking=True
            )

            optimizer.zero_grad(
                set_to_none=True
            )

            representation = decoder(
                xb
            )

            logits = classifier(
                representation
            )

            loss = criterion(
                logits,
                yb
            )

            # ------------------------------------------------
            # Safety check
            # ------------------------------------------------

            if not torch.isfinite(loss):

                raise RuntimeError(
                    f"Non-finite training loss "
                    f"at epoch {epoch}, "
                    f"batch {batch_idx}"
                )

            loss.backward()

            optimizer.step()

            train_loss_sum += (
                loss.item()
                *
                len(yb)
            )

            predictions = logits.argmax(
                dim=1
            )

            train_correct += int(
                (
                    predictions == yb
                )
                .sum()
                .item()
            )

            train_total += len(yb)

        # ====================================================
        # TRAIN METRICS
        # ====================================================

        train_loss = (
            train_loss_sum /
            train_total
        )

        train_accuracy = (
            train_correct /
            train_total
        )

        # ====================================================
        # VALIDATION
        # ====================================================

        decoder.eval()

        classifier.eval()

        val_loss_sum = 0.0

        val_correct = 0

        val_total = 0

        with torch.no_grad():

            for xb, yb in val_loader:

                xb = xb.to(
                    DEVICE,
                    non_blocking=True
                )

                yb = yb.to(
                    DEVICE,
                    non_blocking=True
                )

                representation = decoder(
                    xb
                )

                logits = classifier(
                    representation
                )

                loss = criterion(
                    logits,
                    yb
                )

                if not torch.isfinite(loss):

                    raise RuntimeError(
                        f"Non-finite validation loss "
                        f"at epoch {epoch}"
                    )

                val_loss_sum += (
                    loss.item()
                    *
                    len(yb)
                )

                predictions = logits.argmax(
                    dim=1
                )

                val_correct += int(
                    (
                        predictions == yb
                    )
                    .sum()
                    .item()
                )

                val_total += len(yb)

        # ====================================================
        # VALIDATION METRICS
        # ====================================================

        val_loss = (
            val_loss_sum /
            val_total
        )

        val_accuracy = (
            val_correct /
            val_total
        )

        epoch_time = (
            time.time()
            -
            epoch_start
        )

        # ====================================================
        # BEST MODEL CHECK
        # ====================================================

        improved = (
            val_loss < best_val_loss
        )

        if improved:

            best_val_loss = val_loss

            best_val_accuracy = (
                val_accuracy
            )

        # ====================================================
        # EPOCH RECORD
        # ====================================================

        epoch_record = {

            "epoch":
                int(epoch),

            "train_loss":
                float(train_loss),

            "train_accuracy":
                float(train_accuracy),

            "val_loss":
                float(val_loss),

            "val_accuracy":
                float(val_accuracy),

            "best_val_loss":
                float(best_val_loss),

            "best_val_accuracy":
                float(best_val_accuracy),

            "epoch_seconds":
                float(epoch_time),

            "device":
                str(DEVICE),

            "timestamp":
                time.strftime(
                    "%Y-%m-%d %H:%M:%S"
                )
        }

        history.append(
            epoch_record
        )

        # ====================================================
        # SAVE CSV AFTER EVERY EPOCH
        # ====================================================

        pd.DataFrame(
            history
        ).to_csv(
            HISTORY_PATH,
            index=False
        )

        # ====================================================
        # CREATE CHECKPOINT
        # ====================================================

        checkpoint = {

            "phase":
                "5H.3",

            "status":
                (
                    "COMPLETE"
                    if epoch == EPOCHS
                    else "IN_PROGRESS"
                ),

            "experiment":
                "longer normalized decoder training",

            "epoch_completed":
                int(epoch),

            "epochs_planned":
                int(EPOCHS),

            # ------------------------------------------------
            # MODEL
            # ------------------------------------------------

            "decoder_state_dict":
                {
                    key:
                        value.detach().cpu()
                    for key, value
                    in decoder.state_dict().items()
                },

            "classifier_state_dict":
                {
                    key:
                        value.detach().cpu()
                    for key, value
                    in classifier.state_dict().items()
                },

            # ------------------------------------------------
            # OPTIMIZER
            # ------------------------------------------------

            "optimizer_state_dict":
                optimizer.state_dict(),

            # ------------------------------------------------
            # MODEL CONFIG
            # ------------------------------------------------

            "model_config": {

                "input_channels":
                    INPUT_CHANNELS,

                "input_samples":
                    INPUT_SAMPLES,

                "hidden":
                    2048,

                "output_classes":
                    NUM_CLASSES,

                "time_agg_out":
                    "att",

                "depth":
                    8,

                "kernel_size":
                    3,

                "dilation_period":
                    3,

                "merger_config":
                    None,

                "subject_layers_config":
                    None
            },

            # ------------------------------------------------
            # TRAINING CONFIG
            # ------------------------------------------------

            "training_config": {

                "batch_size":
                    BATCH_SIZE,

                "learning_rate":
                    LEARNING_RATE,

                "weight_decay":
                    WEIGHT_DECAY,

                "epochs_planned":
                    EPOCHS,

                "normalization":
                    "channel-wise per-window z-score",

                "epsilon":
                    NORMALIZATION_EPSILON,

                "seed":
                    SEED,

                "device":
                    str(DEVICE),

                "early_stopping":
                    False
            },

            # ------------------------------------------------
            # DATA
            # ------------------------------------------------

            "split_manifest":
                str(SPLIT_MANIFEST),

            "train_samples":
                len(train_dataset),

            "val_samples":
                len(val_dataset),

            # ------------------------------------------------
            # METRICS
            # ------------------------------------------------

            "last_epoch_metrics":
                epoch_record,

            "best_val_loss":
                float(best_val_loss),

            "best_val_accuracy":
                float(best_val_accuracy),

            "history":
                history,

            # ------------------------------------------------
            # TIMESTAMP
            # ------------------------------------------------

            "timestamp":
                time.strftime(
                    "%Y-%m-%d %H:%M:%S"
                )
        }

        # ====================================================
        # SAVE LATEST CHECKPOINT
        # ====================================================

        torch.save(
            checkpoint,
            LATEST_CHECKPOINT
        )

        # ====================================================
        # SAVE BEST CHECKPOINT
        # ====================================================

        if improved:

            torch.save(
                checkpoint,
                BEST_CHECKPOINT
            )

        # ====================================================
        # PRINT EPOCH RESULT
        # ====================================================

        best_marker = (
            " <-- BEST"
            if improved
            else ""
        )

        print(
            f"Epoch {epoch:02d}/{EPOCHS} | "
            f"Train Loss {train_loss:.6f} | "
            f"Train Acc {train_accuracy:.6f} | "
            f"Val Loss {val_loss:.6f} | "
            f"Val Acc {val_accuracy:.6f} | "
            f"{epoch_time:.1f}s"
            f"{best_marker}"
        )

        print(
            "  Latest checkpoint saved."
        )

        if improved:

            print(
                "  Best checkpoint updated."
            )

        print(
            "  Progress:",
            f"{epoch}/{EPOCHS} epochs completed."
        )

        # ----------------------------------------------------
        # CUDA MEMORY CLEANUP
        # ----------------------------------------------------

        if torch.cuda.is_available():

            torch.cuda.empty_cache()


    # ========================================================
    # 21. FINAL METADATA
    # ========================================================

    final_metadata = {

        "phase":
            "5H.3",

        "status":
            "COMPLETE",

        "epochs_completed":
            int(EPOCHS),

        "epochs_planned":
            int(EPOCHS),

        "early_stopping":
            False,

        "best_val_loss":
            float(best_val_loss),

        "best_val_accuracy":
            float(best_val_accuracy),

        "latest_checkpoint":
            str(LATEST_CHECKPOINT),

        "best_checkpoint":
            str(BEST_CHECKPOINT),

        "history":
            str(HISTORY_PATH),

        "train_samples":
            len(train_dataset),

        "val_samples":
            len(val_dataset),

        "architecture":
            "Phase 5F normalized SimpleConvTimeAgg + Linear(2048,29)",

        "device":
            str(DEVICE),

        "seed":
            SEED,

        "timestamp":
            time.strftime(
                "%Y-%m-%d %H:%M:%S"
            )
    }


    with open(
        METADATA_PATH,
        "w"
    ) as f:

        json.dump(
            final_metadata,
            f,
            indent=2
        )


    # ========================================================
    # 22. FINAL STATUS
    # ========================================================

    print("\n" + "=" * 70)
    print("PHASE 5H.3 COMPLETE")
    print("=" * 70)

    print(
        f"All {EPOCHS} epochs completed."
    )

    print(
        "Best validation loss:",
        f"{best_val_loss:.6f}"
    )

    print(
        "Best validation accuracy:",
        f"{best_val_accuracy:.6f}"
    )

    print("\nSaved files:")

    print(
        "Best checkpoint:",
        BEST_CHECKPOINT
    )

    print(
        "Latest checkpoint:",
        LATEST_CHECKPOINT
    )

    print(
        "Training history:",
        HISTORY_PATH
    )

    print(
        "Final metadata:",
        METADATA_PATH
    )

    print(
        "\nExisting Phase 5F checkpoint was NOT modified:"
    )

    print(
        CHECKPOINT_DIR /
        "phase5_decoder_best.pt"
    )

    print("\nExecution finished normally.")
```

```python
from torch.nn.parameter import (
    UninitializedParameter,
    UninitializedBuffer
)

def safe_state_dict(module):
    raw_state = module.state_dict()

    safe_state = {}
    skipped = []

    for name, value in raw_state.items():

        if isinstance(
            value,
            (UninitializedParameter, UninitializedBuffer)
        ):
            skipped.append(name)
            continue

        safe_state[name] = value.detach().cpu().clone()

    return safe_state, skipped
```

```python
# ============================================================
# NEUROSENTINEL — PHASE 5H.3
# NORMALIZED EEG DECODER RETRAINING
# ============================================================

import os
import sys
import json
import time
import random
import traceback
import importlib
from pathlib import Path
from datetime import datetime

import numpy as np
import pandas as pd

import torch
import torch.nn as nn

from torch.utils.data import Dataset, DataLoader, Subset

from torch.nn.parameter import (
    UninitializedParameter,
    UninitializedBuffer,
)


# ============================================================
# 1. CONFIGURATION
# ============================================================

SEED = 42

BATCH_SIZE = 8
NUM_EPOCHS = 15

LEARNING_RATE = 5e-5
WEIGHT_DECAY = 1e-2

NUM_CLASSES = 29

N_CHANNELS = 61
N_TIME = 25

MIN_STD = 1e-6

EMERGENCY_EVERY_BATCHES = 250


# ============================================================
# 2. PROJECT PATHS
# ============================================================

PROJECT_DIR = Path(
    "/content/gdrive/MyDrive/NeuroSentinel"
)

RECORDINGS_DIR = (
    PROJECT_DIR
    / "06_phase5_utility_labels"
    / "recordings"
)

SPLIT_MANIFEST = (
    PROJECT_DIR
    / "07_phase6_subject_split"
    / "phase6_subject_split_manifest.csv"
)

BRAIN2QWERTY_DIR = (
    PROJECT_DIR
    / "brain2qwerty"
)


# ============================================================
# 3. OUTPUT PATHS
# ============================================================

H3_CHECKPOINT_DIR = (
    PROJECT_DIR
    / "checkpoints"
    / "phase5h3"
)

H3_RESULTS_DIR = (
    PROJECT_DIR
    / "results"
    / "phase5h3"
)

H3_CHECKPOINT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

H3_RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


LATEST_CHECKPOINT = (
    H3_CHECKPOINT_DIR
    / "phase5h3_latest.pt"
)

BEST_CHECKPOINT = (
    H3_CHECKPOINT_DIR
    / "phase5h3_best.pt"
)

HISTORY_FILE = (
    H3_RESULTS_DIR
    / "phase5h3_training_history.csv"
)

CONFIG_FILE = (
    H3_RESULTS_DIR
    / "phase5h3_config.json"
)

INTERRUPTION_FILE = (
    H3_CHECKPOINT_DIR
    / "phase5h3_interruptions.json"
)

ANOMALY_AUDIT_FILE = (
    H3_RESULTS_DIR
    / "phase5h3_anomalous_targets.csv"
)


# ============================================================
# 4. REPRODUCIBILITY
# ============================================================

def seed_everything(seed=42):

    random.seed(seed)

    np.random.seed(seed)

    torch.manual_seed(seed)

    if torch.cuda.is_available():

        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)

    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


seed_everything(SEED)


# ============================================================
# 5. DEVICE
# ============================================================

DEVICE = torch.device(
    "cuda"
    if torch.cuda.is_available()
    else "cpu"
)


print("=" * 70)
print("NEUROSENTINEL — PHASE 5H.3")
print("=" * 70)

print(f"Device: {DEVICE}")

if torch.cuda.is_available():

    print(
        f"GPU: {torch.cuda.get_device_name(0)}"
    )

    print(
        "GPU memory: "
        f"{torch.cuda.get_device_properties(0).total_memory / 1024**3:.2f} GB"
    )

print(
    f"PyTorch: {torch.__version__}"
)

print(
    f"Batch size: {BATCH_SIZE}"
)

print(
    f"Epochs: {NUM_EPOCHS}"
)

print(
    f"Learning rate: {LEARNING_RATE}"
)

print(
    f"Weight decay: {WEIGHT_DECAY}"
)

print(
    f"Seed: {SEED}"
)


# ============================================================
# 6. VERIFY PATHS
# ============================================================

print()
print("=" * 70)
print("VERIFYING PROJECT PATHS")
print("=" * 70)

required_paths = [
    PROJECT_DIR,
    RECORDINGS_DIR,
    SPLIT_MANIFEST,
    BRAIN2QWERTY_DIR,
]

for path in required_paths:

    print(
        f"Checking: {path}"
    )

    if not path.exists():

        raise FileNotFoundError(
            f"Required path does not exist:\n{path}"
        )


print()
print(
    "All required paths exist."
)


# ============================================================
# 7. LOAD PHASE 6 MANIFEST
# ============================================================

manifest = pd.read_csv(
    SPLIT_MANIFEST
)

print()
print("=" * 70)
print("PHASE 6 MANIFEST")
print("=" * 70)

print(
    f"Shape: {manifest.shape}"
)

print(
    f"Columns: {list(manifest.columns)}"
)


required_columns = {
    "recording_id",
    "subject",
    "split",
}

missing_columns = (
    required_columns
    - set(manifest.columns)
)

if missing_columns:

    raise ValueError(
        "Missing required manifest columns: "
        f"{missing_columns}"
    )


manifest["recording_id"] = (
    manifest["recording_id"]
    .astype(str)
    .str.strip()
)

manifest["subject"] = (
    manifest["subject"]
    .astype(int)
)

manifest["split"] = (
    manifest["split"]
    .astype(str)
    .str.strip()
    .str.lower()
)


print()
print("Split counts:")

print(
    manifest["split"].value_counts()
)


# ============================================================
# 8. FROZEN SUBJECT SPLIT
# ============================================================

train_subjects = sorted(
    manifest.loc[
        manifest["split"] == "train",
        "subject"
    ]
    .unique()
    .tolist()
)

val_subjects = sorted(
    manifest.loc[
        manifest["split"] == "val",
        "subject"
    ]
    .unique()
    .tolist()
)

test_subjects = sorted(
    manifest.loc[
        manifest["split"] == "test",
        "subject"
    ]
    .unique()
    .tolist()
)


print()
print("=" * 70)
print("FROZEN PHASE 6 SUBJECT SPLIT")
print("=" * 70)

print(
    "Train subjects:",
    train_subjects
)

print(
    "Validation subjects:",
    val_subjects
)

print(
    "Test subjects:",
    test_subjects
)


assert set(train_subjects).isdisjoint(
    val_subjects
)

assert set(train_subjects).isdisjoint(
    test_subjects
)

assert set(val_subjects).isdisjoint(
    test_subjects
)


print()
print(
    "Subject leakage check: PASSED"
)


# ============================================================
# 9. DISCOVER RECORDINGS
# ============================================================

print()
print("=" * 70)
print("DISCOVERING GENERATED RECORDINGS")
print("=" * 70)


window_files = sorted(
    RECORDINGS_DIR.glob(
        "*_windows.npy"
    )
)

metadata_files = sorted(
    RECORDINGS_DIR.glob(
        "*_metadata.csv"
    )
)


print(
    f"Window files: {len(window_files)}"
)

print(
    f"Metadata files: {len(metadata_files)}"
)


# ============================================================
# 10. RECORDING MAPS
# ============================================================

recording_file_map = {}

for path in window_files:

    recording_id = (
        path.name
        .removesuffix(
            "_windows.npy"
        )
    )

    if recording_id in recording_file_map:

        raise RuntimeError(
            f"Duplicate window file: "
            f"{recording_id}"
        )

    recording_file_map[
        recording_id
    ] = path


metadata_file_map = {}

for path in metadata_files:

    recording_id = (
        path.name
        .removesuffix(
            "_metadata.csv"
        )
    )

    if recording_id in metadata_file_map:

        raise RuntimeError(
            f"Duplicate metadata file: "
            f"{recording_id}"
        )

    metadata_file_map[
        recording_id
    ] = path


# ============================================================
# 11. VERIFY RECORDING MATCH
# ============================================================

manifest_ids = set(
    manifest["recording_id"]
)

window_ids = set(
    recording_file_map.keys()
)

metadata_ids = set(
    metadata_file_map.keys()
)


missing_windows = sorted(
    manifest_ids - window_ids
)

missing_metadata = sorted(
    manifest_ids - metadata_ids
)

extra_windows = sorted(
    window_ids - manifest_ids
)

extra_metadata = sorted(
    metadata_ids - manifest_ids
)


print()
print("=" * 70)
print("RECORDING MAPPING")
print("=" * 70)

print(
    f"Manifest recordings: {len(manifest_ids)}"
)

print(
    f"Window recordings:   {len(window_ids)}"
)

print(
    f"Metadata recordings: {len(metadata_ids)}"
)

print(
    f"Missing windows:     {len(missing_windows)}"
)

print(
    f"Missing metadata:    {len(missing_metadata)}"
)

print(
    f"Extra windows:       {len(extra_windows)}"
)

print(
    f"Extra metadata:      {len(extra_metadata)}"
)


if missing_windows:

    raise RuntimeError(
        "Missing window recordings:\n"
        + "\n".join(missing_windows)
    )


if missing_metadata:

    raise RuntimeError(
        "Missing metadata recordings:\n"
        + "\n".join(missing_metadata)
    )


if extra_windows:

    raise RuntimeError(
        "Unexpected window recordings:\n"
        + "\n".join(extra_windows)
    )


if extra_metadata:

    raise RuntimeError(
        "Unexpected metadata recordings:\n"
        + "\n".join(extra_metadata)
    )


print()
print(
    "Recording mapping check: PASSED"
)


# ============================================================
# 12. 29-CLASS LABEL MAPPING
# ============================================================

CLASS_LABELS = [
    "a", "b", "c", "d", "e", "f", "g",
    "h", "i", "j", "k", "l", "m", "n",
    "o", "p", "q", "r", "s", "t", "u",
    "v", "w", "x", "y", "z",
    "<space>",
    "<number>",
    "<special>",
]


assert len(CLASS_LABELS) == NUM_CLASSES


LABEL_TO_CLASS = {
    label: idx
    for idx, label in enumerate(CLASS_LABELS)
}

CLASS_TO_LABEL = {
    idx: label
    for label, idx in LABEL_TO_CLASS.items()
}


print()
print("=" * 70)
print("29-CLASS LABEL MAPPING")
print("=" * 70)

for idx, label in CLASS_TO_LABEL.items():

    print(
        f"{idx:2d} -> {label}"
    )

print()
print(
    f"Number of classes: {len(CLASS_LABELS)}"
)


# ============================================================
# 13. TARGET LABEL AUDIT
# ============================================================

VALID_LABELS = set(
    LABEL_TO_CLASS.keys()
)

ANOMALOUS_LABELS = {
    "-",
    "`",
}


print()
print("=" * 70)
print("GLOBAL TARGET-LABEL AUDIT")
print("=" * 70)


all_target_values = set()

target_counts = {}

anomalous_rows = []


for metadata_path in metadata_files:

    df = pd.read_csv(
        metadata_path
    )


    if "target" not in df.columns:

        raise ValueError(
            f"'target' column missing from "
            f"{metadata_path.name}"
        )


    values = (
        df["target"]
        .astype(str)
        .str.strip()
    )


    for row_idx, value in values.items():

        all_target_values.add(
            value
        )

        target_counts[value] = (
            target_counts.get(
                value,
                0
            )
            + 1
        )


        if value in ANOMALOUS_LABELS:

            row = df.iloc[row_idx]


            anomalous_rows.append({

                "metadata_file":
                    metadata_path.name,

                "row_index":
                    int(row_idx),

                "recording_id":
                    row.get(
                        "recording_id",
                        None
                    ),

                "subject":
                    row.get(
                        "subject",
                        None
                    ),

                "session":
                    row.get(
                        "session",
                        None
                    ),

                "task":
                    row.get(
                        "task",
                        None
                    ),

                "trial_id":
                    row.get(
                        "trial_id",
                        None
                    ),

                "keystroke_index":
                    row.get(
                        "keystroke_index",
                        None
                    ),

                "timestamp":
                    row.get(
                        "timestamp",
                        None
                    ),

                "target":
                    value,

                "text":
                    row.get(
                        "text",
                        None
                    ),
            })


print(
    "Unique target labels:"
)

for label in sorted(all_target_values):

    print(
        f"{repr(label):15s}"
        f" -> {target_counts[label]}"
    )


unknown_labels = (
    all_target_values
    - VALID_LABELS
)


unexpected_unknowns = (
    unknown_labels
    - ANOMALOUS_LABELS
)


if unexpected_unknowns:

    raise ValueError(
        "Unexpected unknown target labels: "
        f"{sorted(unexpected_unknowns)}"
    )


print()
print(
    "Unknown labels:"
)

if unknown_labels:

    for label in sorted(unknown_labels):

        print(
            f"  {repr(label)} "
            f"({target_counts[label]} occurrences)"
        )

else:

    print(
        "  None"
    )


print()
print(
    f"Anomalous samples to exclude: "
    f"{len(anomalous_rows)}"
)


for row in anomalous_rows:

    print(
        f"  {row['recording_id']} | "
        f"row={row['row_index']} | "
        f"target={repr(row['target'])} | "
        f"text={row['text']}"
    )


pd.DataFrame(
    anomalous_rows
).to_csv(
    ANOMALY_AUDIT_FILE,
    index=False
)


print()
print(
    "Anomaly audit saved:"
)

print(
    ANOMALY_AUDIT_FILE
)


# ============================================================
# 14. RECORDING STORE
# ============================================================

class RecordingStore:

    def __init__(
        self,
        recording_file_map,
        metadata_file_map,
    ):

        self.recording_file_map = (
            recording_file_map
        )

        self.metadata_file_map = (
            metadata_file_map
        )

        self.cache = {}


    def load_recording(
        self,
        recording_id
    ):

        if recording_id in self.cache:

            return self.cache[recording_id]


        window_path = (
            self.recording_file_map[
                recording_id
            ]
        )

        metadata_path = (
            self.metadata_file_map[
                recording_id
            ]
        )


        X = np.load(
            window_path,
            mmap_mode="r"
        )


        metadata = pd.read_csv(
            metadata_path
        )


        if len(X) != len(metadata):

            raise ValueError(
                f"Length mismatch for "
                f"{recording_id}: "
                f"{len(X)} windows vs "
                f"{len(metadata)} metadata rows"
            )


        if X.ndim != 3:

            raise ValueError(
                f"{recording_id}: expected "
                f"3D EEG array, got {X.shape}"
            )


        if X.shape[1] != N_CHANNELS:

            raise ValueError(
                f"{recording_id}: expected "
                f"{N_CHANNELS} channels, "
                f"got {X.shape[1]}"
            )


        if X.shape[2] != N_TIME:

            raise ValueError(
                f"{recording_id}: expected "
                f"{N_TIME} time samples, "
                f"got {X.shape[2]}"
            )


        labels = np.full(
            len(metadata),
            -1,
            dtype=np.int64
        )


        for idx, value in enumerate(
            metadata["target"]
        ):

            value = str(value).strip()


            if value in LABEL_TO_CLASS:

                labels[idx] = (
                    LABEL_TO_CLASS[value]
                )

            elif value in ANOMALOUS_LABELS:

                labels[idx] = -1

            else:

                raise ValueError(
                    f"Unknown target "
                    f"{repr(value)} in "
                    f"{metadata_path.name}"
                )


        X = np.asarray(
            X,
            dtype=np.float32
        ).copy()


        labels = labels.copy()


        self.cache[
            recording_id
        ] = (
            X,
            labels,
            metadata
        )


        return self.cache[
            recording_id
        ]


store = RecordingStore(
    recording_file_map,
    metadata_file_map,
)


# ============================================================
# 15. BUILD GLOBAL SAMPLE INDEX
# ============================================================

def build_index(
    split_name
):

    split_rows = manifest[
        manifest["split"] == split_name
    ].copy()


    index = []

    total_windows = 0

    excluded_windows = 0


    print()
    print(
        f"Building {split_name} index..."
    )


    for _, row in split_rows.iterrows():

        recording_id = str(
            row["recording_id"]
        )


        X, y, metadata = (
            store.load_recording(
                recording_id
            )
        )


        n_windows = len(X)

        total_windows += n_windows


        for local_idx in range(n_windows):

            label = int(
                y[local_idx]
            )


            if label < 0:

                excluded_windows += 1

                continue


            index.append(
                (
                    recording_id,
                    local_idx
                )
            )


    print(
        f"{split_name}: "
        f"{len(index):,} valid samples"
    )

    print(
        f"{split_name}: "
        f"{excluded_windows:,} excluded samples"
    )

    print(
        f"{split_name}: "
        f"{total_windows:,} total samples"
    )


    return index


train_index = build_index("train")
val_index = build_index("val")
test_index = build_index("test")


# ============================================================
# 16. FINAL DATASET SIZE
# ============================================================

print()
print("=" * 70)
print("FINAL DATASET SIZE")
print("=" * 70)

print(
    f"Train: {len(train_index):,}"
)

print(
    f"Val:   {len(val_index):,}"
)

print(
    f"Test:  {len(test_index):,}"
)


total_valid = (
    len(train_index)
    + len(val_index)
    + len(test_index)
)


print(
    f"Total valid: {total_valid:,}"
)


# ============================================================
# 17. VERIFY INDEX
# ============================================================

def verify_index(
    index,
    split_name
):

    invalid = 0

    label_counts = np.zeros(
        NUM_CLASSES,
        dtype=np.int64
    )


    for recording_id, local_idx in index:

        X, y, metadata = (
            store.load_recording(
                recording_id
            )
        )


        label = int(
            y[local_idx]
        )


        if (
            label < 0
            or label >= NUM_CLASSES
        ):

            invalid += 1

        else:

            label_counts[label] += 1


    print()
    print(
        f"{split_name} verification:"
    )

    print(
        f"  Invalid labels: {invalid}"
    )

    print(
        f"  Samples: {len(index):,}"
    )


    if invalid:

        raise RuntimeError(
            f"{split_name} contains "
            f"{invalid} invalid labels."
        )


    return label_counts


train_counts = verify_index(
    train_index,
    "Train"
)

val_counts = verify_index(
    val_index,
    "Validation"
)

test_counts = verify_index(
    test_index,
    "Test"
)


# ============================================================
# 18. CLASS DISTRIBUTION
# ============================================================

print()
print("=" * 70)
print("CLASS DISTRIBUTION")
print("=" * 70)


distribution_table = pd.DataFrame({

    "class_id":
        np.arange(NUM_CLASSES),

    "label":
        CLASS_LABELS,

    "train":
        train_counts,

    "validation":
        val_counts,

    "test":
        test_counts,
})


distribution_table["total"] = (
    distribution_table["train"]
    + distribution_table["validation"]
    + distribution_table["test"]
)


print(
    distribution_table.to_string(
        index=False
    )
)


# ============================================================
# 19. H3 NORMALIZATION
# ============================================================

def normalize_window(
    x
):

    x = np.asarray(
        x,
        dtype=np.float32
    )


    channel_mean = x.mean(
        axis=1,
        keepdims=True
    )


    channel_std = x.std(
        axis=1,
        keepdims=True
    )


    channel_std = np.maximum(
        channel_std,
        MIN_STD
    )


    x_norm = (
        x - channel_mean
    ) / channel_std


    return x_norm.astype(
        np.float32,
        copy=False
    )


# ============================================================
# 20. DATASET
# ============================================================

class EEGWindowDataset(Dataset):

    def __init__(
        self,
        index,
        store,
    ):

        self.index = index
        self.store = store


    def __len__(
        self
    ):

        return len(self.index)


    def __getitem__(
        self,
        idx
    ):

        recording_id, local_idx = (
            self.index[idx]
        )


        X, y, metadata = (
            self.store.load_recording(
                recording_id
            )
        )


        x = X[local_idx]


        label = int(
            y[local_idx]
        )


        if (
            label < 0
            or label >= NUM_CLASSES
        ):

            raise RuntimeError(
                f"Invalid label {label} "
                f"at {recording_id}, "
                f"index {local_idx}"
            )


        x = normalize_window(x)


        return (
            torch.from_numpy(x),
            torch.tensor(
                label,
                dtype=torch.long
            )
        )


train_dataset = EEGWindowDataset(
    train_index,
    store
)

val_dataset = EEGWindowDataset(
    val_index,
    store
)

test_dataset = EEGWindowDataset(
    test_index,
    store
)


# ============================================================
# 21. DATASET SAMPLE CHECK
# ============================================================

sample_x, sample_y = train_dataset[0]


print()
print("=" * 70)
print("DATASET SAMPLE CHECK")
print("=" * 70)

print(
    "Shape:",
    tuple(sample_x.shape)
)

print(
    "Dtype:",
    sample_x.dtype
)

print(
    "Label:",
    sample_y.item()
)

print(
    "Label name:",
    CLASS_TO_LABEL[sample_y.item()]
)

print(
    "Mean:",
    float(sample_x.mean())
)

print(
    "Std:",
    float(sample_x.std())
)

print(
    "Min:",
    float(sample_x.min())
)

print(
    "Max:",
    float(sample_x.max())
)


assert sample_x.shape == (
    N_CHANNELS,
    N_TIME
)

assert (
    0 <= sample_y.item() < NUM_CLASSES
)


print()
print(
    "Dataset validation: PASSED"
)


# ============================================================
# 22. DETERMINISTIC EPOCH PERMUTATION
# ============================================================

def epoch_permutation(
    n_samples,
    epoch
):

    generator = torch.Generator()

    generator.manual_seed(
        SEED + epoch
    )


    return torch.randperm(
        n_samples,
        generator=generator
    )


# ============================================================
# 23. EPOCH LOADER
# ============================================================

def make_epoch_loader(
    dataset,
    epoch,
    start_batch=0
):

    permutation = epoch_permutation(
        len(dataset),
        epoch
    )


    start_index = (
        start_batch * BATCH_SIZE
    )


    remaining_indices = (
        permutation[start_index:].tolist()
    )


    subset = Subset(
        dataset,
        remaining_indices
    )


    loader = DataLoader(
        subset,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=0,
        pin_memory=torch.cuda.is_available(),
        drop_last=False,
    )


    return loader


# ============================================================
# 24. IMPORT CORRECT NEURALTRAIN MODEL
# ============================================================

print()
print("=" * 70)
print("LOADING DECODER CONFIGURATION")
print("=" * 70)

try:
    from neuraltrain.models import SimpleConvTimeAgg

    print(
        "Successfully imported SimpleConvTimeAgg "
        "from neuraltrain.models"
    )

except Exception as exc:
    raise ImportError(
        "Could not import neuraltrain.models.SimpleConvTimeAgg"
    ) from exc


# ============================================================
# 25. DECODER CONFIGURATION
# ============================================================

decoder_config = {
    "hidden": 64,
    "depth": 4,
    "merger_config": None,
    "subject_layers_config": None,
    "initial_linear": 61,
    "backbone_out_channels": 2048,
    "time_agg_out": "att",
    "n_time_groups": None,
    "output_head_config": None,
}

print()
print("Decoder configuration:")
print(json.dumps(decoder_config, indent=2))


# ============================================================
# 26. CREATE CONFIGURATION OBJECT
# ============================================================

decoder_spec = SimpleConvTimeAgg(
    **decoder_config
)

print()
print(
    "SimpleConvTimeAgg configuration created."
)

print(
    "Configuration type:",
    type(decoder_spec)
)


# ============================================================
# 27. BUILD ACTUAL PYTORCH MODEL
# ============================================================

print()
print("=" * 70)
print("BUILDING ACTUAL DECODER MODEL")
print("=" * 70)

decoder = decoder_spec.build(
    n_in_channels=N_CHANNELS,
    n_outputs=2048,
)

print(
    "Decoder model type:",
    type(decoder)
)

if not isinstance(decoder, nn.Module):
    raise TypeError(
        "decoder_spec.build() did not return "
        "a torch.nn.Module. "
        f"Got: {type(decoder)}"
    )


# ============================================================
# 28. MOVE MODEL TO DEVICE
# ============================================================

decoder = decoder.to(DEVICE)

print(
    f"Decoder moved to {DEVICE}"
)


# ============================================================
# 29. INITIALIZE LAZY PARAMETERS
# ============================================================

print()
print("=" * 70)
print("INITIALIZING DECODER")
print("=" * 70)

decoder.eval()

with torch.no_grad():

    dummy_input = torch.zeros(
        2,
        N_CHANNELS,
        N_TIME,
        dtype=torch.float32,
        device=DEVICE,
    )

    dummy_output = decoder(
        dummy_input
    )


if isinstance(
    dummy_output,
    (tuple, list)
):
    dummy_output_tensor = dummy_output[0]
else:
    dummy_output_tensor = dummy_output


print(
    "Decoder output shape:",
    tuple(dummy_output_tensor.shape)
)


if dummy_output_tensor.ndim != 2:
    raise RuntimeError(
        "Expected decoder output with shape "
        "(batch, features), but got "
        f"{tuple(dummy_output_tensor.shape)}"
    )


decoder_output_dim = (
    dummy_output_tensor.shape[-1]
)


print(
    "Decoder feature dimension:",
    decoder_output_dim
)


if decoder_output_dim != 2048:
    raise RuntimeError(
        "Expected 2048 decoder features, "
        f"got {decoder_output_dim}"
    )


print(
    "Decoder forward check: PASSED"
)


# ============================================================
# 30. ATTACH 29-CLASS CLASSIFIER
# ============================================================

classifier = nn.Linear(
    decoder_output_dim,
    NUM_CLASSES
).to(DEVICE)


model = nn.Sequential(
    decoder,
    classifier
).to(DEVICE)


print()
print(
    "Classifier:"
)

print(
    f"{decoder_output_dim} -> {NUM_CLASSES}"
)


# ============================================================
# 31. FINAL MODEL FORWARD CHECK
# ============================================================

model.eval()

with torch.no_grad():

    check_input = torch.zeros(
        2,
        N_CHANNELS,
        N_TIME,
        dtype=torch.float32,
        device=DEVICE,
    )

    check_output = model(
        check_input
    )


if isinstance(
    check_output,
    (tuple, list)
):
    check_output = check_output[0]


print()
print(
    "Final model output shape:",
    tuple(check_output.shape)
)


if tuple(check_output.shape) != (
    2,
    NUM_CLASSES
):
    raise RuntimeError(
        "Expected final output shape "
        f"(2, {NUM_CLASSES}), got "
        f"{tuple(check_output.shape)}"
    )


print(
    "Final model forward check: PASSED"
)


# ============================================================
# 32. CHECK LAZY PARAMETERS
# ============================================================

def find_uninitialized_parameters(module):

    names = []

    for name, parameter in module.named_parameters():

        if isinstance(
            parameter,
            UninitializedParameter
        ):
            names.append(name)

    return names


uninitialized_parameters = (
    find_uninitialized_parameters(model)
)


print()
print("=" * 70)
print("LAZY PARAMETER CHECK")
print("=" * 70)


if uninitialized_parameters:

    print(
        "Uninitialized parameters:"
    )

    for name in uninitialized_parameters:
        print(
            f"  {name}"
        )

else:

    print(
        "No uninitialized parameters."
    )


# ============================================================
# 33. SAFE STATE DICT
# ============================================================

def safe_state_dict(module):

    raw_state = module.state_dict()

    safe_state = {}

    skipped = []

    for key, value in raw_state.items():

        if isinstance(
            value,
            (
                UninitializedParameter,
                UninitializedBuffer,
            )
        ):
            skipped.append(key)
            continue

        safe_state[key] = (
            value.detach()
            .cpu()
            .clone()
        )

    return safe_state, skipped


# ============================================================
# 34. SAFE MODEL LOAD
# ============================================================

def load_model_state(
    module,
    checkpoint
):

    result = module.load_state_dict(
        checkpoint["model_state_dict"],
        strict=False
    )

    allowed_missing = set(
        checkpoint.get(
            "skipped_uninitialized_model_keys",
            []
        )
    )

    actual_missing = set(
        result.missing_keys
    )

    unexpected_missing = (
        actual_missing
        - allowed_missing
    )

    if unexpected_missing:

        raise RuntimeError(
            "Unexpected missing model keys:\n"
            + "\n".join(
                sorted(unexpected_missing)
            )
        )

    if result.unexpected_keys:

        raise RuntimeError(
            "Unexpected checkpoint keys:\n"
            + "\n".join(
                sorted(result.unexpected_keys)
            )
        )

    print(
        "Model checkpoint loaded successfully."
    )


# ============================================================
# 35. LOSS + OPTIMIZER
# ============================================================

criterion = nn.CrossEntropyLoss()

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY,
)


# ============================================================
# 36. TRAINING STATE
# ============================================================

start_epoch = 0
start_batch = 0

best_val_loss = float("inf")

history = []

global_batch = 0


# ============================================================
# 37. RESUME CHECKPOINT
# ============================================================

if LATEST_CHECKPOINT.exists():

    print()
    print("=" * 70)
    print("EXISTING H3 CHECKPOINT FOUND")
    print("=" * 70)

    checkpoint = torch.load(
        LATEST_CHECKPOINT,
        map_location="cpu",
        weights_only=False,
    )

    load_model_state(
        model,
        checkpoint
    )

    if "optimizer_state_dict" in checkpoint:

        optimizer.load_state_dict(
            checkpoint["optimizer_state_dict"]
        )

        for state in optimizer.state.values():

            for key, value in state.items():

                if torch.is_tensor(value):

                    state[key] = value.to(
                        DEVICE
                    )

    start_epoch = int(
        checkpoint.get(
            "epoch",
            0
        )
    )

    start_batch = int(
        checkpoint.get(
            "batch_in_epoch",
            0
        )
    )

    best_val_loss = float(
        checkpoint.get(
            "best_val_loss",
            float("inf")
        )
    )

    global_batch = int(
        checkpoint.get(
            "global_batch",
            0
        )
    )

    history = checkpoint.get(
        "history",
        []
    )

    print(
        f"Resume epoch: {start_epoch}"
    )

    print(
        f"Resume batch: {start_batch}"
    )

    print(
        f"Global batch: {global_batch}"
    )

    print(
        f"Best validation loss: "
        f"{best_val_loss:.6f}"
    )

else:

    print()
    print(
        "No existing H3 checkpoint."
    )

    print(
        "Starting from epoch 0."
    )


# ============================================================
# 38. SAVE CHECKPOINT
# ============================================================

def save_checkpoint(
    path,
    epoch,
    batch_in_epoch,
    best_val_loss,
    global_batch,
):

    model_state, skipped_keys = (
        safe_state_dict(model)
    )

    checkpoint = {
        "phase": "5H.3",

        "timestamp":
            datetime.now().isoformat(),

        "epoch":
            int(epoch),

        "batch_in_epoch":
            int(batch_in_epoch),

        "global_batch":
            int(global_batch),

        "best_val_loss":
            float(best_val_loss),

        "model_state_dict":
            model_state,

        "optimizer_state_dict":
            optimizer.state_dict(),

        "skipped_uninitialized_model_keys":
            skipped_keys,

        "history":
            history,

        "config": {
            "seed": SEED,

            "batch_size":
                BATCH_SIZE,

            "num_epochs":
                NUM_EPOCHS,

            "learning_rate":
                LEARNING_RATE,

            "weight_decay":
                WEIGHT_DECAY,

            "normalization":
                "per-window channel-wise z-score",

            "channels":
                N_CHANNELS,

            "time_samples":
                N_TIME,

            "num_classes":
                NUM_CLASSES,

            "class_labels":
                CLASS_LABELS,

            "anomalous_labels_excluded":
                sorted(ANOMALOUS_LABELS),

            "anomalous_samples_excluded":
                len(anomalous_rows),
        },
    }

    torch.save(
        checkpoint,
        path
    )

    return skipped_keys


# ============================================================
# 39. VALIDATION
# ============================================================

def batch_accuracy(
    logits,
    targets
):

    predictions = logits.argmax(
        dim=1
    )

    correct = (
        predictions == targets
    ).sum().item()

    total = targets.numel()

    return correct, total


def evaluate(dataset):

    model.eval()

    loader = DataLoader(
        dataset,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=0,
        pin_memory=torch.cuda.is_available(),
        drop_last=False,
    )

    total_loss = 0.0
    total_correct = 0
    total_samples = 0

    with torch.no_grad():

        for X, y in loader:

            X = X.to(
                DEVICE,
                non_blocking=True
            )

            y = y.to(
                DEVICE,
                non_blocking=True
            )

            logits = model(X)

            if isinstance(
                logits,
                (tuple, list)
            ):
                logits = logits[0]

            loss = criterion(
                logits,
                y
            )

            correct, total = (
                batch_accuracy(
                    logits,
                    y
                )
            )

            total_loss += (
                loss.item() * total
            )

            total_correct += correct
            total_samples += total

    mean_loss = (
        total_loss
        / max(total_samples, 1)
    )

    accuracy = (
        total_correct
        / max(total_samples, 1)
    )

    return (
        mean_loss,
        accuracy,
        total_correct,
        total_samples,
    )


# ============================================================
# 40. INTERRUPTION LOG
# ============================================================

def save_interruption(
    reason,
    epoch,
    batch,
    global_batch,
    error_text,
):

    record = {
        "timestamp":
            datetime.now().isoformat(),

        "reason":
            reason,

        "epoch":
            int(epoch),

        "batch":
            int(batch),

        "global_batch":
            int(global_batch),

        "error":
            error_text,
    }

    existing = []

    if INTERRUPTION_FILE.exists():

        try:

            with open(
                INTERRUPTION_FILE,
                "r"
            ) as f:

                existing = json.load(f)

        except Exception:

            existing = []

    existing.append(record)

    with open(
        INTERRUPTION_FILE,
        "w"
    ) as f:

        json.dump(
            existing,
            f,
            indent=2
        )


# ============================================================
# 41. TRAINING LOOP
# ============================================================

print()
print("=" * 70)
print("STARTING PHASE 5H.3 TRAINING")
print("=" * 70)

training_start_time = time.time()


try:

    for epoch in range(
        start_epoch,
        NUM_EPOCHS
    ):

        epoch_start_time = time.time()

        print()
        print("=" * 70)
        print(
            f"EPOCH {epoch + 1}/{NUM_EPOCHS}"
        )
        print("=" * 70)


        if epoch == start_epoch:
            resume_batch = start_batch
        else:
            resume_batch = 0


        train_loader = make_epoch_loader(
            train_dataset,
            epoch,
            start_batch=resume_batch
        )


        model.train()

        epoch_loss = 0.0
        epoch_correct = 0
        epoch_samples = 0

        num_batches = len(
            train_loader
        )

        print(
            f"Training batches: {num_batches}"
        )


        for local_batch_idx, (
            X,
            y
        ) in enumerate(train_loader):

            batch_idx = (
                resume_batch
                + local_batch_idx
            )


            X = X.to(
                DEVICE,
                non_blocking=True
            )

            y = y.to(
                DEVICE,
                non_blocking=True
            )


            optimizer.zero_grad(
                set_to_none=True
            )


            logits = model(X)


            if isinstance(
                logits,
                (tuple, list)
            ):
                logits = logits[0]


            if (
                logits.ndim != 2
                or
                logits.shape[1] != NUM_CLASSES
            ):

                raise RuntimeError(
                    "Unexpected model output "
                    f"shape: {logits.shape}"
                )


            loss = criterion(
                logits,
                y
            )


            loss.backward()

            optimizer.step()


            loss_value = loss.item()


            correct, total = (
                batch_accuracy(
                    logits,
                    y
                )
            )


            epoch_loss += (
                loss_value * total
            )

            epoch_correct += correct
            epoch_samples += total

            global_batch += 1


            if (
                (batch_idx + 1) % 50 == 0
                or
                batch_idx == resume_batch
            ):

                running_loss = (
                    epoch_loss
                    / max(epoch_samples, 1)
                )

                running_accuracy = (
                    epoch_correct
                    / max(epoch_samples, 1)
                )

                print(
                    f"Epoch {epoch + 1} | "
                    f"Batch {batch_idx + 1}/{num_batches} | "
                    f"Loss {running_loss:.5f} | "
                    f"Acc {running_accuracy:.5f}"
                )


            # ------------------------------------------------
            # Emergency checkpoint
            # ------------------------------------------------

            if (
                global_batch
                % EMERGENCY_EVERY_BATCHES
                == 0
            ):

                next_batch = batch_idx + 1


                if next_batch >= num_batches:

                    checkpoint_epoch = epoch + 1
                    checkpoint_batch = 0

                else:

                    checkpoint_epoch = epoch
                    checkpoint_batch = next_batch


                skipped_keys = save_checkpoint(
                    LATEST_CHECKPOINT,
                    checkpoint_epoch,
                    checkpoint_batch,
                    best_val_loss,
                    global_batch,
                )


                print()
                print(
                    "[EMERGENCY CHECKPOINT SAVED]"
                )

                print(
                    f"Epoch: {checkpoint_epoch}"
                )

                print(
                    f"Batch: {checkpoint_batch}"
                )

                print(
                    f"Global batch: {global_batch}"
                )


                if skipped_keys:

                    print(
                        "Skipped uninitialized "
                        "parameters:"
                    )

                    for key in skipped_keys:

                        print(
                            f"  {key}"
                        )


        # ====================================================
        # END OF EPOCH
        # ====================================================

        train_loss = (
            epoch_loss
            / max(epoch_samples, 1)
        )

        train_accuracy = (
            epoch_correct
            / max(epoch_samples, 1)
        )


        print()
        print(
            "Running validation..."
        )


        (
            val_loss,
            val_accuracy,
            val_correct,
            val_total,
        ) = evaluate(
            val_dataset
        )


        epoch_time = (
            time.time()
            - epoch_start_time
        )


        record = {
            "epoch":
                epoch + 1,

            "train_loss":
                train_loss,

            "train_accuracy":
                train_accuracy,

            "val_loss":
                val_loss,

            "val_accuracy":
                val_accuracy,

            "train_correct":
                epoch_correct,

            "train_samples":
                epoch_samples,

            "val_correct":
                val_correct,

            "val_samples":
                val_total,

            "epoch_time_seconds":
                epoch_time,

            "global_batch":
                global_batch,
        }


        history.append(record)


        pd.DataFrame(
            history
        ).to_csv(
            HISTORY_FILE,
            index=False
        )


        # ----------------------------------------------------
        # Best checkpoint
        # ----------------------------------------------------

        if val_loss < best_val_loss:

            best_val_loss = val_loss


            save_checkpoint(
                BEST_CHECKPOINT,
                epoch + 1,
                0,
                best_val_loss,
                global_batch,
            )


            print()
            print(
                "*** NEW BEST MODEL ***"
            )

            print(
                f"Best validation loss: "
                f"{best_val_loss:.6f}"
            )


        # ----------------------------------------------------
        # Latest checkpoint
        # ----------------------------------------------------

        save_checkpoint(
            LATEST_CHECKPOINT,
            epoch + 1,
            0,
            best_val_loss,
            global_batch,
        )


        print()
        print(
            "Epoch checkpoint saved."
        )

        print(
            f"Train loss: {train_loss:.6f}"
        )

        print(
            f"Train accuracy: {train_accuracy:.6f}"
        )

        print(
            f"Validation loss: {val_loss:.6f}"
        )

        print(
            f"Validation accuracy: {val_accuracy:.6f}"
        )

        print(
            f"Epoch time: "
            f"{epoch_time / 60:.2f} min"
        )

        print(
            f"Best validation loss: "
            f"{best_val_loss:.6f}"
        )


        start_batch = 0


# ============================================================
# 42. MANUAL INTERRUPTION
# ============================================================

except KeyboardInterrupt:

    print()
    print("=" * 70)
    print("TRAINING INTERRUPTED")
    print("=" * 70)


    save_interruption(
        reason="KeyboardInterrupt",
        epoch=epoch,
        batch=batch_idx,
        global_batch=global_batch,
        error_text="Manual interruption",
    )


    next_batch = batch_idx + 1


    if next_batch >= num_batches:

        save_epoch = epoch + 1
        save_batch = 0

    else:

        save_epoch = epoch
        save_batch = next_batch


    save_checkpoint(
        LATEST_CHECKPOINT,
        save_epoch,
        save_batch,
        best_val_loss,
        global_batch,
    )


    print(
        "Resume checkpoint saved."
    )


# ============================================================
# 43. TRAINING EXCEPTION
# ============================================================

except Exception:

    print()
    print("=" * 70)
    print("TRAINING FAILED")
    print("=" * 70)


    error_text = traceback.format_exc()

    print(
        error_text
    )


    save_interruption(
        reason="Exception",
        epoch=epoch,
        batch=batch_idx,
        global_batch=global_batch,
        error_text=error_text,
    )


    try:

        next_batch = batch_idx + 1


        if next_batch >= num_batches:

            save_epoch = epoch + 1
            save_batch = 0

        else:

            save_epoch = epoch
            save_batch = next_batch


        save_checkpoint(
            LATEST_CHECKPOINT,
            save_epoch,
            save_batch,
            best_val_loss,
            global_batch,
        )


        print(
            "Emergency checkpoint saved."
        )


    except Exception:

        print(
            "Emergency checkpoint save failed:"
        )

        print(
            traceback.format_exc()
        )


    raise


# ============================================================
# 44. FINAL SUMMARY
# ============================================================

elapsed = (
    time.time()
    - training_start_time
)


print()
print("=" * 70)
print("PHASE 5H.3 TRAINING FINISHED")
print("=" * 70)


print(
    f"Runtime: {elapsed / 3600:.2f} hours"
)

print(
    f"Best validation loss: "
    f"{best_val_loss:.6f}"
)

print()
print(
    "Latest checkpoint:"
)

print(
    LATEST_CHECKPOINT
)

print()
print(
    "Best checkpoint:"
)

print(
    BEST_CHECKPOINT
)

print()
print(
    "Training history:"
)

print(
    HISTORY_FILE
)

print()
print(
    "Anomaly audit:"
)

print(
    ANOMALY_AUDIT_FILE
)


# ============================================================
# 45. HISTORY
# ============================================================

if HISTORY_FILE.exists():

    final_history = pd.read_csv(
        HISTORY_FILE
    )

    print()
    print("=" * 70)
    print("TRAINING HISTORY")
    print("=" * 70)

    print(
        final_history.to_string(
            index=False
        )
    )


# ============================================================
# 46. GPU MEMORY
# ============================================================

if torch.cuda.is_available():

    print()
    print("=" * 70)
    print("GPU MEMORY")
    print("=" * 70)

    print(
        f"Allocated: "
        f"{torch.cuda.memory_allocated() / 1024**3:.2f} GB"
    )

    print(
        f"Reserved: "
        f"{torch.cuda.memory_reserved() / 1024**3:.2f} GB"
    )


# ============================================================
# 47. FINAL STATUS
# ============================================================

print()
print("=" * 70)
print("PHASE 5H.3 COMPLETE")
print("=" * 70)

print(
    f"Valid training samples: "
    f"{len(train_index):,}"
)

print(
    f"Valid validation samples: "
    f"{len(val_index):,}"
)

print(
    f"Valid test samples: "
    f"{len(test_index):,}"
)

print(
    f"Excluded anomalous samples: "
    f"{len(anomalous_rows)}"
)

print(
    f"Best validation loss: "
    f"{best_val_loss:.6f}"
)

print()
print(
    "Best checkpoint:"
)

print(
    BEST_CHECKPOINT
)
```

```python
# ============================================================
# PHASE 5H.4 — TEST EVALUATION
# Correct Phase 5 recording path
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

from neuraltrain.models.simpleconv import (
    SimpleConvTimeAgg,
    SimpleConvTimeAggModel,
)

# ============================================================
# 1. CONFIGURATION
# ============================================================

PROJECT_ROOT = "/content/gdrive/MyDrive/NeuroSentinel"

CHECKPOINT_PATH = (
    f"{PROJECT_ROOT}/checkpoints/phase5h3/"
    f"phase5h3_best.pt"
)

MANIFEST_PATH = (
    f"{PROJECT_ROOT}/07_phase6_subject_split/"
    f"phase6_subject_split_manifest.csv"
)

RECORDING_DIR = (
    f"{PROJECT_ROOT}/06_phase5_utility_labels/"
    f"recordings"
)

RESULTS_DIR = (
    f"{PROJECT_ROOT}/results/phase5h3"
)

os.makedirs(
    RESULTS_DIR,
    exist_ok=True
)

RESULT_PATH = (
    f"{RESULTS_DIR}/phase5h4_test_results.json"
)

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

BATCH_SIZE = 8

NUM_CLASSES = 29
IN_CHANNELS = 61
TIME_SAMPLES = 25
EMBEDDING_DIM = 2048

CLASS_NAMES = (
    list("abcdefghijklmnopqrstuvwxyz")
    + ["<space>", "<number>", "<special>"]
)

LABEL_TO_ID = {
    label: idx
    for idx, label in enumerate(CLASS_NAMES)
}

ANOMALOUS_LABELS = {
    "-",
    "`",
}

print("=" * 70)
print("PHASE 5H.4 — TEST EVALUATION")
print("=" * 70)

print("Device:", DEVICE)


# ============================================================
# 2. VERIFY PATHS
# ============================================================

for path in [
    CHECKPOINT_PATH,
    MANIFEST_PATH,
    RECORDING_DIR,
]:

    if not os.path.exists(path):

        raise FileNotFoundError(
            f"Required path not found:\n{path}"
        )

print("\nPaths verified.")


# ============================================================
# 3. LOAD CHECKPOINT
# ============================================================

checkpoint = torch.load(
    CHECKPOINT_PATH,
    map_location="cpu",
    weights_only=False
)

state_dict = checkpoint[
    "model_state_dict"
]

print("\nCheckpoint epoch:")
print(checkpoint["epoch"])

print(
    "Best validation loss:",
    checkpoint["best_val_loss"]
)

print(
    "\nCheckpoint configuration:"
)

print(
    checkpoint["config"]
)


# ============================================================
# 4. LOAD FROZEN PHASE 6 SPLIT
# ============================================================

manifest = pd.read_csv(
    MANIFEST_PATH
)

print(
    "\nManifest:",
    manifest.shape
)

test_manifest = manifest[
    manifest["split"].astype(str).str.lower()
    == "test"
].copy()

print(
    "Test recordings:",
    len(test_manifest)
)

print(
    "\nTest recording IDs:"
)

for rid in test_manifest[
    "recording_id"
].astype(str):

    print(" ", rid)


# ============================================================
# 5. REBUILD TEST DATASET
# ============================================================

test_windows = []
test_labels = []

valid_samples = 0
excluded_samples = 0

print("\n" + "=" * 70)
print("LOADING TEST RECORDINGS")
print("=" * 70)

for _, row in test_manifest.iterrows():

    recording_id = str(
        row["recording_id"]
    )

    window_path = os.path.join(
        RECORDING_DIR,
        f"{recording_id}_windows.npy"
    )

    metadata_path = os.path.join(
        RECORDING_DIR,
        f"{recording_id}_metadata.csv"
    )

    if not os.path.exists(
        window_path
    ):

        raise FileNotFoundError(
            f"Missing windows:\n{window_path}"
        )

    if not os.path.exists(
        metadata_path
    ):

        raise FileNotFoundError(
            f"Missing metadata:\n{metadata_path}"
        )

    windows = np.load(
        window_path,
        mmap_mode="r"
    )

    metadata = pd.read_csv(
        metadata_path
    )

    if len(windows) != len(metadata):

        raise RuntimeError(
            f"Mismatch for {recording_id}: "
            f"{len(windows)} windows vs "
            f"{len(metadata)} metadata rows"
        )

    print(
        f"{recording_id:20s} "
        f"{len(windows):5d} windows"
    )

    for idx in range(
        len(windows)
    ):

        target = str(
            metadata.iloc[idx]["target"]
        )

        # Exclude known anomalies
        if target in ANOMALOUS_LABELS:

            excluded_samples += 1
            continue

        if target not in LABEL_TO_ID:

            raise ValueError(
                f"Unexpected target "
                f"{target!r} in "
                f"{recording_id}, row {idx}"
            )

        window = np.asarray(
            windows[idx],
            dtype=np.float32
        )

        if window.shape != (
            IN_CHANNELS,
            TIME_SAMPLES
        ):

            raise RuntimeError(
                f"Unexpected shape "
                f"{window.shape} in "
                f"{recording_id}, row {idx}"
            )

        # ----------------------------------------------------
        # SAME NORMALIZATION USED IN PHASE 5H.3
        # Per-window channel-wise z-score
        # ----------------------------------------------------

        mean = window.mean(
            axis=1,
            keepdims=True
        )

        std = window.std(
            axis=1,
            keepdims=True
        )

        std = np.maximum(
            std,
            1e-8
        )

        window = (
            window - mean
        ) / std

        window = np.nan_to_num(
            window,
            nan=0.0,
            posinf=0.0,
            neginf=0.0
        )

        test_windows.append(
            window
        )

        test_labels.append(
            LABEL_TO_ID[target]
        )

        valid_samples += 1


# ============================================================
# 6. CONVERT TO ARRAYS
# ============================================================

test_windows = np.asarray(
    test_windows,
    dtype=np.float32
)

test_labels = np.asarray(
    test_labels,
    dtype=np.int64
)

print("\n" + "=" * 70)
print("TEST DATASET")
print("=" * 70)

print(
    "Valid samples:",
    valid_samples
)

print(
    "Excluded samples:",
    excluded_samples
)

print(
    "Windows:",
    test_windows.shape
)

print(
    "Labels:",
    test_labels.shape
)

if valid_samples != 29907:

    raise RuntimeError(
        f"Expected 29,907 valid test "
        f"samples, got {valid_samples}"
    )

if excluded_samples != 2:

    raise RuntimeError(
        f"Expected 2 excluded test "
        f"samples, got {excluded_samples}"
    )


# ============================================================
# 7. DATASET
# ============================================================

class TestDataset(Dataset):

    def __init__(
        self,
        windows,
        labels
    ):

        self.windows = windows
        self.labels = labels

    def __len__(self):

        return len(self.labels)

    def __getitem__(
        self,
        index
    ):

        return (
            torch.from_numpy(
                self.windows[index]
            ),
            torch.tensor(
                self.labels[index],
                dtype=torch.long
            ),
        )


test_dataset = TestDataset(
    test_windows,
    test_labels
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0,
    pin_memory=(
        DEVICE.type == "cuda"
    ),
)

print(
    "\nTest batches:",
    len(test_loader)
)


# ============================================================
# 8. RECREATE EXACT MODEL CONFIG
# ============================================================

model_config = SimpleConvTimeAgg(
    hidden=64,
    depth=4,
    kernel_size=5,
    dilation_growth=2,
    merger_config=None,
    initial_linear=61,
    backbone_out_channels=2048,
    time_agg_out="att",
    n_time_groups=None,
    output_head_config=None,
)

feature_model = SimpleConvTimeAggModel(
    in_channels=61,
    out_channels=2048,
    config=model_config,
)


# ============================================================
# 9. INITIALIZE LAZY ATTENTION PARAMETERS
# ============================================================

init_x, _ = next(
    iter(test_loader)
)

feature_model.eval()

with torch.no_grad():

    init_embedding = feature_model(
        init_x.float()
    )

print(
    "\nEmbedding shape:",
    tuple(init_embedding.shape)
)

if init_embedding.shape[1] != 2048:

    raise RuntimeError(
        "Unexpected embedding dimension."
    )


# ============================================================
# 10. RECREATE EXACT SEQUENTIAL MODEL
# ============================================================

classifier = nn.Linear(
    2048,
    29
)

model = nn.Sequential(
    feature_model,
    classifier
)

# Initialize complete model once
with torch.no_grad():

    init_logits = model(
        init_x.float()
    )

print(
    "Initial logits:",
    tuple(init_logits.shape)
)


# ============================================================
# 11. CHECK STATE-DICT COMPATIBILITY
# ============================================================

model_keys = set(
    model.state_dict().keys()
)

checkpoint_keys = set(
    state_dict.keys()
)

missing = (
    model_keys - checkpoint_keys
)

unexpected = (
    checkpoint_keys - model_keys
)

allowed_missing = set(
    checkpoint.get(
        "skipped_uninitialized_model_keys",
        []
    )
)

unexpected_missing = (
    missing - allowed_missing
)

print("\n" + "=" * 70)
print("CHECKPOINT COMPATIBILITY")
print("=" * 70)

print(
    "Model keys:",
    len(model_keys)
)

print(
    "Checkpoint keys:",
    len(checkpoint_keys)
)

print("\nMissing keys:")

for key in sorted(missing):

    print(" ", key)

print("\nUnexpected keys:")

for key in sorted(unexpected):

    print(" ", key)

if unexpected_missing:

    raise RuntimeError(
        "Unexpected missing keys:\n"
        + "\n".join(
            sorted(unexpected_missing)
        )
    )

if unexpected:

    raise RuntimeError(
        "Unexpected checkpoint keys:\n"
        + "\n".join(
            sorted(unexpected)
        )
    )

print(
    "\nArchitecture compatibility: PASS"
)


# ============================================================
# 12. LOAD CHECKPOINT
# ============================================================

load_result = model.load_state_dict(
    state_dict,
    strict=False
)

print(
    "\nMissing after load:",
    load_result.missing_keys
)

print(
    "Unexpected after load:",
    load_result.unexpected_keys
)

unexpected_after_load = (
    set(load_result.missing_keys)
    - allowed_missing
)

if unexpected_after_load:

    raise RuntimeError(
        "Unexpected missing keys after load:\n"
        + "\n".join(
            sorted(unexpected_after_load)
        )
    )

if load_result.unexpected_keys:

    raise RuntimeError(
        "Unexpected checkpoint keys after load."
    )


# ============================================================
# 13. MOVE MODEL TO GPU
# ============================================================

model = model.to(
    DEVICE
)

model.eval()


# ============================================================
# 14. TEST EVALUATION
# ============================================================

criterion = nn.CrossEntropyLoss()

total_loss = 0.0
total_correct = 0
total_samples = 0

all_predictions = []
all_targets = []
all_confidences = []

print("\n" + "=" * 70)
print("RUNNING TEST EVALUATION")
print("=" * 70)

with torch.no_grad():

    for batch_idx, (
        x,
        y
    ) in enumerate(test_loader):

        x = x.to(
            DEVICE,
            non_blocking=True
        )

        y = y.to(
            DEVICE,
            non_blocking=True
        )

        logits = model(x)

        loss = criterion(
            logits,
            y
        )

        probabilities = torch.softmax(
            logits,
            dim=1
        )

        confidence, prediction = (
            torch.max(
                probabilities,
                dim=1
            )
        )

        correct = prediction.eq(y)

        n = y.size(0)

        total_loss += (
            loss.item() * n
        )

        total_correct += (
            correct.sum().item()
        )

        total_samples += n

        all_predictions.extend(
            prediction.cpu()
            .numpy()
            .tolist()
        )

        all_targets.extend(
            y.cpu()
            .numpy()
            .tolist()
        )

        all_confidences.extend(
            confidence.cpu()
            .numpy()
            .tolist()
        )

        if (
            batch_idx == 0
            or
            (batch_idx + 1) % 500 == 0
            or
            batch_idx == len(test_loader) - 1
        ):

            print(
                f"Batch "
                f"{batch_idx + 1}/"
                f"{len(test_loader)} | "
                f"Loss "
                f"{total_loss / total_samples:.6f} | "
                f"Acc "
                f"{total_correct / total_samples:.6f}"
            )


# ============================================================
# 15. METRICS
# ============================================================

predictions = np.asarray(
    all_predictions,
    dtype=np.int64
)

targets = np.asarray(
    all_targets,
    dtype=np.int64
)

confidences = np.asarray(
    all_confidences,
    dtype=np.float64
)

test_loss = (
    total_loss /
    total_samples
)

test_accuracy = (
    total_correct /
    total_samples
)

prediction_counts = np.bincount(
    predictions,
    minlength=29
)

target_counts = np.bincount(
    targets,
    minlength=29
)

unique_predicted_classes = int(
    np.count_nonzero(
        prediction_counts
    )
)

dominant_class = int(
    np.argmax(
        prediction_counts
    )
)

dominant_count = int(
    prediction_counts[
        dominant_class
    ]
)

dominant_fraction = (
    dominant_count /
    total_samples
)

collapse_detected = (
    unique_predicted_classes == 1
)

correct_mask = (
    predictions == targets
)

correct_confidences = (
    confidences[
        correct_mask
    ]
)

incorrect_confidences = (
    confidences[
        ~correct_mask
    ]
)


# ============================================================
# 16. CONFUSION MATRIX
# ============================================================

confusion_matrix = np.zeros(
    (29, 29),
    dtype=np.int64
)

for true_label, pred_label in zip(
    targets,
    predictions
):

    confusion_matrix[
        true_label,
        pred_label
    ] += 1


# ============================================================
# 17. PRINT RESULTS
# ============================================================

print("\n" + "=" * 70)
print("PHASE 5H.4 RESULTS")
print("=" * 70)

print(
    f"Test samples             : "
    f"{total_samples}"
)

print(
    f"Test loss                : "
    f"{test_loss:.6f}"
)

print(
    f"Test accuracy            : "
    f"{test_accuracy:.6f}"
)

print(
    f"Test accuracy (%)        : "
    f"{test_accuracy * 100:.3f}%"
)

print(
    f"Mean confidence          : "
    f"{confidences.mean():.6f}"
)

print(
    f"Median confidence        : "
    f"{np.median(confidences):.6f}"
)

if len(correct_confidences):

    print(
        f"Correct mean confidence : "
        f"{correct_confidences.mean():.6f}"
    )

if len(incorrect_confidences):

    print(
        f"Incorrect mean confidence: "
        f"{incorrect_confidences.mean():.6f}"
    )

print(
    f"\nUnique predicted classes : "
    f"{unique_predicted_classes}/29"
)

print(
    f"Dominant class           : "
    f"{dominant_class} "
    f"({CLASS_NAMES[dominant_class]})"
)

print(
    f"Dominant count           : "
    f"{dominant_count}"
)

print(
    f"Dominant fraction        : "
    f"{dominant_fraction * 100:.3f}%"
)

print(
    f"Decoder collapse         : "
    f"{'YES' if collapse_detected else 'NO'}"
)


# ============================================================
# 18. PREDICTION DISTRIBUTION
# ============================================================

print("\nPrediction distribution:")
print("-" * 70)

for class_id in range(29):

    count = prediction_counts[
        class_id
    ]

    if count > 0:

        print(
            f"{class_id:2d} "
            f"{CLASS_NAMES[class_id]:9s} "
            f"{count:6d} "
            f"({count / total_samples * 100:7.3f}%)"
        )


# ============================================================
# 19. SAVE RESULTS
# ============================================================

results = {

    "phase": "5H.4",

    "checkpoint": CHECKPOINT_PATH,

    "epoch": int(
        checkpoint["epoch"]
    ),

    "best_validation_loss": float(
        checkpoint["best_val_loss"]
    ),

    "test_samples": int(
        total_samples
    ),

    "test_loss": float(
        test_loss
    ),

    "test_accuracy": float(
        test_accuracy
    ),

    "test_accuracy_percent": float(
        test_accuracy * 100
    ),

    "mean_confidence": float(
        confidences.mean()
    ),

    "median_confidence": float(
        np.median(confidences)
    ),

    "correct_mean_confidence": (
        float(
            correct_confidences.mean()
        )
        if len(correct_confidences)
        else None
    ),

    "incorrect_mean_confidence": (
        float(
            incorrect_confidences.mean()
        )
        if len(incorrect_confidences)
        else None
    ),

    "unique_predicted_classes": (
        unique_predicted_classes
    ),

    "dominant_class_id": (
        dominant_class
    ),

    "dominant_class": (
        CLASS_NAMES[
            dominant_class
        ]
    ),

    "dominant_count": (
        dominant_count
    ),

    "dominant_fraction": (
        float(dominant_fraction)
    ),

    "collapse_detected": (
        bool(collapse_detected)
    ),

    "prediction_distribution": {
        CLASS_NAMES[i]: int(
            prediction_counts[i]
        )
        for i in range(29)
    },

    "target_distribution": {
        CLASS_NAMES[i]: int(
            target_counts[i]
        )
        for i in range(29)
    },

    "confusion_matrix": (
        confusion_matrix.tolist()
    ),

    "excluded_anomalous_samples": (
        int(excluded_samples)
    ),
}

with open(
    RESULT_PATH,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        results,
        f,
        indent=2
    )


# ============================================================
# 20. FINAL
# ============================================================

print("\n" + "=" * 70)
print("PHASE 5H.4 COMPLETE")
print("=" * 70)

print(
    f"Valid test samples : "
    f"{total_samples}"
)

print(
    f"Test loss          : "
    f"{test_loss:.6f}"
)

print(
    f"Test accuracy      : "
    f"{test_accuracy * 100:.3f}%"
)

print(
    f"Unique classes     : "
    f"{unique_predicted_classes}/29"
)

print(
    f"Collapse           : "
    f"{'YES' if collapse_detected else 'NO'}"
)

print(
    "\nResults saved to:"
)

print(
    RESULT_PATH
)

print("=" * 70)
```

```python
# ================================================================
# PHASE 5H.5 — DECODER UTILITY / CONFIDENCE VALIDATION
# ================================================================
#
# Purpose:
#   Validate whether decoder confidence is informative about
#   correctness before generating utility labels for the causal gate.
#
# No model retraining.
# Uses:
#   - Phase 5H.3 normalized decoder checkpoint
#   - Frozen Phase 6 subject-wise test split
#   - Existing recording-level EEG windows + metadata
#
# Outputs:
#   - detailed per-window predictions
#   - per-class metrics
#   - confusion matrix
#   - confidence-bin analysis
#   - ECE
#   - threshold / coverage / accuracy analysis
#   - JSON summary
# ================================================================

import os
import json
import math
import random
import warnings
from pathlib import Path
from collections import Counter

import numpy as np
import pandas as pd

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_recall_fscore_support,
    confusion_matrix,
    classification_report,
)

warnings.filterwarnings("ignore")

print("=" * 70)
print("PHASE 5H.5 — DECODER UTILITY / CONFIDENCE VALIDATION")
print("=" * 70)

# ================================================================
# 1. CONFIGURATION
# ================================================================

SEED = 42

PROJECT_ROOT = Path("/content/gdrive/MyDrive/NeuroSentinel")

RECORDINGS_DIR = (
    PROJECT_ROOT /
    "06_phase5_utility_labels" /
    "recordings"
)

CHECKPOINT_PATH = (
    PROJECT_ROOT /
    "checkpoints" /
    "phase5h3" /
    "phase5h3_best.pt"
)

MANIFEST_PATH = (
    PROJECT_ROOT /
    "07_phase6_subject_split" /
    "phase6_subject_split_manifest.csv"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "results" /
    "phase5h5"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

RESULTS_JSON = OUTPUT_DIR / "phase5h5_decoder_utility_validation.json"
PREDICTIONS_CSV = OUTPUT_DIR / "phase5h5_test_predictions.csv"
CONFUSION_CSV = OUTPUT_DIR / "phase5h5_confusion_matrix.csv"
CLASS_METRICS_CSV = OUTPUT_DIR / "phase5h5_class_metrics.csv"
CONFIDENCE_BINS_CSV = OUTPUT_DIR / "phase5h5_confidence_bins.csv"
THRESHOLD_CSV = OUTPUT_DIR / "phase5h5_threshold_analysis.csv"

WINDOWS_PER_BATCH = 8
NUM_WORKERS = 0

CLASS_LABELS = [
    "a", "b", "c", "d", "e", "f", "g", "h", "i", "j",
    "k", "l", "m", "n", "o", "p", "q", "r", "s", "t",
    "u", "v", "w", "x", "y", "z",
    "<space>", "<number>", "<special>"
]

NUM_CLASSES = 29
CHANNELS = 61
TIME_SAMPLES = 25

ANOMALOUS_LABELS = {"-", "`"}

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Reproducibility
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

print(f"Device: {DEVICE}")
print(f"Project root: {PROJECT_ROOT}")
print(f"Recordings directory: {RECORDINGS_DIR}")
print(f"Checkpoint: {CHECKPOINT_PATH}")

# ================================================================
# 2. VERIFY PATHS
# ================================================================

required_paths = {
    "recordings_dir": RECORDINGS_DIR,
    "checkpoint": CHECKPOINT_PATH,
    "manifest": MANIFEST_PATH,
}

for name, path in required_paths.items():
    if not path.exists():
        raise FileNotFoundError(f"Missing {name}: {path}")

print("\nPaths verified.")

# ================================================================
# 3. IMPORT EXACT MODEL ARCHITECTURE
# ================================================================

from neuraltrain.models.simpleconv import (
    SimpleConvTimeAgg,
    SimpleConvTimeAggModel,
)

print("\nModel classes imported.")

# ================================================================
# 4. LOAD MANIFEST
# ================================================================

manifest = pd.read_csv(MANIFEST_PATH)

print("\nManifest:")
print(manifest.shape)
print(manifest["split"].value_counts())

test_manifest = manifest[
    manifest["split"].astype(str).str.lower() == "test"
].copy()

test_manifest = test_manifest.sort_values("recording_id").reset_index(drop=True)

test_recordings = test_manifest["recording_id"].astype(str).tolist()

print(f"\nTest recordings: {len(test_recordings)}")

for rid in test_recordings:
    print(" ", rid)

# ================================================================
# 5. CHARACTER → CLASS MAPPING
# ================================================================

char_to_class = {
    ch: idx
    for idx, ch in enumerate("abcdefghijklmnopqrstuvwxyz")
}

char_to_class[" "] = 26

# Existing dataset convention
# <number> = 27
# <special> = 28
#
# The metadata may encode these using explicit token strings.

SPECIAL_TOKEN_MAP = {
    "<space>": 26,
    "<number>": 27,
    "<special>": 28,
}

def target_to_class(value):
    """
    Convert metadata target/text label into the 29-class decoder target.
    Returns None for known anomalous labels.
    """

    if pd.isna(value):
        return None

    value = str(value)

    if value in ANOMALOUS_LABELS:
        return None

    if value in SPECIAL_TOKEN_MAP:
        return SPECIAL_TOKEN_MAP[value]

    if value == " ":
        return 26

    if len(value) == 1 and value.lower() in char_to_class:
        return char_to_class[value.lower()]

    # Numeric token conventions
    if value.lower() in {"<number>", "number"}:
        return 27

    if value.lower() in {"<special>", "special"}:
        return 28

    # If metadata uses a numeric target directly
    try:
        integer_value = int(value)

        if 0 <= integer_value < NUM_CLASSES:
            return integer_value
    except Exception:
        pass

    return None


# ================================================================
# 6. DATASET
# ================================================================

class EEGTestDataset(Dataset):

    def __init__(self, windows, labels):

        self.windows = windows
        self.labels = labels

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):

        x = self.windows[idx].astype(np.float32)
        y = int(self.labels[idx])

        # Phase 5H.3 normalization:
        # per-window, channel-wise z-score
        #
        # x shape = [61, 25]
        mean = x.mean(axis=1, keepdims=True)
        std = x.std(axis=1, keepdims=True)

        std = np.where(std < 1e-8, 1.0, std)

        x = (x - mean) / std

        return (
            torch.from_numpy(x),
            torch.tensor(y, dtype=torch.long),
        )


# ================================================================
# 7. LOAD TEST RECORDINGS
# ================================================================

print("\n" + "=" * 70)
print("LOADING TEST RECORDINGS")
print("=" * 70)

all_windows = []
all_labels = []
all_recording_ids = []

excluded_samples = 0

for recording_id in test_recordings:

    windows_path = RECORDINGS_DIR / f"{recording_id}_windows.npy"
    metadata_path = RECORDINGS_DIR / f"{recording_id}_metadata.csv"

    if not windows_path.exists():
        raise FileNotFoundError(windows_path)

    if not metadata_path.exists():
        raise FileNotFoundError(metadata_path)

    windows = np.load(windows_path)
    metadata = pd.read_csv(metadata_path)

    if len(windows) != len(metadata):
        raise ValueError(
            f"{recording_id}: "
            f"{len(windows)} windows vs "
            f"{len(metadata)} metadata rows"
        )

    labels = []

    valid_indices = []

    for idx, row in metadata.iterrows():

        # Prefer target column if present.
        if "target" in metadata.columns:
            raw_target = row["target"]

        elif "label" in metadata.columns:
            raw_target = row["label"]

        elif "text" in metadata.columns:
            raw_target = row["text"]

        else:
            raise ValueError(
                f"No target/label/text column found in {metadata_path}"
            )

        target = target_to_class(raw_target)

        if target is None:
            excluded_samples += 1
            continue

        labels.append(target)
        valid_indices.append(idx)

    windows_valid = windows[valid_indices]

    if len(windows_valid) != len(labels):
        raise RuntimeError(
            f"Label/window mismatch in {recording_id}"
        )

    all_windows.append(windows_valid)
    all_labels.extend(labels)
    all_recording_ids.extend(
        [recording_id] * len(labels)
    )

    print(
        f"{recording_id:20s} "
        f"{len(windows_valid):5d} valid windows"
    )


# Concatenate
X_test = np.concatenate(all_windows, axis=0)
y_test = np.asarray(all_labels, dtype=np.int64)
recording_ids = np.asarray(all_recording_ids)

print("\n" + "=" * 70)
print("TEST DATASET")
print("=" * 70)

print("Valid samples :", len(y_test))
print("Excluded      :", excluded_samples)
print("Windows shape :", X_test.shape)
print("Labels shape  :", y_test.shape)

if X_test.shape[1:] != (CHANNELS, TIME_SAMPLES):
    raise ValueError(
        f"Unexpected window shape: {X_test.shape}"
    )

# ================================================================
# 8. MODEL RECONSTRUCTION
# ================================================================

config = SimpleConvTimeAgg(
    hidden=64,
    depth=4,
    kernel_size=5,
    dilation_growth=2,
    merger_config=None,
    initial_linear=61,
    backbone_out_channels=2048,
    time_agg_out="att",
    n_time_groups=None,
    output_head_config=None,
)

feature_model = SimpleConvTimeAggModel(
    in_channels=CHANNELS,
    out_channels=2048,
    config=config,
)

model = nn.Sequential(
    feature_model,
    nn.Linear(2048, NUM_CLASSES)
)

model = model.to(DEVICE)

# ================================================================
# 9. LOAD CHECKPOINT
# ================================================================

print("\n" + "=" * 70)
print("LOADING CHECKPOINT")
print("=" * 70)

checkpoint = torch.load(
    CHECKPOINT_PATH,
    map_location="cpu",
    weights_only=False,
)

print("Checkpoint epoch:")
print(checkpoint.get("epoch"))

print("Best validation loss:")
print(checkpoint.get("best_val_loss"))

checkpoint_state = checkpoint["model_state_dict"]

model_state = model.state_dict()

missing_model_keys = [
    key for key in model_state
    if key not in checkpoint_state
]

unexpected_checkpoint_keys = [
    key for key in checkpoint_state
    if key not in model_state
]

print("\nModel keys:", len(model_state))
print("Checkpoint keys:", len(checkpoint_state))

print("\nMissing keys:")
for key in missing_model_keys:
    print(" ", key)

print("\nUnexpected keys:")
for key in unexpected_checkpoint_keys:
    print(" ", key)

allowed_missing = {
    "0.time_agg_out.Ua.weight",
    "0.time_agg_out.Ua.bias",
}

unexpected_missing = (
    set(missing_model_keys) - allowed_missing
)

if unexpected_missing:
    raise RuntimeError(
        "Unexpected missing model keys:\n"
        + "\n".join(sorted(unexpected_missing))
    )

if unexpected_checkpoint_keys:
    raise RuntimeError(
        "Unexpected checkpoint keys:\n"
        + "\n".join(sorted(unexpected_checkpoint_keys))
    )

load_result = model.load_state_dict(
    checkpoint_state,
    strict=False,
)

print("\nMissing after load:")
print(load_result.missing_keys)

print("Unexpected after load:")
print(load_result.unexpected_keys)

print("\nArchitecture compatibility: PASS")

# ================================================================
# 10. DATA LOADER
# ================================================================

dataset = EEGTestDataset(
    X_test,
    y_test,
)

loader = DataLoader(
    dataset,
    batch_size=WINDOWS_PER_BATCH,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=torch.cuda.is_available(),
)

print("\nTest batches:", len(loader))

# ================================================================
# 11. INFERENCE
# ================================================================

print("\n" + "=" * 70)
print("RUNNING DECODER INFERENCE")
print("=" * 70)

model.eval()

all_predictions = []
all_confidences = []
all_correct = []
all_true = []
all_recording_output = []

criterion = nn.CrossEntropyLoss(reduction="sum")

total_loss = 0.0
total_samples = 0

with torch.no_grad():

    for batch_idx, (xb, yb) in enumerate(loader):

        xb = xb.to(DEVICE, non_blocking=True)
        yb = yb.to(DEVICE, non_blocking=True)

        logits = model(xb)

        loss = criterion(logits, yb)

        probabilities = torch.softmax(logits, dim=1)

        confidence, predictions = torch.max(
            probabilities,
            dim=1,
        )

        correct = predictions.eq(yb)

        batch_size = yb.size(0)

        total_loss += loss.item()
        total_samples += batch_size

        all_predictions.extend(
            predictions.cpu().numpy().tolist()
        )

        all_confidences.extend(
            confidence.cpu().numpy().tolist()
        )

        all_correct.extend(
            correct.cpu().numpy().astype(bool).tolist()
        )

        all_true.extend(
            yb.cpu().numpy().tolist()
        )

        if (
            batch_idx == 0
            or (batch_idx + 1) % 500 == 0
            or (batch_idx + 1) == len(loader)
        ):

            running_loss = total_loss / total_samples

            running_acc = (
                np.mean(all_correct)
                if all_correct
                else 0.0
            )

            print(
                f"Batch {batch_idx + 1}/{len(loader)} | "
                f"Loss {running_loss:.6f} | "
                f"Acc {running_acc:.6f}"
            )

predictions = np.asarray(
    all_predictions,
    dtype=np.int64,
)

confidences = np.asarray(
    all_confidences,
    dtype=np.float64,
)

correct = np.asarray(
    all_correct,
    dtype=bool,
)

true_labels = np.asarray(
    all_true,
    dtype=np.int64,
)

test_loss = total_loss / total_samples
test_accuracy = float(np.mean(correct))

print("\nInference complete.")

# ================================================================
# 12. BASIC METRICS
# ================================================================

print("\n" + "=" * 70)
print("BASIC DECODER METRICS")
print("=" * 70)

mean_confidence = float(np.mean(confidences))
median_confidence = float(np.median(confidences))

correct_mean_confidence = float(
    np.mean(confidences[correct])
) if np.any(correct) else None

incorrect_mean_confidence = float(
    np.mean(confidences[~correct])
) if np.any(~correct) else None

balanced_acc = float(
    balanced_accuracy_score(
        true_labels,
        predictions,
    )
)

precision, recall, f1, support = precision_recall_fscore_support(
    true_labels,
    predictions,
    labels=np.arange(NUM_CLASSES),
    zero_division=0,
)

macro_f1 = float(np.mean(f1))

weighted_f1 = float(
    np.average(
        f1,
        weights=support,
    )
)

print(f"Test loss                 : {test_loss:.6f}")
print(f"Test accuracy             : {test_accuracy:.6f}")
print(f"Test accuracy (%)         : {test_accuracy * 100:.3f}%")
print(f"Balanced accuracy         : {balanced_acc:.6f}")
print(f"Macro F1                  : {macro_f1:.6f}")
print(f"Weighted F1               : {weighted_f1:.6f}")
print(f"Mean confidence           : {mean_confidence:.6f}")
print(f"Median confidence         : {median_confidence:.6f}")
print(
    f"Correct mean confidence  : "
    f"{correct_mean_confidence:.6f}"
)
print(
    f"Incorrect mean confidence: "
    f"{incorrect_mean_confidence:.6f}"
)

# ================================================================
# 13. PREDICTION DISTRIBUTION
# ================================================================

prediction_counts = np.bincount(
    predictions,
    minlength=NUM_CLASSES,
)

prediction_fractions = (
    prediction_counts / len(predictions)
)

unique_predicted = int(
    np.sum(prediction_counts > 0)
)

dominant_class = int(
    np.argmax(prediction_counts)
)

dominant_count = int(
    prediction_counts[dominant_class]
)

dominant_fraction = float(
    prediction_fractions[dominant_class]
)

print("\n" + "=" * 70)
print("PREDICTION DISTRIBUTION")
print("=" * 70)

print(
    f"Unique predicted classes : "
    f"{unique_predicted}/{NUM_CLASSES}"
)

print(
    f"Dominant class           : "
    f"{dominant_class} "
    f"({CLASS_LABELS[dominant_class]})"
)

print(
    f"Dominant fraction        : "
    f"{dominant_fraction * 100:.3f}%"
)

for idx in range(NUM_CLASSES):

    if prediction_counts[idx] > 0:

        print(
            f"{idx:2d} "
            f"{CLASS_LABELS[idx]:10s} "
            f"{prediction_counts[idx]:6d} "
            f"({prediction_fractions[idx] * 100:7.3f}%)"
        )

# ================================================================
# 14. CONFIDENCE BIN ANALYSIS
# ================================================================

print("\n" + "=" * 70)
print("CONFIDENCE BIN ANALYSIS")
print("=" * 70)

bins = np.arange(
    0.0,
    1.000001,
    0.05,
)

confidence_bin_rows = []

for lower, upper in zip(
    bins[:-1],
    bins[1:],
):

    if upper >= 1.0:
        mask = (
            (confidences >= lower)
            & (confidences <= upper)
        )
    else:
        mask = (
            (confidences >= lower)
            & (confidences < upper)
        )

    count = int(np.sum(mask))

    if count == 0:

        bin_accuracy = None
        mean_conf = None

    else:

        bin_accuracy = float(
            np.mean(correct[mask])
        )

        mean_conf = float(
            np.mean(confidences[mask])
        )

    confidence_bin_rows.append({
        "lower": float(lower),
        "upper": float(upper),
        "samples": count,
        "fraction": (
            float(count / len(confidences))
            if len(confidences)
            else 0.0
        ),
        "mean_confidence": mean_conf,
        "accuracy": bin_accuracy,
    })

    if count > 0:

        print(
            f"{lower:.2f}-{upper:.2f} | "
            f"{count:6d} samples | "
            f"mean conf "
            f"{mean_conf:.4f} | "
            f"accuracy "
            f"{bin_accuracy * 100:.2f}%"
        )

confidence_bins_df = pd.DataFrame(
    confidence_bin_rows
)

confidence_bins_df.to_csv(
    CONFIDENCE_BINS_CSV,
    index=False,
)

# ================================================================
# 15. EXPECTED CALIBRATION ERROR
# ================================================================

print("\n" + "=" * 70)
print("EXPECTED CALIBRATION ERROR")
print("=" * 70)

ece = 0.0

for row in confidence_bin_rows:

    count = row["samples"]

    if count == 0:
        continue

    accuracy_bin = row["accuracy"]
    confidence_bin = row["mean_confidence"]

    ece += (
        count / len(confidences)
    ) * abs(
        accuracy_bin - confidence_bin
    )

ece = float(ece)

print(f"ECE: {ece:.6f}")

# ================================================================
# 16. CONFIDENCE → CORRECTNESS SEPARATION
# ================================================================

print("\n" + "=" * 70)
print("CONFIDENCE / CORRECTNESS SEPARATION")
print("=" * 70)

confidence_difference = (
    correct_mean_confidence -
    incorrect_mean_confidence
)

print(
    "Correct mean confidence   : "
    f"{correct_mean_confidence:.6f}"
)

print(
    "Incorrect mean confidence : "
    f"{incorrect_mean_confidence:.6f}"
)

print(
    "Difference                : "
    f"{confidence_difference:.6f}"
)

# ================================================================
# 17. THRESHOLD / COVERAGE ANALYSIS
# ================================================================

print("\n" + "=" * 70)
print("CONFIDENCE THRESHOLD ANALYSIS")
print("=" * 70)

thresholds = [
    0.05,
    0.10,
    0.15,
    0.20,
    0.25,
    0.30,
    0.35,
    0.40,
    0.45,
    0.50,
    0.60,
    0.70,
    0.80,
    0.90,
]

threshold_rows = []

for threshold in thresholds:

    selected = confidences >= threshold

    selected_count = int(
        np.sum(selected)
    )

    coverage = float(
        selected_count / len(confidences)
    )

    if selected_count > 0:

        selected_accuracy = float(
            np.mean(correct[selected])
        )

        selected_mean_confidence = float(
            np.mean(confidences[selected])
        )

    else:

        selected_accuracy = None
        selected_mean_confidence = None

    threshold_rows.append({
        "threshold": threshold,
        "selected_samples": selected_count,
        "coverage": coverage,
        "coverage_percent": coverage * 100,
        "accuracy": selected_accuracy,
        "accuracy_percent": (
            selected_accuracy * 100
            if selected_accuracy is not None
            else None
        ),
        "mean_confidence": selected_mean_confidence,
    })

    if selected_count > 0:

        print(
            f"Threshold >= {threshold:.2f} | "
            f"Coverage "
            f"{coverage * 100:7.3f}% | "
            f"Accuracy "
            f"{selected_accuracy * 100:7.3f}% | "
            f"N={selected_count}"
        )

    else:

        print(
            f"Threshold >= {threshold:.2f} | "
            f"Coverage   0.000% | "
            f"Accuracy N/A | N=0"
        )

threshold_df = pd.DataFrame(
    threshold_rows
)

threshold_df.to_csv(
    THRESHOLD_CSV,
    index=False,
)

# ================================================================
# 18. CLASS METRICS
# ================================================================

print("\n" + "=" * 70)
print("PER-CLASS METRICS")
print("=" * 70)

class_rows = []

for idx in range(NUM_CLASSES):

    class_rows.append({
        "class_id": idx,
        "class_label": CLASS_LABELS[idx],
        "support": int(support[idx]),
        "predicted_count": int(prediction_counts[idx]),
        "precision": float(precision[idx]),
        "recall": float(recall[idx]),
        "f1": float(f1[idx]),
    })

    print(
        f"{idx:2d} "
        f"{CLASS_LABELS[idx]:10s} | "
        f"true={support[idx]:6d} | "
        f"pred={prediction_counts[idx]:6d} | "
        f"P={precision[idx]:.4f} | "
        f"R={recall[idx]:.4f} | "
        f"F1={f1[idx]:.4f}"
    )

class_metrics_df = pd.DataFrame(
    class_rows
)

class_metrics_df.to_csv(
    CLASS_METRICS_CSV,
    index=False,
)

# ================================================================
# 19. CONFUSION MATRIX
# ================================================================

cm = confusion_matrix(
    true_labels,
    predictions,
    labels=np.arange(NUM_CLASSES),
)

cm_df = pd.DataFrame(
    cm,
    index=CLASS_LABELS,
    columns=CLASS_LABELS,
)

cm_df.to_csv(
    CONFUSION_CSV
)

print("\nConfusion matrix saved.")

# ================================================================
# 20. SAVE DETAILED PREDICTIONS
# ================================================================

print("\n" + "=" * 70)
print("SAVING DETAILED PREDICTIONS")
print("=" * 70)

prediction_df = pd.DataFrame({
    "recording_id": recording_ids,
    "true_class": true_labels,
    "true_label": [
        CLASS_LABELS[x]
        for x in true_labels
    ],
    "predicted_class": predictions,
    "predicted_label": [
        CLASS_LABELS[x]
        for x in predictions
    ],
    "confidence": confidences,
    "correct": correct.astype(int),
})

prediction_df.to_csv(
    PREDICTIONS_CSV,
    index=False,
)

print(
    f"Saved: {PREDICTIONS_CSV}"
)

# ================================================================
# 21. ENTROPY OF PREDICTED CLASS DISTRIBUTION
# ================================================================

pred_probs = prediction_fractions[
    prediction_fractions > 0
]

prediction_entropy = float(
    -np.sum(
        pred_probs *
        np.log2(pred_probs)
    )
)

max_entropy = math.log2(NUM_CLASSES)

normalized_prediction_entropy = (
    prediction_entropy / max_entropy
)

print("\n" + "=" * 70)
print("PREDICTION ENTROPY")
print("=" * 70)

print(
    f"Prediction entropy       : "
    f"{prediction_entropy:.6f} bits"
)

print(
    f"Maximum possible entropy : "
    f"{max_entropy:.6f} bits"
)

print(
    f"Normalized entropy       : "
    f"{normalized_prediction_entropy:.6f}"
)

# ================================================================
# 22. COLLAPSE CHECK
# ================================================================

# This is intentionally stricter than the previous binary
# single-class collapse check.
#
# Single-class collapse:
#   exactly one predicted class.
#
# Severe concentration:
#   one class >= 50% of predictions.
#
# We report both.

single_class_collapse = (
    unique_predicted == 1
)

severe_concentration = (
    dominant_fraction >= 0.50
)

print("\n" + "=" * 70)
print("PREDICTION COLLAPSE / CONCENTRATION")
print("=" * 70)

print(
    "Single-class collapse : "
    f"{'YES' if single_class_collapse else 'NO'}"
)

print(
    "Severe concentration  : "
    f"{'YES' if severe_concentration else 'NO'}"
)

# ================================================================
# 23. SAVE COMPLETE JSON RESULT
# ================================================================

results = {
    "phase": "5H.5",
    "description": "Decoder utility and confidence validation",

    "checkpoint": str(CHECKPOINT_PATH),

    "checkpoint_epoch": (
        int(checkpoint["epoch"])
        if checkpoint.get("epoch") is not None
        else None
    ),

    "best_validation_loss": (
        float(checkpoint["best_val_loss"])
        if checkpoint.get("best_val_loss") is not None
        else None
    ),

    "device": str(DEVICE),

    "seed": SEED,

    "test_recordings": test_recordings,

    "test_samples": int(len(y_test)),

    "excluded_samples": int(excluded_samples),

    "test_loss": float(test_loss),

    "test_accuracy": float(test_accuracy),

    "test_accuracy_percent": float(
        test_accuracy * 100
    ),

    "balanced_accuracy": balanced_acc,

    "macro_f1": macro_f1,

    "weighted_f1": weighted_f1,

    "mean_confidence": mean_confidence,

    "median_confidence": median_confidence,

    "correct_mean_confidence": (
        correct_mean_confidence
    ),

    "incorrect_mean_confidence": (
        incorrect_mean_confidence
    ),

    "confidence_correctness_difference": (
        confidence_difference
    ),

    "expected_calibration_error": ece,

    "unique_predicted_classes": unique_predicted,

    "dominant_class": dominant_class,

    "dominant_class_label": CLASS_LABELS[
        dominant_class
    ],

    "dominant_count": dominant_count,

    "dominant_fraction": dominant_fraction,

    "prediction_entropy_bits": prediction_entropy,

    "maximum_entropy_bits": max_entropy,

    "normalized_prediction_entropy": (
        normalized_prediction_entropy
    ),

    "single_class_collapse": (
        single_class_collapse
    ),

    "severe_prediction_concentration": (
        severe_concentration
    ),

    "prediction_distribution": {
        str(idx): {
            "label": CLASS_LABELS[idx],
            "count": int(prediction_counts[idx]),
            "fraction": float(
                prediction_fractions[idx]
            ),
        }
        for idx in range(NUM_CLASSES)
    },

    "confidence_bins": confidence_bin_rows,

    "threshold_analysis": threshold_rows,

    "class_metrics": class_rows,

    "output_files": {
        "predictions_csv": str(PREDICTIONS_CSV),
        "confusion_matrix_csv": str(CONFUSION_CSV),
        "class_metrics_csv": str(CLASS_METRICS_CSV),
        "confidence_bins_csv": str(CONFIDENCE_BINS_CSV),
        "threshold_analysis_csv": str(THRESHOLD_CSV),
    },

    "next_step": (
        "Determine whether decoder confidence and correctness "
        "support a stable utility-label definition before "
        "training the causal gate."
    ),
}

with open(
    RESULTS_JSON,
    "w",
    encoding="utf-8",
) as f:

    json.dump(
        results,
        f,
        indent=2,
    )

# ================================================================
# 24. FINAL SUMMARY
# ================================================================

print("\n" + "=" * 70)
print("PHASE 5H.5 COMPLETE")
print("=" * 70)

print(
    f"Test samples             : {len(y_test)}"
)

print(
    f"Test accuracy            : "
    f"{test_accuracy * 100:.3f}%"
)

print(
    f"Balanced accuracy        : "
    f"{balanced_acc:.6f}"
)

print(
    f"Macro F1                 : "
    f"{macro_f1:.6f}"
)

print(
    f"Mean confidence          : "
    f"{mean_confidence:.6f}"
)

print(
    f"Correct mean confidence  : "
    f"{correct_mean_confidence:.6f}"
)

print(
    f"Incorrect mean confidence: "
    f"{incorrect_mean_confidence:.6f}"
)

print(
    f"Confidence separation    : "
    f"{confidence_difference:.6f}"
)

print(
    f"ECE                      : "
    f"{ece:.6f}"
)

print(
    f"Unique predicted classes : "
    f"{unique_predicted}/{NUM_CLASSES}"
)

print(
    f"Dominant class           : "
    f"{dominant_class} "
    f"({CLASS_LABELS[dominant_class]})"
)

print(
    f"Dominant fraction        : "
    f"{dominant_fraction * 100:.3f}%"
)

print(
    f"Single-class collapse    : "
    f"{'YES' if single_class_collapse else 'NO'}"
)

print(
    f"Severe concentration     : "
    f"{'YES' if severe_concentration else 'NO'}"
)

print("\nResults saved to:")
print(RESULTS_JSON)

print("\nAdditional outputs:")
print(PREDICTIONS_CSV)
print(CONFUSION_CSV)
print(CLASS_METRICS_CSV)
print(CONFIDENCE_BINS_CSV)
print(THRESHOLD_CSV)

print("=" * 70)
```

```python
# ================================================================
# PHASE 5H.6 — VALIDATION UTILITY CALIBRATION
# ================================================================
#
# Purpose:
#   Determine a utility-label rule using ONLY the frozen validation
#   subjects before generating final utility labels.
#
# Uses:
#   - Phase 5H.3 normalized decoder checkpoint
#   - Frozen Phase 6 validation split
#   - Same preprocessing / normalization as Phase 5H.3
#
# Does NOT:
#   - retrain decoder
#   - use test data
#   - generate final test utility labels
#
# Evaluates:
#   1. Decoder validation performance
#   2. Confidence vs correctness
#   3. Candidate confidence thresholds
#   4. Utility-positive rate
#   5. Precision / recall of high-utility windows
#   6. Candidate utility definitions
#   7. Frozen threshold recommendation
#
# ================================================================

import os
import json
import math
import random
import warnings
from pathlib import Path

import numpy as np
import pandas as pd

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_recall_fscore_support,
    confusion_matrix,
)

warnings.filterwarnings("ignore")

print("=" * 70)
print("PHASE 5H.6 — VALIDATION UTILITY CALIBRATION")
print("=" * 70)

# ================================================================
# 1. CONFIGURATION
# ================================================================

SEED = 42

PROJECT_ROOT = Path(
    "/content/gdrive/MyDrive/NeuroSentinel"
)

RECORDINGS_DIR = (
    PROJECT_ROOT /
    "06_phase5_utility_labels" /
    "recordings"
)

CHECKPOINT_PATH = (
    PROJECT_ROOT /
    "checkpoints" /
    "phase5h3" /
    "phase5h3_best.pt"
)

MANIFEST_PATH = (
    PROJECT_ROOT /
    "07_phase6_subject_split" /
    "phase6_subject_split_manifest.csv"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "results" /
    "phase5h6"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

RESULTS_JSON = (
    OUTPUT_DIR /
    "phase5h6_validation_utility_calibration.json"
)

PREDICTIONS_CSV = (
    OUTPUT_DIR /
    "phase5h6_validation_predictions.csv"
)

THRESHOLD_CSV = (
    OUTPUT_DIR /
    "phase5h6_threshold_analysis.csv"
)

UTILITY_CSV = (
    OUTPUT_DIR /
    "phase5h6_utility_candidates.csv"
)

WINDOWS_PER_BATCH = 8
NUM_WORKERS = 0

CHANNELS = 61
TIME_SAMPLES = 25
NUM_CLASSES = 29

CLASS_LABELS = [
    "a", "b", "c", "d", "e", "f", "g", "h", "i", "j",
    "k", "l", "m", "n", "o", "p", "q", "r", "s", "t",
    "u", "v", "w", "x", "y", "z",
    "<space>", "<number>", "<special>"
]

ANOMALOUS_LABELS = {
    "-",
    "`"
}

# Candidate thresholds.
# These are evaluated on validation only.
THRESHOLDS = [
    0.10,
    0.15,
    0.20,
    0.25,
    0.30,
    0.35,
    0.40,
]

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available()
    else "cpu"
)

# Reproducibility
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

print(f"Device: {DEVICE}")
print(f"Project root: {PROJECT_ROOT}")

# ================================================================
# 2. VERIFY PATHS
# ================================================================

required_paths = {
    "recordings_dir": RECORDINGS_DIR,
    "checkpoint": CHECKPOINT_PATH,
    "manifest": MANIFEST_PATH,
}

for name, path in required_paths.items():

    if not path.exists():
        raise FileNotFoundError(
            f"Missing {name}: {path}"
        )

print("\nPaths verified.")

# ================================================================
# 3. IMPORT EXACT MODEL
# ================================================================

from neuraltrain.models.simpleconv import (
    SimpleConvTimeAgg,
    SimpleConvTimeAggModel,
)

print("Model classes imported.")

# ================================================================
# 4. LOAD FROZEN PHASE 6 SPLIT
# ================================================================

manifest = pd.read_csv(
    MANIFEST_PATH
)

print("\nManifest:")
print(manifest.shape)
print(manifest["split"].value_counts())

validation_manifest = manifest[
    manifest["split"]
    .astype(str)
    .str.lower()
    == "val"
].copy()

validation_manifest = (
    validation_manifest
    .sort_values("recording_id")
    .reset_index(drop=True)
)

validation_recordings = (
    validation_manifest[
        "recording_id"
    ]
    .astype(str)
    .tolist()
)

validation_subjects = sorted(
    validation_manifest[
        "subject"
    ]
    .astype(str)
    .unique()
    .tolist()
)

print(
    f"\nValidation recordings: "
    f"{len(validation_recordings)}"
)

print(
    f"Validation subjects: "
    f"{validation_subjects}"
)

for rid in validation_recordings:
    print(" ", rid)

# ================================================================
# 5. LABEL MAPPING
# ================================================================

char_to_class = {
    ch: idx
    for idx, ch in enumerate(
        "abcdefghijklmnopqrstuvwxyz"
    )
}

char_to_class[" "] = 26

SPECIAL_TOKEN_MAP = {
    "<space>": 26,
    "<number>": 27,
    "<special>": 28,
}

def target_to_class(value):

    if pd.isna(value):
        return None

    value = str(value)

    if value in ANOMALOUS_LABELS:
        return None

    if value in SPECIAL_TOKEN_MAP:
        return SPECIAL_TOKEN_MAP[value]

    if value == " ":
        return 26

    if (
        len(value) == 1
        and value.lower() in char_to_class
    ):
        return char_to_class[
            value.lower()
        ]

    if value.lower() in {
        "number",
        "<number>"
    }:
        return 27

    if value.lower() in {
        "special",
        "<special>"
    }:
        return 28

    try:

        integer_value = int(value)

        if 0 <= integer_value < NUM_CLASSES:
            return integer_value

    except Exception:
        pass

    return None

# ================================================================
# 6. LOAD VALIDATION DATA
# ================================================================

print("\n" + "=" * 70)
print("LOADING VALIDATION RECORDINGS")
print("=" * 70)

all_windows = []
all_labels = []
all_recording_ids = []

excluded_samples = 0

for recording_id in validation_recordings:

    windows_path = (
        RECORDINGS_DIR /
        f"{recording_id}_windows.npy"
    )

    metadata_path = (
        RECORDINGS_DIR /
        f"{recording_id}_metadata.csv"
    )

    if not windows_path.exists():
        raise FileNotFoundError(
            windows_path
        )

    if not metadata_path.exists():
        raise FileNotFoundError(
            metadata_path
        )

    windows = np.load(
        windows_path
    )

    metadata = pd.read_csv(
        metadata_path
    )

    if len(windows) != len(metadata):

        raise ValueError(
            f"{recording_id}: "
            f"{len(windows)} windows vs "
            f"{len(metadata)} metadata rows"
        )

    labels = []
    valid_indices = []

    for idx, row in metadata.iterrows():

        if "target" in metadata.columns:
            raw_target = row["target"]

        elif "label" in metadata.columns:
            raw_target = row["label"]

        elif "text" in metadata.columns:
            raw_target = row["text"]

        else:

            raise ValueError(
                f"No target/label/text column "
                f"in {metadata_path}"
            )

        target = target_to_class(
            raw_target
        )

        if target is None:

            excluded_samples += 1
            continue

        labels.append(target)
        valid_indices.append(idx)

    windows_valid = windows[
        valid_indices
    ]

    all_windows.append(
        windows_valid
    )

    all_labels.extend(labels)

    all_recording_ids.extend(
        [recording_id] * len(labels)
    )

    print(
        f"{recording_id:20s} "
        f"{len(windows_valid):5d} valid windows"
    )

X_val = np.concatenate(
    all_windows,
    axis=0
)

y_val = np.asarray(
    all_labels,
    dtype=np.int64
)

recording_ids = np.asarray(
    all_recording_ids
)

print("\n" + "=" * 70)
print("VALIDATION DATASET")
print("=" * 70)

print(
    "Valid samples :",
    len(y_val)
)

print(
    "Excluded      :",
    excluded_samples
)

print(
    "Windows shape :",
    X_val.shape
)

print(
    "Labels shape  :",
    y_val.shape
)

if X_val.shape[1:] != (
    CHANNELS,
    TIME_SAMPLES
):

    raise ValueError(
        f"Unexpected window shape: "
        f"{X_val.shape}"
    )

# ================================================================
# 7. DATASET
# ================================================================

class EEGValidationDataset(Dataset):

    def __init__(
        self,
        windows,
        labels
    ):

        self.windows = windows
        self.labels = labels

    def __len__(self):

        return len(self.labels)

    def __getitem__(self, idx):

        x = self.windows[
            idx
        ].astype(np.float32)

        y = int(
            self.labels[idx]
        )

        # EXACT Phase 5H.3 normalization:
        # per-window channel-wise z-score
        mean = x.mean(
            axis=1,
            keepdims=True
        )

        std = x.std(
            axis=1,
            keepdims=True
        )

        std = np.where(
            std < 1e-8,
            1.0,
            std
        )

        x = (
            x - mean
        ) / std

        return (
            torch.from_numpy(x),
            torch.tensor(
                y,
                dtype=torch.long
            )
        )

# ================================================================
# 8. RECONSTRUCT EXACT MODEL
# ================================================================

print("\n" + "=" * 70)
print("RECONSTRUCTING MODEL")
print("=" * 70)

config = SimpleConvTimeAgg(
    hidden=64,
    depth=4,
    kernel_size=5,
    dilation_growth=2,
    merger_config=None,
    initial_linear=61,
    backbone_out_channels=2048,
    time_agg_out="att",
    n_time_groups=None,
    output_head_config=None,
)

feature_model = SimpleConvTimeAggModel(
    in_channels=CHANNELS,
    out_channels=2048,
    config=config,
)

model = nn.Sequential(
    feature_model,
    nn.Linear(
        2048,
        NUM_CLASSES
    )
)

model = model.to(
    DEVICE
)

# ================================================================
# 9. LOAD CHECKPOINT
# ================================================================

checkpoint = torch.load(
    CHECKPOINT_PATH,
    map_location="cpu",
    weights_only=False,
)

print(
    "Checkpoint epoch:",
    checkpoint.get("epoch")
)

print(
    "Best validation loss:",
    checkpoint.get(
        "best_val_loss"
    )
)

checkpoint_state = (
    checkpoint[
        "model_state_dict"
    ]
)

model_state = (
    model.state_dict()
)

missing_model_keys = [
    key
    for key in model_state
    if key not in checkpoint_state
]

unexpected_checkpoint_keys = [
    key
    for key in checkpoint_state
    if key not in model_state
]

allowed_missing = {
    "0.time_agg_out.Ua.weight",
    "0.time_agg_out.Ua.bias",
}

unexpected_missing = (
    set(missing_model_keys)
    - allowed_missing
)

if unexpected_missing:

    raise RuntimeError(
        "Unexpected missing model keys:\n"
        + "\n".join(
            sorted(
                unexpected_missing
            )
        )
    )

if unexpected_checkpoint_keys:

    raise RuntimeError(
        "Unexpected checkpoint keys:\n"
        + "\n".join(
            sorted(
                unexpected_checkpoint_keys
            )
        )
    )

load_result = (
    model.load_state_dict(
        checkpoint_state,
        strict=False
    )
)

print(
    "\nMissing after load:",
    load_result.missing_keys
)

print(
    "Unexpected after load:",
    load_result.unexpected_keys
)

print(
    "\nArchitecture compatibility: PASS"
)

# ================================================================
# 10. VALIDATION LOADER
# ================================================================

dataset = EEGValidationDataset(
    X_val,
    y_val
)

loader = DataLoader(
    dataset,
    batch_size=WINDOWS_PER_BATCH,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=torch.cuda.is_available(),
)

print(
    "\nValidation batches:",
    len(loader)
)

# ================================================================
# 11. RUN VALIDATION INFERENCE
# ================================================================

print("\n" + "=" * 70)
print("RUNNING VALIDATION INFERENCE")
print("=" * 70)

model.eval()

all_predictions = []
all_confidences = []
all_correct = []
all_true = []

total_loss = 0.0
total_samples = 0

criterion = nn.CrossEntropyLoss(
    reduction="sum"
)

with torch.no_grad():

    for batch_idx, (
        xb,
        yb
    ) in enumerate(loader):

        xb = xb.to(
            DEVICE,
            non_blocking=True
        )

        yb = yb.to(
            DEVICE,
            non_blocking=True
        )

        logits = model(xb)

        loss = criterion(
            logits,
            yb
        )

        probabilities = torch.softmax(
            logits,
            dim=1
        )

        confidence, predictions = (
            torch.max(
                probabilities,
                dim=1
            )
        )

        batch_correct = (
            predictions.eq(yb)
        )

        batch_size = yb.size(0)

        total_loss += loss.item()
        total_samples += batch_size

        all_predictions.extend(
            predictions
            .cpu()
            .numpy()
            .tolist()
        )

        all_confidences.extend(
            confidence
            .cpu()
            .numpy()
            .tolist()
        )

        all_correct.extend(
            batch_correct
            .cpu()
            .numpy()
            .astype(bool)
            .tolist()
        )

        all_true.extend(
            yb
            .cpu()
            .numpy()
            .tolist()
        )

        if (
            batch_idx == 0
            or (batch_idx + 1) % 500 == 0
            or (batch_idx + 1) == len(loader)
        ):

            running_loss = (
                total_loss /
                total_samples
            )

            running_acc = (
                np.mean(
                    all_correct
                )
            )

            print(
                f"Batch "
                f"{batch_idx + 1}/"
                f"{len(loader)} | "
                f"Loss "
                f"{running_loss:.6f} | "
                f"Acc "
                f"{running_acc:.6f}"
            )

predictions = np.asarray(
    all_predictions,
    dtype=np.int64
)

confidences = np.asarray(
    all_confidences,
    dtype=np.float64
)

correct = np.asarray(
    all_correct,
    dtype=bool
)

true_labels = np.asarray(
    all_true,
    dtype=np.int64
)

validation_loss = (
    total_loss /
    total_samples
)

validation_accuracy = float(
    np.mean(correct)
)

# ================================================================
# 12. BASIC VALIDATION METRICS
# ================================================================

print("\n" + "=" * 70)
print("VALIDATION DECODER METRICS")
print("=" * 70)

mean_confidence = float(
    np.mean(confidences)
)

median_confidence = float(
    np.median(confidences)
)

correct_mean_confidence = float(
    np.mean(
        confidences[correct]
    )
) if np.any(correct) else None

incorrect_mean_confidence = float(
    np.mean(
        confidences[~correct]
    )
) if np.any(~correct) else None

confidence_difference = (
    correct_mean_confidence
    - incorrect_mean_confidence
)

balanced_acc = float(
    balanced_accuracy_score(
        true_labels,
        predictions
    )
)

precision, recall, f1, support = (
    precision_recall_fscore_support(
        true_labels,
        predictions,
        labels=np.arange(
            NUM_CLASSES
        ),
        zero_division=0
    )
)

macro_f1 = float(
    np.mean(f1)
)

weighted_f1 = float(
    np.average(
        f1,
        weights=support
    )
)

print(
    f"Validation loss          : "
    f"{validation_loss:.6f}"
)

print(
    f"Validation accuracy      : "
    f"{validation_accuracy:.6f}"
)

print(
    f"Validation accuracy (%)  : "
    f"{validation_accuracy * 100:.3f}%"
)

print(
    f"Balanced accuracy        : "
    f"{balanced_acc:.6f}"
)

print(
    f"Macro F1                 : "
    f"{macro_f1:.6f}"
)

print(
    f"Weighted F1              : "
    f"{weighted_f1:.6f}"
)

print(
    f"Mean confidence          : "
    f"{mean_confidence:.6f}"
)

print(
    f"Median confidence        : "
    f"{median_confidence:.6f}"
)

print(
    f"Correct mean confidence  : "
    f"{correct_mean_confidence:.6f}"
)

print(
    f"Incorrect mean confidence: "
    f"{incorrect_mean_confidence:.6f}"
)

print(
    f"Confidence separation    : "
    f"{confidence_difference:.6f}"
)

# ================================================================
# 13. CONFIDENCE BIN ANALYSIS
# ================================================================

print("\n" + "=" * 70)
print("VALIDATION CONFIDENCE BINS")
print("=" * 70)

bins = np.arange(
    0.0,
    1.000001,
    0.05
)

confidence_bin_rows = []

for lower, upper in zip(
    bins[:-1],
    bins[1:]
):

    if upper >= 1.0:

        mask = (
            (confidences >= lower)
            &
            (confidences <= upper)
        )

    else:

        mask = (
            (confidences >= lower)
            &
            (confidences < upper)
        )

    count = int(
        np.sum(mask)
    )

    if count == 0:

        mean_conf = None
        bin_accuracy = None

    else:

        mean_conf = float(
            np.mean(
                confidences[mask]
            )
        )

        bin_accuracy = float(
            np.mean(
                correct[mask]
            )
        )

    confidence_bin_rows.append({
        "lower": float(lower),
        "upper": float(upper),
        "samples": count,
        "coverage": (
            count / len(confidences)
        ),
        "mean_confidence": mean_conf,
        "accuracy": bin_accuracy,
    })

    if count > 0:

        print(
            f"{lower:.2f}-{upper:.2f} | "
            f"N={count:6d} | "
            f"conf={mean_conf:.4f} | "
            f"acc={bin_accuracy * 100:.2f}%"
        )

# ================================================================
# 14. ECE
# ================================================================

ece = 0.0

for row in confidence_bin_rows:

    if row["samples"] == 0:
        continue

    ece += (
        row["samples"] /
        len(confidences)
    ) * abs(
        row["accuracy"]
        -
        row["mean_confidence"]
    )

ece = float(ece)

print("\n" + "=" * 70)
print("VALIDATION CALIBRATION")
print("=" * 70)

print(
    f"Expected Calibration Error: "
    f"{ece:.6f}"
)

# ================================================================
# 15. THRESHOLD ANALYSIS
# ================================================================

print("\n" + "=" * 70)
print("VALIDATION THRESHOLD ANALYSIS")
print("=" * 70)

threshold_rows = []

for threshold in THRESHOLDS:

    selected = (
        confidences >= threshold
    )

    selected_count = int(
        np.sum(selected)
    )

    coverage = (
        selected_count /
        len(confidences)
    )

    if selected_count > 0:

        selected_accuracy = float(
            np.mean(
                correct[selected]
            )
        )

        selected_mean_conf = float(
            np.mean(
                confidences[selected]
            )
        )

        # Precision of "HIGH utility" prediction
        # when defining HIGH utility as:
        # confidence >= threshold
        # and evaluating actual correctness.
        utility_precision = (
            selected_accuracy
        )

        # Recall of actual correct windows.
        total_correct = int(
            np.sum(correct)
        )

        utility_recall = (
            np.sum(
                correct[selected]
            ) /
            total_correct
            if total_correct > 0
            else 0.0
        )

    else:

        selected_accuracy = None
        selected_mean_conf = None
        utility_precision = None
        utility_recall = None

    threshold_rows.append({
        "threshold": threshold,
        "selected_samples": selected_count,
        "coverage": float(coverage),
        "coverage_percent": float(
            coverage * 100
        ),
        "accuracy": selected_accuracy,
        "accuracy_percent": (
            selected_accuracy * 100
            if selected_accuracy is not None
            else None
        ),
        "mean_confidence": selected_mean_conf,
        "utility_precision": utility_precision,
        "utility_recall": float(
            utility_recall
        ) if utility_recall is not None else None,
    })

    if selected_count > 0:

        print(
            f"Threshold >= {threshold:.2f} | "
            f"Coverage="
            f"{coverage * 100:7.3f}% | "
            f"Accuracy="
            f"{selected_accuracy * 100:7.3f}% | "
            f"Utility recall="
            f"{utility_recall * 100:7.3f}% | "
            f"N={selected_count}"
        )

    else:

        print(
            f"Threshold >= {threshold:.2f} | "
            f"Coverage=0.000% | "
            f"N=0"
        )

threshold_df = pd.DataFrame(
    threshold_rows
)

threshold_df.to_csv(
    THRESHOLD_CSV,
    index=False
)

# ================================================================
# 16. CANDIDATE UTILITY DEFINITIONS
# ================================================================
#
# Candidate A:
#   correctness only
#
# Candidate B:
#   correctness AND confidence >= threshold
#
# Candidate C:
#   confidence >= threshold
#
# Candidate C is deliberately included only as a diagnostic,
# because it is NOT a correctness-grounded utility definition.
# ================================================================

print("\n" + "=" * 70)
print("UTILITY LABEL CANDIDATES")
print("=" * 70)

utility_rows = []

# Candidate A
utility_a = correct.astype(int)

utility_rows.append({
    "candidate": "A_correctness_only",
    "threshold": None,
    "positive_samples": int(
        np.sum(utility_a)
    ),
    "negative_samples": int(
        len(utility_a) -
        np.sum(utility_a)
    ),
    "positive_rate": float(
        np.mean(utility_a)
    ),
    "description": (
        "HIGH if decoder prediction is correct; "
        "LOW otherwise."
    ),
})

print(
    "A | correctness only | "
    f"positive rate="
    f"{np.mean(utility_a) * 100:.3f}%"
)

# Candidate B
for threshold in THRESHOLDS:

    utility_b = (
        correct
        &
        (confidences >= threshold)
    ).astype(int)

    positive = int(
        np.sum(utility_b)
    )

    negative = int(
        len(utility_b) -
        positive
    )

    utility_rows.append({
        "candidate": (
            "B_correct_and_confidence"
        ),
        "threshold": threshold,
        "positive_samples": positive,
        "negative_samples": negative,
        "positive_rate": float(
            positive / len(utility_b)
        ),
        "description": (
            "HIGH if decoder prediction is "
            "correct AND confidence >= threshold; "
            "LOW otherwise."
        ),
    })

    print(
        f"B | correct AND confidence >= "
        f"{threshold:.2f} | "
        f"positive="
        f"{positive:6d} | "
        f"rate="
        f"{positive / len(utility_b) * 100:.3f}%"
    )

# Candidate C
for threshold in THRESHOLDS:

    utility_c = (
        confidences >= threshold
    ).astype(int)

    positive = int(
        np.sum(utility_c)
    )

    utility_rows.append({
        "candidate": (
            "C_confidence_only_diagnostic"
        ),
        "threshold": threshold,
        "positive_samples": positive,
        "negative_samples": (
            len(utility_c) - positive
        ),
        "positive_rate": float(
            positive / len(utility_c)
        ),
        "description": (
            "Diagnostic only: HIGH if confidence "
            "exceeds threshold, regardless of "
            "correctness."
        ),
    })

utility_df = pd.DataFrame(
    utility_rows
)

utility_df.to_csv(
    UTILITY_CSV,
    index=False
)

# ================================================================
# 17. AUTOMATIC THRESHOLD SCREENING
# ================================================================
#
# This does NOT automatically declare a final threshold.
#
# It only identifies thresholds that satisfy basic dataset
# viability constraints:
#
#   - positive rate >= 5%
#   - positive rate <= 50%
#   - coverage >= 5%
#
# Final selection must be based on project utility criteria.
# ================================================================

print("\n" + "=" * 70)
print("THRESHOLD VIABILITY SCREEN")
print("=" * 70)

viable_thresholds = []

for row in threshold_rows:

    threshold = row["threshold"]
    coverage = row["coverage"]

    utility_positive_rate = (
        row["accuracy"]
        * coverage
        if row["accuracy"] is not None
        else 0.0
    )

    # Positive rate for correctness AND confidence
    selected_mask = (
        confidences >= threshold
    )

    positive_mask = (
        correct &
        selected_mask
    )

    positive_rate = (
        np.mean(
            positive_mask
        )
    )

    viable = (
        positive_rate >= 0.05
        and
        positive_rate <= 0.50
        and
        coverage >= 0.05
    )

    if viable:

        viable_thresholds.append(
            threshold
        )

    print(
        f"Threshold {threshold:.2f} | "
        f"coverage="
        f"{coverage * 100:.3f}% | "
        f"HIGH utility="
        f"{positive_rate * 100:.3f}% | "
        f"viable="
        f"{viable}"
    )

# ================================================================
# 18. PREDICTION DISTRIBUTION
# ================================================================

prediction_counts = np.bincount(
    predictions,
    minlength=NUM_CLASSES
)

unique_predicted = int(
    np.sum(
        prediction_counts > 0
    )
)

dominant_class = int(
    np.argmax(
        prediction_counts
    )
)

dominant_fraction = float(
    prediction_counts[
        dominant_class
    ] / len(predictions)
)

# ================================================================
# 19. SAVE VALIDATION PREDICTIONS
# ================================================================

prediction_df = pd.DataFrame({
    "recording_id": recording_ids,
    "true_class": true_labels,
    "true_label": [
        CLASS_LABELS[x]
        for x in true_labels
    ],
    "predicted_class": predictions,
    "predicted_label": [
        CLASS_LABELS[x]
        for x in predictions
    ],
    "confidence": confidences,
    "correct": correct.astype(int),
})

prediction_df.to_csv(
    PREDICTIONS_CSV,
    index=False
)

# ================================================================
# 20. FINAL JSON
# ================================================================

results = {
    "phase": "5H.6",
    "description": (
        "Validation-only decoder utility "
        "calibration"
    ),

    "checkpoint": str(
        CHECKPOINT_PATH
    ),

    "checkpoint_epoch": (
        int(checkpoint["epoch"])
        if checkpoint.get("epoch")
        is not None
        else None
    ),

    "best_validation_loss_from_training": (
        float(
            checkpoint["best_val_loss"]
        )
        if checkpoint.get(
            "best_val_loss"
        ) is not None
        else None
    ),

    "device": str(DEVICE),

    "seed": SEED,

    "validation_subjects": (
        validation_subjects
    ),

    "validation_recordings": (
        validation_recordings
    ),

    "validation_samples": int(
        len(y_val)
    ),

    "excluded_samples": int(
        excluded_samples
    ),

    "validation_loss": float(
        validation_loss
    ),

    "validation_accuracy": float(
        validation_accuracy
    ),

    "validation_accuracy_percent": float(
        validation_accuracy * 100
    ),

    "balanced_accuracy": balanced_acc,

    "macro_f1": macro_f1,

    "weighted_f1": weighted_f1,

    "mean_confidence": mean_confidence,

    "median_confidence": median_confidence,

    "correct_mean_confidence": (
        correct_mean_confidence
    ),

    "incorrect_mean_confidence": (
        incorrect_mean_confidence
    ),

    "confidence_separation": (
        confidence_difference
    ),

    "expected_calibration_error": ece,

    "unique_predicted_classes": (
        unique_predicted
    ),

    "dominant_predicted_class": (
        dominant_class
    ),

    "dominant_predicted_label": (
        CLASS_LABELS[
            dominant_class
        ]
    ),

    "dominant_prediction_fraction": (
        dominant_fraction
    ),

    "confidence_bins": (
        confidence_bin_rows
    ),

    "threshold_analysis": (
        threshold_rows
    ),

    "utility_candidates": (
        utility_rows
    ),

    "viable_thresholds": (
        viable_thresholds
    ),

    "outputs": {
        "predictions": str(
            PREDICTIONS_CSV
        ),
        "threshold_analysis": str(
            THRESHOLD_CSV
        ),
        "utility_candidates": str(
            UTILITY_CSV
        ),
    },

    "important_note": (
        "Threshold selection is performed on validation "
        "data only. The test set must remain untouched "
        "until the utility rule is frozen."
    ),
}

with open(
    RESULTS_JSON,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        results,
        f,
        indent=2
    )

# ================================================================
# 21. FINAL SUMMARY
# ================================================================

print("\n" + "=" * 70)
print("PHASE 5H.6 COMPLETE")
print("=" * 70)

print(
    f"Validation samples       : "
    f"{len(y_val)}"
)

print(
    f"Validation loss          : "
    f"{validation_loss:.6f}"
)

print(
    f"Validation accuracy      : "
    f"{validation_accuracy * 100:.3f}%"
)

print(
    f"Balanced accuracy        : "
    f"{balanced_acc:.6f}"
)

print(
    f"Macro F1                 : "
    f"{macro_f1:.6f}"
)

print(
    f"Mean confidence          : "
    f"{mean_confidence:.6f}"
)

print(
    f"Correct mean confidence  : "
    f"{correct_mean_confidence:.6f}"
)

print(
    f"Incorrect mean confidence: "
    f"{incorrect_mean_confidence:.6f}"
)

print(
    f"Confidence separation    : "
    f"{confidence_difference:.6f}"
)

print(
    f"ECE                      : "
    f"{ece:.6f}"
)

print(
    f"Unique predicted classes : "
    f"{unique_predicted}/{NUM_CLASSES}"
)

print(
    f"Dominant prediction      : "
    f"{CLASS_LABELS[dominant_class]} "
    f"({dominant_fraction * 100:.3f}%)"
)

print(
    "\nViable candidate "
    "thresholds:"
)

if viable_thresholds:

    print(
        viable_thresholds
    )

else:

    print(
        "None under the current "
        "basic viability constraints."
    )

print("\nOutputs saved to:")

print(
    RESULTS_JSON
)

print(
    PREDICTIONS_CSV
)

print(
    THRESHOLD_CSV
)

print(
    UTILITY_CSV
)

print("=" * 70)
```

```python
# ================================================================
# PHASE 5H.7 — CORRECTNESS-BASED UTILITY LABEL GENERATION
# ================================================================
#
# Frozen utility definition:
#
#   utility_label = 1  if decoder prediction == true label
#   utility_label = 0  otherwise
#
# Confidence is retained as metadata only.
#
# IMPORTANT:
#   - No decoder retraining
#   - No threshold tuning
#   - Frozen Phase 6 subject split
#   - Same Phase 5H.3 normalization
#   - Test set is evaluated but never used for fitting/tuning
#
# Outputs:
#   - train utility labels
#   - validation utility labels
#   - test utility labels
#   - combined manifest
#   - class/utility statistics
#   - JSON summary
# ================================================================

import os
import json
import random
import warnings
from pathlib import Path
from collections import Counter

import numpy as np
import pandas as pd

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

warnings.filterwarnings("ignore")

print("=" * 70)
print("PHASE 5H.7 — CORRECTNESS-BASED UTILITY LABEL GENERATION")
print("=" * 70)

# ================================================================
# 1. CONFIGURATION
# ================================================================

SEED = 42

PROJECT_ROOT = Path(
    "/content/gdrive/MyDrive/NeuroSentinel"
)

RECORDINGS_DIR = (
    PROJECT_ROOT
    / "06_phase5_utility_labels"
    / "recordings"
)

CHECKPOINT_PATH = (
    PROJECT_ROOT
    / "checkpoints"
    / "phase5h3"
    / "phase5h3_best.pt"
)

MANIFEST_PATH = (
    PROJECT_ROOT
    / "07_phase6_subject_split"
    / "phase6_subject_split_manifest.csv"
)

OUTPUT_DIR = (
    PROJECT_ROOT
    / "results"
    / "phase5h7"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

OUTPUT_MANIFEST = (
    OUTPUT_DIR
    / "phase5h7_utility_label_manifest.csv"
)

SUMMARY_JSON = (
    OUTPUT_DIR
    / "phase5h7_utility_label_summary.json"
)

STATS_CSV = (
    OUTPUT_DIR
    / "phase5h7_utility_statistics.csv"
)

RECORDING_STATS_CSV = (
    OUTPUT_DIR
    / "phase5h7_recording_statistics.csv"
)

WINDOW_BATCH_SIZE = 8
NUM_WORKERS = 0

CHANNELS = 61
TIME_SAMPLES = 25
NUM_CLASSES = 29

CLASS_LABELS = [
    "a", "b", "c", "d", "e", "f", "g", "h", "i", "j",
    "k", "l", "m", "n", "o", "p", "q", "r", "s", "t",
    "u", "v", "w", "x", "y", "z",
    "<space>", "<number>", "<special>"
]

ANOMALOUS_LABELS = {
    "-",
    "`"
}

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available()
    else "cpu"
)

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

print(f"Device: {DEVICE}")
print(f"Project root: {PROJECT_ROOT}")

# ================================================================
# 2. VERIFY PATHS
# ================================================================

for name, path in {
    "recordings": RECORDINGS_DIR,
    "checkpoint": CHECKPOINT_PATH,
    "manifest": MANIFEST_PATH,
}.items():

    if not path.exists():
        raise FileNotFoundError(
            f"Missing {name}: {path}"
        )

print("\nPaths verified.")

# ================================================================
# 3. LOAD FROZEN MANIFEST
# ================================================================

manifest = pd.read_csv(
    MANIFEST_PATH
)

required_columns = {
    "recording_id",
    "subject",
    "split",
}

missing_columns = (
    required_columns
    - set(manifest.columns)
)

if missing_columns:
    raise ValueError(
        f"Manifest missing columns: "
        f"{missing_columns}"
    )

print("\nFrozen Phase 6 split:")
print(
    manifest["split"]
    .value_counts()
)

print(
    "\nSubject overlap check:"
)

for split_name in [
    "train",
    "val",
    "test",
]:

    subjects = set(
        manifest[
            manifest["split"]
            .astype(str)
            .str.lower()
            == split_name
        ]["subject"]
        .astype(str)
    )

    print(
        f"{split_name:5s}: "
        f"{sorted(subjects)}"
    )

train_subjects = set(
    manifest[
        manifest["split"]
        .astype(str)
        .str.lower()
        == "train"
    ]["subject"]
    .astype(str)
)

val_subjects = set(
    manifest[
        manifest["split"]
        .astype(str)
        .str.lower()
        == "val"
    ]["subject"]
    .astype(str)
)

test_subjects = set(
    manifest[
        manifest["split"]
        .astype(str)
        .str.lower()
        == "test"
    ]["subject"]
    .astype(str)
)

if train_subjects & val_subjects:
    raise RuntimeError(
        "Train/validation subject overlap detected."
    )

if train_subjects & test_subjects:
    raise RuntimeError(
        "Train/test subject overlap detected."
    )

if val_subjects & test_subjects:
    raise RuntimeError(
        "Validation/test subject overlap detected."
    )

print(
    "\nSubject separation: PASS"
)

# ================================================================
# 4. LABEL MAPPING
# ================================================================

char_to_class = {
    ch: idx
    for idx, ch in enumerate(
        "abcdefghijklmnopqrstuvwxyz"
    )
}

char_to_class[" "] = 26

SPECIAL_TOKEN_MAP = {
    "<space>": 26,
    "<number>": 27,
    "<special>": 28,
}

def target_to_class(value):

    if pd.isna(value):
        return None

    value = str(value)

    if value in ANOMALOUS_LABELS:
        return None

    if value in SPECIAL_TOKEN_MAP:
        return SPECIAL_TOKEN_MAP[value]

    if value == " ":
        return 26

    if (
        len(value) == 1
        and value.lower() in char_to_class
    ):
        return char_to_class[
            value.lower()
        ]

    if value.lower() in {
        "number",
        "<number>",
    }:
        return 27

    if value.lower() in {
        "special",
        "<special>",
    }:
        return 28

    try:

        integer_value = int(value)

        if 0 <= integer_value < NUM_CLASSES:
            return integer_value

    except Exception:
        pass

    return None

# ================================================================
# 5. DATASET FOR INFERENCE
# ================================================================

class EEGInferenceDataset(Dataset):

    def __init__(
        self,
        windows,
        labels,
    ):

        self.windows = windows
        self.labels = labels

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):

        x = self.windows[
            idx
        ].astype(np.float32)

        y = int(
            self.labels[idx]
        )

        # EXACT Phase 5H.3 normalization
        mean = x.mean(
            axis=1,
            keepdims=True
        )

        std = x.std(
            axis=1,
            keepdims=True
        )

        std = np.where(
            std < 1e-8,
            1.0,
            std
        )

        x = (
            x - mean
        ) / std

        return (
            torch.from_numpy(x),
            torch.tensor(
                y,
                dtype=torch.long
            )
        )

# ================================================================
# 6. RECONSTRUCT EXACT DECODER
# ================================================================

print("\n" + "=" * 70)
print("RECONSTRUCTING DECODER")
print("=" * 70)

from neuraltrain.models.simpleconv import (
    SimpleConvTimeAgg,
    SimpleConvTimeAggModel,
)

config = SimpleConvTimeAgg(
    hidden=64,
    depth=4,
    kernel_size=5,
    dilation_growth=2,
    merger_config=None,
    initial_linear=61,
    backbone_out_channels=2048,
    time_agg_out="att",
    n_time_groups=None,
    output_head_config=None,
)

feature_model = SimpleConvTimeAggModel(
    in_channels=CHANNELS,
    out_channels=2048,
    config=config,
)

model = nn.Sequential(
    feature_model,
    nn.Linear(
        2048,
        NUM_CLASSES
    )
)

model = model.to(
    DEVICE
)

# ================================================================
# 7. LOAD CHECKPOINT
# ================================================================

checkpoint = torch.load(
    CHECKPOINT_PATH,
    map_location="cpu",
    weights_only=False,
)

checkpoint_state = (
    checkpoint["model_state_dict"]
)

model_state = (
    model.state_dict()
)

missing_keys = [
    key
    for key in model_state
    if key not in checkpoint_state
]

unexpected_keys = [
    key
    for key in checkpoint_state
    if key not in model_state
]

allowed_missing = {
    "0.time_agg_out.Ua.weight",
    "0.time_agg_out.Ua.bias",
}

unexpected_missing = (
    set(missing_keys)
    - allowed_missing
)

if unexpected_missing:
    raise RuntimeError(
        "Unexpected missing checkpoint keys:\n"
        + "\n".join(
            sorted(unexpected_missing)
        )
    )

if unexpected_keys:
    raise RuntimeError(
        "Unexpected checkpoint keys:\n"
        + "\n".join(
            sorted(unexpected_keys)
        )
    )

load_result = model.load_state_dict(
    checkpoint_state,
    strict=False
)

print(
    "Checkpoint epoch:",
    checkpoint.get("epoch")
)

print(
    "Best validation loss:",
    checkpoint.get("best_val_loss")
)

print(
    "Missing keys:",
    load_result.missing_keys
)

print(
    "Unexpected keys:",
    load_result.unexpected_keys
)

print(
    "Architecture compatibility: PASS"
)

model.eval()

# ================================================================
# 8. PROCESS ONE RECORDING
# ================================================================

def process_recording(
    recording_id,
    split_name,
):

    windows_path = (
        RECORDINGS_DIR
        / f"{recording_id}_windows.npy"
    )

    metadata_path = (
        RECORDINGS_DIR
        / f"{recording_id}_metadata.csv"
    )

    if not windows_path.exists():
        raise FileNotFoundError(
            windows_path
        )

    if not metadata_path.exists():
        raise FileNotFoundError(
            metadata_path
        )

    windows = np.load(
        windows_path
    )

    metadata = pd.read_csv(
        metadata_path
    )

    if len(windows) != len(metadata):

        raise ValueError(
            f"{recording_id}: "
            f"window/metadata mismatch"
        )

    valid_indices = []
    labels = []

    for idx, row in metadata.iterrows():

        if "target" in metadata.columns:
            raw_target = row["target"]

        elif "label" in metadata.columns:
            raw_target = row["label"]

        elif "text" in metadata.columns:
            raw_target = row["text"]

        else:
            raise ValueError(
                f"No target/label/text column "
                f"in {metadata_path}"
            )

        target = target_to_class(
            raw_target
        )

        if target is None:
            continue

        valid_indices.append(idx)
        labels.append(target)

    valid_windows = windows[
        valid_indices
    ]

    labels = np.asarray(
        labels,
        dtype=np.int64
    )

    dataset = EEGInferenceDataset(
        valid_windows,
        labels
    )

    loader = DataLoader(
        dataset,
        batch_size=WINDOW_BATCH_SIZE,
        shuffle=False,
        num_workers=NUM_WORKERS,
        pin_memory=torch.cuda.is_available(),
    )

    predictions = []
    confidences = []

    with torch.no_grad():

        for xb, yb in loader:

            xb = xb.to(
                DEVICE,
                non_blocking=True
            )

            logits = model(xb)

            probabilities = (
                torch.softmax(
                    logits,
                    dim=1
                )
            )

            confidence, prediction = (
                torch.max(
                    probabilities,
                    dim=1
                )
            )

            predictions.extend(
                prediction
                .cpu()
                .numpy()
                .tolist()
            )

            confidences.extend(
                confidence
                .cpu()
                .numpy()
                .tolist()
            )

    predictions = np.asarray(
        predictions,
        dtype=np.int64
    )

    confidences = np.asarray(
        confidences,
        dtype=np.float32
    )

    correctness = (
        predictions == labels
    )

    output = pd.DataFrame({
        "recording_id": [
            recording_id
        ] * len(labels),

        "split": [
            split_name
        ] * len(labels),

        "subject": [
            str(
                manifest[
                    manifest[
                        "recording_id"
                    ].astype(str)
                    == recording_id
                ]["subject"]
                .iloc[0]
            )
        ] * len(labels),

        "window_index": valid_indices,

        "true_class": labels,

        "true_label": [
            CLASS_LABELS[x]
            for x in labels
        ],

        "predicted_class": predictions,

        "predicted_label": [
            CLASS_LABELS[x]
            for x in predictions
        ],

        "decoder_confidence": confidences,

        "decoder_correct": (
            correctness.astype(np.int8)
        ),

        # FROZEN UTILITY DEFINITION
        "utility_label": (
            correctness.astype(np.int8)
        ),
    })

    return (
        output,
        int(len(labels)),
        int(np.sum(correctness)),
    )

# ================================================================
# 9. PROCESS ALL RECORDINGS
# ================================================================

print("\n" + "=" * 70)
print("GENERATING UTILITY LABELS")
print("=" * 70)

all_outputs = []
recording_stats = []

split_order = [
    "train",
    "val",
    "test",
]

for split_name in split_order:

    split_manifest = manifest[
        manifest["split"]
        .astype(str)
        .str.lower()
        == split_name
    ].copy()

    split_manifest = (
        split_manifest
        .sort_values("recording_id")
    )

    print(
        f"\n--- {split_name.upper()} ---"
    )

    for _, row in split_manifest.iterrows():

        recording_id = str(
            row["recording_id"]
        )

        output, count, correct_count = (
            process_recording(
                recording_id,
                split_name,
            )
        )

        all_outputs.append(
            output
        )

        accuracy = (
            correct_count / count
            if count > 0
            else 0.0
        )

        recording_stats.append({
            "recording_id": recording_id,
            "subject": str(
                row["subject"]
            ),
            "split": split_name,
            "valid_windows": count,
            "utility_positive": correct_count,
            "utility_negative": (
                count - correct_count
            ),
            "utility_positive_rate": (
                accuracy
            ),
            "decoder_accuracy": (
                accuracy
            ),
        })

        print(
            f"{recording_id:20s} "
            f"N={count:5d} | "
            f"HIGH={correct_count:5d} | "
            f"rate={accuracy * 100:6.2f}%"
        )

# ================================================================
# 10. COMBINE RESULTS
# ================================================================

utility_df = pd.concat(
    all_outputs,
    ignore_index=True
)

recording_stats_df = pd.DataFrame(
    recording_stats
)

print("\n" + "=" * 70)
print("UTILITY DATASET")
print("=" * 70)

print(
    "Total samples:",
    len(utility_df)
)

print(
    "Columns:",
    list(utility_df.columns)
)

# ================================================================
# 11. VERIFY EXPECTED SAMPLE COUNTS
# ================================================================

print("\n" + "=" * 70)
print("SPLIT STATISTICS")
print("=" * 70)

split_stats = []

for split_name in split_order:

    subset = utility_df[
        utility_df["split"]
        == split_name
    ]

    count = len(subset)

    positives = int(
        subset["utility_label"]
        .sum()
    )

    negatives = (
        count - positives
    )

    positive_rate = (
        positives / count
        if count > 0
        else 0.0
    )

    mean_confidence = float(
        subset[
            "decoder_confidence"
        ].mean()
    )

    split_stats.append({
        "split": split_name,
        "samples": count,
        "utility_positive": positives,
        "utility_negative": negatives,
        "utility_positive_rate": (
            positive_rate
        ),
        "mean_confidence": (
            mean_confidence
        ),
    })

    print(
        f"{split_name:5s} | "
        f"N={count:6d} | "
        f"HIGH={positives:6d} | "
        f"LOW={negatives:6d} | "
        f"HIGH rate="
        f"{positive_rate * 100:6.3f}% | "
        f"mean conf="
        f"{mean_confidence:.6f}"
    )

stats_df = pd.DataFrame(
    split_stats
)

# ================================================================
# 12. VERIFY SUBJECT DISJOINTNESS
# ================================================================

print("\n" + "=" * 70)
print("UTILITY DATASET SUBJECT CHECK")
print("=" * 70)

split_subject_sets = {}

for split_name in split_order:

    split_subject_sets[
        split_name
    ] = set(
        utility_df[
            utility_df["split"]
            == split_name
        ]["subject"]
        .astype(str)
        .unique()
    )

    print(
        f"{split_name:5s}: "
        f"{sorted(split_subject_sets[split_name])}"
    )

if (
    split_subject_sets["train"]
    &
    split_subject_sets["val"]
):
    raise RuntimeError(
        "Train/validation overlap."
    )

if (
    split_subject_sets["train"]
    &
    split_subject_sets["test"]
):
    raise RuntimeError(
        "Train/test overlap."
    )

if (
    split_subject_sets["val"]
    &
    split_subject_sets["test"]
):
    raise RuntimeError(
        "Validation/test overlap."
    )

print(
    "\nSubject separation: PASS"
)

# ================================================================
# 13. CHECK UTILITY LABEL VALIDITY
# ================================================================

print("\n" + "=" * 70)
print("UTILITY LABEL INTEGRITY CHECK")
print("=" * 70)

expected_utility = (
    utility_df[
        "predicted_class"
    ].to_numpy()
    ==
    utility_df[
        "true_class"
    ].to_numpy()
).astype(np.int8)

actual_utility = (
    utility_df[
        "utility_label"
    ].to_numpy()
)

if not np.array_equal(
    expected_utility,
    actual_utility
):

    raise RuntimeError(
        "Utility label integrity check FAILED."
    )

print(
    "utility_label == "
    "(predicted_class == true_class): PASS"
)

print(
    "Utility labels:",
    sorted(
        utility_df[
            "utility_label"
        ].unique()
        .tolist()
    )
)

# ================================================================
# 14. CONFIDENCE STATISTICS BY UTILITY
# ================================================================

print("\n" + "=" * 70)
print("CONFIDENCE BY UTILITY LABEL")
print("=" * 70)

for utility_value in [
    0,
    1,
]:

    subset = utility_df[
        utility_df[
            "utility_label"
        ]
        == utility_value
    ]

    if len(subset) == 0:
        continue

    print(
        f"Utility {utility_value} | "
        f"N={len(subset):6d} | "
        f"mean confidence="
        f"{subset['decoder_confidence'].mean():.6f} | "
        f"median="
        f"{subset['decoder_confidence'].median():.6f}"
    )

# ================================================================
# 15. CLASS DISTRIBUTION BY UTILITY
# ================================================================

print("\n" + "=" * 70)
print("UTILITY LABEL BY TRUE CLASS")
print("=" * 70)

class_utility_rows = []

for class_id in range(
    NUM_CLASSES
):

    subset = utility_df[
        utility_df[
            "true_class"
        ]
        == class_id
    ]

    count = len(subset)

    if count == 0:
        positive = 0
        rate = 0.0
    else:
        positive = int(
            subset[
                "utility_label"
            ].sum()
        )

        rate = (
            positive / count
        )

    class_utility_rows.append({
        "class_id": class_id,
        "class_label": CLASS_LABELS[
            class_id
        ],
        "samples": count,
        "utility_positive": positive,
        "utility_positive_rate": rate,
    })

    print(
        f"{class_id:2d} "
        f"{CLASS_LABELS[class_id]:10s} | "
        f"N={count:6d} | "
        f"HIGH={positive:6d} | "
        f"rate={rate * 100:6.2f}%"
    )

class_utility_df = pd.DataFrame(
    class_utility_rows
)

# ================================================================
# 16. RECORDING-LEVEL UTILITY BALANCE
# ================================================================

print("\n" + "=" * 70)
print("RECORDING-LEVEL UTILITY SUMMARY")
print("=" * 70)

print(
    recording_stats_df[
        [
            "recording_id",
            "split",
            "valid_windows",
            "utility_positive",
            "utility_positive_rate",
        ]
    ].to_string(index=False)
)

# ================================================================
# 17. SAVE OUTPUTS
# ================================================================

print("\n" + "=" * 70)
print("SAVING OUTPUTS")
print("=" * 70)

# Main detailed dataset
utility_df.to_csv(
    OUTPUT_MANIFEST,
    index=False
)

# Split statistics
stats_df.to_csv(
    STATS_CSV,
    index=False
)

# Recording statistics
recording_stats_df.to_csv(
    RECORDING_STATS_CSV,
    index=False
)

print(
    "Utility manifest:",
    OUTPUT_MANIFEST
)

print(
    "Split statistics:",
    STATS_CSV
)

print(
    "Recording statistics:",
    RECORDING_STATS_CSV
)

# ================================================================
# 18. JSON SUMMARY
# ================================================================

summary = {
    "phase": "5H.7",

    "description": (
        "Correctness-based decoder utility-label "
        "generation"
    ),

    "utility_definition": {
        "utility_label_1": (
            "decoder prediction equals true label"
        ),
        "utility_label_0": (
            "decoder prediction does not equal true label"
        ),
        "confidence_used_for_label": False,
    },

    "checkpoint": str(
        CHECKPOINT_PATH
    ),

    "checkpoint_epoch": (
        int(checkpoint["epoch"])
        if checkpoint.get("epoch")
        is not None
        else None
    ),

    "seed": SEED,

    "device": str(
        DEVICE
    ),

    "channels": CHANNELS,

    "time_samples": TIME_SAMPLES,

    "num_classes": NUM_CLASSES,

    "total_samples": int(
        len(utility_df)
    ),

    "split_statistics": (
        stats_df.to_dict(
            orient="records"
        )
    ),

    "recording_statistics": (
        recording_stats_df.to_dict(
            orient="records"
        )
    ),

    "class_utility_statistics": (
        class_utility_df.to_dict(
            orient="records"
        )
    ),

    "subject_sets": {
        split: sorted(
            list(subjects)
        )
        for split, subjects
        in split_subject_sets.items()
    },

    "subject_separation_verified": True,

    "utility_integrity_verified": True,

    "outputs": {
        "utility_manifest": str(
            OUTPUT_MANIFEST
        ),
        "split_statistics": str(
            STATS_CSV
        ),
        "recording_statistics": str(
            RECORDING_STATS_CSV
        ),
    },

    "next_phase": (
        "Prepare the utility-labeled dataset for "
        "non-causal and causal gate training."
    ),
}

with open(
    SUMMARY_JSON,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        summary,
        f,
        indent=2
    )

# ================================================================
# 19. FINAL SUMMARY
# ================================================================

print("\n" + "=" * 70)
print("PHASE 5H.7 COMPLETE")
print("=" * 70)

print(
    f"Total utility samples : "
    f"{len(utility_df)}"
)

for split_name in split_order:

    row = stats_df[
        stats_df["split"]
        == split_name
    ].iloc[0]

    print(
        f"{split_name:5s} | "
        f"N={int(row['samples']):6d} | "
        f"HIGH={int(row['utility_positive']):6d} | "
        f"LOW={int(row['utility_negative']):6d} | "
        f"HIGH rate="
        f"{row['utility_positive_rate'] * 100:6.3f}%"
    )

print(
    "\nUtility definition:"
)

print(
    "HIGH = decoder prediction correct"
)

print(
    "LOW  = decoder prediction incorrect"
)

print(
    "\nSubject separation: VERIFIED"
)

print(
    "Utility integrity: VERIFIED"
)

print(
    "\nResults:"
)

print(
    SUMMARY_JSON
)

print(
    OUTPUT_MANIFEST
)

print("=" * 70)
```

```python
# ============================================================
# NEUROSENTINEL — PHASE 7
# STEP 1: INPUT / HANDOFF VERIFICATION
# ============================================================

import os
import json
import glob
import numpy as np
import pandas as pd
from collections import Counter

# ------------------------------------------------------------
# 1. Paths
# ------------------------------------------------------------

BASE = "/content/gdrive/MyDrive/NeuroSentinel"

H7_DIR = os.path.join(BASE, "results", "phase5h7")
H7_MANIFEST = os.path.join(
    H7_DIR,
    "phase5h7_utility_label_manifest.csv"
)

PHASE6_MANIFEST = os.path.join(
    BASE,
    "07_phase6_subject_split",
    "phase6_subject_split_manifest.csv"
)

WINDOW_DIR = os.path.join(
    BASE,
    "06_phase5_utility_labels",
    "recordings"
)

PHASE7_CHECKPOINT_DIR = os.path.join(
    BASE,
    "checkpoints",
    "phase7"
)

PHASE7_RESULT_DIR = os.path.join(
    BASE,
    "results",
    "phase7"
)

os.makedirs(PHASE7_CHECKPOINT_DIR, exist_ok=True)
os.makedirs(PHASE7_RESULT_DIR, exist_ok=True)

# ------------------------------------------------------------
# 2. Verify files exist
# ------------------------------------------------------------

required_files = {
    "Phase 5H.7 utility manifest": H7_MANIFEST,
    "Phase 6 split manifest": PHASE6_MANIFEST,
}

print("=" * 70)
print("PHASE 7 — INPUT VERIFICATION")
print("=" * 70)

for name, path in required_files.items():
    exists = os.path.isfile(path)
    print(f"{'✓' if exists else '✗'} {name}: {path}")

    if not exists:
        raise FileNotFoundError(path)

# ------------------------------------------------------------
# 3. Load manifests
# ------------------------------------------------------------

h7 = pd.read_csv(H7_MANIFEST)
phase6 = pd.read_csv(PHASE6_MANIFEST)

print("\nH7 manifest shape:", h7.shape)
print("Phase 6 manifest shape:", phase6.shape)

print("\nH7 columns:")
print(list(h7.columns))

print("\nPhase 6 columns:")
print(list(phase6.columns))

# ------------------------------------------------------------
# 4. Identify important columns robustly
# ------------------------------------------------------------

def find_column(df, candidates, description):
    lower_map = {c.lower(): c for c in df.columns}

    for candidate in candidates:
        if candidate.lower() in lower_map:
            return lower_map[candidate.lower()]

    # substring fallback
    for c in df.columns:
        cl = c.lower()
        for candidate in candidates:
            if candidate.lower() in cl:
                return c

    raise KeyError(
        f"Could not identify {description}. "
        f"Available columns: {list(df.columns)}"
    )


recording_col = find_column(
    h7,
    ["recording_id", "recording"],
    "recording ID"
)

window_col = find_column(
    h7,
    ["window_index", "window_idx", "index"],
    "window index"
)

utility_col = find_column(
    h7,
    ["utility_label", "utility"],
    "utility label"
)

split_col = find_column(
    h7,
    ["split"],
    "split"
)

print("\nDetected H7 columns:")
print(" recording :", recording_col)
print(" window    :", window_col)
print(" utility   :", utility_col)
print(" split     :", split_col)

# ------------------------------------------------------------
# 5. Basic utility-label validation
# ------------------------------------------------------------

utility_values = set(
    pd.to_numeric(h7[utility_col], errors="coerce")
    .dropna()
    .unique()
)

print("\nUtility label values:", sorted(utility_values))

if not utility_values.issubset({0, 1}):
    raise ValueError(
        f"Unexpected utility labels: {utility_values}"
    )

# ------------------------------------------------------------
# 6. Split counts
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("SPLIT COUNTS")
print("=" * 70)

split_counts = (
    h7[split_col]
    .value_counts()
    .sort_index()
)

print(split_counts)

expected_counts = {
    "train": 99837,
    "val": 21206,
    "test": 29907,
}

for split_name, expected in expected_counts.items():
    actual = int(split_counts.get(split_name, 0))

    status = "✓" if actual == expected else "✗"

    print(
        f"{status} {split_name:5s}: "
        f"actual={actual:,} | expected={expected:,}"
    )

    if actual != expected:
        raise ValueError(
            f"{split_name} sample count mismatch: "
            f"{actual} != {expected}"
        )

# ------------------------------------------------------------
# 7. Utility distribution by split
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("UTILITY DISTRIBUTION")
print("=" * 70)

utility_table = pd.crosstab(
    h7[split_col],
    h7[utility_col]
)

print(utility_table)

for split_name in ["train", "val", "test"]:
    subset = h7[h7[split_col] == split_name]

    n = len(subset)
    positives = int(
        (subset[utility_col] == 1).sum()
    )
    negatives = int(
        (subset[utility_col] == 0).sum()
    )

    print(
        f"{split_name:5s}: "
        f"HIGH={positives:,} "
        f"({positives/n*100:.3f}%) | "
        f"LOW={negatives:,} "
        f"({negatives/n*100:.3f}%)"
    )

# ------------------------------------------------------------
# 8. Verify recording IDs
# ------------------------------------------------------------

h7_recordings = set(
    h7[recording_col]
    .astype(str)
    .unique()
)

phase6_recordings = set(
    phase6["recording_id"]
    .astype(str)
    .unique()
)

print("\n" + "=" * 70)
print("RECORDING ALIGNMENT")
print("=" * 70)

print("H7 recordings:", len(h7_recordings))
print("Phase 6 recordings:", len(phase6_recordings))

missing_in_phase6 = h7_recordings - phase6_recordings
missing_in_h7 = phase6_recordings - h7_recordings

print(
    "H7 recordings missing from Phase 6:",
    sorted(missing_in_phase6)
)

print(
    "Phase 6 recordings missing from H7:",
    sorted(missing_in_h7)
)

if missing_in_phase6 or missing_in_h7:
    raise ValueError(
        "Recording sets do not match between Phase 5H.7 and Phase 6."
    )

# ------------------------------------------------------------
# 9. Verify split assignment is subject-frozen
# ------------------------------------------------------------

phase6_split_map = (
    phase6
    .set_index("recording_id")["split"]
    .astype(str)
    .to_dict()
)

h7_split_mismatches = []

for _, row in h7.iterrows():
    rid = str(row[recording_col])
    h7_split = str(row[split_col])

    if rid not in phase6_split_map:
        h7_split_mismatches.append(
            (rid, h7_split, "MISSING")
        )
    elif h7_split != phase6_split_map[rid]:
        h7_split_mismatches.append(
            (
                rid,
                h7_split,
                phase6_split_map[rid]
            )
        )

print(
    "\nSplit assignment mismatches:",
    len(h7_split_mismatches)
)

if h7_split_mismatches:
    print(h7_split_mismatches[:10])
    raise ValueError(
        "H7 split assignments do not match frozen Phase 6 split."
    )

print("✓ Frozen Phase 6 split preserved.")

# ------------------------------------------------------------
# 10. Verify recording-level window files
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("WINDOW FILE VERIFICATION")
print("=" * 70)

window_files = {
    os.path.basename(p): p
    for p in glob.glob(
        os.path.join(WINDOW_DIR, "*_windows.npy")
    )
}

print("Window files found:", len(window_files))

missing_window_files = []

for rid in sorted(h7_recordings):
    filename = f"{rid}_windows.npy"

    if filename not in window_files:
        missing_window_files.append(filename)

if missing_window_files:
    print("Missing window files:")
    for f in missing_window_files:
        print(" ", f)

    raise FileNotFoundError(
        "Some recording-level window files are missing."
    )

print("✓ All H7 recordings have window files.")

# ------------------------------------------------------------
# 11. Verify window indices against actual .npy shapes
# ------------------------------------------------------------

print("\nChecking window-index bounds...")

bad_indices = []
recording_shapes = {}

for rid in sorted(h7_recordings):

    path = window_files[f"{rid}_windows.npy"]

    # Memory-map only; do not load the complete array.
    arr = np.load(path, mmap_mode="r")

    recording_shapes[rid] = tuple(arr.shape)

    if arr.ndim != 3:
        raise ValueError(
            f"{rid}: expected 3D windows, got {arr.shape}"
        )

    if arr.shape[1] != 61 or arr.shape[2] != 25:
        raise ValueError(
            f"{rid}: expected (N,61,25), got {arr.shape}"
        )

    rid_rows = h7[
        h7[recording_col].astype(str) == rid
    ]

    indices = pd.to_numeric(
        rid_rows[window_col],
        errors="coerce"
    )

    if indices.isna().any():
        raise ValueError(
            f"{rid}: invalid window indices detected."
        )

    indices = indices.astype(np.int64).to_numpy()

    if len(indices) > 0:
        min_idx = indices.min()
        max_idx = indices.max()

        if min_idx < 0 or max_idx >= arr.shape[0]:
            bad_indices.append({
                "recording_id": rid,
                "n_windows": arr.shape[0],
                "min_index": int(min_idx),
                "max_index": int(max_idx),
            })

    del arr

if bad_indices:
    print(pd.DataFrame(bad_indices))
    raise ValueError(
        "Some H7 window indices are outside the recording arrays."
    )

print("✓ All window indices are valid.")

# ------------------------------------------------------------
# 12. Verify expected tensor shape
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("TENSOR FORMAT")
print("=" * 70)

unique_shapes = Counter(
    recording_shapes.values()
)

for shape, count in unique_shapes.items():
    print(f"{count} recording(s): {shape}")

if not all(
    len(shape) == 3 and
    shape[1] == 61 and
    shape[2] == 25
    for shape in recording_shapes.values()
):
    raise ValueError(
        "Not all recordings have 61 × 25 windows."
    )

# ------------------------------------------------------------
# 13. Check duplicate (recording_id, window_index) rows
# ------------------------------------------------------------

duplicates = h7.duplicated(
    subset=[recording_col, window_col]
).sum()

print(
    "\nDuplicate recording/window rows:",
    int(duplicates)
)

if duplicates != 0:
    raise ValueError(
        "Duplicate recording/window entries detected."
    )

# ------------------------------------------------------------
# 14. Verify one real sample
# ------------------------------------------------------------

sample_row = h7.iloc[0]

sample_rid = str(
    sample_row[recording_col]
)

sample_idx = int(
    sample_row[window_col]
)

sample_path = window_files[
    f"{sample_rid}_windows.npy"
]

sample_memmap = np.load(
    sample_path,
    mmap_mode="r"
)

sample_x = np.asarray(
    sample_memmap[sample_idx],
    dtype=np.float32
)

print("\n" + "=" * 70)
print("SAMPLE CHECK")
print("=" * 70)

print("Recording:", sample_rid)
print("Window index:", sample_idx)
print("Shape:", sample_x.shape)
print("Dtype:", sample_x.dtype)
print("Min:", float(sample_x.min()))
print("Max:", float(sample_x.max()))
print("Mean:", float(sample_x.mean()))
print("Std:", float(sample_x.std()))
print("Utility label:", int(sample_row[utility_col]))

assert sample_x.shape == (61, 25)

# ------------------------------------------------------------
# 15. Final verification summary
# ------------------------------------------------------------

summary = {
    "phase": "Phase 7",
    "stage": "input_verification",
    "h7_manifest": H7_MANIFEST,
    "phase6_manifest": PHASE6_MANIFEST,
    "window_directory": WINDOW_DIR,

    "total_samples": int(len(h7)),

    "train_samples": int(
        (h7[split_col] == "train").sum()
    ),

    "val_samples": int(
        (h7[split_col] == "val").sum()
    ),

    "test_samples": int(
        (h7[split_col] == "test").sum()
    ),

    "recordings": len(h7_recordings),

    "input_channels": 61,
    "input_time_samples": 25,

    "normalization":
        "per-window channel-wise z-score",

    "utility_label":
        "decoder prediction correctness",

    "confidence":
        "metadata only; no threshold used",

    "phase6_split_frozen": True,

    "window_index_alignment": True,

    "duplicate_window_rows": int(duplicates),

    "status": "READY_FOR_PHASE7_TRAINING",
}

summary_path = os.path.join(
    PHASE7_RESULT_DIR,
    "phase7_input_verification.json"
)

with open(summary_path, "w") as f:
    json.dump(
        summary,
        f,
        indent=2
    )

print("\n" + "=" * 70)
print("PHASE 7 INPUT VERIFICATION COMPLETE")
print("=" * 70)

print(json.dumps(summary, indent=2))

print(
    f"\nVerification file saved to:\n{summary_path}"
)

print(
    "\nSTATUS: READY FOR PHASE 7 NON-CAUSAL GATE TRAINING"
)
```

```python
# ============================================================
# NEUROSENTINEL — PHASE 7
# NON-CAUSAL GATE BASELINE
# ============================================================

import os
import json
import random
import time
import warnings
from collections import defaultdict

import numpy as np
import pandas as pd

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
)

warnings.filterwarnings("ignore")

# ============================================================
# 1. CONFIGURATION
# ============================================================

SEED = 42

BASE = "/content/gdrive/MyDrive/NeuroSentinel"

H7_MANIFEST = os.path.join(
    BASE,
    "results",
    "phase5h7",
    "phase5h7_utility_label_manifest.csv"
)

WINDOW_DIR = os.path.join(
    BASE,
    "06_phase5_utility_labels",
    "recordings"
)

PHASE7_CHECKPOINT_DIR = os.path.join(
    BASE,
    "checkpoints",
    "phase7"
)

PHASE7_RESULT_DIR = os.path.join(
    BASE,
    "results",
    "phase7"
)

os.makedirs(PHASE7_CHECKPOINT_DIR, exist_ok=True)
os.makedirs(PHASE7_RESULT_DIR, exist_ok=True)

# Input
N_CHANNELS = 61
N_TIME = 25

# Training
BATCH_SIZE = 256
NUM_WORKERS = 0
NUM_EPOCHS = 25
LEARNING_RATE = 1e-3
WEIGHT_DECAY = 1e-4
PATIENCE = 6

# Normalization
EPS = 1e-6

# Threshold search
THRESHOLDS = np.arange(
    0.05,
    0.951,
    0.01
)

# ============================================================
# 2. REPRODUCIBILITY
# ============================================================

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("=" * 70)
print("NEUROSENTINEL — PHASE 7 NON-CAUSAL GATE")
print("=" * 70)

print("Device:", DEVICE)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))

# ============================================================
# 3. LOAD AND VERIFY MANIFEST
# ============================================================

h7 = pd.read_csv(H7_MANIFEST)

required_columns = [
    "recording_id",
    "split",
    "window_index",
    "utility_label",
]

missing = [
    c for c in required_columns
    if c not in h7.columns
]

if missing:
    raise ValueError(
        f"Missing required H7 columns: {missing}"
    )

h7["recording_id"] = (
    h7["recording_id"]
    .astype(str)
)

h7["window_index"] = (
    pd.to_numeric(
        h7["window_index"],
        errors="raise"
    )
    .astype(np.int64)
)

h7["utility_label"] = (
    pd.to_numeric(
        h7["utility_label"],
        errors="raise"
    )
    .astype(np.int64)
)

if not set(h7["utility_label"].unique()).issubset({0, 1}):
    raise ValueError("Utility labels must be binary 0/1.")

expected_counts = {
    "train": 99837,
    "val": 21206,
    "test": 29907,
}

for split_name, expected in expected_counts.items():

    actual = int(
        (h7["split"] == split_name).sum()
    )

    if actual != expected:
        raise ValueError(
            f"{split_name} count mismatch: "
            f"{actual} != {expected}"
        )

print("\nManifest verified.")
print("Total samples:", len(h7))

# ============================================================
# 4. BUILD RECORDING MEMORY MAPS
# ============================================================

recording_ids = sorted(
    h7["recording_id"].unique()
)

window_arrays = {}

print(
    "\nOpening recording windows using memory mapping..."
)

for rid in recording_ids:

    path = os.path.join(
        WINDOW_DIR,
        f"{rid}_windows.npy"
    )

    if not os.path.isfile(path):
        raise FileNotFoundError(path)

    arr = np.load(
        path,
        mmap_mode="r"
    )

    if arr.ndim != 3:
        raise ValueError(
            f"{rid}: invalid shape {arr.shape}"
        )

    if (
        arr.shape[1] != N_CHANNELS
        or arr.shape[2] != N_TIME
    ):
        raise ValueError(
            f"{rid}: expected "
            f"(N,{N_CHANNELS},{N_TIME}), "
            f"got {arr.shape}"
        )

    window_arrays[rid] = arr

print(
    f"✓ Memory-mapped {len(window_arrays)} recordings."
)

# ============================================================
# 5. DATASET
# ============================================================

class NeuroSentinelGateDataset(Dataset):

    def __init__(
        self,
        dataframe,
        arrays,
        eps=1e-6,
    ):

        self.df = (
            dataframe
            .reset_index(drop=True)
            .copy()
        )

        self.arrays = arrays
        self.eps = eps

        self.recordings = (
            self.df["recording_id"]
            .astype(str)
            .to_numpy()
        )

        self.indices = (
            self.df["window_index"]
            .to_numpy(dtype=np.int64)
        )

        self.labels = (
            self.df["utility_label"]
            .to_numpy(dtype=np.float32)
        )

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):

        rid = self.recordings[idx]
        window_idx = int(self.indices[idx])

        x = np.asarray(
            self.arrays[rid][window_idx],
            dtype=np.float32
        )

        # ----------------------------------------------------
        # Phase 5H.3 normalization:
        # channel-wise z-score across the 25 time samples
        # ----------------------------------------------------

        mean = x.mean(
            axis=1,
            keepdims=True
        )

        std = x.std(
            axis=1,
            keepdims=True
        )

        x = (
            x - mean
        ) / np.maximum(
            std,
            self.eps
        )

        y = self.labels[idx]

        return (
            torch.from_numpy(x),
            torch.tensor(
                y,
                dtype=torch.float32
            )
        )

# ============================================================
# 6. SPLIT DATASETS
# ============================================================

train_df = h7[
    h7["split"] == "train"
].copy()

val_df = h7[
    h7["split"] == "val"
].copy()

test_df = h7[
    h7["split"] == "test"
].copy()

train_dataset = NeuroSentinelGateDataset(
    train_df,
    window_arrays,
    EPS
)

val_dataset = NeuroSentinelGateDataset(
    val_df,
    window_arrays,
    EPS
)

test_dataset = NeuroSentinelGateDataset(
    test_df,
    window_arrays,
    EPS
)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=NUM_WORKERS,
    pin_memory=torch.cuda.is_available(),
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=torch.cuda.is_available(),
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=torch.cuda.is_available(),
)

print("\nDataset sizes:")
print("Train:", len(train_dataset))
print("Val  :", len(val_dataset))
print("Test :", len(test_dataset))

# ============================================================
# 7. CLASS IMBALANCE
# ============================================================

train_positive = int(
    train_df["utility_label"].sum()
)

train_negative = (
    len(train_df)
    - train_positive
)

POS_WEIGHT = (
    train_negative
    / train_positive
)

print("\nClass distribution:")
print(
    f"Train LOW : {train_negative:,}"
)

print(
    f"Train HIGH: {train_positive:,}"
)

print(
    f"Positive weight: {POS_WEIGHT:.6f}"
)

# ============================================================
# 8. NON-CAUSAL CNN
# ============================================================

class NonCausalGate(nn.Module):

    def __init__(
        self,
        in_channels=61,
    ):

        super().__init__()

        # Symmetric padding means temporal context can come
        # from both directions within the EEG window.

        self.features = nn.Sequential(

            nn.Conv1d(
                in_channels,
                64,
                kernel_size=5,
                padding=2,
            ),

            nn.ReLU(),

            nn.Conv1d(
                64,
                64,
                kernel_size=5,
                padding=4,
                dilation=2,
            ),

            nn.ReLU(),

            nn.Conv1d(
                64,
                64,
                kernel_size=5,
                padding=8,
                dilation=4,
            ),

            nn.ReLU(),

            nn.Conv1d(
                64,
                32,
                kernel_size=5,
                padding=16,
                dilation=8,
            ),

            nn.ReLU(),
        )

        self.classifier = nn.Linear(
            32,
            1
        )

    def forward(self, x):

        # x: [B, 61, 25]

        x = self.features(x)

        # Global temporal average pooling
        x = x.mean(
            dim=2
        )

        logits = self.classifier(x)

        return logits.squeeze(1)

model = NonCausalGate(
    in_channels=N_CHANNELS
).to(DEVICE)

parameter_count = sum(
    p.numel()
    for p in model.parameters()
    if p.requires_grad
)

print("\nModel:")
print(model)

print(
    f"\nTrainable parameters: "
    f"{parameter_count:,}"
)

# ============================================================
# 9. LOSS / OPTIMIZER
# ============================================================

criterion = nn.BCEWithLogitsLoss(
    pos_weight=torch.tensor(
        [POS_WEIGHT],
        dtype=torch.float32,
        device=DEVICE,
    )
)

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY,
)

# ============================================================
# 10. EVALUATION FUNCTION
# ============================================================

def collect_predictions(
    model,
    loader,
):

    model.eval()

    all_labels = []
    all_probs = []

    total_loss = 0.0
    total_n = 0

    with torch.no_grad():

        for x, y in loader:

            x = x.to(
                DEVICE,
                non_blocking=True
            )

            y = y.to(
                DEVICE,
                non_blocking=True
            )

            logits = model(x)

            loss = criterion(
                logits,
                y
            )

            probs = torch.sigmoid(
                logits
            )

            batch_n = y.numel()

            total_loss += (
                loss.item()
                * batch_n
            )

            total_n += batch_n

            all_labels.append(
                y.detach()
                .cpu()
                .numpy()
            )

            all_probs.append(
                probs.detach()
                .cpu()
                .numpy()
            )

    labels = np.concatenate(
        all_labels
    )

    probs = np.concatenate(
        all_probs
    )

    avg_loss = (
        total_loss
        / total_n
    )

    return labels, probs, avg_loss

# ============================================================
# 11. TRAINING
# ============================================================

best_val_auprc = -np.inf
best_epoch = None
best_checkpoint = os.path.join(
    PHASE7_CHECKPOINT_DIR,
    "phase7_noncausal_best.pt"
)

history = []

epochs_without_improvement = 0

print("\n" + "=" * 70)
print("TRAINING")
print("=" * 70)

training_start = time.time()

for epoch in range(
    1,
    NUM_EPOCHS + 1
):

    model.train()

    running_loss = 0.0
    running_n = 0

    epoch_start = time.time()

    for x, y in train_loader:

        x = x.to(
            DEVICE,
            non_blocking=True
        )

        y = y.to(
            DEVICE,
            non_blocking=True
        )

        optimizer.zero_grad(
            set_to_none=True
        )

        logits = model(x)

        loss = criterion(
            logits,
            y
        )

        loss.backward()

        optimizer.step()

        batch_n = y.numel()

        running_loss += (
            loss.item()
            * batch_n
        )

        running_n += batch_n

    train_loss = (
        running_loss
        / running_n
    )

    # --------------------------------------------------------
    # Validation
    # --------------------------------------------------------

    val_labels, val_probs, val_loss = (
        collect_predictions(
            model,
            val_loader
        )
    )

    val_preds_05 = (
        val_probs >= 0.5
    ).astype(int)

    val_auprc = average_precision_score(
        val_labels,
        val_probs
    )

    val_auroc = roc_auc_score(
        val_labels,
        val_probs
    )

    val_f1_05 = f1_score(
        val_labels,
        val_preds_05,
        zero_division=0
    )

    val_bal_acc_05 = balanced_accuracy_score(
        val_labels,
        val_preds_05
    )

    epoch_time = (
        time.time()
        - epoch_start
    )

    history.append({
        "epoch": epoch,
        "train_loss": train_loss,
        "val_loss": val_loss,
        "val_auprc": val_auprc,
        "val_auroc": val_auroc,
        "val_f1_threshold_0.5": val_f1_05,
        "val_balanced_accuracy_threshold_0.5":
            val_bal_acc_05,
        "epoch_time_sec": epoch_time,
    })

    improved = (
        val_auprc
        > best_val_auprc
    )

    if improved:

        best_val_auprc = val_auprc
        best_epoch = epoch
        epochs_without_improvement = 0

        torch.save(
            {
                "phase": 7,
                "model_type":
                    "non_causal_dilated_cnn",
                "epoch": epoch,
                "model_state_dict":
                    model.state_dict(),
                "optimizer_state_dict":
                    optimizer.state_dict(),
                "best_val_auprc":
                    float(best_val_auprc),
                "config": {
                    "seed": SEED,
                    "batch_size": BATCH_SIZE,
                    "epochs": NUM_EPOCHS,
                    "learning_rate":
                        LEARNING_RATE,
                    "weight_decay":
                        WEIGHT_DECAY,
                    "patience": PATIENCE,
                    "channels": N_CHANNELS,
                    "time_samples": N_TIME,
                    "normalization":
                        "per-window channel-wise z-score",
                    "epsilon": EPS,
                    "pos_weight":
                        float(POS_WEIGHT),
                },
            },
            best_checkpoint
        )

    else:
        epochs_without_improvement += 1

    marker = "*" if improved else ""

    print(
        f"Epoch {epoch:02d} | "
        f"train_loss={train_loss:.5f} | "
        f"val_loss={val_loss:.5f} | "
        f"AUPRC={val_auprc:.5f} | "
        f"AUROC={val_auroc:.5f} | "
        f"F1@0.5={val_f1_05:.5f} "
        f"{marker} | "
        f"{epoch_time:.1f}s"
    )

    if (
        epochs_without_improvement
        >= PATIENCE
    ):

        print(
            f"\nEarly stopping after "
            f"{epoch} epochs."
        )

        break

training_time = (
    time.time()
    - training_start
)

print(
    f"\nTraining completed in "
    f"{training_time/60:.2f} minutes."
)

print(
    f"Best validation AUPRC: "
    f"{best_val_auprc:.6f}"
)

print(
    f"Best epoch: {best_epoch}"
)

# ============================================================
# 12. LOAD BEST CHECKPOINT
# ============================================================

checkpoint = torch.load(
    best_checkpoint,
    map_location=DEVICE,
    weights_only=False
)

model.load_state_dict(
    checkpoint["model_state_dict"]
)

print(
    "\n✓ Best validation checkpoint restored."
)

# ============================================================
# 13. VALIDATION THRESHOLD SELECTION
# ============================================================

val_labels, val_probs, val_loss = (
    collect_predictions(
        model,
        val_loader
    )
)

threshold_rows = []

for threshold in THRESHOLDS:

    preds = (
        val_probs >= threshold
    ).astype(int)

    threshold_rows.append({
        "threshold": float(threshold),

        "accuracy":
            accuracy_score(
                val_labels,
                preds
            ),

        "balanced_accuracy":
            balanced_accuracy_score(
                val_labels,
                preds
            ),

        "precision":
            precision_score(
                val_labels,
                preds,
                zero_division=0
            ),

        "recall":
            recall_score(
                val_labels,
                preds,
                zero_division=0
            ),

        "f1":
            f1_score(
                val_labels,
                preds,
                zero_division=0
            ),

        "predicted_high_rate":
            float(preds.mean()),
    })

threshold_df = pd.DataFrame(
    threshold_rows
)

# Primary criterion = validation F1
# Tie-break = balanced accuracy
threshold_df = threshold_df.sort_values(
    by=[
        "f1",
        "balanced_accuracy",
        "precision",
    ],
    ascending=False
).reset_index(drop=True)

best_threshold = float(
    threshold_df.iloc[0]["threshold"]
)

best_val_threshold_metrics = (
    threshold_df.iloc[0].to_dict()
)

threshold_path = os.path.join(
    PHASE7_RESULT_DIR,
    "phase7_validation_thresholds.csv"
)

threshold_df.to_csv(
    threshold_path,
    index=False
)

print("\n" + "=" * 70)
print("VALIDATION THRESHOLD SELECTION")
print("=" * 70)

print(
    f"Selected threshold: "
    f"{best_threshold:.2f}"
)

print(
    f"Validation F1: "
    f"{best_val_threshold_metrics['f1']:.6f}"
)

print(
    f"Validation balanced accuracy: "
    f"{best_val_threshold_metrics['balanced_accuracy']:.6f}"
)

print(
    f"Validation precision: "
    f"{best_val_threshold_metrics['precision']:.6f}"
)

print(
    f"Validation recall: "
    f"{best_val_threshold_metrics['recall']:.6f}"
)

# ============================================================
# 14. FINAL TEST EVALUATION
# ============================================================

test_labels, test_probs, test_loss = (
    collect_predictions(
        model,
        test_loader
    )
)

test_preds = (
    test_probs >= best_threshold
).astype(int)

# ------------------------------------------------------------
# Metrics
# ------------------------------------------------------------

test_accuracy = accuracy_score(
    test_labels,
    test_preds
)

test_balanced_accuracy = (
    balanced_accuracy_score(
        test_labels,
        test_preds
    )
)

test_precision = precision_score(
    test_labels,
    test_preds,
    zero_division=0
)

test_recall = recall_score(
    test_labels,
    test_preds,
    zero_division=0
)

test_f1 = f1_score(
    test_labels,
    test_preds,
    zero_division=0
)

test_auroc = roc_auc_score(
    test_labels,
    test_probs
)

test_auprc = average_precision_score(
    test_labels,
    test_probs
)

cm = confusion_matrix(
    test_labels,
    test_preds,
    labels=[0, 1]
)

tn, fp, fn, tp = cm.ravel()

# ============================================================
# 15. BASELINE COMPARISON
# ============================================================

# Always LOW
always_low = np.zeros_like(
    test_labels,
    dtype=int
)

# Always HIGH
always_high = np.ones_like(
    test_labels,
    dtype=int
)

baseline_metrics = {

    "always_low": {
        "accuracy":
            float(
                accuracy_score(
                    test_labels,
                    always_low
                )
            ),
        "balanced_accuracy":
            float(
                balanced_accuracy_score(
                    test_labels,
                    always_low
                )
            ),
        "precision":
            float(
                precision_score(
                    test_labels,
                    always_low,
                    zero_division=0
                )
            ),
        "recall":
            float(
                recall_score(
                    test_labels,
                    always_low,
                    zero_division=0
                )
            ),
        "f1":
            float(
                f1_score(
                    test_labels,
                    always_low,
                    zero_division=0
                )
            ),
    },

    "always_high": {
        "accuracy":
            float(
                accuracy_score(
                    test_labels,
                    always_high
                )
            ),
        "balanced_accuracy":
            float(
                balanced_accuracy_score(
                    test_labels,
                    always_high
                )
            ),
        "precision":
            float(
                precision_score(
                    test_labels,
                    always_high,
                    zero_division=0
                )
            ),
        "recall":
            float(
                recall_score(
                    test_labels,
                    always_high,
                    zero_division=0
                )
            ),
        "f1":
            float(
                f1_score(
                    test_labels,
                    always_high,
                    zero_division=0
                )
            ),
    },
}

# ============================================================
# 16. FINAL RESULTS
# ============================================================

test_metrics = {
    "test_loss": float(test_loss),

    "accuracy": float(
        test_accuracy
    ),

    "balanced_accuracy": float(
        test_balanced_accuracy
    ),

    "precision_high": float(
        test_precision
    ),

    "recall_high": float(
        test_recall
    ),

    "f1_high": float(
        test_f1
    ),

    "auroc": float(
        test_auroc
    ),

    "auprc": float(
        test_auprc
    ),

    "threshold": float(
        best_threshold
    ),

    "true_negative": int(tn),
    "false_positive": int(fp),
    "false_negative": int(fn),
    "true_positive": int(tp),

    "test_samples": int(
        len(test_labels)
    ),

    "actual_high": int(
        test_labels.sum()
    ),

    "actual_low": int(
        len(test_labels)
        - test_labels.sum()
    ),

    "predicted_high": int(
        test_preds.sum()
    ),

    "predicted_low": int(
        len(test_preds)
        - test_preds.sum()
    ),

    "predicted_high_rate": float(
        test_preds.mean()
    ),

    "mean_high_probability": float(
        test_probs.mean()
    ),
}

# ============================================================
# 17. SAVE RESULTS
# ============================================================

history_df = pd.DataFrame(
    history
)

history_path = os.path.join(
    PHASE7_RESULT_DIR,
    "phase7_training_history.csv"
)

history_df.to_csv(
    history_path,
    index=False
)

test_predictions_df = pd.DataFrame({
    "recording_id":
        test_df["recording_id"].to_numpy(),

    "window_index":
        test_df["window_index"].to_numpy(),

    "utility_label":
        test_labels.astype(int),

    "gate_probability_high":
        test_probs,

    "gate_prediction":
        test_preds.astype(int),
})

predictions_path = os.path.join(
    PHASE7_RESULT_DIR,
    "phase7_test_predictions.csv"
)

test_predictions_df.to_csv(
    predictions_path,
    index=False
)

final_results = {

    "phase": 7,

    "experiment":
        "Non-Causal Gate Baseline",

    "model":
        "Dilated CNN with symmetric temporal padding",

    "input_shape": [
        N_CHANNELS,
        N_TIME
    ],

    "normalization":
        "per-window channel-wise z-score",

    "utility_definition":
        "decoder prediction correctness",

    "confidence_used_for_training":
        False,

    "class_imbalance":
        {
            "method":
                "BCEWithLogitsLoss pos_weight",
            "pos_weight":
                float(POS_WEIGHT),
        },

    "model_selection":
        {
            "metric":
                "validation AUPRC",
            "best_epoch":
                int(best_epoch),
            "best_validation_auprc":
                float(best_val_auprc),
        },

    "threshold_selection":
        {
            "metric":
                "validation F1",
            "threshold":
                float(best_threshold),
            "validation_metrics":
                {
                    k: float(v)
                    for k, v
                    in best_val_threshold_metrics.items()
                    if k != "threshold"
                },
        },

    "test_metrics":
        test_metrics,

    "confusion_matrix": cm.tolist(),

    "baselines":
        baseline_metrics,

    "training":
        {
            "seed": SEED,
            "batch_size": BATCH_SIZE,
            "learning_rate": LEARNING_RATE,
            "weight_decay": WEIGHT_DECAY,
            "max_epochs": NUM_EPOCHS,
            "early_stopping_patience":
                PATIENCE,
            "training_time_sec":
                training_time,
        },

    "checkpoint":
        best_checkpoint,

    "test_evaluation_frozen":
        True,
}

results_path = os.path.join(
    PHASE7_RESULT_DIR,
    "phase7_final_results.json"
)

with open(
    results_path,
    "w"
) as f:

    json.dump(
        final_results,
        f,
        indent=2
    )

# ============================================================
# 18. FINAL REPORT
# ============================================================

print("\n" + "=" * 70)
print("PHASE 7 — FINAL TEST RESULTS")
print("=" * 70)

print(
    f"Test loss             : "
    f"{test_loss:.6f}"
)

print(
    f"Accuracy              : "
    f"{test_accuracy:.6f}"
)

print(
    f"Balanced accuracy     : "
    f"{test_balanced_accuracy:.6f}"
)

print(
    f"HIGH precision        : "
    f"{test_precision:.6f}"
)

print(
    f"HIGH recall           : "
    f"{test_recall:.6f}"
)

print(
    f"HIGH F1               : "
    f"{test_f1:.6f}"
)

print(
    f"AUROC                 : "
    f"{test_auroc:.6f}"
)

print(
    f"AUPRC                 : "
    f"{test_auprc:.6f}"
)

print(
    f"Frozen threshold      : "
    f"{best_threshold:.2f}"
)

print("\nConfusion matrix")
print(
    "                 Pred LOW   Pred HIGH"
)

print(
    f"Actual LOW       {tn:9d}   {fp:10d}"
)

print(
    f"Actual HIGH      {fn:9d}   {tp:10d}"
)

print("\nPrediction distribution")
print(
    f"Predicted LOW    : "
    f"{len(test_preds)-test_preds.sum():,}"
)

print(
    f"Predicted HIGH   : "
    f"{test_preds.sum():,}"
)

print(
    f"Predicted HIGH % : "
    f"{test_preds.mean()*100:.3f}%"
)

print("\nAlways-LOW baseline")
print(
    json.dumps(
        baseline_metrics["always_low"],
        indent=2
    )
)

print("\nAlways-HIGH baseline")
print(
    json.dumps(
        baseline_metrics["always_high"],
        indent=2
    )
)

print("\n" + "=" * 70)
print("FILES SAVED")
print("=" * 70)

print(
    "Best checkpoint:"
)
print(best_checkpoint)

print(
    "\nTraining history:"
)
print(history_path)

print(
    "\nValidation threshold table:"
)
print(threshold_path)

print(
    "\nTest predictions:"
)
print(predictions_path)

print(
    "\nFinal results:"
)
print(results_path)

print("\n" + "=" * 70)
print("PHASE 7 COMPLETE")
print("=" * 70)

print(
    "Model and threshold were selected using validation only."
)

print(
    "Test set was evaluated after freezing both."
)

print(
    "STATUS: PHASE7_NON_CAUSAL_BASELINE_COMPLETE"
)
```

```python
# ============================================================
# NEUROSENTINEL — PHASE 7 DIAGNOSTIC
# STEP 1: PROBABILITY SEPARATION ANALYSIS
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    roc_curve,
    precision_recall_curve,
)

# ============================================================
# 1. PATHS
# ============================================================

BASE = "/content/gdrive/MyDrive/NeuroSentinel"

H7_MANIFEST = os.path.join(
    BASE,
    "results",
    "phase5h7",
    "phase5h7_utility_label_manifest.csv"
)

WINDOW_DIR = os.path.join(
    BASE,
    "06_phase5_utility_labels",
    "recordings"
)

PHASE7_CHECKPOINT = os.path.join(
    BASE,
    "checkpoints",
    "phase7",
    "phase7_noncausal_best.pt"
)

PHASE7_RESULT_DIR = os.path.join(
    BASE,
    "results",
    "phase7"
)

DIAGNOSTIC_DIR = os.path.join(
    PHASE7_RESULT_DIR,
    "diagnostics"
)

os.makedirs(
    DIAGNOSTIC_DIR,
    exist_ok=True
)

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available()
    else "cpu"
)

EPS = 1e-6
BATCH_SIZE = 256

print("=" * 70)
print("PHASE 7 DIAGNOSTIC — PROBABILITY SEPARATION")
print("=" * 70)

print("Device:", DEVICE)

# ============================================================
# 2. LOAD MANIFEST
# ============================================================

h7 = pd.read_csv(
    H7_MANIFEST
)

h7["recording_id"] = (
    h7["recording_id"]
    .astype(str)
)

h7["window_index"] = (
    h7["window_index"]
    .astype(np.int64)
)

h7["utility_label"] = (
    h7["utility_label"]
    .astype(np.int64)
)

print(
    "\nTotal samples:",
    len(h7)
)

# ============================================================
# 3. MEMORY-MAP RECORDINGS
# ============================================================

recording_ids = sorted(
    h7["recording_id"].unique()
)

window_arrays = {}

for rid in recording_ids:

    path = os.path.join(
        WINDOW_DIR,
        f"{rid}_windows.npy"
    )

    if not os.path.isfile(path):
        raise FileNotFoundError(path)

    window_arrays[rid] = np.load(
        path,
        mmap_mode="r"
    )

print(
    "Memory-mapped recordings:",
    len(window_arrays)
)

# ============================================================
# 4. DATASET
# ============================================================

class DiagnosticDataset(Dataset):

    def __init__(
        self,
        dataframe,
        arrays,
        eps=1e-6,
    ):

        self.df = (
            dataframe
            .reset_index(drop=True)
        )

        self.arrays = arrays
        self.eps = eps

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):

        row = self.df.iloc[idx]

        rid = str(
            row["recording_id"]
        )

        window_idx = int(
            row["window_index"]
        )

        x = np.asarray(
            self.arrays[rid][window_idx],
            dtype=np.float32
        )

        # Same normalization used in Phase 7
        mean = x.mean(
            axis=1,
            keepdims=True
        )

        std = x.std(
            axis=1,
            keepdims=True
        )

        x = (
            x - mean
        ) / np.maximum(
            std,
            self.eps
        )

        y = int(
            row["utility_label"]
        )

        return (
            torch.from_numpy(x),
            torch.tensor(
                y,
                dtype=torch.float32
            )
        )

# ============================================================
# 5. MODEL
# ============================================================

class NonCausalGate(nn.Module):

    def __init__(
        self,
        in_channels=61,
    ):

        super().__init__()

        self.features = nn.Sequential(

            nn.Conv1d(
                in_channels,
                64,
                kernel_size=5,
                padding=2,
            ),

            nn.ReLU(),

            nn.Conv1d(
                64,
                64,
                kernel_size=5,
                padding=4,
                dilation=2,
            ),

            nn.ReLU(),

            nn.Conv1d(
                64,
                64,
                kernel_size=5,
                padding=8,
                dilation=4,
            ),

            nn.ReLU(),

            nn.Conv1d(
                64,
                32,
                kernel_size=5,
                padding=16,
                dilation=8,
            ),

            nn.ReLU(),
        )

        self.classifier = nn.Linear(
            32,
            1
        )

    def forward(self, x):

        x = self.features(x)

        x = x.mean(
            dim=2
        )

        return self.classifier(
            x
        ).squeeze(1)

model = NonCausalGate(
    in_channels=61
).to(DEVICE)

checkpoint = torch.load(
    PHASE7_CHECKPOINT,
    map_location=DEVICE,
    weights_only=False
)

model.load_state_dict(
    checkpoint["model_state_dict"]
)

model.eval()

print(
    "\n✓ Frozen Phase 7 checkpoint loaded."
)

print(
    "Checkpoint epoch:",
    checkpoint.get("epoch")
)

print(
    "Best validation AUPRC:",
    checkpoint.get("best_val_auprc")
)

# ============================================================
# 6. COLLECT PROBABILITIES
# ============================================================

results = []

for split_name in [
    "train",
    "val",
    "test",
]:

    split_df = h7[
        h7["split"] == split_name
    ].copy()

    dataset = DiagnosticDataset(
        split_df,
        window_arrays,
        EPS
    )

    loader = DataLoader(
        dataset,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=0,
        pin_memory=torch.cuda.is_available(),
    )

    labels = []
    probabilities = []

    with torch.no_grad():

        for x, y in loader:

            x = x.to(
                DEVICE,
                non_blocking=True
            )

            logits = model(x)

            probs = torch.sigmoid(
                logits
            )

            labels.append(
                y.numpy()
            )

            probabilities.append(
                probs.cpu().numpy()
            )

    labels = np.concatenate(
        labels
    ).astype(int)

    probabilities = np.concatenate(
        probabilities
    )

    split_result = split_df[
        [
            "recording_id",
            "window_index",
            "utility_label",
        ]
    ].copy()

    split_result[
        "gate_probability"
    ] = probabilities

    results.append(
        split_result
    )

all_predictions = pd.concat(
    results,
    ignore_index=True
)

prediction_path = os.path.join(
    DIAGNOSTIC_DIR,
    "phase7_probability_diagnostic_predictions.csv"
)

all_predictions.to_csv(
    prediction_path,
    index=False
)

print(
    "\n✓ Probabilities collected."
)

# ============================================================
# 7. PROBABILITY DISTRIBUTION
# ============================================================

print("\n" + "=" * 70)
print("PROBABILITY DISTRIBUTION")
print("=" * 70)

distribution_rows = []

for split_name in [
    "train",
    "val",
    "test",
]:

    subset = all_predictions[
        all_predictions["recording_id"].isin(
            h7.loc[
                h7["split"] == split_name,
                "recording_id"
            ]
        )
    ]

    # Safer split selection
    split_indices = h7[
        h7["split"] == split_name
    ].index

    # Match through recording/window pair
    split_subset = (
        all_predictions
        .merge(
            h7[
                [
                    "recording_id",
                    "window_index",
                    "split",
                ]
            ],
            on=[
                "recording_id",
                "window_index",
            ],
            how="inner"
        )
    )

    split_subset = split_subset[
        split_subset["split"] == split_name
    ]

    for utility in [0, 1]:

        p = split_subset.loc[
            split_subset["utility_label"] == utility,
            "gate_probability"
        ].to_numpy()

        distribution_rows.append({

            "split": split_name,

            "utility_label": utility,

            "n": len(p),

            "mean": float(
                np.mean(p)
            ),

            "std": float(
                np.std(p)
            ),

            "min": float(
                np.min(p)
            ),

            "p10": float(
                np.percentile(p, 10)
            ),

            "p25": float(
                np.percentile(p, 25)
            ),

            "median": float(
                np.median(p)
            ),

            "p75": float(
                np.percentile(p, 75)
            ),

            "p90": float(
                np.percentile(p, 90)
            ),

            "max": float(
                np.max(p)
            ),
        })

distribution_df = pd.DataFrame(
    distribution_rows
)

print(
    distribution_df.to_string(
        index=False
    )
)

distribution_path = os.path.join(
    DIAGNOSTIC_DIR,
    "phase7_probability_distribution.csv"
)

distribution_df.to_csv(
    distribution_path,
    index=False
)

# ============================================================
# 8. SEPARATION STATISTICS
# ============================================================

print("\n" + "=" * 70)
print("SEPARATION STATISTICS")
print("=" * 70)

separation_rows = []

for split_name in [
    "train",
    "val",
    "test",
]:

    subset = all_predictions.merge(
        h7[
            [
                "recording_id",
                "window_index",
                "split",
            ]
        ],
        on=[
            "recording_id",
            "window_index",
        ],
        how="inner"
    )

    subset = subset[
        subset["split"] == split_name
    ]

    y = subset[
        "utility_label"
    ].to_numpy()

    p = subset[
        "gate_probability"
    ].to_numpy()

    auroc = roc_auc_score(
        y,
        p
    )

    auprc = average_precision_score(
        y,
        p
    )

    high_probs = p[
        y == 1
    ]

    low_probs = p[
        y == 0
    ]

    mean_difference = (
        high_probs.mean()
        - low_probs.mean()
    )

    separation_rows.append({

        "split": split_name,

        "AUROC": float(
            auroc
        ),

        "AUPRC": float(
            auprc
        ),

        "mean_probability_HIGH":
            float(
                high_probs.mean()
            ),

        "mean_probability_LOW":
            float(
                low_probs.mean()
            ),

        "mean_difference_HIGH_minus_LOW":
            float(
                mean_difference
            ),

        "median_probability_HIGH":
            float(
                np.median(high_probs)
            ),

        "median_probability_LOW":
            float(
                np.median(low_probs)
            ),
    })

separation_df = pd.DataFrame(
    separation_rows
)

print(
    separation_df.to_string(
        index=False
    )
)

separation_path = os.path.join(
    DIAGNOSTIC_DIR,
    "phase7_probability_separation.csv"
)

separation_df.to_csv(
    separation_path,
    index=False
)

# ============================================================
# 9. TEST ROC / PR CURVE DATA
# ============================================================

test_subset = all_predictions.merge(
    h7[
        [
            "recording_id",
            "window_index",
            "split",
        ]
    ],
    on=[
        "recording_id",
        "window_index",
    ],
    how="inner"
)

test_subset = test_subset[
    test_subset["split"] == "test"
]

y_test = test_subset[
    "utility_label"
].to_numpy()

p_test = test_subset[
    "gate_probability"
].to_numpy()

fpr, tpr, roc_thresholds = roc_curve(
    y_test,
    p_test
)

precision, recall, pr_thresholds = (
    precision_recall_curve(
        y_test,
        p_test
    )
)

roc_df = pd.DataFrame({
    "fpr": fpr,
    "tpr": tpr,
    "threshold": roc_thresholds,
})

pr_df = pd.DataFrame({
    "precision": precision,
    "recall": recall,
})

roc_path = os.path.join(
    DIAGNOSTIC_DIR,
    "phase7_test_roc_curve.csv"
)

pr_path = os.path.join(
    DIAGNOSTIC_DIR,
    "phase7_test_pr_curve.csv"
)

roc_df.to_csv(
    roc_path,
    index=False
)

pr_df.to_csv(
    pr_path,
    index=False
)

# ============================================================
# 10. PROBABILITY QUANTILES
# ============================================================

print("\n" + "=" * 70)
print("TEST PROBABILITY QUANTILES")
print("=" * 70)

quantiles = [
    0.00,
    0.01,
    0.05,
    0.10,
    0.25,
    0.50,
    0.75,
    0.90,
    0.95,
    0.99,
    1.00,
]

quantile_table = pd.DataFrame({
    "quantile": quantiles,
    "probability": [
        float(
            np.quantile(
                p_test,
                q
            )
        )
        for q in quantiles
    ],
})

print(
    quantile_table.to_string(
        index=False
    )
)

quantile_path = os.path.join(
    DIAGNOSTIC_DIR,
    "phase7_test_probability_quantiles.csv"
)

quantile_table.to_csv(
    quantile_path,
    index=False
)

# ============================================================
# 11. CHECK HOW MANY WINDOWS ARE BELOW KEY THRESHOLDS
# ============================================================

print("\n" + "=" * 70)
print("TEST THRESHOLD COVERAGE")
print("=" * 70)

threshold_rows = []

for threshold in [
    0.05,
    0.10,
    0.13,
    0.15,
    0.20,
    0.25,
    0.30,
    0.40,
    0.50,
    0.60,
    0.70,
    0.80,
    0.90,
]:

    pred_high = (
        p_test >= threshold
    )

    threshold_rows.append({

        "threshold": threshold,

        "predicted_high":
            int(
                pred_high.sum()
            ),

        "predicted_low":
            int(
                (~pred_high).sum()
            ),

        "predicted_high_percent":
            float(
                pred_high.mean()
                * 100
            ),
    })

coverage_df = pd.DataFrame(
    threshold_rows
)

print(
    coverage_df.to_string(
        index=False
    )
)

coverage_path = os.path.join(
    DIAGNOSTIC_DIR,
    "phase7_test_threshold_coverage.csv"
)

coverage_df.to_csv(
    coverage_path,
    index=False
)

# ============================================================
# 12. FINAL DIAGNOSTIC SUMMARY
# ============================================================

test_sep = separation_df[
    separation_df["split"] == "test"
].iloc[0]

diagnostic_summary = {

    "phase": 7,

    "diagnostic":
        "probability_separation",

    "checkpoint":
        PHASE7_CHECKPOINT,

    "test_auroc":
        float(test_sep["AUROC"]),

    "test_auprc":
        float(test_sep["AUPRC"]),

    "test_mean_probability_high":
        float(
            test_sep[
                "mean_probability_HIGH"
            ]
        ),

    "test_mean_probability_low":
        float(
            test_sep[
                "mean_probability_LOW"
            ]
        ),

    "test_mean_probability_difference":
        float(
            test_sep[
                "mean_difference_HIGH_minus_LOW"
            ]
        ),

    "interpretation": (
        "Strong separation"
        if (
            test_sep["AUROC"] >= 0.70
            and
            test_sep[
                "mean_difference_HIGH_minus_LOW"
            ] >= 0.05
        )
        else
        "Weak or absent separation"
    ),
}

summary_path = os.path.join(
    DIAGNOSTIC_DIR,
    "phase7_probability_diagnostic_summary.json"
)

with open(
    summary_path,
    "w"
) as f:

    json.dump(
        diagnostic_summary,
        f,
        indent=2
    )

# ============================================================
# 13. FINAL OUTPUT
# ============================================================

print("\n" + "=" * 70)
print("DIAGNOSTIC COMPLETE")
print("=" * 70)

print(
    f"Test AUROC: "
    f"{test_sep['AUROC']:.6f}"
)

print(
    f"Test AUPRC: "
    f"{test_sep['AUPRC']:.6f}"
)

print(
    f"HIGH mean probability: "
    f"{test_sep['mean_probability_HIGH']:.6f}"
)

print(
    f"LOW mean probability: "
    f"{test_sep['mean_probability_LOW']:.6f}"
)

print(
    f"HIGH − LOW difference: "
    f"{test_sep['mean_difference_HIGH_minus_LOW']:.6f}"
)

print(
    "\nInterpretation:",
    diagnostic_summary["interpretation"]
)

print("\nSaved:")
print(prediction_path)
print(distribution_path)
print(separation_path)
print(roc_path)
print(pr_path)
print(quantile_path)
print(coverage_path)
print(summary_path)

print("\nSTATUS: PHASE7_DIAGNOSTIC_STEP1_COMPLETE")
```

```python
# ============================================================
# NEUROSENTINEL — PHASE 7 DIAGNOSTIC STEP 2
# HANDCRAFTED EEG FEATURES + LOGISTIC REGRESSION
# ============================================================

import os
import json
import warnings
import numpy as np
import pandas as pd

from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
)

warnings.filterwarnings("ignore")

# ============================================================
# 1. CONFIGURATION
# ============================================================

BASE = "/content/gdrive/MyDrive/NeuroSentinel"

MANIFEST_PATH = os.path.join(
    BASE,
    "results",
    "phase5h7",
    "phase5h7_utility_label_manifest.csv"
)

WINDOW_DIR = os.path.join(
    BASE,
    "06_phase5_utility_labels",
    "recordings"
)

OUTPUT_DIR = os.path.join(
    BASE,
    "results",
    "phase7",
    "diagnostics"
)

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

SEED = 42
EPS = 1e-6

np.random.seed(SEED)

print("=" * 75)
print("NEUROSENTINEL — PHASE 7 DIAGNOSTIC STEP 2")
print("HANDCRAFTED EEG FEATURES + LOGISTIC REGRESSION")
print("=" * 75)

# ============================================================
# 2. LOAD MANIFEST
# ============================================================

df = pd.read_csv(
    MANIFEST_PATH
)

df["recording_id"] = (
    df["recording_id"].astype(str)
)

df["window_index"] = (
    df["window_index"].astype(int)
)

df["utility_label"] = (
    df["utility_label"].astype(int)
)

print("\nManifest:")
print("  Total:", len(df))

for split in ["train", "val", "test"]:
    s = df[df["split"] == split]
    print(
        f"  {split:5s}: "
        f"{len(s):6d} samples | "
        f"HIGH={s.utility_label.sum():6d} | "
        f"LOW={(s.utility_label == 0).sum():6d}"
    )

# ============================================================
# 3. MEMORY-MAP ALL RECORDINGS
# ============================================================

recording_ids = sorted(
    df["recording_id"].unique()
)

arrays = {}

for rid in recording_ids:

    path = os.path.join(
        WINDOW_DIR,
        f"{rid}_windows.npy"
    )

    if not os.path.isfile(path):
        raise FileNotFoundError(
            f"Missing recording: {path}"
        )

    arrays[rid] = np.load(
        path,
        mmap_mode="r"
    )

print(
    "\n✓ Memory-mapped",
    len(arrays),
    "recordings"
)

# ============================================================
# 4. FEATURE EXTRACTION
#
# Input:
#     61 channels × 25 samples
#
# Features:
#     Per-channel:
#       mean
#       std
#       RMS
#       peak-to-peak
#       absolute mean
#       maximum absolute amplitude
#       temporal-difference RMS
#
#     Global:
#       mean
#       std
#       RMS
#       mean absolute amplitude
#       temporal-difference RMS
#
# Total:
#     61 × 7 + 5 = 432 features
# ============================================================

FEATURE_NAMES = []

for ch in range(61):

    FEATURE_NAMES.extend([
        f"ch{ch:02d}_mean",
        f"ch{ch:02d}_std",
        f"ch{ch:02d}_rms",
        f"ch{ch:02d}_ptp",
        f"ch{ch:02d}_absmean",
        f"ch{ch:02d}_maxabs",
        f"ch{ch:02d}_diff_rms",
    ])

FEATURE_NAMES.extend([
    "global_mean",
    "global_std",
    "global_rms",
    "global_absmean",
    "global_diff_rms",
])

assert len(FEATURE_NAMES) == 432


def extract_features(x):
    """
    x shape:
        (61, 25)

    Returns:
        (432,)
    """

    x = np.asarray(
        x,
        dtype=np.float32
    )

    # --------------------------------------------------------
    # Per-channel features
    # --------------------------------------------------------

    mean = np.mean(
        x,
        axis=1
    )

    std = np.std(
        x,
        axis=1
    )

    rms = np.sqrt(
        np.mean(
            x * x,
            axis=1
        )
    )

    ptp = (
        np.max(
            x,
            axis=1
        )
        -
        np.min(
            x,
            axis=1
        )
    )

    absmean = np.mean(
        np.abs(x),
        axis=1
    )

    maxabs = np.max(
        np.abs(x),
        axis=1
    )

    diff = np.diff(
        x,
        axis=1
    )

    diff_rms = np.sqrt(
        np.mean(
            diff * diff,
            axis=1
        )
    )

    features = np.column_stack([
        mean,
        std,
        rms,
        ptp,
        absmean,
        maxabs,
        diff_rms,
    ]).reshape(-1)

    # --------------------------------------------------------
    # Global features
    # --------------------------------------------------------

    global_mean = float(
        np.mean(x)
    )

    global_std = float(
        np.std(x)
    )

    global_rms = float(
        np.sqrt(
            np.mean(
                x * x
            )
        )
    )

    global_absmean = float(
        np.mean(
            np.abs(x)
        )
    )

    global_diff_rms = float(
        np.sqrt(
            np.mean(
                np.diff(x, axis=1) ** 2
            )
        )
    )

    global_features = np.array([
        global_mean,
        global_std,
        global_rms,
        global_absmean,
        global_diff_rms,
    ], dtype=np.float32)

    return np.concatenate([
        features.astype(np.float32),
        global_features,
    ])

# ============================================================
# 5. EXTRACT FEATURES
# ============================================================

def build_feature_matrix(split_name):

    subset = (
        df[df["split"] == split_name]
        .reset_index(drop=True)
    )

    X = np.empty(
        (
            len(subset),
            len(FEATURE_NAMES)
        ),
        dtype=np.float32
    )

    y = (
        subset["utility_label"]
        .to_numpy(
            dtype=np.int64
        )
    )

    for i, row in subset.iterrows():

        rid = row["recording_id"]

        idx = int(
            row["window_index"]
        )

        x = arrays[rid][idx]

        X[i] = extract_features(x)

        if (i + 1) % 20000 == 0:
            print(
                f"  {split_name}: "
                f"{i + 1}/{len(subset)}"
            )

    return (
        subset,
        X,
        y
    )

print("\n" + "=" * 75)
print("EXTRACTING EEG FEATURES")
print("=" * 75)

train_df, X_train, y_train = (
    build_feature_matrix("train")
)

val_df, X_val, y_val = (
    build_feature_matrix("val")
)

test_df, X_test, y_test = (
    build_feature_matrix("test")
)

print("\nFeature matrices:")

print(
    "  Train:",
    X_train.shape
)

print(
    "  Val:  ",
    X_val.shape
)

print(
    "  Test: ",
    X_test.shape
)

# ============================================================
# 6. SAVE FEATURE STATISTICS
# ============================================================

feature_statistics = pd.DataFrame({

    "feature": FEATURE_NAMES,

    "train_mean": np.mean(
        X_train,
        axis=0
    ),

    "train_std": np.std(
        X_train,
        axis=0
    ),

    "train_high_mean": np.mean(
        X_train[y_train == 1],
        axis=0
    ),

    "train_low_mean": np.mean(
        X_train[y_train == 0],
        axis=0
    ),
})

feature_statistics[
    "train_high_minus_low"
] = (
    feature_statistics["train_high_mean"]
    -
    feature_statistics["train_low_mean"]
)

feature_statistics[
    "standardized_mean_difference"
] = (
    feature_statistics[
        "train_high_minus_low"
    ]
    /
    (
        feature_statistics[
            "train_std"
        ]
        + EPS
    )
)

feature_statistics = (
    feature_statistics
    .replace(
        [np.inf, -np.inf],
        np.nan
    )
)

feature_statistics_path = os.path.join(
    OUTPUT_DIR,
    "phase7_step2_feature_statistics.csv"
)

feature_statistics.to_csv(
    feature_statistics_path,
    index=False
)

# ============================================================
# 7. LOGISTIC REGRESSION
# ============================================================

print("\n" + "=" * 75)
print("TRAINING BALANCED LOGISTIC REGRESSION")
print("=" * 75)

logistic_model = Pipeline([
    (
        "scaler",
        StandardScaler()
    ),

    (
        "classifier",
        LogisticRegression(
            class_weight="balanced",
            max_iter=2000,
            solver="lbfgs",
            random_state=SEED,
        )
    ),
])

logistic_model.fit(
    X_train,
    y_train
)

print(
    "✓ Logistic Regression trained."
)

# ============================================================
# 8. GET PROBABILITIES
# ============================================================

train_prob = (
    logistic_model
    .predict_proba(X_train)[:, 1]
)

val_prob = (
    logistic_model
    .predict_proba(X_val)[:, 1]
)

test_prob = (
    logistic_model
    .predict_proba(X_test)[:, 1]
)

# ============================================================
# 9. METRIC FUNCTION
# ============================================================

def calculate_metrics(
    y_true,
    prob,
    threshold,
):

    pred = (
        prob >= threshold
    ).astype(int)

    tn, fp, fn, tp = (
        confusion_matrix(
            y_true,
            pred,
            labels=[0, 1]
        ).ravel()
    )

    return {

        "threshold": float(
            threshold
        ),

        "accuracy": float(
            accuracy_score(
                y_true,
                pred
            )
        ),

        "balanced_accuracy": float(
            balanced_accuracy_score(
                y_true,
                pred
            )
        ),

        "precision": float(
            precision_score(
                y_true,
                pred,
                zero_division=0
            )
        ),

        "recall": float(
            recall_score(
                y_true,
                pred,
                zero_division=0
            )
        ),

        "f1": float(
            f1_score(
                y_true,
                pred,
                zero_division=0
            )
        ),

        "AUROC": float(
            roc_auc_score(
                y_true,
                prob
            )
        ),

        "AUPRC": float(
            average_precision_score(
                y_true,
                prob
            )
        ),

        "TN": int(tn),
        "FP": int(fp),
        "FN": int(fn),
        "TP": int(tp),

        "predicted_high_rate": float(
            np.mean(pred)
        ),
    }

# ============================================================
# 10. THRESHOLD SELECTION — VALIDATION ONLY
# ============================================================

threshold_rows = []

for threshold in np.arange(
    0.05,
    0.951,
    0.01
):

    m = calculate_metrics(
        y_val,
        val_prob,
        threshold
    )

    threshold_rows.append(m)

threshold_df = pd.DataFrame(
    threshold_rows
)

# Highest validation F1.
# Tie-break:
#     1. balanced accuracy
#     2. threshold closest to 0.5

threshold_df["distance_from_05"] = (
    np.abs(
        threshold_df["threshold"]
        - 0.50
    )
)

threshold_df = (
    threshold_df
    .sort_values(
        [
            "f1",
            "balanced_accuracy",
            "distance_from_05",
        ],
        ascending=[
            False,
            False,
            True,
        ]
    )
    .reset_index(drop=True)
)

selected_threshold = float(
    threshold_df.iloc[0]["threshold"]
)

threshold_output = os.path.join(
    OUTPUT_DIR,
    "phase7_step2_validation_thresholds.csv"
)

threshold_df.to_csv(
    threshold_output,
    index=False
)

print(
    "\nSelected validation threshold:",
    selected_threshold
)

# ============================================================
# 11. FINAL METRICS
# ============================================================

train_metrics = calculate_metrics(
    y_train,
    train_prob,
    selected_threshold
)

val_metrics = calculate_metrics(
    y_val,
    val_prob,
    selected_threshold
)

test_metrics = calculate_metrics(
    y_test,
    test_prob,
    selected_threshold
)

metrics_rows = []

for split_name, metrics in [
    ("train", train_metrics),
    ("val", val_metrics),
    ("test", test_metrics),
]:

    row = {
        "split": split_name
    }

    row.update(metrics)

    metrics_rows.append(
        row
    )

metrics_df = pd.DataFrame(
    metrics_rows
)

print("\n" + "=" * 75)
print("LOGISTIC REGRESSION RESULTS")
print("=" * 75)

print(
    metrics_df[
        [
            "split",
            "AUROC",
            "AUPRC",
            "accuracy",
            "balanced_accuracy",
            "precision",
            "recall",
            "f1",
            "predicted_high_rate",
        ]
    ].to_string(
        index=False
    )
)

# ============================================================
# 12. SHUFFLED-LABEL CONTROL
#
# Train on shuffled TRAIN labels.
# Evaluate on ORIGINAL validation labels.
#
# This tests whether the baseline can produce apparent
# performance without genuine correspondence between EEG
# features and utility labels.
# ============================================================

print("\n" + "=" * 75)
print("SHUFFLED-LABEL CONTROL")
print("=" * 75)

rng = np.random.default_rng(
    SEED
)

y_train_shuffled = (
    rng.permutation(
        y_train
    )
)

shuffle_model = Pipeline([
    (
        "scaler",
        StandardScaler()
    ),

    (
        "classifier",
        LogisticRegression(
            class_weight="balanced",
            max_iter=2000,
            solver="lbfgs",
            random_state=SEED,
        )
    ),
])

shuffle_model.fit(
    X_train,
    y_train_shuffled
)

shuffle_val_prob = (
    shuffle_model
    .predict_proba(X_val)[:, 1]
)

shuffle_test_prob = (
    shuffle_model
    .predict_proba(X_test)[:, 1]
)

shuffle_val_auroc = (
    roc_auc_score(
        y_val,
        shuffle_val_prob
    )
)

shuffle_val_auprc = (
    average_precision_score(
        y_val,
        shuffle_val_prob
    )
)

shuffle_test_auroc = (
    roc_auc_score(
        y_test,
        shuffle_test_prob
    )
)

shuffle_test_auprc = (
    average_precision_score(
        y_test,
        shuffle_test_prob
    )
)

shuffle_results = pd.DataFrame([{

    "control": "train_labels_shuffled",

    "val_AUROC":
        shuffle_val_auroc,

    "val_AUPRC":
        shuffle_val_auprc,

    "test_AUROC":
        shuffle_test_auroc,

    "test_AUPRC":
        shuffle_test_auprc,

}])

print(
    shuffle_results.to_string(
        index=False
    )
)

shuffle_path = os.path.join(
    OUTPUT_DIR,
    "phase7_step2_shuffled_control.csv"
)

shuffle_results.to_csv(
    shuffle_path,
    index=False
)

# ============================================================
# 13. TOP TRAINING FEATURE DIFFERENCES
# ============================================================

print("\n" + "=" * 75)
print("TOP EEG FEATURES BY TRAINING EFFECT SIZE")
print("=" * 75)

top_features = (
    feature_statistics
    .copy()
)

top_features[
    "abs_effect"
] = np.abs(
    top_features[
        "standardized_mean_difference"
    ]
)

top_features = (
    top_features
    .sort_values(
        "abs_effect",
        ascending=False
    )
    .head(20)
)

print(
    top_features[
        [
            "feature",
            "train_high_mean",
            "train_low_mean",
            "standardized_mean_difference",
        ]
    ].to_string(
        index=False
    )
)

top_features_path = os.path.join(
    OUTPUT_DIR,
    "phase7_step2_top_features.csv"
)

top_features.to_csv(
    top_features_path,
    index=False
)

# ============================================================
# 14. SAVE MODEL COEFFICIENTS
# ============================================================

classifier = (
    logistic_model
    .named_steps["classifier"]
)

coefficients = classifier.coef_[0]

coefficient_df = pd.DataFrame({

    "feature": FEATURE_NAMES,

    "coefficient": coefficients,

    "abs_coefficient":
        np.abs(coefficients),
})

coefficient_df = (
    coefficient_df
    .sort_values(
        "abs_coefficient",
        ascending=False
    )
)

coefficient_path = os.path.join(
    OUTPUT_DIR,
    "phase7_step2_logistic_coefficients.csv"
)

coefficient_df.to_csv(
    coefficient_path,
    index=False
)

# ============================================================
# 15. SAVE PREDICTIONS
# ============================================================

prediction_parts = []

for split_name, split_df, prob in [

    (
        "train",
        train_df,
        train_prob
    ),

    (
        "val",
        val_df,
        val_prob
    ),

    (
        "test",
        test_df,
        test_prob
    ),
]:

    temp = split_df[
        [
            "recording_id",
            "window_index",
            "utility_label",
        ]
    ].copy()

    temp["logistic_probability"] = (
        prob
    )

    temp["logistic_prediction"] = (
        prob >= selected_threshold
    ).astype(int)

    temp["split"] = split_name

    prediction_parts.append(
        temp
    )

prediction_df = pd.concat(
    prediction_parts,
    ignore_index=True
)

prediction_path = os.path.join(
    OUTPUT_DIR,
    "phase7_step2_logistic_predictions.csv"
)

prediction_df.to_csv(
    prediction_path,
    index=False
)

# ============================================================
# 16. COMPARE AGAINST PHASE 7 CNN
# ============================================================

comparison = pd.DataFrame([

    {
        "model":
            "Phase7 CNN",

        "validation_AUROC":
            0.521283,

        "validation_AUPRC":
            0.152440,

        "test_AUROC":
            0.525909,

        "test_AUPRC":
            0.167362,
    },

    {
        "model":
            "Handcrafted + Logistic Regression",

        "validation_AUROC":
            val_metrics["AUROC"],

        "validation_AUPRC":
            val_metrics["AUPRC"],

        "test_AUROC":
            test_metrics["AUROC"],

        "test_AUPRC":
            test_metrics["AUPRC"],
    },

    {
        "model":
            "Shuffled-label Logistic Regression",

        "validation_AUROC":
            shuffle_val_auroc,

        "validation_AUPRC":
            shuffle_val_auprc,

        "test_AUROC":
            shuffle_test_auroc,

        "test_AUPRC":
            shuffle_test_auprc,
    },
])

print("\n" + "=" * 75)
print("MODEL COMPARISON")
print("=" * 75)

print(
    comparison.to_string(
        index=False
    )
)

comparison_path = os.path.join(
    OUTPUT_DIR,
    "phase7_step2_model_comparison.csv"
)

comparison.to_csv(
    comparison_path,
    index=False
)

# ============================================================
# 17. AUTOMATIC INTERPRETATION
# ============================================================

test_auc = test_metrics["AUROC"]
val_auc = val_metrics["AUROC"]

test_auprc = test_metrics["AUPRC"]
val_auprc = val_metrics["AUPRC"]

shuffle_auc = shuffle_test_auroc

if test_auc < 0.55:

    interpretation = (
        "Handcrafted EEG features do not provide "
        "meaningful subject-independent utility "
        "prediction with Logistic Regression."
    )

elif test_auc < 0.60:

    interpretation = (
        "Handcrafted EEG features show weak "
        "utility-prediction signal, but not strong "
        "enough evidence for a reliable gate."
    )

else:

    interpretation = (
        "Handcrafted EEG features show potentially "
        "useful utility-prediction signal. "
        "A stronger gate representation should "
        "be investigated."
    )

summary = {

    "diagnostic":
        "phase7_step2_handcrafted_features",

    "seed":
        SEED,

    "feature_count":
        len(FEATURE_NAMES),

    "selected_validation_threshold":
        selected_threshold,

    "validation": {

        "AUROC":
            val_auc,

        "AUPRC":
            val_auprc,

        "balanced_accuracy":
            val_metrics[
                "balanced_accuracy"
            ],

        "F1":
            val_metrics["f1"],
    },

    "test": {

        "AUROC":
            test_auc,

        "AUPRC":
            test_auprc,

        "balanced_accuracy":
            test_metrics[
                "balanced_accuracy"
            ],

        "precision":
            test_metrics["precision"],

        "recall":
            test_metrics["recall"],

        "F1":
            test_metrics["f1"],
    },

    "shuffled_control": {

        "test_AUROC":
            shuffle_auc,

        "test_AUPRC":
            shuffle_test_auprc,
    },

    "interpretation":
        interpretation,
}

summary_path = os.path.join(
    OUTPUT_DIR,
    "phase7_step2_summary.json"
)

with open(
    summary_path,
    "w"
) as f:

    json.dump(
        summary,
        f,
        indent=2
    )

# ============================================================
# 18. FINAL STATUS
# ============================================================

print("\n" + "=" * 75)
print("PHASE 7 DIAGNOSTIC STEP 2 COMPLETE")
print("=" * 75)

print(
    f"Validation AUROC : {val_auc:.6f}"
)

print(
    f"Validation AUPRC  : {val_auprc:.6f}"
)

print(
    f"Test AUROC       : {test_auc:.6f}"
)

print(
    f"Test AUPRC        : {test_auprc:.6f}"
)

print(
    f"Shuffled Test AUROC: {shuffle_auc:.6f}"
)

print(
    "\nInterpretation:"
)

print(
    interpretation
)

print("\nSaved outputs:")

for path in [
    feature_statistics_path,
    threshold_output,
    shuffle_path,
    top_features_path,
    coefficient_path,
    prediction_path,
    comparison_path,
    summary_path,
]:

    print(path)

print(
    "\nSTATUS: "
    "PHASE7_DIAGNOSTIC_STEP2_COMPLETE"
)
```

```python
# ============================================================
# NEUROSENTINEL — PHASE 7 DIAGNOSTIC STEP 3
# UTILITY-LABEL CLASS CONFOUNDING ANALYSIS
# ============================================================

import os
import json
import numpy as np
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
)

# ============================================================
# 1. PATHS
# ============================================================

BASE = "/content/gdrive/MyDrive/NeuroSentinel"

H7_MANIFEST = os.path.join(
    BASE,
    "results",
    "phase5h7",
    "phase5h7_utility_label_manifest.csv"
)

PHASE7_PREDICTIONS = os.path.join(
    BASE,
    "results",
    "phase7",
    "diagnostics",
    "phase7_probability_diagnostic_predictions.csv"
)

OUTPUT_DIR = os.path.join(
    BASE,
    "results",
    "phase7",
    "diagnostics"
)

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

SEED = 42

# Correct 29-class mapping
CLASS_NAMES = (
    list("abcdefghijklmnopqrstuvwxyz")
    + [
        "<space>",
        "<number>",
        "<special>",
    ]
)

assert len(CLASS_NAMES) == 29

print("=" * 75)
print("NEUROSENTINEL — PHASE 7 DIAGNOSTIC STEP 3")
print("UTILITY-LABEL CLASS CONFOUNDING ANALYSIS")
print("=" * 75)

# ============================================================
# 2. LOAD DATA
# ============================================================

h7 = pd.read_csv(
    H7_MANIFEST
)

pred = pd.read_csv(
    PHASE7_PREDICTIONS
)

for frame in [h7, pred]:

    frame["recording_id"] = (
        frame["recording_id"]
        .astype(str)
    )

    frame["window_index"] = (
        frame["window_index"]
        .astype(int)
    )

h7["true_class"] = (
    h7["true_class"]
    .astype(int)
)

h7["predicted_class"] = (
    h7["predicted_class"]
    .astype(int)
)

h7["utility_label"] = (
    h7["utility_label"]
    .astype(int)
)

print("\nH7 manifest:", h7.shape)
print("Prediction file:", pred.shape)

# ============================================================
# 3. MERGE FROZEN GATE PREDICTIONS
# ============================================================

gate = pred[
    [
        "recording_id",
        "window_index",
        "gate_probability",
    ]
].copy()

analysis = h7.merge(
    gate,
    on=[
        "recording_id",
        "window_index",
    ],
    how="inner",
    validate="one_to_one",
)

print(
    "\nMerged rows:",
    len(analysis)
)

assert len(analysis) == len(h7)

# ============================================================
# 4. CLASS LABELS
# ============================================================

analysis["true_class_name"] = (
    analysis["true_class"]
    .map(
        lambda x:
        CLASS_NAMES[int(x)]
        if 0 <= int(x) < 29
        else f"<class_{x}>"
    )
)

analysis["predicted_class_name"] = (
    analysis["predicted_class"]
    .map(
        lambda x:
        CLASS_NAMES[int(x)]
        if 0 <= int(x) < 29
        else f"<class_{x}>"
    )
)

# ============================================================
# 5. UTILITY RATE BY TRUE CLASS
# ============================================================

print("\n" + "=" * 75)
print("UTILITY RATE BY TRUE CLASS")
print("=" * 75)

true_class_rows = []

for class_id in range(29):

    subset = analysis[
        analysis["true_class"] == class_id
    ]

    n = len(subset)

    if n == 0:
        continue

    high = int(
        subset["utility_label"].sum()
    )

    low = n - high

    true_class_rows.append({

        "class_id": class_id,

        "class_name":
            CLASS_NAMES[class_id],

        "n": n,

        "high": high,

        "low": low,

        "utility_rate":
            high / n,

        "predicted_class_accuracy":
            float(
                (
                    subset["predicted_class"]
                    ==
                    subset["true_class"]
                ).mean()
            ),
    })

true_class_df = pd.DataFrame(
    true_class_rows
)

true_class_df[
    "utility_rate_percent"
] = (
    true_class_df["utility_rate"]
    * 100
)

print(
    true_class_df[
        [
            "class_id",
            "class_name",
            "n",
            "high",
            "low",
            "utility_rate_percent",
            "predicted_class_accuracy",
        ]
    ].to_string(
        index=False
    )
)

true_class_path = os.path.join(
    OUTPUT_DIR,
    "phase7_step3_utility_by_true_class.csv"
)

true_class_df.to_csv(
    true_class_path,
    index=False
)

# ============================================================
# 6. UTILITY RATE BY PREDICTED CLASS
# ============================================================

print("\n" + "=" * 75)
print("UTILITY RATE BY PREDICTED CLASS")
print("=" * 75)

pred_class_rows = []

for class_id in range(29):

    subset = analysis[
        analysis["predicted_class"] == class_id
    ]

    n = len(subset)

    if n == 0:
        continue

    high = int(
        subset["utility_label"].sum()
    )

    low = n - high

    pred_class_rows.append({

        "predicted_class_id":
            class_id,

        "predicted_class_name":
            CLASS_NAMES[class_id],

        "n": n,

        "high": high,

        "low": low,

        "utility_rate":
            high / n,

        "true_class_accuracy":
            float(
                (
                    subset["predicted_class"]
                    ==
                    subset["true_class"]
                ).mean()
            ),
    })

pred_class_df = pd.DataFrame(
    pred_class_rows
)

pred_class_df[
    "utility_rate_percent"
] = (
    pred_class_df["utility_rate"]
    * 100
)

print(
    pred_class_df[
        [
            "predicted_class_id",
            "predicted_class_name",
            "n",
            "high",
            "low",
            "utility_rate_percent",
            "true_class_accuracy",
        ]
    ].to_string(
        index=False
    )
)

pred_class_path = os.path.join(
    OUTPUT_DIR,
    "phase7_step3_utility_by_predicted_class.csv"
)

pred_class_df.to_csv(
    pred_class_path,
    index=False
)

# ============================================================
# 7. CLASS DISTRIBUTION
# ============================================================

print("\n" + "=" * 75)
print("TRUE CLASS DISTRIBUTION")
print("=" * 75)

class_distribution = (
    true_class_df[
        [
            "class_id",
            "class_name",
            "n",
            "utility_rate",
        ]
    ]
    .copy()
)

class_distribution[
    "sample_percent"
] = (
    class_distribution["n"]
    / len(analysis)
    * 100
)

print(
    class_distribution.to_string(
        index=False
    )
)

class_distribution_path = os.path.join(
    OUTPUT_DIR,
    "phase7_step3_class_distribution.csv"
)

class_distribution.to_csv(
    class_distribution_path,
    index=False
)

# ============================================================
# 8. CLASS-ONLY BASELINE
#
# Predict utility probability using ONLY the true class.
#
# This is diagnostic only.
#
# IMPORTANT:
# The true class is not available to the actual gate.
# Therefore this is NOT an operational model.
#
# We estimate class utility rates on TRAIN only and apply
# those frozen rates to validation/test.
# ============================================================

print("\n" + "=" * 75)
print("CLASS-ONLY BASELINE")
print("=" * 75)

train_analysis = analysis[
    analysis["split"] == "train"
].copy()

val_analysis = analysis[
    analysis["split"] == "val"
].copy()

test_analysis = analysis[
    analysis["split"] == "test"
].copy()

global_train_rate = (
    train_analysis["utility_label"]
    .mean()
)

train_class_rates = (
    train_analysis
    .groupby("true_class")[
        "utility_label"
    ]
    .mean()
    .to_dict()
)

def class_probability(
    dataframe,
):

    return dataframe[
        "true_class"
    ].map(
        lambda c:
        train_class_rates.get(
            int(c),
            global_train_rate
        )
    ).to_numpy(
        dtype=float
    )

train_class_prob = class_probability(
    train_analysis
)

val_class_prob = class_probability(
    val_analysis
)

test_class_prob = class_probability(
    test_analysis
)

class_baseline_rows = []

for split_name, subset, probability in [

    (
        "train",
        train_analysis,
        train_class_prob
    ),

    (
        "val",
        val_analysis,
        val_class_prob
    ),

    (
        "test",
        test_analysis,
        test_class_prob
    ),
]:

    y = subset[
        "utility_label"
    ].to_numpy()

    # AUROC may fail if a split has one class,
    # but our current splits contain both.
    auc = roc_auc_score(
        y,
        probability
    )

    auprc = average_precision_score(
        y,
        probability
    )

    class_baseline_rows.append({

        "split": split_name,

        "AUROC": auc,

        "AUPRC": auprc,

        "mean_probability":
            probability.mean(),
    })

class_baseline_df = pd.DataFrame(
    class_baseline_rows
)

print(
    class_baseline_df.to_string(
        index=False
    )
)

class_baseline_path = os.path.join(
    OUTPUT_DIR,
    "phase7_step3_class_only_baseline.csv"
)

class_baseline_df.to_csv(
    class_baseline_path,
    index=False
)

# ============================================================
# 9. FROZEN CNN PERFORMANCE WITHIN TRUE CLASS
#
# For each class:
#
#   AUROC
#   AUPRC
#   HIGH/LOW count
#   mean gate probability
#
# AUROC is only meaningful when BOTH utility classes
# are present within the class.
# ============================================================

print("\n" + "=" * 75)
print("CNN PERFORMANCE WITHIN TRUE CLASS")
print("=" * 75)

within_class_rows = []

for class_id in range(29):

    subset = analysis[
        analysis["true_class"] == class_id
    ]

    n = len(subset)

    if n == 0:
        continue

    y = subset[
        "utility_label"
    ].to_numpy()

    p = subset[
        "gate_probability"
    ].to_numpy()

    high = int(
        np.sum(y == 1)
    )

    low = int(
        np.sum(y == 0)
    )

    if high > 0 and low > 0:

        auc = roc_auc_score(
            y,
            p
        )

        auprc = average_precision_score(
            y,
            p
        )

    else:

        auc = np.nan
        auprc = np.nan

    within_class_rows.append({

        "class_id": class_id,

        "class_name":
            CLASS_NAMES[class_id],

        "n": n,

        "high": high,

        "low": low,

        "utility_rate":
            high / n,

        "CNN_AUROC":
            auc,

        "CNN_AUPRC":
            auprc,

        "HIGH_mean_probability":
            float(
                p[y == 1].mean()
            )
            if high > 0
            else np.nan,

        "LOW_mean_probability":
            float(
                p[y == 0].mean()
            )
            if low > 0
            else np.nan,

        "HIGH_LOW_probability_difference":
            (
                float(
                    p[y == 1].mean()
                )
                -
                float(
                    p[y == 0].mean()
                )
            )
            if high > 0 and low > 0
            else np.nan,
    })

within_class_df = pd.DataFrame(
    within_class_rows
)

# ------------------------------------------------------------
# Print separately for validation and test
# ------------------------------------------------------------

for split_name in [
    "val",
    "test",
]:

    print(
        f"\n--- {split_name.upper()} ---"
    )

    split_subset = analysis[
        analysis["split"] == split_name
    ]

    rows = []

    for class_id in range(29):

        subset = split_subset[
            split_subset["true_class"]
            == class_id
        ]

        if len(subset) == 0:
            continue

        y = subset[
            "utility_label"
        ].to_numpy()

        p = subset[
            "gate_probability"
        ].to_numpy()

        high = int(
            np.sum(y == 1)
        )

        low = int(
            np.sum(y == 0)
        )

        if high > 0 and low > 0:

            auc = roc_auc_score(
                y,
                p
            )

        else:

            auc = np.nan

        rows.append({

            "class_id": class_id,

            "class_name":
                CLASS_NAMES[class_id],

            "n": len(subset),

            "high": high,

            "low": low,

            "utility_rate":
                high / len(subset),

            "AUROC": auc,
        })

    split_class_df = pd.DataFrame(
        rows
    )

    print(
        split_class_df.to_string(
            index=False
        )
    )

    split_path = os.path.join(
        OUTPUT_DIR,
        f"phase7_step3_within_class_{split_name}.csv"
    )

    split_class_df.to_csv(
        split_path,
        index=False
    )

within_class_path = os.path.join(
    OUTPUT_DIR,
    "phase7_step3_within_class_all.csv"
)

within_class_df.to_csv(
    within_class_path,
    index=False
)

# ============================================================
# 10. AGGREGATED WITHIN-CLASS PERFORMANCE
# ============================================================

print("\n" + "=" * 75)
print("AGGREGATED WITHIN-CLASS PERFORMANCE")
print("=" * 75)

within_summary_rows = []

for split_name in [
    "train",
    "val",
    "test",
]:

    subset = analysis[
        analysis["split"] == split_name
    ]

    class_aucs = []

    weighted_correct = 0
    weighted_total = 0

    for class_id in range(29):

        class_subset = subset[
            subset["true_class"] == class_id
        ]

        if len(class_subset) == 0:
            continue

        y = class_subset[
            "utility_label"
        ].to_numpy()

        p = class_subset[
            "gate_probability"
        ].to_numpy()

        if (
            np.sum(y == 1) > 0
            and
            np.sum(y == 0) > 0
        ):

            auc = roc_auc_score(
                y,
                p
            )

            class_aucs.append(
                auc
            )

    if len(class_aucs) > 0:

        macro_class_auc = float(
            np.mean(class_aucs)
        )

        median_class_auc = float(
            np.median(class_aucs)
        )

    else:

        macro_class_auc = np.nan
        median_class_auc = np.nan

    within_summary_rows.append({

        "split": split_name,

        "classes_with_both_labels":
            len(class_aucs),

        "macro_within_class_AUROC":
            macro_class_auc,

        "median_within_class_AUROC":
            median_class_auc,
    })

within_summary_df = pd.DataFrame(
    within_summary_rows
)

print(
    within_summary_df.to_string(
        index=False
    )
)

within_summary_path = os.path.join(
    OUTPUT_DIR,
    "phase7_step3_within_class_summary.csv"
)

within_summary_df.to_csv(
    within_summary_path,
    index=False
)

# ============================================================
# 11. UTILITY CONCENTRATION / CLASS VARIATION
# ============================================================

print("\n" + "=" * 75)
print("UTILITY-RATE VARIATION ACROSS TRUE CLASSES")
print("=" * 75)

observed_rates = true_class_df[
    "utility_rate"
].to_numpy()

utility_rate_stats = {

    "number_of_observed_classes":
        int(len(observed_rates)),

    "minimum_class_utility_rate":
        float(np.min(observed_rates)),

    "maximum_class_utility_rate":
        float(np.max(observed_rates)),

    "mean_class_utility_rate":
        float(np.mean(observed_rates)),

    "std_class_utility_rate":
        float(np.std(observed_rates)),

    "median_class_utility_rate":
        float(np.median(observed_rates)),

    "range_class_utility_rate":
        float(
            np.max(observed_rates)
            -
            np.min(observed_rates)
        ),
}

for key, value in utility_rate_stats.items():
    print(
        f"{key}: {value}"
    )

utility_rate_stats_path = os.path.join(
    OUTPUT_DIR,
    "phase7_step3_utility_rate_variation.json"
)

with open(
    utility_rate_stats_path,
    "w"
) as f:

    json.dump(
        utility_rate_stats,
        f,
        indent=2
    )

# ============================================================
# 12. RECORDING-LEVEL UTILITY RATE
# ============================================================

print("\n" + "=" * 75)
print("RECORDING-LEVEL UTILITY RATE")
print("=" * 75)

recording_df = (
    analysis
    .groupby(
        [
            "split",
            "recording_id",
            "subject",
        ]
    )
    .agg(
        n=(
            "utility_label",
            "size"
        ),

        high=(
            "utility_label",
            "sum"
        )
    )
    .reset_index()
)

recording_df["low"] = (
    recording_df["n"]
    -
    recording_df["high"]
)

recording_df["utility_rate"] = (
    recording_df["high"]
    /
    recording_df["n"]
)

print(
    recording_df.to_string(
        index=False
    )
)

recording_path = os.path.join(
    OUTPUT_DIR,
    "phase7_step3_recording_utility_rates.csv"
)

recording_df.to_csv(
    recording_path,
    index=False
)

# ============================================================
# 13. COMPARE CNN VS CLASS-ONLY BASELINE
# ============================================================

print("\n" + "=" * 75)
print("CNN VS CLASS-ONLY BASELINE")
print("=" * 75)

comparison_rows = []

for split_name in [
    "train",
    "val",
    "test",
]:

    subset = analysis[
        analysis["split"] == split_name
    ]

    y = subset[
        "utility_label"
    ].to_numpy()

    cnn_prob = subset[
        "gate_probability"
    ].to_numpy()

    class_prob = class_probability(
        subset
    )

    comparison_rows.append({

        "split": split_name,

        "CNN_AUROC":
            roc_auc_score(
                y,
                cnn_prob
            ),

        "CNN_AUPRC":
            average_precision_score(
                y,
                cnn_prob
            ),

        "ClassOnly_AUROC":
            roc_auc_score(
                y,
                class_prob
            ),

        "ClassOnly_AUPRC":
            average_precision_score(
                y,
                class_prob
            ),
    })

comparison_df = pd.DataFrame(
    comparison_rows
)

print(
    comparison_df.to_string(
        index=False
    )
)

comparison_path = os.path.join(
    OUTPUT_DIR,
    "phase7_step3_cnn_vs_class_only.csv"
)

comparison_df.to_csv(
    comparison_path,
    index=False
)

# ============================================================
# 14. TOP / BOTTOM CLASSES BY UTILITY RATE
# ============================================================

print("\n" + "=" * 75)
print("HIGHEST UTILITY-RATE CLASSES")
print("=" * 75)

print(
    true_class_df
    .sort_values(
        "utility_rate",
        ascending=False
    )
    [
        [
            "class_name",
            "n",
            "high",
            "low",
            "utility_rate_percent",
        ]
    ]
    .head(10)
    .to_string(
        index=False
    )
)

print("\n" + "=" * 75)
print("LOWEST UTILITY-RATE CLASSES")
print("=" * 75)

print(
    true_class_df
    .sort_values(
        "utility_rate",
        ascending=True
    )
    [
        [
            "class_name",
            "n",
            "high",
            "low",
            "utility_rate_percent",
        ]
    ]
    .head(10)
    .to_string(
        index=False
    )
)

# ============================================================
# 15. AUTOMATIC INTERPRETATION
# ============================================================

val_cnn_auc = float(
    comparison_df.loc[
        comparison_df["split"] == "val",
        "CNN_AUROC"
    ].iloc[0]
)

test_cnn_auc = float(
    comparison_df.loc[
        comparison_df["split"] == "test",
        "CNN_AUROC"
    ].iloc[0]
)

val_class_auc = float(
    comparison_df.loc[
        comparison_df["split"] == "val",
        "ClassOnly_AUROC"
    ].iloc[0]
)

test_class_auc = float(
    comparison_df.loc[
        comparison_df["split"] == "test",
        "ClassOnly_AUROC"
    ].iloc[0]
)

test_within_auc = float(
    within_summary_df.loc[
        within_summary_df["split"] == "test",
        "macro_within_class_AUROC"
    ].iloc[0]
)

# ------------------------------------------------------------
# Interpretation logic
# ------------------------------------------------------------

if (
    test_class_auc >= 0.60
    and
    test_cnn_auc < 0.60
):

    interpretation = (
        "Utility is substantially confounded by decoder "
        "class behavior. The class-only baseline explains "
        "more utility ranking than the EEG gate."
    )

elif (
    test_class_auc < 0.55
    and
    test_within_auc < 0.55
):

    interpretation = (
        "There is little evidence that decoder class alone "
        "or within-class EEG gate scores provide reliable "
        "subject-independent utility prediction."
    )

elif test_within_auc >= 0.60:

    interpretation = (
        "There may be useful within-class EEG utility signal. "
        "Further representation/model investigation is justified."
    )

else:

    interpretation = (
        "Utility shows mixed class dependence and weak "
        "within-class predictability. More diagnostic analysis "
        "is required before causal gating."
    )

# ============================================================
# 16. SAVE SUMMARY
# ============================================================

summary = {

    "diagnostic":
        "phase7_step3_class_confounding",

    "total_samples":
        int(len(analysis)),

    "classes":
        29,

    "overall_utility_rate":
        float(
            analysis["utility_label"].mean()
        ),

    "utility_rate_variation":
        utility_rate_stats,

    "validation": {

        "CNN_AUROC":
            val_cnn_auc,

        "ClassOnly_AUROC":
            val_class_auc,

        "macro_within_class_AUROC":
            float(
                within_summary_df.loc[
                    within_summary_df["split"] == "val",
                    "macro_within_class_AUROC"
                ].iloc[0]
            ),
    },

    "test": {

        "CNN_AUROC":
            test_cnn_auc,

        "ClassOnly_AUROC":
            test_class_auc,

        "macro_within_class_AUROC":
            test_within_auc,
    },

    "interpretation":
        interpretation,

    "phase8_status":
        "DEFERRED",
}

summary_path = os.path.join(
    OUTPUT_DIR,
    "phase7_step3_summary.json"
)

with open(
    summary_path,
    "w"
) as f:

    json.dump(
        summary,
        f,
        indent=2
    )

# ============================================================
# 17. FINAL STATUS
# ============================================================

print("\n" + "=" * 75)
print("PHASE 7 DIAGNOSTIC STEP 3 COMPLETE")
print("=" * 75)

print(
    f"Validation CNN AUROC       : "
    f"{val_cnn_auc:.6f}"
)

print(
    f"Validation ClassOnly AUROC : "
    f"{val_class_auc:.6f}"
)

print(
    f"Test CNN AUROC             : "
    f"{test_cnn_auc:.6f}"
)

print(
    f"Test ClassOnly AUROC       : "
    f"{test_class_auc:.6f}"
)

print(
    f"Test Macro Within-Class AUROC: "
    f"{test_within_auc:.6f}"
)

print(
    "\nInterpretation:"
)

print(
    interpretation
)

print("\nSaved outputs:")

for path in [

    true_class_path,
    pred_class_path,
    class_distribution_path,
    class_baseline_path,
    within_class_path,
    within_summary_path,
    utility_rate_stats_path,
    recording_path,
    comparison_path,
    summary_path,

]:

    print(path)

print(
    "\nSTATUS: "
    "PHASE7_DIAGNOSTIC_STEP3_COMPLETE"
)
```

```python
# ============================================================
# NEUROSENTINEL — PHASE 7 DIAGNOSTIC STEP 4
# CLASS-CONDITIONED / RESIDUAL UTILITY ANALYSIS
# ============================================================

import os
import json
import numpy as np
import pandas as pd

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
)

# ============================================================
# 1. PATHS
# ============================================================

BASE = "/content/gdrive/MyDrive/NeuroSentinel"

H7_MANIFEST = os.path.join(
    BASE,
    "results",
    "phase5h7",
    "phase5h7_utility_label_manifest.csv"
)

PHASE7_PREDICTIONS = os.path.join(
    BASE,
    "results",
    "phase7",
    "diagnostics",
    "phase7_probability_diagnostic_predictions.csv"
)

OUTPUT_DIR = os.path.join(
    BASE,
    "results",
    "phase7",
    "diagnostics"
)

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

SEED = 42
rng = np.random.default_rng(SEED)

CLASS_NAMES = (
    list("abcdefghijklmnopqrstuvwxyz")
    + [
        "<space>",
        "<number>",
        "<special>",
    ]
)

assert len(CLASS_NAMES) == 29

print("=" * 78)
print("NEUROSENTINEL — PHASE 7 DIAGNOSTIC STEP 4")
print("CLASS-CONDITIONED / RESIDUAL UTILITY ANALYSIS")
print("=" * 78)

# ============================================================
# 2. LOAD DATA
# ============================================================

h7 = pd.read_csv(
    H7_MANIFEST
)

pred = pd.read_csv(
    PHASE7_PREDICTIONS
)

h7["recording_id"] = (
    h7["recording_id"].astype(str)
)

pred["recording_id"] = (
    pred["recording_id"].astype(str)
)

h7["window_index"] = (
    h7["window_index"].astype(int)
)

pred["window_index"] = (
    pred["window_index"].astype(int)
)

h7["true_class"] = (
    h7["true_class"].astype(int)
)

h7["utility_label"] = (
    h7["utility_label"].astype(int)
)

analysis = h7.merge(
    pred[
        [
            "recording_id",
            "window_index",
            "gate_probability",
        ]
    ],
    on=[
        "recording_id",
        "window_index",
    ],
    how="inner",
    validate="one_to_one",
)

assert len(analysis) == len(h7)

print(
    "\nLoaded samples:",
    len(analysis)
)

print(
    analysis["split"]
    .value_counts()
    .sort_index()
)

# ============================================================
# 3. TRAIN-ONLY CLASS BASELINES
#
# IMPORTANT:
# These rates are estimated ONLY from TRAIN.
# They are frozen before validation/test analysis.
# ============================================================

train = analysis[
    analysis["split"] == "train"
].copy()

val = analysis[
    analysis["split"] == "val"
].copy()

test = analysis[
    analysis["split"] == "test"
].copy()

global_train_rate = float(
    train["utility_label"].mean()
)

train_class_stats = (
    train
    .groupby("true_class")
    .agg(
        train_n=(
            "utility_label",
            "size"
        ),
        train_high=(
            "utility_label",
            "sum"
        ),
        train_utility_rate=(
            "utility_label",
            "mean"
        ),
    )
    .reset_index()
)

train_class_stats["class_name"] = (
    train_class_stats["true_class"]
    .map(
        lambda x: CLASS_NAMES[int(x)]
    )
)

# Ensure all 29 classes exist in the table.
all_classes = pd.DataFrame({
    "true_class": np.arange(29)
})

train_class_stats = (
    all_classes
    .merge(
        train_class_stats,
        on="true_class",
        how="left"
    )
)

train_class_stats["class_name"] = (
    train_class_stats["true_class"]
    .map(
        lambda x: CLASS_NAMES[int(x)]
    )
)

# ============================================================
# 4. DEFINE EXPECTED CLASS UTILITY
#
# For classes absent from train, use global train utility.
# For classes with ZERO training samples, flag them.
# ============================================================

train_class_stats[
    "baseline_source"
] = np.where(
    train_class_stats["train_n"].fillna(0) > 0,
    "train_class_rate",
    "global_train_rate"
)

train_class_stats[
    "expected_utility"
] = (
    train_class_stats[
        "train_utility_rate"
    ]
    .fillna(
        global_train_rate
    )
)

print("\n" + "=" * 78)
print("FROZEN TRAIN-ONLY CLASS BASELINES")
print("=" * 78)

print(
    train_class_stats[
        [
            "true_class",
            "class_name",
            "train_n",
            "train_high",
            "train_utility_rate",
            "expected_utility",
            "baseline_source",
        ]
    ].to_string(
        index=False
    )
)

baseline_path = os.path.join(
    OUTPUT_DIR,
    "phase7_step4_train_class_baselines.csv"
)

train_class_stats.to_csv(
    baseline_path,
    index=False
)

# ============================================================
# 5. ATTACH EXPECTED UTILITY TO ALL SAMPLES
# ============================================================

class_rate_map = dict(
    zip(
        train_class_stats["true_class"],
        train_class_stats["expected_utility"]
    )
)

analysis[
    "expected_class_utility"
] = (
    analysis["true_class"]
    .map(class_rate_map)
)

analysis[
    "expected_class_utility"
] = (
    analysis[
        "expected_class_utility"
    ].fillna(
        global_train_rate
    )
)

# ============================================================
# 6. RAW RESIDUAL
#
# actual utility - expected class utility
#
# correct:
#   1 - expected
#
# incorrect:
#   0 - expected
#
# This is continuous and centered around the class baseline.
# ============================================================

analysis[
    "utility_residual"
] = (
    analysis["utility_label"]
    -
    analysis["expected_class_utility"]
)

# ============================================================
# 7. RESIDUAL SIGN TARGET
#
# Positive residual:
#   actual outcome better than class expectation
#
# Negative residual:
#   actual outcome worse than class expectation
#
# IMPORTANT:
# For a binary actual utility label, this mostly means:
#
#   correct  -> positive
#   incorrect -> negative
#
# But the magnitude depends on the class baseline.
# ============================================================

analysis[
    "residual_positive"
] = (
    analysis["utility_residual"] > 0
).astype(int)

# ============================================================
# 8. CLASS-CONDITIONED STANDARDIZED RESIDUAL
#
# Bernoulli residual variance:
# sqrt(p * (1-p))
#
# This is useful for identifying how surprising the actual
# outcome is relative to its class expectation.
# ============================================================

p = (
    analysis[
        "expected_class_utility"
    ]
    .clip(
        1e-6,
        1 - 1e-6
    )
)

analysis[
    "residual_z"
] = (
    analysis["utility_residual"]
    /
    np.sqrt(
        p * (1 - p)
    )
)

# ============================================================
# 9. SUMMARY OF RESIDUAL TARGET
# ============================================================

print("\n" + "=" * 78)
print("RESIDUAL UTILITY SUMMARY")
print("=" * 78)

residual_summary = []

for split_name in [
    "train",
    "val",
    "test",
]:

    subset = analysis[
        analysis["split"] == split_name
    ]

    residual_summary.append({

        "split": split_name,

        "n": len(subset),

        "mean_residual":
            float(
                subset[
                    "utility_residual"
                ].mean()
            ),

        "std_residual":
            float(
                subset[
                    "utility_residual"
                ].std()
            ),

        "mean_abs_residual":
            float(
                np.abs(
                    subset[
                        "utility_residual"
                    ]
                ).mean()
            ),

        "positive_residual_rate":
            float(
                subset[
                    "residual_positive"
                ].mean()
            ),

        "mean_residual_z":
            float(
                subset[
                    "residual_z"
                ].mean()
            ),

        "std_residual_z":
            float(
                subset[
                    "residual_z"
                ].std()
            ),
    })

residual_summary_df = pd.DataFrame(
    residual_summary
)

print(
    residual_summary_df.to_string(
        index=False
    )
)

residual_summary_path = os.path.join(
    OUTPUT_DIR,
    "phase7_step4_residual_summary.csv"
)

residual_summary_df.to_csv(
    residual_summary_path,
    index=False
)

# ============================================================
# 10. CLASS-CONDITIONED STATISTICS
# ============================================================

print("\n" + "=" * 78)
print("CLASS-CONDITIONED RESIDUAL STATISTICS")
print("=" * 78)

class_rows = []

for class_id in range(29):

    subset = analysis[
        analysis["true_class"] == class_id
    ]

    if len(subset) == 0:
        continue

    expected = float(
        class_rate_map.get(
            class_id,
            global_train_rate
        )
    )

    high = int(
        subset[
            "utility_label"
        ].sum()
    )

    n = len(subset)

    class_rows.append({

        "class_id":
            class_id,

        "class_name":
            CLASS_NAMES[class_id],

        "n":
            n,

        "actual_utility_rate":
            high / n,

        "train_expected_rate":
            expected,

        "observed_minus_expected":
            (high / n) - expected,

        "mean_residual":
            float(
                subset[
                    "utility_residual"
                ].mean()
            ),

        "mean_abs_residual":
            float(
                np.abs(
                    subset[
                        "utility_residual"
                    ]
                ).mean()
            ),

        "positive_residual_rate":
            float(
                subset[
                    "residual_positive"
                ].mean()
            ),
    })

class_residual_df = pd.DataFrame(
    class_rows
)

print(
    class_residual_df.to_string(
        index=False
    )
)

class_residual_path = os.path.join(
    OUTPUT_DIR,
    "phase7_step4_class_conditioned_residuals.csv"
)

class_residual_df.to_csv(
    class_residual_path,
    index=False
)

# ============================================================
# 11. IMPORTANT: TEST WHETHER CNN PROBABILITY PREDICTS
#     CLASS-CONDITIONED RESIDUAL OUTCOMES
#
# Because residual_positive is effectively equivalent to
# correctness for classes with 0 < p < 1, we report this
# carefully and also calculate within-class performance.
# ============================================================

print("\n" + "=" * 78)
print("CNN VS CLASS-CONDITIONED RESIDUAL TARGET")
print("=" * 78)

overall_rows = []

for split_name in [
    "train",
    "val",
    "test",
]:

    subset = analysis[
        analysis["split"] == split_name
    ]

    y = subset[
        "utility_label"
    ].to_numpy()

    p_gate = subset[
        "gate_probability"
    ].to_numpy()

    if (
        np.unique(y).size == 2
    ):

        auc = roc_auc_score(
            y,
            p_gate
        )

        auprc = average_precision_score(
            y,
            p_gate
        )

    else:

        auc = np.nan
        auprc = np.nan

    overall_rows.append({

        "split": split_name,

        "CNN_AUROC_on_actual_correctness":
            auc,

        "CNN_AUPRC_on_actual_correctness":
            auprc,
    })

overall_df = pd.DataFrame(
    overall_rows
)

print(
    overall_df.to_string(
        index=False
    )
)

overall_path = os.path.join(
    OUTPUT_DIR,
    "phase7_step4_overall_cnn_residual_comparison.csv"
)

overall_df.to_csv(
    overall_path,
    index=False
)

# ============================================================
# 12. WITHIN-CLASS CNN AUROC
#
# This is the most important test.
#
# A class must contain both HIGH and LOW examples.
# ============================================================

print("\n" + "=" * 78)
print("WITHIN-CLASS CNN PERFORMANCE")
print("=" * 78)

within_rows = []

for split_name in [
    "train",
    "val",
    "test",
]:

    subset_split = analysis[
        analysis["split"] == split_name
    ]

    for class_id in range(29):

        subset = subset_split[
            subset_split["true_class"]
            == class_id
        ]

        if len(subset) == 0:
            continue

        y = subset[
            "utility_label"
        ].to_numpy()

        p_gate = subset[
            "gate_probability"
        ].to_numpy()

        high = int(
            np.sum(y == 1)
        )

        low = int(
            np.sum(y == 0)
        )

        if high > 0 and low > 0:

            auc = roc_auc_score(
                y,
                p_gate
            )

            auprc = average_precision_score(
                y,
                p_gate
            )

            high_mean = float(
                p_gate[y == 1].mean()
            )

            low_mean = float(
                p_gate[y == 0].mean()
            )

        else:

            auc = np.nan
            auprc = np.nan
            high_mean = np.nan
            low_mean = np.nan

        within_rows.append({

            "split":
                split_name,

            "class_id":
                class_id,

            "class_name":
                CLASS_NAMES[class_id],

            "n":
                len(subset),

            "high":
                high,

            "low":
                low,

            "utility_rate":
                float(y.mean()),

            "AUROC":
                auc,

            "AUPRC":
                auprc,

            "HIGH_mean_gate_probability":
                high_mean,

            "LOW_mean_gate_probability":
                low_mean,

            "probability_difference":
                (
                    high_mean - low_mean
                    if not np.isnan(high_mean)
                    and not np.isnan(low_mean)
                    else np.nan
                ),
        })

within_df = pd.DataFrame(
    within_rows
)

within_path = os.path.join(
    OUTPUT_DIR,
    "phase7_step4_within_class_cnn.csv"
)

within_df.to_csv(
    within_path,
    index=False
)

for split_name in [
    "train",
    "val",
    "test",
]:

    display_df = within_df[
        within_df["split"] == split_name
    ]

    print(
        f"\n--- {split_name.upper()} ---"
    )

    print(
        display_df.to_string(
            index=False
        )
    )

# ============================================================
# 13. MACRO WITHIN-CLASS PERFORMANCE
# ============================================================

macro_rows = []

for split_name in [
    "train",
    "val",
    "test",
]:

    subset = within_df[
        within_df["split"] == split_name
    ]

    valid = subset[
        subset["AUROC"].notna()
    ]

    if len(valid) > 0:

        macro_auc = float(
            valid["AUROC"].mean()
        )

        median_auc = float(
            valid["AUROC"].median()
        )

        macro_auprc = float(
            valid["AUPRC"].mean()
        )

    else:

        macro_auc = np.nan
        median_auc = np.nan
        macro_auprc = np.nan

    macro_rows.append({

        "split":
            split_name,

        "classes_with_both_labels":
            len(valid),

        "macro_within_class_AUROC":
            macro_auc,

        "median_within_class_AUROC":
            median_auc,

        "macro_within_class_AUPRC":
            macro_auprc,
    })

macro_df = pd.DataFrame(
    macro_rows
)

print("\n" + "=" * 78)
print("MACRO WITHIN-CLASS PERFORMANCE")
print("=" * 78)

print(
    macro_df.to_string(
        index=False
    )
)

macro_path = os.path.join(
    OUTPUT_DIR,
    "phase7_step4_macro_within_class.csv"
)

macro_df.to_csv(
    macro_path,
    index=False
)

# ============================================================
# 14. CLASS-BALANCED RESAMPLING DIAGNOSTIC
#
# We construct an evaluation subset where each true class
# contributes equally.
#
# This is diagnostic only.
# ============================================================

print("\n" + "=" * 78)
print("CLASS-BALANCED EVALUATION")
print("=" * 78)

balanced_rows = []

for split_name in [
    "val",
    "test",
]:

    subset_split = analysis[
        analysis["split"] == split_name
    ]

    available_classes = (
        subset_split[
            "true_class"
        ]
        .unique()
    )

    per_class_samples = []

    # Use equal number per class based on the smallest
    # available class.
    class_counts = (
        subset_split
        .groupby("true_class")
        .size()
    )

    if len(class_counts) == 0:
        continue

    n_per_class = int(
        class_counts.min()
    )

    for class_id in sorted(
        available_classes
    ):

        class_subset = subset_split[
            subset_split["true_class"]
            == class_id
        ]

        if len(class_subset) > n_per_class:

            sampled = class_subset.sample(
                n=n_per_class,
                random_state=SEED
            )

        else:

            sampled = class_subset

        per_class_samples.append(
            sampled
        )

    balanced = pd.concat(
        per_class_samples,
        ignore_index=True
    )

    y = balanced[
        "utility_label"
    ].to_numpy()

    p_gate = balanced[
        "gate_probability"
    ].to_numpy()

    if np.unique(y).size == 2:

        auc = roc_auc_score(
            y,
            p_gate
        )

        auprc = average_precision_score(
            y,
            p_gate
        )

    else:

        auc = np.nan
        auprc = np.nan

    balanced_rows.append({

        "split":
            split_name,

        "n_per_class":
            n_per_class,

        "total_balanced_samples":
            len(balanced),

        "utility_rate":
            float(y.mean()),

        "CNN_AUROC":
            auc,

        "CNN_AUPRC":
            auprc,
    })

balanced_df = pd.DataFrame(
    balanced_rows
)

print(
    balanced_df.to_string(
        index=False
    )
)

balanced_path = os.path.join(
    OUTPUT_DIR,
    "phase7_step4_class_balanced_evaluation.csv"
)

balanced_df.to_csv(
    balanced_path,
    index=False
)

# ============================================================
# 15. SHUFFLED-LABEL CONTROL
#
# Shuffle utility labels within TRAIN only.
# The frozen CNN predictions are NOT changed.
#
# This provides a sanity check for downstream statistical
# comparisons, not a retraining experiment.
# ============================================================

print("\n" + "=" * 78)
print("SHUFFLED-LABEL CONTROL")
print("=" * 78)

shuffle_rows = []

for split_name in [
    "val",
    "test",
]:

    subset = analysis[
        analysis["split"] == split_name
    ]

    y_original = subset[
        "utility_label"
    ].to_numpy()

    # Shuffle labels while preserving the split's class
    # distribution. This is a statistical null control.
    y_shuffled = y_original.copy()

    rng.shuffle(
        y_shuffled
    )

    p_gate = subset[
        "gate_probability"
    ].to_numpy()

    auc = roc_auc_score(
        y_shuffled,
        p_gate
    )

    auprc = average_precision_score(
        y_shuffled,
        p_gate
    )

    shuffle_rows.append({

        "split":
            split_name,

        "shuffled_AUROC":
            auc,

        "shuffled_AUPRC":
            auprc,
    })

shuffle_df = pd.DataFrame(
    shuffle_rows
)

print(
    shuffle_df.to_string(
        index=False
    )
)

shuffle_path = os.path.join(
    OUTPUT_DIR,
    "phase7_step4_shuffled_control.csv"
)

shuffle_df.to_csv(
    shuffle_path,
    index=False
)

# ============================================================
# 16. CHECK WHETHER CLASS BASELINE IS STABLE ACROSS SPLITS
#
# This quantifies whether the class confounding seen in train
# transfers to unseen subjects.
# ============================================================

print("\n" + "=" * 78)
print("CLASS UTILITY RATE: TRAIN VS VALIDATION VS TEST")
print("=" * 78)

class_transfer_rows = []

for class_id in range(29):

    row = {
        "class_id":
            class_id,

        "class_name":
            CLASS_NAMES[class_id],
    }

    for split_name in [
        "train",
        "val",
        "test",
    ]:

        subset = analysis[
            (analysis["split"] == split_name)
            &
            (
                analysis["true_class"]
                == class_id
            )
        ]

        if len(subset) > 0:

            row[
                f"{split_name}_n"
            ] = len(subset)

            row[
                f"{split_name}_utility_rate"
            ] = float(
                subset[
                    "utility_label"
                ].mean()
            )

        else:

            row[
                f"{split_name}_n"
            ] = 0

            row[
                f"{split_name}_utility_rate"
            ] = np.nan

    class_transfer_rows.append(
        row
    )

class_transfer_df = pd.DataFrame(
    class_transfer_rows
)

print(
    class_transfer_df.to_string(
        index=False
    )
)

class_transfer_path = os.path.join(
    OUTPUT_DIR,
    "phase7_step4_class_utility_transfer.csv"
)

class_transfer_df.to_csv(
    class_transfer_path,
    index=False
)

# ============================================================
# 17. DETERMINE WHETHER RESIDUAL TARGET IS WELL-FORMED
# ============================================================

well_formed_rows = []

for class_id in range(29):

    subset = train[
        train["true_class"] == class_id
    ]

    if len(subset) == 0:
        continue

    rate = float(
        subset[
            "utility_label"
        ].mean()
    )

    well_formed_rows.append({

        "class_id":
            class_id,

        "class_name":
            CLASS_NAMES[class_id],

        "train_n":
            len(subset),

        "train_utility_rate":
            rate,

        "has_both_outcomes":
            bool(
                subset[
                    "utility_label"
                ].nunique()
                == 2
            ),
    })

well_formed_df = pd.DataFrame(
    well_formed_rows
)

print("\n" + "=" * 78)
print("CLASSES WITH BOTH HIGH AND LOW TRAINING OUTCOMES")
print("=" * 78)

print(
    well_formed_df[
        well_formed_df[
            "has_both_outcomes"
        ]
    ].to_string(
        index=False
    )
)

well_formed_path = os.path.join(
    OUTPUT_DIR,
    "phase7_step4_well_formed_classes.csv"
)

well_formed_df.to_csv(
    well_formed_path,
    index=False
)

# ============================================================
# 18. AUTOMATIC SCIENTIFIC INTERPRETATION
# ============================================================

val_macro_auc = float(
    macro_df.loc[
        macro_df["split"] == "val",
        "macro_within_class_AUROC"
    ].iloc[0]
)

test_macro_auc = float(
    macro_df.loc[
        macro_df["split"] == "test",
        "macro_within_class_AUROC"
    ].iloc[0]
)

test_balanced_auc = float(
    balanced_df.loc[
        balanced_df["split"] == "test",
        "CNN_AUROC"
    ].iloc[0]
)

val_cnn_auc = float(
    overall_df.loc[
        overall_df["split"] == "val",
        "CNN_AUROC_on_actual_correctness"
    ].iloc[0]
)

test_cnn_auc = float(
    overall_df.loc[
        overall_df["split"] == "test",
        "CNN_AUROC_on_actual_correctness"
    ].iloc[0]
)

if (
    test_macro_auc < 0.50
    and
    test_balanced_auc < 0.55
):

    interpretation = (
        "After controlling for character class, the frozen "
        "EEG gate shows no useful subject-independent utility "
        "signal. The class-conditioned analysis does not justify "
        "proceeding directly to causal gating."
    )

elif test_macro_auc < 0.55:

    interpretation = (
        "Class conditioning reduces the apparent utility signal "
        "substantially. Evidence for subject-independent "
        "within-class EEG utility prediction remains weak."
    )

else:

    interpretation = (
        "Some within-class utility signal remains after class "
        "conditioning. Further gate modeling may be justified "
        "before causal gating."
    )

# ============================================================
# 19. SAVE FINAL SUMMARY
# ============================================================

summary = {

    "diagnostic":
        "phase7_step4_class_conditioned_residual_utility",

    "total_samples":
        int(len(analysis)),

    "train_samples":
        int(len(train)),

    "validation_samples":
        int(len(val)),

    "test_samples":
        int(len(test)),

    "global_train_utility_rate":
        global_train_rate,

    "validation": {

        "CNN_AUROC":
            val_cnn_auc,

        "macro_within_class_AUROC":
            val_macro_auc,
    },

    "test": {

        "CNN_AUROC":
            test_cnn_auc,

        "macro_within_class_AUROC":
            test_macro_auc,

        "class_balanced_CNN_AUROC":
            test_balanced_auc,
    },

    "interpretation":
        interpretation,

    "test_data_used_for_baseline_estimation":
        False,

    "test_data_used_for_threshold_selection":
        False,

    "phase8_status":
        "DEFERRED",
}

summary_path = os.path.join(
    OUTPUT_DIR,
    "phase7_step4_summary.json"
)

with open(
    summary_path,
    "w"
) as f:

    json.dump(
        summary,
        f,
        indent=2
    )

# ============================================================
# 20. FINAL REPORT
# ============================================================

print("\n" + "=" * 78)
print("PHASE 7 DIAGNOSTIC STEP 4 COMPLETE")
print("=" * 78)

print(
    f"Validation CNN AUROC: "
    f"{val_cnn_auc:.6f}"
)

print(
    f"Validation Macro Within-Class AUROC: "
    f"{val_macro_auc:.6f}"
)

print(
    f"Test CNN AUROC: "
    f"{test_cnn_auc:.6f}"
)

print(
    f"Test Macro Within-Class AUROC: "
    f"{test_macro_auc:.6f}"
)

print(
    f"Test Class-Balanced CNN AUROC: "
    f"{test_balanced_auc:.6f}"
)

print(
    "\nINTERPRETATION:"
)

print(
    interpretation
)

print(
    "\nSaved outputs:"
)

for path in [

    baseline_path,
    residual_summary_path,
    class_residual_path,
    overall_path,
    within_path,
    macro_path,
    balanced_path,
    shuffle_path,
    class_transfer_path,
    well_formed_path,
    summary_path,

]:

    print(path)

print(
    "\nSTATUS: "
    "PHASE7_DIAGNOSTIC_STEP4_COMPLETE"
)
```

```python
# ============================================================
# NEUROSENTINEL — PHASE 7 DIAGNOSTIC STEP 5
# Subject-Independent EEG Character Decoding
# ============================================================
#
# PURPOSE
# -------
# Determine whether the frozen 61 x 25 EEG representation
# contains subject-independent character information.
#
# IMPORTANT
# ---------
# 1. Phase 6 subject split is frozen.
# 2. Phase 5H.3 normalization is reproduced exactly.
# 3. Test data is NEVER used for model/threshold selection.
# 4. Model selection uses validation macro-F1.
# 5. This is a diagnostic classifier, NOT the production decoder.
#
# OUTPUTS
# -------
# checkpoints/phase7/diagnostic_step5_character_best.pt
# results/phase7/diagnostics/phase7_step5_character_history.csv
# results/phase7/diagnostics/phase7_step5_character_results.json
# results/phase7/diagnostics/phase7_step5_character_confusion_matrix.csv
# results/phase7/diagnostics/phase7_step5_predictions.csv
# results/phase7/diagnostics/phase7_step5_per_class.csv
# results/phase7/diagnostics/phase7_step5_subject_results.csv
# ============================================================

import os
import json
import random
import warnings
from pathlib import Path

import numpy as np
import pandas as pd

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    precision_score,
    recall_score,
    confusion_matrix,
    roc_auc_score,
    average_precision_score,
)

warnings.filterwarnings("ignore")

# ------------------------------------------------------------
# 1. Paths
# ------------------------------------------------------------

BASE = "/content/gdrive/MyDrive/NeuroSentinel"

H7_MANIFEST = (
    f"{BASE}/results/phase5h7/"
    "phase5h7_utility_label_manifest.csv"
)

PHASE6_MANIFEST = (
    f"{BASE}/07_phase6_subject_split/"
    "phase6_subject_split_manifest.csv"
)

WINDOW_DIR = (
    f"{BASE}/06_phase5_utility_labels/recordings"
)

CHECKPOINT_DIR = f"{BASE}/checkpoints/phase7"
RESULT_DIR = f"{BASE}/results/phase7/diagnostics"

os.makedirs(CHECKPOINT_DIR, exist_ok=True)
os.makedirs(RESULT_DIR, exist_ok=True)

BEST_CHECKPOINT = (
    f"{CHECKPOINT_DIR}/diagnostic_step5_character_best.pt"
)

HISTORY_CSV = (
    f"{RESULT_DIR}/phase7_step5_character_history.csv"
)

RESULTS_JSON = (
    f"{RESULT_DIR}/phase7_step5_character_results.json"
)

CM_CSV = (
    f"{RESULT_DIR}/phase7_step5_character_confusion_matrix.csv"
)

PRED_CSV = (
    f"{RESULT_DIR}/phase7_step5_predictions.csv"
)

PER_CLASS_CSV = (
    f"{RESULT_DIR}/phase7_step5_per_class.csv"
)

SUBJECT_CSV = (
    f"{RESULT_DIR}/phase7_step5_subject_results.csv"
)

# ------------------------------------------------------------
# 2. Reproducibility
# ------------------------------------------------------------

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("=" * 70)
print("NEUROSENTINEL — DIAGNOSTIC STEP 5")
print("=" * 70)
print("Device:", DEVICE)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))

# ------------------------------------------------------------
# 3. Load manifests
# ------------------------------------------------------------

h7 = pd.read_csv(H7_MANIFEST)
phase6 = pd.read_csv(PHASE6_MANIFEST)

print("\nH7 manifest:", h7.shape)
print("Phase 6 manifest:", phase6.shape)

# ------------------------------------------------------------
# 4. Preserve frozen Phase 6 subject split
# ------------------------------------------------------------

split_map = (
    phase6[["recording_id", "subject", "split"]]
    .drop_duplicates()
)

if split_map["recording_id"].duplicated().any():
    raise RuntimeError(
        "Phase 6 recording IDs are not unique."
    )

h7 = h7.merge(
    split_map,
    on="recording_id",
    how="left",
    suffixes=("", "_phase6")
)

if h7["split_phase6"].isna().any():
    missing = h7.loc[
        h7["split_phase6"].isna(),
        "recording_id"
    ].unique()

    raise RuntimeError(
        f"Missing Phase 6 split for recordings: {missing}"
    )

if (h7["split"] != h7["split_phase6"]).any():
    raise RuntimeError(
        "H7 split does not match frozen Phase 6 split."
    )

h7["split"] = h7["split_phase6"]
h7.drop(columns=["split_phase6"], inplace=True)

# ------------------------------------------------------------
# 5. Character mapping
# ------------------------------------------------------------

LABEL_NAMES = (
    list("abcdefghijklmnopqrstuvwxyz")
    + ["<space>", "<number>", "<special>"]
)

NUM_CLASSES = 29

assert len(LABEL_NAMES) == NUM_CLASSES

label_to_id = {
    name: i for i, name in enumerate(LABEL_NAMES)
}

id_to_label = {
    i: name for i, name in enumerate(LABEL_NAMES)
}

# true_label is the safest source because it is already mapped
# in the Phase 5H.7 manifest.

if not h7["true_label"].isin(LABEL_NAMES).all():
    bad = h7.loc[
        ~h7["true_label"].isin(LABEL_NAMES),
        "true_label"
    ].unique()

    raise RuntimeError(
        f"Unexpected true labels found: {bad}"
    )

h7["class_id"] = h7["true_label"].map(label_to_id).astype(int)

# ------------------------------------------------------------
# 6. Verify frozen split
# ------------------------------------------------------------

print("\nFROZEN SPLIT")
print("-" * 70)

for split in ["train", "val", "test"]:
    part = h7[h7["split"] == split]

    print(
        f"{split:5s}: "
        f"{len(part):6d} samples | "
        f"{part['subject'].nunique():2d} subjects | "
        f"{part['recording_id'].nunique():2d} recordings"
    )

train_subjects = set(
    h7.loc[h7["split"] == "train", "subject"]
)

val_subjects = set(
    h7.loc[h7["split"] == "val", "subject"]
)

test_subjects = set(
    h7.loc[h7["split"] == "test", "subject"]
)

assert train_subjects.isdisjoint(val_subjects)
assert train_subjects.isdisjoint(test_subjects)
assert val_subjects.isdisjoint(test_subjects)

# ------------------------------------------------------------
# 7. Recording window lookup
# ------------------------------------------------------------

recording_files = {}

for recording_id in h7["recording_id"].unique():
    path = os.path.join(
        WINDOW_DIR,
        f"{recording_id}_windows.npy"
    )

    if not os.path.exists(path):
        raise FileNotFoundError(path)

    recording_files[recording_id] = path

print(
    "\nVerified window files:",
    len(recording_files)
)

# ------------------------------------------------------------
# 8. Dataset
# ------------------------------------------------------------

EPS = 1e-6


class EEGCharacterDataset(Dataset):

    def __init__(self, dataframe):

        self.df = dataframe.reset_index(drop=True)

        self.groups = {}

        for recording_id, idx in self.df.groupby(
            "recording_id"
        ).groups.items():

            self.groups[recording_id] = np.asarray(
                idx,
                dtype=np.int64
            )

        self.mmaps = {}

        for recording_id in self.groups:

            self.mmaps[recording_id] = np.load(
                recording_files[recording_id],
                mmap_mode="r"
            )

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):

        row = self.df.iloc[idx]

        recording_id = row["recording_id"]
        window_index = int(row["window_index"])

        x = self.mmaps[recording_id][window_index]

        x = np.asarray(
            x,
            dtype=np.float32
        )

        # Exact Phase 5H.3 normalization:
        # per-window, channel-wise z-score across time.
        mean = x.mean(axis=1, keepdims=True)
        std = x.std(axis=1, keepdims=True)

        x = (x - mean) / (
            std + EPS
        )

        y = int(row["class_id"])

        return (
            torch.from_numpy(x.copy()),
            torch.tensor(y, dtype=torch.long)
        )


# ------------------------------------------------------------
# 9. Build datasets
# ------------------------------------------------------------

train_df = h7[
    h7["split"] == "train"
].copy()

val_df = h7[
    h7["split"] == "val"
].copy()

test_df = h7[
    h7["split"] == "test"
].copy()

train_ds = EEGCharacterDataset(train_df)
val_ds = EEGCharacterDataset(val_df)
test_ds = EEGCharacterDataset(test_df)

BATCH_SIZE = 256

train_loader = DataLoader(
    train_ds,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=2,
    pin_memory=torch.cuda.is_available(),
    drop_last=False,
)

val_loader = DataLoader(
    val_ds,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=torch.cuda.is_available(),
)

test_loader = DataLoader(
    test_ds,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=torch.cuda.is_available(),
)

print("\nDatasets ready.")

# ------------------------------------------------------------
# 10. Character class statistics
# ------------------------------------------------------------

train_counts = (
    train_df["class_id"]
    .value_counts()
    .reindex(
        range(NUM_CLASSES),
        fill_value=0
    )
    .values
    .astype(np.float64)
)

print("\nTRAIN CLASS COUNTS")
print("-" * 70)

for i, count in enumerate(train_counts):
    print(
        f"{i:2d} {LABEL_NAMES[i]:>9s}: "
        f"{int(count):6d}"
    )

# ------------------------------------------------------------
# 11. Moderate inverse-sqrt class weighting
# ------------------------------------------------------------

# Avoid extremely aggressive inverse-frequency weighting.
weights = 1.0 / np.sqrt(
    np.maximum(train_counts, 1.0)
)

weights = weights / weights.mean()

class_weights = torch.tensor(
    weights,
    dtype=torch.float32,
    device=DEVICE
)

print("\nClass-weight range:")
print(
    float(class_weights.min()),
    "to",
    float(class_weights.max())
)

# ------------------------------------------------------------
# 12. Lightweight character CNN
# ------------------------------------------------------------

class CharacterCNN(nn.Module):

    def __init__(self, num_classes=29):

        super().__init__()

        self.features = nn.Sequential(

            nn.Conv1d(
                61,
                64,
                kernel_size=5,
                padding=2
            ),
            nn.ReLU(),

            nn.Conv1d(
                64,
                64,
                kernel_size=5,
                padding=4,
                dilation=2
            ),
            nn.ReLU(),

            nn.Conv1d(
                64,
                64,
                kernel_size=5,
                padding=8,
                dilation=4
            ),
            nn.ReLU(),

            nn.Conv1d(
                64,
                64,
                kernel_size=5,
                padding=16,
                dilation=8
            ),
            nn.ReLU(),
        )

        self.pool = nn.AdaptiveAvgPool1d(1)

        self.classifier = nn.Linear(
            64,
            num_classes
        )

    def forward(self, x):

        x = self.features(x)

        x = self.pool(x)

        x = x.squeeze(-1)

        return self.classifier(x)


model = CharacterCNN(
    num_classes=NUM_CLASSES
).to(DEVICE)

num_params = sum(
    p.numel()
    for p in model.parameters()
    if p.requires_grad
)

print(
    "\nTrainable parameters:",
    f"{num_params:,}"
)

# ------------------------------------------------------------
# 13. Loss / optimizer
# ------------------------------------------------------------

criterion = nn.CrossEntropyLoss(
    weight=class_weights
)

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=1e-3,
    weight_decay=1e-4
)

MAX_EPOCHS = 20
PATIENCE = 5

# ------------------------------------------------------------
# 14. Evaluation function
# ------------------------------------------------------------

def collect_predictions(
    model,
    loader,
    return_logits=False
):

    model.eval()

    all_y = []
    all_pred = []
    all_prob = []

    total_loss = 0.0
    total_n = 0

    with torch.no_grad():

        for x, y in loader:

            x = x.to(
                DEVICE,
                non_blocking=True
            )

            y = y.to(
                DEVICE,
                non_blocking=True
            )

            logits = model(x)

            loss = criterion(
                logits,
                y
            )

            probs = torch.softmax(
                logits,
                dim=1
            )

            pred = torch.argmax(
                probs,
                dim=1
            )

            n = y.size(0)

            total_loss += (
                loss.item() * n
            )

            total_n += n

            all_y.append(
                y.cpu().numpy()
            )

            all_pred.append(
                pred.cpu().numpy()
            )

            all_prob.append(
                probs.cpu().numpy()
            )

    y_true = np.concatenate(all_y)
    y_pred = np.concatenate(all_pred)
    probs = np.concatenate(all_prob)

    avg_loss = (
        total_loss / total_n
    )

    return (
        avg_loss,
        y_true,
        y_pred,
        probs
    )


def calculate_metrics(
    y_true,
    y_pred,
    probs
):

    metrics = {}

    metrics["accuracy"] = float(
        accuracy_score(
            y_true,
            y_pred
        )
    )

    metrics["balanced_accuracy"] = float(
        balanced_accuracy_score(
            y_true,
            y_pred
        )
    )

    metrics["macro_f1"] = float(
        f1_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        )
    )

    metrics["weighted_f1"] = float(
        f1_score(
            y_true,
            y_pred,
            average="weighted",
            zero_division=0
        )
    )

    metrics["macro_precision"] = float(
        precision_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        )
    )

    metrics["macro_recall"] = float(
        recall_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        )
    )

    # Multiclass one-vs-rest metrics.
    try:
        metrics["macro_auroc"] = float(
            roc_auc_score(
                y_true,
                probs,
                multi_class="ovr",
                average="macro",
                labels=list(range(NUM_CLASSES))
            )
        )
    except Exception:
        metrics["macro_auroc"] = None

    try:
        y_onehot = np.zeros_like(
            probs,
            dtype=np.float64
        )

        y_onehot[
            np.arange(len(y_true)),
            y_true
        ] = 1.0

        metrics["macro_auprc"] = float(
            average_precision_score(
                y_onehot,
                probs,
                average="macro"
            )
        )
    except Exception:
        metrics["macro_auprc"] = None

    return metrics


# ------------------------------------------------------------
# 15. Majority baseline
# ------------------------------------------------------------

majority_class = int(
    np.argmax(train_counts)
)

majority_pred = np.full(
    len(test_df),
    majority_class
)

majority_accuracy = accuracy_score(
    test_df["class_id"].values,
    majority_pred
)

majority_balanced_accuracy = (
    balanced_accuracy_score(
        test_df["class_id"].values,
        majority_pred
    )
)

majority_macro_f1 = f1_score(
    test_df["class_id"].values,
    majority_pred,
    average="macro",
    zero_division=0
)

print("\nMAJORITY BASELINE")
print("-" * 70)
print(
    "Class:",
    majority_class,
    LABEL_NAMES[majority_class]
)
print(
    "Accuracy:",
    f"{majority_accuracy:.6f}"
)
print(
    "Balanced accuracy:",
    f"{majority_balanced_accuracy:.6f}"
)
print(
    "Macro F1:",
    f"{majority_macro_f1:.6f}"
)

# ------------------------------------------------------------
# 16. Training
# ------------------------------------------------------------

history = []

best_val_macro_f1 = -np.inf
best_epoch = None
patience_counter = 0

print("\n")
print("=" * 70)
print("TRAINING")
print("=" * 70)

for epoch in range(
    1,
    MAX_EPOCHS + 1
):

    model.train()

    running_loss = 0.0
    running_n = 0

    for x, y in train_loader:

        x = x.to(
            DEVICE,
            non_blocking=True
        )

        y = y.to(
            DEVICE,
            non_blocking=True
        )

        optimizer.zero_grad(
            set_to_none=True
        )

        logits = model(x)

        loss = criterion(
            logits,
            y
        )

        loss.backward()

        optimizer.step()

        n = y.size(0)

        running_loss += (
            loss.item() * n
        )

        running_n += n

    train_loss = (
        running_loss / running_n
    )

    # Validation
    val_loss, val_y, val_pred, val_prob = (
        collect_predictions(
            model,
            val_loader
        )
    )

    val_metrics = calculate_metrics(
        val_y,
        val_pred,
        val_prob
    )

    row = {
        "epoch": epoch,
        "train_loss": train_loss,
        "val_loss": val_loss,
        **{
            f"val_{k}": v
            for k, v in val_metrics.items()
        }
    }

    history.append(row)

    print(
        f"Epoch {epoch:02d} | "
        f"train_loss={train_loss:.5f} | "
        f"val_loss={val_loss:.5f} | "
        f"val_acc={val_metrics['accuracy']:.5f} | "
        f"val_bal_acc={val_metrics['balanced_accuracy']:.5f} | "
        f"val_macroF1={val_metrics['macro_f1']:.5f}"
    )

    # --------------------------------------------------------
    # Model selection ONLY on validation macro F1
    # --------------------------------------------------------

    score = val_metrics["macro_f1"]

    if score > best_val_macro_f1:

        best_val_macro_f1 = score
        best_epoch = epoch
        patience_counter = 0

        torch.save(
            {
                "model_state_dict": model.state_dict(),
                "epoch": epoch,
                "val_macro_f1": score,
                "seed": SEED,
                "normalization": (
                    "per-window channel-wise "
                    "z-score across 25 samples"
                ),
                "architecture": (
                    "Conv1D 61-64-64-64-64, "
                    "dilations 1/2/4/8, "
                    "global average pooling, "
                    "Linear 64-29"
                ),
                "class_names": LABEL_NAMES,
            },
            BEST_CHECKPOINT
        )

    else:
        patience_counter += 1

    if patience_counter >= PATIENCE:

        print(
            f"\nEarly stopping at epoch {epoch}."
        )

        break

# Save history
history_df = pd.DataFrame(history)
history_df.to_csv(
    HISTORY_CSV,
    index=False
)

print(
    "\nBest validation macro F1:",
    f"{best_val_macro_f1:.6f}"
)

print(
    "Best epoch:",
    best_epoch
)

# ------------------------------------------------------------
# 17. Restore frozen best checkpoint
# ------------------------------------------------------------

checkpoint = torch.load(
    BEST_CHECKPOINT,
    map_location=DEVICE,
    weights_only=False
)

model.load_state_dict(
    checkpoint["model_state_dict"]
)

model.eval()

print(
    "\nRestored best checkpoint."
)

# ------------------------------------------------------------
# 18. Final validation
# ------------------------------------------------------------

val_loss, val_y, val_pred, val_prob = (
    collect_predictions(
        model,
        val_loader
    )
)

val_metrics = calculate_metrics(
    val_y,
    val_pred,
    val_prob
)

# ------------------------------------------------------------
# 19. Final test
# ------------------------------------------------------------

test_loss, test_y, test_pred, test_prob = (
    collect_predictions(
        model,
        test_loader
    )
)

test_metrics = calculate_metrics(
    test_y,
    test_pred,
    test_prob
)

# ------------------------------------------------------------
# 20. Print final metrics
# ------------------------------------------------------------

print("\n")
print("=" * 70)
print("FINAL VALIDATION")
print("=" * 70)

print(
    "Loss:",
    f"{val_loss:.6f}"
)

for k, v in val_metrics.items():
    print(
        f"{k:22s}:",
        "None" if v is None else f"{v:.6f}"
    )

print("\n")
print("=" * 70)
print("FINAL TEST — FROZEN")
print("=" * 70)

print(
    "Loss:",
    f"{test_loss:.6f}"
)

for k, v in test_metrics.items():
    print(
        f"{k:22s}:",
        "None" if v is None else f"{v:.6f}"
    )

# ------------------------------------------------------------
# 21. Predicted class distribution
# ------------------------------------------------------------

print("\nTEST PREDICTION DISTRIBUTION")
print("-" * 70)

unique_pred, pred_counts = np.unique(
    test_pred,
    return_counts=True
)

for cls, count in zip(
    unique_pred,
    pred_counts
):

    print(
        f"{cls:2d} "
        f"{LABEL_NAMES[cls]:>9s}: "
        f"{count:6d} "
        f"({count / len(test_pred) * 100:7.3f}%)"
    )

# ------------------------------------------------------------
# 22. Confusion matrix
# ------------------------------------------------------------

cm = confusion_matrix(
    test_y,
    test_pred,
    labels=list(range(NUM_CLASSES))
)

cm_df = pd.DataFrame(
    cm,
    index=LABEL_NAMES,
    columns=LABEL_NAMES
)

cm_df.to_csv(
    CM_CSV
)

# ------------------------------------------------------------
# 23. Per-class metrics
# ------------------------------------------------------------

per_class_rows = []

for cls in range(NUM_CLASSES):

    mask = test_y == cls

    n = int(mask.sum())

    if n > 0:

        class_recall = (
            (test_pred[mask] == cls).mean()
        )

    else:
        class_recall = np.nan

    predicted_n = int(
        (test_pred == cls).sum()
    )

    per_class_rows.append(
        {
            "class_id": cls,
            "label": LABEL_NAMES[cls],
            "test_count": n,
            "predicted_count": predicted_n,
            "recall": float(class_recall)
            if not np.isnan(class_recall)
            else None,
        }
    )

per_class_df = pd.DataFrame(
    per_class_rows
)

per_class_df.to_csv(
    PER_CLASS_CSV,
    index=False
)

# ------------------------------------------------------------
# 24. Subject-level test performance
# ------------------------------------------------------------

test_prediction_df = test_df.copy()

test_prediction_df["predicted_class"] = test_pred
test_prediction_df["predicted_label"] = [
    LABEL_NAMES[i]
    for i in test_pred
]

subject_rows = []

for subject, group in test_prediction_df.groupby(
    "subject"
):

    y_true_s = group["class_id"].values
    y_pred_s = group["predicted_class"].values

    subject_rows.append(
        {
            "subject": subject,
            "n_samples": len(group),
            "accuracy": accuracy_score(
                y_true_s,
                y_pred_s
            ),
            "balanced_accuracy": balanced_accuracy_score(
                y_true_s,
                y_pred_s
            ),
            "macro_f1": f1_score(
                y_true_s,
                y_pred_s,
                average="macro",
                zero_division=0
            ),
        }
    )

subject_df = pd.DataFrame(
    subject_rows
)

subject_df.to_csv(
    SUBJECT_CSV,
    index=False
)

print("\nSUBJECT-LEVEL TEST RESULTS")
print("-" * 70)
print(subject_df.to_string(index=False))

# ------------------------------------------------------------
# 25. Save predictions
# ------------------------------------------------------------

prediction_output = test_df[
    [
        "recording_id",
        "subject",
        "window_index",
        "true_label",
        "class_id",
    ]
].copy()

prediction_output[
    "predicted_class"
] = test_pred

prediction_output[
    "predicted_label"
] = [
    LABEL_NAMES[i]
    for i in test_pred
]

prediction_output[
    "predicted_confidence"
] = test_prob.max(axis=1)

prediction_output.to_csv(
    PRED_CSV,
    index=False
)

# ------------------------------------------------------------
# 26. Final diagnostic verdict
# ------------------------------------------------------------

test_bal_acc = test_metrics[
    "balanced_accuracy"
]

test_macro_f1 = test_metrics[
    "macro_f1"
]

test_accuracy = test_metrics[
    "accuracy"
]

if (
    test_bal_acc > 0.20
    and test_macro_f1 > 0.10
):

    verdict = (
        "EEG shows evidence of subject-independent "
        "character information. The utility formulation "
        "should be investigated further."
    )

elif (
    test_bal_acc > 0.05
    or test_macro_f1 > 0.03
):

    verdict = (
        "EEG shows weak character information, but "
        "subject-independent decoding is limited."
    )

else:

    verdict = (
        "No convincing subject-independent character "
        "decoding signal was demonstrated by this "
        "diagnostic classifier."
    )

# ------------------------------------------------------------
# 27. Save final JSON
# ------------------------------------------------------------

final_results = {
    "phase": "Phase 7 Diagnostic Step 5",
    "purpose": (
        "Subject-independent EEG character decoding "
        "diagnostic"
    ),

    "seed": SEED,

    "device": str(DEVICE),

    "num_classes": NUM_CLASSES,
    "class_names": LABEL_NAMES,

    "input_shape": [
        61,
        25
    ],

    "normalization": (
        "per-window channel-wise z-score "
        "across 25 time samples, EPS=1e-6"
    ),

    "frozen_split": {
        "train_subjects": sorted(
            list(train_subjects)
        ),
        "val_subjects": sorted(
            list(val_subjects)
        ),
        "test_subjects": sorted(
            list(test_subjects)
        ),
    },

    "sample_counts": {
        "train": len(train_df),
        "val": len(val_df),
        "test": len(test_df),
    },

    "architecture": (
        "Conv1D 61->64 k5 p2; "
        "Conv1D 64->64 k5 dilation2 p4; "
        "Conv1D 64->64 k5 dilation4 p8; "
        "Conv1D 64->64 k5 dilation8 p16; "
        "global average pooling; "
        "Linear 64->29"
    ),

    "trainable_parameters": int(num_params),

    "selection_metric": (
        "validation macro F1"
    ),

    "best_epoch": int(best_epoch),

    "best_validation_macro_f1": float(
        best_val_macro_f1
    ),

    "validation": {
        "loss": float(val_loss),
        **val_metrics,
    },

    "test": {
        "loss": float(test_loss),
        **test_metrics,
    },

    "majority_baseline": {
        "class_id": majority_class,
        "label": LABEL_NAMES[majority_class],
        "accuracy": float(
            majority_accuracy
        ),
        "balanced_accuracy": float(
            majority_balanced_accuracy
        ),
        "macro_f1": float(
            majority_macro_f1
        ),
    },

    "test_unique_predicted_classes": int(
        len(unique_pred)
    ),

    "test_predicted_class_count": {
        LABEL_NAMES[int(cls)]: int(count)
        for cls, count in zip(
            unique_pred,
            pred_counts
        )
    },

    "verdict": verdict,

    "outputs": {
        "checkpoint": BEST_CHECKPOINT,
        "history": HISTORY_CSV,
        "results": RESULTS_JSON,
        "confusion_matrix": CM_CSV,
        "predictions": PRED_CSV,
        "per_class": PER_CLASS_CSV,
        "subject_results": SUBJECT_CSV,
    },
}

with open(
    RESULTS_JSON,
    "w"
) as f:

    json.dump(
        final_results,
        f,
        indent=2
    )

# ------------------------------------------------------------
# 28. Final summary
# ------------------------------------------------------------

print("\n")
print("=" * 70)
print("DIAGNOSTIC STEP 5 COMPLETE")
print("=" * 70)

print(
    f"Best epoch              : {best_epoch}"
)

print(
    f"Validation macro F1     : "
    f"{val_metrics['macro_f1']:.6f}"
)

print(
    f"Test accuracy           : "
    f"{test_accuracy:.6f}"
)

print(
    f"Test balanced accuracy  : "
    f"{test_bal_acc:.6f}"
)

print(
    f"Test macro F1           : "
    f"{test_macro_f1:.6f}"
)

print(
    f"Test macro AUROC        : "
    f"{test_metrics['macro_auroc']}"
)

print(
    f"Test macro AUPRC        : "
    f"{test_metrics['macro_auprc']}"
)

print(
    f"Unique predicted classes: "
    f"{len(unique_pred)}/29"
)

print("\nVERDICT:")
print(verdict)

print("\nSaved:")
print(BEST_CHECKPOINT)
print(HISTORY_CSV)
print(RESULTS_JSON)
print(CM_CSV)
print(PRED_CSV)
print(PER_CLASS_CSV)
print(SUBJECT_CSV)

print("\n" + "=" * 70)
```

```python
# ============================================================
# NEUROSENTINEL — STEP 6
# PHASE 3 ↔ PHASE 5 TIMESTAMP / EVENT RELATIONSHIP
#
# Purpose:
# Determine whether Phase 3 and Phase 5 use different
# recordings/time origins and whether a constant timestamp
# offset can map the two event streams.
#
# NO WINDOW RECONSTRUCTION YET.
# ============================================================

import os
import numpy as np
import pandas as pd

BASE = "/content/gdrive/MyDrive/NeuroSentinel"

RID = "003_S2_task1"

PHASE5_METADATA = (
    f"{BASE}/06_phase5_utility_labels/recordings/"
    f"{RID}_metadata.csv"
)

PHASE3_ALIGNMENT = (
    f"{BASE}/04_phase3_event_alignment/"
    "phase3_keystroke_alignment.csv"
)

meta = pd.read_csv(PHASE5_METADATA)
phase3 = pd.read_csv(PHASE3_ALIGNMENT)

# ------------------------------------------------------------
# 1. Normalize columns
# ------------------------------------------------------------

for c in [
    "trial_id",
    "keystroke_index",
]:
    meta[c] = pd.to_numeric(
        meta[c],
        errors="coerce"
    )

for c in [
    "trial_index",
    "keystroke_index",
]:
    phase3[c] = pd.to_numeric(
        phase3[c],
        errors="coerce"
    )

meta["timestamp"] = pd.to_numeric(
    meta["timestamp"],
    errors="coerce"
)

phase3["timestamp_seconds"] = pd.to_numeric(
    phase3["timestamp_seconds"],
    errors="coerce"
)

phase3["sample_1000hz"] = pd.to_numeric(
    phase3["sample_1000hz"],
    errors="coerce"
)

print("=" * 80)
print("DATASET SUMMARY")
print("=" * 80)

print("Phase 5 rows:", len(meta))
print("Phase 3 rows:", len(phase3))

print(
    "\nPhase 5 time:",
    meta["timestamp"].min(),
    "→",
    meta["timestamp"].max()
)

print(
    "Phase 3 time:",
    phase3["timestamp_seconds"].min(),
    "→",
    phase3["timestamp_seconds"].max()
)

# ------------------------------------------------------------
# 2. Compare trial counts
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("TRIAL STRUCTURE")
print("=" * 80)

print(
    "Phase 5 unique trials:",
    meta["trial_id"].nunique()
)

print(
    "Phase 3 unique trials:",
    phase3["trial_index"].nunique()
)

print("\nPhase 5 trial counts:")
print(
    meta.groupby("trial_id")
    .size()
    .head(20)
    .to_string()
)

print("\nPhase 3 trial counts:")
print(
    phase3.groupby("trial_index")
    .size()
    .head(20)
    .to_string()
)

# ------------------------------------------------------------
# 3. Compare first/last timestamp per trial
# ------------------------------------------------------------

p5_trial = (
    meta.groupby("trial_id")
    .agg(
        p5_first=("timestamp", "min"),
        p5_last=("timestamp", "max"),
        p5_n=("keystroke_index", "count")
    )
    .reset_index()
)

p3_trial = (
    phase3.groupby("trial_index")
    .agg(
        p3_first=("timestamp_seconds", "min"),
        p3_last=("timestamp_seconds", "max"),
        p3_n=("keystroke_index", "count")
    )
    .reset_index()
)

trial_compare = p5_trial.merge(
    p3_trial,
    left_on="trial_id",
    right_on="trial_index",
    how="outer"
)

trial_compare["first_offset"] = (
    trial_compare["p5_first"]
    -
    trial_compare["p3_first"]
)

trial_compare["last_offset"] = (
    trial_compare["p5_last"]
    -
    trial_compare["p3_last"]
)

print("\n" + "=" * 80)
print("TRIAL-LEVEL COMPARISON")
print("=" * 80)

print(
    trial_compare.head(30).to_string(
        index=False
    )
)

# ------------------------------------------------------------
# 4. Trial offset statistics
# ------------------------------------------------------------

valid_offsets = (
    trial_compare["first_offset"]
    .dropna()
    .to_numpy()
)

if len(valid_offsets):

    print("\n" + "=" * 80)
    print("TRIAL FIRST-EVENT OFFSETS")
    print("=" * 80)

    print(
        pd.Series(valid_offsets)
        .describe()
        .to_string()
    )

# ------------------------------------------------------------
# 5. Test likely global time offsets
#
# We test whether adding/subtracting common offsets aligns
# Phase 3 with Phase 5.
# ------------------------------------------------------------

p5_times = (
    meta["timestamp"]
    .dropna()
    .to_numpy(
        dtype=np.float64
    )
)

p3_times = (
    phase3["timestamp_seconds"]
    .dropna()
    .to_numpy(
        dtype=np.float64
    )
)

print("\n" + "=" * 80)
print("GLOBAL OFFSET SEARCH")
print("=" * 80)

# Coarse offset range.
offset_grid = np.arange(
    -100.0,
    100.001,
    0.01
)

# To keep computation manageable, use first 500
# Phase 5 timestamps.
query_times = p5_times[:500]

best_results = []

for offset in offset_grid:

    shifted = (
        query_times[:, None]
        -
        (
            p3_times[None, :]
            + offset
        )
    )

    nearest_error = np.min(
        np.abs(shifted),
        axis=1
    )

    best_results.append({
        "offset": float(offset),
        "median_error": float(
            np.median(nearest_error)
        ),
        "mean_error": float(
            np.mean(nearest_error)
        ),
        "within_5ms": float(
            np.mean(
                nearest_error <= 0.005
            )
        ),
        "within_20ms": float(
            np.mean(
                nearest_error <= 0.020
            )
        ),
        "within_50ms": float(
            np.mean(
                nearest_error <= 0.050
            )
        ),
    })

offset_df = pd.DataFrame(
    best_results
)

# Rank primarily by median error.
offset_df = offset_df.sort_values(
    [
        "median_error",
        "mean_error"
    ]
)

print(
    offset_df.head(20).to_string(
        index=False
    )
)

best_offset = offset_df.iloc[0]

print("\nBEST GLOBAL OFFSET:")
print(
    best_offset.to_string()
)

# ------------------------------------------------------------
# 6. Apply best offset to ALL Phase 5 timestamps
# ------------------------------------------------------------

offset = float(
    best_offset["offset"]
)

shifted_p3 = (
    phase3["timestamp_seconds"]
    .to_numpy(
        dtype=np.float64
    )
    + offset
)

all_p5_times = (
    meta["timestamp"]
    .to_numpy(
        dtype=np.float64
    )
)

nearest_idx = []

nearest_error = []

for t in all_p5_times:

    idx = np.argmin(
        np.abs(
            shifted_p3 - t
        )
    )

    nearest_idx.append(idx)

    nearest_error.append(
        abs(
            shifted_p3[idx] - t
        )
    )

nearest_idx = np.asarray(
    nearest_idx
)

nearest_error = np.asarray(
    nearest_error
)

# ------------------------------------------------------------
# 7. Build timestamp mapping
# ------------------------------------------------------------

mapping = meta[
    [
        "recording_id",
        "trial_id",
        "keystroke_index",
        "timestamp",
        "target",
        "text"
    ]
].copy()

mapping["phase3_row"] = nearest_idx

mapping["phase3_timestamp_raw"] = (
    phase3.iloc[
        nearest_idx
    ]["timestamp_seconds"]
    .to_numpy()
)

mapping["phase3_timestamp_shifted"] = (
    shifted_p3[
        nearest_idx
    ]
)

mapping["phase3_sample_1000hz"] = (
    phase3.iloc[
        nearest_idx
    ]["sample_1000hz"]
    .to_numpy()
)

mapping["phase3_trial"] = (
    phase3.iloc[
        nearest_idx
    ]["trial_index"]
    .to_numpy()
)

mapping["phase3_keystroke"] = (
    phase3.iloc[
        nearest_idx
    ]["keystroke_index"]
    .to_numpy()
)

mapping["phase3_character"] = (
    phase3.iloc[
        nearest_idx
    ]["character"]
    .to_numpy()
)

mapping["timestamp_error"] = (
    mapping["timestamp"]
    -
    mapping["phase3_timestamp_shifted"]
)

# ------------------------------------------------------------
# 8. Mapping statistics
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("BEST-OFFSET MAPPING")
print("=" * 80)

print(
    "Best offset:",
    offset,
    "seconds"
)

print(
    "Median error:",
    np.median(
        nearest_error
    ),
    "seconds"
)

print(
    "Mean error:",
    np.mean(
        nearest_error
    ),
    "seconds"
)

print(
    "Maximum error:",
    np.max(
        nearest_error
    ),
    "seconds"
)

for tolerance in [
    0.001,
    0.005,
    0.010,
    0.020,
    0.050,
    0.100,
]:

    count = (
        nearest_error
        <= tolerance
    ).sum()

    print(
        f"Within ±{tolerance:.3f}s:",
        count,
        "/",
        len(mapping),
        f"({100 * count / len(mapping):.2f}%)"
    )

# ------------------------------------------------------------
# 9. Check trial/keystroke consistency after timestamp match
# ------------------------------------------------------------

same_trial = (
    mapping["trial_id"]
    ==
    mapping["phase3_trial"]
)

same_key = (
    mapping["keystroke_index"]
    ==
    mapping["phase3_keystroke"]
)

print("\n" + "=" * 80)
print("INDEX CONSISTENCY AFTER TIMESTAMP MATCH")
print("=" * 80)

print(
    "Same trial:",
    f"{same_trial.mean() * 100:.2f}%"
)

print(
    "Same keystroke:",
    f"{same_key.mean() * 100:.2f}%"
)

# ------------------------------------------------------------
# 10. Character consistency
# ------------------------------------------------------------

target_chars = (
    mapping["target"]
    .astype(str)
    .str.lower()
)

phase3_chars = (
    mapping["phase3_character"]
    .fillna("")
    .astype(str)
    .str.lower()
)

character_match = (
    target_chars
    ==
    phase3_chars
)

print("\n" + "=" * 80)
print("CHARACTER CONSISTENCY")
print("=" * 80)

print(
    "Character match:",
    f"{character_match.mean() * 100:.2f}%"
)

# ------------------------------------------------------------
# 11. Show first 30 mappings
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("FIRST 30 TIMESTAMP MAPPINGS")
print("=" * 80)

print(
    mapping[
        [
            "trial_id",
            "keystroke_index",
            "timestamp",
            "phase3_timestamp_raw",
            "phase3_timestamp_shifted",
            "timestamp_error",
            "phase3_trial",
            "phase3_keystroke",
            "target",
            "phase3_character"
        ]
    ]
    .head(30)
    .to_string(index=False)
)

# ------------------------------------------------------------
# 12. Show worst mappings
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("WORST 20 TIMESTAMP MAPPINGS")
print("=" * 80)

print(
    mapping[
        [
            "trial_id",
            "keystroke_index",
            "timestamp",
            "phase3_timestamp_shifted",
            "timestamp_error",
            "phase3_trial",
            "phase3_keystroke",
            "target",
            "phase3_character"
        ]
    ]
    .sort_values(
        "timestamp_error",
        key=lambda x: np.abs(x),
        ascending=False
    )
    .head(20)
    .to_string(index=False)
)

# ------------------------------------------------------------
# 13. Check whether timestamp ordering is preserved
# ------------------------------------------------------------

phase3_selected_times = (
    shifted_p3[
        nearest_idx
    ]
)

monotonic = np.all(
    np.diff(
        phase3_selected_times
    ) >= 0
)

print("\n" + "=" * 80)
print("ORDER CHECK")
print("=" * 80)

print(
    "Phase 3 matched timestamps monotonic:",
    monotonic
)

# ------------------------------------------------------------
# 14. Save diagnostic mapping
# ------------------------------------------------------------

mapping_path = (
    f"{BASE}/results/phase6_diagnostic/"
    f"{RID}_timestamp_mapping.csv"
)

mapping.to_csv(
    mapping_path,
    index=False
)

offset_path = (
    f"{BASE}/results/phase6_diagnostic/"
    f"{RID}_offset_search.csv"
)

offset_df.to_csv(
    offset_path,
    index=False
)

print("\n" + "=" * 80)
print("FILES SAVED")
print("=" * 80)

print(mapping_path)
print(offset_path)

print("\n" + "=" * 80)
print("STATUS")
print("=" * 80)

print(
    "Do NOT reconstruct the windows yet."
)

print(
    "The key outputs are the BEST GLOBAL OFFSET, "
    "timestamp match rates, trial/keystroke consistency, "
    "and character consistency."
)
```

```python
# ============================================================
# NEUROSENTINEL — STEP 6A
# VERIFY PHASE 5 WINDOWS AGAINST ORIGINAL BRAINVISION EEG
#
# Purpose:
#   1. Load original BrainVision recording
#   2. Apply the Phase-2 preprocessing pipeline
#   3. Load Phase-5 windows + metadata
#   4. Convert Phase-5 timestamps to raw EEG sample positions
#   5. Reconstruct the existing 61×25 windows
#   6. Compare reconstructed windows numerically with Phase-5
#      saved windows
#
# IMPORTANT:
#   - This cell DOES NOT generate larger windows.
#   - This cell DOES NOT modify any frozen split/labels.
#   - Continue only if reconstruction is numerically consistent.
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import mne

# ------------------------------------------------------------
# CONFIG
# ------------------------------------------------------------

BASE = "/content/gdrive/MyDrive/NeuroSentinel"

RID = "003_S2_task1"

RAW_PATH = (
    f"{BASE}/data/SpanishBCBL/EEG/EEG/"
    "003_DECOMEG_S2_9337_task1.vhdr"
)

WINDOW_PATH = (
    f"{BASE}/06_phase5_utility_labels/recordings/"
    f"{RID}_windows.npy"
)

META_PATH = (
    f"{BASE}/06_phase5_utility_labels/recordings/"
    f"{RID}_metadata.csv"
)

OUTPUT_DIR = (
    f"{BASE}/results/phase6_diagnostic"
)

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

# Existing Phase-5 geometry
BASELINE_TMIN = -0.2
BASELINE_TMAX = 0.3

EXPECTED_SAMPLES = 25
EXPECTED_CHANNELS = 61

# ------------------------------------------------------------
# 1. LOAD PHASE 5 DATA
# ------------------------------------------------------------

print("=" * 80)
print("LOADING PHASE 5 DATA")
print("=" * 80)

phase5_windows = np.load(
    WINDOW_PATH,
    mmap_mode="r"
)

meta = pd.read_csv(
    META_PATH
)

print(
    "Phase 5 windows:",
    phase5_windows.shape,
    phase5_windows.dtype
)

print(
    "Phase 5 metadata:",
    meta.shape
)

print(
    "Columns:",
    list(meta.columns)
)

assert len(phase5_windows) == len(meta), (
    "Window count and metadata count differ."
)

assert phase5_windows.shape[1:] == (
    EXPECTED_CHANNELS,
    EXPECTED_SAMPLES
), (
    f"Unexpected Phase-5 shape: "
    f"{phase5_windows.shape}"
)

# ------------------------------------------------------------
# 2. LOAD ORIGINAL BRAINVISION EEG
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("LOADING ORIGINAL BRAINVISION EEG")
print("=" * 80)

print(
    "Raw:",
    RAW_PATH
)

assert os.path.exists(RAW_PATH), (
    f"Raw file not found:\n{RAW_PATH}"
)

raw = mne.io.read_raw_brainvision(
    RAW_PATH,
    preload=True,
    verbose=False
)

print(raw)

print(
    "\nOriginal channels:",
    len(raw.ch_names)
)

print(
    "Original samples:",
    raw.n_times
)

print(
    "Original sampling rate:",
    raw.info["sfreq"]
)

print(
    "Original duration:",
    raw.times[-1],
    "seconds"
)

print(
    "Original first time:",
    raw.times[0]
)

# ------------------------------------------------------------
# 3. IDENTIFY EOG CHANNELS
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("CHANNEL INSPECTION")
print("=" * 80)

eog_channels = []

for ch in raw.ch_names:

    name = ch.upper()

    if (
        "EOG" in name
        or "HEOG" in name
        or "VEOG" in name
        or "EYE" in name
    ):
        eog_channels.append(ch)

print(
    "Detected EOG channels:",
    eog_channels
)

print(
    "Number of EOG channels:",
    len(eog_channels)
)

# ------------------------------------------------------------
# 4. APPLY PHASE-2 PREPROCESSING
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("APPLYING PHASE-2 PREPROCESSING")
print("=" * 80)

# Remove EOG channels.
if eog_channels:
    raw.drop_channels(
        eog_channels
    )

print(
    "Channels after EOG removal:",
    len(raw.ch_names)
)

# Standard 10-20 montage.
montage = mne.channels.make_standard_montage(
    "standard_1020"
)

raw.set_montage(
    montage,
    on_missing="ignore",
    verbose=False
)

# Filter 0.1–20 Hz.
print("Filtering: 0.1–20 Hz")

raw.filter(
    l_freq=0.1,
    h_freq=20.0,
    verbose=False
)

# Resample to 50 Hz.
print("Resampling: 50 Hz")

raw.resample(
    50.0,
    verbose=False
)

print(
    "\nPreprocessed channels:",
    len(raw.ch_names)
)

print(
    "Preprocessed sampling rate:",
    raw.info["sfreq"]
)

print(
    "Preprocessed samples:",
    raw.n_times
)

print(
    "Preprocessed duration:",
    raw.times[-1]
)

# ------------------------------------------------------------
# 5. VERIFY CHANNEL COUNT
# ------------------------------------------------------------

if len(raw.ch_names) != EXPECTED_CHANNELS:

    print("\nWARNING:")
    print(
        f"Expected {EXPECTED_CHANNELS} EEG channels, "
        f"but found {len(raw.ch_names)}."
    )

    print(
        "Channels:"
    )

    print(
        raw.ch_names
    )

    raise RuntimeError(
        "Channel count does not match Phase-5."
    )

# ------------------------------------------------------------
# 6. RAW TIME AXIS INSPECTION
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("TIME AXIS COMPATIBILITY")
print("=" * 80)

p5_timestamps = (
    pd.to_numeric(
        meta["timestamp"],
        errors="coerce"
    )
    .to_numpy(
        dtype=np.float64
    )
)

print(
    "Phase-5 timestamp range:",
    np.nanmin(p5_timestamps),
    "→",
    np.nanmax(p5_timestamps)
)

print(
    "MNE raw time range:",
    raw.times[0],
    "→",
    raw.times[-1]
)

# ------------------------------------------------------------
# 7. TEST MULTIPLE POSSIBLE TIME ORIGINS
#
# The metadata timestamps may correspond to:
#   A) raw.times directly
#   B) raw.first_samp + time
#   C) another recording-origin offset
#
# First inspect direct compatibility.
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("DIRECT TIMESTAMP → RAW SAMPLE TEST")
print("=" * 80)

sfreq = float(
    raw.info["sfreq"]
)

direct_samples = np.rint(
    p5_timestamps * sfreq
).astype(
    np.int64
)

valid_direct = (
    (direct_samples >= 0)
    &
    (direct_samples < raw.n_times)
)

print(
    "Directly valid timestamps:",
    int(valid_direct.sum()),
    "/",
    len(valid_direct),
    f"({100 * valid_direct.mean():.2f}%)"
)

# ------------------------------------------------------------
# 8. INSPECT FIRST EVENTS
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("FIRST 20 EVENT POSITIONS")
print("=" * 80)

inspect_n = min(
    20,
    len(meta)
)

for i in range(inspect_n):

    t = p5_timestamps[i]

    sample = int(
        direct_samples[i]
    )

    valid = bool(
        valid_direct[i]
    )

    print(
        f"{i:4d} | "
        f"timestamp={t:9.3f} | "
        f"sample={sample:7d} | "
        f"valid={valid} | "
        f"target={meta.iloc[i]['target']}"
    )

# ------------------------------------------------------------
# 9. RECONSTRUCT USING DIRECT TIME
#
# Existing Phase-5 geometry is expected to be:
#   -0.2 s → +0.3 s
#   25 samples at 50 Hz
#
# With 50 Hz:
#   0.5 s × 50 = 25 samples
#
# We use:
#   start = event_sample - 10
#   stop  = start + 25
# ------------------------------------------------------------

pre_samples = int(
    round(
        abs(BASELINE_TMIN)
        * sfreq
    )
)

window_samples = int(
    round(
        (
            BASELINE_TMAX
            -
            BASELINE_TMIN
        )
        * sfreq
    )
)

print("\n" + "=" * 80)
print("WINDOW GEOMETRY")
print("=" * 80)

print(
    "Sampling rate:",
    sfreq
)

print(
    "TMIN:",
    BASELINE_TMIN
)

print(
    "TMAX:",
    BASELINE_TMAX
)

print(
    "Pre-event samples:",
    pre_samples
)

print(
    "Expected window samples:",
    window_samples
)

if window_samples != EXPECTED_SAMPLES:

    raise RuntimeError(
        f"Expected {EXPECTED_SAMPLES} samples, "
        f"calculated {window_samples}."
    )

# ------------------------------------------------------------
# 10. RECONSTRUCT VALID DIRECT WINDOWS
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("RECONSTRUCTING DIRECT-TIME WINDOWS")
print("=" * 80)

reconstructed = []

reconstructed_indices = []

for i, event_sample in enumerate(
    direct_samples
):

    start = (
        event_sample
        -
        pre_samples
    )

    stop = (
        start
        +
        window_samples
    )

    if (
        start < 0
        or stop > raw.n_times
    ):
        continue

    window = raw.get_data(
        start=start,
        stop=stop
    )

    if window.shape != (
        EXPECTED_CHANNELS,
        EXPECTED_SAMPLES
    ):
        continue

    reconstructed.append(
        window
    )

    reconstructed_indices.append(
        i
    )

print(
    "Reconstructed:",
    len(reconstructed),
    "/",
    len(meta)
)

# ------------------------------------------------------------
# 11. COMPARE NUMERICALLY AGAINST PHASE 5
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("NUMERICAL REPRODUCTION TEST")
print("=" * 80)

if len(reconstructed) == 0:

    print(
        "No valid windows were reconstructed."
    )

    print(
        "Direct timestamp interpretation is invalid."
    )

else:

    reconstructed = np.stack(
        reconstructed,
        axis=0
    )

    reconstructed_indices = np.asarray(
        reconstructed_indices
    )

    saved = np.asarray(
        phase5_windows[
            reconstructed_indices
        ]
    )

    print(
        "Reconstructed shape:",
        reconstructed.shape
    )

    print(
        "Saved Phase-5 shape:",
        saved.shape
    )

    # Element-wise difference.
    diff = (
        reconstructed
        -
        saved
    )

    abs_diff = np.abs(
        diff
    )

    mse = np.mean(
        diff ** 2
    )

    mae = np.mean(
        abs_diff
    )

    max_abs_error = np.max(
        abs_diff
    )

    median_abs_error = np.median(
        abs_diff
    )

    print(
        "MSE:",
        mse
    )

    print(
        "MAE:",
        mae
    )

    print(
        "Median absolute error:",
        median_abs_error
    )

    print(
        "Maximum absolute error:",
        max_abs_error
    )

    # Per-window MAE.
    window_mae = np.mean(
        abs_diff,
        axis=(1, 2)
    )

    print(
        "\nPer-window MAE:"
    )

    print(
        pd.Series(
            window_mae
        ).describe().to_string()
    )

    # Correlation on flattened data.
    x = reconstructed.reshape(-1)
    y = saved.reshape(-1)

    if (
        np.std(x) > 0
        and np.std(y) > 0
    ):
        corr = np.corrcoef(
            x,
            y
        )[0, 1]
    else:
        corr = np.nan

    print(
        "\nFlattened correlation:",
        corr
    )

    # Exact-ish equality checks.
    print(
        "\nallclose:",
        np.allclose(
            reconstructed,
            saved,
            rtol=1e-4,
            atol=1e-7
        )
    )

    print(
        "allclose (1e-3):",
        np.allclose(
            reconstructed,
            saved,
            rtol=1e-3,
            atol=1e-6
        )
    )

    # --------------------------------------------------------
    # 12. BEST SAMPLE OFFSET SEARCH
    #
    # If direct reconstruction is not exact, test small
    # integer offsets around the event timestamp.
    # --------------------------------------------------------

    print("\n" + "=" * 80)
    print("LOCAL SAMPLE-OFFSET SEARCH")
    print("=" * 80)

    test_indices = reconstructed_indices[
        :min(
            100,
            len(reconstructed_indices)
        )
    ]

    offsets = range(
        -25,
        26
    )

    offset_results = []

    for offset in offsets:

        errors = []

        for idx in test_indices:

            event_sample = int(
                direct_samples[idx]
            )

            start = (
                event_sample
                -
                pre_samples
                +
                offset
            )

            stop = (
                start
                +
                window_samples
            )

            if (
                start < 0
                or stop > raw.n_times
            ):
                continue

            candidate = raw.get_data(
                start=start,
                stop=stop
            )

            reference = np.asarray(
                phase5_windows[idx]
            )

            if candidate.shape != reference.shape:
                continue

            errors.append(
                np.mean(
                    np.abs(
                        candidate
                        -
                        reference
                    )
                )
            )

        if errors:

            offset_results.append({
                "sample_offset": offset,
                "mean_mae": float(
                    np.mean(errors)
                ),
                "median_mae": float(
                    np.median(errors)
                )
            })

    offset_results = pd.DataFrame(
        offset_results
    )

    if len(offset_results):

        offset_results = (
            offset_results
            .sort_values(
                "mean_mae"
            )
        )

        print(
            offset_results.head(15)
            .to_string(
                index=False
            )
        )

        best_local = (
            offset_results.iloc[0]
        )

        print(
            "\nBEST LOCAL SAMPLE OFFSET:",
            int(
                best_local[
                    "sample_offset"
                ]
            )
        )

        print(
            "Best mean MAE:",
            best_local[
                "mean_mae"
            ]
        )

        offset_results.to_csv(
            f"{OUTPUT_DIR}/"
            f"{RID}_sample_offset_search.csv",
            index=False
        )

# ------------------------------------------------------------
# 13. SAVE BASIC RAW-EEG VERIFICATION
# ------------------------------------------------------------

verification = {
    "recording_id": RID,
    "raw_path": RAW_PATH,
    "raw_original_channels": int(
        len(
            mne.io.read_raw_brainvision(
                RAW_PATH,
                preload=False,
                verbose=False
            ).ch_names
        )
    ),
    "preprocessed_channels": int(
        len(raw.ch_names)
    ),
    "preprocessed_sfreq": float(
        raw.info["sfreq"]
    ),
    "preprocessed_n_times": int(
        raw.n_times
    ),
    "preprocessed_duration": float(
        raw.times[-1]
    ),
    "phase5_windows": int(
        len(phase5_windows)
    ),
    "phase5_timestamp_min": float(
        np.nanmin(p5_timestamps)
    ),
    "phase5_timestamp_max": float(
        np.nanmax(p5_timestamps)
    ),
    "direct_valid_events": int(
        valid_direct.sum()
    ),
    "direct_valid_rate": float(
        valid_direct.mean()
    )
}

verification_path = (
    f"{OUTPUT_DIR}/"
    f"{RID}_raw_eeg_verification.json"
)

with open(
    verification_path,
    "w"
) as f:

    json.dump(
        verification,
        f,
        indent=2
    )

print("\n" + "=" * 80)
print("VERIFICATION FILE")
print("=" * 80)

print(
    verification_path
)

print("\n" + "=" * 80)
print("DECISION")
print("=" * 80)

print(
    "Do NOT generate the larger temporal windows yet."
)

print(
    "The critical values are:"
)

print(
    "1. Directly valid events"
)

print(
    "2. MSE / MAE"
)

print(
    "3. Flattened correlation"
)

print(
    "4. allclose"
)

print(
    "5. BEST LOCAL SAMPLE OFFSET"
)
```

```python
# ============================================================
# NEUROSENTINEL — DIAGNOSTIC STEP 6B
# Temporal Window Expansion — Generate 4 Window Variants
#
# Variants:
#   baseline : [-0.2, +0.3] -> 25 samples
#   W1       : [-0.4, +0.4] -> 40 samples
#   W2       : [-0.5, +0.5] -> 50 samples
#   W3       : [-0.2, +0.5] -> 35 samples
#
# Source:
#   Original BrainVision EEG
#   Phase-5 timestamps / metadata
#
# Preserved:
#   - Phase-6 frozen subject split
#   - Phase-5 event ordering
#   - Phase-2 preprocessing
#   - 61 EEG channels
#   - 50 Hz sampling
#   - exact Phase-5 baseline geometry
#
# ============================================================

import os
import gc
import json
import time
import traceback
from pathlib import Path

import numpy as np
import pandas as pd
import mne


# ============================================================
# 1. CONFIGURATION
# ============================================================

ROOT = Path("/content/gdrive/MyDrive/NeuroSentinel")

PHASE6_DIR = ROOT / "07_phase6_subject_split"
PHASE5_DIR = ROOT / "06_phase5_utility_labels" / "recordings"

RAW_EEG_DIR = ROOT / "data" / "SpanishBCBL" / "EEG" / "EEG"

OUTPUT_ROOT = ROOT / "08_phase6_temporal_windows"
CHECKPOINT_DIR = ROOT / "checkpoints" / "phase6_temporal_windows"

OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)
CHECKPOINT_DIR.mkdir(parents=True, exist_ok=True)

PHASE6_MANIFEST = (
    PHASE6_DIR / "phase6_subject_split_manifest_resolved.csv"
)

if not PHASE6_MANIFEST.exists():
    PHASE6_MANIFEST = (
        PHASE6_DIR / "phase6_subject_split_manifest.csv"
    )

CHECKPOINT_FILE = (
    CHECKPOINT_DIR / "phase6_temporal_windows_checkpoint.json"
)

SUMMARY_FILE = (
    OUTPUT_ROOT / "phase6_temporal_windows_summary.json"
)

# ------------------------------------------------------------
# Exact preprocessing used for Phase 5 reconstruction
# ------------------------------------------------------------

TARGET_SFREQ = 50.0
LOW_FREQ = 0.1
HIGH_FREQ = 20.0

# Expected EEG channels after EOG removal
EXPECTED_EEG_CHANNELS = 61

# ------------------------------------------------------------
# Temporal variants
# ------------------------------------------------------------

WINDOWS = {
    "baseline": {
        "tmin": -0.2,
        "tmax": 0.3,
        "n_samples": 25,
    },
    "W1": {
        "tmin": -0.4,
        "tmax": 0.4,
        "n_samples": 40,
    },
    "W2": {
        "tmin": -0.5,
        "tmax": 0.5,
        "n_samples": 50,
    },
    "W3": {
        "tmin": -0.2,
        "tmax": 0.5,
        "n_samples": 35,
    },
}

# ------------------------------------------------------------
# Memory / numerical configuration
# ------------------------------------------------------------

SAVE_DTYPE = np.float64

# Boundary tolerance:
# events outside the valid larger-window range are skipped.
BOUNDARY_TOL = 1e-9

# ============================================================
# 2. LOAD PHASE-6 MANIFEST
# ============================================================

print("=" * 70)
print("NEUROSENTINEL — STEP 6B")
print("Temporal Window Expansion")
print("=" * 70)

if not PHASE6_MANIFEST.exists():
    raise FileNotFoundError(
        f"Phase-6 manifest not found:\n{PHASE6_MANIFEST}"
    )

phase6 = pd.read_csv(PHASE6_MANIFEST)

print(f"\nPhase-6 manifest:")
print(f"  {PHASE6_MANIFEST}")
print(f"  rows = {len(phase6)}")
print(f"  columns = {list(phase6.columns)}")

required_phase6_cols = [
    "recording_id",
    "split",
    "subject",
]

missing = [
    c for c in required_phase6_cols
    if c not in phase6.columns
]

if missing:
    raise ValueError(
        f"Missing required Phase-6 columns: {missing}"
    )

phase6 = phase6.copy()

phase6["recording_id"] = (
    phase6["recording_id"]
    .astype(str)
    .str.strip()
)

phase6["split"] = (
    phase6["split"]
    .astype(str)
    .str.strip()
)

# Ensure exactly one row per recording
if phase6["recording_id"].duplicated().any():
    dupes = phase6.loc[
        phase6["recording_id"].duplicated(keep=False),
        "recording_id"
    ].unique()

    raise ValueError(
        f"Duplicate recording IDs in Phase-6 manifest: {dupes}"
    )

print("\nFrozen split:")
print(
    phase6.groupby("split")["recording_id"]
    .count()
    .to_string()
)

print(f"\nTotal recordings: {len(phase6)}")


# ============================================================
# 3. LOAD / CREATE CHECKPOINT
# ============================================================

if CHECKPOINT_FILE.exists():
    with open(CHECKPOINT_FILE, "r") as f:
        checkpoint = json.load(f)

    print("\nExisting checkpoint found:")
    print(f"  {CHECKPOINT_FILE}")
else:
    checkpoint = {
        "status": "RUNNING",
        "started_at": time.strftime("%Y-%m-%d %H:%M:%S"),
        "completed_recordings": [],
        "failed_recordings": [],
        "recording_results": {},
    }

completed_recordings = set(
    checkpoint.get("completed_recordings", [])
)

failed_recordings = set(
    checkpoint.get("failed_recordings", [])
)

print(
    f"  completed = {len(completed_recordings)}"
)
print(
    f"  failed = {len(failed_recordings)}"
)


def save_checkpoint():
    checkpoint["last_update"] = time.strftime(
        "%Y-%m-%d %H:%M:%S"
    )

    with open(CHECKPOINT_FILE, "w") as f:
        json.dump(
            checkpoint,
            f,
            indent=2,
        )


# ============================================================
# 4. RAW FILE RESOLUTION
# ============================================================

def resolve_raw_vhdr(row):
    """
    Resolve BrainVision .vhdr path from Phase-6 manifest.

    Prefer an explicit raw path if available.
    Otherwise derive from recording_id.
    """

    # --------------------------------------------------------
    # Case 1: manifest contains raw path
    # --------------------------------------------------------

    possible_columns = [
        "raw_eeg_path",
        "eeg_path",
        "vhdr_path",
        "raw_path",
    ]

    for col in possible_columns:
        if col in row.index:
            value = row[col]

            if pd.notna(value):
                value = str(value).strip()

                if value:
                    p = Path(value)

                    # Correct old /content/drive path
                    if str(p).startswith("/content/drive/"):
                        p = Path(
                            str(p).replace(
                                "/content/drive/",
                                "/content/gdrive/",
                                1,
                            )
                        )

                    if p.exists():
                        return p

    # --------------------------------------------------------
    # Case 2: search by recording ID
    # --------------------------------------------------------

    recording_id = str(row["recording_id"])

    matches = list(
        RAW_EEG_DIR.glob(
            f"*{recording_id.split('_')[0]}*"
        )
    )

    vhdr_matches = [
        p for p in matches
        if p.suffix.lower() == ".vhdr"
    ]

    # More robust search using recording prefix
    if not vhdr_matches:
        subject = str(row.get("subject", "")).strip()

        if subject:
            vhdr_matches = list(
                RAW_EEG_DIR.glob(
                    f"*{recording_id[:3]}*.vhdr"
                )
            )

    if len(vhdr_matches) == 1:
        return vhdr_matches[0]

    # --------------------------------------------------------
    # Fallback: search recursively
    # --------------------------------------------------------

    recursive = list(
        RAW_EEG_DIR.rglob("*.vhdr")
    )

    candidates = [
        p for p in recursive
        if recording_id[:3] in p.name
    ]

    if len(candidates) == 1:
        return candidates[0]

    raise FileNotFoundError(
        f"Could not uniquely resolve raw BrainVision file "
        f"for recording {recording_id}.\n"
        f"Candidates:\n{candidates}"
    )


# ============================================================
# 5. LOAD PHASE-5 METADATA
# ============================================================

def find_phase5_metadata(recording_id):
    path = (
        PHASE5_DIR
        / f"{recording_id}_metadata.csv"
    )

    if not path.exists():
        raise FileNotFoundError(
            f"Phase-5 metadata missing:\n{path}"
        )

    return path


def find_phase5_windows(recording_id):
    path = (
        PHASE5_DIR
        / f"{recording_id}_windows.npy"
    )

    if not path.exists():
        raise FileNotFoundError(
            f"Phase-5 windows missing:\n{path}"
        )

    return path


# ============================================================
# 6. PREPROCESS RAW EEG
# ============================================================

def preprocess_raw(vhdr_path):
    """
    Exact preprocessing required to reconstruct Phase-5 windows.

    Original:
        64 channels
        1000 Hz

    Processing:
        remove EOG
        0.1–20 Hz filter
        resample to 50 Hz

    Result:
        61 EEG channels
        50 Hz
    """

    print(
        f"    Loading BrainVision: {vhdr_path.name}"
    )

    raw = mne.io.read_raw_brainvision(
        str(vhdr_path),
        preload=True,
        verbose="ERROR",
    )

    original_n_channels = len(raw.ch_names)
    original_sfreq = float(raw.info["sfreq"])

    # --------------------------------------------------------
    # Detect EOG channels
    # --------------------------------------------------------

    eog_names = []

    for ch in raw.ch_names:
        ch_lower = ch.lower()

        if (
            "eog" in ch_lower
            or ch_lower in {
                "rveog",
                "lheog",
                "rheog",
            }
        ):
            eog_names.append(ch)

    # Expected SpanishBCBL EOG channels
    if len(eog_names) != 3:
        print(
            f"    WARNING: detected EOG channels = "
            f"{eog_names}"
        )

    if eog_names:
        raw.drop_channels(
            eog_names
        )

    # --------------------------------------------------------
    # Apply filter
    # --------------------------------------------------------

    print(
        f"    Filtering {LOW_FREQ}–{HIGH_FREQ} Hz..."
    )

    raw.filter(
        l_freq=LOW_FREQ,
        h_freq=HIGH_FREQ,
        picks="eeg",
        verbose="ERROR",
    )

    # --------------------------------------------------------
    # Resample
    # --------------------------------------------------------

    print(
        f"    Resampling "
        f"{original_sfreq:g} Hz → "
        f"{TARGET_SFREQ:g} Hz..."
    )

    raw.resample(
        TARGET_SFREQ,
        npad="auto",
        verbose="ERROR",
    )

    # --------------------------------------------------------
    # Validate channel count
    # --------------------------------------------------------

    eeg_picks = mne.pick_types(
        raw.info,
        eeg=True,
        exclude=[],
    )

    if len(eeg_picks) != EXPECTED_EEG_CHANNELS:
        raise ValueError(
            f"Expected {EXPECTED_EEG_CHANNELS} EEG channels "
            f"after EOG removal, got {len(eeg_picks)}.\n"
            f"Channels: {raw.ch_names}"
        )

    data = raw.get_data(
        picks=eeg_picks
    )

    sfreq = float(raw.info["sfreq"])

    if abs(sfreq - TARGET_SFREQ) > 1e-6:
        raise ValueError(
            f"Unexpected sampling rate: {sfreq}"
        )

    print(
        f"    Preprocessed shape: {data.shape}"
    )

    print(
        f"    Duration: "
        f"{data.shape[1] / sfreq:.3f} s"
    )

    return raw, data


# ============================================================
# 7. TIMESTAMP → SAMPLE INDEX
# ============================================================

def timestamp_to_sample(
    timestamps,
    sfreq=TARGET_SFREQ,
):
    """
    Same mapping validated in the successful
    003_S2_task1 reconstruction:

        event_sample = round(timestamp * 50)
    """

    timestamps = np.asarray(
        timestamps,
        dtype=np.float64,
    )

    return np.rint(
        timestamps * sfreq
    ).astype(np.int64)


# ============================================================
# 8. EXTRACT WINDOWS
# ============================================================

def extract_windows(
    eeg_data,
    timestamps,
    tmin,
    tmax,
    sfreq=TARGET_SFREQ,
):
    """
    Extract windows using the same direct timestamp mapping
    validated against Phase-5.

    Output:
        (N, 61, samples)
    """

    timestamps = np.asarray(
        timestamps,
        dtype=np.float64,
    )

    event_samples = timestamp_to_sample(
        timestamps,
        sfreq,
    )

    # --------------------------------------------------------
    # Exact sample geometry
    #
    # Phase-5:
    # [-0.2,+0.3] -> 25 samples
    #
    # General:
    # start = event_sample + round(tmin * sfreq)
    # n     = round((tmax-tmin)*sfreq)
    # --------------------------------------------------------

    pre_samples = int(
        round(abs(tmin) * sfreq)
    )

    n_samples = int(
        round((tmax - tmin) * sfreq)
    )

    starts = (
        event_samples
        + int(round(tmin * sfreq))
    )

    stops = starts + n_samples

    n_total = eeg_data.shape[1]

    valid_mask = (
        (starts >= 0)
        & (stops <= n_total)
    )

    valid_indices = np.flatnonzero(
        valid_mask
    )

    windows = np.empty(
        (
            len(valid_indices),
            eeg_data.shape[0],
            n_samples,
        ),
        dtype=SAVE_DTYPE,
    )

    for j, original_idx in enumerate(
        valid_indices
    ):
        start = int(
            starts[original_idx]
        )
        stop = int(
            stops[original_idx]
        )

        windows[j] = (
            eeg_data[:, start:stop]
        )

    return (
        windows,
        valid_indices,
        event_samples,
        starts,
        stops,
    )


# ============================================================
# 9. BASELINE VERIFICATION
# ============================================================

def verify_baseline(
    recording_id,
    baseline_reconstructed,
):
    """
    Compare reconstructed [-0.2,+0.3] windows against
    the already saved Phase-5 windows.

    This is a critical integrity check.
    """

    phase5_path = find_phase5_windows(
        recording_id
    )

    phase5_windows = np.load(
        phase5_path,
        mmap_mode="r",
    )

    print(
        f"    Phase-5 baseline shape: "
        f"{phase5_windows.shape}"
    )

    print(
        f"    Reconstructed baseline shape: "
        f"{baseline_reconstructed.shape}"
    )

    if (
        phase5_windows.shape
        != baseline_reconstructed.shape
    ):
        raise ValueError(
            f"Baseline shape mismatch for "
            f"{recording_id}: "
            f"{phase5_windows.shape} vs "
            f"{baseline_reconstructed.shape}"
        )

    # Exact numerical comparison
    max_abs_error = float(
        np.max(
            np.abs(
                np.asarray(
                    phase5_windows
                )
                - baseline_reconstructed
            )
        )
    )

    mse = float(
        np.mean(
            (
                np.asarray(phase5_windows)
                - baseline_reconstructed
            ) ** 2
        )
    )

    allclose = bool(
        np.allclose(
            phase5_windows,
            baseline_reconstructed,
            rtol=0,
            atol=0,
        )
    )

    print(
        f"    Baseline max abs error: "
        f"{max_abs_error:.12g}"
    )

    print(
        f"    Baseline MSE: "
        f"{mse:.12g}"
    )

    print(
        f"    Baseline exact allclose: "
        f"{allclose}"
    )

    if not allclose:
        raise ValueError(
            f"Baseline reconstruction FAILED "
            f"for {recording_id}"
        )

    return {
        "phase5_shape": list(
            phase5_windows.shape
        ),
        "reconstructed_shape": list(
            baseline_reconstructed.shape
        ),
        "max_abs_error": max_abs_error,
        "mse": mse,
        "exact_allclose": allclose,
    }


# ============================================================
# 10. METADATA CREATION
# ============================================================

def build_variant_metadata(
    phase5_meta,
    valid_indices,
    recording_id,
    split,
    variant,
    tmin,
    tmax,
):
    """
    Preserve the exact Phase-5 row order for valid windows.
    """

    meta = phase5_meta.iloc[
        valid_indices
    ].copy()

    meta = meta.reset_index(
        drop=True
    )

    # Preserve original window identity
    meta["source_phase5_row"] = (
        valid_indices.astype(np.int64)
    )

    meta["temporal_variant"] = variant
    meta["window_tmin"] = float(tmin)
    meta["window_tmax"] = float(tmax)

    meta["split"] = split

    # Reassign variant-local window index
    meta["window_index"] = np.arange(
        len(meta),
        dtype=np.int64,
    )

    return meta


# ============================================================
# 11. PROCESS ONE RECORDING
# ============================================================

def process_recording(row):
    recording_id = str(
        row["recording_id"]
    )

    split = str(
        row["split"]
    )

    print("\n" + "-" * 70)
    print(
        f"PROCESSING {recording_id} | split={split}"
    )
    print("-" * 70)

    # --------------------------------------------------------
    # Resolve files
    # --------------------------------------------------------

    vhdr_path = resolve_raw_vhdr(row)
    metadata_path = find_phase5_metadata(
        recording_id
    )

    print(
        f"  Raw EEG: {vhdr_path}"
    )

    print(
        f"  Phase-5 metadata: {metadata_path}"
    )

    # --------------------------------------------------------
    # Load Phase-5 metadata
    # --------------------------------------------------------

    phase5_meta = pd.read_csv(
        metadata_path
    )

    if "timestamp" not in phase5_meta.columns:
        raise ValueError(
            f"Phase-5 metadata for "
            f"{recording_id} has no timestamp column."
        )

    timestamps = (
        phase5_meta["timestamp"]
        .to_numpy(dtype=np.float64)
    )

    n_events = len(
        phase5_meta
    )

    print(
        f"  Phase-5 events: {n_events}"
    )

    # --------------------------------------------------------
    # Preprocess raw EEG
    # --------------------------------------------------------

    raw, eeg_data = preprocess_raw(
        vhdr_path
    )

    # --------------------------------------------------------
    # Output directories
    # --------------------------------------------------------

    recording_output = (
        OUTPUT_ROOT
        / split
        / recording_id
    )

    recording_output.mkdir(
        parents=True,
        exist_ok=True,
    )

    variant_results = {}

    # --------------------------------------------------------
    # Generate all variants
    # --------------------------------------------------------

    for variant_name, cfg in WINDOWS.items():

        tmin = cfg["tmin"]
        tmax = cfg["tmax"]
        expected_samples = cfg[
            "n_samples"
        ]

        print(
            f"\n  Variant {variant_name}: "
            f"[{tmin}, {tmax}]"
        )

        (
            windows,
            valid_indices,
            event_samples,
            starts,
            stops,
        ) = extract_windows(
            eeg_data=eeg_data,
            timestamps=timestamps,
            tmin=tmin,
            tmax=tmax,
            sfreq=TARGET_SFREQ,
        )

        # ----------------------------------------------------
        # Validate geometry
        # ----------------------------------------------------

        if windows.ndim != 3:
            raise ValueError(
                f"Unexpected window ndim: "
                f"{windows.ndim}"
            )

        if windows.shape[1] != EXPECTED_EEG_CHANNELS:
            raise ValueError(
                f"Expected 61 channels, got "
                f"{windows.shape[1]}"
            )

        if windows.shape[2] != expected_samples:
            raise ValueError(
                f"Expected {expected_samples} samples "
                f"for {variant_name}, got "
                f"{windows.shape[2]}"
            )

        # ----------------------------------------------------
        # Boundary handling
        # ----------------------------------------------------

        skipped = (
            n_events
            - len(valid_indices)
        )

        if skipped > 0:
            print(
                f"    Boundary-skipped events: "
                f"{skipped}"
            )

        # ----------------------------------------------------
        # Baseline exact verification
        # ----------------------------------------------------

        baseline_verification = None

        if variant_name == "baseline":

            baseline_verification = (
                verify_baseline(
                    recording_id,
                    windows,
                )
            )

        # ----------------------------------------------------
        # Metadata
        # ----------------------------------------------------

        variant_meta = build_variant_metadata(
            phase5_meta=phase5_meta,
            valid_indices=valid_indices,
            recording_id=recording_id,
            split=split,
            variant=variant_name,
            tmin=tmin,
            tmax=tmax,
        )

        # ----------------------------------------------------
        # Verify metadata/window count
        # ----------------------------------------------------

        if len(variant_meta) != len(windows):
            raise ValueError(
                f"Metadata/window mismatch "
                f"for {recording_id} "
                f"{variant_name}: "
                f"{len(variant_meta)} vs "
                f"{len(windows)}"
            )

        # ----------------------------------------------------
        # Save windows
        # ----------------------------------------------------

        windows_path = (
            recording_output
            / f"{recording_id}_{variant_name}_windows.npy"
        )

        metadata_path_out = (
            recording_output
            / f"{recording_id}_{variant_name}_metadata.csv"
        )

        np.save(
            windows_path,
            windows,
        )

        variant_meta.to_csv(
            metadata_path_out,
            index=False,
        )

        # ----------------------------------------------------
        # Statistics
        # ----------------------------------------------------

        result = {
            "variant": variant_name,
            "tmin": float(tmin),
            "tmax": float(tmax),
            "expected_samples": int(
                expected_samples
            ),
            "actual_shape": list(
                windows.shape
            ),
            "total_phase5_events": int(
                n_events
            ),
            "valid_events": int(
                len(valid_indices)
            ),
            "skipped_boundary_events": int(
                skipped
            ),
            "windows_path": str(
                windows_path
            ),
            "metadata_path": str(
                metadata_path_out
            ),
        }

        if baseline_verification is not None:
            result[
                "baseline_verification"
            ] = baseline_verification

        variant_results[
            variant_name
        ] = result

        print(
            f"    Saved: {windows_path.name}"
        )

        print(
            f"    Shape: {windows.shape}"
        )

        # ----------------------------------------------------
        # Cleanup variant
        # ----------------------------------------------------

        del windows
        del variant_meta

        gc.collect()

    # --------------------------------------------------------
    # Recording-level result
    # --------------------------------------------------------

    recording_result = {
        "recording_id": recording_id,
        "split": split,
        "subject": str(
            row["subject"]
        ),
        "raw_vhdr": str(
            vhdr_path
        ),
        "phase5_metadata": str(
            metadata_path
        ),
        "n_phase5_events": int(
            n_events
        ),
        "preprocessed_shape": list(
            eeg_data.shape
        ),
        "sampling_rate": TARGET_SFREQ,
        "n_eeg_channels": int(
            eeg_data.shape[0]
        ),
        "variants": variant_results,
    }

    # --------------------------------------------------------
    # Save recording summary
    # --------------------------------------------------------

    recording_summary_path = (
        recording_output
        / f"{recording_id}_summary.json"
    )

    with open(
        recording_summary_path,
        "w",
    ) as f:
        json.dump(
            recording_result,
            f,
            indent=2,
        )

    # --------------------------------------------------------
    # Cleanup
    # --------------------------------------------------------

    del raw
    del eeg_data
    gc.collect()

    return recording_result


# ============================================================
# 12. MAIN LOOP
# ============================================================

all_start = time.time()

print("\nStarting temporal-window generation...")
print(
    f"Total recordings: {len(phase6)}"
)
print(
    f"Already completed: "
    f"{len(completed_recordings)}"
)

for idx, row in phase6.iterrows():

    recording_id = str(
        row["recording_id"]
    )

    # --------------------------------------------------------
    # Resume support
    # --------------------------------------------------------

    if recording_id in completed_recordings:
        print(
            f"\n[{idx+1}/{len(phase6)}] "
            f"SKIP completed: {recording_id}"
        )
        continue

    # --------------------------------------------------------
    # Process
    # --------------------------------------------------------

    started = time.time()

    try:

        result = process_recording(
            row
        )

        elapsed = (
            time.time() - started
        )

        result["elapsed_seconds"] = (
            elapsed
        )

        checkpoint[
            "recording_results"
        ][recording_id] = result

        completed_recordings.add(
            recording_id
        )

        checkpoint[
            "completed_recordings"
        ] = sorted(
            completed_recordings
        )

        # Remove from failed if rerun succeeds
        failed_recordings.discard(
            recording_id
        )

        checkpoint[
            "failed_recordings"
        ] = sorted(
            failed_recordings
        )

        save_checkpoint()

        print(
            f"\nSUCCESS: {recording_id}"
        )

        print(
            f"Elapsed: {elapsed/60:.2f} min"
        )

    except Exception as e:

        elapsed = (
            time.time() - started
        )

        print(
            f"\nFAILED: {recording_id}"
        )

        print(
            traceback.format_exc()
        )

        failed_recordings.add(
            recording_id
        )

        checkpoint[
            "failed_recordings"
        ] = sorted(
            failed_recordings
        )

        checkpoint[
            "recording_results"
        ][recording_id] = {
            "recording_id": recording_id,
            "split": str(
                row["split"]
            ),
            "status": "FAILED",
            "error": str(e),
            "elapsed_seconds": elapsed,
        }

        save_checkpoint()

        # Continue with remaining recordings
        continue


# ============================================================
# 13. FINAL VALIDATION
# ============================================================

total_elapsed = (
    time.time() - all_start
)

completed = sorted(
    completed_recordings
)

failed = sorted(
    failed_recordings
)

print("\n" + "=" * 70)
print("TEMPORAL WINDOW GENERATION COMPLETE")
print("=" * 70)

print(
    f"\nCompleted: "
    f"{len(completed)}/{len(phase6)}"
)

print(
    f"Failed: "
    f"{len(failed)}"
)

if failed:
    print("\nFailed recordings:")
    for rid in failed:
        print(f"  - {rid}")


# ============================================================
# 14. DATASET-LEVEL SUMMARY
# ============================================================

summary = {
    "status": (
        "COMPLETE"
        if len(completed) == len(phase6)
        else "INCOMPLETE"
    ),
    "created_at": time.strftime(
        "%Y-%m-%d %H:%M:%S"
    ),
    "total_elapsed_seconds": (
        total_elapsed
    ),
    "total_recordings": int(
        len(phase6)
    ),
    "completed_recordings": int(
        len(completed)
    ),
    "failed_recordings": int(
        len(failed)
    ),
    "phase6_manifest": str(
        PHASE6_MANIFEST
    ),
    "raw_eeg_directory": str(
        RAW_EEG_DIR
    ),
    "output_root": str(
        OUTPUT_ROOT
    ),
    "sampling_rate": TARGET_SFREQ,
    "n_eeg_channels": EXPECTED_EEG_CHANNELS,
    "preprocessing": {
        "eog_removed": True,
        "filter_low_hz": LOW_FREQ,
        "filter_high_hz": HIGH_FREQ,
        "resample_hz": TARGET_SFREQ,
    },
    "windows": WINDOWS,
    "completed_recordings_list": completed,
    "failed_recordings_list": failed,
}


# ------------------------------------------------------------
# Split-level counts
# ------------------------------------------------------------

summary["phase6_split_counts"] = (
    phase6.groupby("split")[
        "recording_id"
    ]
    .count()
    .to_dict()
)


# ------------------------------------------------------------
# Variant-level totals
# ------------------------------------------------------------

variant_totals = {}

for variant_name in WINDOWS:

    total_windows = 0
    total_skipped = 0
    shapes = []

    for rid in completed:

        result = checkpoint[
            "recording_results"
        ].get(rid)

        if not result:
            continue

        variant_result = result.get(
            "variants",
            {}
        ).get(
            variant_name
        )

        if not variant_result:
            continue

        total_windows += int(
            variant_result[
                "valid_events"
            ]
        )

        total_skipped += int(
            variant_result[
                "skipped_boundary_events"
            ]
        )

        shapes.append(
            variant_result[
                "actual_shape"
            ]
        )

    variant_totals[
        variant_name
    ] = {
        "total_windows": int(
            total_windows
        ),
        "total_boundary_skipped": int(
            total_skipped
        ),
        "expected_samples": int(
            WINDOWS[
                variant_name
            ]["n_samples"]
        ),
        "unique_shapes": shapes[:10],
    }

summary[
    "variant_totals"
] = variant_totals


# ============================================================
# 15. SAVE FINAL SUMMARY
# ============================================================

with open(
    SUMMARY_FILE,
    "w",
) as f:
    json.dump(
        summary,
        f,
        indent=2,
    )

# Update checkpoint status
checkpoint["status"] = summary[
    "status"
]

checkpoint[
    "finished_at"
] = time.strftime(
    "%Y-%m-%d %H:%M:%S"
)

save_checkpoint()


# ============================================================
# 16. FINAL REPORT
# ============================================================

print(
    f"\nSummary saved:"
)
print(
    f"  {SUMMARY_FILE}"
)

print(
    f"\nCheckpoint saved:"
)
print(
    f"  {CHECKPOINT_FILE}"
)

print("\nVariant totals:")

for variant_name, info in (
    variant_totals.items()
):

    print(
        f"\n  {variant_name}"
    )

    print(
        f"    samples/window: "
        f"{info['expected_samples']}"
    )

    print(
        f"    total windows: "
        f"{info['total_windows']}"
    )

    print(
        f"    boundary skipped: "
        f"{info['total_boundary_skipped']}"
    )


# ============================================================
# 17. CRITICAL INTEGRITY ASSERTIONS
# ============================================================

print("\nRunning final integrity checks...")

# ------------------------------------------------------------
# A. All recordings processed
# ------------------------------------------------------------

assert (
    len(completed) + len(failed)
    == len(phase6)
), (
    "Checkpoint accounting mismatch."
)

# ------------------------------------------------------------
# B. If complete, all variants must exist
# ------------------------------------------------------------

if len(failed) == 0:

    assert (
        len(completed)
        == len(phase6)
    ), (
        "Not all Phase-6 recordings completed."
    )

    for rid in completed:

        result = checkpoint[
            "recording_results"
        ][rid]

        assert (
            set(
                result["variants"].keys()
            )
            == set(WINDOWS.keys())
        ), (
            f"Missing temporal variant "
            f"for {rid}"
        )

        # ----------------------------------------------------
        # Verify each variant geometry
        # ----------------------------------------------------

        for variant_name, cfg in (
            WINDOWS.items()
        ):

            vr = result[
                "variants"
            ][variant_name]

            shape = vr[
                "actual_shape"
            ]

            assert shape[1] == 61, (
                f"{rid}/{variant_name}: "
                f"channel count != 61"
            )

            assert shape[2] == cfg[
                "n_samples"
            ], (
                f"{rid}/{variant_name}: "
                f"sample count mismatch"
            )

    print(
        "  PASS: all 62 recordings processed"
    )

else:
    print(
        "  WARNING: dataset incomplete because "
        f"{len(failed)} recordings failed."
    )


print(
    "\nFinal status:",
    summary["status"],
)

print(
    "\nOutput root:"
)
print(
    OUTPUT_ROOT
)

print(
    "\nNext step after SUCCESS:"
)
print(
    "Train the identical Diagnostic Step 5 "
    "29-class EEG decoder on each temporal variant "
    "using validation-only model selection."
)
```

```python
# ============================================================
# NEUROSENTINEL — STEP 6B
# Generate temporal-window variants from original BrainVision EEG
#
# Variants:
#   baseline: [-0.2, +0.3] = 25 samples
#   W1:       [-0.4, +0.4] = 40 samples
#   W2:       [-0.5, +0.5] = 50 samples
#   W3:       [-0.2, +0.5] = 35 samples
#
# SOURCE:
#   Original BrainVision recordings
#
# EVENT ALIGNMENT:
#   Phase-5 metadata timestamps
#   event_sample = round(timestamp * 50)
#
# FROZEN:
#   Phase-6 subject split
#   Phase-5 preprocessing
#   Phase-5 window/event indexing
# ============================================================

import os
import json
import time
import traceback
from pathlib import Path

import numpy as np
import pandas as pd
import mne


# ============================================================
# 1. PATHS
# ============================================================

ROOT = Path(
    "/content/gdrive/MyDrive/NeuroSentinel"
)

PHASE6_MANIFEST = (
    ROOT
    / "07_phase6_subject_split"
    / "phase6_subject_split_manifest_resolved.csv"
)

PHASE5_RECORDINGS = (
    ROOT
    / "06_phase5_utility_labels"
    / "recordings"
)

OUTPUT_ROOT = (
    ROOT
    / "08_phase6_temporal_windows"
)

CHECKPOINT_PATH = (
    ROOT
    / "checkpoints"
    / "phase6_temporal_windows_checkpoint.json"
)

SUMMARY_PATH = (
    OUTPUT_ROOT
    / "phase6_temporal_windows_summary.json"
)

OUTPUT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

CHECKPOINT_PATH.parent.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# 2. TEMPORAL WINDOWS
# ============================================================

VARIANTS = {
    "baseline": {
        "tmin": -0.2,
        "tmax": 0.3,
        "samples": 25,
    },
    "W1": {
        "tmin": -0.4,
        "tmax": 0.4,
        "samples": 40,
    },
    "W2": {
        "tmin": -0.5,
        "tmax": 0.5,
        "samples": 50,
    },
    "W3": {
        "tmin": -0.2,
        "tmax": 0.5,
        "samples": 35,
    },
}

FS = 50


# ============================================================
# 3. LOAD FROZEN PHASE-6 MANIFEST
# ============================================================

if not PHASE6_MANIFEST.exists():
    raise FileNotFoundError(
        f"Phase-6 resolved manifest not found:\n"
        f"{PHASE6_MANIFEST}"
    )

phase6 = pd.read_csv(
    PHASE6_MANIFEST
)

required_columns = [
    "recording_id",
    "split",
    "raw_resolved",
]

missing = [
    c
    for c in required_columns
    if c not in phase6.columns
]

if missing:
    raise RuntimeError(
        f"Phase-6 manifest missing columns: {missing}"
    )

phase6["recording_id"] = (
    phase6["recording_id"]
    .astype(str)
)

phase6["split"] = (
    phase6["split"]
    .astype(str)
)

phase6["raw_resolved"] = (
    phase6["raw_resolved"]
    .astype(str)
)


# ============================================================
# 4. VERIFY PHASE-6 MANIFEST
# ============================================================

recordings_df = (
    phase6[
        [
            "recording_id",
            "split",
            "raw_resolved",
        ]
    ]
    .drop_duplicates()
    .sort_values("recording_id")
    .reset_index(drop=True)
)

if len(recordings_df) != 62:
    raise RuntimeError(
        f"Expected 62 recordings, "
        f"found {len(recordings_df)}"
    )

print("=" * 75)
print("NEUROSENTINEL — STEP 6B")
print("=" * 75)

print(
    "\nPhase-6 manifest:",
    PHASE6_MANIFEST
)

print(
    "Recordings:",
    len(recordings_df)
)

print("\nSplit counts:")

print(
    recordings_df
    .groupby("split")["recording_id"]
    .count()
    .to_string()
)


# ============================================================
# 5. RESOLVE RAW EEG PATH
# ============================================================

def resolve_raw_path(path_string):

    path_string = str(
        path_string
    )

    p = Path(path_string)

    # Already valid.
    if p.exists():
        return p

    # Convert old Colab mount.
    if path_string.startswith(
        "/content/drive/"
    ):

        candidate = Path(
            path_string.replace(
                "/content/drive/",
                "/content/gdrive/",
                1
            )
        )

        if candidate.exists():
            return candidate

    # Last-resort filename lookup.
    eeg_root = (
        ROOT
        / "data"
        / "SpanishBCBL"
        / "EEG"
        / "EEG"
    )

    candidate = (
        eeg_root
        / p.name
    )

    if candidate.exists():
        return candidate

    return None


# Verify all raw paths BEFORE processing.

print("\nVerifying raw EEG paths...")

raw_path_map = {}

for _, row in recordings_df.iterrows():

    recording_id = row[
        "recording_id"
    ]

    raw_path = resolve_raw_path(
        row["raw_resolved"]
    )

    if raw_path is None:

        raise FileNotFoundError(
            f"{recording_id}: raw EEG not found:\n"
            f"{row['raw_resolved']}"
        )

    raw_path_map[
        recording_id
    ] = raw_path

print(
    "Raw EEG path verification: PASS"
)

print(
    "Resolved raw recordings:",
    len(raw_path_map)
)


# ============================================================
# 6. LOAD / INITIALIZE CHECKPOINT
# ============================================================

if CHECKPOINT_PATH.exists():

    with open(
        CHECKPOINT_PATH,
        "r"
    ) as f:

        checkpoint = json.load(f)

else:

    checkpoint = {
        "completed": {},
        "failed": {},
    }


checkpoint.setdefault(
    "completed",
    {}
)

checkpoint.setdefault(
    "failed",
    {})


# ============================================================
# 7. PHASE-5 METADATA LOADER
# ============================================================

def load_phase5_metadata(
    recording_id
):

    metadata_path = (
        PHASE5_RECORDINGS
        / f"{recording_id}_metadata.csv"
    )

    if not metadata_path.exists():

        raise FileNotFoundError(
            f"Phase-5 metadata missing:\n"
            f"{metadata_path}"
        )

    df = pd.read_csv(
        metadata_path
    )

    required = [
        "recording_id",
        "window_index",
        "timestamp",
        "target",
        "text",
    ]

    missing = [
        c
        for c in required
        if c not in df.columns
    ]

    if missing:

        raise RuntimeError(
            f"{recording_id}: metadata missing "
            f"columns {missing}"
        )

    df = (
        df
        .sort_values("window_index")
        .reset_index(drop=True)
    )

    return df


# ============================================================
# 8. EOG DETECTION
# ============================================================

def detect_eog_channels(raw):

    eog = []

    for ch in raw.info["chs"]:

        if (
            ch["kind"]
            == mne.io.constants.FIFF.FIFFV_EOG_CH
        ):
            eog.append(
                ch["ch_name"]
            )

    # Fallback for BrainVision naming.
    if not eog:

        keywords = [
            "EOG",
            "RVEOG",
            "LHEOG",
            "RHEOG",
        ]

        for name in raw.ch_names:

            upper = name.upper()

            if any(
                k in upper
                for k in keywords
            ):

                eog.append(name)

    return eog


# ============================================================
# 9. SAVE FUNCTION
# ============================================================

def save_variant(
    output_dir,
    recording_id,
    variant_name,
    windows,
    metadata,
):

    npy_path = (
        output_dir
        / f"{recording_id}_{variant_name}_windows.npy"
    )

    csv_path = (
        output_dir
        / f"{recording_id}_{variant_name}_metadata.csv"
    )

    windows = np.asarray(
        windows,
        dtype=np.float32
    )

    np.save(
        npy_path,
        windows
    )

    metadata.to_csv(
        csv_path,
        index=False
    )

    # Verify actual files exist.
    if not npy_path.exists():
        raise RuntimeError(
            f"NPY file was not created:\n{npy_path}"
        )

    if not csv_path.exists():
        raise RuntimeError(
            f"Metadata file was not created:\n{csv_path}"
        )

    # Verify saved shape.
    saved = np.load(
        npy_path,
        mmap_mode="r"
    )

    if saved.shape != windows.shape:

        raise RuntimeError(
            f"Saved shape mismatch:\n"
            f"{saved.shape} != {windows.shape}"
        )

    return npy_path, csv_path


# ============================================================
# 10. PROCESS RECORDINGS
# ============================================================

successful = 0
failed = 0

start_time = time.time()

for position, row in recordings_df.iterrows():

    recording_id = row[
        "recording_id"
    ]

    split = row[
        "split"
    ]

    checkpoint_key = (
        f"{split}/{recording_id}"
    )

    output_dir = (
        OUTPUT_ROOT
        / split
        / recording_id
    )

    print("\n" + "=" * 75)
    print(
        f"[{position + 1}/62] "
        f"{recording_id} "
        f"({split})"
    )
    print("=" * 75)

    # --------------------------------------------------------
    # Resume check.
    # --------------------------------------------------------

    expected_files = [
        output_dir
        / f"{recording_id}_{variant}_windows.npy"
        for variant in VARIANTS
    ]

    if (
        checkpoint_key
        in checkpoint["completed"]
        and all(
            p.exists()
            for p in expected_files
        )
    ):

        print(
            "Already completed; skipping."
        )

        successful += 1
        continue

    try:

        # ----------------------------------------------------
        # Raw path
        # ----------------------------------------------------

        raw_path = raw_path_map[
            recording_id
        ]

        print(
            "Raw EEG:",
            raw_path
        )

        # ----------------------------------------------------
        # Phase-5 metadata
        # ----------------------------------------------------

        metadata = load_phase5_metadata(
            recording_id
        )

        n_events = len(
            metadata
        )

        print(
            "Phase-5 events:",
            n_events
        )

        # ----------------------------------------------------
        # Load BrainVision
        # ----------------------------------------------------

        raw = mne.io.read_raw_brainvision(
            str(raw_path),
            preload=True,
            verbose=False
        )

        original_nchan = (
            raw.info["nchan"]
        )

        original_sfreq = (
            raw.info["sfreq"]
        )

        print(
            "Original:",
            original_nchan,
            "channels |",
            raw.n_times,
            "samples |",
            original_sfreq,
            "Hz"
        )

        # ----------------------------------------------------
        # Detect EOG
        # ----------------------------------------------------

        eog_channels = detect_eog_channels(
            raw
        )

        print(
            "EOG:",
            eog_channels
        )

        if eog_channels:

            raw.drop_channels(
                eog_channels
            )

        # ----------------------------------------------------
        # Same preprocessing as Phase 2
        # ----------------------------------------------------

        raw.filter(
            l_freq=0.1,
            h_freq=20.0,
            verbose=False
        )

        raw.resample(
            FS,
            npad="auto",
            verbose=False
        )

        # ----------------------------------------------------
        # Verify final EEG geometry
        # ----------------------------------------------------

        if raw.info["nchan"] != 61:

            raise RuntimeError(
                f"{recording_id}: expected 61 EEG "
                f"channels, got {raw.info['nchan']}"
            )

        if int(
            round(
                raw.info["sfreq"]
            )
        ) != FS:

            raise RuntimeError(
                f"{recording_id}: expected "
                f"{FS} Hz, got "
                f"{raw.info['sfreq']}"
            )

        eeg = raw.get_data()

        print(
            "Preprocessed:",
            eeg.shape,
            "|",
            raw.info["sfreq"],
            "Hz"
        )

        # ----------------------------------------------------
        # Event samples from Phase-5 timestamps
        # ----------------------------------------------------

        timestamps = (
            metadata["timestamp"]
            .astype(float)
            .to_numpy()
        )

        event_samples = np.rint(
            timestamps * FS
        ).astype(np.int64)

        # ----------------------------------------------------
        # Extract each temporal variant
        # ----------------------------------------------------

        generated = {}

        for variant_name, cfg in VARIANTS.items():

            pre_samples = int(
                round(
                    abs(cfg["tmin"])
                    * FS
                )
            )

            post_samples = int(
                round(
                    cfg["tmax"]
                    * FS
                )
            )

            total_samples = (
                pre_samples
                + post_samples
            )

            if total_samples != cfg[
                "samples"
            ]:

                raise RuntimeError(
                    f"{variant_name}: expected "
                    f"{cfg['samples']} samples, "
                    f"computed {total_samples}"
                )

            windows = []

            for event_sample in event_samples:

                start = (
                    event_sample
                    - pre_samples
                )

                stop = (
                    event_sample
                    + post_samples
                )

                if start < 0:

                    raise RuntimeError(
                        f"{recording_id}/{variant_name}: "
                        f"window begins before recording "
                        f"boundary at sample {start}"
                    )

                if stop > eeg.shape[1]:

                    raise RuntimeError(
                        f"{recording_id}/{variant_name}: "
                        f"window ends after recording "
                        f"boundary at sample {stop}"
                    )

                window = eeg[
                    :,
                    start:stop
                ]

                if window.shape != (
                    61,
                    total_samples
                ):

                    raise RuntimeError(
                        f"{recording_id}/{variant_name}: "
                        f"unexpected window shape "
                        f"{window.shape}"
                    )

                windows.append(
                    window.astype(
                        np.float32,
                        copy=False
                    )
                )

            windows = np.stack(
                windows,
                axis=0
            )

            if windows.shape != (
                n_events,
                61,
                total_samples
            ):

                raise RuntimeError(
                    f"{recording_id}/{variant_name}: "
                    f"final shape mismatch "
                    f"{windows.shape}"
                )

            generated[
                variant_name
            ] = windows

            print(
                f"  {variant_name:8s}: "
                f"{windows.shape}"
            )

        # ----------------------------------------------------
        # EXACT BASELINE VERIFICATION
        # ----------------------------------------------------

        original_baseline_path = (
            PHASE5_RECORDINGS
            / f"{recording_id}_windows.npy"
        )

        if not original_baseline_path.exists():

            raise FileNotFoundError(
                f"Original Phase-5 windows missing:\n"
                f"{original_baseline_path}"
            )

        original_baseline = np.load(
            original_baseline_path,
            mmap_mode="r"
        )

        generated_baseline = generated[
            "baseline"
        ]

        if original_baseline.shape != (
            generated_baseline.shape
        ):

            raise RuntimeError(
                f"{recording_id}: baseline shape mismatch "
                f"with Phase-5"
            )

        if not np.array_equal(
            original_baseline,
            generated_baseline
        ):

            mse = float(
                np.mean(
                    (
                        original_baseline
                        - generated_baseline
                    ) ** 2
                )
            )

            mae = float(
                np.mean(
                    np.abs(
                        original_baseline
                        - generated_baseline
                    )
                )
            )

            raise RuntimeError(
                f"{recording_id}: generated baseline "
                f"does NOT exactly reproduce Phase-5.\n"
                f"MSE={mse}\n"
                f"MAE={mae}"
            )

        print(
            "  Baseline exact reproduction: PASS"
        )

        # ----------------------------------------------------
        # CREATE OUTPUT DIRECTORY
        # ----------------------------------------------------

        output_dir.mkdir(
            parents=True,
            exist_ok=True
        )

        # ----------------------------------------------------
        # SAVE ALL FOUR VARIANTS
        # ----------------------------------------------------

        saved = {}

        for variant_name in VARIANTS:

            windows = generated[
                variant_name
            ]

            npy_path, csv_path = (
                save_variant(
                    output_dir,
                    recording_id,
                    variant_name,
                    windows,
                    metadata
                )
            )

            saved[
                variant_name
            ] = {
                "npy": str(
                    npy_path
                ),
                "metadata": str(
                    csv_path
                ),
                "shape": list(
                    windows.shape
                ),
            }

            print(
                f"  SAVED {variant_name}: "
                f"{npy_path.name}"
            )

        # ----------------------------------------------------
        # Recording summary
        # ----------------------------------------------------

        recording_summary = {
            "recording_id":
                recording_id,
            "split":
                split,
            "raw_path":
                str(raw_path),
            "original_channels":
                int(original_nchan),
            "original_sampling_frequency":
                float(original_sfreq),
            "eog_channels":
                eog_channels,
            "final_eeg_channels":
                int(raw.info["nchan"]),
            "final_sampling_frequency":
                float(raw.info["sfreq"]),
            "phase5_event_count":
                int(n_events),
            "baseline_exact_match":
                True,
            "variants":
                saved,
        }

        summary_path = (
            output_dir
            / f"{recording_id}_summary.json"
        )

        with open(
            summary_path,
            "w"
        ) as f:

            json.dump(
                recording_summary,
                f,
                indent=2
            )

        # ----------------------------------------------------
        # Checkpoint
        # ----------------------------------------------------

        checkpoint[
            "completed"
        ][checkpoint_key] = (
            recording_summary
        )

        checkpoint[
            "failed"
        ].pop(
            checkpoint_key,
            None
        )

        with open(
            CHECKPOINT_PATH,
            "w"
        ) as f:

            json.dump(
                checkpoint,
                f,
                indent=2
            )

        successful += 1

        print(
            f"  COMPLETE: {recording_id}"
        )

    except Exception as exc:

        failed += 1

        error_text = (
            f"{type(exc).__name__}: {exc}"
        )

        print(
            f"\n  FAILED: {recording_id}"
        )

        print(
            " ",
            error_text
        )

        checkpoint[
            "failed"
        ][checkpoint_key] = {
            "recording_id":
                recording_id,
            "split":
                split,
            "error":
                error_text,
        }

        with open(
            CHECKPOINT_PATH,
            "w"
        ) as f:

            json.dump(
                checkpoint,
                f,
                indent=2
            )

        traceback.print_exc()

        # Continue to next recording.
        continue


# ============================================================
# 11. FINAL INVENTORY
# ============================================================

all_npy = sorted(
    OUTPUT_ROOT.rglob(
        "*_windows.npy"
    )
)

all_csv = sorted(
    OUTPUT_ROOT.rglob(
        "*_metadata.csv"
    )
)

expected_recordings = 62
expected_variants = 4

expected_npy = (
    expected_recordings
    * expected_variants
)

print("\n")
print("=" * 75)
print("STEP 6B — FINAL INVENTORY")
print("=" * 75)

print(
    "Successful recordings:",
    successful
)

print(
    "Failed recordings:",
    failed
)

print(
    "NPY files:",
    len(all_npy),
    "/",
    expected_npy
)

print(
    "Metadata CSV files:",
    len(all_csv),
    "/",
    expected_npy
)

print("\nVariant counts:")

variant_counts = {}

for variant_name in VARIANTS:

    count = len(
        list(
            OUTPUT_ROOT.rglob(
                f"*_{variant_name}_windows.npy"
            )
        )
    )

    variant_counts[
        variant_name
    ] = count

    print(
        f"  {variant_name:8s}: "
        f"{count}/62"
    )


# ============================================================
# 12. SAVE MASTER SUMMARY
# ============================================================

master_summary = {
    "experiment":
        "Diagnostic Step 6B",
    "sampling_frequency":
        FS,
    "recordings":
        expected_recordings,
    "successful_recordings":
        successful,
    "failed_recordings":
        failed,
    "expected_npy_files":
        expected_npy,
    "actual_npy_files":
        len(all_npy),
    "expected_metadata_files":
        expected_npy,
    "actual_metadata_files":
        len(all_csv),
    "variant_counts":
        variant_counts,
    "variants":
        VARIANTS,
    "phase6_manifest":
        str(PHASE6_MANIFEST),
    "output_root":
        str(OUTPUT_ROOT),
    "checkpoint":
        str(CHECKPOINT_PATH),
}

with open(
    SUMMARY_PATH,
    "w"
) as f:

    json.dump(
        master_summary,
        f,
        indent=2
    )


# ============================================================
# 13. FINAL STATUS
# ============================================================

print("\n")
print("=" * 75)

if (
    len(all_npy) == expected_npy
    and
    len(all_csv) == expected_npy
    and
    failed == 0
):

    print(
        "STEP 6B STATUS: READY FOR STEP 6C"
    )

    print(
        "All 62 recordings × 4 temporal variants "
        "were generated successfully."
    )

else:

    print(
        "STEP 6B STATUS: INCOMPLETE"
    )

    print(
        "Do NOT run Step 6C yet."
    )

print("=" * 75)

print(
    "\nSummary:",
    SUMMARY_PATH
)

print(
    "Checkpoint:",
    CHECKPOINT_PATH
)

if all_npy:

    print(
        "\nFirst generated files:"
    )

    for p in all_npy[:8]:
        print(
            " ",
            p
        )
```

```python
# ============================================================
# NEUROSENTINEL — STEP 6B
# Temporal Window Expansion from Original BrainVision EEG
#
# Variants:
#   baseline = [-0.2, +0.3] -> 25 samples
#   W1       = [-0.4, +0.4] -> 40 samples
#   W2       = [-0.5, +0.5] -> 50 samples
#   W3       = [-0.2, +0.5] -> 35 samples
#
# Frozen:
#   - Phase-6 subject split
#   - Phase-5 preprocessing
#   - Phase-5 event timestamps
#   - 61 EEG channels
#   - 50 Hz sampling rate
#
# Critical verification:
#   Generated baseline [-0.2,+0.3] must exactly reproduce
#   the existing Phase-5 windows.
# ============================================================

import os
import json
import time
import traceback
from pathlib import Path

import numpy as np
import pandas as pd
import mne


# ============================================================
# 1. PATHS
# ============================================================

ROOT = Path(
    "/content/gdrive/MyDrive/NeuroSentinel"
)

PHASE6_MANIFEST = (
    ROOT
    / "07_phase6_subject_split"
    / "phase6_subject_split_manifest_resolved.csv"
)

PHASE5_RECORDINGS = (
    ROOT
    / "06_phase5_utility_labels"
    / "recordings"
)

OUTPUT_ROOT = (
    ROOT
    / "08_phase6_temporal_windows"
)

CHECKPOINT_PATH = (
    ROOT
    / "checkpoints"
    / "phase6_temporal_windows_checkpoint.json"
)

SUMMARY_PATH = (
    OUTPUT_ROOT
    / "phase6_temporal_windows_summary.json"
)

OUTPUT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

CHECKPOINT_PATH.parent.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# 2. EXPERIMENT CONFIGURATION
# ============================================================

FS = 50

VARIANTS = {
    "baseline": {
        "tmin": -0.2,
        "tmax": 0.3,
        "samples": 25,
    },
    "W1": {
        "tmin": -0.4,
        "tmax": 0.4,
        "samples": 40,
    },
    "W2": {
        "tmin": -0.5,
        "tmax": 0.5,
        "samples": 50,
    },
    "W3": {
        "tmin": -0.2,
        "tmax": 0.5,
        "samples": 35,
    },
}


# ============================================================
# 3. LOAD PHASE-6 RESOLVED MANIFEST
# ============================================================

if not PHASE6_MANIFEST.exists():
    raise FileNotFoundError(
        f"Phase-6 resolved manifest not found:\n"
        f"{PHASE6_MANIFEST}"
    )

phase6 = pd.read_csv(
    PHASE6_MANIFEST
)

required_columns = [
    "recording_id",
    "split",
    "raw_resolved",
]

missing_columns = [
    c
    for c in required_columns
    if c not in phase6.columns
]

if missing_columns:
    raise RuntimeError(
        "Phase-6 manifest missing required columns: "
        f"{missing_columns}\n"
        f"Available columns: {list(phase6.columns)}"
    )


phase6["recording_id"] = (
    phase6["recording_id"]
    .astype(str)
)

phase6["split"] = (
    phase6["split"]
    .astype(str)
)

phase6["raw_resolved"] = (
    phase6["raw_resolved"]
    .astype(str)
)


recordings_df = (
    phase6[
        [
            "recording_id",
            "split",
            "raw_resolved",
        ]
    ]
    .drop_duplicates()
    .sort_values("recording_id")
    .reset_index(drop=True)
)


# ============================================================
# 4. VERIFY PHASE-6 MANIFEST
# ============================================================

if len(recordings_df) != 62:
    raise RuntimeError(
        f"Expected exactly 62 recordings, "
        f"found {len(recordings_df)}"
    )

expected_split_counts = {
    "train": 41,
    "val": 9,
    "test": 12,
}

actual_split_counts = (
    recordings_df
    .groupby("split")["recording_id"]
    .count()
    .to_dict()
)

if actual_split_counts != expected_split_counts:
    raise RuntimeError(
        "Phase-6 split counts do not match frozen split.\n"
        f"Expected: {expected_split_counts}\n"
        f"Actual:   {actual_split_counts}"
    )


print("=" * 75)
print("NEUROSENTINEL — STEP 6B")
print("=" * 75)

print(
    "\nPhase-6 manifest:"
)

print(
    PHASE6_MANIFEST
)

print(
    "\nRecordings:",
    len(recordings_df)
)

print(
    "\nSplit counts:"
)

print(
    recordings_df
    .groupby("split")["recording_id"]
    .count()
    .to_string()
)


# ============================================================
# 5. RESOLVE RAW EEG PATHS
# ============================================================

def resolve_raw_path(path_string):

    path_string = str(
        path_string
    )

    original_path = Path(
        path_string
    )

    # --------------------------------------------------------
    # Case 1: path already exists
    # --------------------------------------------------------

    if original_path.exists():
        return original_path

    # --------------------------------------------------------
    # Case 2: old /content/drive mount
    # --------------------------------------------------------

    if path_string.startswith(
        "/content/drive/"
    ):

        converted = Path(
            path_string.replace(
                "/content/drive/",
                "/content/gdrive/",
                1
            )
        )

        if converted.exists():
            return converted

    # --------------------------------------------------------
    # Case 3: known EEG directory + filename
    # --------------------------------------------------------

    eeg_root = (
        ROOT
        / "data"
        / "SpanishBCBL"
        / "EEG"
        / "EEG"
    )

    filename_candidate = (
        eeg_root
        / original_path.name
    )

    if filename_candidate.exists():
        return filename_candidate

    return None


print(
    "\nVerifying raw EEG paths..."
)

raw_path_map = {}

missing_raw = []

for _, row in recordings_df.iterrows():

    recording_id = row[
        "recording_id"
    ]

    raw_path = resolve_raw_path(
        row["raw_resolved"]
    )

    if raw_path is None:

        missing_raw.append(
            {
                "recording_id":
                    recording_id,
                "path":
                    row["raw_resolved"],
            }
        )

    else:

        raw_path_map[
            recording_id
        ] = raw_path


if missing_raw:

    print(
        "\nMissing raw EEG files:"
    )

    for item in missing_raw:
        print(
            item["recording_id"],
            "->",
            item["path"]
        )

    raise FileNotFoundError(
        f"{len(missing_raw)} raw EEG files could not be resolved."
    )


print(
    "Raw EEG path verification: PASS"
)

print(
    "Resolved raw recordings:",
    len(raw_path_map)
)


# ============================================================
# 6. CHECKPOINT
# ============================================================

if CHECKPOINT_PATH.exists():

    try:

        with open(
            CHECKPOINT_PATH,
            "r"
        ) as f:

            checkpoint = json.load(f)

    except Exception:

        print(
            "Existing checkpoint could not be read."
        )

        print(
            "Starting a fresh checkpoint."
        )

        checkpoint = {
            "completed": {},
            "failed": {},
        }

else:

    checkpoint = {
        "completed": {},
        "failed": {},
    }


checkpoint.setdefault(
    "completed",
    {}
)

checkpoint.setdefault(
    "failed",
    {})


# ============================================================
# 7. LOAD PHASE-5 METADATA
# ============================================================
#
# IMPORTANT:
# Phase-5 metadata does NOT contain window_index.
#
# Existing columns are:
#   recording_id
#   subject
#   session
#   task
#   trial_id
#   keystroke_index
#   timestamp
#   duration
#   target
#   text
#
# We preserve the original row order and create a local
# window_index only for bookkeeping.
# ============================================================

def load_phase5_metadata(
    recording_id
):

    metadata_path = (
        PHASE5_RECORDINGS
        / f"{recording_id}_metadata.csv"
    )

    if not metadata_path.exists():

        raise FileNotFoundError(
            f"Phase-5 metadata missing:\n"
            f"{metadata_path}"
        )

    df = pd.read_csv(
        metadata_path
    )

    required = [
        "recording_id",
        "timestamp",
        "target",
        "text",
    ]

    missing = [
        c
        for c in required
        if c not in df.columns
    ]

    if missing:

        raise RuntimeError(
            f"{recording_id}: metadata missing "
            f"columns {missing}\n"
            f"Available columns: {list(df.columns)}"
        )

    # --------------------------------------------------------
    # Preserve original Phase-5 row order.
    # --------------------------------------------------------

    df = df.reset_index(
        drop=True
    )

    # --------------------------------------------------------
    # Create local bookkeeping index.
    #
    # This does NOT alter the original event alignment.
    # --------------------------------------------------------

    df.insert(
        0,
        "window_index",
        np.arange(
            len(df),
            dtype=np.int64
        )
    )

    # --------------------------------------------------------
    # Verify timestamps.
    # --------------------------------------------------------

    if df["timestamp"].isna().any():

        raise RuntimeError(
            f"{recording_id}: metadata contains "
            "NaN timestamps."
        )

    df["timestamp"] = (
        pd.to_numeric(
            df["timestamp"],
            errors="raise"
        )
    )

    return df


# ============================================================
# 8. DETECT EOG CHANNELS
# ============================================================

def detect_eog_channels(raw):

    eog_channels = []

    # First use MNE channel type.
    for ch in raw.info["chs"]:

        if (
            ch["kind"]
            == mne.io.constants.FIFF.FIFFV_EOG_CH
        ):

            eog_channels.append(
                ch["ch_name"]
            )

    # Fallback based on names.
    if not eog_channels:

        keywords = [
            "EOG",
            "RVEOG",
            "LHEOG",
            "RHEOG",
        ]

        for name in raw.ch_names:

            upper_name = (
                str(name)
                .upper()
            )

            if any(
                keyword in upper_name
                for keyword in keywords
            ):

                eog_channels.append(
                    name
                )

    # Remove duplicates while preserving order.
    eog_channels = list(
        dict.fromkeys(
            eog_channels
        )
    )

    return eog_channels


# ============================================================
# 9. EXTRACT TEMPORAL WINDOWS
# ============================================================

def extract_windows(
    eeg,
    event_samples,
    tmin,
    tmax,
    expected_samples,
    recording_id,
    variant_name,
):

    pre_samples = int(
        round(
            abs(tmin)
            * FS
        )
    )

    post_samples = int(
        round(
            tmax
            * FS
        )
    )

    total_samples = (
        pre_samples
        + post_samples
    )

    if total_samples != expected_samples:

        raise RuntimeError(
            f"{recording_id}/{variant_name}: "
            f"expected {expected_samples} samples, "
            f"computed {total_samples}"
        )

    windows = []

    for event_number, event_sample in enumerate(
        event_samples
    ):

        start = (
            int(event_sample)
            - pre_samples
        )

        stop = (
            int(event_sample)
            + post_samples
        )

        if start < 0:

            raise RuntimeError(
                f"{recording_id}/{variant_name}: "
                f"event {event_number} produces "
                f"negative start={start}"
            )

        if stop > eeg.shape[1]:

            raise RuntimeError(
                f"{recording_id}/{variant_name}: "
                f"event {event_number} produces "
                f"stop={stop}, recording has "
                f"{eeg.shape[1]} samples"
            )

        window = eeg[
            :,
            start:stop
        ]

        if window.shape != (
            61,
            total_samples
        ):

            raise RuntimeError(
                f"{recording_id}/{variant_name}: "
                f"unexpected window shape "
                f"{window.shape}"
            )

        windows.append(
            window.astype(
                np.float32,
                copy=False
            )
        )

    if len(windows) == 0:

        raise RuntimeError(
            f"{recording_id}/{variant_name}: "
            "no windows extracted."
        )

    windows = np.stack(
        windows,
        axis=0
    )

    expected_shape = (
        len(event_samples),
        61,
        total_samples
    )

    if windows.shape != expected_shape:

        raise RuntimeError(
            f"{recording_id}/{variant_name}: "
            f"expected shape {expected_shape}, "
            f"got {windows.shape}"
        )

    return windows


# ============================================================
# 10. SAVE VARIANT
# ============================================================

def save_variant(
    output_dir,
    recording_id,
    variant_name,
    windows,
    metadata,
):

    npy_path = (
        output_dir
        / f"{recording_id}_{variant_name}_windows.npy"
    )

    csv_path = (
        output_dir
        / f"{recording_id}_{variant_name}_metadata.csv"
    )

    # Save EEG windows.
    np.save(
        npy_path,
        np.asarray(
            windows,
            dtype=np.float32
        )
    )

    # Save metadata.
    metadata.to_csv(
        csv_path,
        index=False
    )

    # --------------------------------------------------------
    # Verify files.
    # --------------------------------------------------------

    if not npy_path.exists():

        raise RuntimeError(
            f"NPY file was not created:\n"
            f"{npy_path}"
        )

    if not csv_path.exists():

        raise RuntimeError(
            f"Metadata file was not created:\n"
            f"{csv_path}"
        )

    # --------------------------------------------------------
    # Verify saved array.
    # --------------------------------------------------------

    saved = np.load(
        npy_path,
        mmap_mode="r"
    )

    if saved.shape != windows.shape:

        raise RuntimeError(
            f"Saved shape mismatch:\n"
            f"Expected: {windows.shape}\n"
            f"Saved:    {saved.shape}"
        )

    if saved.dtype != np.float32:

        raise RuntimeError(
            f"Saved dtype mismatch: "
            f"{saved.dtype}"
        )

    return npy_path, csv_path


# ============================================================
# 11. PROCESS ALL 62 RECORDINGS
# ============================================================

successful = 0
failed = 0

overall_start = time.time()

for position, row in recordings_df.iterrows():

    recording_id = (
        row["recording_id"]
    )

    split = (
        row["split"]
    )

    checkpoint_key = (
        f"{split}/{recording_id}"
    )

    output_dir = (
        OUTPUT_ROOT
        / split
        / recording_id
    )

    print("\n")
    print("=" * 75)

    print(
        f"[{position + 1}/62] "
        f"{recording_id} "
        f"({split})"
    )

    print("=" * 75)

    # --------------------------------------------------------
    # Resume support.
    # --------------------------------------------------------

    expected_files = [
        output_dir
        / f"{recording_id}_{variant_name}_windows.npy"
        for variant_name in VARIANTS
    ]

    if (
        checkpoint_key
        in checkpoint["completed"]
        and all(
            p.exists()
            for p in expected_files
        )
    ):

        print(
            "Already completed; skipping."
        )

        successful += 1

        continue

    try:

        recording_start = time.time()

        # ====================================================
        # A. RAW EEG
        # ====================================================

        raw_path = raw_path_map[
            recording_id
        ]

        print(
            "Raw EEG:",
            raw_path
        )

        # ====================================================
        # B. PHASE-5 METADATA
        # ====================================================

        metadata = load_phase5_metadata(
            recording_id
        )

        n_events = len(
            metadata
        )

        print(
            "Phase-5 events:",
            n_events
        )

        # ====================================================
        # C. LOAD ORIGINAL BRAINVISION
        # ====================================================

        raw = (
            mne.io.read_raw_brainvision(
                str(raw_path),
                preload=True,
                verbose=False
            )
        )

        original_nchan = int(
            raw.info["nchan"]
        )

        original_sfreq = float(
            raw.info["sfreq"]
        )

        original_n_times = int(
            raw.n_times
        )

        print(
            "Original:",
            original_nchan,
            "channels |",
            original_n_times,
            "samples |",
            original_sfreq,
            "Hz"
        )

        # ====================================================
        # D. EOG REMOVAL
        # ====================================================

        eog_channels = (
            detect_eog_channels(
                raw
            )
        )

        print(
            "EOG channels:",
            eog_channels
        )

        if eog_channels:

            raw.drop_channels(
                eog_channels
            )

        # ====================================================
        # E. SAME PHASE-5 PREPROCESSING
        # ====================================================

        print(
            "Filtering: 0.1–20 Hz"
        )

        raw.filter(
            l_freq=0.1,
            h_freq=20.0,
            verbose=False
        )

        print(
            "Resampling:",
            FS,
            "Hz"
        )

        raw.resample(
            FS,
            npad="auto",
            verbose=False
        )

        # ====================================================
        # F. FINAL CHANNEL / SAMPLING VERIFICATION
        # ====================================================

        final_nchan = int(
            raw.info["nchan"]
        )

        final_sfreq = float(
            raw.info["sfreq"]
        )

        if final_nchan != 61:

            raise RuntimeError(
                f"{recording_id}: expected "
                f"61 EEG channels after EOG removal, "
                f"got {final_nchan}"
            )

        if int(
            round(final_sfreq)
        ) != FS:

            raise RuntimeError(
                f"{recording_id}: expected "
                f"{FS} Hz after resampling, "
                f"got {final_sfreq}"
            )

        eeg = raw.get_data()

        if eeg.shape[0] != 61:

            raise RuntimeError(
                f"{recording_id}: EEG data has "
                f"{eeg.shape[0]} channels."
            )

        print(
            "Preprocessed EEG:",
            eeg.shape
        )

        # ====================================================
        # G. EVENT ALIGNMENT
        # ====================================================
        #
        # Phase-5 timestamps are already in seconds.
        #
        # event_sample = round(timestamp * 50)
        #
        # This is the same mapping that previously reproduced
        # Phase-5 baseline windows exactly.
        # ====================================================

        timestamps = (
            metadata[
                "timestamp"
            ]
            .astype(float)
            .to_numpy()
        )

        event_samples = np.rint(
            timestamps * FS
        ).astype(
            np.int64
        )

        # ----------------------------------------------------
        # Basic event validation.
        # ----------------------------------------------------

        if not np.all(
            np.isfinite(
                timestamps
            )
        ):

            raise RuntimeError(
                f"{recording_id}: non-finite timestamp."
            )

        if np.any(
            event_samples < 0
        ):

            raise RuntimeError(
                f"{recording_id}: negative event sample."
            )

        if np.any(
            event_samples >= eeg.shape[1]
        ):

            raise RuntimeError(
                f"{recording_id}: event sample "
                "outside EEG recording."
            )

        print(
            "Event mapping: PASS"
        )

        print(
            "Timestamp range:",
            float(timestamps.min()),
            "to",
            float(timestamps.max()),
            "seconds"
        )

        # ====================================================
        # H. GENERATE ALL FOUR VARIANTS
        # ====================================================

        generated = {}

        for variant_name, config in (
            VARIANTS.items()
        ):

            print(
                f"Generating {variant_name}: "
                f"[{config['tmin']}, "
                f"{config['tmax']}]"
            )

            windows = extract_windows(
                eeg=eeg,
                event_samples=event_samples,
                tmin=config["tmin"],
                tmax=config["tmax"],
                expected_samples=config["samples"],
                recording_id=recording_id,
                variant_name=variant_name,
            )

            generated[
                variant_name
            ] = windows

            print(
                f"  Shape: {windows.shape}"
            )

        # ====================================================
        # I. EXACT BASELINE VERIFICATION
        # ====================================================

        print(
            "\nVerifying baseline against "
            "existing Phase-5 windows..."
        )

        original_baseline_path = (
            PHASE5_RECORDINGS
            / f"{recording_id}_windows.npy"
        )

        if not original_baseline_path.exists():

            raise FileNotFoundError(
                f"Original Phase-5 windows missing:\n"
                f"{original_baseline_path}"
            )

        original_baseline = np.load(
            original_baseline_path,
            mmap_mode="r"
        )

        generated_baseline = (
            generated["baseline"]
        )

        if (
            original_baseline.shape
            != generated_baseline.shape
        ):

            raise RuntimeError(
                f"{recording_id}: baseline shape "
                "does not match Phase-5.\n"
                f"Original:   {original_baseline.shape}\n"
                f"Generated:  {generated_baseline.shape}"
            )

        # Exact equality, not approximate equality.
        exact_match = np.array_equal(
            original_baseline,
            generated_baseline
        )

        if not exact_match:

            difference = (
                original_baseline.astype(
                    np.float64
                )
                -
                generated_baseline.astype(
                    np.float64
                )
            )

            mse = float(
                np.mean(
                    difference ** 2
                )
            )

            mae = float(
                np.mean(
                    np.abs(
                        difference
                    )
                )
            )

            max_abs_error = float(
                np.max(
                    np.abs(
                        difference
                    )
                )
            )

            raise RuntimeError(
                f"{recording_id}: EXACT baseline "
                "reproduction FAILED.\n"
                f"MSE: {mse}\n"
                f"MAE: {mae}\n"
                f"Max abs error: {max_abs_error}"
            )

        print(
            "Baseline exact reproduction: PASS"
        )

        # ====================================================
        # J. CREATE OUTPUT DIRECTORY
        # ====================================================

        output_dir.mkdir(
            parents=True,
            exist_ok=True
        )

        # ====================================================
        # K. SAVE ALL FOUR VARIANTS
        # ====================================================

        saved_variants = {}

        for variant_name in VARIANTS:

            windows = generated[
                variant_name
            ]

            npy_path, csv_path = (
                save_variant(
                    output_dir=output_dir,
                    recording_id=recording_id,
                    variant_name=variant_name,
                    windows=windows,
                    metadata=metadata,
                )
            )

            saved_variants[
                variant_name
            ] = {
                "npy":
                    str(npy_path),
                "metadata":
                    str(csv_path),
                "shape":
                    list(
                        windows.shape
                    ),
            }

            print(
                f"SAVED {variant_name}:"
            )

            print(
                " ",
                npy_path
            )

            print(
                " ",
                csv_path
            )

        # ====================================================
        # L. RECORDING SUMMARY
        # ====================================================

        elapsed = (
            time.time()
            - recording_start
        )

        recording_summary = {

            "recording_id":
                recording_id,

            "split":
                split,

            "raw_path":
                str(raw_path),

            "original_channels":
                original_nchan,

            "original_samples":
                original_n_times,

            "original_sampling_frequency":
                original_sfreq,

            "eog_channels":
                eog_channels,

            "final_eeg_channels":
                final_nchan,

            "final_sampling_frequency":
                final_sfreq,

            "phase5_event_count":
                n_events,

            "timestamp_min":
                float(
                    timestamps.min()
                ),

            "timestamp_max":
                float(
                    timestamps.max()
                ),

            "baseline_exact_match":
                True,

            "processing_seconds":
                float(elapsed),

            "variants":
                saved_variants,
        }

        recording_summary_path = (
            output_dir
            / f"{recording_id}_summary.json"
        )

        with open(
            recording_summary_path,
            "w"
        ) as f:

            json.dump(
                recording_summary,
                f,
                indent=2
            )

        # ====================================================
        # M. UPDATE CHECKPOINT
        # ====================================================

        checkpoint[
            "completed"
        ][checkpoint_key] = (
            recording_summary
        )

        checkpoint[
            "failed"
        ].pop(
            checkpoint_key,
            None
        )

        with open(
            CHECKPOINT_PATH,
            "w"
        ) as f:

            json.dump(
                checkpoint,
                f,
                indent=2
            )

        successful += 1

        print(
            f"\nCOMPLETE: {recording_id}"
        )

        print(
            f"Processing time: {elapsed:.1f} s"
        )

    except Exception as exc:

        failed += 1

        error_text = (
            f"{type(exc).__name__}: {exc}"
        )

        print(
            f"\nFAILED: {recording_id}"
        )

        print(
            error_text
        )

        checkpoint[
            "failed"
        ][checkpoint_key] = {

            "recording_id":
                recording_id,

            "split":
                split,

            "error":
                error_text,
        }

        with open(
            CHECKPOINT_PATH,
            "w"
        ) as f:

            json.dump(
                checkpoint,
                f,
                indent=2
            )

        traceback.print_exc()

        # Continue with next recording.
        continue


# ============================================================
# 12. FINAL FILE INVENTORY
# ============================================================

all_npy = sorted(
    OUTPUT_ROOT.rglob(
        "*_windows.npy"
    )
)

all_csv = sorted(
    OUTPUT_ROOT.rglob(
        "*_metadata.csv"
    )
)

expected_recordings = 62
expected_variants = 4
expected_files = (
    expected_recordings
    * expected_variants
)

print("\n")
print("=" * 75)
print("STEP 6B — FINAL INVENTORY")
print("=" * 75)

print(
    "Successful recordings:",
    successful
)

print(
    "Failed recordings:",
    failed
)

print(
    "NPY files:",
    len(all_npy),
    "/",
    expected_files
)

print(
    "Metadata CSV files:",
    len(all_csv),
    "/",
    expected_files
)


# ============================================================
# 13. VERIFY EACH VARIANT COUNT
# ============================================================

variant_counts = {}

print(
    "\nVariant counts:"
)

for variant_name in VARIANTS:

    count = len(
        list(
            OUTPUT_ROOT.rglob(
                f"*_{variant_name}_windows.npy"
            )
        )
    )

    variant_counts[
        variant_name
    ] = count

    print(
        f"  {variant_name:8s}: "
        f"{count}/62"
    )


# ============================================================
# 14. VERIFY EXPECTED SHAPES
# ============================================================

print(
    "\nVerifying generated array shapes..."
)

shape_errors = []

for variant_name, config in (
    VARIANTS.items()
):

    expected_samples = (
        config["samples"]
    )

    files = sorted(
        OUTPUT_ROOT.rglob(
            f"*_{variant_name}_windows.npy"
        )
    )

    for path in files:

        try:

            arr = np.load(
                path,
                mmap_mode="r"
            )

            if (
                arr.ndim != 3
                or arr.shape[1] != 61
                or arr.shape[2]
                != expected_samples
            ):

                shape_errors.append(
                    {
                        "file":
                            str(path),
                        "shape":
                            list(arr.shape),
                        "expected":
                            [
                                "N",
                                61,
                                expected_samples,
                            ],
                    }
                )

        except Exception as exc:

            shape_errors.append(
                {
                    "file":
                        str(path),
                    "error":
                        str(exc),
                }
            )


if shape_errors:

    print(
        "Shape verification: FAILED"
    )

    for item in shape_errors[:20]:
        print(
            item
        )

else:

    print(
        "Shape verification: PASS"
    )


# ============================================================
# 15. MASTER SUMMARY
# ============================================================

overall_elapsed = (
    time.time()
    - overall_start
)

master_summary = {

    "experiment":
        "NeuroSentinel Diagnostic Step 6B",

    "description":
        "Temporal window expansion using original "
        "preprocessed-equivalent BrainVision EEG",

    "sampling_frequency_hz":
        FS,

    "variants":
        VARIANTS,

    "recordings_expected":
        expected_recordings,

    "recordings_successful":
        successful,

    "recordings_failed":
        failed,

    "expected_npy_files":
        expected_files,

    "actual_npy_files":
        len(all_npy),

    "expected_metadata_files":
        expected_files,

    "actual_metadata_files":
        len(all_csv),

    "variant_counts":
        variant_counts,

    "shape_errors":
        shape_errors,

    "phase6_manifest":
        str(
            PHASE6_MANIFEST
        ),

    "phase5_recordings":
        str(
            PHASE5_RECORDINGS
        ),

    "output_root":
        str(
            OUTPUT_ROOT
        ),

    "checkpoint":
        str(
            CHECKPOINT_PATH
        ),

    "elapsed_seconds":
        float(
            overall_elapsed
        ),

    "baseline_verification":
        (
            "exact equality against "
            "Phase-5 saved windows"
        ),
}

with open(
    SUMMARY_PATH,
    "w"
) as f:

    json.dump(
        master_summary,
        f,
        indent=2
    )


# ============================================================
# 16. FINAL STATUS
# ============================================================

print("\n")
print("=" * 75)

ready = (
    successful == 62
    and
    failed == 0
    and
    len(all_npy) == expected_files
    and
    len(all_csv) == expected_files
    and
    len(shape_errors) == 0
    and
    all(
        variant_counts[v] == 62
        for v in VARIANTS
    )
)

if ready:

    print(
        "STEP 6B STATUS: READY FOR STEP 6C"
    )

    print(
        "All 62 recordings × 4 temporal variants "
        "were generated successfully."
    )

else:

    print(
        "STEP 6B STATUS: INCOMPLETE"
    )

    print(
        "Do NOT run Step 6C yet."
    )

print("=" * 75)

print(
    "\nSummary:"
)

print(
    SUMMARY_PATH
)

print(
    "\nCheckpoint:"
)

print(
    CHECKPOINT_PATH
)

if all_npy:

    print(
        "\nFirst generated NPY files:"
    )

    for path in all_npy[:12]:

        print(
            " ",
            path
        )
```

```python
# ============================================================
# NEUROSENTINEL — STEP 6B
# Temporal Window Reconstruction from Original BrainVision EEG
# Corrected baseline verification with floating-point tolerance
# ============================================================

from pathlib import Path
import json
import traceback
import numpy as np
import pandas as pd
import mne
from tqdm.auto import tqdm

# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

ROOT = Path("/content/gdrive/MyDrive/NeuroSentinel")

PHASE6_MANIFEST = (
    ROOT
    / "07_phase6_subject_split"
    / "phase6_subject_split_manifest_resolved.csv"
)

PHASE5_RECORDINGS = (
    ROOT
    / "06_phase5_utility_labels"
    / "recordings"
)

OUTPUT_ROOT = (
    ROOT
    / "08_phase6_temporal_windows"
)

CHECKPOINT_PATH = (
    ROOT
    / "checkpoints"
    / "phase6_temporal_windows_checkpoint.json"
)

SUMMARY_PATH = (
    OUTPUT_ROOT
    / "phase6_temporal_windows_summary.json"
)

OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)
CHECKPOINT_PATH.parent.mkdir(parents=True, exist_ok=True)


# ------------------------------------------------------------
# 2. EXPERIMENT CONFIGURATION
# ------------------------------------------------------------

FS = 50.0
LOWCUT = 0.1
HIGHCUT = 20.0

EXPECTED_EEG_CHANNELS = 61

VARIANTS = {
    "baseline": {
        "tmin": -0.2,
        "tmax": 0.3,
        "samples": 25,
    },
    "W1": {
        "tmin": -0.4,
        "tmax": 0.4,
        "samples": 40,
    },
    "W2": {
        "tmin": -0.5,
        "tmax": 0.5,
        "samples": 50,
    },
    "W3": {
        "tmin": -0.2,
        "tmax": 0.5,
        "samples": 35,
    },
}

EXPECTED_RECORDINGS = 62

# Numerical tolerance for reproducing Phase-5 baseline.
#
# The previous failure had:
#   MSE  = 4.28e-25
#   MAE  = 3.49e-13
#   MAX  = 2.81e-11
#
# Therefore this tolerance is extremely conservative.
BASELINE_RTOL = 1e-5
BASELINE_ATOL = 1e-8


# ------------------------------------------------------------
# 3. LOAD PHASE 6 MANIFEST
# ------------------------------------------------------------

if not PHASE6_MANIFEST.exists():
    raise FileNotFoundError(
        f"Phase 6 manifest not found:\n{PHASE6_MANIFEST}"
    )

phase6 = pd.read_csv(PHASE6_MANIFEST)

required_phase6_cols = {
    "recording_id",
    "split",
    "subject",
    "raw_resolved",
}

missing = required_phase6_cols - set(phase6.columns)

if missing:
    raise RuntimeError(
        f"Phase 6 manifest missing columns: {sorted(missing)}"
    )

phase6 = phase6.copy()

print("=" * 70)
print("PHASE 6 MANIFEST")
print("=" * 70)
print(f"Rows: {len(phase6)}")

print("\nSplit counts:")
print(phase6["split"].value_counts().sort_index())

expected_split_counts = {
    "train": 41,
    "val": 9,
    "test": 12,
}

actual_split_counts = phase6["split"].value_counts().to_dict()

if actual_split_counts != expected_split_counts:
    raise RuntimeError(
        f"Unexpected split counts.\n"
        f"Expected: {expected_split_counts}\n"
        f"Actual:   {actual_split_counts}"
    )

if len(phase6) != EXPECTED_RECORDINGS:
    raise RuntimeError(
        f"Expected {EXPECTED_RECORDINGS} recordings, "
        f"found {len(phase6)}."
    )

if phase6["recording_id"].duplicated().any():
    raise RuntimeError("Duplicate recording IDs in Phase 6 manifest.")

print("Phase 6 manifest verification: PASS")


# ------------------------------------------------------------
# 4. CHECKPOINT LOAD
# ------------------------------------------------------------

if CHECKPOINT_PATH.exists():

    with open(CHECKPOINT_PATH, "r") as f:
        checkpoint = json.load(f)

    print("\nExisting checkpoint found:")
    print(CHECKPOINT_PATH)

else:

    checkpoint = {
        "successful_recordings": [],
        "failed_recordings": {},
        "variants": {
            name: []
            for name in VARIANTS
        },
    }

    print("\nNo previous checkpoint found. Starting fresh.")


successful_recordings = set(
    checkpoint.get("successful_recordings", [])
)

failed_recordings = dict(
    checkpoint.get("failed_recordings", {})
)


# ------------------------------------------------------------
# 5. PATH RESOLUTION
# ------------------------------------------------------------

def resolve_raw_path(raw_path):
    """
    Resolve BrainVision raw EEG path across the current Colab
    /content/gdrive mount.
    """

    if pd.isna(raw_path):
        return None

    raw_path = str(raw_path)

    candidates = []

    # Original path
    candidates.append(Path(raw_path))

    # Drive mount correction
    if raw_path.startswith("/content/drive/"):
        candidates.append(
            Path(
                raw_path.replace(
                    "/content/drive/",
                    "/content/gdrive/",
                    1,
                )
            )
        )

    # Known EEG root fallback
    filename = Path(raw_path).name

    eeg_root = (
        ROOT
        / "data"
        / "SpanishBCBL"
        / "EEG"
        / "EEG"
    )

    candidates.append(eeg_root / filename)

    for candidate in candidates:

        if candidate.exists():
            return candidate

    return None


# ------------------------------------------------------------
# 6. RAW EEG LOADING
# ------------------------------------------------------------

def detect_eog_channels(raw):
    """
    Detect EOG channels using MNE channel types plus
    BrainVision channel naming fallback.
    """

    eog_names = []

    # First use MNE channel types
    try:
        eog_names.extend(
            mne.pick_types(
                raw.info,
                meg=False,
                eeg=False,
                eog=True,
                exclude=[],
            )
        )
    except Exception:
        pass

    eog_names = set(
        raw.ch_names[i]
        for i in eog_names
        if 0 <= i < len(raw.ch_names)
    )

    # Name-based fallback
    for ch in raw.ch_names:

        upper = ch.upper()

        if (
            "EOG" in upper
            or upper in {
                "HEOG",
                "VEOG",
                "LEOG",
                "REOG",
                "LHEOG",
                "RHEOG",
                "LVEOG",
                "RVEOG",
            }
        ):
            eog_names.add(ch)

    return sorted(eog_names)


# ------------------------------------------------------------
# 7. SAVE CHECKPOINT
# ------------------------------------------------------------

def save_checkpoint():

    checkpoint["successful_recordings"] = sorted(
        successful_recordings
    )

    checkpoint["failed_recordings"] = failed_recordings

    checkpoint["variants"] = {
        name: sorted(
            [
                r
                for r in successful_recordings
            ]
        )
        for name in VARIANTS
    }

    checkpoint["configuration"] = {
        "fs": FS,
        "lowcut": LOWCUT,
        "highcut": HIGHCUT,
        "expected_eeg_channels": EXPECTED_EEG_CHANNELS,
        "baseline_rtol": BASELINE_RTOL,
        "baseline_atol": BASELINE_ATOL,
        "variants": VARIANTS,
    }

    with open(CHECKPOINT_PATH, "w") as f:
        json.dump(
            checkpoint,
            f,
            indent=2,
        )


# ------------------------------------------------------------
# 8. MAIN PROCESSING LOOP
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("STEP 6B — TEMPORAL WINDOW RECONSTRUCTION")
print("=" * 70)

print("\nVariants:")

for name, cfg in VARIANTS.items():

    print(
        f"  {name:8s}: "
        f"[{cfg['tmin']:+.1f}, {cfg['tmax']:+.1f}] s "
        f"→ {cfg['samples']} samples"
    )

print(
    f"\nBaseline verification tolerance: "
    f"rtol={BASELINE_RTOL}, atol={BASELINE_ATOL}"
)


for row_idx, row in tqdm(
    phase6.iterrows(),
    total=len(phase6),
    desc="Processing recordings",
):

    recording_id = str(row["recording_id"])
    split = str(row["split"])

    # --------------------------------------------------------
    # Resume
    # --------------------------------------------------------

    if recording_id in successful_recordings:

        print(
            f"\n[{recording_id}] already completed — skipping."
        )

        continue

    print("\n" + "-" * 70)
    print(
        f"Recording {row_idx + 1}/{len(phase6)}: "
        f"{recording_id}"
    )
    print(f"Split: {split}")

    try:

        # ----------------------------------------------------
        # Resolve raw EEG path
        # ----------------------------------------------------

        raw_path = resolve_raw_path(
            row["raw_resolved"]
        )

        if raw_path is None:

            raise FileNotFoundError(
                f"Could not resolve raw EEG path: "
                f"{row['raw_resolved']}"
            )

        print(f"Raw EEG: {raw_path}")

        # ----------------------------------------------------
        # Phase 5 metadata
        # ----------------------------------------------------

        metadata_path = (
            PHASE5_RECORDINGS
            / f"{recording_id}_metadata.csv"
        )

        phase5_windows_path = (
            PHASE5_RECORDINGS
            / f"{recording_id}_windows.npy"
        )

        if not metadata_path.exists():

            raise FileNotFoundError(
                f"Phase 5 metadata not found:\n"
                f"{metadata_path}"
            )

        if not phase5_windows_path.exists():

            raise FileNotFoundError(
                f"Phase 5 windows not found:\n"
                f"{phase5_windows_path}"
            )

        metadata = pd.read_csv(
            metadata_path
        ).reset_index(drop=True)

        required_metadata_cols = {
            "recording_id",
            "timestamp",
            "target",
            "text",
        }

        missing_metadata = (
            required_metadata_cols
            - set(metadata.columns)
        )

        if missing_metadata:

            raise RuntimeError(
                f"{recording_id}: Phase 5 metadata missing "
                f"columns: {sorted(missing_metadata)}"
            )

        # Synthetic index corresponding to Phase 5 row order.
        metadata.insert(
            0,
            "window_index",
            np.arange(len(metadata)),
        )

        timestamps = (
            pd.to_numeric(
                metadata["timestamp"],
                errors="coerce",
            )
            .to_numpy(
                dtype=np.float64
            )
        )

        if not np.all(
            np.isfinite(timestamps)
        ):

            raise RuntimeError(
                f"{recording_id}: non-finite timestamps."
            )

        print(
            f"Phase 5 windows: {len(metadata):,}"
        )

        # ----------------------------------------------------
        # Load original BrainVision EEG
        # ----------------------------------------------------

        print("Loading BrainVision EEG...")

        raw = mne.io.read_raw_brainvision(
            str(raw_path),
            preload=True,
            verbose="ERROR",
        )

        print(
            f"Original channels: {len(raw.ch_names)}"
        )

        print(
            f"Original sampling rate: "
            f"{raw.info['sfreq']}"
        )

        print(
            f"Original samples: "
            f"{raw.n_times:,}"
        )

        # ----------------------------------------------------
        # Detect and remove EOG
        # ----------------------------------------------------

        eog_names = detect_eog_channels(
            raw
        )

        print(
            f"EOG channels detected: "
            f"{eog_names}"
        )

        if len(eog_names) == 0:

            raise RuntimeError(
                f"{recording_id}: no EOG channels detected."
            )

        raw.drop_channels(
            eog_names
        )

        print(
            f"Channels after EOG removal: "
            f"{len(raw.ch_names)}"
        )

        if len(raw.ch_names) != EXPECTED_EEG_CHANNELS:

            raise RuntimeError(
                f"{recording_id}: expected "
                f"{EXPECTED_EEG_CHANNELS} EEG channels, "
                f"found {len(raw.ch_names)}."
            )

        # ----------------------------------------------------
        # Filter
        # ----------------------------------------------------

        print(
            f"Filtering {LOWCUT}–{HIGHCUT} Hz..."
        )

        raw.filter(
            l_freq=LOWCUT,
            h_freq=HIGHCUT,
            verbose="ERROR",
        )

        # ----------------------------------------------------
        # Resample
        # ----------------------------------------------------

        print(
            f"Resampling to {FS} Hz..."
        )

        raw.resample(
            FS,
            npad="auto",
            verbose="ERROR",
        )

        if not np.isclose(
            raw.info["sfreq"],
            FS,
        ):

            raise RuntimeError(
                f"{recording_id}: unexpected sampling "
                f"rate after resampling: "
                f"{raw.info['sfreq']}"
            )

        print(
            f"Preprocessed shape: "
            f"{raw.get_data().shape}"
        )

        # ----------------------------------------------------
        # EEG array
        # ----------------------------------------------------

        eeg = raw.get_data()

        if eeg.shape[0] != EXPECTED_EEG_CHANNELS:

            raise RuntimeError(
                f"{recording_id}: EEG array has "
                f"{eeg.shape[0]} channels, expected "
                f"{EXPECTED_EEG_CHANNELS}."
            )

        # ----------------------------------------------------
        # Event sample mapping
        # ----------------------------------------------------

        event_samples = np.rint(
            timestamps * FS
        ).astype(np.int64)

        if np.any(
            event_samples < 0
        ):

            raise RuntimeError(
                f"{recording_id}: negative event sample."
            )

        if np.any(
            event_samples >= eeg.shape[1]
        ):

            bad = np.where(
                event_samples >= eeg.shape[1]
            )[0][:10]

            raise RuntimeError(
                f"{recording_id}: event sample outside "
                f"EEG bounds. Examples: {bad.tolist()}"
            )

        # ----------------------------------------------------
        # Process each temporal variant
        # ----------------------------------------------------

        for variant_name, cfg in VARIANTS.items():

            tmin = cfg["tmin"]
            tmax = cfg["tmax"]
            n_samples = cfg["samples"]

            pre_samples = int(
                round(
                    abs(tmin) * FS
                )
            )

            expected_samples = int(
                round(
                    (tmax - tmin) * FS
                )
            )

            if expected_samples != n_samples:

                raise RuntimeError(
                    f"{recording_id}/{variant_name}: "
                    f"configuration mismatch. "
                    f"Expected {expected_samples}, "
                    f"configured {n_samples}."
                )

            # ------------------------------------------------
            # Extract windows
            # ------------------------------------------------

            windows = np.empty(
                (
                    len(event_samples),
                    EXPECTED_EEG_CHANNELS,
                    n_samples,
                ),
                dtype=np.float32,
            )

            for i, event_sample in enumerate(
                event_samples
            ):

                start = (
                    event_sample
                    - pre_samples
                )

                stop = (
                    start
                    + n_samples
                )

                if start < 0:

                    raise RuntimeError(
                        f"{recording_id}/{variant_name}: "
                        f"window {i} starts before EEG."
                    )

                if stop > eeg.shape[1]:

                    raise RuntimeError(
                        f"{recording_id}/{variant_name}: "
                        f"window {i} ends after EEG."
                    )

                windows[i] = (
                    eeg[
                        :,
                        start:stop,
                    ]
                )

            # ------------------------------------------------
            # Baseline exact/numerical reproduction
            # ------------------------------------------------

            if variant_name == "baseline":

                original_baseline = np.load(
                    phase5_windows_path,
                    mmap_mode="r",
                )

                if original_baseline.shape != windows.shape:

                    raise RuntimeError(
                        f"{recording_id}: baseline shape "
                        f"mismatch.\n"
                        f"Original:   "
                        f"{original_baseline.shape}\n"
                        f"Generated:  "
                        f"{windows.shape}"
                    )

                baseline_allclose = np.allclose(
                    original_baseline,
                    windows,
                    rtol=BASELINE_RTOL,
                    atol=BASELINE_ATOL,
                )

                if not baseline_allclose:

                    difference = (
                        original_baseline.astype(
                            np.float64
                        )
                        -
                        windows.astype(
                            np.float64
                        )
                    )

                    mse = float(
                        np.mean(
                            difference ** 2
                        )
                    )

                    mae = float(
                        np.mean(
                            np.abs(
                                difference
                            )
                        )
                    )

                    max_abs_error = float(
                        np.max(
                            np.abs(
                                difference
                            )
                        )
                    )

                    raise RuntimeError(
                        f"{recording_id}: "
                        f"EXACT baseline reproduction FAILED "
                        f"beyond tolerance.\n"
                        f"MSE: {mse}\n"
                        f"MAE: {mae}\n"
                        f"Max abs error: {max_abs_error}\n"
                        f"Tolerance: "
                        f"rtol={BASELINE_RTOL}, "
                        f"atol={BASELINE_ATOL}"
                    )

                # Report actual numerical error
                difference = (
                    original_baseline.astype(
                        np.float64
                    )
                    -
                    windows.astype(
                        np.float64
                    )
                )

                mse = float(
                    np.mean(
                        difference ** 2
                    )
                )

                mae = float(
                    np.mean(
                        np.abs(
                            difference
                        )
                    )
                )

                max_abs_error = float(
                    np.max(
                        np.abs(
                            difference
                        )
                    )
                )

                print(
                    f"  baseline verification: PASS "
                    f"(MSE={mse:.3e}, "
                    f"MAE={mae:.3e}, "
                    f"MAX={max_abs_error:.3e})"
                )

                del original_baseline

            # ------------------------------------------------
            # Output directory
            # ------------------------------------------------

            variant_dir = (
                OUTPUT_ROOT
                / split
                / recording_id
            )

            variant_dir.mkdir(
                parents=True,
                exist_ok=True,
            )

            npy_path = (
                variant_dir
                / f"{recording_id}_{variant_name}_windows.npy"
            )

            metadata_path_out = (
                variant_dir
                / f"{recording_id}_{variant_name}_metadata.csv"
            )

            summary_path = (
                variant_dir
                / f"{recording_id}_{variant_name}_summary.json"
            )

            # ------------------------------------------------
            # Save windows
            # ------------------------------------------------

            np.save(
                npy_path,
                windows,
            )

            # ------------------------------------------------
            # Save metadata
            # ------------------------------------------------

            output_metadata = metadata.copy()

            output_metadata[
                "temporal_variant"
            ] = variant_name

            output_metadata[
                "tmin"
            ] = tmin

            output_metadata[
                "tmax"
            ] = tmax

            output_metadata[
                "window_samples"
            ] = n_samples

            output_metadata[
                "event_sample_50hz"
            ] = event_samples

            output_metadata[
                "split"
            ] = split

            output_metadata.to_csv(
                metadata_path_out,
                index=False,
            )

            # ------------------------------------------------
            # Save variant summary
            # ------------------------------------------------

            variant_summary = {
                "recording_id": recording_id,
                "split": split,
                "variant": variant_name,
                "tmin_seconds": tmin,
                "tmax_seconds": tmax,
                "sampling_rate_hz": FS,
                "window_samples": n_samples,
                "n_windows": int(
                    len(windows)
                ),
                "n_channels": int(
                    windows.shape[1]
                ),
                "dtype": str(
                    windows.dtype
                ),
                "npy_path": str(
                    npy_path
                ),
                "metadata_path": str(
                    metadata_path_out
                ),
            }

            with open(
                summary_path,
                "w",
            ) as f:

                json.dump(
                    variant_summary,
                    f,
                    indent=2,
                )

            print(
                f"  {variant_name}: "
                f"{windows.shape} saved."
            )

            del windows

        # ----------------------------------------------------
        # Recording completed
        # ----------------------------------------------------

        successful_recordings.add(
            recording_id
        )

        failed_recordings.pop(
            recording_id,
            None,
        )

        save_checkpoint()

        print(
            f"{recording_id}: SUCCESS"
        )

    except Exception as e:

        failed_recordings[
            recording_id
        ] = {
            "error": str(e),
            "traceback": traceback.format_exc(),
        }

        save_checkpoint()

        print(
            f"\n{recording_id}: FAILED"
        )

        print(
            traceback.format_exc()
        )

        # Continue with remaining recordings
        continue


# ------------------------------------------------------------
# 9. FINAL INVENTORY
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("STEP 6B FINAL INVENTORY")
print("=" * 70)

inventory = {}

for variant_name in VARIANTS:

    npy_files = list(
        OUTPUT_ROOT.glob(
            f"*/ */{variant_name}/*.npy"
        )
    )

# Robust recursive inventory
all_npy = list(
    OUTPUT_ROOT.rglob("*.npy")
)

all_csv = list(
    OUTPUT_ROOT.rglob("*.csv")
)

for variant_name in VARIANTS:

    variant_npy = [
        p
        for p in all_npy
        if f"_{variant_name}_windows.npy"
        in p.name
    ]

    variant_csv = [
        p
        for p in all_csv
        if f"_{variant_name}_metadata.csv"
        in p.name
    ]

    inventory[variant_name] = {
        "npy_count": len(variant_npy),
        "metadata_count": len(variant_csv),
    }

    print(
        f"{variant_name:8s}: "
        f"{len(variant_npy)}/{EXPECTED_RECORDINGS} NPY, "
        f"{len(variant_csv)}/{EXPECTED_RECORDINGS} CSV"
    )


# ------------------------------------------------------------
# 10. SHAPE VERIFICATION
# ------------------------------------------------------------

shape_errors = []

for variant_name, cfg in VARIANTS.items():

    expected_shape_last = (
        EXPECTED_EEG_CHANNELS,
        cfg["samples"],
    )

    for split in [
        "train",
        "val",
        "test",
    ]:

        split_dir = (
            OUTPUT_ROOT
            / split
        )

        if not split_dir.exists():
            continue

        for recording_dir in split_dir.iterdir():

            if not recording_dir.is_dir():
                continue

            recording_id = (
                recording_dir.name
            )

            npy_path = (
                recording_dir
                / f"{recording_id}_{variant_name}_windows.npy"
            )

            if not npy_path.exists():
                continue

            try:

                arr = np.load(
                    npy_path,
                    mmap_mode="r",
                )

                expected_n = len(
                    pd.read_csv(
                        PHASE5_RECORDINGS
                        / f"{recording_id}_metadata.csv"
                    )
                )

                expected_shape = (
                    expected_n,
                    EXPECTED_EEG_CHANNELS,
                    cfg["samples"],
                )

                if arr.shape != expected_shape:

                    shape_errors.append(
                        {
                            "recording_id": recording_id,
                            "variant": variant_name,
                            "expected": expected_shape,
                            "actual": arr.shape,
                        }
                    )

            except Exception as e:

                shape_errors.append(
                    {
                        "recording_id": recording_id,
                        "variant": variant_name,
                        "error": str(e),
                    }
                )


print(
    f"\nShape errors: {len(shape_errors)}"
)


# ------------------------------------------------------------
# 11. FINAL SUMMARY
# ------------------------------------------------------------

final_summary = {
    "phase": "6B",
    "status": (
        "READY_FOR_STEP_6C"
        if (
            len(successful_recordings)
            == EXPECTED_RECORDINGS
            and len(failed_recordings)
            == 0
            and len(shape_errors)
            == 0
            and all(
                inventory[v]["npy_count"]
                == EXPECTED_RECORDINGS
                and inventory[v]["metadata_count"]
                == EXPECTED_RECORDINGS
                for v in VARIANTS
            )
        )
        else
        "INCOMPLETE"
    ),
    "successful_recordings": len(
        successful_recordings
    ),
    "failed_recordings": len(
        failed_recordings
    ),
    "expected_recordings": EXPECTED_RECORDINGS,
    "inventory": inventory,
    "shape_errors": shape_errors,
    "split_counts": actual_split_counts,
    "configuration": {
        "fs": FS,
        "lowcut": LOWCUT,
        "highcut": HIGHCUT,
        "expected_eeg_channels":
            EXPECTED_EEG_CHANNELS,
        "baseline_rtol":
            BASELINE_RTOL,
        "baseline_atol":
            BASELINE_ATOL,
        "variants":
            VARIANTS,
    },
}

with open(
    SUMMARY_PATH,
    "w",
) as f:

    json.dump(
        final_summary,
        f,
        indent=2,
    )


# ------------------------------------------------------------
# 12. FINAL STATUS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print(
    f"Successful recordings: "
    f"{len(successful_recordings)}/{EXPECTED_RECORDINGS}"
)

print(
    f"Failed recordings: "
    f"{len(failed_recordings)}"
)

print(
    f"Shape errors: "
    f"{len(shape_errors)}"
)

print("\nVariant inventory:")

for variant_name, info in inventory.items():

    print(
        f"  {variant_name:8s}: "
        f"{info['npy_count']}/{EXPECTED_RECORDINGS} NPY, "
        f"{info['metadata_count']}/{EXPECTED_RECORDINGS} CSV"
    )

print(
    f"\nSummary:\n{SUMMARY_PATH}"
)

print(
    f"\nCheckpoint:\n{CHECKPOINT_PATH}"
)

if final_summary["status"] == "READY_FOR_STEP_6C":

    print("\n" + "=" * 70)
    print("STEP 6B STATUS: READY FOR STEP 6C")
    print("=" * 70)

else:

    print("\n" + "=" * 70)
    print("STEP 6B STATUS: INCOMPLETE")
    print("=" * 70)

    if failed_recordings:

        print("\nFailed recordings:")

        for recording_id, info in failed_recordings.items():

            print(
                f"  {recording_id}: "
                f"{info.get('error', 'unknown error')}"
            )

    if shape_errors:

        print("\nShape errors:")

        for error in shape_errors[:20]:

            print(error)
```

```python
# ============================================================
# NEUROSENTINEL — STEP 6C
# Temporal Window Decoder Comparison
#
# CORRECTED FOR ACTUAL STEP 6B FILE STRUCTURE
#
# Actual structure:
#   08_phase6_temporal_windows/
#       {split}/
#           {recording_id}/
#               {recording_id}_baseline_windows.npy
#               {recording_id}_baseline_metadata.csv
#               {recording_id}_W1_windows.npy
#               {recording_id}_W1_metadata.csv
#               {recording_id}_W2_windows.npy
#               {recording_id}_W2_metadata.csv
#               {recording_id}_W3_windows.npy
#               {recording_id}_W3_metadata.csv
#
# Variants:
#   baseline = [-0.2, +0.3] -> 25 samples
#   W1       = [-0.4, +0.4] -> 40 samples
#   W2       = [-0.5, +0.5] -> 50 samples
#   W3       = [-0.2, +0.5] -> 35 samples
#
# Protocol:
#   - Frozen Phase 6 subject split
#   - Exact Phase 5H.3 normalization
#   - Same 29-class CNN as Diagnostic Step 5
#   - Same class weighting
#   - Same optimizer/training settings
#   - Validation macro-F1 model selection
#   - Test evaluated only after validation selection
# ============================================================

import os
import json
import random
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from torch.utils.data import Dataset, DataLoader

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    precision_score,
    recall_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
)

warnings.filterwarnings("ignore")


# ============================================================
# 1. CONFIGURATION
# ============================================================

ROOT = Path(
    "/content/gdrive/MyDrive/NeuroSentinel"
)

WINDOW_ROOT = (
    ROOT /
    "08_phase6_temporal_windows"
)

PHASE6_MANIFEST = (
    ROOT /
    "07_phase6_subject_split" /
    "phase6_subject_split_manifest_resolved.csv"
)

if not PHASE6_MANIFEST.exists():
    PHASE6_MANIFEST = (
        ROOT /
        "07_phase6_subject_split" /
        "phase6_subject_split_manifest.csv"
    )

OUTPUT_ROOT = (
    ROOT /
    "results" /
    "phase6_temporal_decoder"
)

CHECKPOINT_ROOT = (
    ROOT /
    "checkpoints" /
    "phase6_temporal_decoder"
)

OUTPUT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

CHECKPOINT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)


# ------------------------------------------------------------
# Frozen protocol
# ------------------------------------------------------------

SEED = 42

N_CHANNELS = 61
N_CLASSES = 29

BATCH_SIZE = 256
NUM_WORKERS = 0

LR = 1e-3
WEIGHT_DECAY = 1e-4

MAX_EPOCHS = 20
PATIENCE = 5

EPS = 1e-6

VARIANTS = {
    "baseline": {
        "tmin": -0.2,
        "tmax": 0.3,
        "samples": 25,
    },
    "W1": {
        "tmin": -0.4,
        "tmax": 0.4,
        "samples": 40,
    },
    "W2": {
        "tmin": -0.5,
        "tmax": 0.5,
        "samples": 50,
    },
    "W3": {
        "tmin": -0.2,
        "tmax": 0.5,
        "samples": 35,
    },
}

DEVICE = torch.device(
    "cuda"
    if torch.cuda.is_available()
    else "cpu"
)


print("=" * 70)
print("NEUROSENTINEL — STEP 6C")
print("Temporal Window Decoder Comparison")
print("=" * 70)

print("Device:", DEVICE)

if torch.cuda.is_available():
    print(
        "GPU:",
        torch.cuda.get_device_name(0)
    )

print(
    "Phase 6 manifest:",
    PHASE6_MANIFEST
)

print(
    "Window root:",
    WINDOW_ROOT
)

print(
    "Output root:",
    OUTPUT_ROOT
)

print()


# ============================================================
# 2. REPRODUCIBILITY
# ============================================================

def set_seed(seed=42):

    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


set_seed(SEED)


# ============================================================
# 3. LOAD FROZEN PHASE 6 SPLIT
# ============================================================

phase6 = pd.read_csv(
    PHASE6_MANIFEST
)

required_phase6 = {
    "recording_id",
    "split",
    "subject",
}

missing = (
    required_phase6 -
    set(phase6.columns)
)

if missing:
    raise RuntimeError(
        "Phase 6 manifest missing columns: "
        f"{sorted(missing)}"
    )


phase6["recording_id"] = (
    phase6["recording_id"]
    .astype(str)
)

phase6["split"] = (
    phase6["split"]
    .astype(str)
    .str.lower()
)


split_map = (
    phase6[
        [
            "recording_id",
            "split",
            "subject",
        ]
    ]
    .drop_duplicates(
        "recording_id"
    )
    .set_index(
        "recording_id"
    )
)


print("Frozen split:")
print(
    split_map["split"]
    .value_counts()
    .sort_index()
)

print()


EXPECTED_SPLITS = {
    "train",
    "val",
    "test",
}

if (
    set(split_map["split"].unique())
    != EXPECTED_SPLITS
):
    raise RuntimeError(
        "Unexpected split values: "
        f"{sorted(split_map['split'].unique())}"
    )


if len(split_map) != 62:
    raise RuntimeError(
        f"Expected 62 recordings, "
        f"found {len(split_map)}"
    )


# ============================================================
# 4. LABEL MAPPING
# ============================================================

IDX_TO_LABEL = {
    0: "a",
    1: "b",
    2: "c",
    3: "d",
    4: "e",
    5: "f",
    6: "g",
    7: "h",
    8: "i",
    9: "j",
    10: "k",
    11: "l",
    12: "m",
    13: "n",
    14: "o",
    15: "p",
    16: "q",
    17: "r",
    18: "s",
    19: "t",
    20: "u",
    21: "v",
    22: "w",
    23: "x",
    24: "y",
    25: "z",
    26: "<space>",
    27: "<number>",
    28: "<special>",
}

LABEL_TO_IDX = {
    v: k
    for k, v in IDX_TO_LABEL.items()
}


# ============================================================
# 5. CORRECT STEP 6B FILE DISCOVERY
# ============================================================

def find_recording_files(
    variant,
    recording_id
):

    split = split_map.loc[
        recording_id,
        "split"
    ]

    recording_dir = (
        WINDOW_ROOT /
        split /
        recording_id
    )

    npy_path = (
        recording_dir /
        f"{recording_id}_{variant}_windows.npy"
    )

    csv_path = (
        recording_dir /
        f"{recording_id}_{variant}_metadata.csv"
    )

    return npy_path, csv_path


# ============================================================
# 6. VERIFY COMPLETE INVENTORY
# ============================================================

print("=" * 70)
print("STEP 6C INPUT INVENTORY")
print("=" * 70)

for variant in VARIANTS:

    npy_count = 0
    csv_count = 0
    missing_npy = []
    missing_csv = []

    for recording_id in split_map.index:

        npy_path, csv_path = (
            find_recording_files(
                variant,
                recording_id
            )
        )

        if npy_path.exists():
            npy_count += 1
        else:
            missing_npy.append(
                recording_id
            )

        if csv_path.exists():
            csv_count += 1
        else:
            missing_csv.append(
                recording_id
            )

    print(
        f"{variant:8s}: "
        f"{npy_count}/62 NPY, "
        f"{csv_count}/62 CSV"
    )

    if missing_npy:
        raise RuntimeError(
            f"{variant}: missing NPY files: "
            f"{missing_npy}"
        )

    if missing_csv:
        raise RuntimeError(
            f"{variant}: missing CSV files: "
            f"{missing_csv}"
        )


print()
print(
    "All 248 NPY and 248 CSV files verified."
)
print()


# ============================================================
# 7. DATASET
# ============================================================

class TemporalEEGDataset(Dataset):

    def __init__(
        self,
        records,
        expected_samples,
    ):

        self.records = records

        self.expected_samples = (
            expected_samples
        )

        self.index = []

        for rec_idx, rec in enumerate(
            records
        ):

            metadata = pd.read_csv(
                rec["csv"]
            )

            if "target" not in metadata.columns:
                raise RuntimeError(
                    "Missing target column: "
                    f"{rec['csv']}"
                )

            metadata = (
                metadata
                .reset_index(drop=True)
            )

            windows = rec["windows"]

            if len(metadata) != len(windows):
                raise RuntimeError(
                    f"{rec['recording_id']}: "
                    f"metadata rows "
                    f"{len(metadata)} != "
                    f"windows "
                    f"{len(windows)}"
                )

            for i, target in enumerate(
                metadata["target"].astype(str)
            ):

                if target in LABEL_TO_IDX:

                    self.index.append(
                        (
                            rec_idx,
                            i,
                            LABEL_TO_IDX[target],
                        )
                    )

        print(
            f"Dataset: "
            f"{len(self.index):,} valid samples "
            f"from "
            f"{len(self.records)} recordings"
        )

    def __len__(self):
        return len(self.index)

    def __getitem__(self, idx):

        rec_idx, window_idx, label = (
            self.index[idx]
        )

        rec = self.records[rec_idx]

        x = rec["windows"][
            window_idx
        ]

        expected_shape = (
            N_CHANNELS,
            self.expected_samples,
        )

        if x.shape != expected_shape:
            raise RuntimeError(
                f"Unexpected window shape: "
                f"{x.shape}; expected "
                f"{expected_shape}"
            )

        # ----------------------------------------------------
        # Exact Phase 5H.3 normalization
        #
        # Channel-wise z-score across time
        # ----------------------------------------------------

        x = np.asarray(
            x,
            dtype=np.float32
        )

        mean = x.mean(
            axis=1,
            keepdims=True
        )

        std = x.std(
            axis=1,
            keepdims=True
        )

        x = (
            x - mean
        ) / (
            std + EPS
        )

        x = np.asarray(
            x,
            dtype=np.float32
        )

        return (
            torch.from_numpy(x),
            torch.tensor(
                label,
                dtype=torch.long
            ),
        )


# ============================================================
# 8. MODEL
# ============================================================

class TemporalCharacterCNN(
    nn.Module
):

    def __init__(self):

        super().__init__()

        self.features = nn.Sequential(

            nn.Conv1d(
                N_CHANNELS,
                64,
                kernel_size=5,
                padding=2,
            ),
            nn.ReLU(),

            nn.Conv1d(
                64,
                64,
                kernel_size=5,
                dilation=2,
                padding=4,
            ),
            nn.ReLU(),

            nn.Conv1d(
                64,
                64,
                kernel_size=5,
                dilation=4,
                padding=8,
            ),
            nn.ReLU(),

            nn.Conv1d(
                64,
                64,
                kernel_size=5,
                dilation=8,
                padding=16,
            ),
            nn.ReLU(),
        )

        self.classifier = nn.Linear(
            64,
            N_CLASSES
        )

    def forward(self, x):

        x = self.features(x)

        x = x.mean(
            dim=2
        )

        return self.classifier(x)


# Parameter verification
_tmp_model = TemporalCharacterCNN()

PARAM_COUNT = sum(
    p.numel()
    for p in _tmp_model.parameters()
    if p.requires_grad
)

del _tmp_model

print(
    f"Decoder trainable parameters: "
    f"{PARAM_COUNT:,}"
)

if PARAM_COUNT != 83101:

    raise RuntimeError(
        "Architecture parameter count "
        f"mismatch: {PARAM_COUNT}; "
        "expected 83101"
    )

print()


# ============================================================
# 9. BUILD RECORDS
# ============================================================

def build_records(
    variant
):

    records = []

    expected_samples = (
        VARIANTS[
            variant
        ]["samples"]
    )

    for recording_id in split_map.index:

        split = split_map.loc[
            recording_id,
            "split"
        ]

        npy_path, csv_path = (
            find_recording_files(
                variant,
                recording_id
            )
        )

        windows = np.load(
            npy_path,
            mmap_mode="r"
        )

        if windows.ndim != 3:
            raise RuntimeError(
                f"{recording_id}: "
                f"unexpected array shape "
                f"{windows.shape}"
            )

        if windows.shape[1] != N_CHANNELS:
            raise RuntimeError(
                f"{recording_id}: expected "
                f"{N_CHANNELS} channels, "
                f"got {windows.shape[1]}"
            )

        if windows.shape[2] != expected_samples:
            raise RuntimeError(
                f"{recording_id}: expected "
                f"{expected_samples} samples, "
                f"got {windows.shape[2]}"
            )

        records.append({
            "recording_id":
                recording_id,

            "split":
                split,

            "npy":
                npy_path,

            "csv":
                csv_path,

            "windows":
                windows,
        })

    return records


# ============================================================
# 10. CLASS WEIGHTS
# ============================================================

def compute_class_weights(
    dataset
):

    counts = np.zeros(
        N_CLASSES,
        dtype=np.float64
    )

    for _, _, label in dataset.index:
        counts[label] += 1

    weights = np.zeros_like(
        counts
    )

    present = counts > 0

    weights[present] = (
        1.0 /
        np.sqrt(
            counts[present]
        )
    )

    # Normalize present-class mean to 1.
    weights[present] /= (
        weights[present].mean()
    )

    return (
        torch.tensor(
            weights,
            dtype=torch.float32
        ),
        counts,
    )


# ============================================================
# 11. METRICS
# ============================================================

def calculate_metrics(
    y_true,
    y_pred,
    probabilities
):

    result = {}

    result["accuracy"] = float(
        accuracy_score(
            y_true,
            y_pred
        )
    )

    result["balanced_accuracy"] = float(
        balanced_accuracy_score(
            y_true,
            y_pred
        )
    )

    result["macro_precision"] = float(
        precision_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        )
    )

    result["macro_recall"] = float(
        recall_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        )
    )

    result["macro_f1"] = float(
        f1_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        )
    )

    result["weighted_f1"] = float(
        f1_score(
            y_true,
            y_pred,
            average="weighted",
            zero_division=0
        )
    )

    y_true_onehot = np.eye(
        N_CLASSES
    )[y_true]

    aucs = []
    aps = []

    for c in range(N_CLASSES):

        yc = y_true_onehot[:, c]

        if yc.min() < yc.max():

            try:
                aucs.append(
                    roc_auc_score(
                        yc,
                        probabilities[:, c]
                    )
                )
            except Exception:
                pass

            try:
                aps.append(
                    average_precision_score(
                        yc,
                        probabilities[:, c]
                    )
                )
            except Exception:
                pass

    result["macro_auroc"] = (
        float(np.mean(aucs))
        if aucs
        else float("nan")
    )

    result["macro_auprc"] = (
        float(np.mean(aps))
        if aps
        else float("nan")
    )

    return result


# ============================================================
# 12. EVALUATION
# ============================================================

@torch.no_grad()
def evaluate(
    model,
    loader,
    criterion
):

    model.eval()

    total_loss = 0.0
    total_n = 0

    all_true = []
    all_pred = []
    all_prob = []

    for x, y in loader:

        x = x.to(
            DEVICE,
            non_blocking=True
        )

        y = y.to(
            DEVICE,
            non_blocking=True
        )

        logits = model(x)

        loss = criterion(
            logits,
            y
        )

        probs = torch.softmax(
            logits,
            dim=1
        )

        preds = torch.argmax(
            probs,
            dim=1
        )

        batch_n = y.size(0)

        total_loss += (
            loss.item() *
            batch_n
        )

        total_n += batch_n

        all_true.append(
            y.cpu().numpy()
        )

        all_pred.append(
            preds.cpu().numpy()
        )

        all_prob.append(
            probs.cpu().numpy()
        )

    y_true = np.concatenate(
        all_true
    )

    y_pred = np.concatenate(
        all_pred
    )

    probabilities = np.concatenate(
        all_prob
    )

    metrics = calculate_metrics(
        y_true,
        y_pred,
        probabilities
    )

    metrics["loss"] = (
        total_loss /
        total_n
    )

    return (
        metrics,
        y_true,
        y_pred,
        probabilities
    )


# ============================================================
# 13. TRAIN ONE VARIANT
# ============================================================

def train_variant(
    variant
):

    print()
    print("=" * 70)
    print(
        f"TRAINING VARIANT: {variant}"
    )
    print("=" * 70)

    cfg = VARIANTS[
        variant
    ]

    print(
        f"Window: "
        f"[{cfg['tmin']}, "
        f"{cfg['tmax']}] s"
    )

    print(
        f"Samples: "
        f"{cfg['samples']}"
    )

    # --------------------------------------------------------
    # Records
    # --------------------------------------------------------

    records = build_records(
        variant
    )

    train_records = [
        r for r in records
        if r["split"] == "train"
    ]

    val_records = [
        r for r in records
        if r["split"] == "val"
    ]

    test_records = [
        r for r in records
        if r["split"] == "test"
    ]

    print(
        f"Recordings: "
        f"train={len(train_records)}, "
        f"val={len(val_records)}, "
        f"test={len(test_records)}"
    )

    # --------------------------------------------------------
    # Datasets
    # --------------------------------------------------------

    train_ds = TemporalEEGDataset(
        train_records,
        cfg["samples"]
    )

    val_ds = TemporalEEGDataset(
        val_records,
        cfg["samples"]
    )

    test_ds = TemporalEEGDataset(
        test_records,
        cfg["samples"]
    )

    # --------------------------------------------------------
    # Class weights
    # --------------------------------------------------------

    class_weights, class_counts = (
        compute_class_weights(
            train_ds
        )
    )

    print()
    print(
        "Train class counts:"
    )

    print(
        pd.Series(
            class_counts,
            index=[
                IDX_TO_LABEL[i]
                for i in range(N_CLASSES)
            ]
        )
    )

    print()
    print(
        "Class weights:"
    )

    print(
        pd.Series(
            class_weights.numpy(),
            index=[
                IDX_TO_LABEL[i]
                for i in range(N_CLASSES)
            ]
        )
    )

    # --------------------------------------------------------
    # DataLoaders
    # --------------------------------------------------------

    train_loader = DataLoader(
        train_ds,
        batch_size=BATCH_SIZE,
        shuffle=True,
        num_workers=NUM_WORKERS,
        pin_memory=torch.cuda.is_available(),
        drop_last=False
    )

    val_loader = DataLoader(
        val_ds,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=NUM_WORKERS,
        pin_memory=torch.cuda.is_available(),
        drop_last=False
    )

    test_loader = DataLoader(
        test_ds,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=NUM_WORKERS,
        pin_memory=torch.cuda.is_available(),
        drop_last=False
    )

    # --------------------------------------------------------
    # Model
    # --------------------------------------------------------

    set_seed(SEED)

    model = (
        TemporalCharacterCNN()
        .to(DEVICE)
    )

    criterion = nn.CrossEntropyLoss(
        weight=class_weights.to(
            DEVICE
        )
    )

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=LR,
        weight_decay=WEIGHT_DECAY
    )

    # --------------------------------------------------------
    # Training
    # --------------------------------------------------------

    best_val_f1 = -np.inf
    best_epoch = None
    best_state = None

    patience_counter = 0

    history = []

    for epoch in range(
        1,
        MAX_EPOCHS + 1
    ):

        model.train()

        train_loss_sum = 0.0
        train_n = 0

        train_true = []
        train_pred = []

        for x, y in train_loader:

            x = x.to(
                DEVICE,
                non_blocking=True
            )

            y = y.to(
                DEVICE,
                non_blocking=True
            )

            optimizer.zero_grad(
                set_to_none=True
            )

            logits = model(x)

            loss = criterion(
                logits,
                y
            )

            loss.backward()

            optimizer.step()

            batch_n = y.size(0)

            train_loss_sum += (
                loss.item() *
                batch_n
            )

            train_n += batch_n

            preds = torch.argmax(
                logits,
                dim=1
            )

            train_true.append(
                y.detach()
                .cpu()
                .numpy()
            )

            train_pred.append(
                preds.detach()
                .cpu()
                .numpy()
            )

        train_true = np.concatenate(
            train_true
        )

        train_pred = np.concatenate(
            train_pred
        )

        train_loss = (
            train_loss_sum /
            train_n
        )

        train_acc = (
            accuracy_score(
                train_true,
                train_pred
            )
        )

        # ----------------------------------------------------
        # Validation
        # ----------------------------------------------------

        val_metrics, _, _, _ = (
            evaluate(
                model,
                val_loader,
                criterion
            )
        )

        row = {
            "variant":
                variant,

            "epoch":
                epoch,

            "train_loss":
                train_loss,

            "train_accuracy":
                train_acc,

            "val_loss":
                val_metrics["loss"],

            "val_accuracy":
                val_metrics["accuracy"],

            "val_balanced_accuracy":
                val_metrics[
                    "balanced_accuracy"
                ],

            "val_macro_f1":
                val_metrics[
                    "macro_f1"
                ],

            "val_weighted_f1":
                val_metrics[
                    "weighted_f1"
                ],

            "val_macro_auroc":
                val_metrics[
                    "macro_auroc"
                ],

            "val_macro_auprc":
                val_metrics[
                    "macro_auprc"
                ],
        }

        history.append(
            row
        )

        print(
            f"Epoch {epoch:02d} | "
            f"train loss={train_loss:.4f} | "
            f"train acc={train_acc:.4f} | "
            f"val loss={val_metrics['loss']:.4f} | "
            f"val acc={val_metrics['accuracy']:.4f} | "
            f"val bal_acc={val_metrics['balanced_accuracy']:.4f} | "
            f"val macro-F1={val_metrics['macro_f1']:.4f}"
        )

        # ----------------------------------------------------
        # Validation-only model selection
        # ----------------------------------------------------

        current_f1 = (
            val_metrics[
                "macro_f1"
            ]
        )

        if current_f1 > best_val_f1:

            best_val_f1 = (
                current_f1
            )

            best_epoch = epoch

            patience_counter = 0

            best_state = {
                k: v.detach()
                .cpu()
                .clone()
                for k, v in
                model.state_dict()
                .items()
            }

        else:

            patience_counter += 1

            if (
                patience_counter
                >= PATIENCE
            ):

                print(
                    "Early stopping at "
                    f"epoch {epoch}."
                )

                break

    # --------------------------------------------------------
    # Restore best validation model
    # --------------------------------------------------------

    if best_state is None:
        raise RuntimeError(
            f"{variant}: no best model saved."
        )

    model.load_state_dict(
        best_state
    )

    checkpoint_path = (
        CHECKPOINT_ROOT /
        f"{variant}_best.pt"
    )

    torch.save(
        {
            "variant":
                variant,

            "best_epoch":
                best_epoch,

            "best_val_macro_f1":
                best_val_f1,

            "model_state_dict":
                best_state,

            "config":
                cfg,

            "seed":
                SEED,

            "n_classes":
                N_CLASSES,

            "n_channels":
                N_CHANNELS,

            "samples":
                cfg["samples"],
        },
        checkpoint_path
    )

    # --------------------------------------------------------
    # Final validation
    # --------------------------------------------------------

    final_val_metrics, _, _, _ = (
        evaluate(
            model,
            val_loader,
            criterion
        )
    )

    # --------------------------------------------------------
    # Frozen test evaluation
    # --------------------------------------------------------

    test_metrics, y_true, y_pred, probabilities = (
        evaluate(
            model,
            test_loader,
            criterion
        )
    )

    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=np.arange(
            N_CLASSES
        )
    )

    # --------------------------------------------------------
    # Test predictions
    # --------------------------------------------------------

    prediction_df = pd.DataFrame({

        "true_class":
            y_true,

        "true_label": [
            IDX_TO_LABEL[
                int(x)
            ]
            for x in y_true
        ],

        "predicted_class":
            y_pred,

        "predicted_label": [
            IDX_TO_LABEL[
                int(x)
            ]
            for x in y_pred
        ],

        "confidence":
            probabilities.max(
                axis=1
            ),
    })

    for c in range(
        N_CLASSES
    ):

        prediction_df[
            f"prob_{c}"
        ] = probabilities[:, c]

    prediction_path = (
        OUTPUT_ROOT /
        f"{variant}_test_predictions.csv"
    )

    prediction_df.to_csv(
        prediction_path,
        index=False
    )

    # --------------------------------------------------------
    # Training history
    # --------------------------------------------------------

    history_df = pd.DataFrame(
        history
    )

    history_path = (
        OUTPUT_ROOT /
        f"{variant}_training_history.csv"
    )

    history_df.to_csv(
        history_path,
        index=False
    )

    # --------------------------------------------------------
    # Result
    # --------------------------------------------------------

    result = {

        "variant":
            variant,

        "window_tmin":
            cfg["tmin"],

        "window_tmax":
            cfg["tmax"],

        "window_samples":
            cfg["samples"],

        "best_epoch":
            int(best_epoch),

        "best_validation_macro_f1":
            float(best_val_f1),

        "validation": {
            k: (
                None
                if not np.isfinite(v)
                else float(v)
            )
            for k, v
            in final_val_metrics.items()
        },

        "test": {
            k: (
                None
                if not np.isfinite(v)
                else float(v)
            )
            for k, v
            in test_metrics.items()
        },

        "confusion_matrix":
            cm.tolist(),

        "train_samples":
            len(train_ds),

        "val_samples":
            len(val_ds),

        "test_samples":
            len(test_ds),

        "train_recordings":
            len(train_records),

        "val_recordings":
            len(val_records),

        "test_recordings":
            len(test_records),

        "parameter_count":
            int(PARAM_COUNT),

        "checkpoint":
            str(checkpoint_path),

        "history":
            str(history_path),

        "test_predictions":
            str(prediction_path),
    }

    result_path = (
        OUTPUT_ROOT /
        f"{variant}_results.json"
    )

    with open(
        result_path,
        "w"
    ) as f:

        json.dump(
            result,
            f,
            indent=2
        )

    # --------------------------------------------------------
    # Print result
    # --------------------------------------------------------

    print()
    print("-" * 70)
    print(
        f"{variant} FINAL RESULT"
    )
    print("-" * 70)

    print(
        f"Best epoch: "
        f"{best_epoch}"
    )

    print(
        f"Best validation macro-F1: "
        f"{best_val_f1:.6f}"
    )

    print()
    print("Validation:")

    print(
        f"  Accuracy:          "
        f"{final_val_metrics['accuracy']:.6f}"
    )

    print(
        f"  Balanced accuracy: "
        f"{final_val_metrics['balanced_accuracy']:.6f}"
    )

    print(
        f"  Macro F1:           "
        f"{final_val_metrics['macro_f1']:.6f}"
    )

    print(
        f"  Macro AUROC:        "
        f"{final_val_metrics['macro_auroc']:.6f}"
    )

    print(
        f"  Macro AUPRC:        "
        f"{final_val_metrics['macro_auprc']:.6f}"
    )

    print()
    print("Test:")

    print(
        f"  Accuracy:          "
        f"{test_metrics['accuracy']:.6f}"
    )

    print(
        f"  Balanced accuracy: "
        f"{test_metrics['balanced_accuracy']:.6f}"
    )

    print(
        f"  Macro F1:           "
        f"{test_metrics['macro_f1']:.6f}"
    )

    print(
        f"  Macro AUROC:        "
        f"{test_metrics['macro_auroc']:.6f}"
    )

    print(
        f"  Macro AUPRC:        "
        f"{test_metrics['macro_auprc']:.6f}"
    )

    return result


# ============================================================
# 14. TRAIN ALL FOUR VARIANTS
# ============================================================

all_results = []

for variant in [
    "baseline",
    "W1",
    "W2",
    "W3",
]:

    result = train_variant(
        variant
    )

    all_results.append(
        result
    )

    if torch.cuda.is_available():
        torch.cuda.empty_cache()


# ============================================================
# 15. COMPARISON TABLE
# ============================================================

comparison_rows = []

for r in all_results:

    comparison_rows.append({

        "variant":
            r["variant"],

        "window":
            (
                f"[{r['window_tmin']}, "
                f"{r['window_tmax']}]"
            ),

        "samples":
            r["window_samples"],

        "best_epoch":
            r["best_epoch"],

        "val_accuracy":
            r["validation"][
                "accuracy"
            ],

        "val_balanced_accuracy":
            r["validation"][
                "balanced_accuracy"
            ],

        "val_macro_f1":
            r["validation"][
                "macro_f1"
            ],

        "val_macro_auroc":
            r["validation"][
                "macro_auroc"
            ],

        "val_macro_auprc":
            r["validation"][
                "macro_auprc"
            ],

        "test_accuracy":
            r["test"][
                "accuracy"
            ],

        "test_balanced_accuracy":
            r["test"][
                "balanced_accuracy"
            ],

        "test_macro_f1":
            r["test"][
                "macro_f1"
            ],

        "test_macro_auroc":
            r["test"][
                "macro_auroc"
            ],

        "test_macro_auprc":
            r["test"][
                "macro_auprc"
            ],
    })


comparison_df = pd.DataFrame(
    comparison_rows
)

comparison_path = (
    OUTPUT_ROOT /
    "phase6_temporal_decoder_comparison.csv"
)

comparison_df.to_csv(
    comparison_path,
    index=False
)


# ============================================================
# 16. BEST VARIANT
# ============================================================

best_idx = comparison_df[
    "val_macro_f1"
].idxmax()

best_variant = comparison_df.loc[
    best_idx,
    "variant"
]

best_val_f1 = comparison_df.loc[
    best_idx,
    "val_macro_f1"
]

best_test_f1 = comparison_df.loc[
    best_idx,
    "test_macro_f1"
]


# ============================================================
# 17. FINAL SUMMARY
# ============================================================

final_summary = {

    "step":
        "6C",

    "status":
        "COMPLETED",

    "protocol": {

        "seed":
            SEED,

        "batch_size":
            BATCH_SIZE,

        "learning_rate":
            LR,

        "weight_decay":
            WEIGHT_DECAY,

        "max_epochs":
            MAX_EPOCHS,

        "patience":
            PATIENCE,

        "normalization":
            "per-window channel-wise z-score",

        "model_selection":
            "validation macro-F1",

        "test_used_for_selection":
            False,
    },

    "best_variant_by_validation_macro_f1":
        str(best_variant),

    "best_validation_macro_f1":
        float(best_val_f1),

    "best_variant_test_macro_f1":
        float(best_test_f1),

    "results":
        all_results,

    "comparison_csv":
        str(comparison_path),
}


summary_path = (
    OUTPUT_ROOT /
    "phase6_temporal_decoder_summary.json"
)

with open(
    summary_path,
    "w"
) as f:

    json.dump(
        final_summary,
        f,
        indent=2
    )


# ============================================================
# 18. FINAL COMPARISON
# ============================================================

print()
print()
print("=" * 90)
print(
    "STEP 6C — FINAL TEMPORAL WINDOW COMPARISON"
)
print("=" * 90)

display_df = (
    comparison_df.copy()
)

metric_columns = [
    "val_accuracy",
    "val_balanced_accuracy",
    "val_macro_f1",
    "val_macro_auroc",
    "val_macro_auprc",
    "test_accuracy",
    "test_balanced_accuracy",
    "test_macro_f1",
    "test_macro_auroc",
    "test_macro_auprc",
]

for col in metric_columns:

    display_df[col] = (
        display_df[col]
        .round(6)
    )


print(
    display_df.to_string(
        index=False
    )
)

print()
print("=" * 90)

print(
    "BEST WINDOW BY "
    "VALIDATION MACRO-F1: "
    f"{best_variant}"
)

print(
    f"Best validation macro-F1: "
    f"{best_val_f1:.6f}"
)

print(
    f"Corresponding test macro-F1: "
    f"{best_test_f1:.6f}"
)

print("=" * 90)

print()
print(
    "Comparison CSV:"
)

print(
    comparison_path
)

print()
print(
    "Summary JSON:"
)

print(
    summary_path
)

print()
print("=" * 90)
print(
    "STEP 6C STATUS: COMPLETED"
)
print("=" *  90)
```

```python
# ============================================================
# NEUROSENTINEL — STEP 6D
# Subject-Independent EEG Representation Diagnosis / Adaptation
#
# Purpose:
#   Diagnose whether subject-specific EEG scale/distribution
#   differences are limiting cross-subject character decoding.
#
# Protocol:
#   - Frozen Phase 6 subject split
#   - Train subjects: training only
#   - Validation subjects: model selection only
#   - Test subjects: evaluated ONCE after model selection
#   - Same preprocessing as Phase 5H.3
#   - Same baseline temporal window: [-0.2, +0.3] s
#   - No test-based selection
#
# Experiments:
#   A. Global channel normalization
#   B. Per-window channel-wise z-score (Phase 5H.3 baseline)
#   C. Subject-wise channel normalization
#   D. Subject-wise channel normalization + per-window z-score
#
# Decoder:
#   Lightweight 29-class CNN
#
# Selection:
#   Best validation macro-F1
#
# Test:
#   Evaluated only after the best validation configuration is frozen.
# ============================================================

import os
import json
import random
import warnings
from pathlib import Path

import numpy as np
import pandas as pd

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    precision_score,
    recall_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
)

warnings.filterwarnings("ignore")

# ============================================================
# 0. CONFIGURATION
# ============================================================

SEED = 42

ROOT = Path("/content/gdrive/MyDrive/NeuroSentinel")

SPLIT_MANIFEST = (
    ROOT / "07_phase6_subject_split" /
    "phase6_subject_split_manifest_resolved.csv"
)

if not SPLIT_MANIFEST.exists():
    SPLIT_MANIFEST = (
        ROOT / "07_phase6_subject_split" /
        "phase6_subject_split_manifest.csv"
    )

WINDOW_ROOT = ROOT / "08_phase6_temporal_windows"

H7_MANIFEST = (
    ROOT / "results" / "phase5h7" /
    "phase5h7_utility_label_manifest.csv"
)

OUTPUT_ROOT = (
    ROOT / "results" / "phase6_representation_diagnosis"
)

CHECKPOINT_ROOT = (
    ROOT / "checkpoints" / "phase6_representation_diagnosis"
)

OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)
CHECKPOINT_ROOT.mkdir(parents=True, exist_ok=True)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

N_CHANNELS = 61
N_CLASSES = 29
WINDOW_SAMPLES = 25

BATCH_SIZE = 256
LR = 1e-3
WEIGHT_DECAY = 1e-4

MAX_EPOCHS = 20
PATIENCE = 5

EPS = 1e-6

# Baseline temporal window from Phase 5H.3
VARIANT = "baseline"

print("=" * 70)
print("NEUROSENTINEL — STEP 6D")
print("Subject-Independent EEG Representation Diagnosis")
print("=" * 70)

print(f"Device: {DEVICE}")

if torch.cuda.is_available():
    print(f"GPU: {torch.cuda.get_device_name(0)}")

# ============================================================
# 1. REPRODUCIBILITY
# ============================================================

def seed_everything(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

seed_everything(SEED)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False

# ============================================================
# 2. LOAD PHASE 6 SUBJECT SPLIT
# ============================================================

split_df = pd.read_csv(SPLIT_MANIFEST)

required_split_cols = [
    "recording_id",
    "subject",
    "session",
    "task",
    "split",
]

missing = [c for c in required_split_cols if c not in split_df.columns]

if missing:
    raise RuntimeError(
        f"Phase 6 manifest missing columns: {missing}"
    )

split_df["recording_id"] = split_df["recording_id"].astype(str)
split_df["subject"] = split_df["subject"].astype(int)
split_df["split"] = split_df["split"].astype(str).str.lower()

print("\nFrozen subject split:")
print(
    split_df.groupby("split")["subject"]
    .unique()
    .to_string()
)

train_recordings = split_df.loc[
    split_df["split"] == "train", "recording_id"
].tolist()

val_recordings = split_df.loc[
    split_df["split"] == "val", "recording_id"
].tolist()

test_recordings = split_df.loc[
    split_df["split"] == "test", "recording_id"
].tolist()

print(f"\nTrain recordings: {len(train_recordings)}")
print(f"Val recordings:   {len(val_recordings)}")
print(f"Test recordings:  {len(test_recordings)}")

# ============================================================
# 3. LOAD H7 MANIFEST
# ============================================================

h7 = pd.read_csv(H7_MANIFEST)

required_h7_cols = [
    "recording_id",
    "split",
    "subject",
    "window_index",
    "true_class",
    "true_label",
]

missing = [c for c in required_h7_cols if c not in h7.columns]

if missing:
    raise RuntimeError(
        f"Phase 5H.7 manifest missing columns: {missing}"
    )

h7["recording_id"] = h7["recording_id"].astype(str)
h7["window_index"] = h7["window_index"].astype(int)
h7["true_class"] = h7["true_class"].astype(int)
h7["subject"] = h7["subject"].astype(int)

# Ensure labels are valid
if not h7["true_class"].between(0, 28).all():
    raise RuntimeError("Invalid class labels detected.")

# ============================================================
# 4. VERIFY FROZEN SPLIT
# ============================================================

recording_to_split = dict(
    zip(split_df["recording_id"], split_df["split"])
)

h7["expected_split"] = h7["recording_id"].map(recording_to_split)

if h7["expected_split"].isna().any():
    raise RuntimeError(
        "H7 manifest contains recordings missing from Phase 6 split."
    )

split_mismatch = (
    h7["split"].astype(str).str.lower()
    != h7["expected_split"].astype(str).str.lower()
)

if split_mismatch.any():
    print(
        h7.loc[
            split_mismatch,
            ["recording_id", "split", "expected_split"]
        ].head()
    )
    raise RuntimeError(
        f"Found {split_mismatch.sum()} split mismatches."
    )

print("\nH7 split verification: PASSED")

# ============================================================
# 5. FIND WINDOW FILES
# ============================================================

def find_window_file(recording_id, variant):
    """
    Actual Step 6C filesystem structure:
      .../{split}/{recording_id}/{recording_id}_{variant}_windows.npy
    """

    split = recording_to_split[recording_id]

    candidates = [
        WINDOW_ROOT / split / recording_id /
        f"{recording_id}_{variant}_windows.npy",

        WINDOW_ROOT / split / recording_id /
        f"{variant}_windows.npy",

        WINDOW_ROOT / recording_id /
        f"{recording_id}_{variant}_windows.npy",

        WINDOW_ROOT / recording_id /
        f"{variant}_windows.npy",
    ]

    for p in candidates:
        if p.exists():
            return p

    # Recursive fallback
    matches = list(
        WINDOW_ROOT.rglob(
            f"{recording_id}_{variant}_windows.npy"
        )
    )

    if matches:
        return matches[0]

    raise FileNotFoundError(
        f"Could not find window file for "
        f"{recording_id}, variant={variant}"
    )

# ============================================================
# 6. BUILD RECORDING INDEX
# ============================================================

recording_index = {}

for recording_id in sorted(h7["recording_id"].unique()):

    window_path = find_window_file(
        recording_id,
        VARIANT
    )

    arr = np.load(
        window_path,
        mmap_mode="r"
    )

    if arr.ndim != 3:
        raise RuntimeError(
            f"{recording_id}: expected 3D array, got {arr.shape}"
        )

    if arr.shape[1] != N_CHANNELS:
        raise RuntimeError(
            f"{recording_id}: expected {N_CHANNELS} channels, "
            f"got {arr.shape[1]}"
        )

    if arr.shape[2] != WINDOW_SAMPLES:
        raise RuntimeError(
            f"{recording_id}: expected {WINDOW_SAMPLES} samples, "
            f"got {arr.shape[2]}"
        )

    recording_index[recording_id] = {
        "path": window_path,
        "shape": tuple(arr.shape),
        "subject": int(
            split_df.loc[
                split_df["recording_id"] == recording_id,
                "subject"
            ].iloc[0]
        ),
        "split": recording_to_split[recording_id],
    }

print(
    f"\nVerified {len(recording_index)}/"
    f"{h7['recording_id'].nunique()} recordings."
)

# ============================================================
# 7. SUBJECT / SPLIT INDEX
# ============================================================

subject_to_recordings = {}

for rid, info in recording_index.items():

    subject = info["subject"]

    subject_to_recordings.setdefault(
        subject, []
    ).append(rid)

# Verify each subject belongs to exactly one split
subject_split = {}

for rid, info in recording_index.items():

    subject = info["subject"]
    split = info["split"]

    if subject in subject_split:
        if subject_split[subject] != split:
            raise RuntimeError(
                f"Subject {subject} appears in multiple splits."
            )
    else:
        subject_split[subject] = split

print("\nSubjects by split:")

for split_name in ["train", "val", "test"]:

    subjects = sorted(
        s for s, sp in subject_split.items()
        if sp == split_name
    )

    print(
        f"  {split_name}: "
        f"{subjects}"
    )

# ============================================================
# 8. CREATE SAMPLE TABLES
# ============================================================

# Keep only rows corresponding to the frozen decoder samples.
# The H7 manifest already contains the valid 150,950 samples.

train_df = h7[h7["split"].str.lower() == "train"].copy()
val_df = h7[h7["split"].str.lower() == "val"].copy()
test_df = h7[h7["split"].str.lower() == "test"].copy()

print("\nSample counts:")
print(f"  Train: {len(train_df):,}")
print(f"  Val:   {len(val_df):,}")
print(f"  Test:  {len(test_df):,}")

# ============================================================
# 9. NORMALIZATION METHODS
# ============================================================
#
# A: GLOBAL_CHANNEL
#    Estimate channel mean/std from TRAIN subjects only.
#
# B: WINDOW_Z
#    Exact Phase 5H.3 normalization:
#    each window independently, channel-wise over time.
#
# C: SUBJECT_CHANNEL
#    Estimate per-subject channel mean/std using each subject's
#    available windows.
#
# D: SUBJECT_CHANNEL_WINDOW_Z
#    Subject normalization followed by per-window channel z-score.
#
# Important:
#    For validation/test subjects, statistics are estimated ONLY
#    from that subject's unlabeled EEG windows.
#
# This is unsupervised test-time normalization, not label-based
# adaptation. No labels are used to estimate normalization stats.
# ============================================================

def compute_channel_stats_from_dataframe(
    df,
    use_subject=None,
    max_samples_per_recording=5000
):
    """
    Estimate channel mean/std.

    Statistics are computed from raw EEG windows.

    Returns:
        mean: [61]
        std:  [61]
    """

    if use_subject is not None:
        selected = df[
            df["subject"] == use_subject
        ]
    else:
        selected = df

    if len(selected) == 0:
        raise RuntimeError(
            f"No samples available for subject={use_subject}"
        )

    sum_x = np.zeros(N_CHANNELS, dtype=np.float64)
    sum_x2 = np.zeros(N_CHANNELS, dtype=np.float64)
    count = 0

    # Limit statistics computation to keep memory bounded.
    # Sampling is deterministic.
    grouped = selected.groupby("recording_id")

    for recording_id, group in grouped:

        info = recording_index[recording_id]

        arr = np.load(
            info["path"],
            mmap_mode="r"
        )

        indices = group["window_index"].to_numpy()

        if len(indices) > max_samples_per_recording:
            rng = np.random.default_rng(
                SEED + int(
                    abs(hash(recording_id)) % 100000
                )
            )

            indices = rng.choice(
                indices,
                size=max_samples_per_recording,
                replace=False
            )

        x = np.asarray(
            arr[indices],
            dtype=np.float32
        )

        # Flatten time into the statistics.
        # Shape: [N, C, T]
        sum_x += x.sum(
            axis=(0, 2),
            dtype=np.float64
        )

        sum_x2 += np.square(x).sum(
            axis=(0, 2),
            dtype=np.float64
        )

        count += x.shape[0] * x.shape[2]

    mean = sum_x / max(count, 1)

    var = (
        sum_x2 / max(count, 1)
        - np.square(mean)
    )

    var = np.maximum(
        var,
        EPS ** 2
    )

    std = np.sqrt(var)

    return (
        mean.astype(np.float32),
        std.astype(np.float32)
    )

# ------------------------------------------------------------
# Global train-only statistics
# ------------------------------------------------------------

print("\nComputing TRAIN-only global channel statistics...")

global_mean, global_std = (
    compute_channel_stats_from_dataframe(
        train_df
    )
)

print(
    "Global mean range:",
    float(global_mean.min()),
    "to",
    float(global_mean.max())
)

print(
    "Global std range:",
    float(global_std.min()),
    "to",
    float(global_std.max())
)

# ------------------------------------------------------------
# Subject-specific statistics
# ------------------------------------------------------------

subject_stats = {}

all_subjects = sorted(
    subject_to_recordings.keys()
)

print("\nComputing subject-specific channel statistics...")

for subject in all_subjects:

    subject_mean, subject_std = (
        compute_channel_stats_from_dataframe(
            h7,
            use_subject=subject
        )
    )

    subject_stats[subject] = {
        "mean": subject_mean,
        "std": subject_std,
    }

    print(
        f"  Subject {subject}: "
        f"mean={subject_mean.mean():.3e}, "
        f"std={subject_std.mean():.3e}"
    )

# ============================================================
# 10. DATASET
# ============================================================

class EEGDataset(Dataset):

    def __init__(
        self,
        dataframe,
        normalization,
    ):

        self.df = dataframe.reset_index(
            drop=True
        )

        self.normalization = normalization

        self.recording_cache = {}

    def __len__(self):
        return len(self.df)

    def _load_recording(self, recording_id):

        if recording_id not in self.recording_cache:

            path = recording_index[
                recording_id
            ]["path"]

            self.recording_cache[
                recording_id
            ] = np.load(
                path,
                mmap_mode="r"
            )

        return self.recording_cache[
            recording_id
        ]

    @staticmethod
    def window_zscore(x):
        """
        Exact Phase 5H.3-style normalization.

        x: [C,T]
        Normalize each channel independently over T.
        """

        mean = x.mean(
            axis=1,
            keepdims=True
        )

        std = x.std(
            axis=1,
            keepdims=True
        )

        return (
            x - mean
        ) / (
            std + EPS
        )

    def __getitem__(self, idx):

        row = self.df.iloc[idx]

        recording_id = row["recording_id"]
        window_index = int(
            row["window_index"]
        )

        subject = int(
            row["subject"]
        )

        arr = self._load_recording(
            recording_id
        )

        x = np.asarray(
            arr[window_index],
            dtype=np.float32
        )

        # ----------------------------------------------------
        # Normalization A
        # ----------------------------------------------------
        if self.normalization == "global_channel":

            mean = global_mean[
                :, None
            ]

            std = global_std[
                :, None
            ]

            x = (
                x - mean
            ) / (
                std + EPS
            )

        # ----------------------------------------------------
        # Normalization B
        # ----------------------------------------------------
        elif self.normalization == "window_z":

            x = self.window_zscore(
                x
            )

        # ----------------------------------------------------
        # Normalization C
        # ----------------------------------------------------
        elif self.normalization == "subject_channel":

            stats = subject_stats[
                subject
            ]

            mean = stats["mean"][
                :, None
            ]

            std = stats["std"][
                :, None
            ]

            x = (
                x - mean
            ) / (
                std + EPS
            )

        # ----------------------------------------------------
        # Normalization D
        # ----------------------------------------------------
        elif self.normalization == (
            "subject_channel_window_z"
        ):

            stats = subject_stats[
                subject
            ]

            mean = stats["mean"][
                :, None
            ]

            std = stats["std"][
                :, None
            ]

            x = (
                x - mean
            ) / (
                std + EPS
            )

            x = self.window_zscore(
                x
            )

        else:
            raise ValueError(
                f"Unknown normalization: "
                f"{self.normalization}"
            )

        y = int(
            row["true_class"]
        )

        return (
            torch.from_numpy(
                np.ascontiguousarray(x)
            ),
            torch.tensor(
                y,
                dtype=torch.long
            )
        )

# ============================================================
# 11. MODEL
# ============================================================

class EEGCharacterCNN(nn.Module):

    def __init__(
        self,
        in_channels=61,
        n_classes=29
    ):

        super().__init__()

        self.encoder = nn.Sequential(

            nn.Conv1d(
                in_channels,
                64,
                kernel_size=5,
                padding=2
            ),

            nn.ReLU(),

            nn.Conv1d(
                64,
                64,
                kernel_size=5,
                padding=4,
                dilation=2
            ),

            nn.ReLU(),

            nn.Conv1d(
                64,
                64,
                kernel_size=5,
                padding=8,
                dilation=4
            ),

            nn.ReLU(),

            nn.Conv1d(
                64,
                64,
                kernel_size=5,
                padding=16,
                dilation=8
            ),

            nn.ReLU(),
        )

        self.classifier = nn.Linear(
            64,
            n_classes
        )

    def forward(self, x):

        z = self.encoder(x)

        z = z.mean(
            dim=-1
        )

        return self.classifier(z)

# ============================================================
# 12. PARAMETER COUNT
# ============================================================

tmp_model = EEGCharacterCNN(
    N_CHANNELS,
    N_CLASSES
)

n_params = sum(
    p.numel()
    for p in tmp_model.parameters()
    if p.requires_grad
)

print(
    f"\nTrainable parameters: "
    f"{n_params:,}"
)

del tmp_model

# ============================================================
# 13. CLASS WEIGHTS
# ============================================================

train_counts = np.bincount(
    train_df["true_class"].to_numpy(),
    minlength=N_CLASSES
).astype(np.float64)

# Same inverse-square-root weighting used in Step 6C.
class_weights = np.zeros(
    N_CLASSES,
    dtype=np.float32
)

valid_classes = train_counts > 0

class_weights[
    valid_classes
] = (
    1.0 /
    np.sqrt(
        train_counts[
            valid_classes
        ]
    )
)

class_weights[
    valid_classes
] /= class_weights[
    valid_classes
].mean()

class_weights = torch.tensor(
    class_weights,
    dtype=torch.float32,
    device=DEVICE
)

print("\nClass weights:")
print(class_weights.detach().cpu().numpy())

# ============================================================
# 14. DATA LOADERS
# ============================================================

def make_loaders(normalization):

    train_dataset = EEGDataset(
        train_df,
        normalization
    )

    val_dataset = EEGDataset(
        val_df,
        normalization
    )

    test_dataset = EEGDataset(
        test_df,
        normalization
    )

    train_loader = DataLoader(
        train_dataset,
        batch_size=BATCH_SIZE,
        shuffle=True,
        num_workers=0,
        pin_memory=torch.cuda.is_available()
    )

    val_loader = DataLoader(
        val_dataset,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=0,
        pin_memory=torch.cuda.is_available()
    )

    test_loader = DataLoader(
        test_dataset,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=0,
        pin_memory=torch.cuda.is_available()
    )

    return (
        train_loader,
        val_loader,
        test_loader
    )

# ============================================================
# 15. METRIC FUNCTIONS
# ============================================================

def compute_multiclass_metrics(
    y_true,
    logits
):

    y_true = np.asarray(
        y_true
    )

    logits = np.asarray(
        logits
    )

    probs = torch.softmax(
        torch.tensor(logits),
        dim=1
    ).numpy()

    y_pred = probs.argmax(
        axis=1
    )

    metrics = {}

    metrics["accuracy"] = accuracy_score(
        y_true,
        y_pred
    )

    metrics["balanced_accuracy"] = (
        balanced_accuracy_score(
            y_true,
            y_pred
        )
    )

    metrics["macro_f1"] = f1_score(
        y_true,
        y_pred,
        average="macro",
        zero_division=0
    )

    metrics["weighted_f1"] = f1_score(
        y_true,
        y_pred,
        average="weighted",
        zero_division=0
    )

    metrics["macro_precision"] = (
        precision_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        )
    )

    metrics["macro_recall"] = (
        recall_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        )
    )

    try:

        metrics["macro_auroc"] = (
            roc_auc_score(
                y_true,
                probs,
                multi_class="ovr",
                average="macro"
            )
        )

    except Exception:

        metrics["macro_auroc"] = np.nan

    try:

        y_onehot = np.eye(
            N_CLASSES
        )[y_true]

        metrics["macro_auprc"] = (
            average_precision_score(
                y_onehot,
                probs,
                average="macro"
            )
        )

    except Exception:

        metrics["macro_auprc"] = np.nan

    return metrics, y_pred, probs

# ============================================================
# 16. TRAIN / EVALUATE
# ============================================================

def run_experiment(
    normalization,
    experiment_name
):

    print("\n")
    print("=" * 70)
    print(
        f"EXPERIMENT: {experiment_name}"
    )
    print(
        f"Normalization: {normalization}"
    )
    print("=" * 70)

    seed_everything(SEED)

    (
        train_loader,
        val_loader,
        test_loader
    ) = make_loaders(
        normalization
    )

    model = EEGCharacterCNN(
        N_CHANNELS,
        N_CLASSES
    ).to(DEVICE)

    criterion = nn.CrossEntropyLoss(
        weight=class_weights
    )

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=LR,
        weight_decay=WEIGHT_DECAY
    )

    best_val_f1 = -np.inf
    best_epoch = None
    patience_counter = 0

    history = []

    checkpoint_path = (
        CHECKPOINT_ROOT /
        f"{experiment_name}_best.pt"
    )

    # --------------------------------------------------------
    # Training
    # --------------------------------------------------------

    for epoch in range(
        1,
        MAX_EPOCHS + 1
    ):

        model.train()

        train_loss_sum = 0.0
        train_n = 0

        for xb, yb in train_loader:

            xb = xb.to(
                DEVICE,
                non_blocking=True
            )

            yb = yb.to(
                DEVICE,
                non_blocking=True
            )

            optimizer.zero_grad(
                set_to_none=True
            )

            logits = model(
                xb
            )

            loss = criterion(
                logits,
                yb
            )

            loss.backward()

            optimizer.step()

            train_loss_sum += (
                loss.item()
                * len(yb)
            )

            train_n += len(yb)

        train_loss = (
            train_loss_sum /
            max(train_n, 1)
        )

        # ----------------------------------------------------
        # Validation
        # ----------------------------------------------------

        model.eval()

        val_loss_sum = 0.0
        val_n = 0

        val_true = []
        val_logits = []

        with torch.no_grad():

            for xb, yb in val_loader:

                xb = xb.to(
                    DEVICE,
                    non_blocking=True
                )

                yb = yb.to(
                    DEVICE,
                    non_blocking=True
                )

                logits = model(
                    xb
                )

                loss = criterion(
                    logits,
                    yb
                )

                val_loss_sum += (
                    loss.item()
                    * len(yb)
                )

                val_n += len(yb)

                val_true.append(
                    yb.cpu().numpy()
                )

                val_logits.append(
                    logits.cpu().numpy()
                )

        val_loss = (
            val_loss_sum /
            max(val_n, 1)
        )

        val_true_np = np.concatenate(
            val_true
        )

        val_logits_np = np.concatenate(
            val_logits
        )

        val_metrics, _, _ = (
            compute_multiclass_metrics(
                val_true_np,
                val_logits_np
            )
        )

        row = {
            "experiment": experiment_name,
            "normalization": normalization,
            "epoch": epoch,
            "train_loss": train_loss,
            "val_loss": val_loss,
            **{
                f"val_{k}": v
                for k, v in val_metrics.items()
            }
        }

        history.append(
            row
        )

        print(
            f"Epoch {epoch:02d} | "
            f"train_loss={train_loss:.4f} | "
            f"val_loss={val_loss:.4f} | "
            f"val_acc={val_metrics['accuracy']:.4f} | "
            f"val_bal={val_metrics['balanced_accuracy']:.4f} | "
            f"val_macroF1={val_metrics['macro_f1']:.4f}"
        )

        # ----------------------------------------------------
        # Validation-only model selection
        # ----------------------------------------------------

        if (
            val_metrics["macro_f1"]
            > best_val_f1
        ):

            best_val_f1 = (
                val_metrics["macro_f1"]
            )

            best_epoch = epoch

            patience_counter = 0

            torch.save(
                {
                    "model_state_dict":
                        model.state_dict(),

                    "epoch":
                        epoch,

                    "best_val_macro_f1":
                        best_val_f1,

                    "normalization":
                        normalization,

                    "experiment":
                        experiment_name,

                    "seed":
                        SEED,
                },
                checkpoint_path
            )

        else:

            patience_counter += 1

            if patience_counter >= PATIENCE:

                print(
                    f"Early stopping at epoch "
                    f"{epoch}."
                )

                break

    # --------------------------------------------------------
    # Save training history
    # --------------------------------------------------------

    history_df = pd.DataFrame(
        history
    )

    history_path = (
        OUTPUT_ROOT /
        f"{experiment_name}_history.csv"
    )

    history_df.to_csv(
        history_path,
        index=False
    )

    # --------------------------------------------------------
    # Load frozen best model
    # --------------------------------------------------------

    checkpoint = torch.load(
        checkpoint_path,
        map_location=DEVICE,
        weights_only=False
    )

    model.load_state_dict(
        checkpoint[
            "model_state_dict"
        ]
    )

    print(
        f"\nFrozen best epoch: "
        f"{checkpoint['epoch']}"
    )

    print(
        f"Best validation macro-F1: "
        f"{checkpoint['best_val_macro_f1']:.6f}"
    )

    # ========================================================
    # IMPORTANT:
    # TEST IS USED ONLY HERE, AFTER VALIDATION SELECTION.
    # ========================================================

    model.eval()

    test_loss_sum = 0.0
    test_n = 0

    test_true = []
    test_logits = []

    with torch.no_grad():

        for xb, yb in test_loader:

            xb = xb.to(
                DEVICE,
                non_blocking=True
            )

            yb = yb.to(
                DEVICE,
                non_blocking=True
            )

            logits = model(
                xb
            )

            loss = criterion(
                logits,
                yb
            )

            test_loss_sum += (
                loss.item()
                * len(yb)
            )

            test_n += len(yb)

            test_true.append(
                yb.cpu().numpy()
            )

            test_logits.append(
                logits.cpu().numpy()
            )

    test_loss = (
        test_loss_sum /
        max(test_n, 1)
    )

    test_true_np = np.concatenate(
        test_true
    )

    test_logits_np = np.concatenate(
        test_logits
    )

    test_metrics, test_pred, test_probs = (
        compute_multiclass_metrics(
            test_true_np,
            test_logits_np
        )
    )

    # --------------------------------------------------------
    # Confusion matrix
    # --------------------------------------------------------

    cm = confusion_matrix(
        test_true_np,
        test_pred,
        labels=np.arange(
            N_CLASSES
        )
    )

    cm_path = (
        OUTPUT_ROOT /
        f"{experiment_name}_confusion_matrix.csv"
    )

    pd.DataFrame(
        cm
    ).to_csv(
        cm_path,
        index=False
    )

    # --------------------------------------------------------
    # Test predictions
    # --------------------------------------------------------

    predictions_df = test_df.copy()

    predictions_df[
        "predicted_class"
    ] = test_pred

    predictions_df[
        "prediction_confidence"
    ] = test_probs.max(
        axis=1
    )

    predictions_path = (
        OUTPUT_ROOT /
        f"{experiment_name}_test_predictions.csv"
    )

    predictions_df.to_csv(
        predictions_path,
        index=False
    )

    # --------------------------------------------------------
    # Result
    # --------------------------------------------------------

    result = {
        "experiment": experiment_name,
        "normalization": normalization,

        "variant": VARIANT,

        "best_epoch": int(
            checkpoint["epoch"]
        ),

        "best_val_macro_f1": float(
            checkpoint[
                "best_val_macro_f1"
            ]
        ),

        "test_loss": float(
            test_loss
        ),

        **{
            f"test_{k}": (
                None
                if np.isnan(v)
                else float(v)
            )
            for k, v in test_metrics.items()
        },

        "train_samples": int(
            len(train_df)
        ),

        "val_samples": int(
            len(val_df)
        ),

        "test_samples": int(
            len(test_df)
        ),

        "train_subjects": sorted(
            set(
                train_df["subject"]
            )
        ),

        "val_subjects": sorted(
            set(
                val_df["subject"]
            )
        ),

        "test_subjects": sorted(
            set(
                test_df["subject"]
            )
        ),

        "checkpoint": str(
            checkpoint_path
        ),
    }

    result_path = (
        OUTPUT_ROOT /
        f"{experiment_name}_result.json"
    )

    with open(
        result_path,
        "w"
    ) as f:

        json.dump(
            result,
            f,
            indent=2
        )

    print("\nTEST RESULT")
    print("-" * 70)

    print(
        f"Test accuracy:          "
        f"{test_metrics['accuracy']:.6f}"
    )

    print(
        f"Test balanced accuracy: "
        f"{test_metrics['balanced_accuracy']:.6f}"
    )

    print(
        f"Test macro-F1:           "
        f"{test_metrics['macro_f1']:.6f}"
    )

    print(
        f"Test weighted-F1:        "
        f"{test_metrics['weighted_f1']:.6f}"
    )

    print(
        f"Test macro AUROC:        "
        f"{test_metrics['macro_auroc']}"
    )

    print(
        f"Test macro AUPRC:        "
        f"{test_metrics['macro_auprc']}"
    )

    return result

# ============================================================
# 17. RUN FOUR REPRESENTATION EXPERIMENTS
# ============================================================

experiments = [

    (
        "global_channel",
        "A_global_channel"
    ),

    (
        "window_z",
        "B_window_z"
    ),

    (
        "subject_channel",
        "C_subject_channel"
    ),

    (
        "subject_channel_window_z",
        "D_subject_channel_window_z"
    ),
]

results = []

for normalization, experiment_name in experiments:

    result = run_experiment(
        normalization,
        experiment_name
    )

    results.append(
        result
    )

# ============================================================
# 18. COMPARISON TABLE
# ============================================================

comparison_rows = []

for r in results:

    comparison_rows.append(
        {
            "experiment":
                r["experiment"],

            "normalization":
                r["normalization"],

            "best_epoch":
                r["best_epoch"],

            "best_val_macro_f1":
                r["best_val_macro_f1"],

            "test_accuracy":
                r["test_accuracy"],

            "test_balanced_accuracy":
                r["test_balanced_accuracy"],

            "test_macro_f1":
                r["test_macro_f1"],

            "test_weighted_f1":
                r["test_weighted_f1"],

            "test_macro_precision":
                r["test_macro_precision"],

            "test_macro_recall":
                r["test_macro_recall"],

            "test_macro_auroc":
                r["test_macro_auroc"],

            "test_macro_auprc":
                r["test_macro_auprc"],
        }
    )

comparison_df = pd.DataFrame(
    comparison_rows
)

comparison_df = comparison_df.sort_values(
    "best_val_macro_f1",
    ascending=False
)

comparison_path = (
    OUTPUT_ROOT /
    "phase6d_representation_comparison.csv"
)

comparison_df.to_csv(
    comparison_path,
    index=False
)

print("\n")
print("=" * 70)
print("STEP 6D FINAL COMPARISON")
print("=" * 70)

print(
    comparison_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.6f}"
    )
)

# ============================================================
# 19. SELECT BEST CONFIGURATION USING VALIDATION ONLY
# ============================================================

best_row = comparison_df.iloc[0]

best_experiment = best_row[
    "experiment"
]

best_normalization = best_row[
    "normalization"
]

best_val_f1 = float(
    best_row[
        "best_val_macro_f1"
    ]
)

best_test_f1 = float(
    best_row[
        "test_macro_f1"
    ]
)

print("\n")
print("=" * 70)
print("VALIDATION-SELECTED CONFIGURATION")
print("=" * 70)

print(
    f"Best experiment: "
    f"{best_experiment}"
)

print(
    f"Normalization: "
    f"{best_normalization}"
)

print(
    f"Validation macro-F1: "
    f"{best_val_f1:.6f}"
)

print(
    f"Corresponding frozen test macro-F1: "
    f"{best_test_f1:.6f}"
)

# ============================================================
# 20. BASELINE COMPARISON
# ============================================================

majority_class = int(
    train_df[
        "true_class"
    ].value_counts().idxmax()
)

test_true = test_df[
    "true_class"
].to_numpy()

majority_pred = np.full(
    len(test_true),
    majority_class
)

majority_acc = accuracy_score(
    test_true,
    majority_pred
)

majority_bal = balanced_accuracy_score(
    test_true,
    majority_pred
)

majority_f1 = f1_score(
    test_true,
    majority_pred,
    average="macro",
    zero_division=0
)

print("\n")
print("=" * 70)
print("MAJORITY-CLASS BASELINE")
print("=" * 70)

print(
    f"Majority class: "
    f"{majority_class}"
)

print(
    f"Test accuracy: "
    f"{majority_acc:.6f}"
)

print(
    f"Test balanced accuracy: "
    f"{majority_bal:.6f}"
)

print(
    f"Test macro-F1: "
    f"{majority_f1:.6f}"
)

# ============================================================
# 21. SUBJECT-LEVEL ANALYSIS FOR BEST CONFIGURATION
# ============================================================

best_predictions_path = (
    OUTPUT_ROOT /
    f"{best_experiment}_test_predictions.csv"
)

best_predictions = pd.read_csv(
    best_predictions_path
)

subject_rows = []

for subject, group in (
    best_predictions
    .groupby("subject")
):

    y_true = group[
        "true_class"
    ].to_numpy()

    y_pred = group[
        "predicted_class"
    ].to_numpy()

    subject_rows.append(
        {
            "subject":
                int(subject),

            "n":
                len(group),

            "accuracy":
                accuracy_score(
                    y_true,
                    y_pred
                ),

            "balanced_accuracy":
                balanced_accuracy_score(
                    y_true,
                    y_pred
                ),

            "macro_f1":
                f1_score(
                    y_true,
                    y_pred,
                    average="macro",
                    zero_division=0
                ),

            "weighted_f1":
                f1_score(
                    y_true,
                    y_pred,
                    average="weighted",
                    zero_division=0
                ),
        }
    )

subject_df = pd.DataFrame(
    subject_rows
)

subject_path = (
    OUTPUT_ROOT /
    "phase6d_best_subject_results.csv"
)

subject_df.to_csv(
    subject_path,
    index=False
)

print("\n")
print("=" * 70)
print("BEST CONFIGURATION — SUBJECT-LEVEL TEST RESULTS")
print("=" * 70)

print(
    subject_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.6f}"
    )
)

# ============================================================
# 22. FINAL DIAGNOSTIC VERDICT
# ============================================================

baseline_row = comparison_df[
    comparison_df["normalization"]
    == "window_z"
].iloc[0]

best_row = comparison_df.iloc[0]

baseline_test_f1 = float(
    baseline_row[
        "test_macro_f1"
    ]
)

best_test_f1 = float(
    best_row[
        "test_macro_f1"
    ]
)

improvement = (
    best_test_f1 -
    baseline_test_f1
)

summary = {
    "step": "6D",
    "purpose":
        "Subject-independent EEG representation diagnosis",

    "protocol": {
        "test_frozen": True,
        "selection_metric":
            "validation macro-F1",
        "temporal_variant":
            VARIANT,
        "normalization_experiments":
            [
                x[0]
                for x in experiments
            ],
    },

    "best_experiment":
        best_experiment,

    "best_normalization":
        best_normalization,

    "best_validation_macro_f1":
        best_val_f1,

    "best_test_macro_f1":
        best_test_f1,

    "phase6c_window_z_test_macro_f1":
        baseline_test_f1,

    "test_macro_f1_difference_vs_window_z":
        improvement,

    "majority_test_accuracy":
        float(majority_acc),

    "majority_test_balanced_accuracy":
        float(majority_bal),

    "majority_test_macro_f1":
        float(majority_f1),

    "interpretation": (
        "Subject-wise normalization improves decoding only if "
        "validation-selected configuration also produces a "
        "meaningful improvement on the frozen test subjects. "
        "Small changes should be treated as insufficient evidence "
        "of improved subject-independent representation."
    ),

    "outputs": {
        "comparison":
            str(comparison_path),

        "subject_results":
            str(subject_path),
    },
}

summary_path = (
    OUTPUT_ROOT /
    "phase6d_summary.json"
)

with open(
    summary_path,
    "w"
) as f:

    json.dump(
        summary,
        f,
        indent=2
    )

# ============================================================
# 23. FINAL STATUS
# ============================================================

print("\n")
print("=" * 70)
print("STEP 6D COMPLETE")
print("=" * 70)

print(
    f"Best validation configuration: "
    f"{best_experiment}"
)

print(
    f"Validation macro-F1: "
    f"{best_val_f1:.6f}"
)

print(
    f"Frozen test macro-F1: "
    f"{best_test_f1:.6f}"
)

print(
    f"Phase 6C window-z test macro-F1: "
    f"{baseline_test_f1:.6f}"
)

print(
    f"Difference: "
    f"{improvement:+.6f}"
)

print("\nOutput files:")
print(
    comparison_path
)
print(
    summary_path
)
print(
    subject_path
)

print("\nSTATUS: READY FOR STEP 6D ANALYSIS")
```

```python
# ============================================================
# NEUROSENTINEL — STEP 6E
# SUBJECT-INVARIANT EEG REPRESENTATION LEARNING
#
# Baseline:
#   EEG -> CNN encoder -> Character classifier
#
# Adversarial:
#   EEG -> CNN encoder
#                  ├── Character classifier
#                  └── Gradient Reversal -> Subject classifier
#
# Protocol:
#   - Frozen Phase 6 subject split
#   - Same baseline temporal window [-0.2,+0.3]
#   - Exact Phase5H.3 per-window channel-wise z-score
#   - Validation-only model selection
#   - Test subjects remain completely frozen
#   - No Phase 8 gating
# ============================================================

import os
import json
import random
import warnings
from pathlib import Path

import numpy as np
import pandas as pd

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    precision_score,
    recall_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
)

warnings.filterwarnings("ignore")

# ============================================================
# 1. CONFIGURATION
# ============================================================

SEED = 42

ROOT = Path("/content/gdrive/MyDrive/NeuroSentinel")

SPLIT_MANIFEST = (
    ROOT / "07_phase6_subject_split" /
    "phase6_subject_split_manifest_resolved.csv"
)

WINDOW_ROOT = (
    ROOT / "08_phase6_temporal_windows"
)

OUTPUT_ROOT = (
    ROOT / "results" / "phase6_subject_invariant"
)

CHECKPOINT_ROOT = (
    ROOT / "checkpoints" / "phase6_subject_invariant"
)

OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)
CHECKPOINT_ROOT.mkdir(parents=True, exist_ok=True)

VARIANT = "baseline"

# Model
N_CHANNELS = 61
N_CLASSES = 29
N_SUBJECTS = 20

HIDDEN = 64
EMBED_DIM = 64

# Training
BATCH_SIZE = 256
LR = 1e-3
WEIGHT_DECAY = 1e-4
MAX_EPOCHS = 20
PATIENCE = 5

# Adversarial strengths to evaluate.
# These are selected using validation performance.
LAMBDA_VALUES = [0.05, 0.10, 0.25, 0.50, 1.00]

NUM_WORKERS = 2

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("=" * 70)
print("NEUROSENTINEL — STEP 6E")
print("SUBJECT-INVARIANT REPRESENTATION LEARNING")
print("=" * 70)

print("Device:", DEVICE)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))

# ============================================================
# 2. REPRODUCIBILITY
# ============================================================

def seed_everything(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


seed_everything(SEED)

# ============================================================
# 3. LOAD FROZEN PHASE 6 SPLIT
# ============================================================

if not SPLIT_MANIFEST.exists():
    fallback = (
        ROOT / "07_phase6_subject_split" /
        "phase6_subject_split_manifest.csv"
    )

    if fallback.exists():
        SPLIT_MANIFEST = fallback
    else:
        raise FileNotFoundError(
            f"Phase 6 manifest not found:\n{SPLIT_MANIFEST}"
        )

split_df = pd.read_csv(SPLIT_MANIFEST)

required_split_cols = [
    "recording_id",
    "subject",
    "session",
    "task",
    "split",
]

missing = [
    c for c in required_split_cols
    if c not in split_df.columns
]

if missing:
    raise ValueError(
        f"Missing split columns: {missing}"
    )

split_df["recording_id"] = (
    split_df["recording_id"].astype(str)
)

split_df["subject"] = (
    split_df["subject"].astype(int)
)

split_df["split"] = (
    split_df["split"].astype(str).str.lower()
)

train_subjects = sorted(
    split_df.loc[
        split_df["split"] == "train", "subject"
    ].unique().tolist()
)

val_subjects = sorted(
    split_df.loc[
        split_df["split"] == "val", "subject"
    ].unique().tolist()
)

test_subjects = sorted(
    split_df.loc[
        split_df["split"] == "test", "subject"
    ].unique().tolist()
)

print("\nFrozen subject split:")
print("Train:", train_subjects)
print("Val  :", val_subjects)
print("Test :", test_subjects)

assert not set(train_subjects) & set(val_subjects)
assert not set(train_subjects) & set(test_subjects)
assert not set(val_subjects) & set(test_subjects)

# ============================================================
# 4. LOAD PHASE 5H.7 UTILITY MANIFEST
#    ONLY TO GET THE FROZEN WINDOW INDEX / LABEL ASSOCIATION.
#    Character labels remain the Phase 5 decoder labels.
# ============================================================

H7_MANIFEST = (
    ROOT / "results" / "phase5h7" /
    "phase5h7_utility_label_manifest.csv"
)

if not H7_MANIFEST.exists():
    raise FileNotFoundError(
        f"H7 manifest not found:\n{H7_MANIFEST}"
    )

h7 = pd.read_csv(H7_MANIFEST)

required_h7_cols = [
    "recording_id",
    "split",
    "subject",
    "window_index",
    "true_class",
    "true_label",
]

missing = [
    c for c in required_h7_cols
    if c not in h7.columns
]

if missing:
    raise ValueError(
        f"Missing H7 columns: {missing}"
    )

h7["recording_id"] = (
    h7["recording_id"].astype(str)
)

h7["subject"] = h7["subject"].astype(int)
h7["window_index"] = h7["window_index"].astype(int)
h7["true_class"] = h7["true_class"].astype(int)

# Verify H7 split agrees with Phase 6.
split_lookup = dict(
    zip(
        split_df["recording_id"],
        split_df["split"]
    )
)

h7["expected_split"] = (
    h7["recording_id"].map(split_lookup)
)

mismatch = (
    h7["split"].astype(str).str.lower()
    != h7["expected_split"]
)

if mismatch.any():
    raise RuntimeError(
        f"Phase 6/H7 split mismatch: {mismatch.sum()}"
    )

print("\nH7 manifest:", h7.shape)

# ============================================================
# 5. RECORDING FILE DISCOVERY
# ============================================================

def find_window_file(recording_id):
    split = split_lookup[recording_id]

    folder = (
        WINDOW_ROOT /
        split /
        recording_id
    )

    candidates = [
        folder / f"{recording_id}_baseline_windows.npy",
        folder / f"{recording_id}_windows.npy",
    ]

    for p in candidates:
        if p.exists():
            return p

    raise FileNotFoundError(
        f"No baseline window file found for {recording_id}"
    )


recording_files = {}

for rid in split_df["recording_id"]:
    recording_files[rid] = find_window_file(rid)

print(
    f"Verified baseline window files: "
    f"{len(recording_files)}/{len(split_df)}"
)

# ============================================================
# 6. BUILD RECORDING-LOCAL INDEX
# ============================================================

# We use the H7 manifest to map:
# recording_id + window_index -> true character class.

groups = {}

for split_name in ["train", "val", "test"]:

    subset = h7[
        h7["split"].astype(str).str.lower()
        == split_name
    ]

    groups[split_name] = []

    for rid, g in subset.groupby("recording_id"):

        g = g.sort_values("window_index")

        groups[split_name].append({
            "recording_id": rid,
            "subject": int(
                g["subject"].iloc[0]
            ),
            "window_file": recording_files[rid],
            "window_indices": (
                g["window_index"]
                .to_numpy(dtype=np.int64)
            ),
            "labels": (
                g["true_class"]
                .to_numpy(dtype=np.int64)
            ),
        })

print("\nRecording counts:")
for s in groups:
    print(
        f"{s:5s}: {len(groups[s])}"
    )

# ============================================================
# 7. EXACT PHASE5H.3 NORMALIZATION
#
# Per-window, per-channel z-score across the temporal axis.
# ============================================================

EPS = 1e-6

def normalize_window(x):
    """
    x: [C,T]
    """
    x = x.astype(np.float32, copy=False)

    mean = x.mean(
        axis=1,
        keepdims=True
    )

    std = x.std(
        axis=1,
        keepdims=True
    )

    x = (
        x - mean
    ) / (
        std + EPS
    )

    return x.astype(np.float32, copy=False)

# ============================================================
# 8. DATASET
# ============================================================

class EEGDataset(Dataset):

    def __init__(self, recording_groups):

        self.samples = []

        for rec in recording_groups:

            rid = rec["recording_id"]
            subject = rec["subject"]

            indices = rec["window_indices"]
            labels = rec["labels"]

            for idx, label in zip(
                indices,
                labels
            ):
                self.samples.append(
                    (
                        rid,
                        int(idx),
                        int(label),
                        int(subject)
                    )
                )

        self._arrays = {}

    def _get_array(self, rid):

        if rid not in self._arrays:

            arr = np.load(
                recording_files[rid],
                mmap_mode="r"
            )

            if arr.ndim != 3:
                raise ValueError(
                    f"{rid}: invalid shape {arr.shape}"
                )

            if arr.shape[1] != N_CHANNELS:
                raise ValueError(
                    f"{rid}: expected "
                    f"{N_CHANNELS} channels, "
                    f"got {arr.shape[1]}"
                )

            self._arrays[rid] = arr

        return self._arrays[rid]

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, i):

        rid, idx, label, subject = (
            self.samples[i]
        )

        arr = self._get_array(rid)

        x = arr[idx]

        x = normalize_window(x)

        x = torch.from_numpy(
            np.ascontiguousarray(x)
        )

        y = torch.tensor(
            label,
            dtype=torch.long
        )

        s = torch.tensor(
            subject,
            dtype=torch.long
        )

        return x, y, s


datasets = {
    split: EEGDataset(groups[split])
    for split in ["train", "val", "test"]
}

print("\nSample counts:")

for split, ds in datasets.items():
    print(
        f"{split:5s}: {len(ds):,}"
    )

assert len(datasets["train"]) == 99837
assert len(datasets["val"]) == 21206
assert len(datasets["test"]) == 29907

# ============================================================
# 9. CLASS WEIGHTS
# ============================================================

train_labels = h7[
    h7["split"].astype(str).str.lower()
    == "train"
]["true_class"].to_numpy()

class_counts = np.bincount(
    train_labels,
    minlength=N_CLASSES
).astype(np.float64)

# Inverse-square-root weighting,
# matching Step 6C / 6D.
class_weights = np.zeros(
    N_CLASSES,
    dtype=np.float64
)

present = class_counts > 0

class_weights[present] = (
    1.0 /
    np.sqrt(class_counts[present])
)

class_weights[present] /= (
    class_weights[present].mean()
)

class_weights = torch.tensor(
    class_weights,
    dtype=torch.float32,
    device=DEVICE
)

print("\nCharacter class counts:")
print(class_counts.astype(int))

# ============================================================
# 10. GRADIENT REVERSAL
# ============================================================

class GradientReversalFunction(
    torch.autograd.Function
):

    @staticmethod
    def forward(
        ctx,
        x,
        lambda_
    ):

        ctx.lambda_ = lambda_

        return x.view_as(x)

    @staticmethod
    def backward(
        ctx,
        grad_output
    ):

        return (
            -ctx.lambda_ * grad_output,
            None
        )


class GradientReversal(nn.Module):

    def __init__(self, lambda_=1.0):

        super().__init__()

        self.lambda_ = lambda_

    def forward(self, x):

        return GradientReversalFunction.apply(
            x,
            self.lambda_
        )

# ============================================================
# 11. CNN ENCODER
# ============================================================

class EEGEncoder(nn.Module):

    def __init__(self):

        super().__init__()

        self.net = nn.Sequential(

            nn.Conv1d(
                N_CHANNELS,
                64,
                kernel_size=5,
                padding=2
            ),

            nn.ReLU(),

            nn.Conv1d(
                64,
                64,
                kernel_size=5,
                padding=4,
                dilation=2
            ),

            nn.ReLU(),

            nn.Conv1d(
                64,
                64,
                kernel_size=5,
                padding=8,
                dilation=4
            ),

            nn.ReLU(),

            nn.Conv1d(
                64,
                EMBED_DIM,
                kernel_size=5,
                padding=16,
                dilation=8
            ),

            nn.ReLU()
        )

    def forward(self, x):

        x = self.net(x)

        # Global temporal average
        x = x.mean(dim=-1)

        return x

# ============================================================
# 12. BASELINE MODEL
# ============================================================

class BaselineModel(nn.Module):

    def __init__(self):

        super().__init__()

        self.encoder = EEGEncoder()

        self.character_head = nn.Linear(
            EMBED_DIM,
            N_CLASSES
        )

    def forward(self, x):

        z = self.encoder(x)

        logits = self.character_head(z)

        return logits

# ============================================================
# 13. ADVERSARIAL MODEL
# ============================================================

class AdversarialModel(nn.Module):

    def __init__(
        self,
        lambda_adv
    ):

        super().__init__()

        self.encoder = EEGEncoder()

        self.character_head = nn.Linear(
            EMBED_DIM,
            N_CLASSES
        )

        self.grl = GradientReversal(
            lambda_adv
        )

        self.subject_head = nn.Sequential(

            nn.Linear(
                EMBED_DIM,
                64
            ),

            nn.ReLU(),

            nn.Linear(
                64,
                N_SUBJECTS + 1
            )
        )

    def forward(self, x):

        z = self.encoder(x)

        char_logits = (
            self.character_head(z)
        )

        subject_logits = self.subject_head(
            self.grl(z)
        )

        return (
            char_logits,
            subject_logits,
            z
        )

# ============================================================
# 14. SUBJECT LABEL MAPPING
# ============================================================

all_subjects = sorted(
    set(
        train_subjects +
        val_subjects +
        test_subjects
    )
)

subject_to_idx = {
    s: i
    for i, s in enumerate(all_subjects)
}

print("\nSubject mapping:")
print(subject_to_idx)

# ============================================================
# 15. DATALOADERS
# ============================================================

loader_kwargs = {
    "batch_size": BATCH_SIZE,
    "num_workers": NUM_WORKERS,
    "pin_memory": torch.cuda.is_available(),
}

train_loader = DataLoader(
    datasets["train"],
    shuffle=True,
    drop_last=False,
    **loader_kwargs
)

val_loader = DataLoader(
    datasets["val"],
    shuffle=False,
    drop_last=False,
    **loader_kwargs
)

test_loader = DataLoader(
    datasets["test"],
    shuffle=False,
    drop_last=False,
    **loader_kwargs
)

# ============================================================
# 16. METRICS
# ============================================================

def safe_auroc(y_true, probs):

    try:
        return float(
            roc_auc_score(
                y_true,
                probs,
                multi_class="ovr",
                average="macro"
            )
        )
    except Exception:
        return float("nan")


def safe_auprc(y_true, probs):

    try:

        scores = []

        for c in range(N_CLASSES):

            binary = (
                y_true == c
            ).astype(np.int32)

            if binary.sum() == 0:
                continue

            scores.append(
                average_precision_score(
                    binary,
                    probs[:, c]
                )
            )

        if not scores:
            return float("nan")

        return float(
            np.mean(scores)
        )

    except Exception:
        return float("nan")


def classification_metrics(
    y_true,
    probs
):

    pred = probs.argmax(axis=1)

    return {
        "accuracy": float(
            accuracy_score(
                y_true,
                pred
            )
        ),

        "balanced_accuracy": float(
            balanced_accuracy_score(
                y_true,
                pred
            )
        ),

        "macro_f1": float(
            f1_score(
                y_true,
                pred,
                average="macro",
                zero_division=0
            )
        ),

        "weighted_f1": float(
            f1_score(
                y_true,
                pred,
                average="weighted",
                zero_division=0
            )
        ),

        "macro_precision": float(
            precision_score(
                y_true,
                pred,
                average="macro",
                zero_division=0
            )
        ),

        "macro_recall": float(
            recall_score(
                y_true,
                pred,
                average="macro",
                zero_division=0
            )
        ),

        "macro_auroc": safe_auroc(
            y_true,
            probs
        ),

        "macro_auprc": safe_auprc(
            y_true,
            probs
        ),
    }

# ============================================================
# 17. EVALUATION FUNCTION
# ============================================================

@torch.no_grad()
def evaluate_baseline(
    model,
    loader
):

    model.eval()

    all_probs = []
    all_y = []

    total_loss = 0.0
    total_n = 0

    criterion = nn.CrossEntropyLoss(
        weight=class_weights
    )

    for x, y, _ in loader:

        x = x.to(
            DEVICE,
            non_blocking=True
        )

        y = y.to(
            DEVICE,
            non_blocking=True
        )

        logits = model(x)

        loss = criterion(
            logits,
            y
        )

        bs = y.size(0)

        total_loss += (
            loss.item() * bs
        )

        total_n += bs

        probs = torch.softmax(
            logits,
            dim=1
        )

        all_probs.append(
            probs.cpu().numpy()
        )

        all_y.append(
            y.cpu().numpy()
        )

    probs = np.concatenate(
        all_probs
    )

    y_true = np.concatenate(
        all_y
    )

    metrics = classification_metrics(
        y_true,
        probs
    )

    metrics["loss"] = (
        total_loss / total_n
    )

    return (
        metrics,
        y_true,
        probs
    )


@torch.no_grad()
def evaluate_adversarial(
    model,
    loader
):

    model.eval()

    all_probs = []
    all_y = []

    all_subject_true = []
    all_subject_pred = []

    total_char_loss = 0.0
    total_subject_loss = 0.0
    total_n = 0

    char_criterion = nn.CrossEntropyLoss(
        weight=class_weights
    )

    subject_criterion = nn.CrossEntropyLoss()

    for x, y, subject in loader:

        x = x.to(
            DEVICE,
            non_blocking=True
        )

        y = y.to(
            DEVICE,
            non_blocking=True
        )

        subject_idx = torch.tensor(
            [
                subject_to_idx[
                    int(s)
                ]
                for s in subject.numpy()
            ],
            dtype=torch.long,
            device=DEVICE
        )

        char_logits, subject_logits, _ = (
            model(x)
        )

        char_loss = char_criterion(
            char_logits,
            y
        )

        subject_loss = subject_criterion(
            subject_logits,
            subject_idx
        )

        bs = y.size(0)

        total_char_loss += (
            char_loss.item() * bs
        )

        total_subject_loss += (
            subject_loss.item() * bs
        )

        total_n += bs

        probs = torch.softmax(
            char_logits,
            dim=1
        )

        all_probs.append(
            probs.cpu().numpy()
        )

        all_y.append(
            y.cpu().numpy()
        )

        subject_pred = (
            subject_logits.argmax(
                dim=1
            )
            .cpu()
            .numpy()
        )

        all_subject_pred.append(
            subject_pred
        )

        all_subject_true.append(
            subject_idx.cpu().numpy()
        )

    probs = np.concatenate(
        all_probs
    )

    y_true = np.concatenate(
        all_y
    )

    subject_true = np.concatenate(
        all_subject_true
    )

    subject_pred = np.concatenate(
        all_subject_pred
    )

    metrics = classification_metrics(
        y_true,
        probs
    )

    metrics["char_loss"] = (
        total_char_loss / total_n
    )

    metrics["subject_loss"] = (
        total_subject_loss / total_n
    )

    metrics["subject_accuracy"] = float(
        accuracy_score(
            subject_true,
            subject_pred
        )
    )

    return (
        metrics,
        y_true,
        probs,
        subject_true,
        subject_pred
    )

# ============================================================
# 18. BASELINE TRAINING
# ============================================================

def train_baseline():

    print("\n" + "=" * 70)
    print("BASELINE CNN")
    print("=" * 70)

    seed_everything(SEED)

    model = BaselineModel().to(
        DEVICE
    )

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=LR,
        weight_decay=WEIGHT_DECAY
    )

    criterion = nn.CrossEntropyLoss(
        weight=class_weights
    )

    best_score = -np.inf
    best_epoch = 0
    patience_count = 0

    history = []

    checkpoint_path = (
        CHECKPOINT_ROOT /
        "step6e_baseline_best.pt"
    )

    for epoch in range(
        1,
        MAX_EPOCHS + 1
    ):

        model.train()

        running_loss = 0.0
        running_n = 0

        for x, y, _ in train_loader:

            x = x.to(
                DEVICE,
                non_blocking=True
            )

            y = y.to(
                DEVICE,
                non_blocking=True
            )

            optimizer.zero_grad(
                set_to_none=True
            )

            logits = model(x)

            loss = criterion(
                logits,
                y
            )

            loss.backward()

            optimizer.step()

            bs = y.size(0)

            running_loss += (
                loss.item() * bs
            )

            running_n += bs

        train_loss = (
            running_loss /
            running_n
        )

        val_metrics, _, _ = (
            evaluate_baseline(
                model,
                val_loader
            )
        )

        row = {
            "epoch": epoch,
            "train_loss": train_loss,
            **{
                f"val_{k}": v
                for k, v
                in val_metrics.items()
            }
        }

        history.append(row)

        score = val_metrics[
            "macro_f1"
        ]

        print(
            f"Epoch {epoch:02d} | "
            f"train_loss={train_loss:.4f} | "
            f"val_macroF1={score:.6f} | "
            f"val_bal={val_metrics['balanced_accuracy']:.6f}"
        )

        if score > best_score:

            best_score = score
            best_epoch = epoch
            patience_count = 0

            torch.save(
                {
                    "model_state_dict":
                        model.state_dict(),
                    "epoch":
                        epoch,
                    "val_macro_f1":
                        score,
                },
                checkpoint_path
            )

        else:

            patience_count += 1

            if patience_count >= PATIENCE:
                print(
                    "Early stopping."
                )
                break

    checkpoint = torch.load(
        checkpoint_path,
        map_location=DEVICE
    )

    model.load_state_dict(
        checkpoint["model_state_dict"]
    )

    test_metrics, test_y, test_probs = (
        evaluate_baseline(
            model,
            test_loader
        )
    )

    print("\nBaseline frozen-test result:")
    print(
        json.dumps(
            test_metrics,
            indent=2
        )
    )

    return {
        "model": model,
        "best_epoch": best_epoch,
        "best_val_macro_f1": best_score,
        "history": history,
        "test_metrics": test_metrics,
        "test_y": test_y,
        "test_probs": test_probs,
    }

# ============================================================
# 19. ADVERSARIAL TRAINING
# ============================================================

def train_adversarial(
    lambda_adv
):

    print("\n" + "=" * 70)
    print(
        f"ADVERSARIAL MODEL "
        f"(lambda={lambda_adv})"
    )
    print("=" * 70)

    seed_everything(SEED)

    model = AdversarialModel(
        lambda_adv=lambda_adv
    ).to(DEVICE)

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=LR,
        weight_decay=WEIGHT_DECAY
    )

    char_criterion = nn.CrossEntropyLoss(
        weight=class_weights
    )

    subject_criterion = nn.CrossEntropyLoss()

    best_score = -np.inf
    best_epoch = 0
    patience_count = 0

    history = []

    checkpoint_path = (
        CHECKPOINT_ROOT /
        f"step6e_adversarial_lambda_{lambda_adv:.2f}.pt"
    )

    for epoch in range(
        1,
        MAX_EPOCHS + 1
    ):

        model.train()

        running_loss = 0.0
        running_char = 0.0
        running_subject = 0.0
        running_n = 0

        for x, y, subject in train_loader:

            x = x.to(
                DEVICE,
                non_blocking=True
            )

            y = y.to(
                DEVICE,
                non_blocking=True
            )

            subject_idx = torch.tensor(
                [
                    subject_to_idx[
                        int(s)
                    ]
                    for s in subject.numpy()
                ],
                dtype=torch.long,
                device=DEVICE
            )

            optimizer.zero_grad(
                set_to_none=True
            )

            (
                char_logits,
                subject_logits,
                _
            ) = model(x)

            char_loss = char_criterion(
                char_logits,
                y
            )

            subject_loss = (
                subject_criterion(
                    subject_logits,
                    subject_idx
                )
            )

            # The GRL reverses the subject-loss
            # gradient before it reaches encoder.
            total_loss = (
                char_loss +
                subject_loss
            )

            total_loss.backward()

            optimizer.step()

            bs = y.size(0)

            running_loss += (
                total_loss.item() * bs
            )

            running_char += (
                char_loss.item() * bs
            )

            running_subject += (
                subject_loss.item() * bs
            )

            running_n += bs

        train_loss = (
            running_loss /
            running_n
        )

        train_char_loss = (
            running_char /
            running_n
        )

        train_subject_loss = (
            running_subject /
            running_n
        )

        (
            val_metrics,
            _,
            _,
            _,
            _
        ) = evaluate_adversarial(
            model,
            val_loader
        )

        row = {
            "epoch": epoch,
            "train_loss": train_loss,
            "train_char_loss":
                train_char_loss,
            "train_subject_loss":
                train_subject_loss,
            **{
                f"val_{k}": v
                for k, v
                in val_metrics.items()
            }
        }

        history.append(row)

        score = val_metrics[
            "macro_f1"
        ]

        print(
            f"Epoch {epoch:02d} | "
            f"train={train_loss:.4f} | "
            f"char={train_char_loss:.4f} | "
            f"subj={train_subject_loss:.4f} | "
            f"val_macroF1={score:.6f} | "
            f"val_subj_acc="
            f"{val_metrics['subject_accuracy']:.4f}"
        )

        if score > best_score:

            best_score = score
            best_epoch = epoch
            patience_count = 0

            torch.save(
                {
                    "model_state_dict":
                        model.state_dict(),
                    "epoch":
                        epoch,
                    "lambda_adv":
                        lambda_adv,
                    "val_macro_f1":
                        score,
                },
                checkpoint_path
            )

        else:

            patience_count += 1

            if patience_count >= PATIENCE:

                print(
                    "Early stopping."
                )

                break

    checkpoint = torch.load(
        checkpoint_path,
        map_location=DEVICE
    )

    model.load_state_dict(
        checkpoint["model_state_dict"]
    )

    (
        test_metrics,
        test_y,
        test_probs,
        test_subject_true,
        test_subject_pred
    ) = evaluate_adversarial(
        model,
        test_loader
    )

    print(
        "\nFrozen-test result:"
    )

    print(
        json.dumps(
            test_metrics,
            indent=2
        )
    )

    return {
        "model": model,
        "lambda_adv": lambda_adv,
        "best_epoch": best_epoch,
        "best_val_macro_f1": best_score,
        "history": history,
        "test_metrics": test_metrics,
        "test_y": test_y,
        "test_probs": test_probs,
        "test_subject_true":
            test_subject_true,
        "test_subject_pred":
            test_subject_pred,
    }

# ============================================================
# 20. RUN BASELINE
# ============================================================

baseline_result = train_baseline()

# ============================================================
# 21. RUN ADVERSARIAL MODELS
# ============================================================

adversarial_results = []

for lambda_adv in LAMBDA_VALUES:

    result = train_adversarial(
        lambda_adv
    )

    adversarial_results.append(
        result
    )

# ============================================================
# 22. MODEL SELECTION
#
# IMPORTANT:
# Select ONLY using validation macro-F1.
# ============================================================

comparison_rows = []

comparison_rows.append({
    "experiment":
        "baseline",
    "lambda_adv":
        0.0,
    "best_epoch":
        baseline_result["best_epoch"],
    "val_macro_f1":
        baseline_result[
            "best_val_macro_f1"
        ],
    **{
        f"test_{k}": v
        for k, v
        in baseline_result[
            "test_metrics"
        ].items()
    }
})

for result in adversarial_results:

    comparison_rows.append({
        "experiment":
            "adversarial",
        "lambda_adv":
            result["lambda_adv"],
        "best_epoch":
            result["best_epoch"],
        "val_macro_f1":
            result[
                "best_val_macro_f1"
            ],
        **{
            f"test_{k}": v
            for k, v
            in result[
                "test_metrics"
            ].items()
        }
    })

comparison_df = pd.DataFrame(
    comparison_rows
)

comparison_df = comparison_df.sort_values(
    "val_macro_f1",
    ascending=False
).reset_index(drop=True)

print("\n" + "=" * 70)
print("STEP 6E MODEL COMPARISON")
print("=" * 70)

display(
    comparison_df[
        [
            "experiment",
            "lambda_adv",
            "best_epoch",
            "val_macro_f1",
            "test_accuracy",
            "test_balanced_accuracy",
            "test_macro_f1",
            "test_weighted_f1",
            "test_macro_auroc",
            "test_macro_auprc",
            "test_subject_accuracy",
        ]
    ]
)

# ============================================================
# 23. SELECT BEST MODEL BY VALIDATION MACRO-F1
# ============================================================

best_row = comparison_df.iloc[0]

print("\nValidation-selected model:")
print(
    best_row[
        [
            "experiment",
            "lambda_adv",
            "best_epoch",
            "val_macro_f1",
        ]
    ].to_string()
)

# ============================================================
# 24. CONFUSION MATRIX FOR SELECTED MODEL
# ============================================================

if best_row["experiment"] == "baseline":

    selected_result = baseline_result

else:

    selected_result = next(
        r
        for r in adversarial_results
        if abs(
            r["lambda_adv"]
            - best_row["lambda_adv"]
        ) < 1e-12
    )

selected_y = (
    selected_result["test_y"]
)

selected_probs = (
    selected_result["test_probs"]
)

selected_pred = (
    selected_probs.argmax(
        axis=1
    )
)

cm = confusion_matrix(
    selected_y,
    selected_pred,
    labels=list(range(N_CLASSES))
)

cm_path = (
    OUTPUT_ROOT /
    "step6e_selected_test_confusion_matrix.csv"
)

pd.DataFrame(
    cm,
    index=[
        f"true_{i}"
        for i in range(N_CLASSES)
    ],
    columns=[
        f"pred_{i}"
        for i in range(N_CLASSES)
    ]
).to_csv(
    cm_path
)

# ============================================================
# 25. SUBJECT-LEVEL TEST PERFORMANCE
# ============================================================

test_meta = h7[
    h7["split"].astype(str).str.lower()
    == "test"
].sort_values(
    ["recording_id", "window_index"]
).reset_index(drop=True)

assert len(test_meta) == len(
    selected_y
)

subject_rows = []

for subject, idxs in test_meta.groupby(
    "subject"
).groups.items():

    idxs = np.asarray(
        list(idxs)
    )

    y_s = selected_y[idxs]

    p_s = selected_probs[idxs]

    m = classification_metrics(
        y_s,
        p_s
    )

    subject_rows.append({
        "subject": int(subject),
        "n": int(len(idxs)),
        **m
    })

subject_df = pd.DataFrame(
    subject_rows
).sort_values(
    "subject"
)

print("\nSelected-model test performance by subject:")

display(subject_df)

# ============================================================
# 26. SAVE HISTORIES
# ============================================================

baseline_history_df = pd.DataFrame(
    baseline_result["history"]
)

baseline_history_df.to_csv(
    OUTPUT_ROOT /
    "step6e_baseline_history.csv",
    index=False
)

for result in adversarial_results:

    lambda_adv = result[
        "lambda_adv"
    ]

    pd.DataFrame(
        result["history"]
    ).to_csv(
        OUTPUT_ROOT /
        f"step6e_adversarial_lambda_{lambda_adv:.2f}_history.csv",
        index=False
    )

# ============================================================
# 27. SAVE TEST PREDICTIONS
# ============================================================

prediction_df = test_meta.copy()

prediction_df[
    "predicted_class"
] = selected_pred

prediction_df[
    "decoder_confidence"
] = selected_probs.max(
    axis=1
)

prediction_df[
    "correct"
] = (
    prediction_df[
        "predicted_class"
    ].to_numpy()
    ==
    prediction_df[
        "true_class"
    ].to_numpy()
)

prediction_path = (
    OUTPUT_ROOT /
    "step6e_selected_test_predictions.csv"
)

prediction_df.to_csv(
    prediction_path,
    index=False
)

# ============================================================
# 28. FINAL SUMMARY JSON
# ============================================================

summary = {

    "experiment":
        "Step 6E — Subject-Invariant "
        "Representation Learning",

    "seed":
        SEED,

    "device":
        str(DEVICE),

    "frozen_split": {

        "train_subjects":
            train_subjects,

        "val_subjects":
            val_subjects,

        "test_subjects":
            test_subjects,

        "train_samples":
            len(datasets["train"]),

        "val_samples":
            len(datasets["val"]),

        "test_samples":
            len(datasets["test"]),
    },

    "window_variant":
        VARIANT,

    "normalization":
        "Phase5H.3 per-window "
        "channel-wise z-score",

    "model": {

        "encoder":
            "4-layer dilated CNN",

        "embedding_dimension":
            EMBED_DIM,

        "character_classes":
            N_CLASSES,

        "subject_classes":
            len(all_subjects),

        "adversarial_method":
            "Gradient Reversal Layer",

        "lambda_values":
            LAMBDA_VALUES,
    },

    "selection_rule":
        "best validation macro-F1",

    "selected_model":
        best_row.to_dict(),

    "comparison":
        comparison_df.to_dict(
            orient="records"
        ),

    "selected_subject_results":
        subject_df.to_dict(
            orient="records"
        ),

    "outputs": {

        "comparison":
            str(
                OUTPUT_ROOT /
                "step6e_model_comparison.csv"
            ),

        "predictions":
            str(
                prediction_path
            ),

        "confusion_matrix":
            str(cm_path),

        "summary":
            str(
                OUTPUT_ROOT /
                "step6e_summary.json"
            ),
    }
}

# Save comparison separately
comparison_df.to_csv(
    OUTPUT_ROOT /
    "step6e_model_comparison.csv",
    index=False
)

with open(
    OUTPUT_ROOT /
    "step6e_summary.json",
    "w"
) as f:

    json.dump(
        summary,
        f,
        indent=2,
        default=lambda x:
            float(x)
            if isinstance(
                x,
                (np.float32, np.float64)
            )
            else x
    )

# ============================================================
# 29. FINAL REPORT
# ============================================================

print("\n" + "=" * 70)
print("STEP 6E COMPLETE")
print("=" * 70)

print(
    f"Selected experiment : "
    f"{best_row['experiment']}"
)

print(
    f"Selected lambda     : "
    f"{best_row['lambda_adv']}"
)

print(
    f"Validation macro-F1 : "
    f"{best_row['val_macro_f1']:.6f}"
)

print(
    f"Test macro-F1       : "
    f"{best_row['test_macro_f1']:.6f}"
)

print(
    f"Test balanced acc   : "
    f"{best_row['test_balanced_accuracy']:.6f}"
)

print(
    f"Test AUROC          : "
    f"{best_row['test_macro_auroc']:.6f}"
)

print(
    f"Test AUPRC           : "
    f"{best_row['test_macro_auprc']:.6f}"
)

if "test_subject_accuracy" in best_row:

    print(
        f"Test subject acc    : "
        f"{best_row['test_subject_accuracy']:.6f}"
    )

print("\nOutputs:")
print(
    OUTPUT_ROOT
)

print("\nStatus:")
print(
    "STEP 6E COMPLETE — "
    "evaluate against Step 6D before proceeding."
)
```

```python
# ============================================================
# NEUROSENTINEL — STEP 6F
# EEG–CHARACTER TEMPORAL ALIGNMENT / INFORMATION DIAGNOSIS
#
# Purpose:
#   Test whether character information is temporally shifted
#   relative to the current EEG event timestamp.
#
# Authoritative event source:
#   Phase 5 recording metadata
#
# Frozen:
#   Phase 6 subject split
#   Phase 5H.3 normalization
#   29-class character mapping
#
# Offsets:
#   -500,-400,-300,-200,-100,0,+100,+200,+300,+400,+500 ms
#
# Model:
#   Same lightweight 4-layer dilated CNN used in Step 6C/6D/6E
#
# Selection:
#   Validation Macro-F1 only
#
# Test:
#   Evaluated once after validation selection
# ============================================================

import os
import json
import random
import warnings
from pathlib import Path

import numpy as np
import pandas as pd

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    precision_score,
    recall_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
)

warnings.filterwarnings("ignore")

# ============================================================
# 1. CONFIGURATION
# ============================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

BASE = "/content/gdrive/MyDrive/NeuroSentinel"

PHASE5_META_ROOT = os.path.join(
    BASE,
    "06_phase5_utility_labels",
    "recordings"
)

WINDOW_ROOT = os.path.join(
    BASE,
    "08_phase6_temporal_windows"
)

PHASE6_MANIFEST = os.path.join(
    BASE,
    "07_phase6_subject_split",
    "phase6_subject_split_manifest_resolved.csv"
)

OUTPUT_ROOT = os.path.join(
    BASE,
    "results",
    "phase6f_temporal_alignment"
)

CHECKPOINT_ROOT = os.path.join(
    BASE,
    "checkpoints",
    "phase6f_temporal_alignment"
)

os.makedirs(OUTPUT_ROOT, exist_ok=True)
os.makedirs(CHECKPOINT_ROOT, exist_ok=True)

# Temporal offsets in milliseconds
OFFSETS_MS = [-500, -400, -300, -200, -100, 0,
               100, 200, 300, 400, 500]

# EEG sampling
FS = 50

# Baseline EEG window
N_CHANNELS = 61
N_TIME = 25

# Training
BATCH_SIZE = 256
LR = 1e-3
WEIGHT_DECAY = 1e-4
MAX_EPOCHS = 20
PATIENCE = 5

# Per-window channel-wise z-score
EPS = 1e-6

# ============================================================
# 2. FROZEN PHASE 6 SPLIT
# ============================================================

phase6 = pd.read_csv(PHASE6_MANIFEST)

split_by_recording = (
    phase6[["recording_id", "split"]]
    .drop_duplicates()
    .set_index("recording_id")["split"]
    .to_dict()
)

print("=" * 70)
print("STEP 6F — EEG–CHARACTER TEMPORAL ALIGNMENT")
print("=" * 70)

print("Device:", DEVICE)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))

print("\nFrozen split:")
print(
    phase6.groupby("split")["recording_id"]
    .nunique()
    .to_string()
)

# ============================================================
# 3. 29-CLASS CHARACTER MAPPING
# ============================================================

CLASS_NAMES = list("abcdefghijklmnopqrstuvwxyz") + [
    "<space>",
    "<number>",
    "<special>"
]

CLASS_TO_ID = {
    ch: i for i, ch in enumerate(CLASS_NAMES)
}

NUM_CLASSES = len(CLASS_NAMES)

assert NUM_CLASSES == 29

def normalize_target(x):
    """
    Map Phase 5 target character into the frozen 29-class space.
    """
    if pd.isna(x):
        return None

    s = str(x)

    if len(s) == 1 and s.isalpha():
        s = s.lower()
        if s in CLASS_TO_ID:
            return s

    if s == " ":
        return "<space>"

    # Known Phase 5 special categories if already represented
    if s.lower() in ["<space>", "space"]:
        return "<space>"

    if s.lower() in ["<number>", "number"]:
        return "<number>"

    if s.lower() in ["<special>", "special"]:
        return "<special>"

    return None


# ============================================================
# 4. FIVE KNOWN ANOMALOUS EVENTS
#
# IMPORTANT:
# These are removed AFTER constructing the temporal
# association from the COMPLETE Phase 5 event sequence.
# ============================================================

ANOMALOUS = {
    ("002_S1_task1", 950, "-"),
    ("011_S1_task1", 1295, "`"),
    ("012_S2_task1", 164, "`"),
    ("012_S2_task2", 1734, "`"),
    ("013_S1_task2", 2313, "`"),
}

print("\nKnown anomalous events:", len(ANOMALOUS))

# ============================================================
# 5. DISCOVER RECORDINGS
# ============================================================

recordings = sorted(split_by_recording.keys())

assert len(recordings) == 62, (
    f"Expected 62 recordings, found {len(recordings)}"
)

print("Recordings:", len(recordings))

# ============================================================
# 6. LOAD PHASE 5 METADATA + EEG WINDOWS
#
# IMPORTANT:
# Phase 5 metadata and EEG arrays are guaranteed to have the
# same row ordering. This is the authoritative alignment.
# ============================================================

recording_data = {}

for rid in recordings:

    split = split_by_recording[rid]

    meta_path = os.path.join(
        PHASE5_META_ROOT,
        f"{rid}_metadata.csv"
    )

    if not os.path.exists(meta_path):
        raise FileNotFoundError(meta_path)

    meta = pd.read_csv(meta_path)

    # Prefer Step 6 temporal-window baseline
    temporal_path = os.path.join(
        WINDOW_ROOT,
        split,
        rid,
        f"{rid}_baseline_windows.npy"
    )

    # Fallback to original Phase 5 windows
    fallback_path = os.path.join(
        PHASE5_META_ROOT,
        f"{rid}_windows.npy"
    )

    if os.path.exists(temporal_path):
        eeg_path = temporal_path
    elif os.path.exists(fallback_path):
        eeg_path = fallback_path
    else:
        raise FileNotFoundError(
            f"No EEG windows found for {rid}"
        )

    eeg = np.load(
        eeg_path,
        mmap_mode="r"
    )

    if eeg.ndim != 3:
        raise RuntimeError(
            f"{rid}: EEG array must be 3D, got {eeg.shape}"
        )

    if eeg.shape[1:] != (N_CHANNELS, N_TIME):
        raise RuntimeError(
            f"{rid}: unexpected EEG shape {eeg.shape}"
        )

    if len(meta) != eeg.shape[0]:
        raise RuntimeError(
            f"{rid}: metadata windows={len(meta)}, "
            f"EEG windows={eeg.shape[0]}"
        )

    if "timestamp" not in meta.columns:
        raise RuntimeError(
            f"{rid}: timestamp column missing"
        )

    if "target" not in meta.columns:
        raise RuntimeError(
            f"{rid}: target column missing"
        )

    recording_data[rid] = {
        "split": split,
        "meta": meta.reset_index(drop=True),
        "eeg": eeg,
        "eeg_path": eeg_path,
    }

print("\nPhase 5 metadata ↔ EEG integrity: PASSED")

# ============================================================
# 7. VERIFY COMPLETE EVENT SEQUENCE
# ============================================================

total_events = 0
total_anomalies = 0

for rid, d in recording_data.items():

    meta = d["meta"]

    total_events += len(meta)

    for row_idx, row in meta.iterrows():

        target = str(row["target"]) if not pd.isna(row["target"]) else ""

        if (rid, row_idx, target) in ANOMALOUS:
            total_anomalies += 1

print("Complete Phase 5 events:", total_events)
print("Known anomalous events found:", total_anomalies)

assert total_anomalies == 5, (
    f"Expected exactly 5 anomalies, found {total_anomalies}"
)

# ============================================================
# 8. BUILD TEMPORAL LABEL ASSOCIATIONS
#
# For each EEG event i:
#
#   shifted_time = EEG_timestamp[i] + offset
#
# For offset = 0:
#   target = same event
#
# For offset != 0:
#   target = nearest OTHER event in timestamp space
#
# This preserves the actual irregular timing of keystrokes.
# ============================================================

def build_offset_samples(offset_ms):

    rows = []

    offset_seconds = offset_ms / 1000.0

    for rid, d in recording_data.items():

        meta = d["meta"]

        timestamps = meta["timestamp"].astype(float).to_numpy()

        targets = meta["target"].astype(str).to_numpy()

        n = len(meta)

        # Ensure timestamps are monotonically increasing
        if not np.all(np.diff(timestamps) >= 0):
            raise RuntimeError(
                f"{rid}: timestamps are not monotonic"
            )

        for i in range(n):

            # ------------------------------------------------
            # Exclude anomalous EEG source events
            # only after the full sequence exists.
            # ------------------------------------------------

            source_target = targets[i]

            if (rid, i, source_target) in ANOMALOUS:
                continue

            if offset_ms == 0:

                target_idx = i

            else:

                desired_time = timestamps[i] + offset_seconds

                # Binary-search neighborhood
                pos = np.searchsorted(
                    timestamps,
                    desired_time
                )

                candidates = []

                if pos > 0:
                    candidates.append(pos - 1)

                if pos < n:
                    candidates.append(pos)

                # Search a few nearby candidates to make sure
                # the current event can be excluded correctly.
                lo = max(0, pos - 3)
                hi = min(n, pos + 4)

                candidates.extend(range(lo, hi))

                candidates = sorted(set(candidates))

                # Remove the source event itself.
                candidates = [
                    j for j in candidates
                    if j != i
                ]

                if not candidates:
                    continue

                target_idx = min(
                    candidates,
                    key=lambda j:
                    abs(timestamps[j] - desired_time)
                )

            target_raw = targets[target_idx]

            target_label = normalize_target(target_raw)

            if target_label is None:
                continue

            target_class = CLASS_TO_ID[target_label]

            rows.append({
                "recording_id": rid,
                "split": d["split"],
                "source_index": i,
                "target_index": int(target_idx),
                "source_timestamp": float(timestamps[i]),
                "target_timestamp": float(timestamps[target_idx]),
                "offset_ms": int(offset_ms),
                "true_label": target_label,
                "true_class": int(target_class),
            })

    return pd.DataFrame(rows)


# ============================================================
# 9. CREATE ALL OFFSET MANIFESTS
# ============================================================

offset_manifests = {}

print("\nBuilding temporal associations...")

for offset in OFFSETS_MS:

    df = build_offset_samples(offset)

    offset_manifests[offset] = df

    print(
        f"{offset:+4d} ms | "
        f"N={len(df):6d} | "
        f"train={sum(df.split=='train'):6d} | "
        f"val={sum(df.split=='val'):6d} | "
        f"test={sum(df.split=='test'):6d}"
    )

# ============================================================
# 10. SAVE ALIGNMENT MANIFESTS
# ============================================================

for offset, df in offset_manifests.items():

    path = os.path.join(
        OUTPUT_ROOT,
        f"step6f_alignment_{offset:+d}ms.csv"
    )

    df.to_csv(path, index=False)

# ============================================================
# 11. DATASET
# ============================================================

class TemporalEEGDataset(Dataset):

    def __init__(
        self,
        manifest,
        recording_data,
    ):

        self.manifest = manifest.reset_index(drop=True)
        self.recording_data = recording_data

    def __len__(self):
        return len(self.manifest)

    def __getitem__(self, idx):

        row = self.manifest.iloc[idx]

        rid = row["recording_id"]

        source_index = int(row["source_index"])

        eeg = self.recording_data[rid]["eeg"]

        x = np.asarray(
            eeg[source_index],
            dtype=np.float32
        )

        # ----------------------------------------------------
        # Exact Phase5H.3 normalization:
        # channel-wise z-score over 25 time samples
        # ----------------------------------------------------

        mean = x.mean(axis=1, keepdims=True)
        std = x.std(axis=1, keepdims=True)

        x = (x - mean) / (std + EPS)

        y = int(row["true_class"])

        return (
            torch.from_numpy(x),
            torch.tensor(y, dtype=torch.long)
        )


# ============================================================
# 12. MODEL
# ============================================================

class EEGCharacterCNN(nn.Module):

    def __init__(self):

        super().__init__()

        self.encoder = nn.Sequential(

            nn.Conv1d(
                61, 64,
                kernel_size=5,
                padding=2
            ),
            nn.ReLU(),

            nn.Conv1d(
                64, 64,
                kernel_size=5,
                dilation=2,
                padding=4
            ),
            nn.ReLU(),

            nn.Conv1d(
                64, 64,
                kernel_size=5,
                dilation=4,
                padding=8
            ),
            nn.ReLU(),

            nn.Conv1d(
                64, 64,
                kernel_size=5,
                dilation=8,
                padding=16
            ),
            nn.ReLU(),
        )

        self.pool = nn.AdaptiveAvgPool1d(1)

        self.classifier = nn.Linear(
            64,
            NUM_CLASSES
        )

    def forward(self, x):

        z = self.encoder(x)

        z = self.pool(z)

        z = z.squeeze(-1)

        return self.classifier(z)


# Parameter count
tmp_model = EEGCharacterCNN()

n_params = sum(
    p.numel()
    for p in tmp_model.parameters()
    if p.requires_grad
)

print("\nTrainable parameters:", n_params)

del tmp_model

# ============================================================
# 13. CLASS WEIGHTS
# ============================================================

def compute_class_weights(manifest):

    counts = (
        manifest["true_class"]
        .value_counts()
        .reindex(
            range(NUM_CLASSES),
            fill_value=0
        )
        .to_numpy(dtype=np.float64)
    )

    # Inverse square-root weighting
    weights = np.zeros(NUM_CLASSES, dtype=np.float64)

    nonzero = counts > 0

    weights[nonzero] = 1.0 / np.sqrt(
        counts[nonzero]
    )

    if nonzero.any():

        weights[nonzero] /= weights[nonzero].mean()

    # Zero-frequency classes receive zero loss weight
    return torch.tensor(
        weights,
        dtype=torch.float32,
        device=DEVICE
    )


# ============================================================
# 14. METRICS
# ============================================================

def calculate_metrics(
    y_true,
    y_pred,
    probs,
):

    result = {}

    result["accuracy"] = float(
        accuracy_score(y_true, y_pred)
    )

    result["balanced_accuracy"] = float(
        balanced_accuracy_score(y_true, y_pred)
    )

    result["macro_precision"] = float(
        precision_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        )
    )

    result["macro_recall"] = float(
        recall_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        )
    )

    result["macro_f1"] = float(
        f1_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        )
    )

    result["weighted_f1"] = float(
        f1_score(
            y_true,
            y_pred,
            average="weighted",
            zero_division=0
        )
    )

    # Macro one-vs-rest AUROC/AUPRC
    aucs = []
    aps = []

    y_true_arr = np.asarray(y_true)
    probs_arr = np.asarray(probs)

    for c in range(NUM_CLASSES):

        y_bin = (
            y_true_arr == c
        ).astype(int)

        if y_bin.sum() == 0:
            continue

        if y_bin.sum() == len(y_bin):
            continue

        try:
            aucs.append(
                roc_auc_score(
                    y_bin,
                    probs_arr[:, c]
                )
            )
        except Exception:
            pass

        try:
            aps.append(
                average_precision_score(
                    y_bin,
                    probs_arr[:, c]
                )
            )
        except Exception:
            pass

    result["macro_auroc"] = (
        float(np.mean(aucs))
        if aucs else float("nan")
    )

    result["macro_auprc"] = (
        float(np.mean(aps))
        if aps else float("nan")
    )

    return result


# ============================================================
# 15. TRAIN / EVALUATE ONE OFFSET
# ============================================================

def run_offset(offset_ms, manifest):

    print("\n" + "=" * 70)
    print(f"OFFSET {offset_ms:+d} ms")
    print("=" * 70)

    train_df = manifest[
        manifest["split"] == "train"
    ].reset_index(drop=True)

    val_df = manifest[
        manifest["split"] == "val"
    ].reset_index(drop=True)

    test_df = manifest[
        manifest["split"] == "test"
    ].reset_index(drop=True)

    print(
        f"N train={len(train_df)} | "
        f"val={len(val_df)} | "
        f"test={len(test_df)}"
    )

    train_ds = TemporalEEGDataset(
        train_df,
        recording_data
    )

    val_ds = TemporalEEGDataset(
        val_df,
        recording_data
    )

    test_ds = TemporalEEGDataset(
        test_df,
        recording_data
    )

    train_loader = DataLoader(
        train_ds,
        batch_size=BATCH_SIZE,
        shuffle=True,
        num_workers=0,
        pin_memory=torch.cuda.is_available()
    )

    val_loader = DataLoader(
        val_ds,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=0,
        pin_memory=torch.cuda.is_available()
    )

    test_loader = DataLoader(
        test_ds,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=0,
        pin_memory=torch.cuda.is_available()
    )

    model = EEGCharacterCNN().to(DEVICE)

    class_weights = compute_class_weights(
        train_df
    )

    criterion = nn.CrossEntropyLoss(
        weight=class_weights
    )

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=LR,
        weight_decay=WEIGHT_DECAY
    )

    best_val_f1 = -np.inf
    best_epoch = -1
    patience_counter = 0

    checkpoint_path = os.path.join(
        CHECKPOINT_ROOT,
        f"phase6f_{offset_ms:+d}ms_best.pt"
    )

    history = []

    # --------------------------------------------------------
    # Evaluation helper
    # --------------------------------------------------------

    def evaluate(loader):

        model.eval()

        total_loss = 0.0
        total_n = 0

        all_true = []
        all_pred = []
        all_probs = []

        with torch.no_grad():

            for x, y in loader:

                x = x.to(
                    DEVICE,
                    non_blocking=True
                )

                y = y.to(
                    DEVICE,
                    non_blocking=True
                )

                logits = model(x)

                loss = criterion(
                    logits,
                    y
                )

                probs = torch.softmax(
                    logits,
                    dim=1
                )

                pred = probs.argmax(
                    dim=1
                )

                bs = y.size(0)

                total_loss += (
                    loss.item() * bs
                )

                total_n += bs

                all_true.append(
                    y.cpu().numpy()
                )

                all_pred.append(
                    pred.cpu().numpy()
                )

                all_probs.append(
                    probs.cpu().numpy()
                )

        y_true = np.concatenate(all_true)
        y_pred = np.concatenate(all_pred)
        probs = np.concatenate(all_probs)

        metrics = calculate_metrics(
            y_true,
            y_pred,
            probs
        )

        metrics["loss"] = (
            total_loss / total_n
        )

        return (
            metrics,
            y_true,
            y_pred,
            probs
        )

    # --------------------------------------------------------
    # Training
    # --------------------------------------------------------

    for epoch in range(1, MAX_EPOCHS + 1):

        model.train()

        running_loss = 0.0
        n_seen = 0

        for x, y in train_loader:

            x = x.to(
                DEVICE,
                non_blocking=True
            )

            y = y.to(
                DEVICE,
                non_blocking=True
            )

            optimizer.zero_grad(
                set_to_none=True
            )

            logits = model(x)

            loss = criterion(
                logits,
                y
            )

            loss.backward()

            optimizer.step()

            bs = y.size(0)

            running_loss += (
                loss.item() * bs
            )

            n_seen += bs

        train_loss = (
            running_loss / n_seen
        )

        val_metrics, _, _, _ = evaluate(
            val_loader
        )

        record = {
            "offset_ms": offset_ms,
            "epoch": epoch,
            "train_loss": train_loss,
            "val_loss": val_metrics["loss"],
            "val_accuracy": val_metrics["accuracy"],
            "val_balanced_accuracy":
                val_metrics["balanced_accuracy"],
            "val_macro_f1":
                val_metrics["macro_f1"],
            "val_macro_auroc":
                val_metrics["macro_auroc"],
            "val_macro_auprc":
                val_metrics["macro_auprc"],
        }

        history.append(record)

        print(
            f"Epoch {epoch:02d} | "
            f"train_loss={train_loss:.4f} | "
            f"val_loss={val_metrics['loss']:.4f} | "
            f"val_acc={val_metrics['accuracy']:.4f} | "
            f"val_bal={val_metrics['balanced_accuracy']:.4f} | "
            f"val_macroF1={val_metrics['macro_f1']:.6f}"
        )

        current_f1 = val_metrics["macro_f1"]

        if current_f1 > best_val_f1:

            best_val_f1 = current_f1

            best_epoch = epoch

            patience_counter = 0

            torch.save(
                {
                    "model_state_dict":
                        model.state_dict(),
                    "offset_ms":
                        offset_ms,
                    "epoch":
                        epoch,
                    "val_macro_f1":
                        current_f1,
                    "seed":
                        SEED,
                },
                checkpoint_path
            )

        else:

            patience_counter += 1

            if patience_counter >= PATIENCE:

                print(
                    f"Early stopping at epoch {epoch}"
                )

                break

    # --------------------------------------------------------
    # Reload best validation checkpoint
    # --------------------------------------------------------

    checkpoint = torch.load(
        checkpoint_path,
        map_location=DEVICE
    )

    model.load_state_dict(
        checkpoint["model_state_dict"]
    )

    # --------------------------------------------------------
    # Frozen validation evaluation
    # --------------------------------------------------------

    best_val_metrics, _, _, _ = evaluate(
        val_loader
    )

    # --------------------------------------------------------
    # Test evaluation — ONLY NOW
    # --------------------------------------------------------

    test_metrics, y_test, pred_test, probs_test = evaluate(
        test_loader
    )

    # --------------------------------------------------------
    # Save history
    # --------------------------------------------------------

    history_df = pd.DataFrame(history)

    history_path = os.path.join(
        OUTPUT_ROOT,
        f"phase6f_{offset_ms:+d}ms_history.csv"
    )

    history_df.to_csv(
        history_path,
        index=False
    )

    # --------------------------------------------------------
    # Save test predictions
    # --------------------------------------------------------

    pred_df = test_df.copy()

    pred_df["predicted_class"] = pred_test

    pred_df["predicted_label"] = [
        CLASS_NAMES[x]
        for x in pred_test
    ]

    pred_df["confidence"] = (
        probs_test.max(axis=1)
    )

    pred_path = os.path.join(
        OUTPUT_ROOT,
        f"phase6f_{offset_ms:+d}ms_test_predictions.csv"
    )

    pred_df.to_csv(
        pred_path,
        index=False
    )

    # --------------------------------------------------------
    # Confusion matrix
    # --------------------------------------------------------

    cm = confusion_matrix(
        y_test,
        pred_test,
        labels=list(range(NUM_CLASSES))
    )

    cm_path = os.path.join(
        OUTPUT_ROOT,
        f"phase6f_{offset_ms:+d}ms_confusion_matrix.csv"
    )

    pd.DataFrame(
        cm,
        index=CLASS_NAMES,
        columns=CLASS_NAMES
    ).to_csv(cm_path)

    # --------------------------------------------------------
    # Result summary
    # --------------------------------------------------------

    result = {
        "offset_ms": offset_ms,
        "best_epoch": best_epoch,

        "train_samples": len(train_df),
        "val_samples": len(val_df),
        "test_samples": len(test_df),

        "val_loss": best_val_metrics["loss"],
        "val_accuracy":
            best_val_metrics["accuracy"],
        "val_balanced_accuracy":
            best_val_metrics["balanced_accuracy"],
        "val_macro_f1":
            best_val_metrics["macro_f1"],
        "val_macro_auroc":
            best_val_metrics["macro_auroc"],
        "val_macro_auprc":
            best_val_metrics["macro_auprc"],

        "test_loss":
            test_metrics["loss"],
        "test_accuracy":
            test_metrics["accuracy"],
        "test_balanced_accuracy":
            test_metrics["balanced_accuracy"],
        "test_macro_precision":
            test_metrics["macro_precision"],
        "test_macro_recall":
            test_metrics["macro_recall"],
        "test_macro_f1":
            test_metrics["macro_f1"],
        "test_weighted_f1":
            test_metrics["weighted_f1"],
        "test_macro_auroc":
            test_metrics["macro_auroc"],
        "test_macro_auprc":
            test_metrics["macro_auprc"],
    }

    result_path = os.path.join(
        OUTPUT_ROOT,
        f"phase6f_{offset_ms:+d}ms_result.json"
    )

    with open(
        result_path,
        "w"
    ) as f:

        json.dump(
            result,
            f,
            indent=2
        )

    print(
        f"\nBEST epoch={best_epoch}"
    )

    print(
        f"VAL  Macro-F1="
        f"{best_val_f1:.6f}"
    )

    print(
        f"TEST Macro-F1="
        f"{test_metrics['macro_f1']:.6f} | "
        f"Balanced Acc="
        f"{test_metrics['balanced_accuracy']:.6f} | "
        f"AUROC="
        f"{test_metrics['macro_auroc']:.6f}"
    )

    del model
    torch.cuda.empty_cache()

    return result


# ============================================================
# 16. RUN ALL TEMPORAL OFFSETS
# ============================================================

all_results = []

for offset in OFFSETS_MS:

    result = run_offset(
        offset,
        offset_manifests[offset]
    )

    all_results.append(result)


# ============================================================
# 17. COMPARISON TABLE
# ============================================================

comparison = pd.DataFrame(
    all_results
)

comparison = comparison.sort_values(
    "val_macro_f1",
    ascending=False
).reset_index(drop=True)

comparison_path = os.path.join(
    OUTPUT_ROOT,
    "phase6f_temporal_alignment_comparison.csv"
)

comparison.to_csv(
    comparison_path,
    index=False
)

print("\n" + "=" * 90)
print("STEP 6F — FINAL COMPARISON")
print("=" * 90)

display_cols = [
    "offset_ms",
    "best_epoch",
    "train_samples",
    "val_samples",
    "test_samples",
    "val_accuracy",
    "val_balanced_accuracy",
    "val_macro_f1",
    "val_macro_auroc",
    "val_macro_auprc",
    "test_accuracy",
    "test_balanced_accuracy",
    "test_macro_f1",
    "test_macro_auroc",
    "test_macro_auprc",
]

print(
    comparison[display_cols].to_string(
        index=False
    )
)

# ============================================================
# 18. VALIDATION-SELECTED OFFSET
# ============================================================

best_row = comparison.iloc[0]

selected_offset = int(
    best_row["offset_ms"]
)

print("\n" + "=" * 70)
print("VALIDATION-SELECTED OFFSET")
print("=" * 70)

print(
    f"Selected offset: {selected_offset:+d} ms"
)

print(
    f"Validation Macro-F1: "
    f"{best_row['val_macro_f1']:.6f}"
)

print(
    f"Corresponding frozen Test Macro-F1: "
    f"{best_row['test_macro_f1']:.6f}"
)

# ============================================================
# 19. ZERO-OFFSET REFERENCE
# ============================================================

zero_row = comparison[
    comparison["offset_ms"] == 0
].iloc[0]

print("\nZero-offset reference:")

print(
    f"Val Macro-F1  : "
    f"{zero_row['val_macro_f1']:.6f}"
)

print(
    f"Test Macro-F1 : "
    f"{zero_row['test_macro_f1']:.6f}"
)

print(
    f"Test AUROC    : "
    f"{zero_row['test_macro_auroc']:.6f}"
)

print(
    f"Test AUPRC    : "
    f"{zero_row['test_macro_auprc']:.6f}"
)

# ============================================================
# 20. IMPROVEMENT RELATIVE TO ZERO OFFSET
# ============================================================

delta_val = (
    best_row["val_macro_f1"]
    - zero_row["val_macro_f1"]
)

delta_test = (
    best_row["test_macro_f1"]
    - zero_row["test_macro_f1"]
)

print("\nDifference from zero-offset baseline:")

print(
    f"Validation Macro-F1 Δ: "
    f"{delta_val:+.6f}"
)

print(
    f"Test Macro-F1 Δ: "
    f"{delta_test:+.6f}"
)

# ============================================================
# 21. SUMMARY JSON
# ============================================================

summary = {
    "experiment": "Phase 6F — EEG–Character Temporal Alignment",
    "seed": SEED,
    "device": str(DEVICE),
    "num_recordings": len(recordings),
    "num_classes": NUM_CLASSES,
    "class_names": CLASS_NAMES,

    "offsets_ms": OFFSETS_MS,

    "authoritative_event_source":
        "Phase 5 recording metadata",

    "normalization":
        "per-window channel-wise z-score over 25 samples",

    "model":
        "4-layer dilated CNN + global average pooling + 29-class classifier",

    "parameter_count": n_params,

    "selection_metric":
        "validation macro-F1",

    "selected_offset_ms":
        selected_offset,

    "selected_val_macro_f1":
        float(best_row["val_macro_f1"]),

    "selected_test_macro_f1":
        float(best_row["test_macro_f1"]),

    "zero_offset_val_macro_f1":
        float(zero_row["val_macro_f1"]),

    "zero_offset_test_macro_f1":
        float(zero_row["test_macro_f1"]),

    "selected_minus_zero_val_macro_f1":
        float(delta_val),

    "selected_minus_zero_test_macro_f1":
        float(delta_test),

    "anomalous_events_excluded":
        5,

    "comparison_csv":
        comparison_path,
}

summary_path = os.path.join(
    OUTPUT_ROOT,
    "phase6f_temporal_alignment_summary.json"
)

with open(
    summary_path,
    "w"
) as f:

    json.dump(
        summary,
        f,
        indent=2
    )

# ============================================================
# 22. FINAL STATUS
# ============================================================

print("\n" + "=" * 70)
print("STEP 6F COMPLETE")
print("=" * 70)

print(
    "Comparison:",
    comparison_path
)

print(
    "Summary:",
    summary_path
)

print(
    "\nNext decision:"
)

if delta_test > 0.005:
    print(
        "TEMPORAL ALIGNMENT MAY BE MEANINGFUL — "
        "inspect the selected offset and subject-level results."
    )
else:
    print(
        "TEMPORAL OFFSET DOES NOT SHOW A MATERIAL "
        "GENERALIZATION IMPROVEMENT."
    )
```

```python
# ============================================================
# NeuroSentinel — Step 6F
# BATCH-WISE RESUMABLE TEMPORAL ALIGNMENT TRAINING
# ============================================================

import os
import json
import math
import random
import copy
import time
import warnings
from pathlib import Path

import numpy as np
import pandas as pd

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    precision_score,
    recall_score,
    roc_auc_score,
    average_precision_score,
)

warnings.filterwarnings("ignore")

# ============================================================
# 1. CONFIGURATION
# ============================================================

SEED = 42

OFFSETS_MS = [
    -500, -400, -300, -200, -100,
       0,
     100,  200,  300,  400,  500
]

BATCH_SIZE = 256
LR = 1e-3
WEIGHT_DECAY = 1e-4
MAX_EPOCHS = 20
PATIENCE = 5

N_CLASSES = 29
N_CHANNELS = 61
WINDOW_SAMPLES = 25
EPS = 1e-6

BASE = Path(
    "/content/gdrive/MyDrive/NeuroSentinel"
)

PHASE6_MANIFEST = (
    BASE /
    "07_phase6_subject_split" /
    "phase6_subject_split_manifest_resolved.csv"
)

PHASE6_MANIFEST_FALLBACK = (
    BASE /
    "07_phase6_subject_split" /
    "phase6_subject_split_manifest.csv"
)

PHASE5_ROOT = (
    BASE /
    "06_phase5_utility_labels" /
    "recordings"
)

TEMPORAL_ROOT = (
    BASE /
    "08_phase6_temporal_windows"
)

RESULT_ROOT = (
    BASE /
    "results" /
    "phase6f_temporal_alignment"
)

CHECKPOINT_ROOT = (
    BASE /
    "checkpoints" /
    "phase6f_temporal_alignment"
)

RESULT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

CHECKPOINT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

# ============================================================
# 2. GPU CHECK
# ============================================================

if not torch.cuda.is_available():
    raise RuntimeError(
        "\nGPU IS NOT AVAILABLE.\n"
        "Reconnect a GPU runtime and run this SAME cell again.\n"
        "Existing checkpoints will be preserved.\n"
    )

DEVICE = torch.device("cuda")

print("=" * 70)
print("NEUROSENTINEL — STEP 6F")
print("BATCH-WISE RESUMABLE TRAINING")
print("=" * 70)
print("Device:", DEVICE)
print("GPU:", torch.cuda.get_device_name(0))
print("PyTorch:", torch.__version__)
print()

# ============================================================
# 3. REPRODUCIBILITY
# ============================================================

def seed_everything(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

seed_everything(SEED)

# ============================================================
# 4. LOAD FROZEN PHASE 6 SPLIT
# ============================================================

if PHASE6_MANIFEST.exists():
    split_df = pd.read_csv(
        PHASE6_MANIFEST
    )
elif PHASE6_MANIFEST_FALLBACK.exists():
    split_df = pd.read_csv(
        PHASE6_MANIFEST_FALLBACK
    )
else:
    raise FileNotFoundError(
        "Phase 6 split manifest not found."
    )

required_cols = {
    "recording_id",
    "subject",
    "split"
}

missing = (
    required_cols -
    set(split_df.columns)
)

if missing:
    raise RuntimeError(
        f"Missing columns: {missing}"
    )

TRAIN_RECORDINGS = sorted(
    split_df.loc[
        split_df["split"].str.lower() == "train",
        "recording_id"
    ].unique()
)

VAL_RECORDINGS = sorted(
    split_df.loc[
        split_df["split"].str.lower() == "val",
        "recording_id"
    ].unique()
)

TEST_RECORDINGS = sorted(
    split_df.loc[
        split_df["split"].str.lower() == "test",
        "recording_id"
    ].unique()
)

print("Frozen split:")
print("  Train:", len(TRAIN_RECORDINGS))
print("  Val  :", len(VAL_RECORDINGS))
print("  Test :", len(TEST_RECORDINGS))
print()

# ============================================================
# 5. KNOWN ANOMALOUS EVENTS
# ============================================================

ANOMALOUS_ROWS = {
    ("002_S1_task1", 950),
    ("011_S1_task1", 1295),
    ("012_S2_task1", 164),
    ("012_S2_task2", 1734),
    ("013_S1_task2", 2313),
}

# ============================================================
# 6. LABEL MAPPING
# ============================================================

LABELS = (
    [chr(ord("a") + i) for i in range(26)]
    + ["<space>", "<number>", "<special>"]
)

LABEL_TO_ID = {
    label: i
    for i, label in enumerate(LABELS)
}

# ============================================================
# 7. LOAD PHASE 5 METADATA + BASELINE WINDOWS
# ============================================================

def get_phase5_metadata(recording_id):

    path = (
        PHASE5_ROOT /
        f"{recording_id}_metadata.csv"
    )

    if not path.exists():
        raise FileNotFoundError(
            f"Missing metadata: {path}"
        )

    return pd.read_csv(
        path
    ).reset_index(drop=True)


def get_windows(recording_id, split):

    temporal_path = (
        TEMPORAL_ROOT /
        split /
        recording_id /
        f"{recording_id}_baseline_windows.npy"
    )

    phase5_path = (
        PHASE5_ROOT /
        f"{recording_id}_windows.npy"
    )

    if temporal_path.exists():
        return np.load(
            temporal_path,
            mmap_mode="r"
        )

    if phase5_path.exists():
        return np.load(
            phase5_path,
            mmap_mode="r"
        )

    raise FileNotFoundError(
        f"No baseline windows for {recording_id}"
    )


recordings = {}

all_recordings = sorted(
    set(TRAIN_RECORDINGS)
    |
    set(VAL_RECORDINGS)
    |
    set(TEST_RECORDINGS)
)

print(
    f"Loading {len(all_recordings)} recordings..."
)

for rid in all_recordings:

    if rid in TRAIN_RECORDINGS:
        split = "train"
    elif rid in VAL_RECORDINGS:
        split = "val"
    else:
        split = "test"

    meta = get_phase5_metadata(
        rid
    )

    windows = get_windows(
        rid,
        split
    )

    if len(meta) != len(windows):
        raise RuntimeError(
            f"{rid}: metadata={len(meta)} "
            f"windows={len(windows)}"
        )

    recordings[rid] = {
        "split": split,
        "meta": meta,
        "windows": windows
    }

print("✓ Phase 5 metadata ↔ EEG integrity passed")
print()

# ============================================================
# 8. BUILD / LOAD TEMPORAL ASSOCIATIONS
# ============================================================

def build_associations(
    offset_ms
):

    offset_sec = (
        offset_ms / 1000.0
    )

    associations = []

    for rid in all_recordings:

        meta = recordings[rid]["meta"]

        timestamps = (
            meta["timestamp"]
            .astype(float)
            .to_numpy()
        )

        n = len(meta)

        valid = np.ones(
            n,
            dtype=bool
        )

        # Exclude known anomalous SOURCE events
        for i in range(n):
            if (
                rid,
                i
            ) in ANOMALOUS_ROWS:
                valid[i] = False

        # ----------------------------------------------------
        # 0 ms = same event
        # ----------------------------------------------------

        if offset_ms == 0:

            for source_idx in range(n):

                if not valid[source_idx]:
                    continue

                target_idx = source_idx

                label = meta.iloc[
                    target_idx
                ]["target"]

                if label not in LABEL_TO_ID:
                    continue

                associations.append(
                    (
                        rid,
                        source_idx,
                        target_idx,
                        LABEL_TO_ID[label]
                    )
                )

            continue

        # ----------------------------------------------------
        # Non-zero offset
        # ----------------------------------------------------

        for source_idx in range(n):

            if not valid[source_idx]:
                continue

            desired = (
                timestamps[source_idx]
                + offset_sec
            )

            pos = np.searchsorted(
                timestamps,
                desired
            )

            candidates = []

            if pos < n:
                candidates.append(pos)

            if pos > 0:
                candidates.append(pos - 1)

            best_idx = None
            best_error = np.inf

            for candidate in candidates:

                if candidate == source_idx:
                    continue

                if not valid[candidate]:
                    continue

                error = abs(
                    timestamps[candidate]
                    - desired
                )

                if error < best_error:

                    best_error = error
                    best_idx = candidate

            if best_idx is None:
                continue

            label = meta.iloc[
                best_idx
            ]["target"]

            if label not in LABEL_TO_ID:
                continue

            associations.append(
                (
                    rid,
                    source_idx,
                    best_idx,
                    LABEL_TO_ID[label]
                )
            )

    return associations


def association_file(
    offset_ms
):

    return (
        CHECKPOINT_ROOT /
        f"associations_{offset_ms:+d}ms.npz"
    )


association_cache = {}

print("=" * 70)
print("TEMPORAL ASSOCIATIONS")
print("=" * 70)

for offset_ms in OFFSETS_MS:

    path = association_file(
        offset_ms
    )

    if path.exists():

        data = np.load(
            path,
            allow_pickle=True
        )

        assoc = list(
            zip(
                data["recording_id"].astype(str),
                data["source_idx"].astype(int),
                data["target_idx"].astype(int),
                data["label"].astype(int)
            )
        )

        association_cache[
            offset_ms
        ] = assoc

        print(
            f"{offset_ms:+4d} ms: "
            f"{len(assoc):,} cached"
        )

    else:

        assoc = build_associations(
            offset_ms
        )

        association_cache[
            offset_ms
        ] = assoc

        np.savez_compressed(
            path,
            recording_id=np.array(
                [x[0] for x in assoc]
            ),
            source_idx=np.array(
                [x[1] for x in assoc],
                dtype=np.int64
            ),
            target_idx=np.array(
                [x[2] for x in assoc],
                dtype=np.int64
            ),
            label=np.array(
                [x[3] for x in assoc],
                dtype=np.int64
            )
        )

        print(
            f"{offset_ms:+4d} ms: "
            f"{len(assoc):,} created"
        )

print()

# ============================================================
# 9. SPLIT ASSOCIATIONS
# ============================================================

def split_associations(
    associations
):

    train_set = set(
        TRAIN_RECORDINGS
    )

    val_set = set(
        VAL_RECORDINGS
    )

    test_set = set(
        TEST_RECORDINGS
    )

    train = []
    val = []
    test = []

    for item in associations:

        rid = item[0]

        if rid in train_set:
            train.append(item)

        elif rid in val_set:
            val.append(item)

        elif rid in test_set:
            test.append(item)

    return train, val, test

# ============================================================
# 10. DATASET
# ============================================================

class TemporalDataset(
    Dataset
):

    def __init__(
        self,
        associations
    ):
        self.associations = associations

    def __len__(self):
        return len(
            self.associations
        )

    def __getitem__(
        self,
        idx
    ):

        rid, source_idx, _, label = (
            self.associations[idx]
        )

        x = np.asarray(
            recordings[rid]["windows"][
                source_idx
            ],
            dtype=np.float32
        )

        # Exact Phase 5H.3 normalization
        mean = x.mean(
            axis=1,
            keepdims=True
        )

        std = x.std(
            axis=1,
            keepdims=True
        )

        x = (
            x - mean
        ) / (
            std + EPS
        )

        return (
            torch.from_numpy(x),
            torch.tensor(
                label,
                dtype=torch.long
            )
        )

# ============================================================
# 11. SAME STEP 6F CNN
# ============================================================

class CharacterCNN(
    nn.Module
):

    def __init__(self):

        super().__init__()

        self.features = nn.Sequential(

            nn.Conv1d(
                61,
                64,
                kernel_size=5,
                padding=2
            ),
            nn.ReLU(),

            nn.Conv1d(
                64,
                64,
                kernel_size=5,
                dilation=2,
                padding=4
            ),
            nn.ReLU(),

            nn.Conv1d(
                64,
                64,
                kernel_size=5,
                dilation=4,
                padding=8
            ),
            nn.ReLU(),

            nn.Conv1d(
                64,
                64,
                kernel_size=5,
                dilation=8,
                padding=16
            ),
            nn.ReLU()
        )

        self.classifier = nn.Linear(
            64,
            N_CLASSES
        )

    def forward(self, x):

        x = self.features(x)

        x = x.mean(
            dim=-1
        )

        return self.classifier(x)


model_check = CharacterCNN()

print(
    "Trainable parameters:",
    sum(
        p.numel()
        for p in model_check.parameters()
        if p.requires_grad
    )
)

del model_check

# ============================================================
# 12. CLASS WEIGHTS
# ============================================================

def make_class_weights(
    train_assoc
):

    counts = np.zeros(
        N_CLASSES,
        dtype=np.float64
    )

    for item in train_assoc:

        counts[
            int(item[3])
        ] += 1

    weights = np.zeros(
        N_CLASSES,
        dtype=np.float64
    )

    present = counts > 0

    weights[present] = (
        1.0 /
        np.sqrt(
            counts[present]
        )
    )

    if present.any():

        weights[present] /= (
            weights[present].mean()
        )

    return (
        torch.tensor(
            weights,
            dtype=torch.float32,
            device=DEVICE
        ),
        counts
    )

# ============================================================
# 13. METRICS
# ============================================================

def calculate_metrics(
    y_true,
    y_pred,
    y_prob
):

    result = {}

    result["accuracy"] = (
        accuracy_score(
            y_true,
            y_pred
        )
    )

    result["balanced_accuracy"] = (
        balanced_accuracy_score(
            y_true,
            y_pred
        )
    )

    result["macro_f1"] = (
        f1_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        )
    )

    result["weighted_f1"] = (
        f1_score(
            y_true,
            y_pred,
            average="weighted",
            zero_division=0
        )
    )

    result["macro_precision"] = (
        precision_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        )
    )

    result["macro_recall"] = (
        recall_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        )
    )

    try:

        onehot = np.eye(
            N_CLASSES
        )[np.asarray(y_true)]

        result["macro_auroc"] = (
            roc_auc_score(
                onehot,
                y_prob,
                average="macro",
                multi_class="ovr"
            )
        )

    except Exception:

        result["macro_auroc"] = np.nan

    try:

        onehot = np.eye(
            N_CLASSES
        )[np.asarray(y_true)]

        result["macro_auprc"] = (
            average_precision_score(
                onehot,
                y_prob,
                average="macro"
            )
        )

    except Exception:

        result["macro_auprc"] = np.nan

    return result

# ============================================================
# 14. EVALUATION
# ============================================================

@torch.no_grad()
def evaluate(
    model,
    loader,
    criterion
):

    model.eval()

    total_loss = 0.0
    total_n = 0

    all_true = []
    all_pred = []
    all_prob = []

    for x, y in loader:

        x = x.to(
            DEVICE,
            non_blocking=True
        )

        y = y.to(
            DEVICE,
            non_blocking=True
        )

        logits = model(x)

        loss = criterion(
            logits,
            y
        )

        probs = torch.softmax(
            logits,
            dim=1
        )

        pred = probs.argmax(
            dim=1
        )

        bs = y.size(0)

        total_loss += (
            loss.item() * bs
        )

        total_n += bs

        all_true.append(
            y.cpu().numpy()
        )

        all_pred.append(
            pred.cpu().numpy()
        )

        all_prob.append(
            probs.cpu().numpy()
        )

    y_true = np.concatenate(
        all_true
    )

    y_pred = np.concatenate(
        all_pred
    )

    y_prob = np.concatenate(
        all_prob
    )

    metrics = calculate_metrics(
        y_true,
        y_pred,
        y_prob
    )

    metrics["loss"] = (
        total_loss /
        max(total_n, 1)
    )

    return metrics

# ============================================================
# 15. DETERMINISTIC BATCH ORDER
# ============================================================

def get_epoch_batches(
    n_samples,
    epoch
):

    generator = torch.Generator()

    generator.manual_seed(
        SEED + epoch
    )

    permutation = torch.randperm(
        n_samples,
        generator=generator
    ).numpy()

    return [
        permutation[
            start:
            min(
                start + BATCH_SIZE,
                n_samples
            )
        ]
        for start in range(
            0,
            n_samples,
            BATCH_SIZE
        )
    ]

# ============================================================
# 16. RNG CHECKPOINTING
# ============================================================

def capture_rng():

    state = {
        "python":
            random.getstate(),

        "numpy":
            np.random.get_state(),

        "torch":
            torch.get_rng_state()
    }

    if torch.cuda.is_available():

        state["cuda"] = (
            torch.cuda.get_rng_state_all()
        )

    return state


def restore_rng(
    state
):

    if state is None:
        return

    random.setstate(
        state["python"]
    )

    np.random.set_state(
        state["numpy"]
    )

    torch.set_rng_state(
        state["torch"]
    )

    if (
        torch.cuda.is_available()
        and "cuda" in state
    ):

        torch.cuda.set_rng_state_all(
            state["cuda"]
        )

# ============================================================
# 17. FILE PATHS
# ============================================================

def tag(
    offset_ms
):

    return f"{offset_ms:+d}ms"


def batch_checkpoint(
    offset_ms
):

    return (
        CHECKPOINT_ROOT /
        f"offset_{tag(offset_ms)}_batch.pt"
    )


def best_checkpoint(
    offset_ms
):

    return (
        CHECKPOINT_ROOT /
        f"offset_{tag(offset_ms)}_best.pt"
    )


def final_result(
    offset_ms
):

    return (
        RESULT_ROOT /
        f"offset_{tag(offset_ms)}_final.json"
    )


def history_file(
    offset_ms
):

    return (
        RESULT_ROOT /
        f"offset_{tag(offset_ms)}_history.csv"
    )

# ============================================================
# 18. TRAIN ONE OFFSET
# ============================================================

def train_offset(
    offset_ms
):

    print()
    print("=" * 70)
    print(
        f"TRAINING OFFSET {offset_ms:+d} ms"
    )
    print("=" * 70)

    associations = (
        association_cache[
            offset_ms
        ]
    )

    train_assoc, val_assoc, test_assoc = (
        split_associations(
            associations
        )
    )

    print(
        f"Train: {len(train_assoc):,}"
    )

    print(
        f"Val  : {len(val_assoc):,}"
    )

    print(
        f"Test : {len(test_assoc):,}"
    )

    # --------------------------------------------------------
    # DATA
    # --------------------------------------------------------

    train_dataset = TemporalDataset(
        train_assoc
    )

    val_dataset = TemporalDataset(
        val_assoc
    )

    test_dataset = TemporalDataset(
        test_assoc
    )

    val_loader = DataLoader(
        val_dataset,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=0,
        pin_memory=True
    )

    test_loader = DataLoader(
        test_dataset,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=0,
        pin_memory=True
    )

    # --------------------------------------------------------
    # MODEL
    # --------------------------------------------------------

    model = CharacterCNN().to(
        DEVICE
    )

    class_weights, class_counts = (
        make_class_weights(
            train_assoc
        )
    )

    criterion = nn.CrossEntropyLoss(
        weight=class_weights
    )

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=LR,
        weight_decay=WEIGHT_DECAY
    )

    # --------------------------------------------------------
    # STATE
    # --------------------------------------------------------

    batch_ckpt = (
        batch_checkpoint(
            offset_ms
        )
    )

    best_ckpt = (
        best_checkpoint(
            offset_ms
        )
    )

    start_epoch = 1
    start_batch = 0

    best_val_f1 = -np.inf
    best_epoch = None
    patience_counter = 0

    history = []

    # --------------------------------------------------------
    # RESUME
    # --------------------------------------------------------

    if batch_ckpt.exists():

        print()
        print("RESUMING FROM:")
        print(batch_ckpt)

        checkpoint = torch.load(
            batch_ckpt,
            map_location=DEVICE,
            weights_only=False
        )

        model.load_state_dict(
            checkpoint["model_state"]
        )

        optimizer.load_state_dict(
            checkpoint["optimizer_state"]
        )

        start_epoch = int(
            checkpoint["epoch"]
        )

        start_batch = int(
            checkpoint["batch_idx"]
        )

        best_val_f1 = float(
            checkpoint[
                "best_val_f1"
            ]
        )

        best_epoch = checkpoint.get(
            "best_epoch"
        )

        patience_counter = int(
            checkpoint.get(
                "patience_counter",
                0
            )
        )

        history = checkpoint.get(
            "history",
            []
        )

        restore_rng(
            checkpoint.get(
                "rng_state"
            )
        )

        print(
            "Resume epoch:",
            start_epoch
        )

        print(
            "Resume batch:",
            start_batch
        )

        print(
            "Best Val Macro-F1:",
            f"{best_val_f1:.6f}"
        )

    else:

        print(
            "No checkpoint found."
        )

        print(
            "Starting from epoch 1."
        )

    n_train = len(
        train_dataset
    )

    n_batches = math.ceil(
        n_train /
        BATCH_SIZE
    )

    # ========================================================
    # TRAINING
    # ========================================================

    for epoch in range(
        start_epoch,
        MAX_EPOCHS + 1
    ):

        model.train()

        epoch_start = time.time()

        batches = get_epoch_batches(
            n_train,
            epoch
        )

        batch_begin = (
            start_batch
            if epoch == start_epoch
            else 0
        )

        running_loss = 0.0
        running_n = 0

        for batch_idx in range(
            batch_begin,
            n_batches
        ):

            indices = batches[
                batch_idx
            ]

            xs = []
            ys = []

            for idx in indices:

                x, y = train_dataset[
                    int(idx)
                ]

                xs.append(x)
                ys.append(y)

            x = torch.stack(
                xs
            ).to(
                DEVICE,
                non_blocking=True
            )

            y = torch.stack(
                ys
            ).to(
                DEVICE,
                non_blocking=True
            )

            optimizer.zero_grad(
                set_to_none=True
            )

            logits = model(x)

            loss = criterion(
                logits,
                y
            )

            loss.backward()

            optimizer.step()

            bs = y.size(0)

            running_loss += (
                loss.item() * bs
            )

            running_n += bs

            # ------------------------------------------------
            # SAVE AFTER EVERY BATCH
            # ------------------------------------------------

            if (
                batch_idx + 1
                >= n_batches
            ):

                next_epoch = (
                    epoch + 1
                )

                next_batch = 0

            else:

                next_epoch = epoch

                next_batch = (
                    batch_idx + 1
                )

            checkpoint_state = {

                "status":
                    "TRAINING",

                "offset_ms":
                    offset_ms,

                "epoch":
                    next_epoch,

                "batch_idx":
                    next_batch,

                "n_batches":
                    n_batches,

                "model_state":
                    model.state_dict(),

                "optimizer_state":
                    optimizer.state_dict(),

                "best_val_f1":
                    best_val_f1,

                "best_epoch":
                    best_epoch,

                "patience_counter":
                    patience_counter,

                "history":
                    history,

                "rng_state":
                    capture_rng(),

                "saved_at":
                    time.strftime(
                        "%Y-%m-%d %H:%M:%S"
                    )
            }

            torch.save(
                checkpoint_state,
                batch_ckpt
            )

            # ------------------------------------------------
            # PROGRESS
            # ------------------------------------------------

            if (
                batch_idx == batch_begin
                or
                (batch_idx + 1) % 20 == 0
                or
                batch_idx == n_batches - 1
            ):

                current_loss = (
                    running_loss /
                    max(
                        running_n,
                        1
                    )
                )

                print(
                    f"Offset {offset_ms:+4d} | "
                    f"Epoch {epoch:02d}/{MAX_EPOCHS} | "
                    f"Batch {batch_idx+1:03d}/{n_batches} | "
                    f"Loss {current_loss:.5f}"
                )

            del x
            del y
            del logits
            del loss

        # ----------------------------------------------------
        # VALIDATION
        # ----------------------------------------------------

        start_batch = 0

        train_loss = (
            running_loss /
            max(
                running_n,
                1
            )
        )

        val_metrics = evaluate(
            model,
            val_loader,
            criterion
        )

        epoch_time = (
            time.time()
            - epoch_start
        )

        print()
        print(
            f"Epoch {epoch:02d} COMPLETE"
        )

        print(
            f"Train Loss : {train_loss:.6f}"
        )

        print(
            f"Val Loss   : "
            f"{val_metrics['loss']:.6f}"
        )

        print(
            f"Val Acc    : "
            f"{val_metrics['accuracy']:.6f}"
        )

        print(
            f"Val Bal    : "
            f"{val_metrics['balanced_accuracy']:.6f}"
        )

        print(
            f"Val MacroF1: "
            f"{val_metrics['macro_f1']:.6f}"
        )

        print(
            f"Time       : "
            f"{epoch_time/60:.2f} min"
        )

        # ----------------------------------------------------
        # HISTORY
        # ----------------------------------------------------

        record = {
            "offset_ms":
                offset_ms,

            "epoch":
                epoch,

            "train_loss":
                train_loss,

            "val_loss":
                val_metrics["loss"],

            "val_accuracy":
                val_metrics["accuracy"],

            "val_balanced_accuracy":
                val_metrics[
                    "balanced_accuracy"
                ],

            "val_macro_f1":
                val_metrics[
                    "macro_f1"
                ],

            "val_weighted_f1":
                val_metrics[
                    "weighted_f1"
                ],

            "val_macro_precision":
                val_metrics[
                    "macro_precision"
                ],

            "val_macro_recall":
                val_metrics[
                    "macro_recall"
                ],

            "val_macro_auroc":
                val_metrics[
                    "macro_auroc"
                ],

            "val_macro_auprc":
                val_metrics[
                    "macro_auprc"
                ],

            "epoch_time_sec":
                epoch_time
        }

        history.append(
            record
        )

        # ----------------------------------------------------
        # BEST MODEL
        # ----------------------------------------------------

        if (
            val_metrics["macro_f1"]
            >
            best_val_f1
        ):

            best_val_f1 = (
                val_metrics[
                    "macro_f1"
                ]
            )

            best_epoch = epoch

            patience_counter = 0

            best_state = {

                "status":
                    "BEST",

                "offset_ms":
                    offset_ms,

                "epoch":
                    epoch,

                "best_val_f1":
                    best_val_f1,

                "model_state":
                    copy.deepcopy(
                        model.state_dict()
                    ),

                "optimizer_state":
                    copy.deepcopy(
                        optimizer.state_dict()
                    ),

                "history":
                    history,

                "class_counts":
                    class_counts
            }

            torch.save(
                best_state,
                best_ckpt
            )

            print(
                "✓ NEW BEST MODEL"
            )

            print(
                f"  Val Macro-F1 = "
                f"{best_val_f1:.6f}"
            )

        else:

            patience_counter += 1

            print(
                f"No improvement: "
                f"{patience_counter}/"
                f"{PATIENCE}"
            )

        # ----------------------------------------------------
        # SAVE EPOCH CHECKPOINT
        # ----------------------------------------------------

        epoch_state = {

            "status":
                "TRAINING",

            "offset_ms":
                offset_ms,

            "epoch":
                epoch + 1,

            "batch_idx":
                0,

            "n_batches":
                n_batches,

            "model_state":
                model.state_dict(),

            "optimizer_state":
                optimizer.state_dict(),

            "best_val_f1":
                best_val_f1,

            "best_epoch":
                best_epoch,

            "patience_counter":
                patience_counter,

            "history":
                history,

            "rng_state":
                capture_rng()
        }

        torch.save(
            epoch_state,
            batch_ckpt
        )

        pd.DataFrame(
            history
        ).to_csv(
            history_file(
                offset_ms
            ),
            index=False
        )

        # ----------------------------------------------------
        # EARLY STOPPING
        # ----------------------------------------------------

        if (
            patience_counter
            >= PATIENCE
        ):

            print(
                "Early stopping."
            )

            break

        torch.cuda.empty_cache()

    # ========================================================
    # LOAD BEST MODEL
    # ========================================================

    if not best_ckpt.exists():

        raise RuntimeError(
            "Best checkpoint does not exist."
        )

    best = torch.load(
        best_ckpt,
        map_location=DEVICE,
        weights_only=False
    )

    model.load_state_dict(
        best["model_state"]
    )

    # ========================================================
    # FROZEN TEST
    # ========================================================

    print()
    print(
        "Running frozen test evaluation..."
    )

    test_metrics = evaluate(
        model,
        test_loader,
        criterion
    )

    result = {

        "status":
            "COMPLETED",

        "offset_ms":
            offset_ms,

        "offset_seconds":
            offset_ms / 1000.0,

        "best_epoch":
            int(best_epoch),

        "best_validation_macro_f1":
            float(best_val_f1),

        "test_loss":
            float(
                test_metrics["loss"]
            ),

        "test_accuracy":
            float(
                test_metrics["accuracy"]
            ),

        "test_balanced_accuracy":
            float(
                test_metrics[
                    "balanced_accuracy"
                ]
            ),

        "test_macro_f1":
            float(
                test_metrics[
                    "macro_f1"
                ]
            ),

        "test_weighted_f1":
            float(
                test_metrics[
                    "weighted_f1"
                ]
            ),

        "test_macro_precision":
            float(
                test_metrics[
                    "macro_precision"
                ]
            ),

        "test_macro_recall":
            float(
                test_metrics[
                    "macro_recall"
                ]
            ),

        "test_auroc":
            (
                None
                if np.isnan(
                    test_metrics[
                        "macro_auroc"
                    ]
                )
                else float(
                    test_metrics[
                        "macro_auroc"
                    ]
                )
            ),

        "test_auprc":
            (
                None
                if np.isnan(
                    test_metrics[
                        "macro_auprc"
                    ]
                )
                else float(
                    test_metrics[
                        "macro_auprc"
                    ]
                )
            ),

        "train_samples":
            len(train_assoc),

        "val_samples":
            len(val_assoc),

        "test_samples":
            len(test_assoc),

        "batch_size":
            BATCH_SIZE,

        "learning_rate":
            LR,

        "weight_decay":
            WEIGHT_DECAY,

        "max_epochs":
            MAX_EPOCHS,

        "patience":
            PATIENCE,

        "seed":
            SEED,

        "normalization":
            "per-window channel-wise z-score",

        "selection_metric":
            "validation_macro_f1"
    }

    with open(
        final_result(offset_ms),
        "w"
    ) as f:

        json.dump(
            result,
            f,
            indent=2
        )

    pd.DataFrame(
        history
    ).to_csv(
        history_file(
            offset_ms
        ),
        index=False
    )

    # Only delete resumable checkpoint AFTER
    # frozen test result has been safely saved.
    if batch_ckpt.exists():
        batch_ckpt.unlink()

    print()
    print("=" * 70)
    print(
        f"✓ OFFSET {offset_ms:+d} ms COMPLETED"
    )
    print("=" * 70)

    print(
        f"Best Val Macro-F1 : "
        f"{best_val_f1:.6f}"
    )

    print(
        f"Best Epoch        : "
        f"{best_epoch}"
    )

    print(
        f"Test Accuracy     : "
        f"{test_metrics['accuracy']:.6f}"
    )

    print(
        f"Test Balanced Acc : "
        f"{test_metrics['balanced_accuracy']:.6f}"
    )

    print(
        f"Test Macro-F1     : "
        f"{test_metrics['macro_f1']:.6f}"
    )

    print(
        f"Test AUROC        : "
        f"{test_metrics['macro_auroc']:.6f}"
    )

    print(
        f"Test AUPRC        : "
        f"{test_metrics['macro_auprc']:.6f}"
    )

    return result

# ============================================================
# 19. FIND COMPLETED OFFSETS
# ============================================================

def completed_offsets():

    completed = []

    for offset_ms in OFFSETS_MS:

        path = final_result(
            offset_ms
        )

        if not path.exists():
            continue

        try:

            with open(
                path,
                "r"
            ) as f:

                data = json.load(f)

            if (
                data.get("status")
                == "COMPLETED"
            ):

                completed.append(
                    offset_ms
                )

        except Exception:
            pass

    return completed

# ============================================================
# 20. AUTOMATICALLY SELECT NEXT OFFSET
# ============================================================

done = completed_offsets()

remaining = [
    x for x in OFFSETS_MS
    if x not in done
]

print()
print("=" * 70)
print("STEP 6F STATUS")
print("=" * 70)

print(
    "Completed:",
    done
)

print(
    "Remaining:",
    remaining
)

print()

if len(remaining) == 0:

    print(
        "ALL OFFSETS COMPLETED."
    )

else:

    next_offset = remaining[0]

    print(
        f"NEXT OFFSET: {next_offset:+d} ms"
    )

    # ========================================================
    # START ONLY THE NEXT OFFSET
    # ========================================================

    train_offset(
        next_offset
    )

# ============================================================
# 21. FINAL STATUS TABLE
# ============================================================

rows = []

for offset_ms in OFFSETS_MS:

    result_path = final_result(
        offset_ms
    )

    checkpoint_path = batch_checkpoint(
        offset_ms
    )

    if result_path.exists():

        try:

            with open(
                result_path,
                "r"
            ) as f:

                d = json.load(f)

            rows.append({
                "offset_ms":
                    offset_ms,

                "status":
                    d.get(
                        "status",
                        "UNKNOWN"
                    ),

                "best_epoch":
                    d.get(
                        "best_epoch"
                    ),

                "val_macro_f1":
                    d.get(
                        "best_validation_macro_f1"
                    ),

                "test_accuracy":
                    d.get(
                        "test_accuracy"
                    ),

                "test_balanced_accuracy":
                    d.get(
                        "test_balanced_accuracy"
                    ),

                "test_macro_f1":
                    d.get(
                        "test_macro_f1"
                    ),

                "test_auroc":
                    d.get(
                        "test_auroc"
                    ),

                "test_auprc":
                    d.get(
                        "test_auprc"
                    )
            })

        except Exception:

            rows.append({
                "offset_ms":
                    offset_ms,

                "status":
                    "ERROR"
            })

    elif checkpoint_path.exists():

        try:

            ckpt = torch.load(
                checkpoint_path,
                map_location="cpu",
                weights_only=False
            )

            rows.append({
                "offset_ms":
                    offset_ms,

                "status":
                    "RESUMABLE",

                "epoch":
                    ckpt.get(
                        "epoch"
                    ),

                "next_batch":
                    ckpt.get(
                        "batch_idx"
                    ),

                "best_val_macro_f1":
                    ckpt.get(
                        "best_val_f1"
                    )
            })

        except Exception:

            rows.append({
                "offset_ms":
                    offset_ms,

                "status":
                    "CHECKPOINT_ERROR"
            })

    else:

        rows.append({
            "offset_ms":
                offset_ms,

            "status":
                "PENDING"
        })

status_df = pd.DataFrame(
    rows
)

print()
print("=" * 70)
print("STEP 6F CURRENT STATUS")
print("=" * 70)

display(
    status_df
)

status_df.to_csv(
    RESULT_ROOT /
    "step6f_resumable_status.csv",
    index=False
)

print()
print(
    "If Colab disconnects:"
)
print(
    "1. Reconnect GPU."
)
print(
    "2. Run this SAME cell."
)
print(
    "3. Training resumes from the saved batch."
)
```

```python
# ============================================================
# NeuroSentinel — Step 6F
# BATCH-WISE RESUMABLE TEMPORAL ALIGNMENT TRAINING
#
# Offsets:
#   -500, -400, -300, -200, -100, 0,
#   +100, +200, +300, +400, +500 ms
#
# IMPORTANT:
#   - Existing completed offsets are preserved.
#   - Training resumes from the last completed batch.
#   - Checkpoint saved after EVERY training batch.
#   - GPU is required.
# ============================================================

import os
import json
import math
import random
import glob
import time
import copy
import warnings
from pathlib import Path

import numpy as np
import pandas as pd

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    precision_score,
    recall_score,
    roc_auc_score,
    average_precision_score,
)

import mne

warnings.filterwarnings("ignore")

# ============================================================
# 1. CONFIGURATION
# ============================================================

SEED = 42

OFFSETS_MS = [
    -500, -400, -300, -200, -100,
       0,
     100,  200,  300,  400,  500
]

# Train configuration — SAME AS ORIGINAL STEP 6F
BATCH_SIZE = 256
LR = 1e-3
WEIGHT_DECAY = 1e-4
MAX_EPOCHS = 20
PATIENCE = 5

N_CLASSES = 29
N_CHANNELS = 61
WINDOW_SAMPLES = 25
EPS = 1e-6

# Paths
BASE = Path("/content/gdrive/MyDrive/NeuroSentinel")

PHASE6_MANIFEST = (
    BASE / "07_phase6_subject_split" /
    "phase6_subject_split_manifest_resolved.csv"
)

PHASE6_MANIFEST_FALLBACK = (
    BASE / "07_phase6_subject_split" /
    "phase6_subject_split_manifest.csv"
)

PHASE5_ROOT = (
    BASE / "06_phase5_utility_labels" /
    "recordings"
)

TEMPORAL_ROOT = (
    BASE / "08_phase6_temporal_windows"
)

RESULT_ROOT = (
    BASE / "results" /
    "phase6f_temporal_alignment"
)

CHECKPOINT_ROOT = (
    BASE / "checkpoints" /
    "phase6f_temporal_alignment"
)

RESULT_ROOT.mkdir(parents=True, exist_ok=True)
CHECKPOINT_ROOT.mkdir(parents=True, exist_ok=True)

# ============================================================
# 2. GPU CHECK
# ============================================================

if not torch.cuda.is_available():
    raise RuntimeError(
        "\nGPU IS NOT AVAILABLE.\n"
        "Colab currently has no CUDA backend available.\n\n"
        "Reconnect to a GPU runtime and run this SAME cell again.\n"
        "The checkpoint system will resume automatically.\n"
    )

DEVICE = torch.device("cuda")

print("=" * 70)
print("STEP 6F — BATCH-WISE RESUMABLE TRAINING")
print("=" * 70)
print("Device:", DEVICE)
print("GPU:", torch.cuda.get_device_name(0))
print("CUDA:", torch.version.cuda)
print("PyTorch:", torch.__version__)
print()

# ============================================================
# 3. REPRODUCIBILITY
# ============================================================

def seed_everything(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

seed_everything(SEED)

# ============================================================
# 4. LOAD FROZEN PHASE 6 SPLIT
# ============================================================

if PHASE6_MANIFEST.exists():
    split_df = pd.read_csv(PHASE6_MANIFEST)
elif PHASE6_MANIFEST_FALLBACK.exists():
    split_df = pd.read_csv(PHASE6_MANIFEST_FALLBACK)
else:
    raise FileNotFoundError(
        "Phase 6 split manifest not found."
    )

required_cols = {
    "recording_id",
    "subject",
    "split",
}

missing = required_cols - set(split_df.columns)

if missing:
    raise RuntimeError(
        f"Phase 6 manifest missing columns: {missing}"
    )

split_df = split_df.copy()

split_map = (
    split_df[
        ["recording_id", "subject", "split"]
    ]
    .drop_duplicates()
    .set_index("recording_id")
)

TRAIN_RECORDINGS = sorted(
    split_df.loc[
        split_df["split"].str.lower() == "train",
        "recording_id"
    ].unique()
)

VAL_RECORDINGS = sorted(
    split_df.loc[
        split_df["split"].str.lower() == "val",
        "recording_id"
    ].unique()
)

TEST_RECORDINGS = sorted(
    split_df.loc[
        split_df["split"].str.lower() == "test",
        "recording_id"
    ].unique()
)

print("Frozen split:")
print("  Train recordings:", len(TRAIN_RECORDINGS))
print("  Val recordings:  ", len(VAL_RECORDINGS))
print("  Test recordings: ", len(TEST_RECORDINGS))
print()

# ============================================================
# 5. KNOWN ANOMALOUS PHASE 5 EVENTS
# ============================================================

ANOMALOUS_ROWS = {
    ("002_S1_task1", 950),
    ("011_S1_task1", 1295),
    ("012_S2_task1", 164),
    ("012_S2_task2", 1734),
    ("013_S1_task2", 2313),
}

print("Known anomalous events:", len(ANOMALOUS_ROWS))
print()

# ============================================================
# 6. LABEL MAPPING
# ============================================================

LABELS = (
    [chr(ord("a") + i) for i in range(26)]
    + ["<space>", "<number>", "<special>"]
)

LABEL_TO_ID = {
    label: i
    for i, label in enumerate(LABELS)
}

# ============================================================
# 7. RECORDING PATH RESOLUTION
# ============================================================

def resolve_raw_path(raw_path):
    if pd.isna(raw_path):
        return None

    p = str(raw_path)

    # Colab mount correction
    p = p.replace(
        "/content/drive/",
        "/content/gdrive/"
    )

    if os.path.exists(p):
        return p

    # Try basename search under EEG directory
    basename = os.path.basename(p)

    eeg_root = (
        BASE /
        "data" /
        "SpanishBCBL" /
        "EEG" /
        "EEG"
    )

    matches = list(eeg_root.rglob(basename))

    if matches:
        return str(matches[0])

    return None


def get_raw_path(recording_id):
    rows = split_df[
        split_df["recording_id"] == recording_id
    ]

    if len(rows) == 0:
        return None

    for col in ["raw", "raw_resolved"]:
        if col in rows.columns:
            for value in rows[col].dropna().unique():
                resolved = resolve_raw_path(value)
                if resolved:
                    return resolved

    return None

# ============================================================
# 8. PHASE 5 METADATA + EEG WINDOW LOADING
# ============================================================

def get_phase5_metadata(recording_id):
    path = (
        PHASE5_ROOT /
        f"{recording_id}_metadata.csv"
    )

    if not path.exists():
        raise FileNotFoundError(
            f"Phase 5 metadata missing: {path}"
        )

    df = pd.read_csv(path)

    required = {
        "recording_id",
        "subject",
        "session",
        "task",
        "keystroke_index",
        "timestamp",
        "target",
        "text",
    }

    missing = required - set(df.columns)

    if missing:
        raise RuntimeError(
            f"{recording_id}: metadata missing {missing}"
        )

    return df.reset_index(drop=True)


def get_baseline_windows(recording_id, split):
    # Preferred Step 6 temporal-window location
    p1 = (
        TEMPORAL_ROOT /
        split /
        recording_id /
        f"{recording_id}_baseline_windows.npy"
    )

    # Phase 5 fallback
    p2 = (
        PHASE5_ROOT /
        f"{recording_id}_windows.npy"
    )

    if p1.exists():
        return np.load(p1, mmap_mode="r")

    if p2.exists():
        return np.load(p2, mmap_mode="r")

    raise FileNotFoundError(
        f"No baseline windows found for {recording_id}"
    )

# ============================================================
# 9. LOAD ALL RECORDING SOURCES
# ============================================================

print("Loading Phase 5 metadata and baseline EEG arrays...")

recordings = {}

all_recordings = sorted(
    set(TRAIN_RECORDINGS)
    | set(VAL_RECORDINGS)
    | set(TEST_RECORDINGS)
)

for rid in all_recordings:

    if rid in TRAIN_RECORDINGS:
        split = "train"
    elif rid in VAL_RECORDINGS:
        split = "val"
    else:
        split = "test"

    meta = get_phase5_metadata(rid)
    windows = get_baseline_windows(rid, split)

    if len(meta) != len(windows):
        raise RuntimeError(
            f"{rid}: metadata windows={len(meta)} "
            f"but EEG windows={len(windows)}"
        )

    recordings[rid] = {
        "split": split,
        "meta": meta,
        "windows": windows,
    }

print(
    f"Loaded {len(recordings)}/{len(all_recordings)} recordings."
)
print()

# ============================================================
# 10. BUILD COMPLETE PHASE 5 EVENT ASSOCIATIONS
# ============================================================

print("=" * 70)
print("BUILDING TEMPORAL ASSOCIATIONS")
print("=" * 70)

def build_offset_associations(offset_ms):

    offset_sec = offset_ms / 1000.0

    associations = []

    for rid in all_recordings:

        rec = recordings[rid]
        meta = rec["meta"]

        timestamps = (
            meta["timestamp"]
            .astype(float)
            .to_numpy()
        )

        n = len(meta)

        # Known anomaly mask
        valid_source = np.ones(n, dtype=bool)

        for i in range(n):
            if (rid, i) in ANOMALOUS_ROWS:
                valid_source[i] = False

        # ----------------------------------------------------
        # Offset 0:
        # source event -> same event
        # ----------------------------------------------------
        if offset_ms == 0:

            for source_idx in range(n):

                if not valid_source[source_idx]:
                    continue

                target_idx = source_idx

                target_label = meta.iloc[
                    target_idx
                ]["target"]

                if target_label not in LABEL_TO_ID:
                    continue

                associations.append(
                    (
                        rid,
                        source_idx,
                        target_idx,
                        LABEL_TO_ID[target_label],
                    )
                )

            continue

        # ----------------------------------------------------
        # Non-zero offset:
        # source timestamp + offset
        # -> nearest OTHER event
        # ----------------------------------------------------

        target_times = timestamps

        for source_idx in range(n):

            if not valid_source[source_idx]:
                continue

            desired_time = (
                timestamps[source_idx]
                + offset_sec
            )

            pos = np.searchsorted(
                target_times,
                desired_time
            )

            candidate_indices = []

            if pos < n:
                candidate_indices.append(pos)

            if pos > 0:
                candidate_indices.append(pos - 1)

            if not candidate_indices:
                continue

            # Select nearest candidate.
            # Source event itself is forbidden for non-zero offset.
            best_idx = None
            best_error = np.inf

            for candidate in candidate_indices:

                if candidate == source_idx:
                    continue

                if not valid_source[candidate]:
                    continue

                error = abs(
                    target_times[candidate]
                    - desired_time
                )

                if error < best_error:
                    best_error = error
                    best_idx = candidate

            if best_idx is None:
                continue

            target_label = meta.iloc[
                best_idx
            ]["target"]

            if target_label not in LABEL_TO_ID:
                continue

            associations.append(
                (
                    rid,
                    source_idx,
                    best_idx,
                    LABEL_TO_ID[target_label],
                )
            )

    return associations


association_cache = {}

for offset_ms in OFFSETS_MS:

    cache_file = (
        CHECKPOINT_ROOT /
        f"associations_{offset_ms:+d}ms.npz"
    )

    if cache_file.exists():

        data = np.load(
            cache_file,
            allow_pickle=True
        )

        rids = data["recording_id"].astype(str)
        source_indices = data["source_idx"].astype(np.int64)
        target_indices = data["target_idx"].astype(np.int64)
        labels = data["label"].astype(np.int64)

        assoc = list(
            zip(
                rids,
                source_indices,
                target_indices,
                labels,
            )
        )

        association_cache[offset_ms] = assoc

        print(
            f"{offset_ms:+4d} ms: loaded cached "
            f"{len(assoc):,} associations"
        )

    else:

        assoc = build_offset_associations(
            offset_ms
        )

        association_cache[offset_ms] = assoc

        rids = np.array(
            [x[0] for x in assoc],
            dtype=str
        )

        source_indices = np.array(
            [x[1] for x in assoc],
            dtype=np.int64
        )

        target_indices = np.array(
            [x[2] for x in assoc],
            dtype=np.int64
        )

        labels = np.array(
            [x[3] for x in assoc],
            dtype=np.int64
        )

        np.savez_compressed(
            cache_file,
            recording_id=rids,
            source_idx=source_indices,
            target_idx=target_indices,
            label=labels,
        )

        print(
            f"{offset_ms:+4d} ms: built and cached "
            f"{len(assoc):,} associations"
        )

print()

# ============================================================
# 11. SPLIT ASSOCIATIONS
# ============================================================

def split_associations(assoc):

    train_ids = set(TRAIN_RECORDINGS)
    val_ids = set(VAL_RECORDINGS)
    test_ids = set(TEST_RECORDINGS)

    train = []
    val = []
    test = []

    for item in assoc:

        rid = item[0]

        if rid in train_ids:
            train.append(item)

        elif rid in val_ids:
            val.append(item)

        elif rid in test_ids:
            test.append(item)

    return train, val, test

# ============================================================
# 12. DATASET
# ============================================================

class TemporalAssociationDataset(Dataset):

    def __init__(self, associations):
        self.associations = associations

    def __len__(self):
        return len(self.associations)

    def __getitem__(self, idx):

        rid, source_idx, target_idx, label = (
            self.associations[idx]
        )

        x = recordings[rid]["windows"][
            source_idx
        ]

        # float32
        x = np.asarray(
            x,
            dtype=np.float32
        )

        # ----------------------------------------------------
        # EXACT PHASE 5H.3 NORMALIZATION
        # Per-window channel-wise z-score across time
        # ----------------------------------------------------

        mean = x.mean(
            axis=1,
            keepdims=True
        )

        std = x.std(
            axis=1,
            keepdims=True
        )

        x = (
            x - mean
        ) / (
            std + EPS
        )

        return (
            torch.from_numpy(x),
            torch.tensor(
                label,
                dtype=torch.long
            ),
        )

# ============================================================
# 13. MODEL — SAME STEP 6F ARCHITECTURE
# ============================================================

class CharacterCNN(nn.Module):

    def __init__(self):
        super().__init__()

        self.features = nn.Sequential(

            nn.Conv1d(
                61, 64,
                kernel_size=5,
                padding=2
            ),
            nn.ReLU(),

            nn.Conv1d(
                64, 64,
                kernel_size=5,
                dilation=2,
                padding=4
            ),
            nn.ReLU(),

            nn.Conv1d(
                64, 64,
                kernel_size=5,
                dilation=4,
                padding=8
            ),
            nn.ReLU(),

            nn.Conv1d(
                64, 64,
                kernel_size=5,
                dilation=8,
                padding=16
            ),
            nn.ReLU(),
        )

        self.classifier = nn.Linear(
            64,
            N_CLASSES
        )

    def forward(self, x):

        x = self.features(x)

        x = x.mean(
            dim=-1
        )

        return self.classifier(x)


# Parameter count
tmp_model = CharacterCNN()

n_params = sum(
    p.numel()
    for p in tmp_model.parameters()
    if p.requires_grad
)

print(
    f"Trainable parameters: {n_params:,}"
)

del tmp_model

# ============================================================
# 14. CLASS WEIGHTS
# ============================================================

def make_class_weights(train_assoc):

    counts = np.zeros(
        N_CLASSES,
        dtype=np.float64
    )

    for item in train_assoc:
        label = int(item[3])
        counts[label] += 1

    weights = np.zeros(
        N_CLASSES,
        dtype=np.float64
    )

    present = counts > 0

    weights[present] = (
        1.0 /
        np.sqrt(counts[present])
    )

    # Normalize mean weight over present classes = 1
    if present.any():
        weights[present] /= (
            weights[present].mean()
        )

    return (
        torch.tensor(
            weights,
            dtype=torch.float32,
            device=DEVICE
        ),
        counts
    )

# ============================================================
# 15. METRIC FUNCTION
# ============================================================

def calculate_metrics(
    y_true,
    y_pred,
    y_prob
):

    result = {}

    result["accuracy"] = accuracy_score(
        y_true,
        y_pred
    )

    result["balanced_accuracy"] = (
        balanced_accuracy_score(
            y_true,
            y_pred
        )
    )

    result["macro_f1"] = f1_score(
        y_true,
        y_pred,
        average="macro",
        zero_division=0
    )

    result["weighted_f1"] = f1_score(
        y_true,
        y_pred,
        average="weighted",
        zero_division=0
    )

    result["macro_precision"] = (
        precision_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        )
    )

    result["macro_recall"] = (
        recall_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        )
    )

    # One-vs-rest AUROC/AUPRC
    try:

        y_true_onehot = np.eye(
            N_CLASSES
        )[np.asarray(y_true)]

        result["macro_auroc"] = (
            roc_auc_score(
                y_true_onehot,
                y_prob,
                average="macro",
                multi_class="ovr"
            )
        )

    except Exception:

        result["macro_auroc"] = float("nan")

    try:

        y_true_onehot = np.eye(
            N_CLASSES
        )[np.asarray(y_true)]

        result["macro_auprc"] = (
            average_precision_score(
                y_true_onehot,
                y_prob,
                average="macro"
            )
        )

    except Exception:

        result["macro_auprc"] = float("nan")

    return result

# ============================================================
# 16. VALIDATION
# ============================================================

@torch.no_grad()
def evaluate_model(
    model,
    loader,
    criterion
):

    model.eval()

    total_loss = 0.0
    total_n = 0

    all_true = []
    all_pred = []
    all_prob = []

    for x, y in loader:

        x = x.to(
            DEVICE,
            non_blocking=True
        )

        y = y.to(
            DEVICE,
            non_blocking=True
        )

        logits = model(x)

        loss = criterion(
            logits,
            y
        )

        bs = y.size(0)

        total_loss += (
            loss.item() * bs
        )

        total_n += bs

        probs = torch.softmax(
            logits,
            dim=1
        )

        pred = probs.argmax(
            dim=1
        )

        all_true.append(
            y.cpu().numpy()
        )

        all_pred.append(
            pred.cpu().numpy()
        )

        all_prob.append(
            probs.cpu().numpy()
        )

    y_true = np.concatenate(
        all_true
    )

    y_pred = np.concatenate(
        all_pred
    )

    y_prob = np.concatenate(
        all_prob
    )

    metrics = calculate_metrics(
        y_true,
        y_pred,
        y_prob
    )

    metrics["loss"] = (
        total_loss /
        max(total_n, 1)
    )

    return metrics

# ============================================================
# 17. DETERMINISTIC BATCH ORDER
# ============================================================

def get_epoch_batches(
    n_samples,
    epoch,
    batch_size,
):

    # Independent deterministic generator for each epoch.
    # This means a disconnected session can recreate
    # EXACTLY the same permutation.
    generator = torch.Generator()

    generator.manual_seed(
        SEED + epoch
    )

    permutation = torch.randperm(
        n_samples,
        generator=generator
    ).numpy()

    batches = []

    for start in range(
        0,
        n_samples,
        batch_size
    ):

        stop = min(
            start + batch_size,
            n_samples
        )

        batches.append(
            permutation[start:stop]
        )

    return batches

# ============================================================
# 18. RNG STATE HELPERS
# ============================================================

def capture_rng_state():

    state = {
        "python": random.getstate(),
        "numpy": np.random.get_state(),
        "torch": torch.get_rng_state(),
    }

    if torch.cuda.is_available():
        state["cuda"] = (
            torch.cuda.get_rng_state_all()
        )

    return state


def restore_rng_state(state):

    if state is None:
        return

    random.setstate(
        state["python"]
    )

    np.random.set_state(
        state["numpy"]
    )

    torch.set_rng_state(
        state["torch"]
    )

    if (
        torch.cuda.is_available()
        and "cuda" in state
    ):
        torch.cuda.set_rng_state_all(
            state["cuda"]
        )

# ============================================================
# 19. CHECKPOINT PATHS
# ============================================================

def offset_tag(offset_ms):

    return f"{offset_ms:+d}ms"


def batch_checkpoint_path(offset_ms):

    return (
        CHECKPOINT_ROOT /
        f"offset_{offset_tag(offset_ms)}_batch.pt"
    )


def best_checkpoint_path(offset_ms):

    return (
        CHECKPOINT_ROOT /
        f"offset_{offset_tag(offset_ms)}_best.pt"
    )


def final_result_path(offset_ms):

    return (
        RESULT_ROOT /
        f"offset_{offset_tag(offset_ms)}_final.json"
    )


def history_path(offset_ms):

    return (
        RESULT_ROOT /
        f"offset_{offset_tag(offset_ms)}_history.csv"
    )

# ============================================================
# 20. COMPLETED OFFSET DETECTION
# ============================================================

def find_completed_offsets():

    completed = []

    for offset_ms in OFFSETS_MS:

        path = final_result_path(
            offset_ms
        )

        if not path.exists():
            continue

        try:

            with open(
                path,
                "r"
            ) as f:
                data = json.load(f)

            # Require explicit completion flag
            if data.get(
                "status"
            ) == "COMPLETED":

                completed.append(
                    offset_ms
                )

        except Exception:
            pass

    return completed


completed_offsets = find_completed_offsets()

print("=" * 70)
print("EXISTING COMPLETED OFFSETS")
print("=" * 70)

if completed_offsets:
    print(
        completed_offsets
    )
else:
    print("None detected.")

print()

# ============================================================
# 21. TRAIN ONE OFFSET — RESUMABLE
# ============================================================

def train_offset(offset_ms):

    print()
    print("=" * 70)
    print(
        f"STARTING OFFSET {offset_ms:+d} ms"
    )
    print("=" * 70)

    assoc = association_cache[
        offset_ms
    ]

    train_assoc, val_assoc, test_assoc = (
        split_associations(assoc)
    )

    print(
        f"Samples: train={len(train_assoc):,} "
        f"val={len(val_assoc):,} "
        f"test={len(test_assoc):,}"
    )

    # --------------------------------------------------------
    # DATASETS
    # --------------------------------------------------------

    train_dataset = (
        TemporalAssociationDataset(
            train_assoc
        )
    )

    val_dataset = (
        TemporalAssociationDataset(
            val_assoc
        )
    )

    test_dataset = (
        TemporalAssociationDataset(
            test_assoc
        )
    )

    val_loader = DataLoader(
        val_dataset,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=0,
        pin_memory=True,
    )

    test_loader = DataLoader(
        test_dataset,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=0,
        pin_memory=True,
    )

    # --------------------------------------------------------
    # MODEL
    # --------------------------------------------------------

    model = CharacterCNN().to(
        DEVICE
    )

    class_weights, class_counts = (
        make_class_weights(
            train_assoc
        )
    )

    criterion = nn.CrossEntropyLoss(
        weight=class_weights
    )

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=LR,
        weight_decay=WEIGHT_DECAY
    )

    # --------------------------------------------------------
    # TRAINING STATE
    # --------------------------------------------------------

    start_epoch = 1
    start_batch = 0

    best_val_f1 = -np.inf
    best_epoch = None
    patience_counter = 0

    history = []

    batch_ckpt = (
        batch_checkpoint_path(
            offset_ms
        )
    )

    best_ckpt = (
        best_checkpoint_path(
            offset_ms
        )
    )

    # --------------------------------------------------------
    # RESUME FROM BATCH CHECKPOINT
    # --------------------------------------------------------

    if batch_ckpt.exists():

        print(
            "\nRESUMING FROM BATCH CHECKPOINT:"
        )
        print(batch_ckpt)

        checkpoint = torch.load(
            batch_ckpt,
            map_location=DEVICE,
            weights_only=False
        )

        model.load_state_dict(
            checkpoint["model_state"]
        )

        optimizer.load_state_dict(
            checkpoint["optimizer_state"]
        )

        start_epoch = int(
            checkpoint["epoch"]
        )

        start_batch = int(
            checkpoint["batch_idx"]
        )

        best_val_f1 = float(
            checkpoint["best_val_f1"]
        )

        best_epoch = checkpoint.get(
            "best_epoch"
        )

        patience_counter = int(
            checkpoint.get(
                "patience_counter",
                0
            )
        )

        history = checkpoint.get(
            "history",
            []
        )

        restore_rng_state(
            checkpoint.get(
                "rng_state"
            )
        )

        print(
            f"Resume epoch: {start_epoch}"
        )
        print(
            f"Resume next batch: {start_batch}"
        )
        print(
            f"Best val Macro-F1: "
            f"{best_val_f1:.6f}"
        )

    else:

        print(
            "\nNo batch checkpoint found."
        )
        print(
            "Starting from epoch 1, batch 0."
        )

    # --------------------------------------------------------
    # TRAINING LOOP
    # --------------------------------------------------------

    n_train = len(
        train_dataset
    )

    n_batches = math.ceil(
        n_train /
        BATCH_SIZE
    )

    for epoch in range(
        start_epoch,
        MAX_EPOCHS + 1
    ):

        model.train()

        epoch_start = time.time()

        batches = get_epoch_batches(
            n_train,
            epoch,
            BATCH_SIZE
        )

        # If resuming a previous epoch,
        # skip batches already completed.
        batch_begin = (
            start_batch
            if epoch == start_epoch
            else 0
        )

        running_loss = 0.0
        running_n = 0

        for batch_idx in range(
            batch_begin,
            n_batches
        ):

            batch_indices = batches[
                batch_idx
            ]

            xs = []
            ys = []

            for dataset_idx in batch_indices:

                x, y = train_dataset[
                    int(dataset_idx)
                ]

                xs.append(x)
                ys.append(y)

            x = torch.stack(
                xs,
                dim=0
            ).to(
                DEVICE,
                non_blocking=True
            )

            y = torch.stack(
                ys,
                dim=0
            ).to(
                DEVICE,
                non_blocking=True
            )

            optimizer.zero_grad(
                set_to_none=True
            )

            logits = model(x)

            loss = criterion(
                logits,
                y
            )

            loss.backward()

            optimizer.step()

            bs = y.size(0)

            running_loss += (
                loss.item() * bs
            )

            running_n += bs

            # ------------------------------------------------
            # CRITICAL:
            # SAVE AFTER EVERY BATCH
            # ------------------------------------------------

            next_epoch = epoch
            next_batch = batch_idx + 1

            if next_batch >= n_batches:

                # Epoch completed.
                checkpoint_epoch = (
                    epoch + 1
                )
                checkpoint_batch = 0

            else:

                checkpoint_epoch = epoch
                checkpoint_batch = next_batch

            batch_state = {
                "status": "TRAINING",

                "offset_ms": offset_ms,

                "epoch": checkpoint_epoch,

                "batch_idx": checkpoint_batch,

                "n_batches": n_batches,

                "global_batch":
                    (
                        (epoch - 1)
                        * n_batches
                        + batch_idx
                        + 1
                    ),

                "model_state":
                    model.state_dict(),

                "optimizer_state":
                    optimizer.state_dict(),

                "best_val_f1":
                    best_val_f1,

                "best_epoch":
                    best_epoch,

                "patience_counter":
                    patience_counter,

                "history":
                    history,

                "rng_state":
                    capture_rng_state(),

                "saved_at":
                    time.strftime(
                        "%Y-%m-%d %H:%M:%S"
                    ),
            }

            torch.save(
                batch_state,
                batch_ckpt
            )

            # ------------------------------------------------
            # PROGRESS
            # ------------------------------------------------

            if (
                batch_idx == batch_begin
                or
                (batch_idx + 1) % 20 == 0
                or
                batch_idx == n_batches - 1
            ):

                current_loss = (
                    running_loss /
                    max(running_n, 1)
                )

                print(
                    f"Offset {offset_ms:+4d} ms | "
                    f"Epoch {epoch:02d}/{MAX_EPOCHS} | "
                    f"Batch {batch_idx+1:03d}/{n_batches} | "
                    f"Loss {current_loss:.5f}"
                )

            del x, y, logits, loss

        # ----------------------------------------------------
        # EPOCH COMPLETE
        # ----------------------------------------------------

        start_batch = 0

        epoch_train_loss = (
            running_loss /
            max(running_n, 1)
        )

        val_metrics = evaluate_model(
            model,
            val_loader,
            criterion
        )

        epoch_time = (
            time.time()
            - epoch_start
        )

        print(
            f"\nOffset {offset_ms:+4d} ms | "
            f"Epoch {epoch:02d} COMPLETE | "
            f"Train Loss {epoch_train_loss:.6f} | "
            f"Val Loss {val_metrics['loss']:.6f} | "
            f"Val Acc {val_metrics['accuracy']:.6f} | "
            f"Val Bal {val_metrics['balanced_accuracy']:.6f} | "
            f"Val Macro-F1 {val_metrics['macro_f1']:.6f} | "
            f"Time {epoch_time/60:.2f} min"
        )

        record = {
            "offset_ms": offset_ms,
            "epoch": epoch,
            "train_loss": epoch_train_loss,
            "val_loss": val_metrics["loss"],
            "val_accuracy":
                val_metrics["accuracy"],
            "val_balanced_accuracy":
                val_metrics["balanced_accuracy"],
            "val_macro_f1":
                val_metrics["macro_f1"],
            "val_weighted_f1":
                val_metrics["weighted_f1"],
            "val_macro_precision":
                val_metrics["macro_precision"],
            "val_macro_recall":
                val_metrics["macro_recall"],
            "val_macro_auroc":
                val_metrics["macro_auroc"],
            "val_macro_auprc":
                val_metrics["macro_auprc"],
            "epoch_time_sec": epoch_time,
        }

        history.append(record)

        # ----------------------------------------------------
        # VALIDATION MODEL SELECTION
        # ----------------------------------------------------

        if (
            val_metrics["macro_f1"]
            >
            best_val_f1
        ):

            best_val_f1 = (
                val_metrics["macro_f1"]
            )

            best_epoch = epoch

            patience_counter = 0

            best_state = {
                "status": "BEST",

                "offset_ms": offset_ms,

                "epoch": epoch,

                "best_val_f1":
                    best_val_f1,

                "model_state":
                    copy.deepcopy(
                        model.state_dict()
                    ),

                "optimizer_state":
                    copy.deepcopy(
                        optimizer.state_dict()
                    ),

                "history":
                    history,

                "class_counts":
                    class_counts,

                "saved_at":
                    time.strftime(
                        "%Y-%m-%d %H:%M:%S"
                    ),
            }

            torch.save(
                best_state,
                best_ckpt
            )

            print(
                f"  NEW BEST MODEL — "
                f"Val Macro-F1 = "
                f"{best_val_f1:.6f}"
            )

        else:

            patience_counter += 1

            print(
                f"  No improvement. "
                f"Patience "
                f"{patience_counter}/{PATIENCE}"
            )

        # ----------------------------------------------------
        # UPDATE BATCH CHECKPOINT AFTER EPOCH
        # ----------------------------------------------------

        epoch_state = {
            "status": "TRAINING",

            "offset_ms": offset_ms,

            "epoch": epoch + 1,

            "batch_idx": 0,

            "n_batches": n_batches,

            "global_batch":
                epoch * n_batches,

            "model_state":
                model.state_dict(),

            "optimizer_state":
                optimizer.state_dict(),

            "best_val_f1":
                best_val_f1,

            "best_epoch":
                best_epoch,

            "patience_counter":
                patience_counter,

            "history":
                history,

            "rng_state":
                capture_rng_state(),

            "saved_at":
                time.strftime(
                    "%Y-%m-%d %H:%M:%S"
                ),
        }

        torch.save(
            epoch_state,
            batch_ckpt
        )

        # Save history
        pd.DataFrame(
            history
        ).to_csv(
            history_path(offset_ms),
            index=False
        )

        # ----------------------------------------------------
        # EARLY STOPPING
        # ----------------------------------------------------

        if patience_counter >= PATIENCE:

            print(
                "\nEarly stopping triggered."
            )

            break

        # ----------------------------------------------------
        # EMPTY CUDA CACHE
        # ----------------------------------------------------

        torch.cuda.empty_cache()

    # ========================================================
    # TRAINING COMPLETE — LOAD BEST MODEL
    # ========================================================

    if not best_ckpt.exists():

        raise RuntimeError(
            f"No best checkpoint found for "
            f"{offset_ms:+d} ms."
        )

    print(
        "\nLoading validation-selected "
        "best model..."
    )

    best = torch.load(
        best_ckpt,
        map_location=DEVICE,
        weights_only=False
    )

    model.load_state_dict(
        best["model_state"]
    )

    # ========================================================
    # FINAL FROZEN TEST
    # ========================================================

    print(
        "\nRunning FROZEN test evaluation..."
    )

    test_metrics = evaluate_model(
        model,
        test_loader,
        criterion
    )

    final_result = {

        "status": "COMPLETED",

        "offset_ms": offset_ms,

        "offset_seconds":
            offset_ms / 1000.0,

        "best_epoch":
            int(best_epoch),

        "best_validation_macro_f1":
            float(best_val_f1),

        "test_loss":
            float(test_metrics["loss"]),

        "test_accuracy":
            float(test_metrics["accuracy"]),

        "test_balanced_accuracy":
            float(
                test_metrics[
                    "balanced_accuracy"
                ]
            ),

        "test_macro_f1":
            float(
                test_metrics[
                    "macro_f1"
                ]
            ),

        "test_weighted_f1":
            float(
                test_metrics[
                    "weighted_f1"
                ]
            ),

        "test_macro_precision":
            float(
                test_metrics[
                    "macro_precision"
                ]
            ),

        "test_macro_recall":
            float(
                test_metrics[
                    "macro_recall"
                ]
            ),

        "test_macro_auroc":
            float(
                test_metrics[
                    "macro_auroc"
                ]
            )
            if not np.isnan(
                test_metrics[
                    "macro_auroc"
                ]
            )
            else None,

        "test_macro_auprc":
            float(
                test_metrics[
                    "macro_auprc"
                ]
            )
            if not np.isnan(
                test_metrics[
                    "macro_auprc"
                ]
            )
            else None,

        "train_samples":
            len(train_assoc),

        "val_samples":
            len(val_assoc),

        "test_samples":
            len(test_assoc),

        "batch_size":
            BATCH_SIZE,

        "learning_rate":
            LR,

        "weight_decay":
            WEIGHT_DECAY,

        "max_epochs":
            MAX_EPOCHS,

        "patience":
            PATIENCE,

        "seed":
            SEED,

        "normalization":
            "per-window channel-wise z-score",

        "architecture":
            "4-layer dilated CNN 61->64->64->64->64 + global mean pooling + 29-class linear head",

        "selection_metric":
            "validation_macro_f1",

        "timestamp":
            time.strftime(
                "%Y-%m-%d %H:%M:%S"
            ),
    }

    with open(
        final_result_path(offset_ms),
        "w"
    ) as f:

        json.dump(
            final_result,
            f,
            indent=2
        )

    # Save final history
    pd.DataFrame(
        history
    ).to_csv(
        history_path(offset_ms),
        index=False
    )

    # Remove batch checkpoint ONLY after the
    # final test result has been safely written.
    if batch_ckpt.exists():
        batch_ckpt.unlink()

    print()
    print("=" * 70)
    print(
        f"OFFSET {offset_ms:+d} ms COMPLETED"
    )
    print("=" * 70)
    print(
        f"Best Val Macro-F1 : "
        f"{best_val_f1:.6f}"
    )
    print(
        f"Best Epoch        : "
        f"{best_epoch}"
    )
    print(
        f"Test Accuracy     : "
        f"{test_metrics['accuracy']:.6f}"
    )
    print(
        f"Test Balanced Acc : "
        f"{test_metrics['balanced_accuracy']:.6f}"
    )
    print(
        f"Test Macro-F1     : "
        f"{test_metrics['macro_f1']:.6f}"
    )
    print(
        f"Test AUROC        : "
        f"{test_metrics['macro_auroc']:.6f}"
    )
    print(
        f"Test AUPRC        : "
        f"{test_metrics['macro_auprc']:.6f}"
    )
    print()

    return final_result

# ============================================================
# 22. MAIN RESUMABLE DRIVER
# ============================================================

print("=" * 70)
print("STEP 6F RESUMABLE DRIVER")
print("=" * 70)

completed_offsets = find_completed_offsets()

remaining_offsets = [
    x for x in OFFSETS_MS
    if x not in completed_offsets
]

print(
    "Completed:",
    completed_offsets
)

print(
    "Remaining:",
    remaining_offsets
)

print()

# ------------------------------------------------------------
# Process only ONE offset per Colab session invocation.
#
# This is intentional:
#   - reduces runtime risk
#   - allows disconnect/reconnect
#   - checkpoint survives
#   - next run automatically continues
# ------------------------------------------------------------

if len(remaining_offsets) == 0:

    print(
        "ALL STEP 6F OFFSETS ARE ALREADY COMPLETE."
    )

else:

    next_offset = remaining_offsets[0]

    print(
        f"Next offset to process: "
        f"{next_offset:+d} ms"
    )

    print(
        "\nThe cell will process ONLY this offset."
    )

    result = train_offset(
        next_offset
    )

# ============================================================
# 23. CURRENT SUMMARY
# ============================================================

print()
print("=" * 70)
print("STEP 6F CURRENT STATUS")
print("=" * 70)

rows = []

for offset_ms in OFFSETS_MS:

    path = final_result_path(
        offset_ms
    )

    if path.exists():

        try:

            with open(
                path,
                "r"
            ) as f:
                d = json.load(f)

            rows.append({
                "offset_ms":
                    offset_ms,
                "status":
                    d.get(
                        "status",
                        "UNKNOWN"
                    ),
                "best_epoch":
                    d.get(
                        "best_epoch"
                    ),
                "val_macro_f1":
                    d.get(
                        "best_validation_macro_f1"
                    ),
                "test_accuracy":
                    d.get(
                        "test_accuracy"
                    ),
                "test_balanced_accuracy":
                    d.get(
                        "test_balanced_accuracy"
                    ),
                "test_macro_f1":
                    d.get(
                        "test_macro_f1"
                    ),
                "test_auroc":
                    d.get(
                        "test_macro_auroc"
                    ),
                "test_auprc":
                    d.get(
                        "test_macro_auprc"
                    ),
            })

        except Exception:

            rows.append({
                "offset_ms":
                    offset_ms,
                "status":
                    "CORRUPT/UNREADABLE"
            })

    else:

        batch_path = (
            batch_checkpoint_path(
                offset_ms
            )
        )

        if batch_path.exists():

            try:

                ckpt = torch.load(
                    batch_path,
                    map_location="cpu",
                    weights_only=False
                )

                rows.append({
                    "offset_ms":
                        offset_ms,
                    "status":
                        "RESUMABLE",
                    "epoch":
                        ckpt.get(
                            "epoch"
                        ),
                    "next_batch":
                        ckpt.get(
                            "batch_idx"
                        ),
                    "best_val_macro_f1":
                        ckpt.get(
                            "best_val_f1"
                        ),
                })

            except Exception:

                rows.append({
                    "offset_ms":
                        offset_ms,
                    "status":
                        "CHECKPOINT_PRESENT"
                })

        else:

            rows.append({
                "offset_ms":
                    offset_ms,
                "status":
                    "PENDING"
            })

status_df = pd.DataFrame(rows)

display(
    status_df
)

status_df.to_csv(
    RESULT_ROOT /
    "step6f_resumable_status.csv",
    index=False
)

print()
print(
    "IMPORTANT:"
)
print(
    "If Colab disconnects, reconnect GPU and run "
    "this SAME cell again."
)
print(
    "It will automatically resume from the saved "
    "batch checkpoint."
)
```

```python
# ============================================================
# NEUROSENTINEL — STEP 6G
# WITHIN-SUBJECT CHARACTER DECODING DIAGNOSTIC
#
# Purpose:
#   Determine whether character information is present within
#   subjects but fails to generalize across subjects.
#
# Important:
#   - Official Phase 6 cross-subject split is NOT modified.
#   - This is a diagnostic experiment only.
#   - Splitting is at RECORDING level, never window level.
#   - Same 61 x 25 baseline windows.
#   - Same Phase 5H.3 per-window channel-wise z-score.
#   - Same 29-character mapping.
#   - Same CNN architecture.
#   - Batch-wise resumable checkpoints.
# ============================================================

import os
import json
import math
import random
import copy
import time
import warnings
from pathlib import Path

import numpy as np
import pandas as pd

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    precision_score,
    recall_score,
    roc_auc_score,
    average_precision_score,
)

warnings.filterwarnings("ignore")

# ============================================================
# 1. CONFIGURATION
# ============================================================

SEED = 42

# The three frozen official test subjects from Phase 6.
TEST_SUBJECTS = [
    "002",
    "011",
    "021",
]

# Recording-level proportions.
TRAIN_FRAC = 0.60
VAL_FRAC = 0.20
TEST_FRAC = 0.20

BATCH_SIZE = 256

LR = 1e-3
WEIGHT_DECAY = 1e-4

MAX_EPOCHS = 20
PATIENCE = 5

N_CLASSES = 29
N_CHANNELS = 61
WINDOW_SAMPLES = 25

EPS = 1e-6

# ============================================================
# 2. PATHS
# ============================================================

BASE = Path(
    "/content/gdrive/MyDrive/NeuroSentinel"
)

PHASE6_MANIFEST = (
    BASE /
    "07_phase6_subject_split" /
    "phase6_subject_split_manifest_resolved.csv"
)

PHASE6_MANIFEST_FALLBACK = (
    BASE /
    "07_phase6_subject_split" /
    "phase6_subject_split_manifest.csv"
)

PHASE5_ROOT = (
    BASE /
    "06_phase5_utility_labels" /
    "recordings"
)

RESULT_ROOT = (
    BASE /
    "results" /
    "phase6g_within_subject"
)

CHECKPOINT_ROOT = (
    BASE /
    "checkpoints" /
    "phase6g_within_subject"
)

RESULT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

CHECKPOINT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

# ============================================================
# 3. GPU
# ============================================================

if not torch.cuda.is_available():

    raise RuntimeError(
        "\nGPU IS NOT AVAILABLE.\n"
        "Reconnect a GPU runtime and run this cell again.\n"
        "Existing checkpoints are preserved.\n"
    )

DEVICE = torch.device("cuda")

print("=" * 70)
print("NEUROSENTINEL — STEP 6G")
print("WITHIN-SUBJECT CHARACTER DECODING")
print("=" * 70)

print(
    "Device:",
    DEVICE
)

print(
    "GPU:",
    torch.cuda.get_device_name(0)
)

print(
    "PyTorch:",
    torch.__version__
)

print()

# ============================================================
# 4. REPRODUCIBILITY
# ============================================================

def seed_everything(
    seed=42
):

    random.seed(seed)
    np.random.seed(seed)

    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)


seed_everything(
    SEED
)

# ============================================================
# 5. LABEL MAPPING
# ============================================================

LABELS = (
    [
        chr(ord("a") + i)
        for i in range(26)
    ]
    +
    [
        "<space>",
        "<number>",
        "<special>"
    ]
)

LABEL_TO_ID = {
    label: i
    for i, label in enumerate(
        LABELS
    )
}

# ============================================================
# 6. KNOWN ANOMALOUS SOURCE EVENTS
# ============================================================

ANOMALOUS_ROWS = {
    ("002_S1_task1", 950),
    ("011_S1_task1", 1295),
    ("012_S2_task1", 164),
    ("012_S2_task2", 1734),
    ("013_S1_task2", 2313),
}

# ============================================================
# 7. LOAD FROZEN PHASE 6 MANIFEST
# ============================================================

if PHASE6_MANIFEST.exists():

    split_df = pd.read_csv(
        PHASE6_MANIFEST
    )

elif PHASE6_MANIFEST_FALLBACK.exists():

    split_df = pd.read_csv(
        PHASE6_MANIFEST_FALLBACK
    )

else:

    raise FileNotFoundError(
        "Phase 6 manifest was not found."
    )

required_columns = {
    "recording_id",
    "subject",
    "split",
}

missing_columns = (
    required_columns -
    set(split_df.columns)
)

if missing_columns:

    raise RuntimeError(
        f"Missing required columns: "
        f"{missing_columns}"
    )

split_df["subject"] = (
    split_df["subject"]
    .astype(str)
    .str.zfill(3)
)

split_df["recording_id"] = (
    split_df["recording_id"]
    .astype(str)
)

print(
    "Frozen Phase 6 manifest loaded."
)

print()

# ============================================================
# 8. VERIFY TEST SUBJECTS
# ============================================================

for subject in TEST_SUBJECTS:

    subject_rows = split_df[
        split_df["subject"] == subject
    ]

    if len(subject_rows) == 0:

        raise RuntimeError(
            f"Subject {subject} "
            f"not found in Phase 6 manifest."
        )

    print(
        f"Subject {subject}: "
        f"{len(subject_rows)} recordings"
    )

print()

# ============================================================
# 9. RECORDING DISCOVERY
# ============================================================

subject_recordings = {}

for subject in TEST_SUBJECTS:

    rows = split_df[
        split_df["subject"] == subject
    ]

    recordings = sorted(
        rows["recording_id"]
        .unique()
        .tolist()
    )

    subject_recordings[
        subject
    ] = recordings

    print(
        f"{subject}:",
        recordings
    )

print()

# ============================================================
# 10. DETERMINISTIC RECORDING-LEVEL SPLIT
#
# We deliberately split recordings, not windows.
#
# For each subject:
#   - 60% train
#   - 20% validation
#   - 20% test
#
# With very small recording counts, we guarantee that each
# split contains at least one recording whenever possible.
# ============================================================

def make_recording_split(
    recordings,
    seed=42
):

    recordings = list(
        sorted(recordings)
    )

    n = len(
        recordings
    )

    if n < 3:

        raise RuntimeError(
            f"Need at least 3 recordings "
            f"for recording-level train/val/test "
            f"split, found {n}."
        )

    rng = random.Random(
        seed
    )

    shuffled = recordings.copy()

    rng.shuffle(
        shuffled
    )

    # --------------------------------------------------------
    # Determine counts.
    # --------------------------------------------------------

    n_train = max(
        1,
        int(
            round(
                n * TRAIN_FRAC
            )
        )
    )

    n_val = max(
        1,
        int(
            round(
                n * VAL_FRAC
            )
        )
    )

    n_test = (
        n -
        n_train -
        n_val
    )

    # --------------------------------------------------------
    # Adjust so test has at least one.
    # --------------------------------------------------------

    if n_test < 1:

        n_test = 1

        if n_train > n_val:
            n_train -= 1
        else:
            n_val -= 1

    # --------------------------------------------------------
    # Safety correction.
    # --------------------------------------------------------

    while (
        n_train +
        n_val +
        n_test
        > n
    ):

        if n_train > 1:
            n_train -= 1

        elif n_val > 1:
            n_val -= 1

        else:
            n_test -= 1

    while (
        n_train +
        n_val +
        n_test
        < n
    ):

        n_train += 1

    train = shuffled[
        :n_train
    ]

    val = shuffled[
        n_train:
        n_train + n_val
    ]

    test = shuffled[
        n_train + n_val:
    ]

    if not train or not val or not test:

        raise RuntimeError(
            "Recording split failed "
            "to create all three partitions."
        )

    return (
        sorted(train),
        sorted(val),
        sorted(test)
    )


subject_splits = {}

for i, subject in enumerate(
    TEST_SUBJECTS
):

    recs = subject_recordings[
        subject
    ]

    train_recs, val_recs, test_recs = (
        make_recording_split(
            recs,
            seed=SEED + i
        )
    )

    subject_splits[
        subject
    ] = {
        "train": train_recs,
        "val": val_recs,
        "test": test_recs,
    }

    print("=" * 70)

    print(
        f"SUBJECT {subject}"
    )

    print(
        "Train:",
        train_recs
    )

    print(
        "Val  :",
        val_recs
    )

    print(
        "Test :",
        test_recs
    )

print()

# ============================================================
# 11. SAVE SPLIT MANIFEST
# ============================================================

split_rows = []

for subject in TEST_SUBJECTS:

    for split_name in [
        "train",
        "val",
        "test"
    ]:

        for recording_id in (
            subject_splits[
                subject
            ][split_name]
        ):

            split_rows.append({
                "subject":
                    subject,

                "recording_id":
                    recording_id,

                "within_subject_split":
                    split_name
            })

within_subject_split_df = pd.DataFrame(
    split_rows
)

within_subject_split_path = (
    RESULT_ROOT /
    "step6g_recording_split.csv"
)

within_subject_split_df.to_csv(
    within_subject_split_path,
    index=False
)

print(
    "Recording split saved:"
)

print(
    within_subject_split_path
)

print()

# ============================================================
# 12. LOAD RECORDING DATA
# ============================================================

recordings = {}

all_recording_ids = sorted(
    within_subject_split_df[
        "recording_id"
    ].unique()
)

print(
    f"Loading "
    f"{len(all_recording_ids)} "
    f"recordings..."
)

for recording_id in (
    all_recording_ids
):

    metadata_path = (
        PHASE5_ROOT /
        f"{recording_id}_metadata.csv"
    )

    windows_path = (
        PHASE5_ROOT /
        f"{recording_id}_windows.npy"
    )

    if not metadata_path.exists():

        raise FileNotFoundError(
            f"Missing metadata:\n"
            f"{metadata_path}"
        )

    if not windows_path.exists():

        raise FileNotFoundError(
            f"Missing windows:\n"
            f"{windows_path}"
        )

    metadata = pd.read_csv(
        metadata_path
    ).reset_index(
        drop=True
    )

    windows = np.load(
        windows_path,
        mmap_mode="r"
    )

    if len(metadata) != len(windows):

        raise RuntimeError(
            f"{recording_id}: "
            f"metadata={len(metadata)}, "
            f"windows={len(windows)}"
        )

    # --------------------------------------------------------
    # Shape verification.
    # --------------------------------------------------------

    if (
        windows.ndim != 3
        or
        windows.shape[1] != N_CHANNELS
        or
        windows.shape[2] != WINDOW_SAMPLES
    ):

        raise RuntimeError(
            f"{recording_id}: "
            f"unexpected shape "
            f"{windows.shape}"
        )

    recordings[
        recording_id
    ] = {
        "metadata":
            metadata,

        "windows":
            windows,
    }

print(
    "✓ All recording data loaded."
)

print()

# ============================================================
# 13. BUILD ASSOCIATIONS
#
# Each valid Phase 5 event becomes one character example.
#
# The target is the character belonging to THAT SAME event.
# ============================================================

def build_associations(
    recording_ids
):

    associations = []

    for recording_id in (
        recording_ids
    ):

        metadata = recordings[
            recording_id
        ]["metadata"]

        for row_idx in range(
            len(metadata)
        ):

            # -----------------------------------------------
            # Remove the five known anomalous events.
            # -----------------------------------------------

            if (
                recording_id,
                row_idx
            ) in ANOMALOUS_ROWS:

                continue

            target = metadata.iloc[
                row_idx
            ]["target"]

            if target not in LABEL_TO_ID:

                continue

            associations.append(
                (
                    recording_id,
                    row_idx,
                    LABEL_TO_ID[target]
                )
            )

    return associations

# ============================================================
# 14. DATASET
# ============================================================

class CharacterDataset(
    Dataset
):

    def __init__(
        self,
        associations
    ):

        self.associations = (
            associations
        )

    def __len__(
        self
    ):

        return len(
            self.associations
        )

    def __getitem__(
        self,
        idx
    ):

        recording_id, row_idx, label = (
            self.associations[idx]
        )

        x = np.asarray(
            recordings[
                recording_id
            ]["windows"][
                row_idx
            ],
            dtype=np.float32
        )

        # ----------------------------------------------------
        # EXACT Phase 5H.3 normalization:
        # channel-wise z-score within each window.
        # ----------------------------------------------------

        mean = x.mean(
            axis=1,
            keepdims=True
        )

        std = x.std(
            axis=1,
            keepdims=True
        )

        x = (
            x - mean
        ) / (
            std + EPS
        )

        return (
            torch.from_numpy(x),
            torch.tensor(
                label,
                dtype=torch.long
            )
        )

# ============================================================
# 15. CNN ARCHITECTURE
#
# Same lightweight character decoder family used in
# the previous subject-independent diagnostic.
# ============================================================

class CharacterCNN(
    nn.Module
):

    def __init__(
        self
    ):

        super().__init__()

        self.features = nn.Sequential(

            nn.Conv1d(
                61,
                64,
                kernel_size=5,
                padding=2
            ),

            nn.ReLU(),

            nn.Conv1d(
                64,
                64,
                kernel_size=5,
                dilation=2,
                padding=4
            ),

            nn.ReLU(),

            nn.Conv1d(
                64,
                64,
                kernel_size=5,
                dilation=4,
                padding=8
            ),

            nn.ReLU(),

            nn.Conv1d(
                64,
                64,
                kernel_size=5,
                dilation=8,
                padding=16
            ),

            nn.ReLU()
        )

        self.classifier = nn.Linear(
            64,
            N_CLASSES
        )

    def forward(
        self,
        x
    ):

        x = self.features(
            x
        )

        x = x.mean(
            dim=-1
        )

        return self.classifier(
            x
        )

# ============================================================
# 16. METRICS
# ============================================================

def calculate_metrics(
    y_true,
    y_pred,
    y_prob
):

    metrics = {}

    metrics["accuracy"] = (
        accuracy_score(
            y_true,
            y_pred
        )
    )

    metrics["balanced_accuracy"] = (
        balanced_accuracy_score(
            y_true,
            y_pred
        )
    )

    metrics["macro_f1"] = (
        f1_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        )
    )

    metrics["weighted_f1"] = (
        f1_score(
            y_true,
            y_pred,
            average="weighted",
            zero_division=0
        )
    )

    metrics["macro_precision"] = (
        precision_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        )
    )

    metrics["macro_recall"] = (
        recall_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        )
    )

    # --------------------------------------------------------
    # AUROC / AUPRC.
    # Only calculate if all required classes are present.
    # --------------------------------------------------------

    try:

        present_classes = np.unique(
            y_true
        )

        if len(present_classes) >= 2:

            onehot = np.eye(
                N_CLASSES
            )[y_true]

            metrics["macro_auroc"] = (
                roc_auc_score(
                    onehot,
                    y_prob,
                    average="macro",
                    multi_class="ovr"
                )
            )

        else:

            metrics[
                "macro_auroc"
            ] = np.nan

    except Exception:

        metrics[
            "macro_auroc"
        ] = np.nan

    try:

        onehot = np.eye(
            N_CLASSES
        )[y_true]

        metrics["macro_auprc"] = (
            average_precision_score(
                onehot,
                y_prob,
                average="macro"
            )
        )

    except Exception:

        metrics[
            "macro_auprc"
        ] = np.nan

    return metrics

# ============================================================
# 17. EVALUATION
# ============================================================

@torch.no_grad()
def evaluate(
    model,
    loader,
    criterion
):

    model.eval()

    total_loss = 0.0
    total_n = 0

    true_values = []
    pred_values = []
    probability_values = []

    for x, y in loader:

        x = x.to(
            DEVICE,
            non_blocking=True
        )

        y = y.to(
            DEVICE,
            non_blocking=True
        )

        logits = model(
            x
        )

        loss = criterion(
            logits,
            y
        )

        probabilities = (
            torch.softmax(
                logits,
                dim=1
            )
        )

        predictions = (
            probabilities.argmax(
                dim=1
            )
        )

        batch_size = y.size(0)

        total_loss += (
            loss.item()
            *
            batch_size
        )

        total_n += (
            batch_size
        )

        true_values.append(
            y.cpu().numpy()
        )

        pred_values.append(
            predictions.cpu().numpy()
        )

        probability_values.append(
            probabilities.cpu().numpy()
        )

    y_true = np.concatenate(
        true_values
    )

    y_pred = np.concatenate(
        pred_values
    )

    y_prob = np.concatenate(
        probability_values
    )

    metrics = calculate_metrics(
        y_true,
        y_pred,
        y_prob
    )

    metrics["loss"] = (
        total_loss /
        max(
            total_n,
            1
        )
    )

    return (
        metrics,
        y_true,
        y_pred,
        y_prob
    )

# ============================================================
# 18. CLASS WEIGHTS
# ============================================================

def make_class_weights(
    associations
):

    counts = np.zeros(
        N_CLASSES,
        dtype=np.float64
    )

    for (
        recording_id,
        row_idx,
        label
    ) in associations:

        counts[
            label
        ] += 1

    weights = np.zeros(
        N_CLASSES,
        dtype=np.float64
    )

    present = (
        counts > 0
    )

    weights[
        present
    ] = (
        1.0 /
        np.sqrt(
            counts[present]
        )
    )

    if present.any():

        weights[
            present
        ] /= (
            weights[
                present
            ].mean()
        )

    return (
        torch.tensor(
            weights,
            dtype=torch.float32,
            device=DEVICE
        ),
        counts
    )

# ============================================================
# 19. RNG CHECKPOINTING
# ============================================================

def capture_rng():

    state = {

        "python":
            random.getstate(),

        "numpy":
            np.random.get_state(),

        "torch":
            torch.get_rng_state()
    }

    if torch.cuda.is_available():

        state["cuda"] = (
            torch.cuda.get_rng_state_all()
        )

    return state


def restore_rng(
    state
):

    if state is None:
        return

    random.setstate(
        state["python"]
    )

    np.random.set_state(
        state["numpy"]
    )

    torch.set_rng_state(
        state["torch"]
    )

    if (
        torch.cuda.is_available()
        and
        "cuda" in state
    ):

        torch.cuda.set_rng_state_all(
            state["cuda"]
        )

# ============================================================
# 20. DETERMINISTIC BATCH ORDER
# ============================================================

def get_epoch_batches(
    n_samples,
    epoch,
    seed
):

    generator = torch.Generator()

    generator.manual_seed(
        seed + epoch
    )

    permutation = (
        torch.randperm(
            n_samples,
            generator=generator
        )
        .numpy()
    )

    batches = []

    for start in range(
        0,
        n_samples,
        BATCH_SIZE
    ):

        batches.append(
            permutation[
                start:
                min(
                    start + BATCH_SIZE,
                    n_samples
                )
            ]
        )

    return batches

# ============================================================
# 21. FILE HELPERS
# ============================================================

def subject_tag(
    subject
):

    return (
        f"subject_{subject}"
    )


def batch_checkpoint_path(
    subject
):

    return (
        CHECKPOINT_ROOT /
        f"{subject_tag(subject)}_batch.pt"
    )


def best_checkpoint_path(
    subject
):

    return (
        CHECKPOINT_ROOT /
        f"{subject_tag(subject)}_best.pt"
    )


def final_result_path(
    subject
):

    return (
        RESULT_ROOT /
        f"{subject_tag(subject)}_final.json"
    )


def history_path(
    subject
):

    return (
        RESULT_ROOT /
        f"{subject_tag(subject)}_history.csv"
    )

# ============================================================
# 22. TRAIN ONE SUBJECT
# ============================================================

def train_subject(
    subject
):

    print()
    print("=" * 70)
    print(
        f"STEP 6G — SUBJECT {subject}"
    )
    print("=" * 70)

    splits = subject_splits[
        subject
    ]

    # --------------------------------------------------------
    # Build associations.
    # --------------------------------------------------------

    train_assoc = (
        build_associations(
            splits["train"]
        )
    )

    val_assoc = (
        build_associations(
            splits["val"]
        )
    )

    test_assoc = (
        build_associations(
            splits["test"]
        )
    )

    print(
        "Train windows:",
        f"{len(train_assoc):,}"
    )

    print(
        "Val windows  :",
        f"{len(val_assoc):,}"
    )

    print(
        "Test windows :",
        f"{len(test_assoc):,}"
    )

    if (
        len(train_assoc) == 0
        or
        len(val_assoc) == 0
        or
        len(test_assoc) == 0
    ):

        raise RuntimeError(
            f"Subject {subject} "
            f"has an empty split."
        )

    # --------------------------------------------------------
    # Dataset / loaders.
    # --------------------------------------------------------

    train_dataset = CharacterDataset(
        train_assoc
    )

    val_dataset = CharacterDataset(
        val_assoc
    )

    test_dataset = CharacterDataset(
        test_assoc
    )

    val_loader = DataLoader(
        val_dataset,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=0,
        pin_memory=True
    )

    test_loader = DataLoader(
        test_dataset,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=0,
        pin_memory=True
    )

    # --------------------------------------------------------
    # Model.
    # --------------------------------------------------------

    model = CharacterCNN().to(
        DEVICE
    )

    class_weights, class_counts = (
        make_class_weights(
            train_assoc
        )
    )

    criterion = nn.CrossEntropyLoss(
        weight=class_weights
    )

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=LR,
        weight_decay=WEIGHT_DECAY
    )

    # --------------------------------------------------------
    # Checkpoints.
    # --------------------------------------------------------

    batch_ckpt = (
        batch_checkpoint_path(
            subject
        )
    )

    best_ckpt = (
        best_checkpoint_path(
            subject
        )
    )

    start_epoch = 1
    start_batch = 0

    best_val_f1 = -np.inf
    best_epoch = None
    patience_counter = 0

    history = []

    # --------------------------------------------------------
    # Resume if checkpoint exists.
    # --------------------------------------------------------

    if batch_ckpt.exists():

        print()
        print(
            "RESUMING FROM CHECKPOINT:"
        )

        print(
            batch_ckpt
        )

        checkpoint = torch.load(
            batch_ckpt,
            map_location=DEVICE,
            weights_only=False
        )

        model.load_state_dict(
            checkpoint[
                "model_state"
            ]
        )

        optimizer.load_state_dict(
            checkpoint[
                "optimizer_state"
            ]
        )

        start_epoch = int(
            checkpoint[
                "epoch"
            ]
        )

        start_batch = int(
            checkpoint[
                "batch_idx"
            ]
        )

        best_val_f1 = float(
            checkpoint[
                "best_val_f1"
            ]
        )

        best_epoch = checkpoint.get(
            "best_epoch"
        )

        patience_counter = int(
            checkpoint.get(
                "patience_counter",
                0
            )
        )

        history = checkpoint.get(
            "history",
            []
        )

        restore_rng(
            checkpoint.get(
                "rng_state"
            )
        )

        print(
            "Resume epoch:",
            start_epoch
        )

        print(
            "Resume batch:",
            start_batch
        )

        print(
            "Best validation Macro-F1:",
            f"{best_val_f1:.6f}"
        )

    else:

        print(
            "Starting from epoch 1."
        )

    n_train = len(
        train_dataset
    )

    n_batches = math.ceil(
        n_train /
        BATCH_SIZE
    )

    # ========================================================
    # TRAINING LOOP
    # ========================================================

    for epoch in range(
        start_epoch,
        MAX_EPOCHS + 1
    ):

        model.train()

        epoch_start = time.time()

        batches = get_epoch_batches(
            n_train,
            epoch,
            SEED
        )

        batch_begin = (
            start_batch
            if epoch == start_epoch
            else 0
        )

        running_loss = 0.0
        running_n = 0

        for batch_idx in range(
            batch_begin,
            n_batches
        ):

            indices = batches[
                batch_idx
            ]

            xs = []
            ys = []

            for index in indices:

                x, y = (
                    train_dataset[
                        int(index)
                    ]
                )

                xs.append(x)
                ys.append(y)

            x = torch.stack(
                xs
            ).to(
                DEVICE,
                non_blocking=True
            )

            y = torch.stack(
                ys
            ).to(
                DEVICE,
                non_blocking=True
            )

            optimizer.zero_grad(
                set_to_none=True
            )

            logits = model(
                x
            )

            loss = criterion(
                logits,
                y
            )

            loss.backward()

            optimizer.step()

            batch_size = y.size(0)

            running_loss += (
                loss.item()
                *
                batch_size
            )

            running_n += (
                batch_size
            )

            # ------------------------------------------------
            # Determine next checkpoint location.
            # ------------------------------------------------

            if (
                batch_idx + 1
                >= n_batches
            ):

                next_epoch = (
                    epoch + 1
                )

                next_batch = 0

            else:

                next_epoch = epoch

                next_batch = (
                    batch_idx + 1
                )

            # ------------------------------------------------
            # SAVE AFTER EVERY BATCH
            # ------------------------------------------------

            checkpoint_state = {

                "status":
                    "TRAINING",

                "subject":
                    subject,

                "epoch":
                    next_epoch,

                "batch_idx":
                    next_batch,

                "n_batches":
                    n_batches,

                "model_state":
                    model.state_dict(),

                "optimizer_state":
                    optimizer.state_dict(),

                "best_val_f1":
                    best_val_f1,

                "best_epoch":
                    best_epoch,

                "patience_counter":
                    patience_counter,

                "history":
                    history,

                "rng_state":
                    capture_rng(),

                "saved_at":
                    time.strftime(
                        "%Y-%m-%d %H:%M:%S"
                    )
            }

            torch.save(
                checkpoint_state,
                batch_ckpt
            )

            # ------------------------------------------------
            # Progress.
            # ------------------------------------------------

            if (
                batch_idx == batch_begin
                or
                (batch_idx + 1) % 20 == 0
                or
                batch_idx == n_batches - 1
            ):

                current_loss = (
                    running_loss /
                    max(
                        running_n,
                        1
                    )
                )

                print(
                    f"Subject {subject} | "
                    f"Epoch {epoch:02d}/"
                    f"{MAX_EPOCHS} | "
                    f"Batch {batch_idx+1:03d}/"
                    f"{n_batches} | "
                    f"Loss {current_loss:.5f}"
                )

            del x
            del y
            del logits
            del loss

        # ----------------------------------------------------
        # Reset batch index after completed epoch.
        # ----------------------------------------------------

        start_batch = 0

        train_loss = (
            running_loss /
            max(
                running_n,
                1
            )
        )

        # ----------------------------------------------------
        # Validation.
        # ----------------------------------------------------

        (
            val_metrics,
            _,
            _,
            _
        ) = evaluate(
            model,
            val_loader,
            criterion
        )

        epoch_time = (
            time.time()
            -
            epoch_start
        )

        print()
        print(
            f"Epoch {epoch:02d} COMPLETE"
        )

        print(
            f"Train Loss : "
            f"{train_loss:.6f}"
        )

        print(
            f"Val Loss   : "
            f"{val_metrics['loss']:.6f}"
        )

        print(
            f"Val Acc    : "
            f"{val_metrics['accuracy']:.6f}"
        )

        print(
            f"Val BalAcc : "
            f"{val_metrics['balanced_accuracy']:.6f}"
        )

        print(
            f"Val MacroF1: "
            f"{val_metrics['macro_f1']:.6f}"
        )

        print(
            f"Time       : "
            f"{epoch_time/60:.2f} min"
        )

        # ----------------------------------------------------
        # Save history.
        # ----------------------------------------------------

        record = {

            "subject":
                subject,

            "epoch":
                epoch,

            "train_loss":
                train_loss,

            "val_loss":
                val_metrics["loss"],

            "val_accuracy":
                val_metrics[
                    "accuracy"
                ],

            "val_balanced_accuracy":
                val_metrics[
                    "balanced_accuracy"
                ],

            "val_macro_f1":
                val_metrics[
                    "macro_f1"
                ],

            "val_weighted_f1":
                val_metrics[
                    "weighted_f1"
                ],

            "val_macro_precision":
                val_metrics[
                    "macro_precision"
                ],

            "val_macro_recall":
                val_metrics[
                    "macro_recall"
                ],

            "val_macro_auroc":
                val_metrics[
                    "macro_auroc"
                ],

            "val_macro_auprc":
                val_metrics[
                    "macro_auprc"
                ],

            "epoch_time_sec":
                epoch_time
        }

        history.append(
            record
        )

        # ----------------------------------------------------
        # Best validation model.
        # ----------------------------------------------------

        if (
            val_metrics["macro_f1"]
            >
            best_val_f1
        ):

            best_val_f1 = (
                val_metrics[
                    "macro_f1"
                ]
            )

            best_epoch = (
                epoch
            )

            patience_counter = 0

            best_state = {

                "status":
                    "BEST",

                "subject":
                    subject,

                "epoch":
                    epoch,

                "best_val_f1":
                    best_val_f1,

                "model_state":
                    copy.deepcopy(
                        model.state_dict()
                    ),

                "optimizer_state":
                    copy.deepcopy(
                        optimizer.state_dict()
                    ),

                "history":
                    history,

                "class_counts":
                    class_counts,

                "train_recordings":
                    splits["train"],

                "val_recordings":
                    splits["val"],

                "test_recordings":
                    splits["test"]
            }

            torch.save(
                best_state,
                best_ckpt
            )

            print(
                "✓ NEW BEST MODEL"
            )

            print(
                f"  Val Macro-F1 = "
                f"{best_val_f1:.6f}"
            )

        else:

            patience_counter += 1

            print(
                f"No improvement: "
                f"{patience_counter}/"
                f"{PATIENCE}"
            )

        # ----------------------------------------------------
        # Epoch checkpoint.
        # ----------------------------------------------------

        epoch_state = {

            "status":
                "TRAINING",

            "subject":
                subject,

            "epoch":
                epoch + 1,

            "batch_idx":
                0,

            "n_batches":
                n_batches,

            "model_state":
                model.state_dict(),

            "optimizer_state":
                optimizer.state_dict(),

            "best_val_f1":
                best_val_f1,

            "best_epoch":
                best_epoch,

            "patience_counter":
                patience_counter,

            "history":
                history,

            "rng_state":
                capture_rng()
        }

        torch.save(
            epoch_state,
            batch_ckpt
        )

        pd.DataFrame(
            history
        ).to_csv(
            history_path(
                subject
            ),
            index=False
        )

        # ----------------------------------------------------
        # Early stopping.
        # ----------------------------------------------------

        if (
            patience_counter
            >= PATIENCE
        ):

            print(
                "Early stopping."
            )

            break

        torch.cuda.empty_cache()

    # ========================================================
    # LOAD BEST MODEL
    # ========================================================

    if not best_ckpt.exists():

        raise RuntimeError(
            f"No best checkpoint "
            f"for subject {subject}."
        )

    best = torch.load(
        best_ckpt,
        map_location=DEVICE,
        weights_only=False
    )

    model.load_state_dict(
        best["model_state"]
    )

    # ========================================================
    # FROZEN TEST
    # ========================================================

    print()
    print(
        "Running frozen test evaluation..."
    )

    (
        test_metrics,
        y_true,
        y_pred,
        y_prob
    ) = evaluate(
        model,
        test_loader,
        criterion
    )

    # --------------------------------------------------------
    # Prediction distribution.
    # --------------------------------------------------------

    prediction_counts = np.bincount(
        y_pred,
        minlength=N_CLASSES
    )

    true_counts = np.bincount(
        y_true,
        minlength=N_CLASSES
    )

    prediction_distribution = {
        LABELS[i]:
            int(prediction_counts[i])
        for i in range(
            N_CLASSES
        )
        if prediction_counts[i] > 0
    }

    true_distribution = {
        LABELS[i]:
            int(true_counts[i])
        for i in range(
            N_CLASSES
        )
        if true_counts[i] > 0
    }

    # ========================================================
    # SAVE FINAL RESULT
    # ========================================================

    result = {

        "status":
            "COMPLETED",

        "experiment":
            "Step 6G within-subject diagnostic",

        "subject":
            subject,

        "train_recordings":
            splits["train"],

        "val_recordings":
            splits["val"],

        "test_recordings":
            splits["test"],

        "train_samples":
            len(train_assoc),

        "val_samples":
            len(val_assoc),

        "test_samples":
            len(test_assoc),

        "best_epoch":
            int(best_epoch),

        "best_validation_macro_f1":
            float(best_val_f1),

        "test_loss":
            float(
                test_metrics[
                    "loss"
                ]
            ),

        "test_accuracy":
            float(
                test_metrics[
                    "accuracy"
                ]
            ),

        "test_balanced_accuracy":
            float(
                test_metrics[
                    "balanced_accuracy"
                ]
            ),

        "test_macro_f1":
            float(
                test_metrics[
                    "macro_f1"
                ]
            ),

        "test_weighted_f1":
            float(
                test_metrics[
                    "weighted_f1"
                ]
            ),

        "test_macro_precision":
            float(
                test_metrics[
                    "macro_precision"
                ]
            ),

        "test_macro_recall":
            float(
                test_metrics[
                    "macro_recall"
                ]
            ),

        "test_auroc":
            (
                None
                if np.isnan(
                    test_metrics[
                        "macro_auroc"
                    ]
                )
                else float(
                    test_metrics[
                        "macro_auroc"
                    ]
                )
            ),

        "test_auprc":
            (
                None
                if np.isnan(
                    test_metrics[
                        "macro_auprc"
                    ]
                )
                else float(
                    test_metrics[
                        "macro_auprc"
                    ]
                )
            ),

        "prediction_distribution":
            prediction_distribution,

        "true_distribution":
            true_distribution,

        "architecture":
            "Conv1D 61->64 -> "
            "Conv1D 64->64 d2 -> "
            "Conv1D 64->64 d4 -> "
            "Conv1D 64->64 d8 -> "
            "global temporal mean -> "
            "Linear 64->29",

        "normalization":
            "per-window channel-wise z-score",

        "batch_size":
            BATCH_SIZE,

        "learning_rate":
            LR,

        "weight_decay":
            WEIGHT_DECAY,

        "max_epochs":
            MAX_EPOCHS,

        "patience":
            PATIENCE,

        "seed":
            SEED,

        "selection_metric":
            "validation_macro_f1",

        "split_unit":
            "recording",

        "split_strategy":
            "deterministic approximately 60/20/20 "
            "recording-level split"
    }

    with open(
        final_result_path(
            subject
        ),
        "w"
    ) as f:

        json.dump(
            result,
            f,
            indent=2
        )

    pd.DataFrame(
        history
    ).to_csv(
        history_path(
            subject
        ),
        index=False
    )

    # --------------------------------------------------------
    # Save detailed predictions.
    # --------------------------------------------------------

    prediction_df = pd.DataFrame({

        "true_label_id":
            y_true,

        "predicted_label_id":
            y_pred,

        "true_label":
            [
                LABELS[i]
                for i in y_true
            ],

        "predicted_label":
            [
                LABELS[i]
                for i in y_pred
            ],

        "correct":
            (
                y_true == y_pred
            ),

        "confidence":
            np.max(
                y_prob,
                axis=1
            )
    })

    prediction_df.to_csv(
        RESULT_ROOT /
        f"{subject_tag(subject)}_predictions.csv",
        index=False
    )

    # --------------------------------------------------------
    # Delete resumable checkpoint ONLY after final result
    # has been safely written.
    # --------------------------------------------------------

    if batch_ckpt.exists():

        batch_ckpt.unlink()

    print()
    print("=" * 70)

    print(
        f"✓ SUBJECT {subject} COMPLETED"
    )

    print("=" * 70)

    print(
        f"Best Epoch       : "
        f"{best_epoch}"
    )

    print(
        f"Best Val MacroF1 : "
        f"{best_val_f1:.6f}"
    )

    print(
        f"Test Accuracy    : "
        f"{test_metrics['accuracy']:.6f}"
    )

    print(
        f"Test BalancedAcc : "
        f"{test_metrics['balanced_accuracy']:.6f}"
    )

    print(
        f"Test MacroF1     : "
        f"{test_metrics['macro_f1']:.6f}"
    )

    print(
        f"Test AUROC       : "
        f"{test_metrics['macro_auroc']}"
    )

    print(
        f"Test AUPRC       : "
        f"{test_metrics['macro_auprc']}"
    )

    return result

# ============================================================
# 23. CHECK COMPLETED SUBJECTS
# ============================================================

def completed_subjects():

    completed = []

    for subject in TEST_SUBJECTS:

        path = (
            final_result_path(
                subject
            )
        )

        if not path.exists():
            continue

        try:

            with open(
                path,
                "r"
            ) as f:

                data = json.load(
                    f
                )

            if (
                data.get("status")
                ==
                "COMPLETED"
            ):

                completed.append(
                    subject
                )

        except Exception:

            pass

    return completed

# ============================================================
# 24. AUTOMATICALLY SELECT NEXT SUBJECT
# ============================================================

done = completed_subjects()

remaining = [
    subject
    for subject in TEST_SUBJECTS
    if subject not in done
]

print()
print("=" * 70)
print("STEP 6G CURRENT STATUS")
print("=" * 70)

print(
    "Completed subjects:",
    done
)

print(
    "Remaining subjects:",
    remaining
)

print()

if len(remaining) == 0:

    print(
        "ALL STEP 6G SUBJECTS COMPLETED."
    )

else:

    next_subject = (
        remaining[0]
    )

    print(
        f"NEXT SUBJECT: "
        f"{next_subject}"
    )

    train_subject(
        next_subject
    )

# ============================================================
# 25. CREATE STATUS TABLE
# ============================================================

status_rows = []

for subject in TEST_SUBJECTS:

    result_path = (
        final_result_path(
            subject
        )
    )

    checkpoint_path = (
        batch_checkpoint_path(
            subject
        )
    )

    if result_path.exists():

        try:

            with open(
                result_path,
                "r"
            ) as f:

                data = json.load(
                    f
                )

            status_rows.append({

                "subject":
                    subject,

                "status":
                    data.get(
                        "status",
                        "UNKNOWN"
                    ),

                "best_epoch":
                    data.get(
                        "best_epoch"
                    ),

                "val_macro_f1":
                    data.get(
                        "best_validation_macro_f1"
                    ),

                "test_accuracy":
                    data.get(
                        "test_accuracy"
                    ),

                "test_balanced_accuracy":
                    data.get(
                        "test_balanced_accuracy"
                    ),

                "test_macro_f1":
                    data.get(
                        "test_macro_f1"
                    ),

                "test_auroc":
                    data.get(
                        "test_auroc"
                    ),

                "test_auprc":
                    data.get(
                        "test_auprc"
                    )
            })

        except Exception:

            status_rows.append({

                "subject":
                    subject,

                "status":
                    "ERROR"
            })

    elif checkpoint_path.exists():

        try:

            checkpoint = torch.load(
                checkpoint_path,
                map_location="cpu",
                weights_only=False
            )

            status_rows.append({

                "subject":
                    subject,

                "status":
                    "RESUMABLE",

                "epoch":
                    checkpoint.get(
                        "epoch"
                    ),

                "next_batch":
                    checkpoint.get(
                        "batch_idx"
                    ),

                "best_val_macro_f1":
                    checkpoint.get(
                        "best_val_f1"
                    )
            })

        except Exception:

            status_rows.append({

                "subject":
                    subject,

                "status":
                    "CHECKPOINT_ERROR"
            })

    else:

        status_rows.append({

            "subject":
                subject,

            "status":
                "PENDING"
        })

status_df = pd.DataFrame(
    status_rows
)

print()
print("=" * 70)
print("STEP 6G CURRENT STATUS")
print("=" * 70)

display(
    status_df
)

status_df.to_csv(
    RESULT_ROOT /
    "step6g_status.csv",
    index=False
)

print()
print(
    "If Colab disconnects:"
)

print(
    "1. Reconnect GPU."
)

print(
    "2. Run this SAME cell."
)

print(
    "3. It resumes the current subject "
    "from its latest batch checkpoint."
)

print()
print(
    "Results directory:"
)

print(
    RESULT_ROOT
)

print()
print(
    "Checkpoint directory:"
)

print(
    CHECKPOINT_ROOT
)
```

```python
# ============================================================
# STEP 7A — CONTINUE AFTER FEATURE EXTRACTION
# No EEG feature extraction is repeated.
# ============================================================

import numpy as np
import pandas as pd

from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
)

# ------------------------------------------------------------
# 1. VERIFY EXISTING DATA
# ------------------------------------------------------------

if "h7" not in globals():
    raise RuntimeError(
        "H7 dataframe is missing. "
        "Run the full corrected Step 7A cell."
    )

if "X_eeg" not in globals():
    raise RuntimeError(
        "X_eeg is missing. "
        "Run the full corrected Step 7A cell."
    )

print("=" * 70)
print("STEP 7A — CONTINUING FROM EXTRACTED EEG FEATURES")
print("=" * 70)

print(
    "H7 rows      :",
    len(h7)
)

print(
    "EEG features :",
    X_eeg.shape
)

assert len(h7) == len(X_eeg), (
    "H7 and EEG feature counts do not match."
)

# ------------------------------------------------------------
# 2. RECREATE SPLIT MASKS
# ------------------------------------------------------------

split_array = (
    h7["split"]
    .astype(str)
    .to_numpy()
)

train_mask = (
    split_array == "train"
)

val_mask = (
    split_array == "val"
)

test_mask = (
    split_array == "test"
)

print("\nSplit sizes:")
print(
    "Train:",
    int(train_mask.sum())
)
print(
    "Val  :",
    int(val_mask.sum())
)
print(
    "Test :",
    int(test_mask.sum())
)

# ------------------------------------------------------------
# 3. CORRECT NUMERIC TARGET
# ------------------------------------------------------------

# true_label contains strings such as "l", "a", "<space>".
# true_class contains the required numeric 0..28 labels.

classes = (
    h7["true_class"]
    .astype(int)
    .to_numpy()
)

y = (
    h7["utility_label"]
    .astype(int)
    .to_numpy()
)

print("\nTarget:")
print(
    "true_class range:",
    int(classes.min()),
    "to",
    int(classes.max())
)

print(
    "Number of classes:",
    len(np.unique(classes))
)

print(
    "Utility labels:",
    np.unique(y)
)

# ------------------------------------------------------------
# 4. METRIC HELPERS
# ------------------------------------------------------------

def safe_auc(y_true, p):
    if len(np.unique(y_true)) < 2:
        return np.nan
    return roc_auc_score(y_true, p)


def safe_auprc(y_true, p):
    if len(np.unique(y_true)) < 2:
        return np.nan
    return average_precision_score(y_true, p)


def evaluate(y_true, p, threshold):
    pred = (
        p >= threshold
    ).astype(np.int8)

    return {
        "accuracy":
            accuracy_score(y_true, pred),

        "balanced_accuracy":
            balanced_accuracy_score(y_true, pred),

        "precision":
            precision_score(
                y_true,
                pred,
                zero_division=0
            ),

        "recall":
            recall_score(
                y_true,
                pred,
                zero_division=0
            ),

        "f1":
            f1_score(
                y_true,
                pred,
                zero_division=0
            ),

        "auroc":
            safe_auc(y_true, p),

        "auprc":
            safe_auprc(y_true, p),
    }


def select_threshold(y_true, p):
    best_t = 0.5
    best_f1 = -1

    for t in np.arange(
        0.05,
        0.951,
        0.01
    ):

        pred = (
            p >= t
        ).astype(np.int8)

        f1 = f1_score(
            y_true,
            pred,
            zero_division=0
        )

        if f1 > best_f1:
            best_f1 = f1
            best_t = float(t)

    return best_t, best_f1


# ============================================================
# 5. CLASS-ONLY BASELINE
# ============================================================

print("\n" + "=" * 70)
print("A. CLASS-ONLY BASELINE")
print("=" * 70)

train_class_rates = (
    h7.loc[train_mask]
    .groupby("true_class")[
        "utility_label"
    ]
    .mean()
)

class_results = []

for name, mask in [
    ("train", train_mask),
    ("val", val_mask),
    ("test", test_mask),
]:

    yy = y[mask]

    p = (
        h7.loc[
            mask,
            "true_class"
        ]
        .map(train_class_rates)
        .fillna(
            y[train_mask].mean()
        )
        .to_numpy()
    )

    if name == "train":
        class_threshold, _ = (
            select_threshold(
                yy,
                p
            )
        )

    metrics = evaluate(
        yy,
        p,
        class_threshold
    )

    class_results.append({
        "model":
            "class_only",
        "split":
            name,
        "threshold":
            class_threshold,
        **metrics
    })

    print(
        f"{name:5s} | "
        f"AUROC={metrics['auroc']:.6f} | "
        f"AUPRC={metrics['auprc']:.6f} | "
        f"F1={metrics['f1']:.6f} | "
        f"BalAcc={metrics['balanced_accuracy']:.6f}"
    )


# ============================================================
# 6. CONFIDENCE-ONLY BASELINE
# ============================================================

confidence_results = []

if "decoder_confidence" in h7.columns:

    print("\n" + "=" * 70)
    print("B. CONFIDENCE-ONLY BASELINE")
    print("=" * 70)

    confidence = (
        h7["decoder_confidence"]
        .astype(float)
        .to_numpy()
    )

    for name, mask in [
        ("train", train_mask),
        ("val", val_mask),
        ("test", test_mask),
    ]:

        yy = y[mask]
        p = confidence[mask]

        if name == "train":
            confidence_threshold, _ = (
                select_threshold(
                    yy,
                    p
                )
            )

        metrics = evaluate(
            yy,
            p,
            confidence_threshold
        )

        confidence_results.append({
            "model":
                "confidence_only",
            "split":
                name,
            "threshold":
                confidence_threshold,
            **metrics
        })

        print(
            f"{name:5s} | "
            f"AUROC={metrics['auroc']:.6f} | "
            f"AUPRC={metrics['auprc']:.6f} | "
            f"F1={metrics['f1']:.6f} | "
            f"BalAcc={metrics['balanced_accuracy']:.6f}"
        )


# ============================================================
# 7. EEG-ONLY MODEL
# ============================================================

print("\n" + "=" * 70)
print("C. EEG-ONLY UTILITY PREDICTABILITY")
print("=" * 70)

eeg_model = Pipeline([
    (
        "scaler",
        StandardScaler()
    ),
    (
        "classifier",
        LogisticRegression(
            max_iter=1000,
            class_weight="balanced",
            random_state=42,
            n_jobs=-1
        )
    )
])

eeg_model.fit(
    X_eeg[train_mask],
    y[train_mask]
)

p_train = eeg_model.predict_proba(
    X_eeg[train_mask]
)[:, 1]

p_val = eeg_model.predict_proba(
    X_eeg[val_mask]
)[:, 1]

p_test = eeg_model.predict_proba(
    X_eeg[test_mask]
)[:, 1]

eeg_threshold, eeg_val_f1 = (
    select_threshold(
        y[val_mask],
        p_val
    )
)

print(
    f"Selected threshold: {eeg_threshold:.3f}"
)

print(
    f"Validation F1: {eeg_val_f1:.6f}"
)

eeg_results = []

for name, yy, p in [
    ("train", y[train_mask], p_train),
    ("val", y[val_mask], p_val),
    ("test", y[test_mask], p_test),
]:

    metrics = evaluate(
        yy,
        p,
        eeg_threshold
    )

    eeg_results.append({
        "model":
            "EEG_only",
        "split":
            name,
        "threshold":
            eeg_threshold,
        **metrics
    })

    print(
        f"{name:5s} | "
        f"AUROC={metrics['auroc']:.6f} | "
        f"AUPRC={metrics['auprc']:.6f} | "
        f"F1={metrics['f1']:.6f} | "
        f"BalAcc={metrics['balanced_accuracy']:.6f}"
    )


# ============================================================
# 8. EEG + CHARACTER CLASS
# ============================================================

print("\n" + "=" * 70)
print("D. EEG + CHARACTER CLASS")
print("=" * 70)

class_onehot = np.zeros(
    (
        len(classes),
        29
    ),
    dtype=np.float32
)

rows = np.arange(
    len(classes)
)

valid = (
    (classes >= 0)
    &
    (classes < 29)
)

class_onehot[
    rows[valid],
    classes[valid]
] = 1.0

X_combined = np.concatenate(
    [
        X_eeg,
        class_onehot
    ],
    axis=1
)

print(
    "Combined features:",
    X_combined.shape
)

combined_model = Pipeline([
    (
        "scaler",
        StandardScaler()
    ),
    (
        "classifier",
        LogisticRegression(
            max_iter=1000,
            class_weight="balanced",
            random_state=42,
            n_jobs=-1
        )
    )
])

combined_model.fit(
    X_combined[train_mask],
    y[train_mask]
)

pc_train = combined_model.predict_proba(
    X_combined[train_mask]
)[:, 1]

pc_val = combined_model.predict_proba(
    X_combined[val_mask]
)[:, 1]

pc_test = combined_model.predict_proba(
    X_combined[test_mask]
)[:, 1]

combined_threshold, combined_val_f1 = (
    select_threshold(
        y[val_mask],
        pc_val
    )
)

print(
    f"Selected threshold: {combined_threshold:.3f}"
)

print(
    f"Validation F1: {combined_val_f1:.6f}"
)

combined_results = []

for name, yy, p in [
    ("train", y[train_mask], pc_train),
    ("val", y[val_mask], pc_val),
    ("test", y[test_mask], pc_test),
]:

    metrics = evaluate(
        yy,
        p,
        combined_threshold
    )

    combined_results.append({
        "model":
            "EEG_plus_class",
        "split":
            name,
        "threshold":
            combined_threshold,
        **metrics
    })

    print(
        f"{name:5s} | "
        f"AUROC={metrics['auroc']:.6f} | "
        f"AUPRC={metrics['auprc']:.6f} | "
        f"F1={metrics['f1']:.6f} | "
        f"BalAcc={metrics['balanced_accuracy']:.6f}"
    )


# ============================================================
# 9. SHUFFLED-LABEL CONTROL
# ============================================================

print("\n" + "=" * 70)
print("E. SHUFFLED-LABEL CONTROL")
print("=" * 70)

shuffle_results = []

for seed in [42, 43, 44]:

    rng = np.random.default_rng(seed)

    shuffled_y = (
        y[train_mask]
        .copy()
    )

    rng.shuffle(
        shuffled_y
    )

    model = Pipeline([
        (
            "scaler",
            StandardScaler()
        ),
        (
            "classifier",
            LogisticRegression(
                max_iter=1000,
                class_weight="balanced",
                random_state=seed,
                n_jobs=-1
            )
        )
    ])

    model.fit(
        X_eeg[train_mask],
        shuffled_y
    )

    p = model.predict_proba(
        X_eeg[test_mask]
    )[:, 1]

    auc = safe_auc(
        y[test_mask],
        p
    )

    auprc = safe_auprc(
        y[test_mask],
        p
    )

    shuffle_results.append({
        "seed":
            seed,
        "test_auroc":
            auc,
        "test_auprc":
            auprc
    })

    print(
        f"Seed {seed}: "
        f"AUROC={auc:.6f} | "
        f"AUPRC={auprc:.6f}"
    )

shuffle_df = pd.DataFrame(
    shuffle_results
)


# ============================================================
# 10. FINAL COMPARISON
# ============================================================

results_df = pd.DataFrame(
    class_results
    + confidence_results
    + eeg_results
    + combined_results
)

print("\n" + "=" * 70)
print("STEP 7A FINAL COMPARISON")
print("=" * 70)

display(
    results_df[
        [
            "model",
            "split",
            "threshold",
            "accuracy",
            "balanced_accuracy",
            "precision",
            "recall",
            "f1",
            "auroc",
            "auprc"
        ]
    ].round(6)
)

print(
    "\nSHUFFLED-LABEL CONTROL"
)

display(
    shuffle_df.round(6)
)


# ============================================================
# 11. CRITICAL DIAGNOSTIC
# ============================================================

def test_metric(
    model_name,
    metric
):

    row = results_df[
        (
            results_df["model"]
            == model_name
        )
        &
        (
            results_df["split"]
            == "test"
        )
    ]

    if len(row) == 0:
        return np.nan

    return float(
        row.iloc[0][metric]
    )


class_auc = test_metric(
    "class_only",
    "auroc"
)

eeg_auc = test_metric(
    "EEG_only",
    "auroc"
)

combined_auc = test_metric(
    "EEG_plus_class",
    "auroc"
)

class_auprc = test_metric(
    "class_only",
    "auprc"
)

eeg_auprc = test_metric(
    "EEG_only",
    "auprc"
)

combined_auprc = test_metric(
    "EEG_plus_class",
    "auprc"
)

print("\n" + "=" * 70)
print("CRITICAL STEP 7A RESULT")
print("=" * 70)

print(
    f"Class-only AUROC      : {class_auc:.6f}"
)

print(
    f"EEG-only AUROC        : {eeg_auc:.6f}"
)

print(
    f"EEG + class AUROC     : {combined_auc:.6f}"
)

print()

print(
    f"Class-only AUPRC      : {class_auprc:.6f}"
)

print(
    f"EEG-only AUPRC        : {eeg_auprc:.6f}"
)

print(
    f"EEG + class AUPRC     : {combined_auprc:.6f}"
)

print()

print(
    f"Incremental EEG AUROC: "
    f"{combined_auc - class_auc:+.6f}"
)

print(
    f"Incremental EEG AUPRC: "
    f"{combined_auprc - class_auprc:+.6f}"
)


# ============================================================
# 12. SAVE
# ============================================================

BASE = Path(
    "/content/gdrive/MyDrive/NeuroSentinel"
)

RESULT_ROOT = (
    BASE /
    "results" /
    "step7a_utility_predictability"
)

RESULT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

results_df.to_csv(
    RESULT_ROOT /
    "step7a_model_comparison.csv",
    index=False
)

shuffle_df.to_csv(
    RESULT_ROOT /
    "step7a_shuffled_control.csv",
    index=False
)

print("\nSaved results:")
print(
    RESULT_ROOT /
    "step7a_model_comparison.csv"
)

print(
    RESULT_ROOT /
    "step7a_shuffled_control.csv"
)

print("\n" + "=" * 70)
print("STEP 7A COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# STEP 7B — UTILITY-LABEL REDESIGN
# ============================================================
#
# Goal:
#   Determine whether a less class-confounded utility definition
#   is more predictable from EEG.
#
# Candidate utility definitions:
#
#   U1: Original correctness
#       decoder_correct
#
#   U2: Global confidence-high
#       decoder confidence above TRAIN 75th percentile
#
#   U3: Class-conditioned confidence-high
#       confidence above the TRAIN median confidence for that
#       character class
#
#   U4: Class-conditioned confidence top-quartile
#       confidence above the TRAIN 75th percentile for that class
#
#   U5: Class-conditioned confidence residual
#       confidence - TRAIN class median confidence > 0
#
# U3/U4/U5 are diagnostic labels.
# They are NOT directly deployable because the true character
# class is not known before the expensive decoder runs.
#
# Selection:
#   - label definition / thresholds use TRAIN only
#   - model threshold selected using VAL only
#   - TEST is evaluated once after freezing
#
# ============================================================

import os
import json
from pathlib import Path

import numpy as np
import pandas as pd

from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
)

# ------------------------------------------------------------
# 0. PROJECT PATH
# ------------------------------------------------------------

PROJECT_ROOT = Path(
    "/content/gdrive/MyDrive/NeuroSentinel"
)

RESULT_ROOT = (
    PROJECT_ROOT /
    "results" /
    "step7b_utility_redesign"
)

RESULT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

print("=" * 75)
print("STEP 7B — UTILITY-LABEL REDESIGN")
print("=" * 75)

# ------------------------------------------------------------
# 1. LOAD H7 MANIFEST
# ------------------------------------------------------------

H7_PATH = (
    PROJECT_ROOT /
    "results" /
    "phase5h7" /
    "phase5h7_utility_label_manifest.csv"
)

if not H7_PATH.exists():
    raise FileNotFoundError(
        f"H7 manifest not found:\n{H7_PATH}"
    )

h7 = pd.read_csv(H7_PATH)

print("\nH7 manifest:")
print(H7_PATH)

print(
    "Shape:",
    h7.shape
)

print(
    "Columns:",
    list(h7.columns)
)

# ------------------------------------------------------------
# 2. VERIFY / LOAD EEG FEATURES
# ------------------------------------------------------------
#
# X_eeg was generated in Step 7A.
# If it still exists in the Colab runtime, reuse it.
#
# We deliberately do not silently reconstruct a different
# feature definition because Step 7A's 43-feature representation
# should remain fixed for a fair comparison.

if "X_eeg" not in globals():

    # Try loading a previously saved feature matrix if present.
    possible_feature_paths = [
        RESULT_ROOT / "X_eeg.npy",
        PROJECT_ROOT / "results" /
            "step7a_utility_predictability" /
            "X_eeg.npy",
    ]

    found_feature_path = None

    for p in possible_feature_paths:
        if p.exists():
            found_feature_path = p
            break

    if found_feature_path is not None:

        print(
            "\nLoading EEG features:",
            found_feature_path
        )

        X_eeg = np.load(
            found_feature_path
        )

    else:

        raise RuntimeError(
            "\nX_eeg is not available in the current runtime.\n"
            "Run the Step 7A EEG feature-extraction cell first.\n"
            "Do NOT replace it with a different feature definition "
            "for this comparison."
        )

print(
    "\nEEG feature matrix:",
    X_eeg.shape
)

if len(h7) != len(X_eeg):
    raise RuntimeError(
        f"H7 rows ({len(h7)}) != EEG rows ({len(X_eeg)})"
    )

# ------------------------------------------------------------
# 3. BASIC ARRAYS
# ------------------------------------------------------------

split = (
    h7["split"]
    .astype(str)
    .to_numpy()
)

train_mask = split == "train"
val_mask = split == "val"
test_mask = split == "test"

true_class = (
    h7["true_class"]
    .astype(int)
    .to_numpy()
)

confidence = (
    h7["decoder_confidence"]
    .astype(float)
    .to_numpy()
)

original_correct = (
    h7["decoder_correct"]
    .astype(int)
    .to_numpy()
)

print("\nSplit sizes:")
print(
    "Train:",
    train_mask.sum()
)
print(
    "Val  :",
    val_mask.sum()
)
print(
    "Test :",
    test_mask.sum()
)

print(
    "\nConfidence range:",
    confidence.min(),
    "to",
    confidence.max()
)

# ------------------------------------------------------------
# 4. METRIC HELPERS
# ------------------------------------------------------------

def safe_auc(y_true, p):

    if len(np.unique(y_true)) < 2:
        return np.nan

    return roc_auc_score(
        y_true,
        p
    )


def safe_auprc(y_true, p):

    if len(np.unique(y_true)) < 2:
        return np.nan

    return average_precision_score(
        y_true,
        p
    )


def select_threshold(
    y_true,
    probability
):

    best_threshold = 0.50
    best_f1 = -1.0

    for threshold in np.arange(
        0.05,
        0.951,
        0.01
    ):

        pred = (
            probability >= threshold
        ).astype(np.int8)

        f1 = f1_score(
            y_true,
            pred,
            zero_division=0
        )

        if f1 > best_f1:

            best_f1 = f1
            best_threshold = float(
                threshold
            )

    return (
        best_threshold,
        best_f1
    )


def evaluate_predictions(
    y_true,
    probability,
    threshold
):

    prediction = (
        probability >= threshold
    ).astype(np.int8)

    return {
        "accuracy":
            accuracy_score(
                y_true,
                prediction
            ),

        "balanced_accuracy":
            balanced_accuracy_score(
                y_true,
                prediction
            ),

        "precision":
            precision_score(
                y_true,
                prediction,
                zero_division=0
            ),

        "recall":
            recall_score(
                y_true,
                prediction,
                zero_division=0
            ),

        "f1":
            f1_score(
                y_true,
                prediction,
                zero_division=0
            ),

        "auroc":
            safe_auc(
                y_true,
                probability
            ),

        "auprc":
            safe_auprc(
                y_true,
                probability
            ),
    }


# ============================================================
# 5. TRAIN-ONLY CLASS-CONDITIONED CONFIDENCE STATISTICS
# ============================================================

print("\n" + "=" * 75)
print("TRAIN-ONLY CONFIDENCE STATISTICS")
print("=" * 75)

train_classes = true_class[train_mask]
train_conf = confidence[train_mask]

class_median = {}
class_q75 = {}

for c in range(29):

    values = train_conf[
        train_classes == c
    ]

    if len(values) == 0:
        continue

    class_median[c] = float(
        np.median(values)
    )

    class_q75[c] = float(
        np.percentile(
            values,
            75
        )
    )

print(
    "Classes with training statistics:",
    len(class_median)
)

# ------------------------------------------------------------
# GLOBAL TRAIN QUANTILES
# ------------------------------------------------------------

global_q50 = float(
    np.percentile(
        train_conf,
        50
    )
)

global_q75 = float(
    np.percentile(
        train_conf,
        75
    )
)

print(
    f"Global train median : {global_q50:.6f}"
)

print(
    f"Global train Q75    : {global_q75:.6f}"
)


# ============================================================
# 6. CREATE CANDIDATE UTILITY LABELS
# ============================================================

candidate_labels = {}

# ------------------------------------------------------------
# U1 — ORIGINAL CORRECTNESS
# ------------------------------------------------------------

candidate_labels[
    "U1_original_correctness"
] = original_correct.copy()


# ------------------------------------------------------------
# U2 — GLOBAL CONFIDENCE HIGH
# ------------------------------------------------------------

candidate_labels[
    "U2_global_confidence_high"
] = (
    confidence >= global_q75
).astype(np.int8)


# ------------------------------------------------------------
# U3 — CLASS-CONDITIONED ABOVE MEDIAN
# ------------------------------------------------------------

u3 = np.zeros(
    len(h7),
    dtype=np.int8
)

for c, median_value in class_median.items():

    mask = (
        true_class == c
    )

    u3[mask] = (
        confidence[mask]
        >= median_value
    ).astype(np.int8)

candidate_labels[
    "U3_class_conditioned_median"
] = u3


# ------------------------------------------------------------
# U4 — CLASS-CONDITIONED TOP QUARTILE
# ------------------------------------------------------------

u4 = np.zeros(
    len(h7),
    dtype=np.int8
)

for c, q75_value in class_q75.items():

    mask = (
        true_class == c
    )

    u4[mask] = (
        confidence[mask]
        >= q75_value
    ).astype(np.int8)

candidate_labels[
    "U4_class_conditioned_q75"
] = u4


# ------------------------------------------------------------
# U5 — CLASS-CONDITIONED CONFIDENCE RESIDUAL
# ------------------------------------------------------------
#
# This is equivalent to asking:
#
#   "Is this window more confident than the typical decoder
#    confidence for this character?"
#
# Positive residual = 1
# Negative residual = 0
#
# ------------------------------------------------------------

confidence_residual = np.zeros(
    len(h7),
    dtype=np.float32
)

for c, median_value in class_median.items():

    mask = (
        true_class == c
    )

    confidence_residual[mask] = (
        confidence[mask]
        - median_value
    )

u5 = (
    confidence_residual >= 0
).astype(np.int8)

candidate_labels[
    "U5_class_conditioned_residual"
] = u5


# ============================================================
# 7. CHARACTERIZE LABEL DISTRIBUTIONS
# ============================================================

print("\n" + "=" * 75)
print("CANDIDATE UTILITY-LABEL DISTRIBUTIONS")
print("=" * 75)

label_summary = []

for name, labels in candidate_labels.items():

    print(
        f"\n{name}"
    )

    for split_name, mask in [
        ("train", train_mask),
        ("val", val_mask),
        ("test", test_mask),
    ]:

        yy = labels[mask]

        rate = (
            float(np.mean(yy))
            if len(yy)
            else np.nan
        )

        label_summary.append({
            "utility_definition":
                name,
            "split":
                split_name,
            "N":
                len(yy),
            "high":
                int(yy.sum()),
            "low":
                int(len(yy) - yy.sum()),
            "high_rate":
                rate
        })

        print(
            f"  {split_name:5s}: "
            f"N={len(yy):6d}, "
            f"HIGH={yy.sum():6d}, "
            f"rate={rate:.6f}"
        )

label_summary_df = pd.DataFrame(
    label_summary
)

# ============================================================
# 8. EEG-ONLY PREDICTABILITY FOR EACH CANDIDATE
# ============================================================

print("\n" + "=" * 75)
print("EEG-ONLY PREDICTABILITY")
print("=" * 75)

all_results = []

models = {}

for utility_name, utility_target in (
    candidate_labels.items()
):

    print(
        "\n" + "-" * 75
    )

    print(
        utility_name
    )

    y_train = utility_target[
        train_mask
    ]

    y_val = utility_target[
        val_mask
    ]

    y_test = utility_target[
        test_mask
    ]

    # --------------------------------------------------------
    # Verify target is not degenerate
    # --------------------------------------------------------

    if len(np.unique(y_train)) < 2:

        print(
            "Skipped: training target has only one class."
        )

        continue

    # --------------------------------------------------------
    # Logistic EEG model
    # --------------------------------------------------------

    model = Pipeline([
        (
            "scaler",
            StandardScaler()
        ),
        (
            "classifier",
            LogisticRegression(
                max_iter=1000,
                class_weight="balanced",
                random_state=42,
                n_jobs=-1
            )
        )
    ])

    model.fit(
        X_eeg[train_mask],
        y_train
    )

    p_train = model.predict_proba(
        X_eeg[train_mask]
    )[:, 1]

    p_val = model.predict_proba(
        X_eeg[val_mask]
    )[:, 1]

    p_test = model.predict_proba(
        X_eeg[test_mask]
    )[:, 1]

    # --------------------------------------------------------
    # Threshold ONLY on validation
    # --------------------------------------------------------

    threshold, val_f1 = (
        select_threshold(
            y_val,
            p_val
        )
    )

    models[
        utility_name
    ] = model

    for split_name, yy, pp in [
        (
            "train",
            y_train,
            p_train
        ),
        (
            "val",
            y_val,
            p_val
        ),
        (
            "test",
            y_test,
            p_test
        ),
    ]:

        metrics = evaluate_predictions(
            yy,
            pp,
            threshold
        )

        all_results.append({
            "utility_definition":
                utility_name,

            "model":
                "EEG_only_logistic",

            "split":
                split_name,

            "threshold":
                threshold,

            **metrics
        })

        print(
            f"{split_name:5s} | "
            f"AUROC={metrics['auroc']:.6f} | "
            f"AUPRC={metrics['auprc']:.6f} | "
            f"F1={metrics['f1']:.6f} | "
            f"BalAcc={metrics['balanced_accuracy']:.6f}"
        )


# ============================================================
# 9. CONFIDENCE-ONLY PREDICTABILITY OF EACH REDESIGNED LABEL
# ============================================================
#
# This is NOT an EEG gate.
#
# It verifies whether the candidate label actually represents
# decoder confidence information as intended.
#
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("CONFIDENCE-ONLY DIAGNOSTIC")
print("=" * 75)

for utility_name, utility_target in (
    candidate_labels.items()
):

    y_train = utility_target[
        train_mask
    ]

    y_val = utility_target[
        val_mask
    ]

    y_test = utility_target[
        test_mask
    ]

    if len(np.unique(y_train)) < 2:
        continue

    # Confidence itself is used as the retrospective predictor.
    p_train = confidence[
        train_mask
    ]

    p_val = confidence[
        val_mask
    ]

    p_test = confidence[
        test_mask
    ]

    threshold, _ = select_threshold(
        y_val,
        p_val
    )

    for split_name, yy, pp in [
        (
            "train",
            y_train,
            p_train
        ),
        (
            "val",
            y_val,
            p_val
        ),
        (
            "test",
            y_test,
            p_test
        ),
    ]:

        metrics = evaluate_predictions(
            yy,
            pp,
            threshold
        )

        all_results.append({
            "utility_definition":
                utility_name,

            "model":
                "confidence_only",

            "split":
                split_name,

            "threshold":
                threshold,

            **metrics
        })


# ============================================================
# 10. SHUFFLED-LABEL CONTROL
# ============================================================

print("\n" + "=" * 75)
print("SHUFFLED-LABEL CONTROL")
print("=" * 75)

shuffle_results = []

# Use the most promising class-conditioned candidate:
# U5.
#
# If EEG performance is real, shuffling the training labels
# should destroy the relationship.

shuffle_target = candidate_labels[
    "U5_class_conditioned_residual"
]

for seed in [
    42,
    43,
    44
]:

    rng = np.random.default_rng(
        seed
    )

    shuffled_train_y = (
        shuffle_target[train_mask]
        .copy()
    )

    rng.shuffle(
        shuffled_train_y
    )

    shuffled_model = Pipeline([
        (
            "scaler",
            StandardScaler()
        ),
        (
            "classifier",
            LogisticRegression(
                max_iter=1000,
                class_weight="balanced",
                random_state=seed,
                n_jobs=-1
            )
        )
    ])

    shuffled_model.fit(
        X_eeg[train_mask],
        shuffled_train_y
    )

    p_test = shuffled_model.predict_proba(
        X_eeg[test_mask]
    )[:, 1]

    auc = safe_auc(
        shuffle_target[test_mask],
        p_test
    )

    auprc = safe_auprc(
        shuffle_target[test_mask],
        p_test
    )

    shuffle_results.append({
        "seed":
            seed,
        "test_auroc":
            auc,
        "test_auprc":
            auprc
    })

    print(
        f"Seed {seed}: "
        f"AUROC={auc:.6f} | "
        f"AUPRC={auprc:.6f}"
    )

shuffle_df = pd.DataFrame(
    shuffle_results
)


# ============================================================
# 11. FINAL COMPARISON TABLE
# ============================================================

results_df = pd.DataFrame(
    all_results
)

print("\n" + "=" * 75)
print("STEP 7B — FINAL EEG PREDICTABILITY")
print("=" * 75)

test_results = results_df[
    results_df["split"] == "test"
].copy()

display(
    test_results[
        [
            "utility_definition",
            "model",
            "threshold",
            "accuracy",
            "balanced_accuracy",
            "precision",
            "recall",
            "f1",
            "auroc",
            "auprc"
        ]
    ].sort_values(
        [
            "model",
            "auroc"
        ],
        ascending=[
            True,
            False
        ]
    ).round(6)
)


# ============================================================
# 12. VALIDATION RANKING
# ============================================================

print("\n" + "=" * 75)
print("VALIDATION RANKING — MODEL/LABEL SELECTION")
print("=" * 75)

val_results = results_df[
    (
        results_df["split"]
        == "val"
    )
    &
    (
        results_df["model"]
        == "EEG_only_logistic"
    )
].copy()

if len(val_results):

    display(
        val_results[
            [
                "utility_definition",
                "threshold",
                "accuracy",
                "balanced_accuracy",
                "precision",
                "recall",
                "f1",
                "auroc",
                "auprc"
            ]
        ]
        .sort_values(
            "auroc",
            ascending=False
        )
        .round(6)
    )


# ============================================================
# 13. TEST RESULTS FOR EEG-ONLY MODELS
# ============================================================

eeg_test = results_df[
    (
        results_df["split"]
        == "test"
    )
    &
    (
        results_df["model"]
        == "EEG_only_logistic"
    )
].copy()

print("\n" + "=" * 75)
print("EEG-ONLY TEST RESULTS")
print("=" * 75)

display(
    eeg_test[
        [
            "utility_definition",
            "threshold",
            "accuracy",
            "balanced_accuracy",
            "precision",
            "recall",
            "f1",
            "auroc",
            "auprc"
        ]
    ]
    .sort_values(
        "auroc",
        ascending=False
    )
    .round(6)
)


# ============================================================
# 14. AUTOMATIC INTERPRETATION
# ============================================================

print("\n" + "=" * 75)
print("STEP 7B INTERPRETATION")
print("=" * 75)

if len(val_results):

    best_val_row = (
        val_results
        .sort_values(
            "auroc",
            ascending=False
        )
        .iloc[0]
    )

    best_label = (
        best_val_row[
            "utility_definition"
        ]
    )

    best_val_auc = float(
        best_val_row[
            "auroc"
        ]
    )

    print(
        f"Validation-best utility label: "
        f"{best_label}"
    )

    print(
        f"Validation AUROC: "
        f"{best_val_auc:.6f}"
    )

    best_test = eeg_test[
        eeg_test[
            "utility_definition"
        ]
        == best_label
    ]

    if len(best_test):

        best_test_auc = float(
            best_test.iloc[0][
                "auroc"
            ]
        )

        best_test_auprc = float(
            best_test.iloc[0][
                "auprc"
            ]
        )

        print(
            f"Frozen test AUROC: "
            f"{best_test_auc:.6f}"
        )

        print(
            f"Frozen test AUPRC: "
            f"{best_test_auprc:.6f}"
        )

        if best_test_auc <= 0.55:

            print(
                "\nRESULT:"
            )

            print(
                "The redesigned utility target remains "
                "essentially unpredictable from EEG."
            )

        elif best_test_auc < 0.65:

            print(
                "\nRESULT:"
            )

            print(
                "The redesigned utility target shows "
                "weak EEG predictability."
            )

        else:

            print(
                "\nRESULT:"
            )

            print(
                "The redesigned utility target shows "
                "potential EEG-predictable signal and "
                "should be investigated further."
            )


# ============================================================
# 15. IMPORTANT CAUSALITY WARNING
# ============================================================

print("\n" + "=" * 75)
print("CAUSALITY STATUS")
print("=" * 75)

print(
    """
U1 original correctness:
    Retrospective decoder outcome.

U2 global confidence:
    Confidence is produced by the expensive decoder.

U3 class-conditioned median:
    Uses true character class and is therefore not causal.

U4 class-conditioned Q75:
    Uses true character class and is therefore not causal.

U5 class-conditioned residual:
    Uses true character class and is therefore not causal.

Therefore:
    These redesigned labels are diagnostic targets only.

A future deployable gate must predict utility from information
available BEFORE invoking the expensive decoder.
"""
)


# ============================================================
# 16. SAVE ALL RESULTS
# ============================================================

label_summary_path = (
    RESULT_ROOT /
    "step7b_candidate_label_summary.csv"
)

results_path = (
    RESULT_ROOT /
    "step7b_model_comparison.csv"
)

shuffle_path = (
    RESULT_ROOT /
    "step7b_shuffled_control.csv"
)

label_summary_df.to_csv(
    label_summary_path,
    index=False
)

results_df.to_csv(
    results_path,
    index=False
)

shuffle_df.to_csv(
    shuffle_path,
    index=False
)

# ------------------------------------------------------------
# Save metadata
# ------------------------------------------------------------

metadata = {
    "step":
        "7B",

    "description":
        "Utility-label redesign and EEG predictability diagnostic",

    "n_samples":
        int(len(h7)),

    "n_eeg_features":
        int(X_eeg.shape[1]),

    "train_n":
        int(train_mask.sum()),

    "val_n":
        int(val_mask.sum()),

    "test_n":
        int(test_mask.sum()),

    "candidate_labels":
        list(candidate_labels.keys()),

    "selection_rule":
        "Label statistics from train; EEG model threshold selected on validation; test evaluated after freezing",

    "causal_status":
        "All candidate labels are retrospective diagnostic targets; class-conditioned labels are not deployable gate inputs",

    "best_validation_label":
        (
            str(best_label)
            if len(val_results)
            else None
        ),

    "best_validation_auroc":
        (
            float(best_val_auc)
            if len(val_results)
            else None
        )
}

with open(
    RESULT_ROOT /
    "step7b_summary.json",
    "w"
) as f:

    json.dump(
        metadata,
        f,
        indent=2
    )

print("\n" + "=" * 75)
print("FILES SAVED")
print("=" * 75)

print(
    label_summary_path
)

print(
    results_path
)

print(
    shuffle_path
)

print(
    RESULT_ROOT /
    "step7b_summary.json"
)

print("\n" + "=" * 75)
print("STEP 7B COMPLETE")
print("=" * 75)
```

```python
# ============================================================
# NEUROSENTINEL — STEP 7C
# DEFINE AND VALIDATE A CAUSAL / PRE-DECODER UTILITY TARGET
# ============================================================
#
# Goal:
#   Identify EEG-only utility targets that are available BEFORE
#   invoking the expensive decoder.
#
# STRICTLY FORBIDDEN:
#   - true character / true_class
#   - decoder prediction
#   - decoder confidence
#   - decoder correctness
#   - decoder-derived utility labels
#   - future EEG / future keystroke information
#
# Frozen split:
#   Phase 6 subject-independent split
#
# Evaluation:
#   - Train / validation / frozen test
#   - AUROC
#   - AUPRC
#   - Accuracy
#   - Balanced accuracy
#   - Precision
#   - Recall
#   - F1
#   - shuffled-label control
#   - subject-level analysis
#
# Candidate causal EEG-only utility proxies:
#
#   C1: High robust EEG amplitude / energy
#   C2: High temporal stability
#   C3: High spatial coherence
#   C4: High physiological-band concentration
#   C5: Low spectral entropy
#   C6: Composite EEG quality score
#
# NOTE:
# These are PRE-DECODER EEG QUALITY / INFORMATION proxies.
# They are not decoder correctness labels.
# ============================================================


# ============================================================
# 0. INSTALL / IMPORT
# ============================================================

import os
import json
import math
import random
import warnings
from pathlib import Path

import numpy as np
import pandas as pd

from scipy.signal import welch
from scipy.stats import entropy

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

warnings.filterwarnings("ignore")


# ============================================================
# 1. REPRODUCIBILITY
# ============================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("=" * 70)
print("NEUROSENTINEL STEP 7C")
print("=" * 70)
print("Device:", DEVICE)
print("Seed:", SEED)


# ============================================================
# 2. PATHS — ROBUST GOOGLE DRIVE RESOLUTION
# ============================================================

from pathlib import Path

# Detect the mounted Google Drive root automatically
DRIVE_CANDIDATES = [
    Path("/content/gdrive/MyDrive"),
    Path("/content/drive/MyDrive"),
]

DRIVE_ROOT = None

for candidate in DRIVE_CANDIDATES:
    if candidate.exists():
        DRIVE_ROOT = candidate
        break

if DRIVE_ROOT is None:
    raise FileNotFoundError(
        "Google Drive MyDrive was not found.\n"
        "Expected either:\n"
        "  /content/gdrive/MyDrive\n"
        "or\n"
        "  /content/drive/MyDrive"
    )

ROOT = DRIVE_ROOT / "NeuroSentinel"

print("Detected Drive root:", DRIVE_ROOT)
print("NeuroSentinel root:", ROOT)

# ------------------------------------------------------------
# Phase 6 manifest
# ------------------------------------------------------------

PHASE6_MANIFEST = (
    ROOT /
    "07_phase6_subject_split" /
    "phase6_subject_split_manifest_resolved.csv"
)

# ------------------------------------------------------------
# H7 manifest
# ------------------------------------------------------------

H7_MANIFEST = (
    ROOT /
    "results" /
    "phase5h7" /
    "phase5h7_utility_label_manifest.csv"
)

# ------------------------------------------------------------
# Temporal windows
# ------------------------------------------------------------

TEMPORAL_ROOT = (
    ROOT /
    "08_phase6_temporal_windows"
)

# ------------------------------------------------------------
# Step 7C outputs
# ------------------------------------------------------------

OUTPUT_ROOT = (
    ROOT /
    "results" /
    "step7c_causal_utility"
)

CHECKPOINT_ROOT = (
    ROOT /
    "checkpoints" /
    "step7c"
)

OUTPUT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

CHECKPOINT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

print("\nPaths:")
print("ROOT:", ROOT)
print("Phase 6 manifest:", PHASE6_MANIFEST)
print("H7 manifest:", H7_MANIFEST)
print("Temporal root:", TEMPORAL_ROOT)
print("Output:", OUTPUT_ROOT)
print("Checkpoints:", CHECKPOINT_ROOT)

# ------------------------------------------------------------
# Verify required files BEFORE continuing
# ------------------------------------------------------------

print("\nChecking required files...")

if not PHASE6_MANIFEST.exists():

    # Search automatically in case the file has another location
    matches = list(
        ROOT.rglob(
            "phase6_subject_split_manifest_resolved.csv"
        )
    )

    if len(matches) > 0:

        PHASE6_MANIFEST = matches[0]

        print(
            "Found Phase 6 manifest automatically:",
            PHASE6_MANIFEST
        )

    else:

        # Also search for any Phase 6 manifest
        possible = list(
            ROOT.rglob(
                "*phase6*manifest*.csv"
            )
        )

        print(
            "\nAvailable Phase 6-like manifests:"
        )

        for p in possible:
            print(" -", p)

        raise FileNotFoundError(
            "\nCould not find "
            "phase6_subject_split_manifest_resolved.csv"
        )

if not H7_MANIFEST.exists():

    matches = list(
        ROOT.rglob(
            "phase5h7_utility_label_manifest.csv"
        )
    )

    if len(matches) > 0:

        H7_MANIFEST = matches[0]

        print(
            "Found H7 manifest automatically:",
            H7_MANIFEST
        )

    else:

        raise FileNotFoundError(
            f"Could not find H7 manifest under {ROOT}"
        )

print("\nRequired files verified successfully.")


# ============================================================
# 3. VERIFY PHASE 6 SPLIT
# ============================================================

assert PHASE6_MANIFEST.exists(), (
    f"Missing Phase 6 manifest: {PHASE6_MANIFEST}"
)

phase6 = pd.read_csv(PHASE6_MANIFEST)

required_phase6_cols = {
    "recording_id",
    "subject",
    "session",
    "task",
    "raw",
    "log",
    "split",
    "raw_resolved"
}

missing = required_phase6_cols - set(phase6.columns)

assert not missing, (
    f"Phase 6 manifest missing columns: {missing}"
)

print("\nPhase 6 split")
print(phase6.groupby("split")["recording_id"].count())

print("\nSubjects by split")

for split_name in ["train", "val", "test"]:
    subjects = sorted(
        phase6.loc[
            phase6["split"] == split_name,
            "subject"
        ].astype(str).unique()
    )
    print(split_name, subjects)


# ============================================================
# 4. LOAD H7 ONLY FOR ALIGNMENT / VALIDATION
# ============================================================
#
# H7 is NOT used as the causal target.
#
# It is loaded only to verify that Step 7C does not accidentally
# depend on decoder-derived information.
#
# ============================================================

assert H7_MANIFEST.exists(), (
    f"Missing H7 manifest: {H7_MANIFEST}"
)

h7 = pd.read_csv(H7_MANIFEST)

print("\nH7 shape:", h7.shape)

print("H7 columns:")
print(list(h7.columns))


# ============================================================
# 5. DISCOVER RECORDING WINDOW FILES
# ============================================================
#
# We preferentially use the original Phase 5 baseline windows:
#
#   61 channels × 25 samples
#
# These are exactly the windows used in the decoder experiments.
#
# ============================================================

def find_baseline_window_file(recording_id):
    """
    Find the Phase 5 baseline EEG window file for one recording.
    """

    candidates = [
        ROOT /
        "06_phase5_utility_labels" /
        "recordings" /
        f"{recording_id}_windows.npy",

        ROOT /
        "06_phase5_utility_labels" /
        "recordings" /
        recording_id /
        f"{recording_id}_windows.npy",

        ROOT /
        "06_phase5_utility_labels" /
        f"{recording_id}_windows.npy",
    ]

    for p in candidates:
        if p.exists():
            return p

    # Recursive fallback
    matches = list(
        (
            ROOT /
            "06_phase5_utility_labels"
        ).rglob(f"{recording_id}_windows.npy")
    )

    if len(matches) > 0:
        return matches[0]

    return None


recording_files = {}

for rid in phase6["recording_id"].astype(str):
    p = find_baseline_window_file(rid)

    if p is None:
        print("WARNING: missing windows:", rid)
    else:
        recording_files[rid] = p

print("\nWindow files found:",
      len(recording_files),
      "/",
      len(phase6))


assert len(recording_files) == len(phase6), (
    "Not all Phase 6 recordings have baseline window files."
)


# ============================================================
# 6. WINDOW NORMALIZATION
# ============================================================
#
# Same normalization used in Phase 5H.3:
#
# channel-wise z-score within each window
#
# For every sample:
#   mean/std computed over 25 temporal samples separately
#   for each of the 61 channels.
#
# ============================================================

def channelwise_window_zscore(x):
    """
    x:
        (N, C, T)

    Returns:
        float32 normalized array
    """

    x = np.asarray(x, dtype=np.float32)

    mean = x.mean(axis=2, keepdims=True)
    std = x.std(axis=2, keepdims=True)

    std = np.maximum(std, 1e-6)

    return (x - mean) / std


# ============================================================
# 7. EEG-ONLY FEATURE EXTRACTION
# ============================================================
#
# Features are calculated only from the current EEG window.
#
# NO:
#   character
#   decoder
#   future event
#   confidence
#   correctness
#
# ============================================================

FS = 50.0

EPS = 1e-12


def safe_z(x):
    x = np.asarray(x, dtype=np.float64)

    med = np.nanmedian(x)
    mad = np.nanmedian(np.abs(x - med))

    if mad < EPS:
        return np.zeros_like(x)

    return (x - med) / (1.4826 * mad)


def spectral_features_one_window(window):
    """
    window shape:
        (61, 25)

    Returns a dictionary of EEG-only features.
    """

    x = np.asarray(window, dtype=np.float64)

    C, T = x.shape

    # --------------------------------------------------------
    # Basic amplitude features
    # --------------------------------------------------------

    channel_std = np.std(x, axis=1)

    rms = np.sqrt(np.mean(x ** 2, axis=1))

    abs_mean = np.mean(np.abs(x), axis=1)

    peak_to_peak = (
        np.max(x, axis=1) -
        np.min(x, axis=1)
    )

    # Robust amplitude
    q95 = np.percentile(np.abs(x), 95, axis=1)
    q50 = np.percentile(np.abs(x), 50, axis=1)

    # --------------------------------------------------------
    # Temporal stability
    # --------------------------------------------------------

    if T > 1:
        temporal_diff = np.diff(x, axis=1)

        diff_rms = np.sqrt(
            np.mean(temporal_diff ** 2, axis=1)
        )

        temporal_corrs = []

        for c in range(C):
            a = x[c, :-1]
            b = x[c, 1:]

            sa = np.std(a)
            sb = np.std(b)

            if sa < EPS or sb < EPS:
                temporal_corrs.append(0.0)
            else:
                temporal_corrs.append(
                    np.corrcoef(a, b)[0, 1]
                )

        temporal_corrs = np.asarray(
            temporal_corrs
        )

    else:
        diff_rms = np.zeros(C)
        temporal_corrs = np.zeros(C)

    # --------------------------------------------------------
    # Spatial coherence
    # --------------------------------------------------------

    corr = np.corrcoef(x)

    upper = corr[
        np.triu_indices(C, k=1)
    ]

    upper = upper[
        np.isfinite(upper)
    ]

    if len(upper) == 0:
        spatial_corr_mean = 0.0
        spatial_corr_abs = 0.0
    else:
        spatial_corr_mean = np.mean(upper)
        spatial_corr_abs = np.mean(np.abs(upper))

    # --------------------------------------------------------
    # Frequency-domain features
    # --------------------------------------------------------
    #
    # 25 samples at 50 Hz is a short window.
    # Welch is therefore computed with nperseg <= T.
    #
    # --------------------------------------------------------

    freqs, psd = welch(
        x,
        fs=FS,
        axis=1,
        nperseg=min(T, 25),
        noverlap=0
    )

    # Remove DC
    valid = freqs > 0

    freqs2 = freqs[valid]
    psd2 = psd[:, valid]

    total_power = np.trapezoid(
        psd2,
        freqs2,
        axis=1
    )

    total_power = np.maximum(
        total_power,
        EPS
    )

    # Frequency bands compatible with 0.1–20 Hz preprocessing

    def band_power(lo, hi):
        mask = (
            (freqs2 >= lo) &
            (freqs2 < hi)
        )

        if not np.any(mask):
            return np.zeros(C)

        return np.trapezoid(
            psd2[:, mask],
            freqs2[mask],
            axis=1
        )

    delta = band_power(1, 4)
    theta = band_power(4, 8)
    alpha = band_power(8, 13)
    beta = band_power(13, 20)

    physiological_power = (
        theta +
        alpha +
        beta
    )

    phys_ratio = (
        physiological_power /
        total_power
    )

    high_freq_ratio = (
        beta /
        total_power
    )

    low_freq_ratio = (
        delta /
        total_power
    )

    # --------------------------------------------------------
    # Spectral entropy
    # --------------------------------------------------------

    psd_norm = (
        psd2 /
        np.maximum(
            np.sum(psd2, axis=1, keepdims=True),
            EPS
        )
    )

    spectral_entropy = -np.sum(
        psd_norm *
        np.log(
            np.maximum(psd_norm, EPS)
        ),
        axis=1
    )

    spectral_entropy /= np.log(
        max(psd_norm.shape[1], 2)
    )

    # --------------------------------------------------------
    # Aggregate features
    # --------------------------------------------------------

    features = {
        # amplitude
        "amp_std_mean": np.mean(channel_std),
        "amp_std_median": np.median(channel_std),
        "amp_std_std": np.std(channel_std),

        "rms_mean": np.mean(rms),
        "rms_median": np.median(rms),

        "abs_mean": np.mean(abs_mean),

        "peak_to_peak_mean": np.mean(peak_to_peak),
        "peak_to_peak_median": np.median(peak_to_peak),

        "q95_mean": np.mean(q95),
        "q50_mean": np.mean(q50),

        # temporal
        "diff_rms_mean": np.mean(diff_rms),
        "diff_rms_median": np.median(diff_rms),

        "temporal_corr_mean": np.mean(
            temporal_corrs
        ),

        "temporal_corr_median": np.median(
            temporal_corrs
        ),

        # spatial
        "spatial_corr_mean": spatial_corr_mean,
        "spatial_corr_abs": spatial_corr_abs,

        # frequency
        "total_power_mean": np.mean(
            total_power
        ),

        "delta_ratio_mean": np.mean(
            delta / total_power
        ),

        "theta_ratio_mean": np.mean(
            theta / total_power
        ),

        "alpha_ratio_mean": np.mean(
            alpha / total_power
        ),

        "beta_ratio_mean": np.mean(
            beta / total_power
        ),

        "phys_ratio_mean": np.mean(
            phys_ratio
        ),

        "high_freq_ratio_mean": np.mean(
            high_freq_ratio
        ),

        "low_freq_ratio_mean": np.mean(
            low_freq_ratio
        ),

        "spectral_entropy_mean": np.mean(
            spectral_entropy
        ),

        "spectral_entropy_std": np.std(
            spectral_entropy
        ),
    }

    return features


# ============================================================
# 8. EXTRACT ALL EEG FEATURES
# ============================================================

feature_rows = []

window_count_check = []

print("\nExtracting EEG-only features...")

for idx, row in phase6.iterrows():

    rid = str(row["recording_id"])
    split = str(row["split"])
    subject = str(row["subject"])

    window_path = recording_files[rid]

    X = np.load(
        window_path,
        mmap_mode="r"
    )

    X = np.asarray(X)

    assert X.ndim == 3, (
        f"{rid}: expected 3D array, got {X.shape}"
    )

    assert X.shape[1] == 61, (
        f"{rid}: expected 61 channels, got {X.shape}"
    )

    assert X.shape[2] == 25, (
        f"{rid}: expected 25 samples, got {X.shape}"
    )

    window_count_check.append(
        {
            "recording_id": rid,
            "split": split,
            "subject": subject,
            "n_windows": X.shape[0]
        }
    )

    for i in range(X.shape[0]):

        feats = spectral_features_one_window(
            X[i]
        )

        feats["recording_id"] = rid
        feats["subject"] = subject
        feats["split"] = split
        feats["window_index"] = i

        feature_rows.append(feats)

    print(
        f"[{idx + 1:02d}/{len(phase6):02d}] "
        f"{rid}: {X.shape[0]} windows"
    )


features_df = pd.DataFrame(feature_rows)

print("\nFeature matrix:")
print(features_df.shape)

print("\nFeature columns:")
print([
    c for c in features_df.columns
    if c not in [
        "recording_id",
        "subject",
        "split",
        "window_index"
    ]
])


# ============================================================
# 9. SAVE FEATURES
# ============================================================

FEATURE_PATH = (
    OUTPUT_ROOT /
    "step7c_eeg_only_features.csv"
)

features_df.to_csv(
    FEATURE_PATH,
    index=False
)

print("\nSaved:")
print(FEATURE_PATH)


# ============================================================
# 10. CREATE CAUSAL EEG-ONLY TARGETS
# ============================================================
#
# These targets deliberately do NOT depend on:
#
#   true_class
#   predicted_class
#   confidence
#   decoder_correct
#   utility_label
#
# Candidate targets:
#
# C1:
#   robust amplitude / signal-energy quality
#
# C2:
#   temporal stability
#
# C3:
#   spatial coherence
#
# C4:
#   physiological-band concentration
#
# C5:
#   low spectral entropy
#
# C6:
#   composite EEG quality score
#
# Thresholds are determined ONLY from TRAINING data.
#
# Q75 is used for "high utility" to create a sufficiently
# selective binary target.
#
# ============================================================

df = features_df.copy()

train_mask = df["split"].values == "train"
val_mask = df["split"].values == "val"
test_mask = df["split"].values == "test"


def train_quantile_threshold(
    values,
    quantile=0.75
):
    return float(
        np.nanquantile(
            values[train_mask],
            quantile
        )
    )


# ------------------------------------------------------------
# Candidate continuous quality scores
# ------------------------------------------------------------

df["C1_energy"] = (
    np.log1p(
        np.maximum(
            df["rms_mean"].values,
            0
        )
    )
)

df["C2_temporal_stability"] = (
    df["temporal_corr_mean"].values
    -
    df["diff_rms_mean"].values
)

df["C3_spatial_coherence"] = (
    df["spatial_corr_mean"].values
)

df["C4_physiological_concentration"] = (
    df["phys_ratio_mean"].values
)

df["C5_low_spectral_entropy"] = (
    -df["spectral_entropy_mean"].values
)


# ------------------------------------------------------------
# Composite EEG quality score
# ------------------------------------------------------------
#
# Standardize candidate components using TRAIN ONLY.
#
# This avoids letting validation/test distributions influence
# target construction.
# ------------------------------------------------------------

composite_components = [
    "C1_energy",
    "C2_temporal_stability",
    "C3_spatial_coherence",
    "C4_physiological_concentration",
    "C5_low_spectral_entropy"
]

composite_train_stats = {}

for col in composite_components:

    vals = df.loc[
        train_mask,
        col
    ].values.astype(float)

    mu = np.nanmean(vals)
    sd = np.nanstd(vals)

    if sd < EPS:
        sd = 1.0

    composite_train_stats[col] = {
        "mean": float(mu),
        "std": float(sd)
    }

    df[col + "_z"] = (
        df[col] - mu
    ) / sd


df["C6_composite_quality"] = (
    df["C1_energy_z"] +
    df["C2_temporal_stability_z"] +
    df["C3_spatial_coherence_z"] +
    df["C4_physiological_concentration_z"] +
    df["C5_low_spectral_entropy_z"]
) / 5.0


# ============================================================
# 11. CREATE BINARY TARGETS
# ============================================================
#
# Threshold:
#   training Q75
#
# Therefore:
#   1 = high EEG quality / information proxy
#   0 = otherwise
#
# Threshold is frozen before validation/test.
# ============================================================

candidate_scores = [
    "C1_energy",
    "C2_temporal_stability",
    "C3_spatial_coherence",
    "C4_physiological_concentration",
    "C5_low_spectral_entropy",
    "C6_composite_quality"
]

target_thresholds = {}

for score_col in candidate_scores:

    threshold = train_quantile_threshold(
        df[score_col].values,
        quantile=0.75
    )

    target_thresholds[score_col] = threshold

    target_col = (
        score_col +
        "_target"
    )

    df[target_col] = (
        df[score_col] >= threshold
    ).astype(np.int64)


# ============================================================
# 12. TARGET DISTRIBUTION
# ============================================================

target_distribution_rows = []

for score_col in candidate_scores:

    target_col = score_col + "_target"

    for split in ["train", "val", "test"]:

        m = df["split"] == split

        y = df.loc[
            m,
            target_col
        ].values

        target_distribution_rows.append(
            {
                "candidate": score_col,
                "split": split,
                "n": len(y),
                "positive": int(y.sum()),
                "negative": int((1 - y).sum()),
                "positive_rate": float(y.mean())
            }
        )


target_distribution = pd.DataFrame(
    target_distribution_rows
)

print("\n" + "=" * 70)
print("CAUSAL TARGET DISTRIBUTIONS")
print("=" * 70)

print(
    target_distribution.to_string(
        index=False
    )
)


target_distribution.to_csv(
    OUTPUT_ROOT /
    "step7c_target_distribution.csv",
    index=False
)


# ============================================================
# 13. PREPARE EEG INPUTS
# ============================================================
#
# Same Phase 5H.3 normalization.
#
# The target is derived from EEG-only metrics.
# The gate receives normalized EEG.
#
# ============================================================

class EEGWindowDataset(Dataset):

    def __init__(
        self,
        arrays,
        targets
    ):

        self.X = arrays
        self.y = np.asarray(
            targets,
            dtype=np.float32
        )

    def __len__(self):
        return len(self.y)

    def __getitem__(self, idx):

        x = self.X[idx]

        return (
            torch.tensor(
                x,
                dtype=torch.float32
            ),
            torch.tensor(
                self.y[idx],
                dtype=torch.float32
            )
        )


# ============================================================
# 14. LIGHTWEIGHT CAUSAL GATE
# ============================================================
#
# Same general family used in Phase 7:
#
#   Conv1D 61 -> 64
#   Conv1D 64 -> 64 dilation 2
#   Conv1D 64 -> 64 dilation 4
#   Conv1D 64 -> 32 dilation 8
#   Global temporal mean
#   Linear 32 -> 1
#
# ============================================================

class CausalUtilityGate(nn.Module):

    def __init__(self):

        super().__init__()

        self.net = nn.Sequential(

            nn.Conv1d(
                61,
                64,
                kernel_size=5,
                padding=2
            ),

            nn.ReLU(),

            nn.Conv1d(
                64,
                64,
                kernel_size=5,
                dilation=2,
                padding=4
            ),

            nn.ReLU(),

            nn.Conv1d(
                64,
                64,
                kernel_size=5,
                dilation=4,
                padding=8
            ),

            nn.ReLU(),

            nn.Conv1d(
                64,
                32,
                kernel_size=5,
                dilation=8,
                padding=16
            ),

            nn.ReLU()
        )

        self.head = nn.Linear(
            32,
            1
        )

    def forward(self, x):

        z = self.net(x)

        z = z.mean(
            dim=2
        )

        return self.head(
            z
        ).squeeze(1)


# ============================================================
# 15. BUILD RECORDING-LEVEL INDEX
# ============================================================

# We reload each recording as required rather than keeping the
# complete EEG dataset simultaneously in RAM.

recording_index = {}

for rid in phase6["recording_id"].astype(str):

    row = phase6[
        phase6["recording_id"].astype(str) == rid
    ].iloc[0]

    X = np.load(
        recording_files[rid]
    )

    X = np.asarray(
        X,
        dtype=np.float32
    )

    X = channelwise_window_zscore(
        X
    )

    recording_index[rid] = {
        "X": X,
        "split": str(row["split"]),
        "subject": str(row["subject"])
    }

print(
    "\nLoaded normalized recordings:",
    len(recording_index)
)


# ============================================================
# 16. CREATE ROW-TO-EEG LOOKUP
# ============================================================

# Map:
#   recording_id + window_index
# to row index in df.

df = df.reset_index(drop=True)

row_lookup = {}

for i, r in df.iterrows():

    key = (
        str(r["recording_id"]),
        int(r["window_index"])
    )

    row_lookup[key] = i


# ============================================================
# 17. METRIC FUNCTION
# ============================================================

def calculate_binary_metrics(
    y_true,
    y_prob,
    threshold=0.5
):

    y_true = np.asarray(
        y_true,
        dtype=int
    )

    y_prob = np.asarray(
        y_prob,
        dtype=float
    )

    y_pred = (
        y_prob >= threshold
    ).astype(int)

    result = {}

    result["accuracy"] = accuracy_score(
        y_true,
        y_pred
    )

    result["balanced_accuracy"] = (
        balanced_accuracy_score(
            y_true,
            y_pred
        )
    )

    result["precision"] = precision_score(
        y_true,
        y_pred,
        zero_division=0
    )

    result["recall"] = recall_score(
        y_true,
        y_pred,
        zero_division=0
    )

    result["f1"] = f1_score(
        y_true,
        y_pred,
        zero_division=0
    )

    if len(np.unique(y_true)) == 2:

        result["auroc"] = roc_auc_score(
            y_true,
            y_prob
        )

        result["auprc"] = (
            average_precision_score(
                y_true,
                y_prob
            )
        )

    else:

        result["auroc"] = np.nan
        result["auprc"] = np.nan

    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1]
    )

    result["tn"] = int(cm[0, 0])
    result["fp"] = int(cm[0, 1])
    result["fn"] = int(cm[1, 0])
    result["tp"] = int(cm[1, 1])

    return result


# ============================================================
# 18. MODEL DATA LOADER
# ============================================================

def make_split_arrays(
    split,
    target_col
):

    subset = df[
        df["split"] == split
    ].copy()

    X_list = []
    y_list = []

    for _, row in subset.iterrows():

        rid = str(
            row["recording_id"]
        )

        wi = int(
            row["window_index"]
        )

        X_list.append(
            recording_index[rid]["X"][wi]
        )

        y_list.append(
            row[target_col]
        )

    X = np.asarray(
        X_list,
        dtype=np.float32
    )

    y = np.asarray(
        y_list,
        dtype=np.float32
    )

    return X, y


# ============================================================
# 19. TRAIN ONE CANDIDATE TARGET
# ============================================================

BATCH_SIZE = 256
MAX_EPOCHS = 20
PATIENCE = 5

LEARNING_RATE = 1e-3
WEIGHT_DECAY = 1e-4


def train_candidate(
    candidate_name
):

    target_col = (
        candidate_name +
        "_target"
    )

    print("\n")
    print("=" * 70)
    print("TRAINING:", candidate_name)
    print("=" * 70)

    X_train, y_train = make_split_arrays(
        "train",
        target_col
    )

    X_val, y_val = make_split_arrays(
        "val",
        target_col
    )

    X_test, y_test = make_split_arrays(
        "test",
        target_col
    )

    print(
        "Train:",
        X_train.shape,
        "positive:",
        int(y_train.sum())
    )

    print(
        "Val:",
        X_val.shape,
        "positive:",
        int(y_val.sum())
    )

    print(
        "Test:",
        X_test.shape,
        "positive:",
        int(y_test.sum())
    )

    # --------------------------------------------------------
    # Class weighting
    # --------------------------------------------------------

    n_pos = max(
        int(y_train.sum()),
        1
    )

    n_neg = max(
        int(len(y_train) - y_train.sum()),
        1
    )

    pos_weight = (
        n_neg /
        n_pos
    )

    print(
        "pos_weight:",
        pos_weight
    )

    # --------------------------------------------------------
    # Datasets
    # --------------------------------------------------------

    train_ds = EEGWindowDataset(
        X_train,
        y_train
    )

    val_ds = EEGWindowDataset(
        X_val,
        y_val
    )

    test_ds = EEGWindowDataset(
        X_test,
        y_test
    )

    train_loader = DataLoader(
        train_ds,
        batch_size=BATCH_SIZE,
        shuffle=True,
        num_workers=0,
        pin_memory=torch.cuda.is_available()
    )

    val_loader = DataLoader(
        val_ds,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=0,
        pin_memory=torch.cuda.is_available()
    )

    test_loader = DataLoader(
        test_ds,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=0,
        pin_memory=torch.cuda.is_available()
    )

    # --------------------------------------------------------
    # Model
    # --------------------------------------------------------

    model = CausalUtilityGate().to(
        DEVICE
    )

    criterion = nn.BCEWithLogitsLoss(
        pos_weight=torch.tensor(
            pos_weight,
            dtype=torch.float32,
            device=DEVICE
        )
    )

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=LEARNING_RATE,
        weight_decay=WEIGHT_DECAY
    )

    # --------------------------------------------------------
    # Training
    # --------------------------------------------------------

    best_val_auprc = -np.inf
    best_val_f1 = -np.inf
    best_epoch = 0
    patience_counter = 0

    checkpoint_path = (
        CHECKPOINT_ROOT /
        f"{candidate_name}_best.pt"
    )

    history = []

    for epoch in range(
        1,
        MAX_EPOCHS + 1
    ):

        model.train()

        train_loss_sum = 0.0
        train_count = 0

        for xb, yb in train_loader:

            xb = xb.to(
                DEVICE,
                non_blocking=True
            )

            yb = yb.to(
                DEVICE,
                non_blocking=True
            )

            optimizer.zero_grad(
                set_to_none=True
            )

            logits = model(
                xb
            )

            loss = criterion(
                logits,
                yb
            )

            loss.backward()

            optimizer.step()

            train_loss_sum += (
                loss.item() *
                len(yb)
            )

            train_count += len(yb)

        train_loss = (
            train_loss_sum /
            max(train_count, 1)
        )

        # ----------------------------------------------------
        # Validation
        # ----------------------------------------------------

        model.eval()

        val_probs = []
        val_targets = []

        val_loss_sum = 0.0
        val_count = 0

        with torch.no_grad():

            for xb, yb in val_loader:

                xb = xb.to(
                    DEVICE,
                    non_blocking=True
                )

                yb = yb.to(
                    DEVICE,
                    non_blocking=True
                )

                logits = model(
                    xb
                )

                loss = criterion(
                    logits,
                    yb
                )

                probs = torch.sigmoid(
                    logits
                )

                val_loss_sum += (
                    loss.item() *
                    len(yb)
                )

                val_count += len(yb)

                val_probs.extend(
                    probs.cpu().numpy()
                )

                val_targets.extend(
                    yb.cpu().numpy()
                )

        val_loss = (
            val_loss_sum /
            max(val_count, 1)
        )

        val_probs = np.asarray(
            val_probs
        )

        val_targets = np.asarray(
            val_targets,
            dtype=int
        )

        val_metrics = calculate_binary_metrics(
            val_targets,
            val_probs,
            threshold=0.5
        )

        # ----------------------------------------------------
        # Validation threshold selection
        # ----------------------------------------------------
        #
        # Threshold is selected ONLY on validation.
        # Test threshold is never optimized.
        #
        # ----------------------------------------------------

        thresholds = np.arange(
            0.05,
            0.951,
            0.01
        )

        best_threshold = 0.5
        best_threshold_f1 = -np.inf

        for threshold in thresholds:

            pred = (
                val_probs >= threshold
            ).astype(int)

            f1 = f1_score(
                val_targets,
                pred,
                zero_division=0
            )

            if f1 > best_threshold_f1:

                best_threshold_f1 = f1
                best_threshold = float(
                    threshold
                )

        val_threshold_metrics = (
            calculate_binary_metrics(
                val_targets,
                val_probs,
                threshold=best_threshold
            )
        )

        print(
            f"Epoch {epoch:02d} | "
            f"train_loss={train_loss:.4f} | "
            f"val_loss={val_loss:.4f} | "
            f"val_AUROC={val_metrics['auroc']:.4f} | "
            f"val_AUPRC={val_metrics['auprc']:.4f} | "
            f"val_F1={val_threshold_metrics['f1']:.4f} | "
            f"thr={best_threshold:.2f}"
        )

        history.append(
            {
                "epoch": epoch,
                "train_loss": train_loss,
                "val_loss": val_loss,
                "val_auroc": val_metrics["auroc"],
                "val_auprc": val_metrics["auprc"],
                "val_f1": val_threshold_metrics["f1"],
                "val_balanced_accuracy":
                    val_threshold_metrics[
                        "balanced_accuracy"
                    ],
                "threshold": best_threshold
            }
        )

        # ----------------------------------------------------
        # Model selection:
        # validation AUPRC
        # ----------------------------------------------------

        selection_score = (
            val_metrics["auprc"]
        )

        if selection_score > best_val_auprc:

            best_val_auprc = (
                selection_score
            )

            best_val_f1 = (
                val_threshold_metrics["f1"]
            )

            best_epoch = epoch

            patience_counter = 0

            torch.save(
                {
                    "candidate": candidate_name,
                    "epoch": epoch,
                    "model_state_dict":
                        model.state_dict(),
                    "optimizer_state_dict":
                        optimizer.state_dict(),
                    "val_auprc":
                        val_metrics["auprc"],
                    "val_auroc":
                        val_metrics["auroc"],
                    "val_f1":
                        val_threshold_metrics["f1"],
                    "threshold":
                        best_threshold,
                    "seed": SEED
                },
                checkpoint_path
            )

        else:

            patience_counter += 1

        if patience_counter >= PATIENCE:

            print(
                "Early stopping."
            )

            break

    # ========================================================
    # LOAD BEST VALIDATION MODEL
    # ========================================================

    checkpoint = torch.load(
        checkpoint_path,
        map_location=DEVICE
    )

    model.load_state_dict(
        checkpoint["model_state_dict"]
    )

    frozen_threshold = float(
        checkpoint["threshold"]
    )

    # ========================================================
    # FINAL VALIDATION
    # ========================================================

    model.eval()

    val_probs = []
    val_targets = []

    with torch.no_grad():

        for xb, yb in val_loader:

            xb = xb.to(
                DEVICE,
                non_blocking=True
            )

            logits = model(
                xb
            )

            probs = torch.sigmoid(
                logits
            )

            val_probs.extend(
                probs.cpu().numpy()
            )

            val_targets.extend(
                yb.numpy()
            )

    val_probs = np.asarray(
        val_probs
    )

    val_targets = np.asarray(
        val_targets,
        dtype=int
    )

    final_val_metrics = (
        calculate_binary_metrics(
            val_targets,
            val_probs,
            threshold=frozen_threshold
        )
    )

    # ========================================================
    # FROZEN TEST
    # ========================================================
    #
    # IMPORTANT:
    # threshold was selected from validation only.
    #
    # ========================================================

    test_probs = []
    test_targets = []

    with torch.no_grad():

        for xb, yb in test_loader:

            xb = xb.to(
                DEVICE,
                non_blocking=True
            )

            logits = model(
                xb
            )

            probs = torch.sigmoid(
                logits
            )

            test_probs.extend(
                probs.cpu().numpy()
            )

            test_targets.extend(
                yb.numpy()
            )

    test_probs = np.asarray(
        test_probs
    )

    test_targets = np.asarray(
        test_targets,
        dtype=int
    )

    final_test_metrics = (
        calculate_binary_metrics(
            test_targets,
            test_probs,
            threshold=frozen_threshold
        )
    )

    # ========================================================
    # SAVE PROBABILITIES
    # ========================================================

    probability_df = pd.DataFrame(
        {
            "candidate": candidate_name,
            "split": (
                ["val"] * len(val_probs) +
                ["test"] * len(test_probs)
            ),
            "target": np.concatenate(
                [
                    val_targets,
                    test_targets
                ]
            ),
            "probability": np.concatenate(
                [
                    val_probs,
                    test_probs
                ]
            )
        }
    )

    probability_df.to_csv(
        OUTPUT_ROOT /
        f"{candidate_name}_probabilities.csv",
        index=False
    )

    # ========================================================
    # RESULT
    # ========================================================

    result = {
        "candidate": candidate_name,

        "best_epoch": int(
            checkpoint["epoch"]
        ),

        "validation_auroc": float(
            final_val_metrics["auroc"]
        ),

        "validation_auprc": float(
            final_val_metrics["auprc"]
        ),

        "validation_accuracy": float(
            final_val_metrics["accuracy"]
        ),

        "validation_balanced_accuracy":
            float(
                final_val_metrics[
                    "balanced_accuracy"
                ]
            ),

        "validation_precision": float(
            final_val_metrics[
                "precision"
            ]
        ),

        "validation_recall": float(
            final_val_metrics[
                "recall"
            ]
        ),

        "validation_f1": float(
            final_val_metrics["f1"]
        ),

        "test_auroc": float(
            final_test_metrics["auroc"]
        ),

        "test_auprc": float(
            final_test_metrics["auprc"]
        ),

        "test_accuracy": float(
            final_test_metrics["accuracy"]
        ),

        "test_balanced_accuracy":
            float(
                final_test_metrics[
                    "balanced_accuracy"
                ]
            ),

        "test_precision": float(
            final_test_metrics[
                "precision"
            ]
        ),

        "test_recall": float(
            final_test_metrics[
                "recall"
            ]
        ),

        "test_f1": float(
            final_test_metrics["f1"]
        ),

        "threshold": frozen_threshold,

        "target_threshold":
            float(
                target_thresholds[
                    candidate_name
                ]
            ),

        "train_positive_rate":
            float(
                df.loc[
                    df["split"] == "train",
                    target_col
                ].mean()
            ),

        "val_positive_rate":
            float(
                df.loc[
                    df["split"] == "val",
                    target_col
                ].mean()
            ),

        "test_positive_rate":
            float(
                df.loc[
                    df["split"] == "test",
                    target_col
                ].mean()
            )
    }

    print("\nFINAL:", candidate_name)

    print(
        "Validation AUROC:",
        f"{result['validation_auroc']:.6f}"
    )

    print(
        "Validation AUPRC:",
        f"{result['validation_auprc']:.6f}"
    )

    print(
        "Validation F1:",
        f"{result['validation_f1']:.6f}"
    )

    print(
        "Frozen Test AUROC:",
        f"{result['test_auroc']:.6f}"
    )

    print(
        "Frozen Test AUPRC:",
        f"{result['test_auprc']:.6f}"
    )

    print(
        "Frozen Test F1:",
        f"{result['test_f1']:.6f}"
    )

    return result, model


# ============================================================
# 20. RUN ALL CANDIDATES
# ============================================================

all_results = []
trained_models = {}

for candidate in candidate_scores:

    result, model = train_candidate(
        candidate
    )

    all_results.append(
        result
    )

    trained_models[candidate] = model

    # Free GPU memory between candidates
    if torch.cuda.is_available():

        torch.cuda.empty_cache()


# ============================================================
# 21. SAVE OVERALL RESULTS
# ============================================================

results_df = pd.DataFrame(
    all_results
)

results_df = results_df.sort_values(
    "validation_auprc",
    ascending=False
).reset_index(drop=True)

RESULT_PATH = (
    OUTPUT_ROOT /
    "step7c_candidate_results.csv"
)

results_df.to_csv(
    RESULT_PATH,
    index=False
)

print("\n")
print("=" * 70)
print("STEP 7C CANDIDATE RESULTS")
print("=" * 70)

display(
    results_df[
        [
            "candidate",
            "validation_auroc",
            "validation_auprc",
            "validation_balanced_accuracy",
            "validation_f1",
            "test_auroc",
            "test_auprc",
            "test_balanced_accuracy",
            "test_f1"
        ]
    ]
)


# ============================================================
# 22. SHUFFLED-LABEL CONTROL
# ============================================================
#
# Critical control:
#
# For the validation-best candidate, shuffle the target labels
# within TRAIN ONLY and train another model.
#
# The validation/test labels remain untouched.
#
# Expected:
#   shuffled control should approach chance.
#
# If shuffled control remains strong, there is likely leakage,
# an indexing problem, or a target construction problem.
#
# ============================================================

best_candidate = results_df.iloc[0]["candidate"]

print("\n")
print("=" * 70)
print("VALIDATION-BEST CANDIDATE")
print("=" * 70)
print(best_candidate)

best_target_col = (
    best_candidate +
    "_target"
)

X_train, y_train = make_split_arrays(
    "train",
    best_target_col
)

X_val, y_val = make_split_arrays(
    "val",
    best_target_col
)

X_test, y_test = make_split_arrays(
    "test",
    best_target_col
)

# Shuffle ONLY train labels
rng = np.random.default_rng(
    SEED + 777
)

y_train_shuffled = (
    y_train.copy()
)

rng.shuffle(
    y_train_shuffled
)

n_pos = max(
    int(y_train_shuffled.sum()),
    1
)

n_neg = max(
    int(len(y_train_shuffled) - y_train_shuffled.sum()),
    1
)

shuffled_pos_weight = (
    n_neg /
    n_pos
)

shuffle_model = (
    CausalUtilityGate().to(DEVICE)
)

shuffle_criterion = (
    nn.BCEWithLogitsLoss(
        pos_weight=torch.tensor(
            shuffled_pos_weight,
            dtype=torch.float32,
            device=DEVICE
        )
    )
)

shuffle_optimizer = torch.optim.AdamW(
    shuffle_model.parameters(),
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY
)

shuffle_train_ds = EEGWindowDataset(
    X_train,
    y_train_shuffled
)

shuffle_val_ds = EEGWindowDataset(
    X_val,
    y_val
)

shuffle_train_loader = DataLoader(
    shuffle_train_ds,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=0
)

shuffle_val_loader = DataLoader(
    shuffle_val_ds,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0
)

best_shuffle_auprc = -np.inf
best_shuffle_state = None

shuffle_patience = 0

for epoch in range(
    1,
    MAX_EPOCHS + 1
):

    shuffle_model.train()

    for xb, yb in shuffle_train_loader:

        xb = xb.to(DEVICE)
        yb = yb.to(DEVICE)

        shuffle_optimizer.zero_grad(
            set_to_none=True
        )

        logits = shuffle_model(
            xb
        )

        loss = shuffle_criterion(
            logits,
            yb
        )

        loss.backward()

        shuffle_optimizer.step()

    shuffle_model.eval()

    probs = []
    targets = []

    with torch.no_grad():

        for xb, yb in shuffle_val_loader:

            xb = xb.to(DEVICE)

            logits = shuffle_model(
                xb
            )

            p = torch.sigmoid(
                logits
            )

            probs.extend(
                p.cpu().numpy()
            )

            targets.extend(
                yb.numpy()
            )

    probs = np.asarray(probs)
    targets = np.asarray(
        targets,
        dtype=int
    )

    m = calculate_binary_metrics(
        targets,
        probs,
        threshold=0.5
    )

    if m["auprc"] > best_shuffle_auprc:

        best_shuffle_auprc = m["auprc"]

        best_shuffle_state = {
            k: v.cpu().clone()
            for k, v
            in shuffle_model.state_dict().items()
        }

        shuffle_patience = 0

    else:

        shuffle_patience += 1

    if shuffle_patience >= PATIENCE:
        break


# ============================================================
# 23. EVALUATE SHUFFLED CONTROL ON TEST
# ============================================================

shuffle_model.load_state_dict(
    best_shuffle_state
)

shuffle_model.eval()

shuffle_test_ds = EEGWindowDataset(
    X_test,
    y_test
)

shuffle_test_loader = DataLoader(
    shuffle_test_ds,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0
)

shuffle_test_probs = []
shuffle_test_targets = []

with torch.no_grad():

    for xb, yb in shuffle_test_loader:

        xb = xb.to(DEVICE)

        logits = shuffle_model(
            xb
        )

        p = torch.sigmoid(
            logits
        )

        shuffle_test_probs.extend(
            p.cpu().numpy()
        )

        shuffle_test_targets.extend(
            yb.numpy()
        )

shuffle_test_probs = np.asarray(
    shuffle_test_probs
)

shuffle_test_targets = np.asarray(
    shuffle_test_targets,
    dtype=int
)

shuffle_test_metrics = (
    calculate_binary_metrics(
        shuffle_test_targets,
        shuffle_test_probs,
        threshold=0.5
    )
)

shuffle_result = {
    "candidate": best_candidate,
    "control": "shuffled_train_labels",
    "test_auroc":
        float(
            shuffle_test_metrics[
                "auroc"
            ]
        ),
    "test_auprc":
        float(
            shuffle_test_metrics[
                "auprc"
            ]
        ),
    "test_accuracy":
        float(
            shuffle_test_metrics[
                "accuracy"
            ]
        ),
    "test_balanced_accuracy":
        float(
            shuffle_test_metrics[
                "balanced_accuracy"
            ]
        ),
    "test_precision":
        float(
            shuffle_test_metrics[
                "precision"
            ]
        ),
    "test_recall":
        float(
            shuffle_test_metrics[
                "recall"
            ]
        ),
    "test_f1":
        float(
            shuffle_test_metrics[
                "f1"
            ]
        )
}

print("\n")
print("=" * 70)
print("SHUFFLED-LABEL CONTROL")
print("=" * 70)

for k, v in shuffle_result.items():
    print(
        f"{k}: {v}"
    )


pd.DataFrame(
    [shuffle_result]
).to_csv(
    OUTPUT_ROOT /
    "step7c_shuffled_control.csv",
    index=False
)


# ============================================================
# 24. SUBJECT-LEVEL ANALYSIS
# ============================================================
#
# This is critical because the main project bottleneck is
# subject-independent generalization.
#
# We evaluate the frozen validation-selected model separately
# on official test subjects:
#
#   002
#   011
#   021
#
# No retraining.
# No subject-specific threshold.
# No test-based model selection.
#
# ============================================================

best_model = trained_models[
    best_candidate
]

best_model.eval()

subject_rows = []

best_target_col = (
    best_candidate +
    "_target"
)

test_df = df[
    df["split"] == "test"
].copy()

test_subjects = sorted(
    test_df["subject"].astype(str).unique()
)

for subject in test_subjects:

    subject_df = test_df[
        test_df["subject"].astype(str) ==
        subject
    ]

    X_list = []
    y_list = []

    for _, row in subject_df.iterrows():

        rid = str(
            row["recording_id"]
        )

        wi = int(
            row["window_index"]
        )

        X_list.append(
            recording_index[rid]["X"][wi]
        )

        y_list.append(
            row[best_target_col]
        )

    X_sub = np.asarray(
        X_list,
        dtype=np.float32
    )

    y_sub = np.asarray(
        y_list,
        dtype=np.float32
    )

    sub_ds = EEGWindowDataset(
        X_sub,
        y_sub
    )

    sub_loader = DataLoader(
        sub_ds,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=0
    )

    probs = []
    targets = []

    with torch.no_grad():

        for xb, yb in sub_loader:

            xb = xb.to(DEVICE)

            logits = best_model(
                xb
            )

            p = torch.sigmoid(
                logits
            )

            probs.extend(
                p.cpu().numpy()
            )

            targets.extend(
                yb.numpy()
            )

    probs = np.asarray(probs)

    targets = np.asarray(
        targets,
        dtype=int
    )

    m = calculate_binary_metrics(
        targets,
        probs,
        threshold=float(
            results_df.iloc[0]["threshold"]
        )
    )

    subject_rows.append(
        {
            "candidate": best_candidate,
            "subject": subject,
            "n": len(targets),
            "positive_rate":
                float(targets.mean()),
            "auroc":
                float(m["auroc"]),
            "auprc":
                float(m["auprc"]),
            "accuracy":
                float(m["accuracy"]),
            "balanced_accuracy":
                float(
                    m["balanced_accuracy"]
                ),
            "precision":
                float(m["precision"]),
            "recall":
                float(m["recall"]),
            "f1":
                float(m["f1"])
        }
    )


subject_results = pd.DataFrame(
    subject_rows
)

print("\n")
print("=" * 70)
print("SUBJECT-LEVEL TEST RESULTS")
print("=" * 70)

display(
    subject_results
)

subject_results.to_csv(
    OUTPUT_ROOT /
    "step7c_subject_level_results.csv",
    index=False
)


# ============================================================
# 25. SIMPLE EEG-ONLY LOGISTIC REGRESSION BASELINE
# ============================================================
#
# This provides an independent model family.
#
# It uses the extracted EEG-only features rather than the CNN.
#
# Scaling is fit ONLY on TRAIN.
#
# Target remains causal EEG-only.
#
# ============================================================

feature_columns = [
    c for c in features_df.columns
    if c not in [
        "recording_id",
        "subject",
        "split",
        "window_index"
    ]
    and not c.startswith("C")
]

# Use original handcrafted EEG features only.

X_lr = df[
    feature_columns
].replace(
    [np.inf, -np.inf],
    np.nan
).fillna(0).values

y_lr = df[
    best_target_col
].values.astype(int)

split_arr = df[
    "split"
].values

train_idx = np.where(
    split_arr == "train"
)[0]

val_idx = np.where(
    split_arr == "val"
)[0]

test_idx = np.where(
    split_arr == "test"
)[0]

lr_model = Pipeline(
    [
        (
            "scaler",
            StandardScaler()
        ),
        (
            "classifier",
            LogisticRegression(
                max_iter=1000,
                class_weight="balanced",
                random_state=SEED
            )
        )
    ]
)

lr_model.fit(
    X_lr[train_idx],
    y_lr[train_idx]
)

lr_val_probs = lr_model.predict_proba(
    X_lr[val_idx]
)[:, 1]

lr_test_probs = lr_model.predict_proba(
    X_lr[test_idx]
)[:, 1]

lr_val_metrics = calculate_binary_metrics(
    y_lr[val_idx],
    lr_val_probs,
    threshold=0.5
)

lr_test_metrics = calculate_binary_metrics(
    y_lr[test_idx],
    lr_test_probs,
    threshold=0.5
)

lr_result = {
    "candidate": best_candidate,
    "model": "EEG_feature_logistic_regression",
    "validation_auroc":
        float(
            lr_val_metrics["auroc"]
        ),
    "validation_auprc":
        float(
            lr_val_metrics["auprc"]
        ),
    "validation_f1":
        float(
            lr_val_metrics["f1"]
        ),
    "test_auroc":
        float(
            lr_test_metrics["auroc"]
        ),
    "test_auprc":
        float(
            lr_test_metrics["auprc"]
        ),
    "test_f1":
        float(
            lr_test_metrics["f1"]
        ),
    "test_balanced_accuracy":
        float(
            lr_test_metrics[
                "balanced_accuracy"
            ]
        )
}

print("\n")
print("=" * 70)
print("EEG FEATURE LOGISTIC REGRESSION")
print("=" * 70)

for k, v in lr_result.items():
    print(
        f"{k}: {v}"
    )

pd.DataFrame(
    [lr_result]
).to_csv(
    OUTPUT_ROOT /
    "step7c_logistic_baseline.csv",
    index=False
)


# ============================================================
# 26. FINAL STEP 7C DECISION TABLE
# ============================================================

summary_rows = []

for _, r in results_df.iterrows():

    summary_rows.append(
        {
            "candidate": r["candidate"],
            "val_AUROC": r["validation_auroc"],
            "val_AUPRC": r["validation_auprc"],
            "val_balanced_acc":
                r["validation_balanced_accuracy"],
            "val_F1": r["validation_f1"],
            "test_AUROC": r["test_auroc"],
            "test_AUPRC": r["test_auprc"],
            "test_balanced_acc":
                r["test_balanced_accuracy"],
            "test_F1": r["test_f1"],
            "target_threshold":
                r["target_threshold"]
        }
    )

summary_df = pd.DataFrame(
    summary_rows
)

summary_df.to_csv(
    OUTPUT_ROOT /
    "step7c_final_decision_table.csv",
    index=False
)


# ============================================================
# 27. SAVE COMPLETE STEP 7C STATUS
# ============================================================

best_row = results_df.iloc[0]

status = {
    "phase": "Step 7C",
    "status": "COMPLETED",

    "objective":
        "Evaluate pre-decoder EEG-only utility targets",

    "forbidden_information": [
        "true_class",
        "true_label",
        "predicted_class",
        "predicted_label",
        "decoder_confidence",
        "decoder_correct",
        "utility_label",
        "future_EEG",
        "future_keystroke"
    ],

    "input_shape": [
        61,
        25
    ],

    "sampling_rate_hz": 50,

    "normalization":
        "Phase 5H.3 channel-wise per-window z-score",

    "frozen_split":
        "Phase 6 subject-independent split",

    "selection_metric":
        "Validation AUPRC",

    "threshold_selection":
        "Validation F1 only",

    "best_candidate":
        str(best_candidate),

    "best_validation_auroc":
        float(
            best_row["validation_auroc"]
        ),

    "best_validation_auprc":
        float(
            best_row["validation_auprc"]
        ),

    "best_validation_f1":
        float(
            best_row["validation_f1"]
        ),

    "best_test_auroc":
        float(
            best_row["test_auroc"]
        ),

    "best_test_auprc":
        float(
            best_row["test_auprc"]
        ),

    "best_test_f1":
        float(
            best_row["test_f1"]
        ),

    "best_test_balanced_accuracy":
        float(
            best_row["test_balanced_accuracy"]
        ),

    "candidate_targets":
        candidate_scores,

    "target_thresholds":
        target_thresholds,

    "shuffle_control":
        shuffle_result,

    "output_directory":
        str(OUTPUT_ROOT)
}

STATUS_PATH = (
    OUTPUT_ROOT /
    "STEP7C_STATUS.json"
)

with open(
    STATUS_PATH,
    "w"
) as f:

    json.dump(
        status,
        f,
        indent=2
    )


# ============================================================
# 28. FINAL REPORT
# ============================================================

print("\n")
print("=" * 70)
print("STEP 7C COMPLETE")
print("=" * 70)

print(
    "\nValidation-best candidate:",
    best_candidate
)

print(
    "Validation AUROC:",
    f"{best_row['validation_auroc']:.6f}"
)

print(
    "Validation AUPRC:",
    f"{best_row['validation_auprc']:.6f}"
)

print(
    "Validation F1:",
    f"{best_row['validation_f1']:.6f}"
)

print(
    "\nFrozen test AUROC:",
    f"{best_row['test_auroc']:.6f}"
)

print(
    "Frozen test AUPRC:",
    f"{best_row['test_auprc']:.6f}"
)

print(
    "Frozen test balanced accuracy:",
    f"{best_row['test_balanced_accuracy']:.6f}"
)

print(
    "Frozen test F1:",
    f"{best_row['test_f1']:.6f}"
)

print(
    "\nShuffled-control test AUROC:",
    f"{shuffle_result['test_auroc']:.6f}"
)

print(
    "Shuffled-control test AUPRC:",
    f"{shuffle_result['test_auprc']:.6f}"
)

print("\nOutputs:")
print(
    OUTPUT_ROOT
)

print("\nFiles:")
for p in sorted(
    OUTPUT_ROOT.iterdir()
):
    print(
        " -",
        p.name
    )

print("\n" + "=" * 70)
print("IMPORTANT INTERPRETATION RULE")
print("=" * 70)

print(
    """
A strong result here means that an EEG-only pre-decoder
quality/information property is predictable from the EEG.

It does NOT by itself prove that the expensive decoder can
be safely skipped on the negative class.

For Phase 8 to be justified, the selected target must have:

1. Stable validation performance.
2. Strong frozen-test performance.
3. Subject-independent behavior.
4. Shuffled-label control near chance.
5. A defensible causal interpretation.
6. A meaningful relationship to downstream decoder utility.

If the best EEG-only target is weak or unstable, do NOT
claim that a causal utility gate has been solved.
"""
)

print("=" * 70)
```

```python
# ================================================================
# NEUROSENTINEL — STEP 7C FINAL RUN
# Causal Utility Target Validation
#
# IMPORTANT:
# - Uses existing Step 7C EEG-only feature file.
# - Uses current 150,955-window dataset.
# - Retrains ALL C1-C6 from scratch.
# - Does NOT use the old incompatible C1 checkpoint.
# - No decoder prediction/confidence/correctness is used.
# - No true character/class is used.
# - Thresholds are derived from TRAIN ONLY.
# - Candidate selection is based ONLY on validation AUPRC.
# - TEST is evaluated only after the candidate is frozen.
# - Includes shuffled-label control and subject-level test analysis.
# ================================================================

import os
import json
import random
import warnings
from pathlib import Path

import numpy as np
import pandas as pd

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    f1_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    accuracy_score,
)

warnings.filterwarnings("ignore")


# ================================================================
# 1. CONFIGURATION
# ================================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("=" * 70)
print("NEUROSENTINEL STEP 7C — FINAL RUN")
print("=" * 70)
print("Device:", DEVICE)
print("Seed:", SEED)


# ================================================================
# 2. ROBUST PATH DETECTION
# ================================================================

DRIVE_CANDIDATES = [
    Path("/content/gdrive/MyDrive"),
    Path("/content/drive/MyDrive"),
]

DRIVE = None

for p in DRIVE_CANDIDATES:
    if (p / "NeuroSentinel").exists():
        DRIVE = p
        break

if DRIVE is None:
    raise FileNotFoundError(
        "Could not find NeuroSentinel under:\n"
        "/content/gdrive/MyDrive\n"
        "or\n"
        "/content/drive/MyDrive"
    )

ROOT = DRIVE / "NeuroSentinel"

print("ROOT:", ROOT)


# ================================================================
# 3. FILE SEARCH
# ================================================================

def find_file(filename):

    matches = list(ROOT.rglob(filename))

    if not matches:
        raise FileNotFoundError(
            f"Could not find {filename} under {ROOT}"
        )

    if len(matches) > 1:
        print(f"\nMultiple matches found for {filename}:")
        for m in matches:
            print(" ", m)
        print("Using:", matches[0])

    return matches[0]


PHASE6_MANIFEST = find_file(
    "phase6_subject_split_manifest_resolved.csv"
)

H7_MANIFEST = find_file(
    "phase5h7_utility_label_manifest.csv"
)

FEATURE_FILE = find_file(
    "step7c_eeg_only_features.csv"
)

OUT_DIR = (
    ROOT
    / "results"
    / "step7c_causal_utility"
)

CKPT_DIR = (
    ROOT
    / "checkpoints"
    / "step7c_final"
)

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

CKPT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

print("\nPaths")
print("Phase 6 manifest:", PHASE6_MANIFEST)
print("H7 manifest:", H7_MANIFEST)
print("Feature file:", FEATURE_FILE)
print("Output:", OUT_DIR)
print("Fresh checkpoints:", CKPT_DIR)


# ================================================================
# 4. LOAD FEATURE DATA
# ================================================================

features = pd.read_csv(
    FEATURE_FILE
)

print("\nFeature dataset:")
print("Shape:", features.shape)

print("Columns:")
print(features.columns.tolist())


FEATURE_COLUMNS = [
    "amp_std_mean",
    "amp_std_median",
    "amp_std_std",
    "rms_mean",
    "rms_median",
    "abs_mean",
    "peak_to_peak_mean",
    "peak_to_peak_median",
    "q95_mean",
    "q50_mean",
    "diff_rms_mean",
    "diff_rms_median",
    "temporal_corr_mean",
    "temporal_corr_median",
    "spatial_corr_mean",
    "spatial_corr_abs",
    "total_power_mean",
    "delta_ratio_mean",
    "theta_ratio_mean",
    "alpha_ratio_mean",
    "beta_ratio_mean",
    "phys_ratio_mean",
    "high_freq_ratio_mean",
    "low_freq_ratio_mean",
    "spectral_entropy_mean",
    "spectral_entropy_std",
]

missing = [
    c for c in FEATURE_COLUMNS
    if c not in features.columns
]

if missing:
    raise RuntimeError(
        "Missing feature columns:\n"
        + "\n".join(missing)
    )

print(
    "\nFeature columns verified:",
    len(FEATURE_COLUMNS)
)


# ================================================================
# 5. LOAD PHASE 6 MANIFEST
# ================================================================

phase6 = pd.read_csv(
    PHASE6_MANIFEST
)

print(
    "\nPhase 6 manifest shape:",
    phase6.shape
)

print(
    "Phase 6 columns:",
    phase6.columns.tolist()
)


def normalize_id(x):
    return str(x).strip()


features["recording_id"] = (
    features["recording_id"]
    .map(normalize_id)
)

phase6["recording_id"] = (
    phase6["recording_id"]
    .map(normalize_id)
)


# ================================================================
# 6. VERIFY RECORDING → SPLIT MAPPING
# ================================================================

split_map = (
    phase6[
        [
            "recording_id",
            "subject",
            "split"
        ]
    ]
    .drop_duplicates()
)

if split_map["recording_id"].duplicated().any():

    raise RuntimeError(
        "Duplicate recording IDs in Phase 6 manifest."
    )

split_dict = dict(
    zip(
        split_map["recording_id"],
        split_map["split"]
    )
)

subject_dict = dict(
    zip(
        split_map["recording_id"],
        split_map["subject"]
    )
)

features["split_from_manifest"] = (
    features["recording_id"]
    .map(split_dict)
)

features["subject_from_manifest"] = (
    features["recording_id"]
    .map(subject_dict)
)

if features["split_from_manifest"].isna().any():

    missing_ids = sorted(
        features.loc[
            features["split_from_manifest"].isna(),
            "recording_id"
        ].unique()
    )

    raise RuntimeError(
        "Missing split mapping for:\n"
        + "\n".join(missing_ids)
    )

# ------------------------------------------------
# Verify feature CSV split agrees with Phase 6
# ------------------------------------------------

if "split" in features.columns:

    mismatch = (
        features["split"].astype(str)
        != features["split_from_manifest"].astype(str)
    )

    mismatch_count = int(mismatch.sum())

    print(
        "\nFeature CSV vs Phase 6 split mismatches:",
        mismatch_count
    )

    if mismatch_count > 0:

        raise RuntimeError(
            "Feature split information does not agree "
            "with the frozen Phase 6 split."
        )

# Use authoritative Phase 6 split
features["split"] = (
    features["split_from_manifest"]
)

features["subject"] = (
    features["subject_from_manifest"]
)


# ================================================================
# 7. LOCATE PHASE 5 WINDOWS
# ================================================================

print(
    "\nLocating Phase 5 baseline window files..."
)

WINDOW_ROOTS = [
    ROOT / "06_phase5_utility_labels" / "recordings",
    ROOT / "06_phase5_utility_labels",
]

WINDOW_ROOT = None

for candidate in WINDOW_ROOTS:

    if candidate.exists():

        files = list(
            candidate.rglob(
                "*_windows.npy"
            )
        )

        if files:

            WINDOW_ROOT = candidate
            break

if WINDOW_ROOT is None:

    raise FileNotFoundError(
        "Could not locate Phase 5 *_windows.npy files."
    )

window_files = sorted(
    WINDOW_ROOT.rglob(
        "*_windows.npy"
    )
)

print(
    "Window root:",
    WINDOW_ROOT
)

print(
    "Window files:",
    len(window_files)
)

if len(window_files) != 62:

    print(
        "WARNING: expected 62 recordings, "
        f"found {len(window_files)}"
    )


window_map = {}

for p in window_files:

    rid = p.name.replace(
        "_windows.npy",
        ""
    )

    window_map[rid] = p


feature_recordings = set(
    features["recording_id"].unique()
)

missing_windows = sorted(
    feature_recordings
    - set(window_map.keys())
)

if missing_windows:

    raise RuntimeError(
        "Missing window files:\n"
        + "\n".join(missing_windows)
    )

print(
    "All feature recordings have corresponding EEG windows."
)


# ================================================================
# 8. LOAD EEG WINDOWS
# ================================================================

print(
    "\nLoading EEG recordings..."
)

recording_cache = {}

recording_ids = sorted(
    feature_recordings
)

for i, rid in enumerate(
    recording_ids,
    start=1
):

    X = np.load(
        window_map[rid],
        mmap_mode="r"
    )

    if X.ndim != 3:

        raise RuntimeError(
            f"{rid}: unexpected shape {X.shape}"
        )

    if X.shape[1:] != (
        61,
        25
    ):

        raise RuntimeError(
            f"{rid}: expected "
            "(N,61,25), got "
            f"{X.shape}"
        )

    recording_cache[rid] = X

    if (
        i % 10 == 0
        or i == len(recording_ids)
    ):

        print(
            f"Loaded {i}/{len(recording_ids)} recordings"
        )

print(
    "\nLoaded normalized recordings:",
    len(recording_cache)
)


# ================================================================
# 9. VERIFY FEATURE / EEG ALIGNMENT
# ================================================================

print(
    "\nValidating feature/window alignment..."
)

aligned = []

for rid in recording_ids:

    f = (
        features[
            features["recording_id"] == rid
        ]
        .sort_values("window_index")
    )

    X = recording_cache[rid]

    if len(f) != X.shape[0]:

        raise RuntimeError(
            f"{rid}: feature rows={len(f)}, "
            f"EEG windows={X.shape[0]}"
        )

    expected = np.arange(
        X.shape[0]
    )

    actual = f[
        "window_index"
    ].to_numpy()

    if not np.array_equal(
        actual,
        expected
    ):

        raise RuntimeError(
            f"{rid}: window indices are not aligned."
        )

    aligned.append(f)

features = pd.concat(
    aligned,
    ignore_index=True
)

print(
    "Feature/window alignment verified."
)

print(
    "Total aligned rows:",
    len(features)
)


# ================================================================
# 10. BUILD GLOBAL EEG ARRAY
# ================================================================

print(
    "\nBuilding global EEG array..."
)

X_parts = []
M_parts = []

for rid in recording_ids:

    f = (
        features[
            features["recording_id"] == rid
        ]
        .sort_values("window_index")
    )

    X = np.asarray(
        recording_cache[rid],
        dtype=np.float32
    )

    X_parts.append(X)
    M_parts.append(f)

X_all = np.concatenate(
    X_parts,
    axis=0
)

meta = pd.concat(
    M_parts,
    ignore_index=True
)

print(
    "Global EEG:",
    X_all.shape
)

print(
    "Global metadata:",
    meta.shape
)


# ================================================================
# 11. SPLIT COUNTS
# ================================================================

split_array = (
    meta["split"]
    .astype(str)
    .to_numpy()
)

print(
    "\nSplit counts:"
)

for split in [
    "train",
    "val",
    "test"
]:

    n = int(
        (
            split_array == split
        ).sum()
    )

    print(
        f"{split:5s}: {n}"
    )


# ================================================================
# 12. CAUSAL SCORE CONSTRUCTION
# ================================================================
#
# These scores use ONLY EEG-derived information available before
# the expensive decoder is invoked.
#
# They do NOT use:
#   true class
#   decoder prediction
#   decoder confidence
#   decoder correctness
#   future EEG
#   future keystroke
# ================================================================

f = features

# ------------------------------------------------
# C1 — Energy
# ------------------------------------------------

C1 = (
    f["rms_mean"].to_numpy(
        dtype=np.float64
    )
    +
    f["abs_mean"].to_numpy(
        dtype=np.float64
    )
    +
    f["total_power_mean"].to_numpy(
        dtype=np.float64
    )
)

# ------------------------------------------------
# C2 — Temporal stability
# ------------------------------------------------

C2 = (
    f["temporal_corr_mean"].to_numpy(
        dtype=np.float64
    )
    +
    f["temporal_corr_median"].to_numpy(
        dtype=np.float64
    )
)

# ------------------------------------------------
# C3 — Spatial coherence
# ------------------------------------------------

C3 = (
    f["spatial_corr_mean"].to_numpy(
        dtype=np.float64
    )
    +
    f["spatial_corr_abs"].to_numpy(
        dtype=np.float64
    )
)

# ------------------------------------------------
# C4 — Physiological concentration
# ------------------------------------------------

C4 = (
    f["delta_ratio_mean"].to_numpy(
        dtype=np.float64
    )
    +
    f["theta_ratio_mean"].to_numpy(
        dtype=np.float64
    )
    +
    f["alpha_ratio_mean"].to_numpy(
        dtype=np.float64
    )
    +
    f["beta_ratio_mean"].to_numpy(
        dtype=np.float64
    )
    +
    f["phys_ratio_mean"].to_numpy(
        dtype=np.float64
    )
)

# ------------------------------------------------
# C5 — Low spectral entropy
# ------------------------------------------------

C5 = -f[
    "spectral_entropy_mean"
].to_numpy(
    dtype=np.float64
)


# ================================================================
# 13. TRAIN-ONLY STANDARDIZATION FOR C6
# ================================================================

train_mask = (
    split_array == "train"
)


def train_zscore(
    values,
    train_mask
):

    values = np.asarray(
        values,
        dtype=np.float64
    )

    train_values = values[
        train_mask
    ]

    mu = np.nanmean(
        train_values
    )

    sd = np.nanstd(
        train_values
    )

    if (
        not np.isfinite(sd)
        or sd < 1e-12
    ):

        sd = 1.0

    return (
        values - mu
    ) / sd


Z1 = train_zscore(
    C1,
    train_mask
)

Z2 = train_zscore(
    C2,
    train_mask
)

Z3 = train_zscore(
    C3,
    train_mask
)

Z4 = train_zscore(
    C4,
    train_mask
)

Z5 = train_zscore(
    C5,
    train_mask
)

C6 = (
    Z1
    + Z2
    + Z3
    + Z4
    + Z5
) / 5.0


candidate_scores = [
    "C1_energy",
    "C2_temporal_stability",
    "C3_spatial_coherence",
    "C4_physiological_concentration",
    "C5_low_spectral_entropy",
    "C6_composite_quality",
]

score_arrays = {
    "C1_energy": C1,
    "C2_temporal_stability": C2,
    "C3_spatial_coherence": C3,
    "C4_physiological_concentration": C4,
    "C5_low_spectral_entropy": C5,
    "C6_composite_quality": C6,
}


# ================================================================
# 14. CREATE TARGETS
# ================================================================
#
# IMPORTANT:
# Q75 threshold is calculated ONLY on TRAIN.
# ================================================================

TARGETS = {}

distribution_rows = []

print("\n")
print("=" * 70)
print("CAUSAL TARGET DISTRIBUTIONS")
print("=" * 70)

for candidate in candidate_scores:

    score = score_arrays[
        candidate
    ]

    threshold = np.nanpercentile(
        score[train_mask],
        75
    )

    y = (
        score >= threshold
    ).astype(
        np.int64
    )

    TARGETS[candidate] = y

    for split in [
        "train",
        "val",
        "test"
    ]:

        mask = (
            split_array == split
        )

        n = int(mask.sum())

        positive = int(
            y[mask].sum()
        )

        negative = (
            n - positive
        )

        distribution_rows.append({
            "candidate": candidate,
            "split": split,
            "n": n,
            "positive": positive,
            "negative": negative,
            "positive_rate":
                positive / n,
            "train_q75_threshold":
                float(threshold),
        })

distribution_df = pd.DataFrame(
    distribution_rows
)

print(
    distribution_df[
        [
            "candidate",
            "split",
            "n",
            "positive",
            "negative",
            "positive_rate",
        ]
    ].to_string(
        index=False
    )
)

distribution_df.to_csv(
    OUT_DIR
    / "step7c_target_distributions_final.csv",
    index=False
)


# ================================================================
# 15. DATASET
# ================================================================

class EEGDataset(
    Dataset
):

    def __init__(
        self,
        X,
        y
    ):

        self.X = torch.from_numpy(
            np.asarray(
                X,
                dtype=np.float32
            )
        )

        self.y = torch.from_numpy(
            np.asarray(
                y,
                dtype=np.float32
            )
        )

    def __len__(self):

        return len(
            self.y
        )

    def __getitem__(
        self,
        idx
    ):

        return (
            self.X[idx],
            self.y[idx]
        )


# ================================================================
# 16. MODEL
# ================================================================

class CausalGateCNN(
    nn.Module
):

    def __init__(self):

        super().__init__()

        self.encoder = nn.Sequential(

            nn.Conv1d(
                61,
                64,
                kernel_size=5,
                padding=2
            ),

            nn.ReLU(),

            nn.Conv1d(
                64,
                64,
                kernel_size=5,
                dilation=2,
                padding=4
            ),

            nn.ReLU(),

            nn.Conv1d(
                64,
                64,
                kernel_size=5,
                dilation=4,
                padding=8
            ),

            nn.ReLU(),

            nn.Conv1d(
                64,
                32,
                kernel_size=5,
                dilation=8,
                padding=16
            ),

            nn.ReLU(),
        )

        self.head = nn.Linear(
            32,
            1
        )

    def forward(
        self,
        x
    ):

        x = self.encoder(
            x
        )

        x = x.mean(
            dim=2
        )

        return self.head(
            x
        ).squeeze(1)


# Parameter count

parameter_count = sum(
    p.numel()
    for p in CausalGateCNN().parameters()
)

print(
    "\nCausalGateCNN parameters:",
    parameter_count
)


# ================================================================
# 17. METRIC HELPERS
# ================================================================

def safe_auc(
    y,
    p
):

    try:

        return roc_auc_score(
            y,
            p
        )

    except Exception:

        return np.nan


def safe_auprc(
    y,
    p
):

    try:

        return average_precision_score(
            y,
            p
        )

    except Exception:

        return np.nan


def find_best_threshold(
    y,
    p
):

    thresholds = np.arange(
        0.05,
        0.951,
        0.01
    )

    best_f1 = -1.0
    best_threshold = 0.5

    for threshold in thresholds:

        pred = (
            p >= threshold
        ).astype(int)

        f1 = f1_score(
            y,
            pred,
            zero_division=0
        )

        if f1 > best_f1:

            best_f1 = f1
            best_threshold = (
                float(threshold)
            )

    return (
        best_threshold,
        best_f1
    )


def classification_metrics(
    y,
    p,
    threshold
):

    pred = (
        p >= threshold
    ).astype(int)

    return {

        "accuracy":
            accuracy_score(
                y,
                pred
            ),

        "balanced_accuracy":
            balanced_accuracy_score(
                y,
                pred
            ),

        "precision":
            precision_score(
                y,
                pred,
                zero_division=0
            ),

        "recall":
            recall_score(
                y,
                pred,
                zero_division=0
            ),

        "f1":
            f1_score(
                y,
                pred,
                zero_division=0
            ),

        "auroc":
            safe_auc(
                y,
                p
            ),

        "auprc":
            safe_auprc(
                y,
                p
            ),

        "threshold":
            float(threshold),
    }


# ================================================================
# 18. DATA LOADERS
# ================================================================

BATCH_SIZE = 256

LR = 1e-3

WEIGHT_DECAY = 1e-4

MAX_EPOCHS = 20

PATIENCE = 5


def make_loader(
    X,
    y,
    shuffle=False
):

    return DataLoader(
        EEGDataset(
            X,
            y
        ),
        batch_size=BATCH_SIZE,
        shuffle=shuffle,
        num_workers=0,
        pin_memory=torch.cuda.is_available(),
    )


def predict_model(
    model,
    loader
):

    model.eval()

    probabilities = []
    targets = []

    with torch.no_grad():

        for xb, yb in loader:

            xb = xb.to(
                DEVICE,
                non_blocking=True
            )

            logits = model(
                xb
            )

            probability = torch.sigmoid(
                logits
            )

            probabilities.append(
                probability
                .cpu()
                .numpy()
            )

            targets.append(
                yb.numpy()
            )

    return (
        np.concatenate(
            targets
        ),
        np.concatenate(
            probabilities
        )
    )


# ================================================================
# 19. TRAIN ONE CANDIDATE
# ================================================================

def train_candidate(
    candidate
):

    print("\n")
    print("=" * 70)
    print(
        "TRAINING:",
        candidate
    )
    print("=" * 70)

    y = TARGETS[
        candidate
    ]

    train_idx = np.where(
        split_array == "train"
    )[0]

    val_idx = np.where(
        split_array == "val"
    )[0]

    test_idx = np.where(
        split_array == "test"
    )[0]

    X_train = X_all[
        train_idx
    ]

    X_val = X_all[
        val_idx
    ]

    X_test = X_all[
        test_idx
    ]

    y_train = y[
        train_idx
    ]

    y_val = y[
        val_idx
    ]

    y_test = y[
        test_idx
    ]

    print(
        "Train:",
        X_train.shape,
        "positive:",
        int(y_train.sum())
    )

    print(
        "Val:",
        X_val.shape,
        "positive:",
        int(y_val.sum())
    )

    print(
        "Test:",
        X_test.shape,
        "positive:",
        int(y_test.sum())
    )

    positives = max(
        int(y_train.sum()),
        1
    )

    negatives = max(
        len(y_train)
        - positives,
        1
    )

    pos_weight = (
        negatives
        / positives
    )

    print(
        "pos_weight:",
        pos_weight
    )

    train_loader = make_loader(
        X_train,
        y_train,
        shuffle=True
    )

    val_loader = make_loader(
        X_val,
        y_val,
        shuffle=False
    )

    test_loader = make_loader(
        X_test,
        y_test,
        shuffle=False
    )

    model = CausalGateCNN().to(
        DEVICE
    )

    criterion = (
        nn.BCEWithLogitsLoss(
            pos_weight=torch.tensor(
                pos_weight,
                dtype=torch.float32,
                device=DEVICE
            )
        )
    )

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=LR,
        weight_decay=WEIGHT_DECAY
    )

    # ------------------------------------------------------------
    # IMPORTANT:
    # Fresh checkpoint path.
    # The previous incompatible C1 checkpoint is NOT used.
    # ------------------------------------------------------------

    checkpoint_path = (
        CKPT_DIR
        / f"{candidate}_best.pt"
    )

    history = []

    best_val_auprc = -np.inf

    best_epoch = 0

    patience_counter = 0

    # ------------------------------------------------------------
    # TRAIN FROM SCRATCH
    # ------------------------------------------------------------

    for epoch in range(
        1,
        MAX_EPOCHS + 1
    ):

        model.train()

        losses = []

        for xb, yb in train_loader:

            xb = xb.to(
                DEVICE,
                non_blocking=True
            )

            yb = yb.to(
                DEVICE,
                non_blocking=True
            )

            optimizer.zero_grad(
                set_to_none=True
            )

            logits = model(
                xb
            )

            loss = criterion(
                logits,
                yb
            )

            loss.backward()

            optimizer.step()

            losses.append(
                loss.item()
            )

        train_loss = float(
            np.mean(
                losses
            )
        )

        # --------------------------------------------------------
        # Validation
        # --------------------------------------------------------

        yv, pv = predict_model(
            model,
            val_loader
        )

        val_auroc = safe_auc(
            yv,
            pv
        )

        val_auprc = safe_auprc(
            yv,
            pv
        )

        val_threshold, val_f1 = (
            find_best_threshold(
                yv,
                pv
            )
        )

        val_metrics = (
            classification_metrics(
                yv,
                pv,
                val_threshold
            )
        )

        history.append({

            "epoch": epoch,

            "train_loss":
                train_loss,

            "val_auroc":
                val_metrics[
                    "auroc"
                ],

            "val_auprc":
                val_metrics[
                    "auprc"
                ],

            "val_f1":
                val_metrics[
                    "f1"
                ],

            "val_balanced_accuracy":
                val_metrics[
                    "balanced_accuracy"
                ],

            "threshold":
                val_threshold,
        })

        print(
            f"Epoch {epoch:02d} | "
            f"train_loss={train_loss:.4f} | "
            f"val_AUROC={val_auroc:.4f} | "
            f"val_AUPRC={val_auprc:.4f} | "
            f"val_F1={val_f1:.4f} | "
            f"thr={val_threshold:.2f}"
        )

        # --------------------------------------------------------
        # Select model ONLY using validation AUPRC
        # --------------------------------------------------------

        if (
            np.isfinite(
                val_auprc
            )
            and
            val_auprc
            > best_val_auprc
        ):

            best_val_auprc = (
                val_auprc
            )

            best_epoch = (
                epoch
            )

            patience_counter = 0

            torch.save(
                {
                    "candidate":
                        candidate,

                    "model_state":
                        model.state_dict(),

                    "best_val_auprc":
                        best_val_auprc,

                    "best_epoch":
                        best_epoch,

                    "history":
                        history,
                },
                checkpoint_path
            )

            print(
                "  Saved validation-best checkpoint."
            )

        else:

            patience_counter += 1

            if (
                patience_counter
                >= PATIENCE
            ):

                print(
                    "Early stopping."
                )

                break

    # ============================================================
    # 20. LOAD BEST VALIDATION CHECKPOINT
    # ============================================================

    print(
        "\nLoading validation-best checkpoint..."
    )

    checkpoint = torch.load(
        checkpoint_path,
        map_location=DEVICE,
        weights_only=False
    )

    model.load_state_dict(
        checkpoint[
            "model_state"
        ]
    )

    best_epoch = checkpoint[
        "best_epoch"
    ]

    best_val_auprc = checkpoint[
        "best_val_auprc"
    ]

    # ------------------------------------------------------------
    # Validation metrics
    # ------------------------------------------------------------

    yv, pv = predict_model(
        model,
        val_loader
    )

    val_threshold, _ = (
        find_best_threshold(
            yv,
            pv
        )
    )

    val_metrics = (
        classification_metrics(
            yv,
            pv,
            val_threshold
        )
    )

    # ------------------------------------------------------------
    # FROZEN TEST
    # ------------------------------------------------------------

    yt, pt = predict_model(
        model,
        test_loader
    )

    test_metrics = (
        classification_metrics(
            yt,
            pt,
            val_threshold
        )
    )

    print(
        "\nValidation-best:"
    )

    print(
        json.dumps(
            val_metrics,
            indent=2
        )
    )

    print(
        "\nFrozen test:"
    )

    print(
        json.dumps(
            test_metrics,
            indent=2
        )
    )

    # ------------------------------------------------------------
    # Save test predictions
    # ------------------------------------------------------------

    prediction_df = pd.DataFrame({

        "y_true":
            yt.astype(int),

        "probability":
            pt,

        "prediction":
            (
                pt
                >= val_threshold
            ).astype(int),
    })

    prediction_df.to_csv(
        OUT_DIR
        / f"{candidate}_test_predictions.csv",
        index=False
    )

    result = {

        "candidate":
            candidate,

        "best_epoch":
            int(best_epoch),

        "val_auroc":
            float(
                val_metrics[
                    "auroc"
                ]
            ),

        "val_auprc":
            float(
                val_metrics[
                    "auprc"
                ]
            ),

        "val_f1":
            float(
                val_metrics[
                    "f1"
                ]
            ),

        "val_balanced_accuracy":
            float(
                val_metrics[
                    "balanced_accuracy"
                ]
            ),

        "val_precision":
            float(
                val_metrics[
                    "precision"
                ]
            ),

        "val_recall":
            float(
                val_metrics[
                    "recall"
                ]
            ),

        "val_threshold":
            float(
                val_threshold
            ),

        "test_auroc":
            float(
                test_metrics[
                    "auroc"
                ]
            ),

        "test_auprc":
            float(
                test_metrics[
                    "auprc"
                ]
            ),

        "test_f1":
            float(
                test_metrics[
                    "f1"
                ]
            ),

        "test_balanced_accuracy":
            float(
                test_metrics[
                    "balanced_accuracy"
                ]
            ),

        "test_precision":
            float(
                test_metrics[
                    "precision"
                ]
            ),

        "test_recall":
            float(
                test_metrics[
                    "recall"
                ]
            ),

        "test_accuracy":
            float(
                test_metrics[
                    "accuracy"
                ]
            ),
    }

    return (
        result,
        model
    )


# ================================================================
# 21. TRAIN ALL SIX CANDIDATES
# ================================================================

results = []

trained_models = {}

for candidate in candidate_scores:

    result, model = train_candidate(
        candidate
    )

    results.append(
        result
    )

    trained_models[
        candidate
    ] = model

    # Save after EVERY candidate
    pd.DataFrame(
        results
    ).to_csv(
        OUT_DIR
        / "step7c_candidate_results_partial.csv",
        index=False
    )

    print(
        "\nCompleted:",
        candidate
    )


# ================================================================
# 22. VALIDATION-ONLY CANDIDATE SELECTION
# ================================================================

results_df = pd.DataFrame(
    results
)

results_df = (
    results_df
    .sort_values(
        "val_auprc",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)

print("\n")
print("=" * 70)
print("STEP 7C CANDIDATE COMPARISON")
print("=" * 70)

print(
    results_df.to_string(
        index=False
    )
)

results_df.to_csv(
    OUT_DIR
    / "step7c_candidate_results.csv",
    index=False
)

best_candidate = str(
    results_df.iloc[0][
        "candidate"
    ]
)

print(
    "\nValidation-best candidate:",
    best_candidate
)

print(
    "Selection metric: Validation AUPRC"
)


# ================================================================
# 23. SHUFFLED-LABEL CONTROL
# ================================================================

print("\n")
print("=" * 70)
print("SHUFFLED-LABEL CONTROL")
print("=" * 70)

best_model = (
    trained_models[
        best_candidate
    ]
)

y_best = TARGETS[
    best_candidate
]

train_idx = np.where(
    split_array == "train"
)[0]

val_idx = np.where(
    split_array == "val"
)[0]

test_idx = np.where(
    split_array == "test"
)[0]


# ------------------------------------------------
# Shuffle training labels ONLY
# ------------------------------------------------

rng = np.random.default_rng(
    SEED
)

y_shuffle_train = (
    y_best[
        train_idx
    ].copy()
)

rng.shuffle(
    y_shuffle_train
)

# ------------------------------------------------
# Data
# ------------------------------------------------

X_train = X_all[
    train_idx
]

X_val = X_all[
    val_idx
]

y_val = y_best[
    val_idx
]

shuffle_model = (
    CausalGateCNN()
    .to(DEVICE)
)

positives = max(
    int(y_shuffle_train.sum()),
    1
)

negatives = max(
    len(y_shuffle_train)
    - positives,
    1
)

shuffle_pos_weight = (
    negatives
    / positives
)

criterion = (
    nn.BCEWithLogitsLoss(
        pos_weight=torch.tensor(
            shuffle_pos_weight,
            dtype=torch.float32,
            device=DEVICE
        )
    )
)

optimizer = torch.optim.AdamW(
    shuffle_model.parameters(),
    lr=LR,
    weight_decay=WEIGHT_DECAY
)

shuffle_train_loader = make_loader(
    X_train,
    y_shuffle_train,
    shuffle=True
)

shuffle_val_loader = make_loader(
    X_val,
    y_val,
    shuffle=False
)

best_shuffle_auprc = -np.inf

best_shuffle_state = None

shuffle_patience = 0

for epoch in range(
    1,
    MAX_EPOCHS + 1
):

    shuffle_model.train()

    losses = []

    for xb, yb in (
        shuffle_train_loader
    ):

        xb = xb.to(
            DEVICE,
            non_blocking=True
        )

        yb = yb.to(
            DEVICE,
            non_blocking=True
        )

        optimizer.zero_grad(
            set_to_none=True
        )

        logits = shuffle_model(
            xb
        )

        loss = criterion(
            logits,
            yb
        )

        loss.backward()

        optimizer.step()

        losses.append(
            loss.item()
        )

    yv, pv = predict_model(
        shuffle_model,
        shuffle_val_loader
    )

    auprc = safe_auprc(
        yv,
        pv
    )

    print(
        f"Shuffle epoch {epoch:02d} | "
        f"val_AUPRC={auprc:.4f}"
    )

    if auprc > best_shuffle_auprc:

        best_shuffle_auprc = (
            auprc
        )

        best_shuffle_state = {
            k:
                v.detach()
                .cpu()
                .clone()

            for k, v in
            shuffle_model
            .state_dict()
            .items()
        }

        shuffle_patience = 0

    else:

        shuffle_patience += 1

        if (
            shuffle_patience
            >= PATIENCE
        ):

            break


shuffle_model.load_state_dict(
    best_shuffle_state
)

y_shuffle_val, p_shuffle_val = (
    predict_model(
        shuffle_model,
        shuffle_val_loader
    )
)

shuffle_threshold, _ = (
    find_best_threshold(
        y_shuffle_val,
        p_shuffle_val
    )
)

shuffle_metrics = (
    classification_metrics(
        y_shuffle_val,
        p_shuffle_val,
        shuffle_threshold
    )
)

print(
    "\nShuffled-label validation metrics:"
)

print(
    json.dumps(
        shuffle_metrics,
        indent=2
    )
)


# ================================================================
# 24. SUBJECT-LEVEL TEST ANALYSIS
# ================================================================

print("\n")
print("=" * 70)
print("SUBJECT-LEVEL TEST ANALYSIS")
print("=" * 70)

X_test = X_all[
    test_idx
]

y_test = y_best[
    test_idx
]

test_loader = make_loader(
    X_test,
    y_test,
    shuffle=False
)

yt, pt = predict_model(
    best_model,
    test_loader
)

frozen_threshold = float(
    results_df.iloc[0][
        "val_threshold"
    ]
)

test_meta = meta.iloc[
    test_idx
].copy()

test_meta = (
    test_meta
    .reset_index(
        drop=True
    )
)

test_meta[
    "y_true"
] = yt.astype(int)

test_meta[
    "probability"
] = pt

test_meta[
    "prediction"
] = (
    pt
    >= frozen_threshold
).astype(int)


subject_rows = []

for subject, group in (
    test_meta
    .groupby("subject")
):

    ys = group[
        "y_true"
    ].to_numpy()

    ps = group[
        "probability"
    ].to_numpy()

    preds = group[
        "prediction"
    ].to_numpy()

    subject_rows.append({

        "subject":
            str(subject),

        "n":
            int(len(group)),

        "positive_rate":
            float(ys.mean()),

        "accuracy":
            float(
                accuracy_score(
                    ys,
                    preds
                )
            ),

        "balanced_accuracy":
            float(
                balanced_accuracy_score(
                    ys,
                    preds
                )
            ),

        "precision":
            float(
                precision_score(
                    ys,
                    preds,
                    zero_division=0
                )
            ),

        "recall":
            float(
                recall_score(
                    ys,
                    preds,
                    zero_division=0
                )
            ),

        "f1":
            float(
                f1_score(
                    ys,
                    preds,
                    zero_division=0
                )
            ),

        "auroc":
            float(
                safe_auc(
                    ys,
                    ps
                )
            ),

        "auprc":
            float(
                safe_auprc(
                    ys,
                    ps
                )
            ),
    })


subject_df = pd.DataFrame(
    subject_rows
)

print(
    subject_df.to_string(
        index=False
    )
)

subject_df.to_csv(
    OUT_DIR
    / "step7c_subject_level_test.csv",
    index=False
)


# ================================================================
# 25. FINAL SUMMARY
# ================================================================

best_row = (
    results_df.iloc[0]
    .to_dict()
)

summary = {

    "phase":
        "Step 7C",

    "seed":
        SEED,

    "device":
        str(DEVICE),

    "dataset_rows":
        int(len(features)),

    "dataset_shape":
        list(features.shape),

    "eeg_shape":
        list(X_all.shape),

    "model_parameters":
        int(parameter_count),

    "candidates":
        candidate_scores,

    "selection_metric":
        "validation_AUPRC",

    "best_candidate":
        best_candidate,

    "best_candidate_validation":
        best_row,

    "shuffled_label_control":
        shuffle_metrics,

    "subject_level_test":
        subject_df.to_dict(
            orient="records"
        ),

    "test_was_used_for_selection":
        False,

    "decoder_information_used":
        False,

    "true_character_used":
        False,

    "future_information_used":
        False,

    "old_checkpoint_used":
        False,

    "fresh_training":
        True,
}


with open(
    OUT_DIR
    / "step7c_final_summary.json",
    "w"
) as f:

    json.dump(
        summary,
        f,
        indent=2,
        default=float
    )


# ================================================================
# 26. SAVE FINAL TARGET FOR THE SELECTED CANDIDATE
# ================================================================

selected_target_df = meta[
    [
        "recording_id",
        "subject",
        "split",
        "window_index",
    ]
].copy()

selected_target_df[
    "causal_candidate"
] = best_candidate

selected_target_df[
    "causal_target"
] = TARGETS[
    best_candidate
]

selected_target_df[
    "causal_score"
] = score_arrays[
    best_candidate
]

selected_target_df.to_csv(
    OUT_DIR
    / "step7c_selected_causal_target.csv",
    index=False
)


# ================================================================
# 27. FINAL REPORT
# ================================================================

print("\n")
print("=" * 70)
print("STEP 7C FINAL RUN COMPLETE")
print("=" * 70)

print(
    "\nValidation-best candidate:",
    best_candidate
)

print(
    "Validation AUROC:",
    f"{best_row['val_auroc']:.6f}"
)

print(
    "Validation AUPRC:",
    f"{best_row['val_auprc']:.6f}"
)

print(
    "Validation F1:",
    f"{best_row['val_f1']:.6f}"
)

print(
    "Validation balanced accuracy:",
    f"{best_row['val_balanced_accuracy']:.6f}"
)

print(
    "\nFrozen test AUROC:",
    f"{best_row['test_auroc']:.6f}"
)

print(
    "Frozen test AUPRC:",
    f"{best_row['test_auprc']:.6f}"
)

print(
    "Frozen test F1:",
    f"{best_row['test_f1']:.6f}"
)

print(
    "Frozen test balanced accuracy:",
    f"{best_row['test_balanced_accuracy']:.6f}"
)

print(
    "\nShuffled-label AUROC:",
    f"{shuffle_metrics['auroc']:.6f}"
)

print(
    "Shuffled-label AUPRC:",
    f"{shuffle_metrics['auprc']:.6f}"
)

print(
    "\nOutputs:"
)

print(
    OUT_DIR
)

print(
    "\nImportant files:"
)

print(
    "1.",
    OUT_DIR
    / "step7c_candidate_results.csv"
)

print(
    "2.",
    OUT_DIR
    / "step7c_target_distributions_final.csv"
)

print(
    "3.",
    OUT_DIR
    / "step7c_subject_level_test.csv"
)

print(
    "4.",
    OUT_DIR
    / "step7c_selected_causal_target.csv"
)

print(
    "5.",
    OUT_DIR
    / "step7c_final_summary.json"
)

print("\n")
print("=" * 70)
print("PHASE 7C DONE")
print("=" * 70)
```

```python

from pathlib import Path
import os

DRIVE_ROOT = Path("/content/gdrive/MyDrive")

skip_dirs = {
    ".git",
    ".ipynb_checkpoints",
    "brain2qwerty",
    ".Trash",
}

matches = []

for root, dirs, files in os.walk(DRIVE_ROOT):
    dirs[:] = [d for d in dirs if d not in skip_dirs]

    for filename in files:
        path = Path(root) / filename

        if path.suffix.lower() in {".ipynb", ".py"}:
            # Exclude the cloned dependency and Git internals
            if "brain2qwerty" in path.parts or ".git" in path.parts:
                continue
            matches.append(path)

print("Python files and notebooks found in My Drive:", len(matches))

for path in sorted(matches):
    print(path)

```
